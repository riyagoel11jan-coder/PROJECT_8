```python
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
logger_name = "JobMarketProject"
BASE_DIR = Path.cwd()
DATA_DIR = BASE_DIR
EXPORT_DIR = BASE_DIR / "exports"
CHART_DIR = BASE_DIR / "charts"
MODEL_DIR = BASE_DIR / "models"
REPORT_DIR = BASE_DIR / "reports"
LOG_DIR = BASE_DIR / "logs"
for directory in [
    EXPORT_DIR,
    CHART_DIR,
    MODEL_DIR,
    REPORT_DIR,
    LOG_DIR
]:
    directory.mkdir(parents=True, exist_ok=True)
DATA_FILE = DATA_DIR / "JD.csv"
LOG_FILE = LOG_DIR / "job_market_analysis.log"
logger = logging.getLogger(logger_name)
logger.setLevel(logging.INFO)
if logger.hasHandlers():
    logger.handlers.clear()
file_handler = logging.FileHandler(
    LOG_FILE,
    mode="a",
    encoding="utf-8"
)
file_handler.setLevel(logging.INFO)
console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)
log_formatter = logging.Formatter(
    "%(asctime)s | %(levelname)s | %(name)s | %(message)s"
)
file_handler.setFormatter(log_formatter)
console_handler.setFormatter(log_formatter)
logger.addHandler(file_handler)
logger.addHandler(console_handler)
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.max_colwidth", 120)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")
logger.info("Job Market Analysis project initialized successfully.")
```

```python
if DATA_FILE.exists():

    file_size_mb = DATA_FILE.stat().st_size / (1024 ** 2)

    logger.info(
        "Dataset found successfully | File: %s | Size: %.2f MB",
        DATA_FILE.name,
        file_size_mb
    )

    print("=" * 65)
    print("DATASET CHECK")
    print("=" * 65)

    print("Status    : Dataset found")
    print(f"File      : {DATA_FILE.name}")
    print(f"Size      : {file_size_mb:.2f} MB")
    print(f"Location  : {DATA_FILE}")

    print("=" * 65)

else:

    logger.error(
        "Dataset not found at %s",
        DATA_FILE
    )

    raise FileNotFoundError(
        f"\nJD.csv was not found at:\n{DATA_FILE}\n\n"
        "Place JD.csv inside the project folder "
        "or update DATA_DIR in the configuration cell."
    )
```

```python
EXPECTED_COLUMNS = [
    "title",
    "link",
    "published_date",
    "is_hourly",
    "hourly_low",
    "hourly_high",
    "budget",
    "country"
]
if not DATA_FILE.exists():

    logger.error("Dataset not found: %s", DATA_FILE)

    raise FileNotFoundError(
        f"JD.csv was not found at:\n{DATA_FILE}\n\n"
        "Place the dataset in the configured DATA_DIR "
        "or update the project path."
    )

file_size_mb = DATA_FILE.stat().st_size / (1024 ** 2)
logger.info(
    "Dataset located | File=%s | Size=%.2f MB",
    DATA_FILE.name,
    file_size_mb
)
try:

    df_raw = pd.read_csv(
        DATA_FILE,
        low_memory=False
    )

    logger.info(
        "Dataset loaded successfully | Rows=%s | Columns=%s",
        f"{df_raw.shape[0]:,}",
        df_raw.shape[1]
    )

except Exception as exc:

    logger.exception("Dataset loading failed.")
    raise RuntimeError(
        "Unable to load JD.csv."
    ) from exc

df = df_raw.copy(deep=True)
logger.info(
    "Working dataframe created from immutable raw dataset."
)
actual_columns = df_raw.columns.tolist()
missing_columns = [
    column
    for column in EXPECTED_COLUMNS
    if column not in actual_columns
]
unexpected_columns = [
    column
    for column in actual_columns
    if column not in EXPECTED_COLUMNS
]
schema_valid = (
    len(missing_columns) == 0
    and len(unexpected_columns) == 0
)
if schema_valid:

    logger.info(
        "Schema validation passed."
    )
else:

    logger.warning(
        "Schema discrepancy detected | Missing=%s | Unexpected=%s",
        missing_columns,
        unexpected_columns
    )
dataset_metadata = {
    "file_name": DATA_FILE.name,
    "file_size_mb": round(file_size_mb, 2),
    "rows": int(df_raw.shape[0]),
    "columns": int(df_raw.shape[1]),
    "schema_valid": schema_valid,
    "expected_columns": EXPECTED_COLUMNS,
    "actual_columns": actual_columns,
    "missing_columns": missing_columns,
    "unexpected_columns": unexpected_columns
}
dataset_metadata.update({
    "exact_duplicate_rows":
        int(df_raw.duplicated().sum()),
    "missing_title":
        int(df_raw["title"].isna().sum()),
    "missing_link":
        int(df_raw["link"].isna().sum()),
    "missing_country":
        int(df_raw["country"].isna().sum()),
    "unique_non_null_links":
        int(df_raw["link"].nunique(dropna=True)),
    "unique_titles":
        int(df_raw["title"].nunique(dropna=True))
})
logger.info(
    "Initial audit completed | "
    "Rows=%s | Columns=%s | Duplicates=%s | "
    "Missing link=%s | Missing country=%s",
    f"{dataset_metadata['rows']:,}",
    dataset_metadata["columns"],
    dataset_metadata["exact_duplicate_rows"],
    dataset_metadata["missing_link"],
    dataset_metadata["missing_country"]
)
metadata_file = EXPORT_DIR / "dataset_metadata.json"
with open(
    metadata_file,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        dataset_metadata,
        file,
        indent=4,
        default=str
    )
logger.info(
    "Dataset metadata exported to %s",
    metadata_file
)
```

```python
if "df_raw" not in globals():
    raise NameError("Run the Data Ingestion cell first — df_raw is missing.")

EXPORT_DIR = globals().get("EXPORT_DIR", Path.cwd() / "exports")
EXPORT_DIR.mkdir(parents=True, exist_ok=True)

EXPECTED_COLUMNS = [
    "title", "link", "published_date", "is_hourly",
    "hourly_low", "hourly_high", "budget", "country"
]

schema_valid = set(EXPECTED_COLUMNS).issubset(df_raw.columns)
n, p = df_raw.shape
missing_pct = df_raw.isna().mean().mul(100)
completeness = 100 - missing_pct.mean()
duplicates = int(df_raw.duplicated().sum())
duplicate_links = int(df_raw["link"].dropna().duplicated().sum())
title_complete = (1 - df_raw["title"].isna().mean()) * 100
link_complete = (1 - df_raw["link"].isna().mean()) * 100
country_coverage = (1 - df_raw["country"].isna().mean()) * 100
countries = int(df_raw["country"].nunique(dropna=True))

dates = pd.to_datetime(
    df_raw["published_date"],
    errors="coerce",
    utc=True
)

date_validity = dates.notna().mean() * 100
date_start, date_end = dates.min(), dates.max()

date_span = (
    (date_end.normalize() - date_start.normalize()).days + 1
    if dates.notna().any()
    else 0
)

refresh_time = pd.Timestamp.now(
    tz="Asia/Kolkata"
).strftime("%d %b %Y • %I:%M %p IST")

data_through = (
    date_end.strftime("%d %b %Y")
    if pd.notna(date_end)
    else "Unavailable"
)

hourly = (
    df_raw["is_hourly"]
    .astype(str)
    .str.strip()
    .str.lower()
    .map({
        "true": True, "1": True, "yes": True,
        "false": False, "0": False, "no": False
    })
)

hourly_n = int(hourly.eq(True).sum())
fixed_n = int(hourly.eq(False).sum())

hourly_complete = int(
    (
        hourly.eq(True)
        & df_raw["hourly_low"].notna()
        & df_raw["hourly_high"].notna()
    ).sum()
)

fixed_complete = int(
    (
        hourly.eq(False)
        & df_raw["budget"].notna()
    ).sum()
)

hourly_coverage = hourly_complete / hourly_n * 100 if hourly_n else 0
fixed_coverage = fixed_complete / fixed_n * 100 if fixed_n else 0

quality_df = pd.DataFrame({
    "Quality Check": [
        "Expected Schema",
        "Exact Duplicate Rows",
        "Duplicate Posting Links",
        "Title Completeness",
        "Link Completeness",
        "Country Coverage",
        "Published-Date Validity",
        "Hourly Pay Coverage",
        "Fixed Budget Coverage"
    ],
    "Result": [
        "PASS" if schema_valid else "REVIEW",
        f"{duplicates:,}",
        f"{duplicate_links:,}",
        f"{title_complete:.2f}%",
        f"{link_complete:.2f}%",
        f"{country_coverage:.2f}%",
        f"{date_validity:.2f}%",
        f"{hourly_coverage:.2f}%",
        f"{fixed_coverage:.2f}%"
    ]
})

missing_df = (
    missing_pct.rename("Missing %")
    .reset_index()
    .rename(columns={"index": "Column"})
    .sort_values("Missing %")
    .reset_index(drop=True)
)

worst = missing_df.iloc[-1]

health_label = (
    "READY FOR CONTROLLED ANALYSIS"
    if schema_valid and duplicates == 0
    else "REVIEW REQUIRED"
)

def panel(title, label, body, bg="#ffffff", border="#c9ada7",
          text="#22223b", accent="#4a4e69"):

    return HTML(f"""
    <div style="
        font-family:Arial;
        background:{bg};
        border-left:5px solid {border};
        border-radius:9px;
        padding:22px 25px;
    ">
        <div style="
            color:{accent};
            font-size:10px;
            font-weight:700;
            letter-spacing:1.3px;">
            {label}
        </div>

        <div style="
            color:{text};
            font-size:19px;
            font-weight:700;
            margin-top:5px;">
            {title}
        </div>

        <div style="
            color:{text};
            font-size:14px;
            line-height:1.8;
            margin-top:12px;">
            {body}
        </div>
    </div>
    """)

def kpi(label, value, note):

    return f"""
    <div style="
        background:#ffffff;
        padding:15px 17px;
        border-radius:9px;
        border-bottom:3px solid #4a4e69;
    ">
        <div style="font-size:9px;color:#4a4e69;font-weight:700;">
            {label}
        </div>

        <div style="
            font-size:24px;
            color:#22223b;
            font-weight:700;
            margin-top:5px;">
            {value}
        </div>

        <div style="font-size:9px;color:#4a4e69;">
            {note}
        </div>
    </div>
    """

kpis = "".join([
    kpi("TOTAL RECORDS", f"{n:,}", "Job postings"),
    kpi("COMPLETENESS", f"{completeness:.2f}%", "Overall dataset"),
    kpi("COUNTRIES", f"{countries:,}", f"{country_coverage:.2f}% coverage"),
    kpi("DUPLICATES", f"{duplicates:,}", "Exact rows"),
    kpi("DATE SPAN", f"{date_span}", "Calendar days")
])

display(HTML(f"""
<div style="font-family:Arial;margin:18px 0 12px 0;">

    <div style="
        background:linear-gradient(135deg,#22223b,#4a4e69);
        padding:27px 30px;
        border-radius:15px 15px 0 0;
        display:flex;
        justify-content:space-between;
        gap:20px;
        flex-wrap:wrap;
    ">

        <div>
            <div style="
                color:#c9ada7;
                font-size:10px;
                font-weight:700;
                letter-spacing:1.6px;">
                03 • DATA QUALITY INTELLIGENCE
            </div>

            <div style="
                color:#ffffff;
                font-size:28px;
                font-weight:700;
                margin-top:6px;">
                Dataset Health Dashboard
            </div>

            <div style="
                color:#f2e9e4;
                font-size:12px;
                margin-top:8px;">
                Structural Integrity • Completeness • Analytical Readiness
            </div>
        </div>

        <div style="text-align:right;">
            <div style="
                display:inline-block;
                background:#f2e9e4;
                color:#22223b;
                padding:6px 11px;
                border-radius:20px;
                font-size:9px;
                font-weight:700;">
                {health_label}
            </div>

            <div style="
                color:#c9ada7;
                font-size:9px;
                font-weight:700;
                margin-top:11px;">
                LAST REFRESHED
            </div>

            <div style="color:#ffffff;font-size:11px;margin-top:3px;">
                {refresh_time}
            </div>

            <div style="color:#f2e9e4;font-size:10px;margin-top:4px;">
                Source data through {data_through}
            </div>
        </div>

    </div>

    <div style="
        display:grid;
        grid-template-columns:repeat(auto-fit,minmax(145px,1fr));
        gap:9px;
        background:#f6f3f2;
        padding:13px;
        border:1px solid #ebe4e1;
        border-top:none;
        border-radius:0 0 15px 15px;">
        {kpis}
    </div>

</div>
"""))

display(HTML("""
<div style="font-family:Arial;margin:17px 0 7px;">
    <div style="
        color:#4a4e69;
        font-size:9px;
        font-weight:700;
        letter-spacing:1.3px;">
        DASHBOARD VIEW
    </div>
    <div style="color:#22223b;font-size:12px;margin-top:4px;">
        Select one view to explore the dataset health assessment.
    </div>
</div>
"""))

view_selector = widgets.ToggleButtons(
    options=[
        ("TABLE", "table"),
        ("CHART", "chart"),
        ("INSIGHTS", "insights"),
        ("INTERPRETATION", "interpretation"),
        ("RECOMMENDATIONS", "recommendations")
    ],
    value="table",
    layout=widgets.Layout(width="100%"),
    style={
        "button_width": "19%",
        "font_weight": "700"
    }
)

dashboard_output = widgets.Output(
    layout=widgets.Layout(
        width="100%",
        border="1px solid #e8e1de",
        padding="18px",
        margin="8px 0 0 0"
    )
)

display(view_selector, dashboard_output)

def update_dashboard(change=None):

    selected = view_selector.value

    with dashboard_output:

        clear_output(wait=True)

        # TABLE
        if selected == "table":

            table_html = quality_df.to_html(
                index=False,
                border=0,
                classes="quality-table"
            )

            display(HTML(f"""
            <style>
                .quality-table {{
                    width:100%;
                    border-collapse:collapse;
                    font-family:Arial;
                    font-size:12px;
                }}

                .quality-table th {{
                    background:#22223b;
                    color:white;
                    text-align:left;
                    padding:11px 13px;
                }}

                .quality-table td {{
                    color:#22223b;
                    padding:10px 13px;
                    border-bottom:1px solid #eee8e5;
                }}

                .quality-table tr:nth-child(even) td {{
                    background:#f7f4f3;
                }}

                .quality-table tr:hover td {{
                    background:#f2e9e4;
                }}
            </style>

            <div style="
                font-family:Arial;
                color:#4a4e69;
                font-size:10px;
                font-weight:700;">
                QUALITY CONTROL
            </div>

            <h3 style="
                font-family:Arial;
                color:#22223b;
                margin:5px 0 14px;">
                Dataset Validation Summary
            </h3>

            {table_html}
            """))

        elif selected == "chart":

            display(panel(
                "Field-Level Missingness",
                "COMPLETENESS PROFILE",
                "Percentage of missing observations across each source field."
            ))

            fig = px.bar(
                missing_df,
                x="Missing %",
                y="Column",
                orientation="h",
                text="Missing %"
            )

            fig.update_traces(
                marker_color="#4a4e69",
                texttemplate="%{x:.1f}%",
                textposition="outside",
                hovertemplate="<b>%{y}</b><br>Missing: %{x:.2f}%<extra></extra>"
            )

            fig.update_layout(
                template="plotly_white",
                height=430,
                margin=dict(l=25, r=80, t=30, b=40),
                font=dict(family="Arial", color="#22223b"),
                xaxis_title="Missing values (%)",
                yaxis_title="",
                showlegend=False
            )

            display(fig)

        elif selected == "insights":

            body = f"""
            • <b>{n:,}</b> job postings across <b>{p}</b> variables.<br><br>
            • Overall completeness: <b>{completeness:.2f}%</b>.<br><br>
            • Highest missingness: <b>{worst["Column"]}</b>
              ({worst["Missing %"]:.2f}%).<br><br>
            • Country coverage: <b>{country_coverage:.2f}%</b>.<br><br>
            • Exact duplicates: <b>{duplicates:,}</b>;
              duplicate links: <b>{duplicate_links:,}</b>.<br><br>
            • Valid publication dates: <b>{date_validity:.2f}%</b>.
            """

            display(panel(
                "Dataset Quality Summary",
                "KEY INSIGHTS",
                body,
                bg="#f2e9e4",
                border="#4a4e69"
            ))

        elif selected == "interpretation":

            body = f"""
            Publication dates are valid for
            <b>{date_validity:.2f}%</b> of records across
            <b>{date_span} calendar days</b>.<br><br>

            Missing compensation fields reflect two different payment
            structures: <b>hourly-rate jobs</b> and
            <b>fixed-price projects</b>.<br><br>

            Hourly analysis should use records with both
            <b>hourly_low</b> and <b>hourly_high</b>.
            Fixed-price analysis should use <b>budget</b> separately.<br><br>

            These compensation groups should remain separate during
            cleaning, analysis and modelling.
            """

            display(panel(
                "Analytical Interpretation",
                "INTERPRETATION",
                body
            ))

        elif selected == "recommendations":

            body = """
            <b>01.</b> Preserve <b>df_raw</b> as the immutable source.<br><br>
            <b>02.</b> Clean and engineer features on a separate dataframe.<br><br>
            <b>03.</b> Quarantine records missing critical identifiers.<br><br>
            <b>04.</b> Do not globally impute compensation fields.<br><br>
            <b>05.</b> Keep hourly and fixed-price compensation separate.<br><br>
            <b>06.</b> Use hourly midpoint only when both bounds exist.<br><br>
            <b>07.</b> Handle missing countries explicitly.<br><br>
            <b>08.</b> Reuse dashboard exports in downstream analysis and Streamlit.
            """

            display(panel(
                "Data-Governance Actions",
                "RECOMMENDATIONS",
                body,
                bg="#22223b",
                border="#c9ada7",
                text="#f2e9e4",
                accent="#c9ada7"
            ))

view_selector.observe(update_dashboard, names="value")
update_dashboard()

quality_df.to_csv(
    EXPORT_DIR / "data_quality_checks.csv",
    index=False
)

missing_df.to_csv(
    EXPORT_DIR / "data_quality_missingness.csv",
    index=False
)

metrics = {
    "last_refreshed_ist": refresh_time,
    "records": int(n),
    "columns": int(p),
    "schema_valid": bool(schema_valid),
    "completeness_pct": round(float(completeness), 2),
    "exact_duplicates": duplicates,
    "duplicate_links": duplicate_links,
    "title_completeness_pct": round(float(title_complete), 2),
    "link_completeness_pct": round(float(link_complete), 2),
    "country_coverage_pct": round(float(country_coverage), 2),
    "countries": countries,
    "date_validity_pct": round(float(date_validity), 2),
    "date_start": date_start.isoformat() if pd.notna(date_start) else None,
    "date_end": date_end.isoformat() if pd.notna(date_end) else None,
    "date_span_days": int(date_span),
    "hourly_records": hourly_n,
    "fixed_price_records": fixed_n,
    "hourly_complete_records": hourly_complete,
    "fixed_budget_complete_records": fixed_complete,
    "hourly_pay_coverage_pct": round(float(hourly_coverage), 2),
    "fixed_budget_coverage_pct": round(float(fixed_coverage), 2),
    "highest_missingness_column": str(worst["Column"]),
    "highest_missingness_pct": round(float(worst["Missing %"]), 2)
}

with open(
    EXPORT_DIR / "data_quality_metrics.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(metrics, f, indent=4)

if "logger" in globals():
    logger.info(
        "Data Quality Dashboard | %s | Records=%s | "
        "Completeness=%.2f%% | Duplicates=%s",
        refresh_time,
        f"{n:,}",
        completeness,
        duplicates
    )
```

```python
if "df_raw" not in globals():
    raise NameError("Run Data Ingestion first.")
if "coco" not in globals():
    raise NameError("Add 'import country_converter as coco' to Master Imports.")

df_clean = df_raw.copy(deep=True)
source_n = len(df_clean)

for c in ["title", "link", "country"]:
    df_clean[c] = df_clean[c].astype("string").str.strip().replace("", pd.NA)

df_clean["published_date"] = pd.to_datetime(
    df_clean["published_date"], errors="coerce", utc=True
)

for c in ["hourly_low", "hourly_high", "budget"]:
    df_clean[c] = pd.to_numeric(df_clean[c], errors="coerce")

pay_flag = (
    df_clean["is_hourly"].astype(str).str.strip().str.lower()
    .map({
        "true": True, "1": True, "1.0": True, "yes": True,
        "false": False, "0": False, "0.0": False, "no": False
    })
)

df_clean["pay_type"] = (
    pay_flag.map({True: "Hourly", False: "Fixed Price"})
    .fillna("Unknown")
)

hourly_evidence = df_clean[["hourly_low", "hourly_high"]].notna().any(axis=1)
fixed_evidence = df_clean["budget"].notna()

df_clean.loc[
    df_clean["pay_type"].eq("Unknown") & hourly_evidence & ~fixed_evidence,
    "pay_type"
] = "Hourly"

df_clean.loc[
    df_clean["pay_type"].eq("Unknown") & fixed_evidence & ~hourly_evidence,
    "pay_type"
] = "Fixed Price"

duplicate_n = int(df_clean.duplicated().sum())
df_clean = df_clean.drop_duplicates().copy()

critical = (
    df_clean["title"].isna()
    | df_clean["link"].isna()
    | df_clean["published_date"].isna()
)

quarantine_df = df_clean.loc[critical].copy()
df_clean = df_clean.loc[~critical].copy()
critical_n = len(quarantine_df)

df_clean["title_normalized"] = (
    df_clean["title"].str.lower()
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

df_clean["country_original"] = (
    df_clean["country"].astype("string").str.strip().replace("", pd.NA)
)
df_clean["country_was_missing"] = df_clean["country_original"].isna()
country_file = Path(coco.__file__).resolve().parent / "country_data.tsv"
country_ref = pd.read_csv(
    country_file, sep="\t", dtype=str, keep_default_na=False
)

def norm_country(x):
    if pd.isna(x):
        return pd.NA
    x = str(x).lower().strip()
    for old, new in [
        (".", ""), (",", ""), ("'", ""),
        ("_", " "), ("-", " ")
    ]:
        x = x.replace(old, new)
    return " ".join(x.split())

country_name_lookup = {}

for _, row in country_ref.iterrows():
    standard = row["name_short"]
    if not standard:
        continue
    for col in ["name_short", "name_official", "ISO2", "ISO3"]:
        if row[col]:
            country_name_lookup[norm_country(row[col])] = standard

df_clean["country_standardized"] = (
    df_clean["country_original"]
    .map(lambda x: country_name_lookup.get(norm_country(x), pd.NA))
)

link_country_lookup = {}

for _, row in country_ref.iterrows():
    standard = row["name_short"]
    if not standard:
        continue
    for col in ["name_short", "name_official", "ISO3"]:
        value = row[col]
        if value:
            token = (
                str(value).lower().strip()
                .replace("%20", "-")
                .replace(" ", "-")
                .replace("_", "-")
                .replace(".", "")
                .replace(",", "")
                .replace("'", "")
            )

            if len(token) >= 3:
                link_country_lookup[token] = standard

link_country_lookup = dict(
    sorted(
        link_country_lookup.items(),
        key=lambda x: len(x[0]),
        reverse=True
    )
)

def country_from_link(link):
    if pd.isna(link):
        return pd.NA
    text = str(link).lower()
    for char in ["/", "?", "&", "=", "+", "_", ".", "%20"]:
        text = text.replace(char, "-")
    while "--" in text:
        text = text.replace("--", "-")
    text = f"-{text.strip('-')}-"
    for token, country_name in link_country_lookup.items():
        if f"-{token}-" in text:
            return country_name
    return pd.NA

needs_recovery = df_clean["country_standardized"].isna()
df_clean["country_from_link"] = pd.NA

df_clean.loc[needs_recovery, "country_from_link"] = (
    df_clean.loc[needs_recovery, "link"].apply(country_from_link)
)

df_clean["country_clean"] = (
    df_clean["country_standardized"]
    .fillna(df_clean["country_from_link"])
)

country_mode = (
    df_clean["country_standardized"].dropna().mode().iloc[0]
)

df_clean["country_clean"] = (
    df_clean["country_clean"].fillna(country_mode)
)

df_clean["country_imputation_method"] = "Original / Standardized"

df_clean.loc[
    df_clean["country_standardized"].isna()
    & df_clean["country_from_link"].notna(),
    "country_imputation_method"
] = "Link Derived"

df_clean.loc[
    df_clean["country_original"].isna()
    & df_clean["country_from_link"].isna(),
    "country_imputation_method"
] = "Mode Imputed - Missing"

df_clean.loc[
    df_clean["country_original"].notna()
    & df_clean["country_standardized"].isna()
    & df_clean["country_from_link"].isna(),
    "country_imputation_method"
] = "Mode Imputed - Unmapped"

df_clean["country_observed_flag"] = (
    df_clean["country_imputation_method"].eq("Original / Standardized")
)

df_clean["country_link_recovered_flag"] = (
    df_clean["country_imputation_method"].eq("Link Derived")
)

df_clean["country_mode_imputed_flag"] = (
    df_clean["country_imputation_method"].str.startswith("Mode Imputed")
)

df_clean["country"] = df_clean["country_clean"].astype("string")

if df_clean["country_clean"].isna().any():
    raise ValueError("country_clean still contains missing values.")

for src, dst in [
    ("hourly_low", "hourly_low_original"),
    ("hourly_high", "hourly_high_original"),
    ("budget", "budget_original")
]:
    df_clean[dst] = df_clean[src]

hourly_rows = df_clean["pay_type"].eq("Hourly")
fixed_rows = df_clean["pay_type"].eq("Fixed Price")

valid_hourly = (
    hourly_rows
    & df_clean["hourly_low"].notna()
    & df_clean["hourly_high"].notna()
    & df_clean["hourly_low"].ge(0)
    & df_clean["hourly_high"].ge(df_clean["hourly_low"])
)

valid_fixed = (
    fixed_rows
    & df_clean["budget"].notna()
    & df_clean["budget"].ge(0)
)

df_clean["hourly_midpoint"] = (
    (df_clean["hourly_low"] + df_clean["hourly_high"]) / 2
).where(valid_hourly)

df_clean["fixed_budget"] = df_clean["budget"].where(valid_fixed)

hourly_midpoint_median = df_clean.loc[
    valid_hourly, "hourly_midpoint"
].median()

fixed_budget_median = df_clean.loc[
    valid_fixed, "fixed_budget"
].median()

df_clean["hourly_midpoint_imputed"] = df_clean["hourly_midpoint"]
df_clean.loc[
    hourly_rows & df_clean["hourly_midpoint_imputed"].isna(),
    "hourly_midpoint_imputed"
] = hourly_midpoint_median

df_clean["fixed_budget_imputed"] = df_clean["fixed_budget"]
df_clean.loc[
    fixed_rows & df_clean["fixed_budget_imputed"].isna(),
    "fixed_budget_imputed"
] = fixed_budget_median

df_clean["hourly_comp_imputed_flag"] = (
    hourly_rows
    & df_clean["hourly_midpoint"].isna()
    & df_clean["hourly_midpoint_imputed"].notna()
)

df_clean["fixed_comp_imputed_flag"] = (
    fixed_rows
    & df_clean["fixed_budget"].isna()
    & df_clean["fixed_budget_imputed"].notna()
)

dt = df_clean["published_date"].dt.tz_convert(None)
df_clean["published_day"] = df_clean["published_date"].dt.floor("D")
df_clean["published_week"] = dt.dt.to_period("W").astype(str)
df_clean["published_month"] = dt.dt.to_period("M").astype(str)
final_n = len(df_clean)
country_link_n = int(df_clean["country_link_recovered_flag"].sum())
country_mode_n = int(df_clean["country_mode_imputed_flag"].sum())
hourly_imputed_n = int(df_clean["hourly_comp_imputed_flag"].sum())
fixed_imputed_n = int(df_clean["fixed_comp_imputed_flag"].sum())

cleaning_summary = pd.DataFrame({
    "Stage": [
        "Source Records",
        "Duplicates Removed",
        "Critical Records Quarantined",
        "Final Clean Records"
    ],
    "Records": [
        source_n,
        duplicate_n,
        critical_n,
        final_n
    ]
})

missing_strategy = pd.DataFrame({
    "Variable": [
        "title", "link", "published_date", "country",
        "hourly compensation", "fixed budget", "pay_type"
    ],
    "Treatment": [
        "Quarantine", "Quarantine", "Quarantine",
        "Standardize → Link → Mode",
        "Original + Median helper",
        "Original + Median helper",
        "Infer only when structure is clear"
    ],
    "Reason": [
        "Critical analytical field",
        "Critical identifier",
        "Required for temporal analysis",
        "Recover categorical information before fallback",
        "Preserve original pay evidence",
        "Preserve original project value",
        "Avoid arbitrary classification"
    ]
})

country_imputation_summary = (
    df_clean["country_imputation_method"]
    .value_counts()
    .rename_axis("Method")
    .reset_index(name="Records")
)

country_imputation_summary["Share %"] = (
    country_imputation_summary["Records"]
    / len(df_clean) * 100
).round(2)

compensation_summary = pd.DataFrame({
    "Measure": [
        "Original valid hourly midpoint",
        "Median-imputed hourly helper",
        "Original valid fixed budget",
        "Median-imputed fixed helper"
    ],
    "Records": [
        int(df_clean["hourly_midpoint"].notna().sum()),
        hourly_imputed_n,
        int(df_clean["fixed_budget"].notna().sum()),
        fixed_imputed_n
    ]
})

refresh_time = pd.Timestamp.now(
    tz="Asia/Kolkata"
).strftime("%d %b %Y • %I:%M %p")

def c51_card(label, value, note):
    return (
        f'<div class="c"><small>{label}</small>'
        f'<b>{value}</b><span>{note}</span></div>'
    )

display(HTML("""
<style>
.d51{border:1px solid #ded8d5;border-radius:15px;overflow:hidden;background:#fff;
box-shadow:0 6px 20px rgba(34,34,59,.08)}
.h51{background:linear-gradient(120deg,#22223b,#30324d,#4a4e69);
padding:23px 26px;display:flex;justify-content:space-between;font-family:Arial}
.h51 .k{color:#c9ada7;font-size:8px;font-weight:700;letter-spacing:1.5px}
.h51 h2{color:#fff;font-size:23px;margin:6px 0}
.h51 p,.h51 .m{color:#f2e9e4;font-size:10px;line-height:1.6}
.cards{display:grid;grid-template-columns:repeat(6,1fr);border-bottom:1px solid #e8e2df}
.c{padding:14px;border-right:1px solid #ebe6e4;font-family:Arial}
.c small{display:block;color:#9a8c98;font-size:7px;font-weight:700}
.c b{display:block;color:#22223b;font-size:19px;margin-top:5px}
.c span{display:block;color:#4a4e69;font-size:8px;margin-top:3px}
.btn51{margin:13px 18px 8px!important}.focus51{margin:0 18px 13px!important}
.out51{border-top:1px solid #eee9e7!important}
.t51{width:100%;border-collapse:collapse;font:11px Arial}
.t51 th{background:#22223b;color:#f2e9e4;padding:10px;text-align:left}
.t51 td{padding:10px;border-bottom:1px solid #eee9e7;color:#22223b}
.txt51{font:12px/1.8 Arial;color:#22223b}
</style>
"""))
clean_header = widgets.HTML(f"""
<div class="h51">
<div>
<div class="k">05.1 • DATA CLEANING & MISSING-VALUE TREATMENT</div>
<h2>Controlled Cleaning & Imputation Framework</h2>
<p>Critical quarantine • Country standardization • Link recovery •
Mode fallback • Median compensation helpers • Audit controls</p>
</div>
<div class="m">LAST REFRESHED<br><b>{refresh_time} IST</b></div>
</div>
<div class="cards">
{c51_card("SOURCE", f"{source_n:,}", "Input records")}
{c51_card("CLEAN", f"{final_n:,}", "Retained")}
{c51_card("QUARANTINED", f"{critical_n:,}", "Critical missing")}
{c51_card("COUNTRY COMPLETE", f"{df_clean['country_clean'].notna().mean()*100:.1f}%", "After treatment")}
{c51_card("LINK RECOVERED", f"{country_link_n:,}", "Countries")}
{c51_card("MODE FALLBACK", f"{country_mode_n:,}", f"Mode: {country_mode}")}
</div>
""")

clean_view = widgets.ToggleButtons(
    options=[
        ("TABLE", "table"),
        ("CHART", "chart"),
        ("INSIGHTS", "insights"),
        ("INTERPRETATION", "interpretation"),
        ("RECOMMENDATIONS", "recommendations")
    ],
    value="table",
    layout=widgets.Layout(width="calc(100% - 36px)"),
    style={"button_width": "19%", "font_weight": "700"}
)
clean_view.add_class("btn51")

clean_focus = widgets.Dropdown(
    options=[
        ("Cleaning Flow", "flow"),
        ("Missing-Value Strategy", "missing"),
        ("Country Treatment", "country"),
        ("Compensation Treatment", "comp")
    ],
    value="missing",
    description="Focus:",
    layout=widgets.Layout(width="350px")
)
clean_focus.add_class("focus51")

clean_output = widgets.Output(
    layout=widgets.Layout(padding="20px 23px 23px")
)
clean_output.add_class("out51")

def update_clean_dashboard(change=None):
    with clean_output:
        clear_output(wait=True)
        view, focus = clean_view.value, clean_focus.value
        if view == "table":
            tables = {
                "flow": cleaning_summary,
                "missing": missing_strategy,
                "country": country_imputation_summary,
                "comp": compensation_summary
            }

            display(HTML(
                tables[focus].to_html(
                    index=False,
                    border=0,
                    classes="t51"
                )
            ))

        elif view == "chart":
            fig = go.Figure(go.Waterfall(
                measure=["absolute", "relative", "relative", "total"],
                x=["Source", "Duplicates", "Quarantined", "Final Clean"],
                y=[source_n, -duplicate_n, -critical_n, 0],
                text=[
                    f"{source_n:,}",
                    f"-{duplicate_n:,}",
                    f"-{critical_n:,}",
                    f"{final_n:,}"
                ],
                textposition="outside",
                connector={"line": {"color": "#9a8c98"}},
                increasing={"marker": {"color": "#c9ada7"}},
                decreasing={"marker": {"color": "#9a8c98"}},
                totals={"marker": {"color": "#4a4e69"}}
            ))

            fig.update_layout(
                height=440,
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",
                showlegend=False,
                font={"family": "Arial", "color": "#22223b"},
                margin={"l": 30, "r": 30, "t": 20, "b": 35}
            )
            display(fig)

        elif view == "insights":
            display(HTML(f"""
            <div class="txt51">
            <b>Country completeness: {df_clean['country_clean'].notna().mean()*100:.1f}%.</b><br><br>
            <b>{country_link_n:,}</b> countries recovered from job links.<br><br>
            <b>{country_mode_n:,}</b> unresolved countries filled using mode:
            <b>{country_mode}</b>.<br><br>
            <b>{hourly_imputed_n:,}</b> hourly and
            <b>{fixed_imputed_n:,}</b> fixed-price records received
            median helper values.<br><br>
            Original compensation remains unchanged.
            </div>
            """))

        elif view == "interpretation":
            display(HTML("""
            <div class="txt51">
            <b>Critical fields:</b> missing title, link or publication date → quarantine.<br><br>
            <b>Country:</b> standardized source → link evidence → mode fallback.<br><br>
            <b>Compensation:</b> median helpers are used because pay distributions may be skewed.<br><br>
            <b>Structural missingness:</b> hourly and fixed-price fields remain separated.
            </div>
            """))

        else:
            display(HTML("""
            <div class="txt51">
            <b>01.</b> Use original compensation for reported pay statistics.<br>
            <b>02.</b> Use imputed helpers only where complete model inputs are required.<br>
            <b>03.</b> Preserve country_imputation_method for auditability.<br>
            <b>04.</b> Flag mode-imputed geography in sensitive country comparisons.<br>
            <b>05.</b> Preserve quarantine_df as cleaning evidence.<br>
            <b>06.</b> Revalidate everything in Section 05.2.
            </div>
            """))

clean_view.observe(update_clean_dashboard, names="value")
clean_focus.observe(update_clean_dashboard, names="value")
update_clean_dashboard()

dashboard_051 = widgets.VBox([
    clean_header,
    clean_view,
    clean_focus,
    clean_output
])
dashboard_051.add_class("d51")
display(dashboard_051)

exports = {
    "cleaned_job_market.csv": df_clean,
    "quarantined_records.csv": quarantine_df,
    "cleaning_summary.csv": cleaning_summary,
    "missing_value_strategy.csv": missing_strategy,
    "country_imputation_summary.csv": country_imputation_summary,
    "compensation_imputation_summary.csv": compensation_summary
}

for name, data in exports.items():
    data.to_csv(EXPORT_DIR / name, index=False)

if "logger" in globals():
    logger.info(
        "DATA_CLEANING | source=%s | clean=%s | duplicates=%s | "
        "quarantined=%s | country_complete=%.2f%% | "
        "link_recovered=%s | mode_fallback=%s | "
        "hourly_imputed=%s | fixed_imputed=%s",
        f"{source_n:,}",
        f"{final_n:,}",
        duplicate_n,
        critical_n,
        df_clean["country_clean"].notna().mean() * 100,
        country_link_n,
        country_mode_n,
        hourly_imputed_n,
        fixed_imputed_n
    )
```

```python
if "df_clean" not in globals():
    raise NameError("Run Section 05.1 first.")

df_final = df_clean.copy(deep=True)

def missing_check(label, column, mask=None):
    mask = pd.Series(True, index=df_final.index) if mask is None else mask
    n = int(mask.sum())
    m = int(df_final.loc[mask, column].isna().sum())
    return [label, n, m, round(m / n * 100, 2) if n else 0]

hourly_rows = df_final["pay_type"].eq("Hourly")
fixed_rows = df_final["pay_type"].eq("Fixed Price")

missing_value_audit = pd.DataFrame([
    missing_check("Title", "title"),
    missing_check("Link", "link"),
    missing_check("Published Date", "published_date"),
    missing_check("Country Clean", "country_clean"),
    missing_check("Hourly Midpoint - Original", "hourly_midpoint", hourly_rows),
    missing_check("Hourly Midpoint - Helper", "hourly_midpoint_imputed", hourly_rows),
    missing_check("Fixed Budget - Original", "fixed_budget", fixed_rows),
    missing_check("Fixed Budget - Helper", "fixed_budget_imputed", fixed_rows)
], columns=["Field", "Eligible Records", "Missing", "Missing %"])

exact_duplicates = int(df_final.duplicated().sum())
duplicate_links = int(df_final["link"].duplicated().sum())

duplicate_integrity = pd.DataFrame({
    "Check": ["Exact Duplicate Rows", "Duplicate Job Links"],
    "Records": [exact_duplicates, duplicate_links],
    "Expected": [0, 0]
})

hourly_valid_original = (
    hourly_rows
    & df_final["hourly_midpoint"].notna()
)

fixed_valid_original = (
    fixed_rows
    & df_final["fixed_budget"].notna()
)

pay_validity = pd.DataFrame({
    "Measure": [
        "Hourly Jobs",
        "Original Valid Hourly Pay",
        "Hourly Helper Complete",
        "Fixed-Price Jobs",
        "Original Valid Fixed Budget",
        "Fixed Helper Complete"
    ],
    "Records": [
        int(hourly_rows.sum()),
        int(hourly_valid_original.sum()),
        int(df_final.loc[hourly_rows, "hourly_midpoint_imputed"].notna().sum()),
        int(fixed_rows.sum()),
        int(fixed_valid_original.sum()),
        int(df_final.loc[fixed_rows, "fixed_budget_imputed"].notna().sum())
    ]
})

def outlier_control(series):
    valid = series.dropna()
    if len(valid) < 4:
        return (
            pd.Series(False, index=series.index),
            series.copy(),
            np.nan, np.nan, np.nan, np.nan
        )

    q1, q3 = valid.quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    p01, p99 = valid.quantile([0.01, 0.99])

    flag = series.notna() & (
        (series < lower) | (series > upper)
    )

    return flag, series.clip(p01, p99), lower, upper, p01, p99

(
    df_final["hourly_outlier_flag"],
    df_final["hourly_midpoint_capped"],
    h_lower, h_upper, h_p01, h_p99
) = outlier_control(df_final["hourly_midpoint"])

(
    df_final["fixed_outlier_flag"],
    df_final["fixed_budget_capped"],
    f_lower, f_upper, f_p01, f_p99
) = outlier_control(df_final["fixed_budget"])

outlier_thresholds = pd.DataFrame({
    "Measure": ["Hourly Midpoint", "Fixed Budget"],
    "IQR Lower": [h_lower, f_lower],
    "IQR Upper": [h_upper, f_upper],
    "P01 Cap": [h_p01, f_p01],
    "P99 Cap": [h_p99, f_p99],
    "Outliers": [
        int(df_final["hourly_outlier_flag"].sum()),
        int(df_final["fixed_outlier_flag"].sum())
    ]
})

daily_quality = (
    df_final.groupby("published_day")
    .size()
    .rename("postings")
    .reset_index()
    .sort_values("published_day")
)

dense_start = pd.Timestamp("2024-02-13", tz="UTC")
latest_ts = df_final["published_date"].max()
latest_day = latest_ts.floor("D")

partial_final_day = (
    latest_ts.hour < 23
    or latest_ts.minute < 59
)

trend_end = (
    latest_day - pd.Timedelta(days=1)
    if partial_final_day else latest_day
)

dense_daily = daily_quality[
    daily_quality["published_day"].between(
        dense_start, trend_end
    )
].copy()

dense_median = dense_daily["postings"].median()

daily_quality["dense_window_flag"] = (
    daily_quality["published_day"].ge(dense_start)
)

daily_quality["partial_final_day_flag"] = (
    daily_quality["published_day"].eq(latest_day)
    & partial_final_day
)

daily_quality["low_volume_flag"] = (
    daily_quality["dense_window_flag"]
    & ~daily_quality["partial_final_day_flag"]
    & daily_quality["postings"].lt(dense_median * 0.25)
)

dense_share = (
    df_final["published_date"].ge(dense_start).mean() * 100
)

low_volume_days = int(
    daily_quality["low_volume_flag"].sum()
)

temporal_summary = pd.DataFrame({
    "Measure": [
        "Dataset Start",
        "Dataset End",
        "Dense Window Start",
        "Dense-Window Coverage",
        "Partial Final Day",
        "Low-Volume Dense Days"
    ],
    "Result": [
        df_final["published_date"].min().strftime("%d %b %Y"),
        latest_ts.strftime("%d %b %Y %H:%M UTC"),
        dense_start.strftime("%d %b %Y"),
        f"{dense_share:.2f}%",
        "Yes" if partial_final_day else "No",
        low_volume_days
    ]
})

critical_missing = int(
    df_final[
        ["title", "link", "published_date"]
    ].isna().any(axis=1).sum()
)

country_missing = int(
    df_final["country_clean"].isna().sum()
)

hourly_helper_missing = int(
    df_final.loc[
        hourly_rows,
        "hourly_midpoint_imputed"
    ].isna().sum()
)

fixed_helper_missing = int(
    df_final.loc[
        fixed_rows,
        "fixed_budget_imputed"
    ].isna().sum()
)

checks = [
    ("Critical fields complete", critical_missing == 0,
     f"{critical_missing:,} unresolved"),

    ("Country complete", country_missing == 0,
     f"{country_missing:,} unresolved"),

    ("Exact duplicates removed", exact_duplicates == 0,
     f"{exact_duplicates:,} duplicates"),

    ("Job links unique", duplicate_links == 0,
     f"{duplicate_links:,} duplicates"),

    ("Hourly helper complete", hourly_helper_missing == 0,
     f"{hourly_helper_missing:,} missing"),

    ("Fixed helper complete", fixed_helper_missing == 0,
     f"{fixed_helper_missing:,} missing"),

    ("Dense-window coverage", dense_share >= 95,
     f"{dense_share:.2f}%"),

    ("Outlier controls created", True,
     "Flags + capped variants available")
]

final_validation_checks = pd.DataFrame(
    checks,
    columns=["Validation Check", "Passed", "Details"]
)

final_validation_checks["Status"] = (
    final_validation_checks["Passed"]
    .map({True: "PASS", False: "REVIEW"})
)

passed_checks = int(
    final_validation_checks["Passed"].sum()
)

total_checks = len(final_validation_checks)

readiness_pct = (
    passed_checks / total_checks * 100
)

refresh_time = pd.Timestamp.now(
    tz="Asia/Kolkata"
).strftime("%d %b %Y • %I:%M %p")

def c52_card(label, value, note):
    return (
        f'<div class="c52"><small>{label}</small>'
        f'<b>{value}</b><span>{note}</span></div>'
    )

display(HTML("""
<style>
.d52{border:1px solid #ded8d5;border-radius:15px;overflow:hidden;background:#fff;
box-shadow:0 6px 20px rgba(34,34,59,.08)}
.h52{background:linear-gradient(120deg,#22223b,#30324d,#4a4e69);
padding:23px 26px;display:flex;justify-content:space-between;font-family:Arial}
.h52 .k{color:#c9ada7;font-size:8px;font-weight:700;letter-spacing:1.5px}
.h52 h2{color:#fff;font-size:23px;margin:6px 0}
.h52 p,.h52 .m{color:#f2e9e4;font-size:10px;line-height:1.6}
.cards52{display:grid;grid-template-columns:repeat(6,1fr);border-bottom:1px solid #e8e2df}
.c52{padding:14px;border-right:1px solid #ebe6e4;font-family:Arial}
.c52 small{display:block;color:#9a8c98;font-size:7px;font-weight:700}
.c52 b{display:block;color:#22223b;font-size:19px;margin-top:5px}
.c52 span{display:block;color:#4a4e69;font-size:8px;margin-top:3px}
.btn52{margin:13px 18px 8px!important}.focus52{margin:0 18px 13px!important}
.out52{border-top:1px solid #eee9e7!important}
.t52{width:100%;border-collapse:collapse;font:11px Arial}
.t52 th{background:#22223b;color:#f2e9e4;padding:10px;text-align:left}
.t52 td{padding:10px;border-bottom:1px solid #eee9e7;color:#22223b}
.txt52{font:12px/1.8 Arial;color:#22223b}
</style>
"""))
control_header = widgets.HTML(f"""
<div class="h52">
<div>
<div class="k">05.2 • DATA CONTROL CENTRE</div>
<h2>Validation, Outlier & Readiness Control</h2>
<p>Completeness • Duplicate integrity • Pay validity •
Outlier control • Temporal quality • Final readiness</p>
</div>
<div class="m">LAST REFRESHED<br><b>{refresh_time} IST</b></div>
</div>
<div class="cards52">
{c52_card("FINAL RECORDS", f"{len(df_final):,}", "Validated rows")}
{c52_card("COUNTRY COMPLETE", f"{df_final['country_clean'].notna().mean()*100:.1f}%", "Operational field")}
{c52_card("DUPLICATES", f"{exact_duplicates:,}", "Exact rows")}
{c52_card("HOURLY OUTLIERS", f"{df_final['hourly_outlier_flag'].sum():,}", "Flagged only")}
{c52_card("DENSE WINDOW", f"{dense_share:.1f}%", "Trend coverage")}
{c52_card("CONTROLS PASSED", f"{passed_checks}/{total_checks}", f"{readiness_pct:.0f}% readiness")}
</div>
""")

control_view = widgets.ToggleButtons(
    options=[
        ("TABLE", "table"),
        ("CHART", "chart"),
        ("INSIGHTS", "insights"),
        ("INTERPRETATION", "interpretation"),
        ("RECOMMENDATIONS", "recommendations")
    ],
    value="table",
    layout=widgets.Layout(width="calc(100% - 36px)"),
    style={"button_width": "19%", "font_weight": "700"}
)
control_view.add_class("btn52")

control_focus = widgets.Dropdown(
    options=[
        ("Missing Values", "missing"),
        ("Duplicate Integrity", "duplicates"),
        ("Hourly Outliers", "hourly"),
        ("Fixed-Price Outliers", "fixed"),
        ("Pay Validity", "pay"),
        ("Temporal Quality", "temporal"),
        ("Final Readiness", "readiness")
    ],
    value="readiness",
    description="Focus:",
    layout=widgets.Layout(width="350px")
)
control_focus.add_class("focus52")

control_output = widgets.Output(
    layout=widgets.Layout(padding="20px 23px 23px")
)
control_output.add_class("out52")

def update_control_dashboard(change=None):
    with control_output:
        clear_output(wait=True)
        view = control_view.value
        focus = control_focus.value

        if view == "table":
            tables = {
                "missing": missing_value_audit,
                "duplicates": duplicate_integrity,
                "hourly": outlier_thresholds.iloc[[0]],
                "fixed": outlier_thresholds.iloc[[1]],
                "pay": pay_validity,
                "temporal": temporal_summary,
                "readiness": final_validation_checks[
                    ["Validation Check", "Status", "Details"]
                ]
            }

            display(HTML(
                tables[focus].to_html(
                    index=False,
                    border=0,
                    classes="t52"
                )
            ))

        elif view == "chart":

            if focus in ["hourly", "fixed"]:
                col = (
                    "hourly_midpoint"
                    if focus == "hourly"
                    else "fixed_budget"
                )

                label = (
                    "Hourly Midpoint"
                    if focus == "hourly"
                    else "Fixed Budget"
                )

                sample = (
                    df_final[col]
                    .dropna()
                    .sample(
                        min(5000, df_final[col].notna().sum()),
                        random_state=42
                    )
                )

                fig = go.Figure(go.Box(
                    x=sample,
                    name=label,
                    boxpoints="outliers",
                    marker_color="#4a4e69",
                    line_color="#22223b"
                ))

                fig.update_layout(
                    height=420,
                    paper_bgcolor="#fff",
                    plot_bgcolor="#fff",
                    font={"family": "Arial", "color": "#22223b"},
                    margin={"l": 30, "r": 30, "t": 25, "b": 35}
                )

            elif focus == "temporal":
                fig = go.Figure(go.Scatter(
                    x=daily_quality["published_day"],
                    y=daily_quality["postings"],
                    mode="lines",
                    fill="tozeroy",
                    line={"color": "#4a4e69"},
                    fillcolor="rgba(154,140,152,.25)"
                ))

                fig.update_layout(
                    height=420,
                    paper_bgcolor="#fff",
                    plot_bgcolor="#fff",
                    font={"family": "Arial", "color": "#22223b"},
                    xaxis_title="Posting Date",
                    yaxis_title="Daily Job Postings",
                    margin={"l": 45, "r": 25, "t": 25, "b": 45}
                )

            elif focus == "readiness":
                fig = go.Figure(go.Indicator(
                    mode="gauge+number",
                    value=readiness_pct,
                    number={"suffix": "%"},
                    title={"text": "Analytical Readiness"},
                    gauge={
                        "axis": {"range": [0, 100]},
                        "bar": {"color": "#4a4e69"},
                        "bgcolor": "#f2e9e4",
                        "threshold": {
                            "line": {"color": "#22223b", "width": 3},
                            "value": 95
                        }
                    }
                ))

                fig.update_layout(
                    height=360,
                    paper_bgcolor="#fff",
                    font={"family": "Arial", "color": "#22223b"},
                    margin={"l": 60, "r": 60, "t": 60, "b": 30}
                )

            else:
                chart_df = (
                    missing_value_audit
                    .copy()
                )

                fig = go.Figure(go.Bar(
                    x=100 - chart_df["Missing %"],
                    y=chart_df["Field"],
                    orientation="h",
                    marker_color="#9a8c98"
                ))

                fig.update_layout(
                    height=420,
                    paper_bgcolor="#fff",
                    plot_bgcolor="#fff",
                    xaxis_title="Available / Valid %",
                    xaxis_range=[0, 100],
                    font={"family": "Arial", "color": "#22223b"},
                    margin={"l": 170, "r": 25, "t": 25, "b": 40}
                )

            display(fig)


        elif view == "insights":
            display(HTML(f"""
            <div class="txt52">
            <b>{passed_checks}/{total_checks}</b> validation controls currently pass.<br><br>
            Country is <b>{df_final['country_clean'].notna().mean()*100:.1f}% complete</b>
            after the controlled 05.1 treatment.<br><br>
            <b>{int(df_final['hourly_outlier_flag'].sum()):,}</b> hourly and
            <b>{int(df_final['fixed_outlier_flag'].sum()):,}</b> fixed-price
            observations are flagged as statistical outliers rather than deleted.<br><br>
            <b>{dense_share:.2f}%</b> of records fall inside the preferred
            dense trend window beginning 13 February 2024.<br><br>
            The final observation day is
            <b>{"partial" if partial_final_day else "complete"}</b>
            and is controlled separately in time-series analysis.
            </div>
            """))

        elif view == "interpretation":
            display(HTML("""
            <div class="txt52">
            <b>Missingness:</b> structural compensation gaps are evaluated only
            within their relevant pay type rather than across every record.<br><br>
            <b>Outliers:</b> extreme pay observations are preserved and flagged.
            P01/P99 capped alternatives are provided for sensitivity analysis.<br><br>
            <b>Temporal quality:</b> the sparse early period and partial final day
            must not distort trend or forecasting results.<br><br>
            <b>Validation:</b> this section checks cleaning quality; it does not
            create business features. Feature creation begins in 05.3.
            </div>
            """))

        else:
            display(HTML("""
            <div class="txt52">
            <b>01.</b> Use original pay fields for descriptive salary reporting.<br>
            <b>02.</b> Use capped pay fields only for robustness or model sensitivity tests.<br>
            <b>03.</b> Exclude the partial final day from completed-day trend comparisons.<br>
            <b>04.</b> Investigate flagged low-volume days before forecasting.<br>
            <b>05.</b> Retain imputation and outlier flags for auditability.<br>
            <b>06.</b> Pass df_final directly into Section 05.3 Feature Engineering.
            </div>
            """))

control_view.observe(update_control_dashboard, names="value")
control_focus.observe(update_control_dashboard, names="value")
update_control_dashboard()

dashboard_052 = widgets.VBox([
    control_header,
    control_view,
    control_focus,
    control_output
])
dashboard_052.add_class("d52")

display(dashboard_052)

df_final.to_csv(
    EXPORT_DIR / "final_job_market.csv",
    index=False
)

missing_value_audit.to_csv(
    EXPORT_DIR / "final_missing_value_audit.csv",
    index=False
)

final_validation_checks.to_csv(
    EXPORT_DIR / "final_validation_checks.csv",
    index=False
)

daily_quality.to_csv(
    EXPORT_DIR / "temporal_quality_daily.csv",
    index=False
)

outlier_thresholds.to_csv(
    EXPORT_DIR / "outlier_thresholds.csv",
    index=False
)

outlier_thresholds.to_json(
    EXPORT_DIR / "outlier_thresholds.json",
    orient="records",
    indent=2
)

if "logger" in globals():
    logger.info(
        "DATA_CONTROL | records=%s | passed=%s/%s | "
        "hourly_outliers=%s | fixed_outliers=%s | "
        "dense_coverage=%.2f%% | partial_final_day=%s",
        f"{len(df_final):,}",
        passed_checks,
        total_checks,
        int(df_final["hourly_outlier_flag"].sum()),
        int(df_final["fixed_outlier_flag"].sum()),
        dense_share,
        partial_final_day
    )
```

```python
if "df_final" not in globals():
    raise NameError("Run Section 05.2 first.")
if "df_clean" not in globals():
    raise NameError("Run Section 05.1 first.")

control_cols = [
    "hourly_outlier_flag", "fixed_outlier_flag",
    "hourly_midpoint_capped", "fixed_budget_capped"]

base_cols = list(df_clean.columns) + control_cols
df_features = df_final[
    [c for c in base_cols if c in df_final.columns]
].copy()

base_feature_set = set(df_features.columns)
title = df_features["title_normalized"].fillna("")
df_features["title_word_count"] = title.str.split().str.len()
df_features["title_char_count"] = title.str.len()

seniority_rules = [
    (r"\b(chief|ceo|cto|cfo|director|vp|vice president|head)\b", "Executive"),
    (r"\b(lead|manager|principal|supervisor)\b", "Lead / Manager"),
    (r"\b(senior|sr)\b", "Senior"),
    (r"\b(junior|jr|entry level|entry-level|intern|trainee)\b", "Junior / Entry")]

df_features["seniority_level"] = "Unspecified"

for pattern, label in seniority_rules:
    mask = (
        df_features["seniority_level"].eq("Unspecified")
        & title.str.contains(pattern, regex=True, na=False)
    )
    df_features.loc[mask, "seniority_level"] = label

family_rules = {
    "Data & AI":
        r"\b(data|analyst|analytics|machine learning|ml|ai|artificial intelligence|python|sql|power bi|tableau|data science|deep learning|nlp)\b",
    "Software Development":
        r"\b(developer|software|frontend|front end|backend|back end|full stack|fullstack|web|wordpress|shopify|react|javascript|php|java|node|api|android|ios)\b",
    "Design & Creative":
        r"\b(design|designer|graphic|logo|video|animation|figma|ui|ux|illustrator|photoshop|creative)\b",
    "Marketing & Sales":
        r"\b(marketing|seo|sem|sales|advertising|social media|lead generation|google ads|facebook ads|email marketing)\b",
    "Writing & Translation":
        r"\b(writer|writing|content|copywriter|copywriting|translation|translator|editor|proofread|blog)\b",
    "Finance & Accounting":
        r"\b(accounting|accountant|finance|financial|bookkeeping|quickbooks|tax|payroll)\b",
    "Admin & Support":
        r"\b(assistant|virtual assistant|admin|customer support|customer service|data entry|appointment setter)\b",
    "Engineering & Architecture":
        r"\b(engineer|engineering|architect|architecture|cad|autocad|civil|mechanical|electrical|3d model)\b",
    "Legal":
        r"\b(legal|lawyer|attorney|paralegal|contract law)\b"
}

df_features["job_family"] = "Other"

for family, pattern in family_rules.items():
    mask = (
        df_features["job_family"].eq("Other")
        & title.str.contains(pattern, regex=True, na=False)
    )
    df_features.loc[mask, "job_family"] = family

df_features["hourly_spread"] = (
    df_features["hourly_high"]
    - df_features["hourly_low"]
).where(
    df_features["hourly_midpoint"].notna()
)

df_features["hourly_spread_pct"] = (
    df_features["hourly_spread"]
    / df_features["hourly_low"].replace(0, np.nan)
    * 100
)

df_features["has_hourly_pay"] = (
    df_features["hourly_midpoint"].notna()
)

df_features["has_fixed_budget"] = (
    df_features["fixed_budget"].notna()
)

df_features["has_compensation"] = (
    df_features["has_hourly_pay"]
    | df_features["has_fixed_budget"]
)

dt = df_features["published_date"]
df_features["day_of_week"] = dt.dt.day_name()
df_features["day_of_week_num"] = dt.dt.dayofweek
df_features["is_weekend"] = df_features["day_of_week_num"].ge(5)
df_features["posting_hour_utc"] = dt.dt.hour
start_date = dt.min().floor("D")

df_features["days_from_start"] = (
    dt.dt.floor("D") - start_date
).dt.days

df_features["week_index"] = (
    df_features["days_from_start"] // 7
)

dense_start = pd.Timestamp("2024-02-13", tz="UTC")

df_features["dense_window_flag"] = (
    dt.ge(dense_start)
)

if "country_ref" not in globals():
    country_file = Path(coco.__file__).resolve().parent / "country_data.tsv"
    country_ref = pd.read_csv(
        country_file, sep="\t",
        dtype=str, keep_default_na=False
    )

country_lookup = country_ref[
    ["name_short", "ISO3", "continent", "Continent_7", "UNregion"]
].copy()

country_lookup["country_key"] = (
    country_lookup["name_short"].map(norm_country)
)

country_lookup = (
    country_lookup
    .drop_duplicates("country_key")
    .rename(columns={
        "ISO3": "country_code",
        "Continent_7": "continent_7",
        "UNregion": "un_region"
    })
)

df_features["country_key"] = (df_features["country_clean"].map(norm_country))

df_features = df_features.merge(
    country_lookup[
        [
            "country_key",
            "country_code",
            "continent",
            "continent_7",
            "un_region"
        ]
    ],
    on="country_key",
    how="left"
)

mode_geo = country_lookup.loc[
    country_lookup["country_key"].eq(
        norm_country(country_mode)
    )
]

if mode_geo.empty:
    raise ValueError(
        f"Mode country '{country_mode}' could not be mapped."
    )

mode_geo = mode_geo.iloc[0]

geo_cols = [
    "country_code",
    "continent",
    "continent_7",
    "un_region"
]

df_features["geo_fallback_flag"] = (
    df_features[geo_cols].isna().any(axis=1)
)

for col in geo_cols:
    df_features[col] = (
        df_features[col]
        .fillna(mode_geo[col])
    )

df_features["geo_mapping_method"] = "Direct Mapping"

df_features.loc[
    df_features["country_link_recovered_flag"],
    "geo_mapping_method"
] = "Link-Derived Country"

df_features.loc[
    df_features["country_mode_imputed_flag"],
    "geo_mapping_method"
] = "Mode-Imputed Country"

df_features.loc[
    df_features["geo_fallback_flag"],
    "geo_mapping_method"
] = "Mode Geography Fallback"

df_features["continent_missing_flag"] = (
    df_features["continent"].isna()
)

latest_ts = df_features["published_date"].max()
latest_day = latest_ts.floor("D")

partial_last_day = (
    latest_ts.hour < 23
    or latest_ts.minute < 59
)

anchor = (
    latest_day - pd.Timedelta(days=1)
    if partial_last_day
    else latest_day
)

recent_start = anchor - pd.Timedelta(days=6)
previous_start = anchor - pd.Timedelta(days=13)
previous_end = anchor - pd.Timedelta(days=7)

country_market = (
    df_features.groupby("country_clean", dropna=False)
    .agg(
        country_posting_count=("link", "size"),
        country_hourly_share=(
            "pay_type",
            lambda x: x.eq("Hourly").mean() * 100
        ),
        country_fixed_share=(
            "pay_type",
            lambda x: x.eq("Fixed Price").mean() * 100
        ),
        country_median_hourly=(
            "hourly_midpoint", "median"
        ),
        country_median_budget=(
            "fixed_budget", "median"
        ),
        country_hourly_volume=(
            "hourly_midpoint", "count"
        ),
        country_fixed_volume=(
            "fixed_budget", "count"
        ),
        country_active_days=(
            "published_day", "nunique"
        ),
        country_last_posting=(
            "published_day", "max"
        )
    )
    .reset_index()
)

country_market["country_posting_share"] = (
    country_market["country_posting_count"]
    / len(df_features) * 100
)

country_market["country_market_rank"] = (
    country_market["country_posting_count"]
    .rank(method="dense", ascending=False)
    .astype(int)
)

country_market["country_market_tier"] = np.select(
    [
        country_market["country_market_rank"].le(10),
        country_market["country_market_rank"].le(50)
    ],
    [
        "Tier 1",
        "Tier 2"
    ],
    default="Tier 3"
)

country_market["country_recency_days"] = (
    anchor
    - country_market["country_last_posting"]
).dt.days

recent = (
    df_features[
        df_features["published_day"]
        .between(recent_start, anchor)
    ]
    .groupby("country_clean")
    .size()
)

previous = (
    df_features[
        df_features["published_day"]
        .between(previous_start, previous_end)
    ]
    .groupby("country_clean")
    .size()
)

country_market["country_recent_7d"] = (
    country_market["country_clean"]
    .map(recent)
    .fillna(0)
    .astype(int)
)

country_market["country_previous_7d"] = (
    country_market["country_clean"]
    .map(previous)
    .fillna(0)
    .astype(int)
)

country_market["country_growth_pct"] = np.where(
    country_market["country_previous_7d"].gt(0),
    (
        (
            country_market["country_recent_7d"]
            - country_market["country_previous_7d"]
        )
        / country_market["country_previous_7d"]
        * 100
    ),
    np.nan
)

country_daily = (
    df_features.groupby(
        ["country_clean", "published_day"]
    )
    .size()
    .rename("daily_postings")
    .reset_index()
)

country_volatility = (
    country_daily.groupby("country_clean")["daily_postings"]
    .agg(["mean", "std"])
)

country_volatility["country_demand_volatility"] = (
    country_volatility["std"]
    / country_volatility["mean"].replace(0, np.nan)
)

country_market = country_market.merge(
    country_volatility[
        ["country_demand_volatility"]
    ],
    left_on="country_clean",
    right_index=True,
    how="left"
)

country_market["country_sample_eligible"] = (
    country_market["country_posting_count"].ge(100)
)

country_market["country_hourly_sample_eligible"] = (
    country_market["country_hourly_volume"].ge(30)
)

country_market["country_fixed_sample_eligible"] = (
    country_market["country_fixed_volume"].ge(30)
)

df_features = df_features.merge(
    country_market,
    on="country_clean",
    how="left"
)

continent_market = (
    df_features.groupby("continent")
    .agg(
        continent_posting_count=("link", "size"),
        continent_country_count=("country_clean", "nunique"),
        continent_hourly_share=(
            "pay_type",
            lambda x: x.eq("Hourly").mean() * 100
        ),

        continent_fixed_share=(
            "pay_type",
            lambda x: x.eq("Fixed Price").mean() * 100
        ),

        continent_median_hourly=("hourly_midpoint", "median"),
        continent_median_budget=("fixed_budget", "median"),
        continent_active_days=("published_day", "nunique")
    )
    .reset_index()
)

continent_market["continent_posting_share"] = (
    continent_market["continent_posting_count"]
    / len(df_features) * 100
)

continent_market["continent_market_rank"] = (
    continent_market["continent_posting_count"]
    .rank(method="dense", ascending=False)
    .astype(int)
)

continent_market["continent_median_hourly"] = (
    continent_market["continent_median_hourly"]
    .fillna(df_features["hourly_midpoint"].median())
)

continent_market["continent_median_budget"] = (
    continent_market["continent_median_budget"]
    .fillna(df_features["fixed_budget"].median())
)

df_features = df_features.merge(
    continent_market,
    on="continent",
    how="left"
)

geo_final_cols = [
    "country_code",
    "continent",
    "continent_7",
    "un_region",
    "continent_fixed_share",
    "continent_country_count",
    "continent_hourly_share",
    "continent_posting_count",
    "continent_median_hourly",
    "continent_median_budget",
    "continent_active_days",
    "continent_posting_share"
]

geo_missing = (
    df_features[geo_final_cols]
    .isna()
    .mean()
    .mul(100)
)

if geo_missing.gt(0).any():
    raise ValueError(
        "Geography features still contain missing values:\n"
        + geo_missing[geo_missing.gt(0)].to_string()
    )

df_final = df_features.copy(deep=True)

new_features = [
    c for c in df_final.columns
    if c not in base_feature_set
]

def feature_group(col):
    if col in [
        "title_word_count", "title_char_count",
        "seniority_level", "job_family"
    ]:
        return "Text & Job"

    if (
        "hourly" in col
        or "budget" in col
        or col.startswith("has_")
    ):
        return "Compensation"

    if col in [
        "day_of_week", "day_of_week_num",
        "is_weekend", "posting_hour_utc",
        "days_from_start", "week_index",
        "dense_window_flag"
    ]:
        return "Temporal"

    if (
        col.startswith("country_")
        or col in [
            "country_code",
            "continent",
            "continent_7",
            "un_region",
            "geo_fallback_flag",
            "geo_mapping_method"
        ]
    ):
        return "Geography & Country"

    if col.startswith("continent_"):
        return "Continent Market"

    return "Other"

feature_inventory = pd.DataFrame({
    "Feature": new_features
})

feature_inventory["Group"] = (
    feature_inventory["Feature"]
    .map(feature_group)
)

feature_inventory["Data Type"] = (
    feature_inventory["Feature"]
    .map(lambda c: str(df_final[c].dtype))
)

feature_inventory["Missing %"] = (
    feature_inventory["Feature"]
    .map(
        lambda c:
        round(
            df_final[c].isna().mean() * 100,
            2
        )
    )
)

geo_coverage = (
    df_final["continent"].notna().mean() * 100
)

pay_coverage = (
    df_final["has_compensation"].mean() * 100
)

dense_share = (
    df_final["dense_window_flag"].mean() * 100
)

family_share = (
    df_final["job_family"].ne("Other").mean() * 100
)

seniority_share = (
    df_final["seniority_level"].ne(
        "Unspecified"
    ).mean() * 100
)

eligible_markets = int(
    country_market["country_sample_eligible"].sum()
)

geo_fallback_n = int(
    df_final["geo_fallback_flag"].sum()
)

feature_quality_audit = pd.DataFrame({
    "Control": [
        "Geographic completeness",
        "Compensation availability",
        "Dense-window coverage",
        "Job-family classification",
        "Explicit seniority detection",
        "Comparison-ready country markets",
        "Geography fallback records"
    ],

    "Result": [
        f"{geo_coverage:.2f}%",
        f"{pay_coverage:.2f}%",
        f"{dense_share:.2f}%",
        f"{family_share:.2f}%",
        f"{seniority_share:.2f}%",
        f"{eligible_markets:,}",
        f"{geo_fallback_n:,}"
    ],

    "Interpretation": [
        "Country and continent fields complete",
        "Records with usable original compensation",
        "Records inside preferred trend window",
        "Titles classified beyond Other",
        "Titles containing explicit seniority signals",
        "Countries with at least 100 postings",
        "Rows requiring final geography fallback"
    ]
})

feature_lineage = pd.DataFrame([
    ["title", "job_family / seniority_level", "Category analysis & recommendations"],
    ["hourly_low + hourly_high", "hourly_spread / midpoint", "Compensation analysis"],
    ["published_date", "weekday / hour / dense-window features", "Trend & forecasting"],
    ["country", "country_code / continent / region", "Geographic analysis"],
    ["country + date", "recent demand / growth / volatility", "Market dynamics"],
    ["country + pay", "country median pay / eligibility", "Country comparison"],
    ["continent + pay", "continent market aggregates", "Regional comparison"]
], columns=[
    "Source",
    "Engineered Feature",
    "Primary Use"
])

job_profile = (
    df_final.groupby("job_family")
    .agg(
        Postings=("link", "size"),
        Seniority_Groups=("seniority_level", "nunique"),
        Median_Hourly=("hourly_midpoint", "median"),
        Median_Budget=("fixed_budget", "median")
    )
    .reset_index()
    .sort_values("Postings", ascending=False)
)

comp_profile = pd.DataFrame({
    "Metric": [
        "Hourly jobs",
        "Hourly pay available",
        "Fixed-price jobs",
        "Fixed budget available",
        "Any original compensation"
    ],
    "Records": [
        int(df_final["pay_type"].eq("Hourly").sum()),
        int(df_final["has_hourly_pay"].sum()),
        int(df_final["pay_type"].eq("Fixed Price").sum()),
        int(df_final["has_fixed_budget"].sum()),
        int(df_final["has_compensation"].sum())
    ]
})

temporal_profile = (
    df_final.groupby(
        ["day_of_week_num", "day_of_week"]
    )
    .agg(
        Postings=("link", "size"),
        Active_Days=("published_day", "nunique")
    )
    .reset_index()
    .sort_values("day_of_week_num")
    .drop(columns="day_of_week_num")
)

refresh_time = pd.Timestamp.now(
    tz="Asia/Kolkata"
).strftime("%d %b %Y • %I:%M %p")

def c53_card(label, value, note):
    return (
        f'<div class="c53"><small>{label}</small>'
        f'<b>{value}</b><span>{note}</span></div>'
    )

display(HTML("""
<style>
.d53{border:1px solid #ded8d5;border-radius:15px;overflow:hidden;background:#fff;
box-shadow:0 6px 20px rgba(34,34,59,.08)}
.h53{background:linear-gradient(120deg,#22223b,#30324d,#4a4e69);
padding:23px 26px;display:flex;justify-content:space-between;font-family:Arial}
.h53 .k{color:#c9ada7;font-size:8px;font-weight:700;letter-spacing:1.5px}
.h53 h2{color:#fff;font-size:23px;margin:6px 0}
.h53 p,.h53 .m{color:#f2e9e4;font-size:10px;line-height:1.6}
.cards53{display:grid;grid-template-columns:repeat(8,1fr);border-bottom:1px solid #e8e2df}
.c53{padding:13px;border-right:1px solid #ebe6e4;font-family:Arial}
.c53 small{display:block;color:#9a8c98;font-size:7px;font-weight:700}
.c53 b{display:block;color:#22223b;font-size:17px;margin-top:5px}
.c53 span{display:block;color:#4a4e69;font-size:8px;margin-top:3px}
.btn53{margin:13px 18px 8px!important}.focus53{margin:0 18px 13px!important}
.out53{border-top:1px solid #eee9e7!important}
.t53{width:100%;border-collapse:collapse;font:11px Arial}
.t53 th{background:#22223b;color:#f2e9e4;padding:9px;text-align:left}
.t53 td{padding:9px;border-bottom:1px solid #eee9e7;color:#22223b}
.txt53{font:12px/1.8 Arial;color:#22223b}
</style>
"""))
feature_header = widgets.HTML(f"""
<div class="h53">
<div>
<div class="k">05.3 • FEATURE ENGINEERING</div>
<h2>Feature Intelligence Control Center</h2>
<p>Text • Compensation • Temporal • Geography •
Country markets • Continent markets • Demand signals</p>
</div>
<div class="m">LAST REFRESHED<br><b>{refresh_time} IST</b></div>
</div>
<div class="cards53">
{c53_card("NEW FEATURES", f"{len(new_features):,}", "Engineered")}
{c53_card("JOB FAMILIES", f"{df_final['job_family'].nunique():,}", "Including Other")}
{c53_card("COUNTRIES", f"{df_final['country_clean'].nunique():,}", "Markets")}
{c53_card("CONTINENTS", f"{df_final['continent'].nunique():,}", "Mapped")}
{c53_card("GEO COVERAGE", f"{geo_coverage:.1f}%", "Complete")}
{c53_card("PAY COVERAGE", f"{pay_coverage:.1f}%", "Original pay")}
{c53_card("DENSE WINDOW", f"{dense_share:.1f}%", "Trend ready")}
{c53_card("ELIGIBLE MARKETS", f"{eligible_markets:,}", "N ≥ 100")}
</div>
""")

feature_view = widgets.ToggleButtons(
    options=[
        ("OVERVIEW", "overview"),
        ("TABLE", "table"),
        ("CHART", "chart"),
        ("INSIGHTS", "insights"),
        ("QUALITY", "quality"),
        ("LINEAGE", "lineage"),
        ("INTERPRETATION", "interpretation"),
        ("RECOMMENDATIONS", "recommendations")
    ],
    value="overview",
    layout=widgets.Layout(width="calc(100% - 36px)"),
    style={"button_width": "12%", "font_weight": "700"}
)
feature_view.add_class("btn53")

feature_focus = widgets.Dropdown(
    options=[
        ("Geography & Market", "geo"),
        ("Job Families & Seniority", "jobs"),
        ("Compensation Features", "comp"),
        ("Temporal Features", "time"),
        ("Feature Inventory", "inventory")
    ],
    value="geo",
    description="Focus:",
    layout=widgets.Layout(width="360px")
)
feature_focus.add_class("focus53")

feature_output = widgets.Output(
    layout=widgets.Layout(padding="20px 23px 23px")
)
feature_output.add_class("out53")

def update_feature_dashboard(change=None):
    with feature_output:
        clear_output(wait=True)
        view = feature_view.value
        focus = feature_focus.value
        if view == "overview":
            group_counts = (
                feature_inventory["Group"]
                .value_counts()
                .rename_axis("Feature Group")
                .reset_index(name="Features")
            )
            display(HTML(
                group_counts.to_html(
                    index=False,
                    border=0,
                    classes="t53"
                )
            ))
        elif view == "table":
            tables = {
                "geo":
                    country_market.sort_values(
                        "country_posting_count",
                        ascending=False
                    ).head(15),
                "jobs":
                    job_profile,
                "comp":
                    comp_profile,
                "time":
                    temporal_profile,
                "inventory":
                    feature_inventory
            }
            display(HTML(
                tables[focus].to_html(
                    index=False,
                    border=0,
                    classes="t53"
                )
            ))

        elif view == "chart":
            if focus == "geo":
                sankey_df = (
                    df_final.groupby(
                        [
                            "continent",
                            "country_market_tier",
                            "job_family"
                        ]
                    )
                    .size()
                    .rename("value")
                    .reset_index()
                )

                top_families = (
                    sankey_df.groupby("job_family")["value"]
                    .sum()
                    .nlargest(8)
                    .index
                )

                sankey_df = sankey_df[
                    sankey_df["job_family"].isin(
                        top_families
                    )
                ]

                continents = [
                    f"C:{x}"
                    for x in sankey_df["continent"].unique()
                ]

                tiers = [
                    f"T:{x}"
                    for x in sankey_df["country_market_tier"].unique()
                ]

                families = [
                    f"J:{x}"
                    for x in sankey_df["job_family"].unique()
                ]

                nodes = continents + tiers + families
                node_index = {
                    n: i for i, n in enumerate(nodes)
                }

                left = (
                    sankey_df.groupby(
                        ["continent", "country_market_tier"]
                    )["value"].sum().reset_index()
                )

                right = (
                    sankey_df.groupby(
                        ["country_market_tier", "job_family"]
                    )["value"].sum().reset_index()
                )

                source = (
                    [node_index[f"C:{x}"] for x in left["continent"]]
                    + [node_index[f"T:{x}"] for x in right["country_market_tier"]]
                )

                target = (
                    [node_index[f"T:{x}"] for x in left["country_market_tier"]]
                    + [node_index[f"J:{x}"] for x in right["job_family"]]
                )

                values = (
                    left["value"].tolist()
                    + right["value"].tolist()
                )

                fig = go.Figure(go.Sankey(
                    node=dict(
                        label=[
                            x.split(":", 1)[1]
                            for x in nodes
                        ],
                        color="#9a8c98"
                    ),
                    link=dict(
                        source=source,
                        target=target,
                        value=values
                    )
                ))

            elif focus == "jobs":

                job_tree = (
                    df_final.groupby(
                        ["seniority_level", "job_family"]
                    )
                    .size()
                    .rename("value")
                    .reset_index()
                )

                labels = ["All Jobs"]
                parents = [""]
                values = [job_tree["value"].sum()]

                for seniority in job_tree["seniority_level"].unique():

                    part = job_tree[
                        job_tree["seniority_level"].eq(
                            seniority
                        )
                    ]

                    labels.append(seniority)
                    parents.append("All Jobs")
                    values.append(part["value"].sum())

                    for _, row in part.iterrows():
                        labels.append(
                            f"{seniority} • {row['job_family']}"
                        )
                        parents.append(seniority)
                        values.append(row["value"])

                fig = go.Figure(go.Sunburst(
                    labels=labels,
                    parents=parents,
                    values=values,
                    branchvalues="total"
                ))


            elif focus == "comp":

                bubble = country_market[
                    country_market[
                        "country_hourly_sample_eligible"
                    ]
                    & country_market[
                        "country_median_hourly"
                    ].notna()
                ].copy()

                size = np.sqrt(
                    bubble["country_hourly_volume"]
                )

                fig = go.Figure(go.Scatter(
                    x=bubble["country_posting_count"],
                    y=bubble["country_median_hourly"],
                    mode="markers",
                    text=bubble["country_clean"],
                    marker=dict(
                        size=size,
                        sizemode="diameter",
                        color="#4a4e69",
                        opacity=.7
                    ),
                    hovertemplate=(
                        "<b>%{text}</b><br>"
                        "Postings: %{x:,}<br>"
                        "Median hourly: %{y:.2f}"
                        "<extra></extra>"
                    )
                ))

                fig.update_xaxes(
                    title="Country Posting Volume"
                )

                fig.update_yaxes(
                    title="Median Hourly Rate"
                )


            elif focus == "time":

                order = [
                    "Monday", "Tuesday", "Wednesday",
                    "Thursday", "Friday", "Saturday",
                    "Sunday"
                ]

                heat = (
                    df_final.groupby(
                        ["day_of_week", "posting_hour_utc"]
                    )
                    .size()
                    .unstack(fill_value=0)
                    .reindex(order)
                )

                fig = go.Figure(go.Heatmap(
                    z=heat.values,
                    x=heat.columns,
                    y=heat.index,
                    colorscale=[
                        [0, "#f2e9e4"],
                        [.5, "#9a8c98"],
                        [1, "#22223b"]
                    ]
                ))

                fig.update_xaxes(
                    title="UTC Posting Hour"
                )


            else:

                funnel = (
                    feature_inventory["Group"]
                    .value_counts()
                    .reset_index()
                )

                funnel.columns = [
                    "Group",
                    "Features"
                ]

                fig = go.Figure(go.Funnel(
                    y=funnel["Group"],
                    x=funnel["Features"],
                    marker={"color": "#4a4e69"}
                ))


            fig.update_layout(
                height=500,
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",
                font={
                    "family": "Arial",
                    "color": "#22223b"
                },
                margin={
                    "l": 40,
                    "r": 35,
                    "t": 30,
                    "b": 45
                }
            )

            display(fig)


        elif view == "quality":

            display(HTML(
                feature_quality_audit.to_html(
                    index=False,
                    border=0,
                    classes="t53"
                )
            ))


        elif view == "lineage":

            display(HTML(
                feature_lineage.to_html(
                    index=False,
                    border=0,
                    classes="t53"
                )
            ))


        elif view == "insights":

            display(HTML(f"""
            <div class="txt53">

            <b>{len(new_features):,} engineered features</b>
            now form the analytical layer.<br><br>

            Geographic fields are
            <b>{geo_coverage:.2f}% complete</b>,
            including continent and region mappings.<br><br>

            <b>{eligible_markets:,} countries</b>
            meet the minimum 100-posting threshold for stronger
            country-level comparisons.<br><br>

            <b>{family_share:.2f}%</b> of job titles are classified
            beyond the generic Other category.<br><br>

            The preferred dense trend window contains
            <b>{dense_share:.2f}%</b> of all records.

            </div>
            """))


        elif view == "interpretation":

            interpretations = {
                "geo":
                    "Geographic enrichment supports country and regional comparison. "
                    "Fallback flags preserve transparency where location was imputed.",

                "jobs":
                    "Job families and seniority are title-derived classifications, "
                    "not verified employer skills or experience requirements.",

                "comp":
                    "Hourly and fixed-price compensation remain analytically separate. "
                    "Original pay should remain the primary descriptive measure.",

                "time":
                    "Dense-window and calendar features support trend analysis while "
                    "protecting later forecasting from the sparse early period.",

                "inventory":
                    "The feature inventory provides a reproducible record of the variables "
                    "introduced during engineering."
            }

            display(HTML(
                f'<div class="txt53">{interpretations[focus]}</div>'
            ))


        else:

            display(HTML("""
            <div class="txt53">

            <b>01.</b> Use job_family for emerging-category analysis.<br>
            <b>02.</b> Use country_code for the Task 4 choropleth.<br>
            <b>03.</b> Apply country sample-eligibility flags before pay comparisons.<br>
            <b>04.</b> Use dense-window temporal features for forecasting.<br>
            <b>05.</b> Use recent growth and volatility for market-dynamics analysis.<br>
            <b>06.</b> Keep hourly and fixed-price compensation separate.<br>
            <b>07.</b> Recalculate target-like market aggregates inside training folds when modelling.

            </div>
            """))


feature_view.observe(
    update_feature_dashboard,
    names="value"
)

feature_focus.observe(
    update_feature_dashboard,
    names="value"
)

update_feature_dashboard()

dashboard_053 = widgets.VBox([
    feature_header,
    feature_view,
    feature_focus,
    feature_output
])

dashboard_053.add_class("d53")

display(dashboard_053)

exports_053 = {
    "country_market_features.csv":
        country_market,

    "continent_market_features.csv":
        continent_market,

    "feature_inventory.csv":
        feature_inventory,

    "feature_quality_audit.csv":
        feature_quality_audit,

    "feature_lineage.csv":
        feature_lineage,

    "final_feature_engineered_job_market.csv":
        df_final
}

for name, data in exports_053.items():
    data.to_csv(
        EXPORT_DIR / name,
        index=False
    )

if "logger" in globals():
    logger.info(
        "FEATURE_ENGINEERING | records=%s | "
        "new_features=%s | geo_coverage=%.2f%% | "
        "job_family_coverage=%.2f%% | "
        "eligible_markets=%s | geo_fallback=%s",
        f"{len(df_final):,}",
        len(new_features),
        geo_coverage,
        family_share,
        eligible_markets,
        geo_fallback_n
    )
```

```python
if "df_final" not in globals():
    raise NameError("Run Section 05.3 first.")

eda=df_final.copy()
hourly_eda=eda[eda["hourly_midpoint"].notna()]
fixed_eda=eda[eda["fixed_budget"].notna()]
n_records=len(eda)
n_countries=eda["country_clean"].nunique()
n_families=eda["job_family"].nunique()
hourly_share=eda["pay_type"].eq("Hourly").mean()*100
median_hourly=hourly_eda["hourly_midpoint"].median()
median_fixed=fixed_eda["fixed_budget"].median()
original_pay_coverage=(eda["hourly_midpoint"].notna()|eda["fixed_budget"].notna()).mean()*100
model_pay_coverage=(eda["hourly_midpoint_imputed"].notna()|eda["fixed_budget_imputed"].notna()).mean()*100

compensation_eda=pd.DataFrame({
    "Measure":["Hourly postings","Valid hourly pay","Median hourly midpoint","Fixed-price postings",
               "Valid fixed budget","Median fixed budget","Original pay coverage","Model-ready coverage"],
    "Result":[f"{eda['pay_type'].eq('Hourly').sum():,}",f"{len(hourly_eda):,}",f"{median_hourly:,.2f}",
              f"{eda['pay_type'].eq('Fixed Price').sum():,}",f"{len(fixed_eda):,}",f"{median_fixed:,.2f}",
              f"{original_pay_coverage:.2f}%",f"{model_pay_coverage:.2f}%"]
})

job_family_eda=(eda.groupby("job_family")
    .agg(Postings=("link","size"),Countries=("country_clean","nunique"),
         Median_Hourly=("hourly_midpoint","median"),Median_Fixed=("fixed_budget","median"))
    .reset_index())
job_family_eda["Share %"]=(job_family_eda["Postings"]/n_records*100).round(2)
job_family_eda=job_family_eda.sort_values("Postings",ascending=False)

country_eda=(eda.groupby("country_clean")
    .agg(Postings=("link","size"),Job_Families=("job_family","nunique"),
         Median_Hourly=("hourly_midpoint","median"),Median_Fixed=("fixed_budget","median"))
    .reset_index().sort_values("Postings",ascending=False))
country_eda["Share %"]=(country_eda["Postings"]/n_records*100).round(2)
country_eda["Cumulative Share %"]=country_eda["Share %"].cumsum()

weekday_order=["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]
temporal_eda=(eda.groupby("day_of_week")
    .agg(Postings=("link","size"),Active_Days=("published_day","nunique"),
         Median_Hourly=("hourly_midpoint","median"))
    .reindex(weekday_order).reset_index())
temporal_eda["Share %"]=(temporal_eda["Postings"]/temporal_eda["Postings"].sum()*100).round(2)

outlier_eda=pd.DataFrame({
    "Measure":["Hourly observations","Hourly outliers","Hourly outlier share",
               "Fixed observations","Fixed outliers","Fixed outlier share"],
    "Result":[f"{len(hourly_eda):,}",f"{eda['hourly_outlier_flag'].sum():,}",
              f"{eda.loc[eda['hourly_midpoint'].notna(),'hourly_outlier_flag'].mean()*100:.2f}%",
              f"{len(fixed_eda):,}",f"{eda['fixed_outlier_flag'].sum():,}",
              f"{eda.loc[eda['fixed_budget'].notna(),'fixed_outlier_flag'].mean()*100:.2f}%"]
})

country_control=country_eda.merge(
    country_market[["country_clean","country_sample_eligible",
                    "country_hourly_sample_eligible","country_fixed_sample_eligible"]],
    on="country_clean",how="left"
)

refresh_time=pd.Timestamp.now(tz="Asia/Kolkata").strftime("%d %b %Y • %I:%M %p")

def card(a,b,c):
    return f'<div class="c6"><small>{a}</small><b>{b}</b><span>{c}</span></div>'

def show_table(df,n=None):
    display(HTML((df.head(n) if n else df).to_html(
        index=False,border=0,classes="t6",float_format=lambda x:f"{x:,.2f}"
    )))

display(HTML("""
<style>
.d6{border:1px solid #ded8d5;border-radius:15px;overflow:hidden;background:#fff;box-shadow:0 6px 20px #22223b14}
.h6{background:linear-gradient(120deg,#22223b,#30324d,#4a4e69);padding:22px 25px;display:flex;justify-content:space-between;font-family:Arial}
.h6 .k{color:#c9ada7;font-size:8px;font-weight:700;letter-spacing:1.5px}.h6 h2{color:#fff;margin:5px 0;font-size:23px}.h6 p,.h6 .m{color:#f2e9e4;font-size:10px;line-height:1.6}
.cards6{display:grid;grid-template-columns:repeat(6,1fr);border-bottom:1px solid #e8e2df}.c6{padding:13px;border-right:1px solid #ebe6e4;font-family:Arial;min-width:0}
.c6 small{display:block;color:#9a8c98;font-size:7px;font-weight:700}.c6 b{display:block;color:#22223b;font-size:17px;margin-top:5px;overflow-wrap:anywhere}.c6 span{display:block;color:#4a4e69;font-size:8px;margin-top:3px}
.ctrl6{padding:12px 18px 3px}.btn6{margin:7px 18px!important}.out6{border-top:1px solid #eee9e7!important}
.t6{width:100%;border-collapse:collapse;font:11px Arial}.t6 th{background:#22223b;color:#fff;padding:9px;text-align:left}.t6 td{padding:9px;border-bottom:1px solid #eee9e7;color:#22223b}
.txt6{font:12px/1.8 Arial;color:#22223b}.tag6{display:inline-block;background:#f2e9e4;color:#4a4e69;padding:5px 9px;border-radius:12px;margin:3px;font:9px Arial}
.info6{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}.box6{background:#f8f6f5;border-left:4px solid #4a4e69;padding:14px;border-radius:7px;font:11px/1.6 Arial;color:#22223b}
.box6 b{display:block;color:#4a4e69;font-size:9px;margin-bottom:5px}
</style>
"""))
header=widgets.HTML(f"""
<div class="h6">
<div><div class="k">SECTION 06 • EXPLORATORY DATA ANALYSIS</div>
<h2>Job Market Exploratory Intelligence</h2>
<p>Compensation • Job Demand • Geography • Time • Outliers</p></div>
<div class="m">RECORDS<br><b>{n_records:,}</b><br><br>UPDATED<br><b>{refresh_time} IST</b></div>
</div>
""")

focus=widgets.Dropdown(
    options=[("Compensation","comp"),("Job Families","jobs"),("Geographic Markets","country"),
             ("Temporal Activity","time"),("Outlier Analysis","outliers")],
    value="comp",description="Analysis:",layout=widgets.Layout(width="290px")
)

metric=widgets.Dropdown(description="Metric:",layout=widgets.Layout(width="310px"))

basis=widgets.Dropdown(
    options=[("Original","original"),("Robust / Capped","capped")],
    value="original",description="Pay View:",layout=widgets.Layout(width="260px")
)

top_n=widgets.IntSlider(
    value=10,min=5,max=30,step=5,description="Top N:",
    continuous_update=False,layout=widgets.Layout(width="290px")
)

market_filter=widgets.Dropdown(
    options=[("All Markets","all"),("N ≥ 100","eligible"),
             ("Hourly N ≥ 30","hourly"),("Fixed N ≥ 30","fixed")],
    value="all",description="Market:",layout=widgets.Layout(width="270px")
)

view=widgets.ToggleButtons(
    options=[("OVERVIEW","overview"),("EXPLORE","chart"),("DATA","table"),
             ("INSIGHTS","insights"),("FINDINGS","findings"),("RECOMMENDATIONS","recommendations"),
             ("INTERPRETATION","interpretation"),("METHOD","method")],
    value="overview",layout=widgets.Layout(width="100%"),
    style={"button_width":"12%","font_weight":"700"}
)
view.add_class("btn6")

controls=widgets.HBox(
    [focus,metric,basis,top_n,market_filter],
    layout=widgets.Layout(padding="8px 18px",flex_flow="row wrap",gap="10px")
)

kpi_box=widgets.HTML()
output=widgets.Output(layout=widgets.Layout(padding="20px 23px"))
output.add_class("out6")

def set_controls(*_):
    f=focus.value
    metric.options={
        "comp":[("Hourly Distribution","hourly"),("Fixed Budget Distribution","fixed")],
        "jobs":[("Posting Volume","volume"),("Median Hourly Pay","hourly"),("Median Fixed Budget","fixed")],
        "country":[("Posting Volume","volume"),("Median Hourly Pay","hourly"),
                   ("Median Fixed Budget","fixed"),("Market Share","share")],
        "time":[("Posting Volume","volume"),("Median Hourly Pay","hourly")],
        "outliers":[("Hourly Pay","hourly"),("Fixed Budget","fixed")]
    }[f]
    top_n.layout.display="" if f in ["jobs","country"] else "none"
    market_filter.layout.display="" if f=="country" else "none"
    basis.layout.display="" if f in ["comp","outliers"] else "none"

def filtered_countries():
    d=country_control.copy()
    col={"eligible":"country_sample_eligible","hourly":"country_hourly_sample_eligible",
         "fixed":"country_fixed_sample_eligible"}.get(market_filter.value)
    return d[d[col].fillna(False)] if col else d

def update_kpis():
    f=focus.value

    if f=="comp":
        cards=[
            ("HOURLY JOBS",f"{eda['pay_type'].eq('Hourly').sum():,}","Posting type"),
            ("FIXED JOBS",f"{eda['pay_type'].eq('Fixed Price').sum():,}","Posting type"),
            ("MEDIAN HOURLY",f"{median_hourly:,.2f}","Original"),
            ("MEDIAN FIXED",f"{median_fixed:,.0f}","Original"),
            ("SOURCE PAY",f"{original_pay_coverage:.1f}%","Observed"),
            ("MODEL READY",f"{model_pay_coverage:.1f}%","After imputation")
        ]

    elif f=="jobs":
        x=job_family_eda.iloc[0]
        cards=[
            ("JOB FAMILIES",n_families,"Derived groups"),
            ("TOP FAMILY",x["job_family"],"Highest demand"),
            ("TOP VOLUME",f"{int(x['Postings']):,}","Postings"),
            ("CLASSIFIED",f"{eda['job_family'].ne('Other').mean()*100:.1f}%","Beyond Other"),
            ("OTHER",f"{eda['job_family'].eq('Other').mean()*100:.1f}%","Fallback"),
            ("COUNTRIES",n_countries,"Coverage")
        ]

    elif f=="country":
        d=filtered_countries()
        x=d.iloc[0] if len(d) else country_eda.iloc[0]
        cards=[
            ("MARKETS",len(d),"Current filter"),
            ("TOP MARKET",x["country_clean"],"By demand"),
            ("TOP POSTINGS",f"{int(x['Postings']):,}","Jobs"),
            ("TOP 5 SHARE",f"{d.head(5)['Share %'].sum():.1f}%","Concentration"),
            ("ELIGIBLE",f"{country_market['country_sample_eligible'].sum():,}","N ≥ 100"),
            ("GEO COMPLETE",f"{eda['continent'].notna().mean()*100:.1f}%","Mapped")
        ]

    elif f=="time":
        x=temporal_eda.loc[temporal_eda["Postings"].idxmax()]
        cards=[
            ("DENSE WINDOW",f"{eda['dense_window_flag'].mean()*100:.1f}%","Trend ready"),
            ("PEAK DAY",x["day_of_week"],"Most active"),
            ("PEAK VOLUME",f"{int(x['Postings']):,}","Postings"),
            ("WEEKEND",f"{eda['is_weekend'].mean()*100:.1f}%","Posting share"),
            ("ACTIVE DAYS",eda["published_day"].nunique(),"Observed"),
            ("UTC HOURS",eda["posting_hour_utc"].nunique(),"Coverage")
        ]

    else:
        cards=[
            ("HOURLY OUTLIERS",f"{eda['hourly_outlier_flag'].sum():,}","IQR flagged"),
            ("FIXED OUTLIERS",f"{eda['fixed_outlier_flag'].sum():,}","IQR flagged"),
            ("HOURLY RATE",f"{eda.loc[eda['hourly_midpoint'].notna(),'hourly_outlier_flag'].mean()*100:.1f}%","Outlier share"),
            ("FIXED RATE",f"{eda.loc[eda['fixed_budget'].notna(),'fixed_outlier_flag'].mean()*100:.1f}%","Outlier share"),
            ("METHOD","IQR","Detection"),("ROBUSTNESS","P01–P99","Sensitivity")
        ]

    kpi_box.value='<div class="cards6">'+"".join(card(*x) for x in cards)+"</div>"

def update_dashboard(*_):
    update_kpis()

    with output:
        clear_output(wait=True)
        f,v,m,n=focus.value,view.value,metric.value,top_n.value

        if v=="overview":
            q,e,c={
                "comp":("How is pay structured?","Hourly vs fixed-price compensation.","Keep both pay types separate."),
                "jobs":("Where is demand concentrated?","Title-derived job families.","Other is a fallback category."),
                "country":("Which markets dominate?","Country demand and compensation.","Use sample-size controls."),
                "time":("When is activity strongest?","Weekday and dense-window patterns.","Sparse periods need caution."),
                "outliers":("How extreme is pay?","Original versus robust compensation.","Outliers remain preserved.")
            }[f]
            display(HTML(f"""
            <div class="info6">
            <div class="box6"><b>ANALYTICAL QUESTION</b>{q}</div>
            <div class="box6"><b>EVIDENCE</b>{e}</div>
            <div class="box6"><b>CONTROL</b>{c}</div>
            </div><br>
            <span class="tag6">DESCRIPTIVE EDA</span><span class="tag6">INTERACTIVE</span>
            <span class="tag6">AUDITABLE</span><span class="tag6">CLIENT READY</span>
            """))

        elif v=="table":
            if f=="comp":
                show_table(compensation_eda)
            elif f=="jobs":
                show_table(job_family_eda[
                    ["job_family","Postings","Share %","Countries","Median_Hourly","Median_Fixed"]
                ],n)
            elif f=="country":
                show_table(filtered_countries()[
                    ["country_clean","Postings","Share %","Cumulative Share %",
                     "Job_Families","Median_Hourly","Median_Fixed"]
                ],n)
            elif f=="time":
                show_table(temporal_eda)
            else:
                show_table(outlier_eda)

        elif v=="chart":

            if f=="comp":
                col="hourly_midpoint" if m=="hourly" else "fixed_budget"
                if basis.value=="capped":
                    col="hourly_midpoint_capped" if m=="hourly" else "fixed_budget_capped"
                d=eda[col].dropna()
                if basis.value=="original":
                    d=d[d<=d.quantile(.99)]

                fig=go.Figure(go.Histogram(x=d,nbinsx=50,marker_color="#4a4e69",opacity=.85))
                fig.update_xaxes(title=col.replace("_"," ").title())
                fig.update_yaxes(title="Job Postings")

            elif f=="jobs":
                col={"volume":"Postings","hourly":"Median_Hourly","fixed":"Median_Fixed"}[m]
                d=job_family_eda.dropna(subset=[col]).nlargest(n,col).sort_values(col)

                fig=go.Figure(go.Scatter(
                    x=d[col],y=d["job_family"],mode="markers",
                    marker=dict(size=np.sqrt(d["Postings"])*1.3,color="#4a4e69",
                                opacity=.78,line=dict(color="#22223b",width=1)),
                    text=d["Postings"],customdata=d["Share %"],
                    hovertemplate="<b>%{y}</b><br>Value: %{x:,.2f}<br>Postings: %{text:,}<br>Share: %{customdata:.2f}%<extra></extra>"
                ))
                fig.update_xaxes(title=col.replace("_"," "))

            elif f=="country":
                d=filtered_countries()

                if m=="volume":
                    d=d.nlargest(n,"Postings").sort_values("Postings")
                    fig=go.Figure(go.Bar(
                        x=d["Postings"],y=d["country_clean"],orientation="h",
                        marker_color="#4a4e69",text=d["Share %"]
                    ))
                    fig.update_xaxes(title="Job Postings")

                elif m=="share":
                    d=d.head(n).copy()
                    d["Rank"]=range(1,len(d)+1)

                    fig=go.Figure(go.Scatter(
                        x=d["Rank"],y=d["Cumulative Share %"],mode="lines+markers",
                        line=dict(color="#4a4e69",width=3),
                        marker=dict(color="#9a8c98",size=8),text=d["country_clean"]
                    ))
                    fig.update_xaxes(title="Market Rank")
                    fig.update_yaxes(title="Cumulative Share (%)",range=[0,100])

                else:
                    col="Median_Hourly" if m=="hourly" else "Median_Fixed"
                    d=d.dropna(subset=[col]).nlargest(n,col).sort_values(col)

                    fig=go.Figure(go.Scatter(
                        x=d[col],y=d["country_clean"],mode="markers",
                        marker=dict(size=np.sqrt(d["Postings"])*1.2,color="#9a8c98",opacity=.8),
                        text=d["Postings"]
                    ))
                    fig.update_xaxes(title=col.replace("_"," "))

            elif f=="time":
                if m=="volume":
                    fig=go.Figure(go.Barpolar(
                        r=temporal_eda["Postings"],theta=temporal_eda["day_of_week"],
                        marker_color="#4a4e69",opacity=.82
                    ))
                else:
                    fig=go.Figure(go.Scatter(
                        x=temporal_eda["day_of_week"],y=temporal_eda["Median_Hourly"],
                        mode="lines+markers",line=dict(color="#4a4e69",width=3),
                        marker=dict(size=9,color="#9a8c98")
                    ))
                    fig.update_yaxes(title="Median Hourly Pay")

            else:
                col="hourly_midpoint" if m=="hourly" else "fixed_budget"
                cap="hourly_midpoint_capped" if m=="hourly" else "fixed_budget_capped"
                d=eda[[col,cap]].dropna()
                d=d.sample(min(7000,len(d)),random_state=42)

                fig=go.Figure([
                    go.Violin(y=d[col],name="Original",box_visible=True,
                              meanline_visible=True,line_color="#22223b"),
                    go.Violin(y=d[cap],name="P01–P99 Capped",box_visible=True,
                              meanline_visible=True,line_color="#9a8c98")
                ])
                fig.update_yaxes(title=col.replace("_"," ").title())

            fig.update_layout(
                height=480,paper_bgcolor="#fff",plot_bgcolor="#fff",
                font=dict(family="Arial",color="#22223b"),
                margin=dict(l=55,r=30,t=25,b=45),
                hoverlabel=dict(bgcolor="#22223b",font_color="#fff")
            )
            display(fig)

        elif v=="insights":
            top_family=job_family_eda.iloc[0]
            top_country=country_eda.iloc[0]

            text={
                "comp":f"<b>{hourly_share:.1f}%</b> of postings are hourly.<br><br>Median hourly pay: <b>{median_hourly:,.2f}</b>; median fixed budget: <b>{median_fixed:,.2f}</b>.<br><br>Observed pay coverage: <b>{original_pay_coverage:.2f}%</b>; model-ready: <b>{model_pay_coverage:.2f}%</b>.",
                "jobs":f"<b>{top_family['job_family']}</b> leads demand with <b>{int(top_family['Postings']):,}</b> postings.<br><br><b>{eda['job_family'].ne('Other').mean()*100:.2f}%</b> classify beyond Other.",
                "country":f"<b>{top_country['country_clean']}</b> is the largest market with <b>{int(top_country['Postings']):,}</b> postings.<br><br>Top five markets represent <b>{country_eda.head(5)['Share %'].sum():.2f}%</b> of demand.",
                "time":f"<b>{temporal_eda.loc[temporal_eda['Postings'].idxmax(),'day_of_week']}</b> has the highest activity.<br><br><b>{eda['dense_window_flag'].mean()*100:.2f}%</b> of records are in the dense period.",
                "outliers":f"<b>{eda['hourly_outlier_flag'].sum():,}</b> hourly and <b>{eda['fixed_outlier_flag'].sum():,}</b> fixed-price records are flagged.<br><br>Original values remain preserved."
            }
            display(HTML(f'<div class="txt6">{text[f]}</div>'))

        elif v=="findings":
            top=job_family_eda.iloc[0]
            peak=temporal_eda.loc[temporal_eda["Postings"].idxmax(),"day_of_week"]

            findings={
                "comp":[f"Hourly work represents {hourly_share:.1f}% of postings.",
                        f"Median hourly midpoint is {median_hourly:,.2f}.",
                        f"Median fixed budget is {median_fixed:,.2f}.",
                        f"{original_pay_coverage:.1f}% have observed compensation."],
                "jobs":[f"{top['job_family']} is the largest job family.",
                        f"It contains {int(top['Postings']):,} postings.",
                        f"{eda['job_family'].ne('Other').mean()*100:.1f}% classify beyond Other.",
                        "Demand and compensation can be compared by family."],
                "country":[f"{country_eda.iloc[0]['country_clean']} is the largest market.",
                           f"Top five markets account for {country_eda.head(5)['Share %'].sum():.1f}%.",
                           f"{int(country_market['country_sample_eligible'].sum())} markets have N ≥ 100.",
                           "Market concentration matters in comparisons."],
                "time":[f"{peak} has the highest weekday activity.",
                        f"{eda['dense_window_flag'].mean()*100:.1f}% of records are in the dense period.",
                        f"Weekend postings represent {eda['is_weekend'].mean()*100:.1f}%.",
                        "Trend modelling should use the dense window."],
                "outliers":[f"{int(eda['hourly_outlier_flag'].sum()):,} hourly outliers.",
                            f"{int(eda['fixed_outlier_flag'].sum()):,} fixed-price outliers.",
                            "Outliers remain preserved.",
                            "Capped fields support sensitivity testing."]
            }

            display(HTML('<div class="txt6">'+"".join(
                f"<b>{i+1:02d}.</b> {x}<br><br>" for i,x in enumerate(findings[f])
            )+"</div>"))

        elif v=="recommendations":
            recs={
                "comp":["Report hourly and fixed compensation separately.",
                        "Use original pay for descriptive conclusions.",
                        "Use imputed pay only when models require complete inputs.",
                        "Use capped pay for robustness checks."],
                "jobs":["Use job_family for emerging-category analysis.",
                        "Review frequent titles classified as Other.",
                        "Analyse demand growth by major job family.",
                        "Do not interpret title categories as verified skills."],
                "country":["Apply N ≥ 100 for broad country comparison.",
                           "Use pay-specific eligibility for compensation rankings.",
                           "Use country_code for Task 4 mapping.",
                           "Flag imputed geography in sensitive conclusions."],
                "time":["Use the dense window beginning 13 February 2024.",
                        "Exclude the partial final day from complete-day analysis.",
                        "Investigate low-volume days before forecasting.",
                        "Prefer short-horizon forecasting."],
                "outliers":["Do not globally delete extreme values.",
                            "Compare original and capped results.",
                            "Prefer robust statistics such as medians.",
                            "Document model-specific outlier handling."]
            }

            display(HTML('<div class="txt6"><b>Recommended next actions</b><br><br>'+
                "".join(f"<b>{i+1:02d}.</b> {x}<br><br>" for i,x in enumerate(recs[f]))+
                "</div>"))

        elif v=="interpretation":
            text={
                "comp":"Hourly and fixed-price compensation measure different quantities and should remain separate.",
                "jobs":"Job families are title-derived analytical groups, not verified employer categories.",
                "country":"Country conclusions require sample-size controls because market volumes differ.",
                "time":"Temporal conclusions should emphasize the dense February–March period.",
                "outliers":"Preserving originals plus robust alternatives avoids premature information loss."
            }
            display(HTML(f'<div class="txt6"><b>Analytical interpretation</b><br><br>{text[f]}</div>'))

        else:
            methods={
                "comp":"Original valid hourly_midpoint and fixed_budget drive descriptive analysis; imputed helpers remain separate.",
                "jobs":"Job families and seniority are derived from normalized title keywords using transparent rules.",
                "country":"Countries are standardized, link-recovered where possible, then controlled fallback and eligibility thresholds are applied.",
                "time":"Calendar features use UTC timestamps with dense-window and partial-final-day controls.",
                "outliers":"IQR identifies extremes; P01–P99 capped fields support robustness analysis."
            }
            display(HTML(f'<div class="txt6"><b>Methodology</b><br><br>{methods[f]}</div>'))

for w in [metric,basis,top_n,market_filter,view]:
    w.observe(update_dashboard,names="value")

focus.observe(set_controls,names="value")
focus.observe(update_dashboard,names="value")
set_controls()
update_dashboard()

dashboard_06=widgets.VBox([
    header,kpi_box,
    widgets.HTML('<div class="ctrl6"><b style="font:11px Arial;color:#22223b">INTERACTIVE ANALYSIS CONTROLS</b></div>'),
    controls,view,output
])
dashboard_06.add_class("d6")
display(dashboard_06)

for name,data in {
    "eda_compensation_summary.csv":compensation_eda,
    "eda_job_family_summary.csv":job_family_eda,
    "eda_country_summary.csv":country_eda,
    "eda_temporal_summary.csv":temporal_eda,
    "eda_outlier_summary.csv":outlier_eda
}.items():
    data.to_csv(EXPORT_DIR/name,index=False)
```

```python
if "df_final" not in globals():
    raise NameError("Run Section 05.3 first.")

task1=df_final.reset_index(drop=True).copy()
task1["job_id"]=np.arange(len(task1))

stopwords={
    "the","and","for","with","from","that","this","you","your","our","are","into",
    "using","need","needed","looking","wanted","required","job","work","project",
    "expert","professional","specialist","help","someone","create","based"
}

tokens=(task1[["job_id","title_normalized"]]
    .assign(keyword=lambda x:x["title_normalized"].fillna("").str.findall(r"\b[a-z][a-z0-9+#.-]{2,}\b"))
    .explode("keyword").dropna(subset=["keyword"]))
tokens=tokens[~tokens["keyword"].isin(stopwords)].drop_duplicates(["job_id","keyword"])

def keyword_stats(col):
    pay=task1[["job_id",col]].dropna().rename(columns={col:"pay"})
    x=tokens.merge(pay,on="job_id")
    n=len(pay); total=pay["pay"].sum(); median=pay["pay"].median(); sd=pay["pay"].std(ddof=0)

    g=(x.groupby("keyword")["pay"]
       .agg(Count="count",Mean_With="mean",Median_With="median",Sum_With="sum")
       .reset_index())

    g["Count_Without"]=n-g["Count"]
    g["Mean_Without"]=(total-g["Sum_With"])/g["Count_Without"].replace(0,np.nan)
    g["Correlation"]=(g["Mean_With"]-g["Mean_Without"])/sd*np.sqrt(g["Count"]*g["Count_Without"])/n
    g["Median_Lift"]=g["Median_With"]-median
    g["Median_Lift_%"]=np.where(median!=0,g["Median_Lift"]/median*100,np.nan)
    g["Sample_%"]=g["Count"]/n*100
    g["Abs_Correlation"]=g["Correlation"].abs()
    return g.sort_values("Abs_Correlation",ascending=False)

pay_cols={
    ("hourly","original"):"hourly_midpoint",
    ("hourly","capped"):"hourly_midpoint_capped",
    ("fixed","original"):"fixed_budget",
    ("fixed","capped"):"fixed_budget_capped"
}
kw_tables={k:keyword_stats(v) for k,v in pay_cols.items()}

refresh_time=pd.Timestamp.now(tz="Asia/Kolkata").strftime("%d %b %Y • %I:%M %p")

def card7(a,b,c):
    return f'<div class="c7"><small>{a}</small><b>{b}</b><span>{c}</span></div>'

def table7(df,n=None):
    display(HTML((df.head(n) if n else df).to_html(
        index=False,border=0,classes="t7",float_format=lambda x:f"{x:,.3f}"
    )))

display(HTML("""
<style>
.d7{border:1px solid #ded8d5;border-radius:15px;overflow:hidden;background:#fff;box-shadow:0 6px 20px #22223b14}
.h7{background:linear-gradient(120deg,#22223b,#30324d,#4a4e69);padding:22px 25px;display:flex;justify-content:space-between;font-family:Arial}
.h7 .k{color:#c9ada7;font-size:8px;font-weight:700;letter-spacing:1.5px}.h7 h2{color:#fff;margin:5px 0;font-size:23px}.h7 p,.h7 .m{color:#f2e9e4;font-size:10px;line-height:1.6}
.cards7{display:grid;grid-template-columns:repeat(6,1fr);border-bottom:1px solid #e8e2df}.c7{padding:13px;border-right:1px solid #ebe6e4;font-family:Arial;min-width:0}
.c7 small{display:block;color:#9a8c98;font-size:7px;font-weight:700}.c7 b{display:block;color:#22223b;font-size:17px;margin-top:5px;overflow-wrap:anywhere}.c7 span{display:block;color:#4a4e69;font-size:8px;margin-top:3px}
.ctrl7{padding:12px 18px 3px}.btn7{margin:7px 18px!important}.out7{border-top:1px solid #eee9e7!important}
.t7{width:100%;border-collapse:collapse;font:11px Arial}.t7 th{background:#22223b;color:#fff;padding:9px;text-align:left}.t7 td{padding:9px;border-bottom:1px solid #eee9e7;color:#22223b}
.txt7{font:12px/1.8 Arial;color:#22223b}.info7{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}
.box7{background:#f8f6f5;border-left:4px solid #4a4e69;padding:14px;border-radius:7px;font:11px/1.6 Arial;color:#22223b}.box7 b{display:block;color:#4a4e69;font-size:9px;margin-bottom:5px}
</style>
"""))

header7=widgets.HTML(f"""
<div class="h7">
<div><div class="k">TASK 01 • KEYWORD–COMPENSATION ANALYSIS</div>
<h2>Title Keyword Compensation Intelligence</h2>
<p>Frequency • Word Cloud • Pay Lift • Correlation • Robustness</p></div>
<div class="m">RECORDS<br><b>{len(task1):,}</b><br><br>UPDATED<br><b>{refresh_time} IST</b></div>
</div>
""")

pay_type=widgets.Dropdown(
    options=[("Hourly Rate","hourly"),("Fixed Budget","fixed")],
    value="hourly",description="Pay Type:",layout=widgets.Layout(width="260px")
)
pay_basis=widgets.Dropdown(
    options=[("Original","original"),("Robust / Capped","capped")],
    value="original",description="Pay View:",layout=widgets.Layout(width="270px")
)
metric7=widgets.Dropdown(
    options=[("Correlation","corr"),("Median Pay Lift %","lift"),
             ("Median Pay","median"),("Keyword Frequency","frequency")],
    value="corr",description="Metric:",layout=widgets.Layout(width="280px")
)
min_n=widgets.IntSlider(
    value=100,min=25,max=1000,step=25,description="Min N:",
    continuous_update=False,layout=widgets.Layout(width="280px")
)
top_n7=widgets.IntSlider(
    value=15,min=5,max=30,step=5,description="Top N:",
    continuous_update=False,layout=widgets.Layout(width="270px")
)
keyword_pick=widgets.Dropdown(description="Keyword:",layout=widgets.Layout(width="280px"))

view7=widgets.ToggleButtons(
    options=[("OVERVIEW","overview"),("CHART","chart"),("WORD CLOUD","cloud"),
             ("DATA","table"),("KEYWORD DETAIL","detail"),("FINDINGS","findings"),
             ("RECOMMENDATIONS","recommendations"),("INTERPRETATION","interpretation"),
             ("METHOD","method")],
    value="chart",layout=widgets.Layout(width="100%"),
    style={"button_width":"11%","font_weight":"700"}
)
view7.add_class("btn7")

controls7=widgets.HBox(
    [pay_type,pay_basis,metric7,min_n,top_n7,keyword_pick],
    layout=widgets.Layout(padding="8px 18px",flex_flow="row wrap",gap="10px")
)

kpi7=widgets.HTML()
output7=widgets.Output(layout=widgets.Layout(padding="20px 23px"))
output7.add_class("out7")

def current_stats():
    return kw_tables[(pay_type.value,pay_basis.value)].query("Count>=@min_n.value").copy()

def update_keywords(*_):
    d=current_stats()
    keyword_pick.options=d.nlargest(100,"Abs_Correlation")["keyword"].tolist() or ["No eligible keyword"]

def update7(*_):
    d=current_stats()

    with output7:
        clear_output(wait=True)

        if d.empty:
            kpi7.value='<div class="cards7">'+card7("STATUS","NO DATA","Lower Min N")+"</div>"
            display(HTML('<div class="txt7">No keywords meet the selected sample threshold.</div>'))
            return

        pos=d.nlargest(1,"Correlation").iloc[0]
        neg=d.nsmallest(1,"Correlation").iloc[0]
        col=pay_cols[(pay_type.value,pay_basis.value)]
        overall=task1[col].median()

        kpi7.value='<div class="cards7">'+"".join([
            card7("ELIGIBLE KEYWORDS",f"{len(d):,}",f"N ≥ {min_n.value}"),
            card7("STRONGEST +",pos["keyword"],f"r = {pos['Correlation']:.3f}"),
            card7("STRONGEST −",neg["keyword"],f"r = {neg['Correlation']:.3f}"),
            card7("MEDIAN PAY",f"{overall:,.2f}",pay_basis.label),
            card7("MAX SAMPLE",f"{int(d['Count'].max()):,}","Keyword jobs"),
            card7("PAY TYPE",pay_type.label,"Current analysis")
        ])+"</div>"

        v=view7.value

        if v=="overview":
            display(HTML("""
            <div class="info7">
            <div class="box7"><b>QUESTION</b>Which title keywords are associated with higher or lower compensation?</div>
            <div class="box7"><b>EVIDENCE</b>Frequency, median pay lift and point-biserial correlation.</div>
            <div class="box7"><b>CONTROL</b>Hourly and fixed compensation remain separate with minimum sample thresholds.</div>
            </div>
            """))

        elif v=="chart":
            m=metric7.value

            if m=="corr":
                x=d.nlargest(top_n7.value,"Abs_Correlation").sort_values("Correlation")
                fig=go.Figure(go.Scatter(
                    x=x["Correlation"],y=x["keyword"],mode="markers",
                    marker=dict(size=np.sqrt(x["Count"])*1.3,color="#4a4e69",opacity=.8,
                                line=dict(color="#22223b",width=1)),
                    text=x["Count"],customdata=x["Median_Lift_%"],
                    hovertemplate="<b>%{y}</b><br>r: %{x:.3f}<br>N: %{text:,}<br>Median lift: %{customdata:.2f}%<extra></extra>"
                ))
                fig.add_vline(x=0,line_color="#9a8c98")
                fig.update_xaxes(title="Point-Biserial Correlation")

            elif m=="lift":
                x=d.loc[d["Median_Lift_%"].abs().nlargest(top_n7.value).index].sort_values("Median_Lift_%")
                fig=go.Figure(go.Bar(
                    x=x["Median_Lift_%"],y=x["keyword"],orientation="h",
                    marker_color="#4a4e69"
                ))
                fig.add_vline(x=0,line_color="#9a8c98")
                fig.update_xaxes(title="Median Pay Lift (%)")

            elif m=="median":
                x=d.nlargest(top_n7.value,"Median_With").sort_values("Median_With")
                fig=go.Figure(go.Scatter(
                    x=x["Median_With"],y=x["keyword"],mode="markers",
                    marker=dict(size=np.sqrt(x["Count"])*1.2,color="#9a8c98",opacity=.8),
                    text=x["Count"],
                    hovertemplate="<b>%{y}</b><br>Median: %{x:,.2f}<br>N: %{text:,}<extra></extra>"
                ))
                fig.update_xaxes(title="Median Compensation")

            else:
                x=d.nlargest(top_n7.value,"Count").sort_values("Count")
                fig=go.Figure(go.Bar(
                    x=x["Count"],y=x["keyword"],orientation="h",
                    marker_color="#4a4e69"
                ))
                fig.update_xaxes(title="Keyword Posting Count")

            fig.update_layout(
                height=500,paper_bgcolor="#fff",plot_bgcolor="#fff",
                font=dict(family="Arial",color="#22223b"),
                margin=dict(l=120,r=30,t=25,b=45),
                hoverlabel=dict(bgcolor="#22223b",font_color="#fff")
            )
            display(fig)

        elif v=="cloud":
            x=d.nlargest(min(100,len(d)),"Count")
            freq=dict(zip(x["keyword"],x["Count"]))
            corr=dict(zip(x["keyword"],x["Correlation"]))

            def wc_color(word,**kwargs):
                r=corr.get(word,0)
                return "#4a4e69" if r>.02 else "#c9ada7" if r<-.02 else "#9a8c98"

            wc=WordCloud(
                width=1200,height=520,background_color="white",
                prefer_horizontal=.9,random_state=42,
                color_func=wc_color
            ).generate_from_frequencies(freq)

            fig,ax=plt.subplots(figsize=(13,5.5))
            ax.imshow(wc,interpolation="bilinear")
            ax.axis("off")
            plt.tight_layout()
            plt.show()

            display(HTML("""
            <div class="txt7" style="text-align:center">
            <b style="color:#4a4e69">Positive association</b> &nbsp;•&nbsp;
            <b style="color:#c9ada7">Negative association</b> &nbsp;•&nbsp;
            <b style="color:#9a8c98">Near-neutral</b><br>
            Word size represents keyword frequency.
            </div>
            """))

        elif v=="table":
            table7(d.sort_values("Abs_Correlation",ascending=False)[
                ["keyword","Count","Sample_%","Median_With","Median_Lift",
                 "Median_Lift_%","Correlation","Mean_With","Mean_Without"]
            ],top_n7.value)

        elif v=="detail":
            x=d[d["keyword"].eq(keyword_pick.value)]

            if len(x):
                x=x.iloc[0]
                display(HTML(f"""
                <div class="info7">
                <div class="box7"><b>KEYWORD</b>{x['keyword']}<br>N = {int(x['Count']):,} ({x['Sample_%']:.2f}%)</div>
                <div class="box7"><b>PAY EFFECT</b>Median: {x['Median_With']:,.2f}<br>Lift: {x['Median_Lift_%']:.2f}%</div>
                <div class="box7"><b>RELATIONSHIP</b>Correlation: {x['Correlation']:.3f}<br>Mean without: {x['Mean_Without']:,.2f}</div>
                </div>
                """))

        elif v=="findings":
            pos=d.nlargest(5,"Correlation")
            neg=d.nsmallest(5,"Correlation")

            display(HTML(
                '<div class="txt7"><b>Strongest positive associations</b><br>'+
                "<br>".join(f"{r.keyword}: r={r.Correlation:.3f}, N={int(r.Count):,}" for r in pos.itertuples())+
                '<br><br><b>Strongest negative associations</b><br>'+
                "<br>".join(f"{r.keyword}: r={r.Correlation:.3f}, N={int(r.Count):,}" for r in neg.itertuples())+
                "</div>"
            ))

        elif v=="recommendations":
            display(HTML("""
            <div class="txt7">
            <b>01.</b> Prioritize keywords with both meaningful association and sufficient volume.<br><br>
            <b>02.</b> Use the word cloud for market prominence, not as evidence of salary impact.<br><br>
            <b>03.</b> Use correlation and pay-lift charts for compensation conclusions.<br><br>
            <b>04.</b> Keep hourly and fixed-price keyword effects separate.<br><br>
            <b>05.</b> Compare original and capped results before accepting extreme-pay relationships.<br><br>
            <b>06.</b> Validate important keyword effects within job families before client recommendations.
            </div>
            """))

        elif v=="interpretation":
            display(HTML("""
            <div class="txt7">
            <b>Correlation shows association, not causation.</b><br><br>
            Positive values indicate that postings containing the keyword tend to have higher compensation;
            negative values indicate lower compensation.<br><br>
            Keyword effects can also reflect job family, seniority, geography or project complexity.
            The word cloud visualizes prominence; it does not establish compensation effects.
            </div>
            """))

        else:
            display(HTML("""
            <div class="txt7">
            <b>Extraction:</b> normalized titles are tokenized and generic stopwords removed.<br><br>
            <b>Deduplication:</b> each keyword counts once per posting.<br><br>
            <b>Frequency:</b> word-cloud size represents the number of eligible postings containing the keyword.<br><br>
            <b>Correlation:</b> point-biserial correlation measures binary keyword presence against continuous pay.<br><br>
            <b>Pay lift:</b> keyword median compensation is compared with the overall median.<br><br>
            <b>Reliability:</b> Min N prevents tiny keyword groups from dominating results.<br><br>
            <b>Robustness:</b> original and P01–P99 capped compensation can be compared.
            </div>
            """))

for w in [pay_type,pay_basis,min_n]:
    w.observe(update_keywords,names="value")
    w.observe(update7,names="value")

for w in [metric7,top_n7,keyword_pick,view7]:
    w.observe(update7,names="value")

update_keywords()
update7()

dashboard_07=widgets.VBox([
    header7,kpi7,
    widgets.HTML('<div class="ctrl7"><b style="font:11px Arial;color:#22223b">TASK 01 ANALYSIS CONTROLS</b></div>'),
    controls7,view7,output7
])
dashboard_07.add_class("d7")
display(dashboard_07)

for (ptype,basis_name),data in kw_tables.items():
    data.to_csv(EXPORT_DIR/f"task1_keywords_{ptype}_{basis_name}.csv",index=False)
```

```python
if "df_final" not in globals():
    raise NameError("Run Section 05.3 first.")

task2=df_final.copy()
task2["published_day"]=pd.to_datetime(task2["published_day"],utc=True)
last_ts=pd.to_datetime(task2["published_date"],utc=True).max()
last_day=last_ts.normalize()
trend_end=last_day-pd.Timedelta(days=1) if last_ts.hour<23 else last_day
dense_start=pd.Timestamp("2024-02-13",tz="UTC")
t2=task2[task2["published_day"].between(dense_start,trend_end)].copy()
t2["week_start"]=t2["published_day"]-pd.to_timedelta(t2["published_day"].dt.dayofweek,unit="D")
recent_start=trend_end-pd.Timedelta(days=13)
prev_end=recent_start-pd.Timedelta(days=1)
prev_start=prev_end-pd.Timedelta(days=13)
recent=t2[t2["published_day"].between(recent_start,trend_end)].groupby("job_family").size()
previous=t2[t2["published_day"].between(prev_start,prev_end)].groupby("job_family").size()
category_trends=pd.DataFrame({"Recent_Postings":recent,"Previous_Postings":previous}).fillna(0).astype(int)
category_trends["Volume_Change"]=category_trends["Recent_Postings"]-category_trends["Previous_Postings"]
category_trends["Growth_%"]=np.where(
    category_trends["Previous_Postings"]>0,
    category_trends["Volume_Change"]/category_trends["Previous_Postings"]*100,np.nan
)

rt=category_trends["Recent_Postings"].sum()
pt=category_trends["Previous_Postings"].sum()
category_trends["Recent_Share_%"]=category_trends["Recent_Postings"]/rt*100
category_trends["Previous_Share_%"]=category_trends["Previous_Postings"]/pt*100
category_trends["Share_Change_pp"]=category_trends["Recent_Share_%"]-category_trends["Previous_Share_%"]
category_trends["Recent_Avg_Day"]=category_trends["Recent_Postings"]/14
category_trends["Previous_Avg_Day"]=category_trends["Previous_Postings"]/14
g=category_trends["Growth_%"]
category_trends["Status"]=np.select(
    [
        (category_trends["Previous_Postings"]==0)&(category_trends["Recent_Postings"]>0),
        (g>=20)&(category_trends["Share_Change_pp"]>0),
        g>5,g<-5
    ],
    ["New / Re-emerging","Emerging","Growing","Declining"],
    default="Stable"
)

category_trends=category_trends.reset_index().sort_values("Recent_Postings",ascending=False)

weekly_trends=(t2.groupby(["week_start","job_family"]).size()
    .rename("Postings").reset_index())

refresh_time=pd.Timestamp.now(tz="Asia/Kolkata").strftime("%d %b %Y • %I:%M %p")

def card8(a,b,c):
    return f'<div class="c8"><small>{a}</small><b>{b}</b><span>{c}</span></div>'

def table8(df,n=None):
    display(HTML((df.head(n) if n else df).to_html(
        index=False,border=0,classes="t8",float_format=lambda x:f"{x:,.2f}"
    )))

display(HTML("""
<style>
.d8{border:1px solid #ded8d5;border-radius:15px;overflow:hidden;background:#fff;box-shadow:0 6px 20px #22223b14}
.h8{background:linear-gradient(120deg,#22223b,#30324d,#4a4e69);padding:22px 25px;display:flex;justify-content:space-between;font-family:Arial}
.h8 .k{color:#c9ada7;font-size:8px;font-weight:700;letter-spacing:1.5px}.h8 h2{color:#fff;margin:5px 0;font-size:23px}.h8 p,.h8 .m{color:#f2e9e4;font-size:10px;line-height:1.6}
.cards8{display:grid;grid-template-columns:repeat(6,1fr);border-bottom:1px solid #e8e2df}.c8{padding:13px;border-right:1px solid #ebe6e4;font-family:Arial;min-width:0}
.c8 small{display:block;color:#9a8c98;font-size:7px;font-weight:700}.c8 b{display:block;color:#22223b;font-size:17px;margin-top:5px;overflow-wrap:anywhere}.c8 span{display:block;color:#4a4e69;font-size:8px;margin-top:3px}
.ctrl8{padding:12px 18px 3px}.btn8{margin:7px 18px!important}.out8{border-top:1px solid #eee9e7!important}
.t8{width:100%;border-collapse:collapse;font:11px Arial}.t8 th{background:#22223b;color:#fff;padding:9px;text-align:left}.t8 td{padding:9px;border-bottom:1px solid #eee9e7;color:#22223b}
.txt8{font:12px/1.8 Arial;color:#22223b}.info8{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}
.box8{background:#f8f6f5;border-left:4px solid #4a4e69;padding:14px;border-radius:7px;font:11px/1.6 Arial;color:#22223b}.box8 b{display:block;color:#4a4e69;font-size:9px;margin-bottom:5px}
</style>
"""))

header8=widgets.HTML(f"""
<div class="h8">
<div><div class="k">TASK 02 • EMERGING CATEGORY ANALYSIS</div>
<h2>Job Category Demand Intelligence</h2>
<p>Growth • Market Share • Weekly Trends • Emerging Categories • Demand Momentum</p></div>
<div class="m">DENSE WINDOW<br><b>{dense_start:%d %b} – {trend_end:%d %b %Y}</b><br><br>UPDATED<br><b>{refresh_time} IST</b></div>
</div>
""")

chart8=widgets.Dropdown(
    options=[("Weekly Demand Trend","trend"),("Growth vs Volume","matrix"),
             ("Market Share Shift","share")],
    value="trend",description="Chart:",layout=widgets.Layout(width="300px")
)

min_n8=widgets.IntSlider(
    value=100,min=25,max=1000,step=25,description="Min Recent:",
    continuous_update=False,layout=widgets.Layout(width="300px")
)

top_n8=widgets.IntSlider(
    value=5,min=3,max=10,step=1,description="Top N:",
    continuous_update=False,layout=widgets.Layout(width="270px")
)

status8=widgets.Dropdown(
    options=["All","Emerging","Growing","Stable","Declining","New / Re-emerging"],
    value="All",description="Status:",layout=widgets.Layout(width="270px")
)

category8=widgets.Dropdown(description="Category:",layout=widgets.Layout(width="300px"))

view8=widgets.ToggleButtons(
    options=[("OVERVIEW","overview"),("CHART","chart"),("DATA","table"),
             ("CATEGORY DETAIL","detail"),("FINDINGS","findings"),
             ("RECOMMENDATIONS","recommendations"),
             ("INTERPRETATION","interpretation"),("METHOD","method")],
    value="chart",layout=widgets.Layout(width="100%"),
    style={"button_width":"12%","font_weight":"700"}
)
view8.add_class("btn8")

controls8=widgets.HBox(
    [chart8,min_n8,top_n8,status8,category8],
    layout=widgets.Layout(padding="8px 18px",flex_flow="row wrap",gap="10px")
)

kpi8=widgets.HTML()
output8=widgets.Output(layout=widgets.Layout(padding="20px 23px"))
output8.add_class("out8")

def current8():
    d=category_trends[category_trends["Recent_Postings"]>=min_n8.value].copy()
    if status8.value!="All":
        d=d[d["Status"].eq(status8.value)]
    return d

def update_categories(*_):
    d=current8()
    category8.options=d["job_family"].tolist() or ["No eligible category"]

def update8(*_):
    d=current8()

    with output8:
        clear_output(wait=True)

        if d.empty:
            kpi8.value='<div class="cards8">'+card8("STATUS","NO DATA","Adjust filters")+"</div>"
            display(HTML('<div class="txt8">No categories meet the selected controls.</div>'))
            return

        emerging=d[(d["Growth_%"]>0)&(d["Share_Change_pp"]>0)].sort_values("Growth_%",ascending=False)
        top=emerging.iloc[0] if len(emerging) else d.iloc[0]

        kpi8.value='<div class="cards8">'+"".join([
            card8("ELIGIBLE",f"{len(d):,}",f"N ≥ {min_n8.value}"),
            card8("TOP EMERGING",top["job_family"],"Current filters"),
            card8("GROWTH",f"{top['Growth_%']:.1f}%" if pd.notna(top["Growth_%"]) else "NEW","Recent vs prior"),
            card8("RECENT JOBS",f"{int(top['Recent_Postings']):,}","14 days"),
            card8("SHARE SHIFT",f"{top['Share_Change_pp']:+.2f} pp","Market share"),
            card8("WINDOW","14 vs 14","Days compared")
        ])+"</div>"

        v=view8.value

        if v=="overview":
            display(HTML("""
            <div class="info8">
            <div class="box8"><b>QUESTION</b>Which job categories are gaining or losing demand?</div>
            <div class="box8"><b>EVIDENCE</b>Recent volume, prior volume, growth rate, share change and weekly posting trends.</div>
            <div class="box8"><b>CONTROL</b>Only the reliable dense period is used and the partial final day is excluded.</div>
            </div>
            """))

        elif v=="chart":

            if chart8.value=="trend":
                cats=(emerging if len(emerging) else d).head(top_n8.value)["job_family"]
                x=weekly_trends[weekly_trends["job_family"].isin(cats)]
                palette=["#22223b","#4a4e69","#9a8c98","#c9ada7","#6d6875",
                         "#8d99ae","#b5838d","#7f8c8d","#a5a58d","#6b705c"]

                fig=go.Figure()
                for i,c in enumerate(cats):
                    z=x[x["job_family"].eq(c)]
                    fig.add_trace(go.Scatter(
                        x=z["week_start"],y=z["Postings"],name=c,
                        mode="lines",stackgroup="one",
                        line=dict(width=1.5,color=palette[i%len(palette)]),
                        hovertemplate=f"<b>{c}</b><br>%{{x|%d %b}}<br>Postings: %{{y:,}}<extra></extra>"
                    ))
                fig.update_yaxes(title="Weekly Postings")

            elif chart8.value=="matrix":
                x=d.dropna(subset=["Growth_%"])
                fig=go.Figure(go.Scatter(
                    x=x["Recent_Postings"],y=x["Growth_%"],mode="markers",
                    marker=dict(size=np.sqrt(x["Recent_Postings"])*1.6,
                                color="#4a4e69",opacity=.75,
                                line=dict(color="#22223b",width=1)),
                    text=x["job_family"],customdata=x["Share_Change_pp"],
                    hovertemplate="<b>%{text}</b><br>Recent: %{x:,}<br>Growth: %{y:.1f}%<br>Share shift: %{customdata:+.2f} pp<extra></extra>"
                ))
                fig.add_hline(y=0,line_color="#9a8c98")
                fig.update_xaxes(title="Recent 14-Day Postings")
                fig.update_yaxes(title="Growth (%)")

            else:
                x=d.reindex(d["Share_Change_pp"].abs().nlargest(top_n8.value).index).sort_values("Share_Change_pp")
                fig=go.Figure(go.Bar(
                    x=x["Share_Change_pp"],y=x["job_family"],orientation="h",
                    marker_color="#4a4e69",
                    hovertemplate="<b>%{y}</b><br>Share shift: %{x:+.2f} pp<extra></extra>"
                ))
                fig.add_vline(x=0,line_color="#9a8c98")
                fig.update_xaxes(title="Market Share Change (percentage points)")

            fig.update_layout(
                height=500,paper_bgcolor="#fff",plot_bgcolor="#fff",
                font=dict(family="Arial",color="#22223b"),
                margin=dict(l=110,r=30,t=25,b=45),
                hoverlabel=dict(bgcolor="#22223b",font_color="#fff")
            )
            display(fig)

        elif v=="table":
            table8(d.sort_values(["Growth_%","Recent_Postings"],ascending=False)[
                ["job_family","Status","Recent_Postings","Previous_Postings",
                 "Volume_Change","Growth_%","Recent_Share_%",
                 "Previous_Share_%","Share_Change_pp"]
            ])

        elif v=="detail":
            x=category_trends[category_trends["job_family"].eq(category8.value)]
            if len(x):
                x=x.iloc[0]
                display(HTML(f"""
                <div class="info8">
                <div class="box8"><b>CATEGORY</b>{x['job_family']}<br>Status: {x['Status']}</div>
                <div class="box8"><b>DEMAND</b>Recent: {int(x['Recent_Postings']):,}<br>Previous: {int(x['Previous_Postings']):,}<br>Change: {int(x['Volume_Change']):+,}</div>
                <div class="box8"><b>MOMENTUM</b>Growth: {x['Growth_%']:.2f}%<br>Share shift: {x['Share_Change_pp']:+.2f} pp</div>
                </div>
                """))

        elif v=="findings":
            grow=d.dropna(subset=["Growth_%"]).nlargest(5,"Growth_%")
            fall=d.dropna(subset=["Growth_%"]).nsmallest(5,"Growth_%")

            display(HTML(
                '<div class="txt8"><b>Fastest-growing categories</b><br>'+
                "<br>".join(f"{r.job_family}: {r._6:+.1f}% ({int(r.Recent_Postings):,} recent postings)"
                           for r in grow.itertuples())+
                '<br><br><b>Fastest-declining categories</b><br>'+
                "<br>".join(f"{r.job_family}: {r._6:+.1f}% ({int(r.Recent_Postings):,} recent postings)"
                           for r in fall.itertuples())+
                "</div>"
            ))

        elif v=="recommendations":
            display(HTML("""
            <div class="txt8">
            <b>01.</b> Prioritize categories showing both positive growth and positive market-share movement.<br><br>
            <b>02.</b> Require meaningful recent volume before labelling a category as emerging.<br><br>
            <b>03.</b> Track high-growth categories weekly rather than relying on one comparison window.<br><br>
            <b>04.</b> Use emerging categories as candidates for Task 3 demand forecasting.<br><br>
            <b>05.</b> Combine category growth with compensation insights from Task 1 when building recommendations.<br><br>
            <b>06.</b> Treat newly appearing categories separately from categories with measurable prior-period growth.
            </div>
            """))

        elif v=="interpretation":
            display(HTML("""
            <div class="txt8">
            <b>Growth alone does not define an emerging market.</b><br><br>
            A category is more convincing when recent posting volume is substantial, growth is positive and its share of total demand is also increasing.<br><br>
            Short observation history means these results describe recent momentum rather than long-term labour-market trends.
            </div>
            """))

        else:
            display(HTML(f"""
            <div class="txt8">
            <b>Analysis window:</b> {dense_start:%d %b %Y} to {trend_end:%d %b %Y}; the incomplete final day is excluded.<br><br>
            <b>Comparison:</b> most recent 14 complete days versus the preceding 14 days.<br><br>
            <b>Growth:</b> percentage change in posting volume between the two windows.<br><br>
            <b>Share shift:</b> change in each category's percentage of total postings.<br><br>
            <b>Emerging rule:</b> ≥20% growth plus positive market-share movement; minimum volume remains user-controlled.<br><br>
            <b>Trend chart:</b> weekly posting frequency within the dense observation period.
            </div>
            """))

for w in [min_n8,status8]:
    w.observe(update_categories,names="value")
    w.observe(update8,names="value")

for w in [chart8,top_n8,category8,view8]:
    w.observe(update8,names="value")

update_categories()
update8()

dashboard_08=widgets.VBox([
    header8,kpi8,
    widgets.HTML('<div class="ctrl8"><b style="font:11px Arial;color:#22223b">TASK 02 ANALYSIS CONTROLS</b></div>'),
    controls8,view8,output8
])
dashboard_08.add_class("d8")
display(dashboard_08)

category_trends.to_csv(EXPORT_DIR/"task2_category_growth.csv",index=False)
weekly_trends.to_csv(EXPORT_DIR/"task2_weekly_category_trends.csv",index=False)

if "logger" in globals():
    logger.info(
        "TASK2_COMPLETE | dense_records=%s | categories=%s | trend_end=%s",
        f"{len(t2):,}",category_trends["job_family"].nunique(),trend_end.date()
    )
```

<div style="color:#f2e9e4;font-size:12px;line-height:1.65;margin-top:8px">Forecasting near-term job demand using daily category volumes, weekday seasonality and trend signals, with chronological holdout validation and comparison against a lag-7 benchmark.</div>

```python
if "df_final" not in globals():
    raise NameError("Run Section 05.3 first.")

task3=df_final.copy()
task3["published_day"]=pd.to_datetime(task3["published_day"],utc=True)
last_ts=pd.to_datetime(task3["published_date"],utc=True).max()
last_day=last_ts.normalize()
trend_end=last_day-pd.Timedelta(days=1) if last_ts.hour<23 else last_day
dense_start=pd.Timestamp("2024-02-13",tz="UTC")
t3=task3[task3["published_day"].between(dense_start,trend_end)].copy()
top_roles=t3["job_family"].value_counts().head(10).index.tolist()
dates=pd.date_range(dense_start,trend_end,freq="D",tz="UTC")

daily=(t3[t3["job_family"].isin(top_roles)]
       .groupby(["published_day","job_family"]).size()
       .rename("Actual").reset_index())

grid=pd.MultiIndex.from_product(
    [dates,top_roles],names=["published_day","job_family"]
).to_frame(index=False)

daily=grid.merge(daily,on=["published_day","job_family"],how="left").fillna({"Actual":0})
daily["Actual"]=daily["Actual"].astype(int)

def Xmat(d,start):
    t=(d-start).dt.days.to_numpy(dtype=float)
    dow=d.dt.dayofweek.to_numpy()
    return np.column_stack([np.ones(len(d)),t]+[(dow==i).astype(int) for i in range(6)])

horizon=7
validation=[]
forecasts=[]

for role,g in daily.groupby("job_family"):
    g=g.sort_values("published_day").reset_index(drop=True)
    train,test=g.iloc[:-horizon].copy(),g.iloc[-horizon:].copy()

    Xtr=Xmat(train["published_day"],train["published_day"].iloc[0])
    Xte=Xmat(test["published_day"],train["published_day"].iloc[0])
    beta=np.linalg.lstsq(Xtr,train["Actual"].to_numpy(),rcond=None)[0]
    pred=np.clip(Xte@beta,0,None)

    actual=test["Actual"].to_numpy()
    lag7=g["Actual"].shift(7).iloc[-horizon:].to_numpy()

    mae=np.mean(np.abs(actual-pred))
    baseline_mae=np.mean(np.abs(actual-lag7))
    wape=np.abs(actual-pred).sum()/max(actual.sum(),1)*100

    validation.append([
        role,round(mae,2),round(baseline_mae,2),round(wape,2),
        "Model" if mae<baseline_mae else "Lag-7"
    ])

    Xall=Xmat(g["published_day"],g["published_day"].iloc[0])
    beta=np.linalg.lstsq(Xall,g["Actual"].to_numpy(),rcond=None)[0]

    future_dates=pd.date_range(
        trend_end+pd.Timedelta(days=1),periods=horizon,freq="D",tz="UTC"
    )
    Xf=Xmat(pd.Series(future_dates),g["published_day"].iloc[0])
    fp=np.clip(Xf@beta,0,None)

    forecasts.append(pd.DataFrame({
        "job_family":role,
        "forecast_day":future_dates,
        "Forecast":fp,
        "Last_7d_Total":g["Actual"].tail(7).sum()
    }))

validation_df=pd.DataFrame(
    validation,
    columns=["job_family","MAE","Lag7_MAE","WAPE_%","Best_Method"]
)

forecast_daily=pd.concat(forecasts,ignore_index=True)

forecast_summary=(forecast_daily.groupby("job_family")
    .agg(Forecast_7d=("Forecast","sum"),
         Forecast_Avg_Day=("Forecast","mean"),
         Last_7d_Total=("Last_7d_Total","first"))
    .reset_index())

forecast_summary["Forecast_Change_%"]=np.where(
    forecast_summary["Last_7d_Total"]>0,
    (forecast_summary["Forecast_7d"]-forecast_summary["Last_7d_Total"])/
    forecast_summary["Last_7d_Total"]*100,np.nan
)

forecast_summary=(forecast_summary
    .merge(validation_df,on="job_family",how="left")
    .sort_values("Forecast_7d",ascending=False))

refresh_time=pd.Timestamp.now(tz="Asia/Kolkata").strftime("%d %b %Y • %I:%M %p")

def card9(a,b,c):
    return f'<div class="c9"><small>{a}</small><b>{b}</b><span>{c}</span></div>'

def table9(df,n=None):
    display(HTML((df.head(n) if n else df).to_html(
        index=False,border=0,classes="t9",
        float_format=lambda x:f"{x:,.2f}"
    )))

display(HTML("""
<style>
.d9{border:1px solid #ded8d5;border-radius:15px;overflow:hidden;background:#fff;box-shadow:0 6px 20px #22223b14}
.h9{background:linear-gradient(120deg,#22223b,#30324d,#4a4e69);padding:22px 25px;display:flex;justify-content:space-between;font-family:Arial}
.h9 .k{color:#c9ada7;font-size:8px;font-weight:700;letter-spacing:1.5px}.h9 h2{color:#fff;margin:5px 0;font-size:23px}.h9 p,.h9 .m{color:#f2e9e4;font-size:10px;line-height:1.6}
.cards9{display:grid;grid-template-columns:repeat(6,1fr);border-bottom:1px solid #e8e2df}.c9{padding:13px;border-right:1px solid #ebe6e4;font-family:Arial;min-width:0}
.c9 small{display:block;color:#9a8c98;font-size:7px;font-weight:700}.c9 b{display:block;color:#22223b;font-size:17px;margin-top:5px;overflow-wrap:anywhere}.c9 span{display:block;color:#4a4e69;font-size:8px;margin-top:3px}
.ctrl9{padding:12px 18px 3px}.btn9{margin:7px 18px!important}.out9{border-top:1px solid #eee9e7!important}
.t9{width:100%;border-collapse:collapse;font:11px Arial}.t9 th{background:#22223b;color:#fff;padding:9px;text-align:left}.t9 td{padding:9px;border-bottom:1px solid #eee9e7;color:#22223b}
.txt9{font:12px/1.8 Arial;color:#22223b}.info9{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}
.box9{background:#f8f6f5;border-left:4px solid #4a4e69;padding:14px;border-radius:7px;font:11px/1.6 Arial;color:#22223b}.box9 b{display:block;color:#4a4e69;font-size:9px;margin-bottom:5px}
</style>
"""))

header9=widgets.HTML(f"""
<div class="h9">
<div><div class="k">TASK 03 • DEMAND FORECASTING</div>
<h2>High-Demand Role Forecast Intelligence</h2>
<p>Daily Demand • Seasonality • 7-Day Forecast • Validation • Benchmarking</p></div>
<div class="m">TRAINING END<br><b>{trend_end:%d %b %Y}</b><br><br>UPDATED<br><b>{refresh_time} IST</b></div>
</div>
""")

chart9=widgets.Dropdown(
    options=[("Actual + Forecast","forecast"),("Forecast Ranking","ranking"),
             ("Validation Performance","validation")],
    value="forecast",description="Chart:",layout=widgets.Layout(width="300px")
)

role9=widgets.Dropdown(
    options=forecast_summary["job_family"].tolist(),
    description="Role:",layout=widgets.Layout(width="300px")
)

top_n9=widgets.IntSlider(
    value=5,min=3,max=10,step=1,description="Top N:",
    continuous_update=False,layout=widgets.Layout(width="270px")
)

view9=widgets.ToggleButtons(
    options=[("OVERVIEW","overview"),("CHART","chart"),("FORECAST","forecast"),
             ("VALIDATION","validation"),("DATA","data"),("FINDINGS","findings"),
             ("RECOMMENDATIONS","recommendations"),("METHOD","method")],
    value="chart",layout=widgets.Layout(width="100%"),
    style={"button_width":"12%","font_weight":"700"}
)
view9.add_class("btn9")

controls9=widgets.HBox(
    [chart9,role9,top_n9],
    layout=widgets.Layout(padding="8px 18px",flex_flow="row wrap",gap="10px")
)

kpi9=widgets.HTML()
output9=widgets.Output(layout=widgets.Layout(padding="20px 23px"))
output9.add_class("out9")

def update9(*_):
    with output9:
        clear_output(wait=True)

        top=forecast_summary.iloc[0]
        avg_mae=validation_df["MAE"].mean()
        beat=(validation_df["MAE"]<validation_df["Lag7_MAE"]).mean()*100

        kpi9.value='<div class="cards9">'+"".join([
            card9("FORECAST ROLES",len(forecast_summary),"Top families"),
            card9("TOP FORECAST",top["job_family"],"Next 7 days"),
            card9("FORECAST JOBS",f"{top['Forecast_7d']:,.0f}","7-day total"),
            card9("EXPECTED CHANGE",f"{top['Forecast_Change_%']:+.1f}%","vs last 7 days"),
            card9("AVG MAE",f"{avg_mae:.1f}","Holdout error"),
            card9("BEATS BASELINE",f"{beat:.0f}%","Roles")
        ])+"</div>"

        v=view9.value

        if v=="overview":
            display(HTML("""
            <div class="info9">
            <div class="box9"><b>QUESTION</b>Which high-demand job families are likely to receive the most postings next?</div>
            <div class="box9"><b>MODEL</b>Daily trend plus weekday seasonality fitted independently by job family.</div>
            <div class="box9"><b>VALIDATION</b>Latest seven complete days are held out and compared with a lag-7 benchmark.</div>
            </div>
            """))

        elif v=="chart":
            if chart9.value=="forecast":
                role=role9.value
                hist=daily[daily["job_family"].eq(role)]
                future=forecast_daily[forecast_daily["job_family"].eq(role)]

                fig=go.Figure()
                fig.add_trace(go.Scatter(
                    x=hist["published_day"],y=hist["Actual"],
                    mode="lines",name="Actual",
                    line=dict(color="#22223b",width=2)
                ))
                fig.add_trace(go.Scatter(
                    x=future["forecast_day"],y=future["Forecast"],
                    mode="lines+markers",name="Forecast",
                    line=dict(color="#c9ada7",width=3,dash="dash"),
                    marker=dict(size=7,color="#4a4e69")
                ))
                fig.add_vline(x=trend_end,line_color="#9a8c98",line_dash="dot")
                fig.update_yaxes(title="Daily Job Postings")

            elif chart9.value=="ranking":
                d=forecast_summary.head(top_n9.value).sort_values("Forecast_7d")

                fig=go.Figure(go.Bar(
                    x=d["Forecast_7d"],y=d["job_family"],
                    orientation="h",marker_color="#4a4e69",
                    customdata=d["Forecast_Change_%"],
                    hovertemplate="<b>%{y}</b><br>Forecast: %{x:,.0f}<br>Change: %{customdata:+.1f}%<extra></extra>"
                ))
                fig.update_xaxes(title="Forecast Postings — Next 7 Days")

            else:
                d=validation_df.sort_values("MAE")

                fig=go.Figure()
                fig.add_trace(go.Scatter(
                    x=d["MAE"],y=d["job_family"],mode="markers",
                    name="Model",marker=dict(size=11,color="#4a4e69")
                ))
                fig.add_trace(go.Scatter(
                    x=d["Lag7_MAE"],y=d["job_family"],mode="markers",
                    name="Lag-7",
                    marker=dict(size=11,color="#c9ada7",symbol="diamond")
                ))
                fig.update_xaxes(title="MAE — Lower is Better")

            fig.update_layout(
                height=500,paper_bgcolor="#fff",plot_bgcolor="#fff",
                font=dict(family="Arial",color="#22223b"),
                margin=dict(l=110,r=30,t=25,b=45),
                hoverlabel=dict(bgcolor="#22223b",font_color="#fff")
            )
            display(fig)

        elif v=="forecast":
            table9(forecast_summary[
                ["job_family","Forecast_7d","Forecast_Avg_Day","Last_7d_Total",
                 "Forecast_Change_%","MAE","WAPE_%","Best_Method"]
            ],top_n9.value)

        elif v=="validation":
            table9(validation_df.sort_values("MAE"))

        elif v=="data":
            table9(forecast_daily[
                ["job_family","forecast_day","Forecast","Last_7d_Total"]
            ].sort_values(["job_family","forecast_day"]))

        elif v=="findings":
            demand=forecast_summary.nlargest(5,"Forecast_7d")
            up=forecast_summary.dropna(subset=["Forecast_Change_%"]).nlargest(5,"Forecast_Change_%")

            display(HTML(
                '<div class="txt9"><b>Highest forecast demand</b><br>'+
                "<br>".join(
                    f"{r['job_family']}: {r['Forecast_7d']:,.0f} predicted postings"
                    for _,r in demand.iterrows()
                )+
                '<br><br><b>Strongest expected growth</b><br>'+
                "<br>".join(
                    f"{r['job_family']}: {r['Forecast_Change_%']:+.1f}%"
                    for _,r in up.iterrows()
                )+
                "</div>"
            ))

        elif v=="recommendations":
            display(HTML("""
            <div class="txt9">
            <b>01.</b> Prioritize roles with both high forecast volume and positive expected growth.<br><br>
            <b>02.</b> Treat forecasts as short-term demand intelligence rather than long-term labour-market projections.<br><br>
            <b>03.</b> Retain the lag-7 benchmark for transparent model comparison.<br><br>
            <b>04.</b> Refit forecasts when new complete-day data becomes available.<br><br>
            <b>05.</b> Combine Task 3 demand with Task 1 compensation and Task 2 momentum for recommendations.<br><br>
            <b>06.</b> Avoid forecasting low-volume categories without additional history.
            </div>
            """))

        else:
            display(HTML(f"""
            <div class="txt9">
            <b>Training window:</b> {dense_start:%d %b %Y} to {trend_end:%d %b %Y}.<br><br>
            <b>Frequency:</b> daily job-family posting counts; missing category-days are zero.<br><br>
            <b>Model:</b> least-squares regression with time trend and weekday indicators.<br><br>
            <b>Validation:</b> final seven complete days are held out chronologically; no random split.<br><br>
            <b>Benchmark:</b> lag-7 uses the corresponding day from the previous week.<br><br>
            <b>Metrics:</b> MAE and WAPE measure forecast error.<br><br>
            <b>Horizon:</b> seven days because the reliable history is short.<br><br>
            <b>Constraint:</b> negative predictions are clipped to zero.
            </div>
            """))

for w in [chart9,role9,top_n9,view9]:
    w.observe(update9,names="value")

update9()

dashboard_09=widgets.VBox([
    header9,kpi9,
    widgets.HTML('<div class="ctrl9"><b style="font:11px Arial;color:#22223b">TASK 03 FORECAST CONTROLS</b></div>'),
    controls9,view9,output9
])
dashboard_09.add_class("d9")
display(dashboard_09)
forecast_summary.to_csv(EXPORT_DIR/"task3_role_forecast_summary.csv",index=False)
forecast_daily.to_csv(EXPORT_DIR/"task3_daily_forecasts.csv",index=False)
validation_df.to_csv(EXPORT_DIR/"task3_forecast_validation.csv",index=False)

if "logger" in globals():
    logger.info(
        "TASK3_COMPLETE | roles=%s | horizon=%s | avg_mae=%.2f | baseline_win=%.1f%%",
        len(forecast_summary),horizon,validation_df["MAE"].mean(),
        (validation_df["MAE"]<validation_df["Lag7_MAE"]).mean()*100
    )
```

```python
if "df_final" not in globals():
    raise NameError("Run Section 05.3 first.")

task4=df_final.copy()

country_map=(task4.groupby(["country_clean","country_code","continent"],dropna=False)
    .agg(
        Total_Postings=("link","size"),
        Hourly_Postings=("pay_type",lambda x:x.eq("Hourly").sum()),
        Hourly_N=("hourly_midpoint","count"),
        Mean_Hourly=("hourly_midpoint","mean"),
        Median_Hourly=("hourly_midpoint","median"),
        P25_Hourly=("hourly_midpoint",lambda x:x.quantile(.25)),
        P75_Hourly=("hourly_midpoint",lambda x:x.quantile(.75)),
        Job_Families=("job_family","nunique")
    ).reset_index())

country_map["Hourly_Share_%"]=country_map["Hourly_Postings"]/country_map["Total_Postings"]*100
country_map["Market_Share_%"]=country_map["Total_Postings"]/country_map["Total_Postings"].sum()*100
country_map["Market_Rank"]=country_map["Total_Postings"].rank(ascending=False,method="min").astype(int)
country_map["Sample_Quality"]=np.select(
    [country_map["Hourly_N"]>=100,country_map["Hourly_N"]>=30,country_map["Hourly_N"]>=10],
    ["Strong","Reliable","Limited"],default="Insufficient"
)
country_map["Pay_Rank"]=country_map["Median_Hourly"].where(
    country_map["Hourly_N"]>=30
).rank(ascending=False,method="min")
country_map=country_map.sort_values("Total_Postings",ascending=False)

if country_map["country_code"].isna().any():
    raise ValueError("Missing country codes. Recheck Section 05.3.")

metric_labels={
    "Median_Hourly":"Median Reported Hourly",
    "Mean_Hourly":"Mean Reported Hourly",
    "Total_Postings":"Total Posting Volume",
    "Hourly_Postings":"Hourly Posting Volume",
    "Hourly_Share_%":"Hourly Posting Share",
    "Hourly_N":"Observed Hourly Sample",
    "Market_Share_%":"Market Share"
}
pay_metrics=["Median_Hourly","Mean_Hourly"]
refresh_time=pd.Timestamp.now(tz="Asia/Kolkata").strftime("%d %b %Y • %I:%M %p")

def card10(a,b,c):
    return f'<div class="c10"><small>{a}</small><b>{b}</b><span>{c}</span></div>'

def table10(df,n=None):
    display(HTML((df.head(n) if n else df).to_html(
        index=False,border=0,classes="t10",float_format=lambda x:f"{x:,.2f}"
    )))

display(HTML("""
<style>
.d10{border:1px solid #ded8d5;border-radius:17px;overflow:hidden;background:#fff;box-shadow:0 8px 24px #22223b17}
.h10{background:linear-gradient(120deg,#22223b,#30324d 55%,#4a4e69);padding:24px 27px;display:flex;justify-content:space-between;font-family:Arial}
.h10 .k{color:#c9ada7;font-size:8px;font-weight:700;letter-spacing:1.6px}.h10 h2{color:#fff;margin:6px 0;font-size:24px}.h10 p,.h10 .m{color:#f2e9e4;font-size:10px;line-height:1.65}
.cards10{display:grid;grid-template-columns:repeat(6,1fr);border-bottom:1px solid #e8e2df}.c10{padding:14px;border-right:1px solid #ebe6e4;font-family:Arial;min-width:0}
.c10 small{display:block;color:#9a8c98;font-size:7px;font-weight:700}.c10 b{display:block;color:#22223b;font-size:17px;margin-top:5px;overflow-wrap:anywhere}.c10 span{display:block;color:#4a4e69;font-size:8px;margin-top:4px}
.ctrl10{padding:13px 19px 3px;background:#faf9f8}.btn10{margin:7px 18px!important}.out10{border-top:1px solid #eee9e7!important}
.t10{width:100%;border-collapse:collapse;font:11px Arial}.t10 th{background:#22223b;color:#fff;padding:9px;text-align:left}.t10 td{padding:9px;border-bottom:1px solid #eee9e7;color:#22223b}
.txt10{font:12px/1.8 Arial;color:#22223b}.info10{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}
.box10{background:#f8f6f5;border-left:4px solid #4a4e69;padding:14px;border-radius:8px;font:11px/1.65 Arial;color:#22223b}.box10 b{display:block;color:#4a4e69;font-size:9px;margin-bottom:6px}
.tag10{display:inline-block;background:#f2e9e4;color:#4a4e69;padding:5px 9px;border-radius:12px;margin:3px;font:9px Arial}
.note10{background:#f8f6f5;border:1px solid #e5dfdc;border-radius:8px;padding:12px 14px;font:10px/1.6 Arial;color:#4a4e69}
@media(max-width:900px){.cards10{grid-template-columns:repeat(3,1fr)}.info10{grid-template-columns:1fr}}
</style>
"""))

header10=widgets.HTML(f"""
<div class="h10">
<div><div class="k">TASK 04 • GEOGRAPHIC MARKET ANALYSIS</div>
<h2>Global Hourly Rate & Market Intelligence</h2>
<p>Choropleth • Market Scale • Reported Pay • Geographic Benchmarking • Sample Quality</p></div>
<div class="m">COUNTRIES<br><b>{country_map['country_clean'].nunique():,}</b><br><br>
UPDATED<br><b>{refresh_time} IST</b></div>
</div>
""")

metric10=widgets.Dropdown(
    options=[(v,k) for k,v in metric_labels.items()],
    value="Median_Hourly",description="Metric:",layout=widgets.Layout(width="315px")
)
quality10=widgets.Dropdown(
    options=[("All Samples","all"),("Reliable+ (N ≥ 30)","reliable"),("Strong (N ≥ 100)","strong")],
    value="reliable",description="Quality:",layout=widgets.Layout(width="295px")
)
min_n10=widgets.IntSlider(
    value=30,min=10,max=200,step=10,description="Min Hourly N:",
    continuous_update=False,layout=widgets.Layout(width="300px")
)
continent10=widgets.Dropdown(
    options=["All"]+sorted(country_map["continent"].dropna().unique().tolist()),
    value="All",description="Continent:",layout=widgets.Layout(width="260px")
)
top_n10=widgets.IntSlider(
    value=10,min=5,max=25,step=5,description="Top N:",
    continuous_update=False,layout=widgets.Layout(width="270px")
)
country10=widgets.Dropdown(description="Country:",layout=widgets.Layout(width="300px"))

view10=widgets.ToggleButtons(
    options=[
        ("OVERVIEW","overview"),("MAP","map"),("RANKING","ranking"),
        ("MARKET MATRIX","matrix"),("DATA","data"),("COUNTRY DETAIL","detail"),
        ("FINDINGS","findings"),("RECOMMENDATIONS","recommendations"),
        ("INTERPRETATION","interpretation"),("METHOD","method")
    ],
    value="map",layout=widgets.Layout(width="100%"),
    style={"button_width":"10%","font_weight":"700"}
)
view10.add_class("btn10")

controls10=widgets.HBox(
    [metric10,quality10,min_n10,continent10,top_n10,country10],
    layout=widgets.Layout(padding="8px 18px",flex_flow="row wrap",gap="10px")
)

kpi10=widgets.HTML()
output10=widgets.Output(layout=widgets.Layout(padding="20px 23px"))
output10.add_class("out10")

def current10():
    d=country_map.copy()

    if continent10.value!="All":
        d=d[d["continent"].eq(continent10.value)]

    if quality10.value=="reliable":
        d=d[d["Hourly_N"]>=30]
    elif quality10.value=="strong":
        d=d[d["Hourly_N"]>=100]

    if metric10.value in pay_metrics:
        d=d[d["Hourly_N"]>=min_n10.value]

    return d.dropna(subset=[metric10.value])

def set_controls10(*_):
    min_n10.layout.display="" if metric10.value in pay_metrics else "none"

def update_countries10(*_):
    d=current10()
    country10.options=d["country_clean"].tolist() or ["No eligible country"]

def update10(*_):
    d=current10()

    with output10:
        clear_output(wait=True)

        if d.empty:
            kpi10.value='<div class="cards10">'+card10("STATUS","NO DATA","Adjust filters")+"</div>"
            display(HTML('<div class="txt10">No countries meet the selected controls.</div>'))
            return

        col=metric10.value
        label=metric_labels[col]
        top=d.nlargest(1,col).iloc[0]
        typical=d[col].median()
        coverage=d["Total_Postings"].sum()/country_map["Total_Postings"].sum()*100

        kpi10.value='<div class="cards10">'+"".join([
            card10("VISIBLE MARKETS",f"{len(d):,}","Current filters"),
            card10("TOP MARKET",top["country_clean"],label),
            card10("TOP VALUE",f"{top[col]:,.2f}",label),
            card10("TYPICAL MARKET",f"{typical:,.2f}","Visible-market median"),
            card10("HOURLY SAMPLE",f"{int(d['Hourly_N'].sum()):,}","Observed records"),
            card10("MARKET COVERAGE",f"{coverage:.1f}%","Posting volume")
        ])+"</div>"

        v=view10.value

        if v=="overview":
            display(HTML(f"""
            <div class="info10">
            <div class="box10"><b>ANALYTICAL QUESTION</b>How do reported hourly compensation and job-market scale differ geographically?</div>
            <div class="box10"><b>CURRENT VIEW</b>{label}<br>{len(d):,} eligible markets<br>Geography: {continent10.value}</div>
            <div class="box10"><b>QUALITY CONTROL</b>{quality10.label}<br>{"Minimum hourly N = "+str(min_n10.value) if col in pay_metrics else "No pay-specific threshold"}</div>
            </div><br>
            <span class="tag10">CHOROPLETH</span><span class="tag10">SAMPLE CONTROLLED</span>
            <span class="tag10">ORIGINAL PAY</span><span class="tag10">CLIENT READY</span><br><br>
            <div class="note10"><b>Currency limitation:</b> no currency field exists. Reported hourly values are not FX-normalized or purchasing-power adjusted.</div>
            """))

        elif v=="map":
            fig=go.Figure(go.Choropleth(
                locations=d["country_code"],locationmode="ISO-3",
                z=d[col],text=d["country_clean"],
                customdata=d[
                    ["continent","Total_Postings","Hourly_Postings","Hourly_N",
                     "Median_Hourly","Mean_Hourly","Hourly_Share_%","Sample_Quality"]
                ],
                colorscale=[
                    [0,"#f2e9e4"],[.30,"#c9ada7"],
                    [.62,"#9a8c98"],[1,"#4a4e69"]
                ],
                marker_line_color="#fff",marker_line_width=.4,
                colorbar=dict(title=label,thickness=13,len=.72),
                hovertemplate=(
                    "<b>%{text}</b><br>"+label+": %{z:,.2f}<br>"
                    "Continent: %{customdata[0]}<br>"
                    "Total postings: %{customdata[1]:,}<br>"
                    "Hourly postings: %{customdata[2]:,}<br>"
                    "Observed hourly N: %{customdata[3]:,}<br>"
                    "Median hourly: %{customdata[4]:,.2f}<br>"
                    "Mean hourly: %{customdata[5]:,.2f}<br>"
                    "Hourly share: %{customdata[6]:.1f}%<br>"
                    "Sample quality: %{customdata[7]}<extra></extra>"
                )
            ))

            fig.update_geos(
                showframe=False,showcoastlines=True,
                coastlinecolor="#c9ada7",
                showland=True,landcolor="#f8f6f5",
                showocean=True,oceancolor="#fff",
                projection_type="natural earth"
            )
            fig.update_layout(
                height=570,paper_bgcolor="#fff",
                margin=dict(l=0,r=0,t=10,b=0)
            )
            display(fig)

        elif v=="ranking":
            x=d.nlargest(top_n10.value,col).sort_values(col)

            fig=go.Figure(go.Bar(
                x=x[col],y=x["country_clean"],orientation="h",
                marker_color="#4a4e69",
                text=x[col].map(lambda z:f"{z:,.1f}"),
                textposition="outside",
                customdata=x[["Hourly_N","Total_Postings","Hourly_Share_%","Sample_Quality"]],
                hovertemplate=(
                    "<b>%{y}</b><br>"+label+": %{x:,.2f}<br>"
                    "Hourly N: %{customdata[0]:,}<br>"
                    "Total postings: %{customdata[1]:,}<br>"
                    "Hourly share: %{customdata[2]:.1f}%<br>"
                    "Quality: %{customdata[3]}<extra></extra>"
                )
            ))

            fig.update_xaxes(title=label)
            fig.update_layout(
                height=max(430,top_n10.value*31),
                paper_bgcolor="#fff",plot_bgcolor="#fff",
                font=dict(family="Arial",color="#22223b"),
                margin=dict(l=130,r=55,t=20,b=45)
            )
            display(fig)

        elif v=="matrix":
            x=d[d["Hourly_N"]>=max(30,min_n10.value)].dropna(subset=["Median_Hourly"])

            fig=go.Figure(go.Scatter(
                x=x["Total_Postings"],y=x["Median_Hourly"],mode="markers",
                marker=dict(
                    size=np.clip(np.sqrt(x["Hourly_N"])*1.4,8,42),
                    color=x["Hourly_Share_%"],
                    colorscale=[[0,"#f2e9e4"],[.45,"#c9ada7"],[.75,"#9a8c98"],[1,"#4a4e69"]],
                    showscale=True,colorbar=dict(title="Hourly Share %"),
                    opacity=.82,line=dict(color="#22223b",width=.6)
                ),
                text=x["country_clean"],
                customdata=x[["Hourly_N","Market_Share_%","Sample_Quality"]],
                hovertemplate=(
                    "<b>%{text}</b><br>"
                    "Total postings: %{x:,}<br>"
                    "Median reported hourly: %{y:,.2f}<br>"
                    "Hourly N: %{customdata[0]:,}<br>"
                    "Market share: %{customdata[1]:.2f}%<br>"
                    "Quality: %{customdata[2]}<extra></extra>"
                )
            ))

            fig.update_xaxes(title="Total Job Postings")
            fig.update_yaxes(title="Median Reported Hourly")
            fig.update_layout(
                height=520,paper_bgcolor="#fff",plot_bgcolor="#fff",
                font=dict(family="Arial",color="#22223b"),
                margin=dict(l=65,r=30,t=20,b=55)
            )
            display(fig)

        elif v=="data":
            table10(d.sort_values(col,ascending=False)[
                ["country_clean","country_code","continent","Sample_Quality",
                 "Total_Postings","Market_Share_%","Hourly_Postings","Hourly_N",
                 "Hourly_Share_%","Mean_Hourly","Median_Hourly",
                 "P25_Hourly","P75_Hourly","Job_Families"]
            ])

        elif v=="detail":
            x=country_map[country_map["country_clean"].eq(country10.value)]

            if len(x):
                x=x.iloc[0]
                reliable=country_map[country_map["Hourly_N"]>=30]
                benchmark=reliable["Median_Hourly"].median()

                diff=(x["Median_Hourly"]-benchmark)/benchmark*100 \
                    if pd.notna(x["Median_Hourly"]) and benchmark else np.nan

                rank="N/A" if pd.isna(x["Pay_Rank"]) else f"#{int(x['Pay_Rank'])}"
                med="N/A" if pd.isna(x["Median_Hourly"]) else f"{x['Median_Hourly']:,.2f}"
                mean="N/A" if pd.isna(x["Mean_Hourly"]) else f"{x['Mean_Hourly']:,.2f}"
                p25="N/A" if pd.isna(x["P25_Hourly"]) else f"{x['P25_Hourly']:,.2f}"
                p75="N/A" if pd.isna(x["P75_Hourly"]) else f"{x['P75_Hourly']:,.2f}"
                diff_txt="N/A" if pd.isna(diff) else f"{diff:+.2f}%"

                display(HTML(f"""
                <div class="info10">
                <div class="box10"><b>MARKET PROFILE</b>{x['country_clean']} • {x['country_code']}<br>{x['continent']}<br>Sample quality: {x['Sample_Quality']}</div>
                <div class="box10"><b>DEMAND PROFILE</b>Total postings: {int(x['Total_Postings']):,}<br>Market rank: #{int(x['Market_Rank'])}<br>Market share: {x['Market_Share_%']:.2f}%<br>Job families: {int(x['Job_Families']):,}</div>
                <div class="box10"><b>HOURLY PROFILE</b>Hourly postings: {int(x['Hourly_Postings']):,}<br>Observed hourly N: {int(x['Hourly_N']):,}<br>Hourly share: {x['Hourly_Share_%']:.2f}%<br>Pay rank: {rank}</div>
                </div><br>

                <div class="info10">
                <div class="box10"><b>REPORTED PAY</b>Median: {med}<br>Mean: {mean}<br>Reliable-market median: {benchmark:,.2f}</div>
                <div class="box10"><b>BENCHMARK DIFFERENCE</b>{diff_txt}<br>versus median of reliable country markets</div>
                <div class="box10"><b>DISTRIBUTION</b>P25: {p25}<br>P75: {p75}<br>Observed N: {int(x['Hourly_N']):,}</div>
                </div><br>

                <div class="note10">Reported compensation is not currency-normalized. Pay rank requires at least 30 observed hourly records.</div>
                """))

        elif v=="findings":
            reliable=country_map[country_map["Hourly_N"]>=max(30,min_n10.value)]
            pay_top=reliable.nlargest(5,"Median_Hourly")
            volume_top=country_map.nlargest(5,"Total_Postings")
            hourly_top=country_map[country_map["Hourly_N"]>=30].nlargest(5,"Hourly_Share_%")

            pay_lines="<br>".join(
                f"{r['country_clean']}: {r['Median_Hourly']:,.2f} (N={int(r['Hourly_N']):,})"
                for _,r in pay_top.iterrows()
            )

            volume_lines="<br>".join(
                f"{r['country_clean']}: {int(r['Total_Postings']):,} postings ({r['Market_Share_%']:.2f}% share)"
                for _,r in volume_top.iterrows()
            )

            hourly_lines="<br>".join(
                f"{r['country_clean']}: {r['Hourly_Share_%']:.1f}% hourly"
                for _,r in hourly_top.iterrows()
            )

            display(HTML(
                f'<div class="txt10"><b>Highest median reported hourly values</b><br>{pay_lines}'
                f'<br><br><b>Largest posting markets</b><br>{volume_lines}'
                f'<br><br><b>Highest hourly-job concentration</b><br>{hourly_lines}</div>'
            ))

        elif v=="recommendations":
            display(HTML("""
            <div class="txt10">
            <b>01.</b> Use the choropleth for geographic pattern recognition and Ranking for precise comparisons.<br><br>
            <b>02.</b> Use median reported hourly as the primary compensation statistic because pay is skewed.<br><br>
            <b>03.</b> Require N ≥ 30 for exploratory country-pay comparisons and prefer N ≥ 100 for stronger conclusions.<br><br>
            <b>04.</b> Interpret compensation together with posting volume and sample quality.<br><br>
            <b>05.</b> Use Market Matrix to identify markets combining demand, compensation and hourly-job concentration.<br><br>
            <b>06.</b> Do not present reported values as standardized international wages because currency information is unavailable.<br><br>
            <b>07.</b> Feed sample-controlled country signals into the recommendation engine.
            </div>
            """))

        elif v=="interpretation":
            display(HTML("""
            <div class="txt10">
            <b>Map colour represents the selected metric, not overall market attractiveness.</b><br><br>
            Compensation views use observed source pay and sample-quality controls. Volume metrics measure market activity, while hourly-share measures posting composition.<br><br>
            High reported hourly compensation in a small market should not automatically be interpreted as a stronger opportunity than a large, reliable market.<br><br>
            Cross-country compensation is not FX-normalized because the source contains no currency field.
            </div>
            """))

        else:
            display(HTML("""
            <div class="txt10">
            <b>Source:</b> validated feature-engineered job-market dataset.<br><br>
            <b>Geography:</b> standardized country names and ISO3 country codes from Section 05.<br><br>
            <b>Pay:</b> valid original hourly midpoint only; imputed values are excluded from descriptive geographic compensation statistics.<br><br>
            <b>Primary pay measure:</b> median reported hourly value; mean, P25 and P75 provide context.<br><br>
            <b>Sample quality:</b> Strong = N ≥ 100; Reliable = N ≥ 30; Limited = N ≥ 10; Insufficient = N &lt; 10.<br><br>
            <b>Market share:</b> country postings divided by total postings.<br><br>
            <b>Choropleth:</b> ISO3 codes map the selected country metric.<br><br>
            <b>Currency constraint:</b> no FX or purchasing-power normalization is performed.
            </div>
            """))

for w in [metric10,quality10,min_n10,continent10]:
    w.observe(update_countries10,names="value")
    w.observe(update10,names="value")

metric10.observe(set_controls10,names="value")

for w in [top_n10,country10,view10]:
    w.observe(update10,names="value")

set_controls10()
update_countries10()
update10()

dashboard_10=widgets.VBox([
    header10,kpi10,
    widgets.HTML('<div class="ctrl10"><b style="font:11px Arial;color:#22223b">GEOGRAPHIC INTELLIGENCE CONTROLS</b></div>'),
    controls10,view10,output10
])
dashboard_10.add_class("d10")
display(dashboard_10)
country_map.to_csv(EXPORT_DIR/"task4_country_market_analysis.csv",index=False)
country_map[country_map["Hourly_N"]>=30].to_csv(
    EXPORT_DIR/"task4_country_hourly_reliable.csv",index=False
)
country_map[country_map["Hourly_N"]>=100].to_csv(
    EXPORT_DIR/"task4_country_hourly_strong.csv",index=False
)

if "logger" in globals():
    logger.info(
        "TASK4_COMPLETE | countries=%s | reliable=%s | strong=%s",
        country_map["country_clean"].nunique(),
        (country_map["Hourly_N"]>=30).sum(),
        (country_map["Hourly_N"]>=100).sum()
    )
```

<div style="color:#f2e9e4;font-size:12px;line-height:1.65;margin-top:8px">Building a transparent content-based recommendation engine using job-title similarity with optional country, pay-type, seniority and recency preferences, plus explainable match scoring and API-ready outputs.</div>

```python
if "df_final" not in globals():
    raise NameError("Run Section 05.3 first.")

rec=df_final[[
    "title","title_normalized","link","published_date","country_clean",
    "pay_type","job_family","seniority_level","hourly_midpoint","fixed_budget"
]].copy().reset_index(drop=True)

rec["job_id"]=np.arange(len(rec))
rec["title_normalized"]=rec["title_normalized"].fillna("").astype(str)
rec["published_date"]=pd.to_datetime(rec["published_date"],utc=True)

vectorizer=TfidfVectorizer(
    stop_words="english",ngram_range=(1,2),
    min_df=2,max_features=12000,sublinear_tf=True
)

tfidf_matrix=vectorizer.fit_transform(rec["title_normalized"])
latest_date=rec["published_date"].max()
age=(latest_date-rec["published_date"]).dt.total_seconds()/86400
rec["recency_score"]=1-(age/age.max()).clip(0,1)

def recommend_jobs(query,country="Any",pay="Any",seniority="Any",n=10,mode="boost"):
    q=str(query).strip().lower()
    if len(q)<2:
        return pd.DataFrame()

    sim=cosine_similarity(vectorizer.transform([q]),tfidf_matrix).ravel()
    d=rec.copy()
    d["Title_Similarity"]=sim
    d["Country_Match"]=1.0 if country=="Any" else d["country_clean"].eq(country).astype(float)
    d["Pay_Match"]=1.0 if pay=="Any" else d["pay_type"].eq(pay).astype(float)
    d["Seniority_Match"]=1.0 if seniority=="Any" else d["seniority_level"].eq(seniority).astype(float)
    if mode=="strict":
        if country!="Any": d=d[d["Country_Match"].eq(1)]
        if pay!="Any": d=d[d["Pay_Match"].eq(1)]
        if seniority!="Any": d=d[d["Seniority_Match"].eq(1)]

    d["Match_Score"]=(
        .75*d["Title_Similarity"]+
        .10*d["Country_Match"]+
        .05*d["Pay_Match"]+
        .05*d["Seniority_Match"]+
        .05*d["recency_score"]
    )

    d=d[d["Title_Similarity"]>0].nlargest(n,"Match_Score").copy()
    d["Match_%"]=(d["Match_Score"]*100).round(1)

    return d[[
        "job_id","title","country_clean","pay_type","job_family","seniority_level",
        "hourly_midpoint","fixed_budget","published_date","link",
        "Title_Similarity","Country_Match","Pay_Match","Seniority_Match",
        "recency_score","Match_Score","Match_%"
    ]]

refresh_time=pd.Timestamp.now(tz="Asia/Kolkata").strftime("%d %b %Y • %I:%M %p")
state={"results":pd.DataFrame()}

def card11(a,b,c):
    return f'<div class="c11"><small>{a}</small><b>{b}</b><span>{c}</span></div>'

def table11(df,n=None):
    display(HTML((df.head(n) if n else df).to_html(
        index=False,border=0,classes="t11",float_format=lambda x:f"{x:,.3f}"
    )))

display(HTML("""
<style>
.d11{border:1px solid #ded8d5;border-radius:17px;overflow:hidden;background:#fff;box-shadow:0 8px 24px #22223b17}
.h11{background:linear-gradient(120deg,#22223b,#30324d 55%,#4a4e69);padding:24px 27px;display:flex;justify-content:space-between;font-family:Arial}
.h11 .k{color:#c9ada7;font-size:8px;font-weight:700;letter-spacing:1.6px}.h11 h2{color:#fff;margin:6px 0;font-size:24px}.h11 p,.h11 .m{color:#f2e9e4;font-size:10px;line-height:1.65}
.cards11{display:grid;grid-template-columns:repeat(6,1fr);border-bottom:1px solid #e8e2df}.c11{padding:14px;border-right:1px solid #ebe6e4;font-family:Arial;min-width:0}
.c11 small{display:block;color:#9a8c98;font-size:7px;font-weight:700}.c11 b{display:block;color:#22223b;font-size:17px;margin-top:5px;overflow-wrap:anywhere}.c11 span{display:block;color:#4a4e69;font-size:8px;margin-top:4px}
.ctrl11{padding:13px 19px 4px;background:#faf9f8}.btn11{margin:7px 18px!important}.out11{border-top:1px solid #eee9e7!important}
.t11{width:100%;border-collapse:collapse;font:11px Arial}.t11 th{background:#22223b;color:#fff;padding:9px;text-align:left}.t11 td{padding:9px;border-bottom:1px solid #eee9e7;color:#22223b}
.txt11{font:12px/1.8 Arial;color:#22223b}.info11{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}
.box11{background:#f8f6f5;border-left:4px solid #4a4e69;padding:14px;border-radius:8px;font:11px/1.65 Arial;color:#22223b}.box11 b{display:block;color:#4a4e69;font-size:9px;margin-bottom:6px}
.note11{background:#f8f6f5;border:1px solid #e5dfdc;border-radius:8px;padding:12px 14px;font:10px/1.6 Arial;color:#4a4e69}
</style>
"""))
header11=widgets.HTML(f"""
<div class="h11">
<div><div class="k">TASK 05 • RECOMMENDATION ENGINE</div>
<h2>Job Match Intelligence</h2>
<p>TF-IDF Similarity • Preference Matching • Recency • Explainability • API-Ready Output</p></div>
<div class="m">CATALOGUE<br><b>{len(rec):,}</b><br><br>UPDATED<br><b>{refresh_time} IST</b></div>
</div>
""")

query11=widgets.Text(
    value="data analyst",description="Target Role:",
    placeholder="e.g. data analyst",
    layout=widgets.Layout(width="330px")
)

country11=widgets.Dropdown(
    options=["Any"]+sorted(rec["country_clean"].dropna().unique().tolist()),
    value="Any",description="Country:",
    layout=widgets.Layout(width="270px")
)

pay11=widgets.Dropdown(
    options=["Any"]+sorted(rec["pay_type"].dropna().unique().tolist()),
    value="Any",description="Pay Type:",
    layout=widgets.Layout(width="240px")
)

seniority11=widgets.Dropdown(
    options=["Any"]+sorted(rec["seniority_level"].dropna().unique().tolist()),
    value="Any",description="Seniority:",
    layout=widgets.Layout(width="270px")
)

mode11=widgets.Dropdown(
    options=[("Preference Boost","boost"),("Strict Filters","strict")],
    value="boost",description="Mode:",
    layout=widgets.Layout(width="260px")
)

top_n11=widgets.IntSlider(
    value=10,min=5,max=25,step=5,description="Results:",
    continuous_update=False,layout=widgets.Layout(width="270px")
)

run11=widgets.Button(
    description="FIND MATCHES",
    button_style="",
    layout=widgets.Layout(width="170px",height="36px")
)

pick11=widgets.Dropdown(
    description="Inspect:",
    layout=widgets.Layout(width="440px")
)

view11=widgets.ToggleButtons(
    options=[
        ("OVERVIEW","overview"),("MATCHES","matches"),
        ("MATCH PROFILE","profile"),("DATA","data"),
        ("WHY MATCHED","why"),("API PREVIEW","api"),
        ("RECOMMENDATIONS","recommendations"),("METHOD","method")
    ],
    value="matches",layout=widgets.Layout(width="100%"),
    style={"button_width":"12%","font_weight":"700"}
)
view11.add_class("btn11")

controls11=widgets.HBox(
    [query11,country11,pay11,seniority11,mode11,top_n11,run11],
    layout=widgets.Layout(padding="8px 18px",flex_flow="row wrap",gap="10px")
)

inspect11=widgets.HBox(
    [pick11],
    layout=widgets.Layout(padding="0 18px 8px")
)

kpi11=widgets.HTML()
output11=widgets.Output(layout=widgets.Layout(padding="20px 23px"))
output11.add_class("out11")

def run_recommender11(*_):
    d=recommend_jobs(
        query11.value,country11.value,pay11.value,
        seniority11.value,top_n11.value,mode11.value
    )
    state["results"]=d
    if len(d):
        pick11.options=[
            (f"{r['title'][:70]} • {r['country_clean']} • {r['Match_%']:.1f}%",int(r["job_id"]))
            for _,r in d.iterrows()
        ]
    else:
        pick11.options=[]

    update11()

def selected11():
    d=state["results"]
    if d.empty or pick11.value is None:
        return None
    x=d[d["job_id"].eq(pick11.value)]
    return None if x.empty else x.iloc[0]

def update11(*_):
    d=state["results"]
    with output11:
        clear_output(wait=True)
        if d.empty:
            kpi11.value='<div class="cards11">'+card11("STATUS","NO MATCHES","Change query or filters")+"</div>"
            display(HTML("""
            <div class="txt11">
            Enter a target role and select <b>FIND MATCHES</b>.
            If Strict Filters returns no jobs, switch to Preference Boost.
            </div>
            """))
            return
        top=d.iloc[0]
        avg=d["Match_%"].mean()
        countries=d["country_clean"].nunique()
        families=d["job_family"].nunique()
        kpi11.value='<div class="cards11">'+"".join([
            card11("RESULTS",len(d),"Returned matches"),
            card11("TOP MATCH",f"{top['Match_%']:.1f}%","Composite score"),
            card11("AVG MATCH",f"{avg:.1f}%","Current results"),
            card11("COUNTRIES",countries,"Result diversity"),
            card11("JOB FAMILIES",families,"Result diversity"),
            card11("MODE","BOOST" if mode11.value=="boost" else "STRICT","Preference handling")
        ])+"</div>"

        v=view11.value
        x=selected11()

        if v=="overview":
            display(HTML(f"""
            <div class="info11">
            <div class="box11"><b>TARGET ROLE</b>{query11.value}<br>Title-based semantic matching</div>
            <div class="box11"><b>PREFERENCES</b>Country: {country11.value}<br>Pay: {pay11.value}<br>Seniority: {seniority11.value}</div>
            <div class="box11"><b>SCORING</b>75% title similarity<br>10% country<br>5% pay type<br>5% seniority<br>5% recency</div>
            </div><br>
            <div class="note11">
            The dataset has no employer, skills, description, experience or remote-status field.
            Recommendations therefore use only attributes actually available in the source.
            </div>
            """))

        elif v=="matches":
            show=d.copy()
            show["Pay"]=np.where(
                show["pay_type"].eq("Hourly"),
                show["hourly_midpoint"].round(2),
                show["fixed_budget"].round(2)
            )

            xshow=show[
                ["title","country_clean","pay_type","job_family",
                 "seniority_level","Pay","Match_%"]
            ].copy()

            fig=go.Figure(go.Bar(
                x=xshow["Match_%"][::-1],
                y=xshow["title"][::-1],
                orientation="h",
                marker_color="#4a4e69",
                customdata=xshow[
                    ["country_clean","pay_type","job_family","seniority_level"]
                ][::-1],
                hovertemplate=(
                    "<b>%{y}</b><br>"
                    "Match: %{x:.1f}%<br>"
                    "Country: %{customdata[0]}<br>"
                    "Pay type: %{customdata[1]}<br>"
                    "Family: %{customdata[2]}<br>"
                    "Seniority: %{customdata[3]}<extra></extra>"
                )
            ))

            fig.update_xaxes(title="Recommendation Match (%)")
            fig.update_layout(
                height=max(450,len(xshow)*38),
                paper_bgcolor="#fff",plot_bgcolor="#fff",
                font=dict(family="Arial",color="#22223b"),
                margin=dict(l=250,r=30,t=20,b=45)
            )
            display(fig)

        elif v=="profile" and x is not None:
            labels=["Title Similarity","Country Match","Pay Match","Seniority Match","Recency"]
            values=[
                x["Title_Similarity"]*100,x["Country_Match"]*100,
                x["Pay_Match"]*100,x["Seniority_Match"]*100,
                x["recency_score"]*100
            ]

            fig=go.Figure(go.Scatterpolar(
                r=values+[values[0]],
                theta=labels+[labels[0]],
                fill="toself",
                line=dict(color="#4a4e69",width=2),
                fillcolor="rgba(154,140,152,.25)"
            ))

            fig.update_layout(
                height=470,
                polar=dict(radialaxis=dict(range=[0,100],ticksuffix="%")),
                paper_bgcolor="#fff",
                font=dict(family="Arial",color="#22223b"),
                showlegend=False
            )
            display(fig)

        elif v=="data":
            table11(d[[
                "title","country_clean","pay_type","job_family","seniority_level",
                "hourly_midpoint","fixed_budget","Match_%","Title_Similarity"
            ]])

        elif v=="why" and x is not None:
            pay_value=x["hourly_midpoint"] if x["pay_type"]=="Hourly" else x["fixed_budget"]
            pay_text="N/A" if pd.isna(pay_value) else f"{pay_value:,.2f}"

            display(HTML(f"""
            <div class="info11">
            <div class="box11"><b>RECOMMENDED JOB</b>{x['title']}<br>{x['country_clean']} • {x['job_family']}</div>
            <div class="box11"><b>MATCH QUALITY</b>Overall: {x['Match_%']:.1f}%<br>Title similarity: {x['Title_Similarity']*100:.1f}%<br>Recency: {x['recency_score']*100:.1f}%</div>
            <div class="box11"><b>PREFERENCES</b>Country match: {"Yes" if x['Country_Match'] else "No"}<br>Pay match: {"Yes" if x['Pay_Match'] else "No"}<br>Seniority match: {"Yes" if x['Seniority_Match'] else "No"}</div>
            </div><br>

            <div class="info11">
            <div class="box11"><b>PAY TYPE</b>{x['pay_type']}<br>Reported value: {pay_text}</div>
            <div class="box11"><b>SENIORITY</b>{x['seniority_level']}</div>
            <div class="box11"><b>PUBLISHED</b>{x['published_date']:%d %b %Y}</div>
            </div>
            """))

        elif v=="api":
            example=d.iloc[0]

            display(HTML(f"""
            <div class="txt11">
            <b>Example API request</b>
            <pre style="background:#f8f6f5;padding:13px;border-radius:7px">{{
  "query": "{query11.value}",
  "country": "{country11.value}",
  "pay_type": "{pay11.value}",
  "seniority": "{seniority11.value}",
  "mode": "{mode11.value}",
  "top_n": {top_n11.value}
}}</pre>

            <b>Example response structure</b>
            <pre style="background:#f8f6f5;padding:13px;border-radius:7px">{{
  "title": "{str(example['title'])[:80]}",
  "country": "{example['country_clean']}",
  "job_family": "{example['job_family']}",
  "pay_type": "{example['pay_type']}",
  "match_score": {example['Match_%']:.1f}
}}</pre>
            </div>
            """))

        elif v=="recommendations":
            display(HTML("""
            <div class="txt11">
            <b>01.</b> Use title similarity as the primary signal because title is the richest textual field available.<br><br>
            <b>02.</b> Use Preference Boost for discovery and Strict Filters when user requirements are mandatory.<br><br>
            <b>03.</b> Show match explanations rather than exposing only a single opaque recommendation score.<br><br>
            <b>04.</b> Do not infer skills, remote status, company quality or required experience from fields that are absent.<br><br>
            <b>05.</b> Add Task 1 compensation, Task 2 momentum and Task 3 demand signals later as optional ranking features.<br><br>
            <b>06.</b> Refit TF-IDF when the job catalogue is refreshed.<br><br>
            <b>07.</b> Use this scoring function consistently in the later Streamlit and Flask implementations.
            </div>
            """))

        else:
            display(HTML("""
            <div class="txt11">
            <b>Model type:</b> content-based recommendation engine.<br><br>
            <b>Text representation:</b> TF-IDF using title unigrams and bigrams; no full pairwise similarity matrix is created.<br><br>
            <b>Title similarity:</b> cosine similarity between the user query and each job title vector.<br><br>
            <b>Composite score:</b> 75% title similarity + 10% country + 5% pay type + 5% seniority + 5% recency.<br><br>
            <b>Boost mode:</b> preferences increase ranking but do not eliminate other jobs.<br><br>
            <b>Strict mode:</b> selected country, pay-type and seniority preferences become hard filters.<br><br>
            <b>Explainability:</b> each score component is retained for inspection.<br><br>
            <b>Limitation:</b> recommendations are title-based because descriptions, employers, skills and experience requirements are unavailable.
            </div>
            """))


run11.on_click(run_recommender11)

for w in [pick11,view11]:
    w.observe(update11,names="value")

run_recommender11()

dashboard_11=widgets.VBox([
    header11,kpi11,
    widgets.HTML('<div class="ctrl11"><b style="font:11px Arial;color:#22223b">RECOMMENDATION CONTROLS</b></div>'),
    controls11,inspect11,view11,output11
])
dashboard_11.add_class("d11")
display(dashboard_11)
rec.to_csv(EXPORT_DIR/"task5_recommendation_catalog.csv",index=False)

pd.DataFrame({
    "Component":["Title similarity","Country preference","Pay-type preference","Seniority preference","Recency"],
    "Weight":[.75,.10,.05,.05,.05]
}).to_csv(EXPORT_DIR/"task5_recommendation_weights.csv",index=False)

if "logger" in globals():
    logger.info(
        "TASK5_COMPLETE | catalogue=%s | vocabulary=%s",
        f"{len(rec):,}",len(vectorizer.vocabulary_)
    )
```

```python
if "df_final" not in globals():
    raise NameError("Run Section 05.3 first.")

task6=df_final.copy()
task6["published_day"]=pd.to_datetime(task6["published_day"],utc=True)
last_ts=pd.to_datetime(task6["published_date"],utc=True).max()
last_day=last_ts.normalize()
trend_end=last_day-pd.Timedelta(days=1) if last_ts.hour<23 else last_day
dense_start=pd.Timestamp("2024-02-13",tz="UTC")
dense6=task6[task6["published_day"].between(dense_start,trend_end)].copy()
dense6["week_start"]=dense6["published_day"]-pd.to_timedelta(
    dense6["published_day"].dt.dayofweek,unit="D"
)
task6["month"]=task6["published_day"].dt.strftime("%Y-%m")
weekly_market=(dense6.groupby("week_start")
    .agg(
        Postings=("link","size"),
        Countries=("country_clean","nunique"),
        Job_Families=("job_family","nunique"),
        Hourly_Postings=("pay_type",lambda x:x.eq("Hourly").sum()),
        Median_Hourly=("hourly_midpoint","median"),
        Median_Fixed=("fixed_budget","median")
    ).reset_index())

weekly_market["Hourly_Share_%"]=weekly_market["Hourly_Postings"]/weekly_market["Postings"]*100
weekly_market["WoW_Growth_%"]=weekly_market["Postings"].pct_change()*100
weekly_market["Rolling_3W"]=weekly_market["Postings"].rolling(3,min_periods=1).mean()
family_weekly=(dense6.groupby(["week_start","job_family"]).size()
    .rename("Postings").reset_index())

family_weekly["Week_Total"]=family_weekly.groupby("week_start")["Postings"].transform("sum")
family_weekly["Share_%"]=family_weekly["Postings"]/family_weekly["Week_Total"]*100
family_weekly=family_weekly.sort_values(["job_family","week_start"])
family_weekly["WoW_Growth_%"]=family_weekly.groupby("job_family")["Postings"].pct_change()*100
family_weekly["Share_Change_pp"]=family_weekly.groupby("job_family")["Share_%"].diff()
latest_week=family_weekly["week_start"].max()
previous_week=latest_week-pd.Timedelta(days=7)

latest=family_weekly[family_weekly["week_start"].eq(latest_week)][
    ["job_family","Postings","Share_%"]
].rename(columns={"Postings":"Latest_Postings","Share_%":"Latest_Share_%"})

previous=family_weekly[family_weekly["week_start"].eq(previous_week)][
    ["job_family","Postings","Share_%"]
].rename(columns={"Postings":"Previous_Postings","Share_%":"Previous_Share_%"})

momentum=(latest.merge(previous,on="job_family",how="outer").fillna(0))
momentum["Volume_Change"]=momentum["Latest_Postings"]-momentum["Previous_Postings"]
momentum["Growth_%"]=np.where(
    momentum["Previous_Postings"]>0,
    momentum["Volume_Change"]/momentum["Previous_Postings"]*100,np.nan
)
momentum["Share_Change_pp"]=momentum["Latest_Share_%"]-momentum["Previous_Share_%"]
momentum["Momentum"]=np.select(
    [
        (momentum["Growth_%"]>10)&(momentum["Share_Change_pp"]>0),
        (momentum["Growth_%"]<-10)&(momentum["Share_Change_pp"]<0)
    ],
    ["Accelerating","Cooling"],default="Stable"
)
momentum=momentum.sort_values("Latest_Postings",ascending=False)
monthly_snapshot=(task6.groupby("month")
    .agg(
        Postings=("link","size"),
        Countries=("country_clean","nunique"),
        Job_Families=("job_family","nunique"),
        Hourly_Postings=("pay_type",lambda x:x.eq("Hourly").sum()),
        Dense_Records=("dense_window_flag","sum"),
        Median_Hourly=("hourly_midpoint","median")
    ).reset_index())

monthly_snapshot["Hourly_Share_%"]=monthly_snapshot["Hourly_Postings"]/monthly_snapshot["Postings"]*100
monthly_snapshot["Dense_Share_%"]=monthly_snapshot["Dense_Records"]/monthly_snapshot["Postings"]*100
monthly_snapshot["Reliability"]=np.select(
    [monthly_snapshot["Dense_Share_%"]>=95,monthly_snapshot["Dense_Share_%"]>=50],
    ["High","Mixed"],default="Sparse / Caution"
)

refresh_time=pd.Timestamp.now(tz="Asia/Kolkata").strftime("%d %b %Y • %I:%M %p")

def card12(a,b,c):
    return f'<div class="c12"><small>{a}</small><b>{b}</b><span>{c}</span></div>'

def table12(df,n=None):
    display(HTML((df.head(n) if n else df).to_html(
        index=False,border=0,classes="t12",float_format=lambda x:f"{x:,.2f}"
    )))

display(HTML("""
<style>
.d12{border:1px solid #ded8d5;border-radius:17px;overflow:hidden;background:#fff;box-shadow:0 8px 24px #22223b17}
.h12{background:linear-gradient(120deg,#22223b,#30324d 55%,#4a4e69);padding:24px 27px;display:flex;justify-content:space-between;font-family:Arial}
.h12 .k{color:#c9ada7;font-size:8px;font-weight:700;letter-spacing:1.6px}.h12 h2{color:#fff;margin:6px 0;font-size:24px}.h12 p,.h12 .m{color:#f2e9e4;font-size:10px;line-height:1.65}
.cards12{display:grid;grid-template-columns:repeat(6,1fr);border-bottom:1px solid #e8e2df}.c12{padding:14px;border-right:1px solid #ebe6e4;font-family:Arial;min-width:0}
.c12 small{display:block;color:#9a8c98;font-size:7px;font-weight:700}.c12 b{display:block;color:#22223b;font-size:17px;margin-top:5px;overflow-wrap:anywhere}.c12 span{display:block;color:#4a4e69;font-size:8px;margin-top:4px}
.ctrl12{padding:13px 19px 4px;background:#faf9f8}.btn12{margin:7px 18px!important}.out12{border-top:1px solid #eee9e7!important}
.t12{width:100%;border-collapse:collapse;font:11px Arial}.t12 th{background:#22223b;color:#fff;padding:9px;text-align:left}.t12 td{padding:9px;border-bottom:1px solid #eee9e7;color:#22223b}
.txt12{font:12px/1.8 Arial;color:#22223b}.info12{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}
.box12{background:#f8f6f5;border-left:4px solid #4a4e69;padding:14px;border-radius:8px;font:11px/1.65 Arial;color:#22223b}.box12 b{display:block;color:#4a4e69;font-size:9px;margin-bottom:6px}
.note12{background:#f8f6f5;border:1px solid #e5dfdc;border-radius:8px;padding:12px 14px;font:10px/1.6 Arial;color:#4a4e69}
</style>
"""))

header12=widgets.HTML(f"""
<div class="h12">
<div><div class="k">TASK 06 • MARKET DYNAMICS</div>
<h2>Recurring Market Dynamics Intelligence</h2>
<p>Demand Mix • Weekly Momentum • Market Composition • Monthly Monitoring</p></div>
<div class="m">DENSE WINDOW<br><b>{dense_start:%d %b} – {trend_end:%d %b %Y}</b><br><br>
UPDATED<br><b>{refresh_time} IST</b></div>
</div>
""")

chart12=widgets.Dropdown(
    options=[
        ("Demand Mix Stream","stream"),
        ("Market Activity","activity"),
        ("Category Momentum","momentum"),
        ("Pay-Type Mix","paymix"),
        ("Monthly Snapshot","monthly")
    ],
    value="stream",description="Chart:",
    layout=widgets.Layout(width="300px")
)

top_n12=widgets.IntSlider(
    value=5,min=3,max=10,step=1,description="Top N:",
    continuous_update=False,layout=widgets.Layout(width="270px")
)

family12=widgets.Dropdown(
    options=["All"]+sorted(dense6["job_family"].dropna().unique().tolist()),
    value="All",description="Family:",
    layout=widgets.Layout(width="300px")
)

view12=widgets.ToggleButtons(
    options=[
        ("OVERVIEW","overview"),("CHART","chart"),("MOMENTUM","momentum"),
        ("MONTHLY","monthly"),("DATA","data"),("FINDINGS","findings"),
        ("RECOMMENDATIONS","recommendations"),("INTERPRETATION","interpretation"),
        ("METHOD","method")
    ],
    value="chart",layout=widgets.Layout(width="100%"),
    style={"button_width":"11%","font_weight":"700"}
)
view12.add_class("btn12")

controls12=widgets.HBox(
    [chart12,top_n12,family12],
    layout=widgets.Layout(padding="8px 18px",flex_flow="row wrap",gap="10px")
)

kpi12=widgets.HTML()
output12=widgets.Output(layout=widgets.Layout(padding="20px 23px"))
output12.add_class("out12")

def update12(*_):
    with output12:
        clear_output(wait=True)

        current=weekly_market.iloc[-1]
        previous=weekly_market.iloc[-2] if len(weekly_market)>1 else current
        top_family=momentum.iloc[0]

        kpi12.value='<div class="cards12">'+"".join([
            card12("LATEST WEEK",f"{int(current['Postings']):,}","Postings"),
            card12("WEEKLY CHANGE",f"{current['WoW_Growth_%']:+.1f}%" if pd.notna(current["WoW_Growth_%"]) else "N/A","vs prior week"),
            card12("TOP FAMILY",top_family["job_family"],"Latest volume"),
            card12("ACTIVE COUNTRIES",f"{int(current['Countries']):,}","Latest week"),
            card12("HOURLY SHARE",f"{current['Hourly_Share_%']:.1f}%","Latest week"),
            card12("3W AVG",f"{current['Rolling_3W']:,.0f}","Posting baseline")
        ])+"</div>"

        v=view12.value

        if v=="overview":
            display(HTML(f"""
            <div class="info12">
            <div class="box12"><b>QUESTION</b>How are demand, job-family mix and market composition changing over time?</div>
            <div class="box12"><b>PRIMARY WINDOW</b>{dense_start:%d %b %Y} – {trend_end:%d %b %Y}<br>Weekly monitoring cadence</div>
            <div class="box12"><b>CONTROL</b>Dense-window trends are separated from sparse historical months to avoid misleading growth conclusions.</div>
            </div><br>
            <div class="note12">Monthly history is retained for context, but reliability is explicitly labelled because collection intensity changed sharply across the dataset.</div>
            """))

        elif v=="chart":

            if chart12.value=="stream":
                top=dense6["job_family"].value_counts().head(top_n12.value).index
                x=family_weekly[family_weekly["job_family"].isin(top)]
                palette=["#22223b","#4a4e69","#9a8c98","#c9ada7","#6d6875",
                         "#8d99ae","#b5838d","#7f8c8d","#a5a58d","#6b705c"]

                fig=go.Figure()
                for i,fam in enumerate(top):
                    z=x[x["job_family"].eq(fam)]
                    fig.add_trace(go.Scatter(
                        x=z["week_start"],y=z["Share_%"],
                        mode="lines",stackgroup="one",
                        groupnorm="percent",name=fam,
                        line=dict(width=1.4,color=palette[i%len(palette)]),
                        hovertemplate=f"<b>{fam}</b><br>%{{x|%d %b}}<br>Demand mix: %{{y:.1f}}%<extra></extra>"
                    ))
                fig.update_yaxes(title="Demand Mix (%)",range=[0,100])

            elif chart12.value=="activity":
                fig=go.Figure()
                fig.add_trace(go.Scatter(
                    x=weekly_market["week_start"],y=weekly_market["Postings"],
                    mode="lines+markers",name="Weekly Postings",
                    line=dict(color="#22223b",width=2),
                    marker=dict(size=7,color="#4a4e69")
                ))
                fig.add_trace(go.Scatter(
                    x=weekly_market["week_start"],y=weekly_market["Rolling_3W"],
                    mode="lines",name="3-Week Average",
                    line=dict(color="#c9ada7",width=3,dash="dash")
                ))
                fig.update_yaxes(title="Job Postings")

            elif chart12.value=="momentum":
                x=momentum.dropna(subset=["Growth_%"]).copy()
                x=x.nlargest(top_n12.value,"Latest_Postings")

                fig=go.Figure(go.Scatter(
                    x=x["Latest_Postings"],y=x["Growth_%"],
                    mode="markers+text",
                    text=x["job_family"],textposition="top center",
                    marker=dict(
                        size=np.clip(np.sqrt(x["Latest_Postings"])*1.8,12,45),
                        color=x["Share_Change_pp"],
                        colorscale=[[0,"#c9ada7"],[.5,"#f2e9e4"],[1,"#4a4e69"]],
                        showscale=True,colorbar=dict(title="Share Δ pp"),
                        line=dict(color="#22223b",width=.7)
                    ),
                    hovertemplate="<b>%{text}</b><br>Latest jobs: %{x:,}<br>Growth: %{y:+.1f}%<extra></extra>"
                ))
                fig.add_hline(y=0,line_color="#9a8c98")
                fig.update_xaxes(title="Latest Week Postings")
                fig.update_yaxes(title="Week-over-Week Growth (%)")

            elif chart12.value=="paymix":
                x=(dense6.groupby(["week_start","pay_type"]).size()
                    .rename("Postings").reset_index())
                pivot=x.pivot(index="week_start",columns="pay_type",values="Postings").fillna(0)
                share=pivot.div(pivot.sum(axis=1),axis=0)*100

                fig=go.Figure()
                palette=["#4a4e69","#c9ada7","#9a8c98"]
                for i,c in enumerate(share.columns):
                    fig.add_trace(go.Bar(
                        x=share.index,y=share[c],name=c,
                        marker_color=palette[i%len(palette)]
                    ))
                fig.update_layout(barmode="stack")
                fig.update_yaxes(title="Posting Mix (%)",range=[0,100])

            else:
                x=monthly_snapshot.copy()

                fig=go.Figure(go.Bar(
                    x=x["month"],y=x["Postings"],
                    marker_color="#4a4e69",
                    text=x["Reliability"],
                    customdata=x[["Dense_Share_%","Countries","Job_Families"]],
                    hovertemplate=(
                        "<b>%{x}</b><br>Postings: %{y:,}<br>"
                        "Reliability: %{text}<br>"
                        "Dense share: %{customdata[0]:.1f}%<br>"
                        "Countries: %{customdata[1]:,}<br>"
                        "Job families: %{customdata[2]:,}<extra></extra>"
                    )
                ))
                fig.update_yaxes(title="Monthly Postings")

            fig.update_layout(
                height=510,paper_bgcolor="#fff",plot_bgcolor="#fff",
                font=dict(family="Arial",color="#22223b"),
                margin=dict(l=65,r=35,t=25,b=50),
                hoverlabel=dict(bgcolor="#22223b",font_color="#fff")
            )
            display(fig)

        elif v=="momentum":
            d=momentum.copy()
            if family12.value!="All":
                d=d[d["job_family"].eq(family12.value)]

            table12(d[
                ["job_family","Momentum","Latest_Postings","Previous_Postings",
                 "Volume_Change","Growth_%","Latest_Share_%",
                 "Previous_Share_%","Share_Change_pp"]
            ].sort_values("Latest_Postings",ascending=False))

        elif v=="monthly":
            table12(monthly_snapshot[
                ["month","Postings","Countries","Job_Families",
                 "Hourly_Share_%","Median_Hourly",
                 "Dense_Share_%","Reliability"]
            ])

        elif v=="data":
            if family12.value=="All":
                table12(weekly_market)
            else:
                table12(family_weekly[
                    family_weekly["job_family"].eq(family12.value)
                ])

        elif v=="findings":
            accelerating=momentum[momentum["Momentum"].eq("Accelerating")].head(5)
            cooling=momentum[momentum["Momentum"].eq("Cooling")].head(5)

            a="<br>".join(
                f"{r['job_family']}: {int(r['Latest_Postings']):,} jobs, {r['Growth_%']:+.1f}% growth"
                for _,r in accelerating.iterrows()
            ) or "No categories meet the accelerating rule."

            c="<br>".join(
                f"{r['job_family']}: {int(r['Latest_Postings']):,} jobs, {r['Growth_%']:+.1f}% growth"
                for _,r in cooling.iterrows()
            ) or "No categories meet the cooling rule."

            display(HTML(f"""
            <div class="txt12">
            <b>Accelerating categories</b><br>{a}<br><br>
            <b>Cooling categories</b><br>{c}<br><br>
            <b>Latest market activity:</b> {int(current['Postings']):,} postings with
            {current['Hourly_Share_%']:.1f}% hourly share.<br><br>
            <b>Rolling baseline:</b> {current['Rolling_3W']:,.0f} postings per week.
            </div>
            """))

        elif v=="recommendations":
            display(HTML("""
            <div class="txt12">
            <b>01.</b> Monitor weekly category share as well as raw posting volume; volume growth can reflect overall platform expansion.<br><br>
            <b>02.</b> Flag categories as accelerating only when both posting growth and market-share movement are positive.<br><br>
            <b>03.</b> Use the 3-week rolling baseline to distinguish persistent movement from one-week noise.<br><br>
            <b>04.</b> Keep sparse historical months as context only; do not compare them directly with dense months as if collection intensity were constant.<br><br>
            <b>05.</b> Use recurring weekly market dynamics as an input to forecast refreshes and recommendation ranking.<br><br>
            <b>06.</b> Refresh the dashboard whenever a new complete observation week becomes available.
            </div>
            """))

        elif v=="interpretation":
            display(HTML("""
            <div class="txt12">
            <b>Market dynamics should be interpreted as movement within this platform snapshot, not as the entire labour market.</b><br><br>
            Rising posting volume indicates greater observed activity, while increasing share means a category is gaining importance relative to other categories.<br><br>
            Weekly changes are more reliable inside the dense collection period. Earlier monthly totals are retained but explicitly labelled for collection reliability.
            </div>
            """))

        else:
            display(HTML(f"""
            <div class="txt12">
            <b>Dense trend window:</b> {dense_start:%d %b %Y} to {trend_end:%d %b %Y}.<br><br>
            <b>Cadence:</b> Monday-aligned weekly posting counts for recurring monitoring.<br><br>
            <b>Demand share:</b> job-family postings divided by all postings in the same week.<br><br>
            <b>Momentum:</b> latest complete week compared with the preceding week.<br><br>
            <b>Accelerating:</b> growth above 10% with positive market-share movement.<br><br>
            <b>Cooling:</b> decline below -10% with negative market-share movement.<br><br>
            <b>Rolling baseline:</b> three-week moving average of total posting volume.<br><br>
            <b>Monthly reliability:</b> months are labelled using the proportion of records falling inside the dense observation window.
            </div>
            """))

for w in [chart12,top_n12,family12,view12]:
    w.observe(update12,names="value")

update12()

dashboard_12=widgets.VBox([
    header12,kpi12,
    widgets.HTML('<div class="ctrl12"><b style="font:11px Arial;color:#22223b">MARKET DYNAMICS CONTROLS</b></div>'),
    controls12,view12,output12
])
dashboard_12.add_class("d12")
display(dashboard_12)
weekly_market.to_csv(EXPORT_DIR/"task6_weekly_market_dynamics.csv",index=False)
family_weekly.to_csv(EXPORT_DIR/"task6_job_family_weekly_dynamics.csv",index=False)
momentum.to_csv(EXPORT_DIR/"task6_latest_market_momentum.csv",index=False)
monthly_snapshot.to_csv(EXPORT_DIR/"task6_monthly_market_snapshot.csv",index=False)

if "logger" in globals():
    logger.info(
        "TASK6_COMPLETE | dense_records=%s | weeks=%s | families=%s",
        f"{len(dense6):,}",
        weekly_market["week_start"].nunique(),
        dense6["job_family"].nunique()
    )
```

```python
if "df_final" not in globals():
    raise NameError("Run Section 05.3 first.")

task7=df_final.copy()
task7["published_day"]=pd.to_datetime(task7["published_day"],utc=True)
title=task7["title_normalized"].fillna("").str.lower()
proxy=r"\bremote\b|\bwork\s*from\s*home\b|\bwfh\b|\bhome[\s-]*based\b|\btelecommut(?:e|ing|er)?\b"
exclude=r"\bremote\s+(sensing|control|desktop|access|monitoring)\b"
task7["remote_title_flag"]=title.str.contains(proxy,regex=True,na=False)&~title.str.contains(exclude,regex=True,na=False)
task7["remote_proxy_label"]=np.where(task7["remote_title_flag"],"Remote Signal","No Signal")
remote_count=int(task7["remote_title_flag"].sum())
remote_share=remote_count/len(task7)*100

def segment_summary(col):
    x=(task7.groupby(col,dropna=False)
       .agg(Total_Postings=("link","size"),Remote_Proxy=("remote_title_flag","sum"))
       .reset_index())
    x["Remote_Proxy_Share_%"]=x["Remote_Proxy"]/x["Total_Postings"]*100
    return x.sort_values("Remote_Proxy",ascending=False)

segments={
    "family":segment_summary("job_family"),
    "country":segment_summary("country_clean"),
    "seniority":segment_summary("seniority_level"),
    "pay":segment_summary("pay_type")
}
segment_cols={
    "family":"job_family",
    "country":"country_clean",
    "seniority":"seniority_level",
    "pay":"pay_type"
}

last_ts=pd.to_datetime(task7["published_date"],utc=True).max()
last_day=last_ts.normalize()
trend_end=last_day-pd.Timedelta(days=1) if last_ts.hour<23 else last_day
dense_start=pd.Timestamp("2024-02-13",tz="UTC")
trend7=task7[task7["published_day"].between(dense_start,trend_end)].copy()
trend7["week_start"]=trend7["published_day"]-pd.to_timedelta(
    trend7["published_day"].dt.dayofweek,unit="D"
)

weekly_remote=(trend7.groupby("week_start")
    .agg(Total_Postings=("link","size"),Remote_Proxy=("remote_title_flag","sum"))
    .reset_index())
weekly_remote["Remote_Proxy_Share_%"]=weekly_remote["Remote_Proxy"]/weekly_remote["Total_Postings"]*100

def pay_summary(col):
    x=(task7.dropna(subset=[col])
       .groupby(["job_family","remote_proxy_label"])[col]
       .agg(["median","count"]).reset_index())
    med=x.pivot(index="job_family",columns="remote_proxy_label",values="median")
    cnt=x.pivot(index="job_family",columns="remote_proxy_label",values="count")
    out=pd.DataFrame(index=med.index)
    for g in ["Remote Signal","No Signal"]:
        out[f"{g}_Median"]=med[g] if g in med else np.nan
        out[f"{g}_N"]=cnt[g] if g in cnt else 0
    return out.reset_index()

pay_tables={
    "hourly":pay_summary("hourly_midpoint"),
    "fixed":pay_summary("fixed_budget")
}

refresh_time=pd.Timestamp.now(tz="Asia/Kolkata").strftime("%d %b %Y • %I:%M %p")

def card13(a,b,c):
    return f'<div class="c13"><small>{a}</small><b>{b}</b><span>{c}</span></div>'

def table13(df):
    display(HTML(df.to_html(
        index=False,border=0,classes="t13",
        float_format=lambda x:f"{x:,.2f}"
    )))

display(HTML("""
<style>
.d13{border:1px solid #ddd7d4;border-radius:18px;overflow:hidden;background:#fff;box-shadow:0 8px 25px #22223b18}
.h13{background:linear-gradient(120deg,#22223b,#30324d 55%,#4a4e69);padding:25px 28px;display:flex;justify-content:space-between;font-family:Arial}
.h13 .k{color:#c9ada7;font:700 8px Arial;letter-spacing:1.6px}.h13 h2{color:#fff;margin:6px 0;font:700 24px Arial}.h13 p,.h13 .m{color:#f2e9e4;font:10px/1.6 Arial}
.cards13{display:grid;grid-template-columns:repeat(6,1fr);border-bottom:1px solid #ece7e4}.c13{padding:14px;border-right:1px solid #eee9e7;font-family:Arial;min-width:0}
.c13 small{color:#9a8c98;font-size:7px;font-weight:700}.c13 b{display:block;color:#22223b;font-size:17px;margin-top:5px;overflow-wrap:anywhere}.c13 span{display:block;color:#4a4e69;font-size:8px;margin-top:4px}
.ctrl13{padding:12px 18px 3px;background:#faf9f8}.btn13{margin:7px 18px!important}.out13{border-top:1px solid #eee9e7!important}
.t13{width:100%;border-collapse:collapse;font:11px Arial}.t13 th{background:#22223b;color:#fff;padding:9px;text-align:left}.t13 td{padding:9px;border-bottom:1px solid #eee9e7;color:#22223b}
.txt13{font:12px/1.8 Arial;color:#22223b}.grid13{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}
.box13{background:#f8f6f5;border-left:4px solid #4a4e69;padding:14px;border-radius:8px;font:11px/1.65 Arial;color:#22223b}.box13 b{display:block;color:#4a4e69;font-size:9px;margin-bottom:5px}
.note13{background:#f2e9e4;padding:12px 14px;border-radius:8px;font:10px/1.6 Arial;color:#4a4e69}
@media(max-width:900px){.cards13{grid-template-columns:repeat(3,1fr)}.grid13{grid-template-columns:1fr}}
</style>
"""))
header13=widgets.HTML(f"""
<div class="h13">
<div>
<div class="k">TASK 07 • REMOTE-WORK PROXY ANALYSIS</div>
<h2>Remote-Work Landscape Intelligence</h2>
<p>Market Presence • Segment Concentration • Weekly Trend • Compensation Comparison</p>
</div>
<div class="m">PROXY RECORDS<br><b>{remote_count:,}</b><br><br>UPDATED<br><b>{refresh_time} IST</b></div>
</div>
""")

segment13=widgets.Dropdown(
    options=[("Job Family","family"),("Country","country"),
             ("Seniority","seniority"),("Pay Type","pay")],
    value="family",description="Segment:",
    layout=widgets.Layout(width="280px")
)

metric13=widgets.Dropdown(
    options=[("Proxy Share","share"),("Proxy Volume","volume")],
    value="share",description="Metric:",
    layout=widgets.Layout(width="250px")
)

pay13=widgets.Dropdown(
    options=[("Hourly","hourly"),("Fixed Budget","fixed")],
    value="hourly",description="Pay:",
    layout=widgets.Layout(width="230px")
)

min_n13=widgets.IntSlider(
    value=30,min=10,max=200,step=10,
    description="Min N:",continuous_update=False,
    layout=widgets.Layout(width="280px")
)

top_n13=widgets.IntSlider(
    value=10,min=5,max=20,step=5,
    description="Top N:",continuous_update=False,
    layout=widgets.Layout(width="260px")
)

view13=widgets.ToggleButtons(
    options=[
        ("OVERVIEW","overview"),("TREND","trend"),
        ("SEGMENTS","segments"),("COMPENSATION","compensation"),
        ("DATA","data"),("FINDINGS","findings"),
        ("RECOMMENDATIONS","recommendations"),("METHOD","method")
    ],
    value="overview",
    layout=widgets.Layout(width="100%"),
    style={"button_width":"12%","font_weight":"700"}
)
view13.add_class("btn13")

controls13=widgets.HBox(
    [segment13,metric13,pay13,min_n13,top_n13],
    layout=widgets.Layout(padding="8px 18px",flex_flow="row wrap",gap="10px")
)

kpi13=widgets.HTML()
output13=widgets.Output(layout=widgets.Layout(padding="20px 23px"))
output13.add_class("out13")

def update13(*_):
    with output13:
        clear_output(wait=True)
        family=segments["family"]
        country=segments["country"]
        ef=family[family["Remote_Proxy"]>=min_n13.value]
        ec=country[country["Remote_Proxy"]>=min_n13.value]
        top_family=(
            ef.nlargest(1,"Remote_Proxy_Share_%").iloc[0]["job_family"]
            if len(ef) else "N/A"
        )
        top_country=(
            ec.nlargest(1,"Remote_Proxy_Share_%").iloc[0]["country_clean"]
            if len(ec) else "N/A"
        )

        latest=weekly_remote.iloc[-1]["Remote_Proxy_Share_%"] if len(weekly_remote) else np.nan
        first=weekly_remote.iloc[0]["Remote_Proxy_Share_%"] if len(weekly_remote) else np.nan
        movement=latest-first if pd.notna(latest) and pd.notna(first) else np.nan
        kpi13.value='<div class="cards13">'+"".join([
            card13("REMOTE SIGNALS",f"{remote_count:,}","Explicit title proxy"),
            card13("MARKET SHARE",f"{remote_share:.2f}%","All postings"),
            card13("TOP FAMILY",top_family,f"N ≥ {min_n13.value}"),
            card13("TOP COUNTRY",top_country,f"N ≥ {min_n13.value}"),
            card13("LATEST WEEK",f"{latest:.2f}%","Proxy share"),
            card13("TREND SHIFT",f"{movement:+.2f} pp","Dense window")
        ])+"</div>"

        v=view13.value
        if v=="overview":
            fig=go.Figure(go.Pie(
                labels=["Remote Signal","No Explicit Signal"],
                values=[remote_count,len(task7)-remote_count],
                hole=.68,
                marker=dict(
                    colors=["#4a4e69","#f2e9e4"],
                    line=dict(color="#fff",width=2)
                ),
                textinfo="percent",
                hovertemplate="<b>%{label}</b><br>Jobs: %{value:,}<br>%{percent}<extra></extra>"
            ))

            fig.add_annotation(
                text=f"<b>{remote_share:.2f}%</b><br>Proxy Share",
                x=.5,y=.5,showarrow=False,
                font=dict(size=18,color="#22223b")
            )

            fig.update_layout(
                height=380,paper_bgcolor="#fff",
                margin=dict(l=10,r=10,t=10,b=10),
                font=dict(family="Arial",color="#22223b")
            )
            display(fig)
            display(HTML("""
            <div class="grid13">
            <div class="box13"><b>WHAT IT MEASURES</b>Explicit remote-work language visible in job titles.</div>
            <div class="box13"><b>WHAT IT DOES NOT MEASURE</b>Verified remote, hybrid or on-site work status.</div>
            <div class="box13"><b>FALSE-POSITIVE CONTROL</b>Remote sensing, control, desktop, access and monitoring are excluded.</div>
            </div>
            """))

        elif v=="trend":
            fig=go.Figure(go.Scatter(
                x=weekly_remote["week_start"],
                y=weekly_remote["Remote_Proxy_Share_%"],
                mode="lines+markers",
                line=dict(color="#4a4e69",width=3),
                marker=dict(size=8,color="#22223b"),
                fill="tozeroy",fillcolor="rgba(201,173,167,.20)",
                customdata=weekly_remote[["Remote_Proxy","Total_Postings"]],
                hovertemplate=(
                    "<b>%{x|%d %b %Y}</b><br>"
                    "Proxy share: %{y:.2f}%<br>"
                    "Proxy jobs: %{customdata[0]:,}<br>"
                    "Total jobs: %{customdata[1]:,}<extra></extra>"
                )
            ))

            fig.update_yaxes(title="Remote Proxy Share (%)")
            fig.update_xaxes(title="Week")
            fig.update_layout(
                height=480,paper_bgcolor="#fff",plot_bgcolor="#fff",
                font=dict(family="Arial",color="#22223b"),
                margin=dict(l=65,r=30,t=20,b=45)
            )
            display(fig)

        elif v=="segments":
            d=segments[segment13.value].copy()
            col=segment_cols[segment13.value]
            d=d[d["Remote_Proxy"]>=min_n13.value]
            measure="Remote_Proxy_Share_%" if metric13.value=="share" else "Remote_Proxy"
            x=d.nlargest(top_n13.value,measure).sort_values(measure)

            if x.empty:
                display(HTML('<div class="txt13">No segments meet the selected sample threshold.</div>'))
                return

            fig=go.Figure(go.Bar(
                x=x[measure],y=x[col].astype(str),orientation="h",
                marker_color="#4a4e69",
                customdata=x[["Remote_Proxy","Total_Postings","Remote_Proxy_Share_%"]],
                hovertemplate=(
                    "<b>%{y}</b><br>"
                    "Proxy jobs: %{customdata[0]:,}<br>"
                    "Total jobs: %{customdata[1]:,}<br>"
                    "Proxy share: %{customdata[2]:.2f}%<extra></extra>"
                )
            ))

            fig.update_xaxes(
                title="Remote Proxy Share (%)" if metric13.value=="share" else "Remote Proxy Jobs"
            )
            fig.update_layout(
                height=max(430,len(x)*32),
                paper_bgcolor="#fff",plot_bgcolor="#fff",
                font=dict(family="Arial",color="#22223b"),
                margin=dict(l=145,r=30,t=20,b=45)
            )
            display(fig)

        elif v=="compensation":
            d=pay_tables[pay13.value].copy()

            d=d[
                (d["Remote Signal_N"]>=min_n13.value)&
                (d["No Signal_N"]>=min_n13.value)
            ].dropna(subset=["Remote Signal_Median","No Signal_Median"])

            d["Difference"]=d["Remote Signal_Median"]-d["No Signal_Median"]
            d=d.reindex(
                d["Difference"].abs().nlargest(top_n13.value).index
            ).sort_values("Remote Signal_Median")

            if d.empty:
                display(HTML('<div class="txt13">No job families meet the selected pay sample threshold.</div>'))
                return

            fig=go.Figure()

            for _,r in d.iterrows():
                fig.add_trace(go.Scatter(
                    x=[r["No Signal_Median"],r["Remote Signal_Median"]],
                    y=[r["job_family"],r["job_family"]],
                    mode="lines",line=dict(color="#c9ada7",width=3),
                    showlegend=False,hoverinfo="skip"
                ))

            fig.add_trace(go.Scatter(
                x=d["No Signal_Median"],y=d["job_family"],
                mode="markers",name="No Signal",
                marker=dict(size=11,color="#9a8c98")
            ))

            fig.add_trace(go.Scatter(
                x=d["Remote Signal_Median"],y=d["job_family"],
                mode="markers",name="Remote Signal",
                marker=dict(size=12,color="#4a4e69")
            ))

            fig.update_xaxes(
                title="Median Hourly" if pay13.value=="hourly" else "Median Fixed Budget"
            )
            fig.update_layout(
                height=max(440,len(d)*38),
                paper_bgcolor="#fff",plot_bgcolor="#fff",
                font=dict(family="Arial",color="#22223b"),
                margin=dict(l=155,r=30,t=20,b=45)
            )
            display(fig)

        elif v=="data":
            d=segments[segment13.value]
            col=segment_cols[segment13.value]

            table13(
                d[d["Remote_Proxy"]>=min_n13.value][
                    [col,"Total_Postings","Remote_Proxy","Remote_Proxy_Share_%"]
                ].sort_values("Remote_Proxy",ascending=False)
            )

        elif v=="findings":
            fam=family[family["Remote_Proxy"]>=min_n13.value].nlargest(
                5,"Remote_Proxy_Share_%"
            )
            ctry=country[country["Remote_Proxy"]>=min_n13.value].nlargest(
                5,"Remote_Proxy_Share_%"
            )

            fam_text="<br>".join(
                f"{r['job_family']}: {r['Remote_Proxy_Share_%']:.2f}% "
                f"({int(r['Remote_Proxy']):,} jobs)"
                for _,r in fam.iterrows()
            ) or "No eligible categories."

            country_text="<br>".join(
                f"{r['country_clean']}: {r['Remote_Proxy_Share_%']:.2f}% "
                f"({int(r['Remote_Proxy']):,} jobs)"
                for _,r in ctry.iterrows()
            ) or "No eligible countries."

            display(HTML(f"""
            <div class="grid13">
            <div class="box13"><b>OVERALL SIGNAL</b>{remote_count:,} jobs<br>{remote_share:.2f}% of all postings</div>
            <div class="box13"><b>DENSE-WINDOW SHIFT</b>{movement:+.2f} percentage points</div>
            <div class="box13"><b>LATEST WEEK</b>{latest:.2f}% explicit remote-title share</div>
            </div><br>
            <div class="txt13">
            <b>Highest job-family concentration</b><br>{fam_text}
            <br><br><b>Highest country concentration</b><br>{country_text}
            </div>
            """))

        elif v=="recommendations":
            display(HTML("""
            <div class="txt13">
            <b>01.</b> Treat this as a title-based remote-work proxy, not verified work-location classification.<br><br>
            <b>02.</b> Evaluate proxy percentage together with absolute posting volume and sample size.<br><br>
            <b>03.</b> Prioritize segments showing both meaningful proxy volume and high concentration.<br><br>
            <b>04.</b> Keep hourly and fixed-price compensation comparisons separate.<br><br>
            <b>05.</b> Do not treat No Signal as On-Site; it only means the title contains no explicit remote wording.<br><br>
            <b>06.</b> Add job descriptions or a verified remote-status field for definitive remote-market analysis.<br><br>
            <b>07.</b> Use the proxy only as an optional recommendation-engine signal until better data is available.
            </div>
            """))

        else:
            display(HTML(f"""
            <div class="txt13">
            <b>Proxy phrases:</b> Remote, Work From Home, WFH, Home Based and Telecommute.<br><br>
            <b>Excluded technical phrases:</b> Remote Sensing, Remote Control, Remote Desktop, Remote Access and Remote Monitoring.<br><br>
            <b>No Signal:</b> absence of explicit title language; it does not mean On-Site.<br><br>
            <b>Trend window:</b> {dense_start:%d %b %Y} to {trend_end:%d %b %Y}; partial final day excluded.<br><br>
            <b>Compensation:</b> original valid hourly midpoint and fixed budget only; imputed pay is excluded.<br><br>
            <b>Sample control:</b> minimum N is user-controlled for segment and compensation comparisons.<br><br>
            <b>Main limitation:</b> the dataset contains no verified remote/hybrid/on-site field.
            </div>
            """))

for w in [segment13,metric13,pay13,min_n13,top_n13,view13]:
    w.observe(update13,names="value")

update13()

dashboard_13=widgets.VBox([
    header13,kpi13,
    widgets.HTML(
        '<div class="ctrl13"><b style="font:11px Arial;color:#22223b">'
        'REMOTE-WORK INTELLIGENCE CONTROLS</b></div>'
    ),
    controls13,view13,output13
])

dashboard_13.add_class("d13")
display(dashboard_13)

pd.DataFrame({
    "Metric":["Total Postings","Remote Proxy Jobs","Remote Proxy Share %"],
    "Value":[len(task7),remote_count,remote_share]
}).to_csv(EXPORT_DIR/"task7_remote_proxy_summary.csv",index=False)

segments["family"].to_csv(
    EXPORT_DIR/"task7_remote_proxy_by_job_family.csv",index=False
)
segments["country"].to_csv(
    EXPORT_DIR/"task7_remote_proxy_by_country.csv",index=False
)
weekly_remote.to_csv(
    EXPORT_DIR/"task7_remote_proxy_weekly_trend.csv",index=False
)

task7.loc[
    task7["remote_title_flag"],
    ["title","link","country_clean","job_family","seniority_level",
     "pay_type","hourly_midpoint","fixed_budget","published_date"]
].to_csv(
    EXPORT_DIR/"task7_remote_proxy_jobs.csv",index=False
)

if "logger" in globals():
    logger.info(
        "TASK7_COMPLETE | proxy_jobs=%s | proxy_share=%.2f | weeks=%s",
        f"{remote_count:,}",remote_share,
        weekly_remote["week_start"].nunique()
    )
```

```python
if "df_final" not in globals():
    raise NameError("Run Section 05.3 first.")

task8=df_final.copy()
task8["published_day"]=pd.to_datetime(task8["published_day"],utc=True)
last_ts=pd.to_datetime(task8["published_date"],utc=True).max()
last_day=last_ts.normalize()
trend_end=last_day-pd.Timedelta(days=1) if last_ts.hour<23 else last_day
dense_start=pd.Timestamp("2024-02-13",tz="UTC")
x=task8[task8["published_day"].between(dense_start,trend_end)].copy()
recent_start=trend_end-pd.Timedelta(days=13)
prev_end=recent_start-pd.Timedelta(days=1)
prev_start=prev_end-pd.Timedelta(days=13)
r=x[x["published_day"].between(recent_start,trend_end)].groupby("job_family").size()
p=x[x["published_day"].between(prev_start,prev_end)].groupby("job_family").size()
scenario=pd.DataFrame({"Recent_14d":r,"Previous_14d":p}).fillna(0)
scenario["Growth_%"]=np.where(
    scenario["Previous_14d"]>0,
    (scenario["Recent_14d"]-scenario["Previous_14d"])/scenario["Previous_14d"]*100,
    np.nan
)
scenario["Recent_Daily_Avg"]=scenario["Recent_14d"]/14
days=pd.date_range(recent_start,trend_end,freq="D",tz="UTC")
daily8=(x.groupby(["published_day","job_family"]).size()
        .rename("Postings").reset_index())

grid8=pd.MultiIndex.from_product(
    [days,scenario.index],names=["published_day","job_family"]
).to_frame(index=False)

recent_daily=(grid8.merge(
    daily8,on=["published_day","job_family"],how="left"
).fillna({"Postings":0}))

v=recent_daily.groupby("job_family")["Postings"].agg(["mean","std"])
scenario["Volatility"]=(v["std"]/v["mean"].replace(0,np.nan)).replace(
    [np.inf,-np.inf],np.nan
).fillna(0)

scenario["Momentum"]=scenario["Growth_%"].fillna(0).clip(-50,50)/100
scenario["Base_Band_%"]=np.clip(
    .10+.12*scenario["Volatility"]+.30*scenario["Momentum"].abs(),.10,.35
)

scenario["Direction"]=np.select(
    [scenario["Growth_%"]>=15,scenario["Growth_%"]<=-15],
    ["Expansion","Contraction"],default="Stable"
)
scenario["Outlook"]=np.select(
    [
        (scenario["Growth_%"]>=15)&(scenario["Recent_14d"]>=100),
        (scenario["Growth_%"]<=-15)&(scenario["Recent_14d"]>=100)
    ],
    ["Positive","Watch"],default="Neutral"
)
scenario["Risk"]=np.select(
    [scenario["Base_Band_%"]<=.17,scenario["Base_Band_%"]<=.25],
    ["Low","Moderate"],default="High"
)

scenario["Opportunity_Score"]=(
    100*(
        .45*scenario["Recent_14d"].rank(pct=True).fillna(0)+
        .35*scenario["Growth_%"].fillna(0).rank(pct=True).fillna(0)+
        .20*(1-scenario["Volatility"].rank(pct=True)).fillna(.5)
    )
).round(1)

scenario=scenario.reset_index().sort_values("Opportunity_Score",ascending=False)

refresh_time=pd.Timestamp.now(
    tz="Asia/Kolkata"
).strftime("%d %b %Y • %I:%M %p")


def card14(a,b,c):
    return f'<div class="c14"><small>{a}</small><b>{b}</b><span>{c}</span></div>'


def table14(d):
    display(HTML(d.to_html(
        index=False,border=0,classes="t14",
        float_format=lambda z:f"{z:,.2f}"
    )))


def layout14(fig,h=500,l=70):
    fig.update_layout(
        height=h,paper_bgcolor="#fff",plot_bgcolor="#fff",
        font=dict(family="Arial",color="#22223b"),
        margin=dict(l=l,r=35,t=30,b=50),
        hoverlabel=dict(bgcolor="#22223b",font_color="#fff"),
        legend=dict(orientation="h",y=1.08)
    )
    return fig


display(HTML("""
<style>
.d14{border:1px solid #ddd7d4;border-radius:18px;overflow:hidden;background:#fff;box-shadow:0 8px 26px #22223b18}
.h14{background:linear-gradient(120deg,#22223b,#30324d 56%,#4a4e69);padding:26px 29px;display:flex;justify-content:space-between;font-family:Arial}
.h14 .k{color:#c9ada7;font:700 8px Arial;letter-spacing:1.7px}.h14 h2{color:#fff;margin:6px 0;font:700 25px Arial}.h14 p,.h14 .m{color:#f2e9e4;font:10px/1.65 Arial}
.cards14{display:grid;grid-template-columns:repeat(6,1fr);border-bottom:1px solid #ece7e4}.c14{padding:14px 13px;border-right:1px solid #eee9e7;font-family:Arial;min-width:0}
.c14 small{color:#9a8c98;font-size:7px;font-weight:700}.c14 b{display:block;color:#22223b;font-size:17px;margin-top:5px;overflow-wrap:anywhere}.c14 span{display:block;color:#4a4e69;font-size:8px;margin-top:4px}
.ctrl14{padding:13px 18px 4px;background:#faf9f8}.btn14{margin:8px 15px!important}.out14{border-top:1px solid #eee9e7!important}
.t14{width:100%;border-collapse:collapse;font:11px Arial}.t14 th{background:#22223b;color:#fff;padding:9px;text-align:left}.t14 td{padding:9px;border-bottom:1px solid #eee9e7;color:#22223b}
.txt14{font:12px/1.8 Arial;color:#22223b}.grid14{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}
.box14{background:#f8f6f5;border-left:4px solid #4a4e69;padding:14px;border-radius:9px;font:11px/1.65 Arial;color:#22223b}.box14 b{display:block;color:#4a4e69;font-size:9px;margin-bottom:6px}
.note14{background:#f2e9e4;padding:12px 14px;border-radius:9px;font:10px/1.65 Arial;color:#4a4e69}
.badge14{display:inline-block;padding:5px 9px;border-radius:12px;background:#f2e9e4;color:#4a4e69;font:700 9px Arial;margin:3px}
@media(max-width:900px){.cards14{grid-template-columns:repeat(3,1fr)}.grid14{grid-template-columns:1fr}}
</style>
"""))

header14=widgets.HTML(f"""
<div class="h14">
<div><div class="k">TASK 08 • FUTURE SCENARIO ANALYSIS</div>
<h2>Future Demand Scenario Intelligence</h2>
<p>Opportunity • Momentum • Risk • Scenario Range • Strategic Decision Support</p></div>
<div class="m">DATA THROUGH<br><b>{trend_end:%d %b %Y}</b><br><br>
UPDATED<br><b>{refresh_time} IST</b></div>
</div>
""")

horizon14=widgets.ToggleButtons(
    options=[("7 Days",7),("14 Days",14)],value=7,
    description="Horizon:",style={"button_width":"72px"}
)
band14=widgets.Dropdown(
    options=[("Tight",.80),("Balanced",1.),("Wide",1.25)],
    value=1.,description="Range:",layout=widgets.Layout(width="220px")
)
outlook14=widgets.Dropdown(
    options=["All","Positive","Neutral","Watch"],
    value="All",description="Outlook:",layout=widgets.Layout(width="215px")
)
direction14=widgets.Dropdown(
    options=["All","Expansion","Stable","Contraction"],
    value="All",description="Direction:",layout=widgets.Layout(width="240px")
)
risk14=widgets.Dropdown(
    options=["All","Low","Moderate","High"],
    value="All",description="Risk:",layout=widgets.Layout(width="205px")
)
min_n14=widgets.IntSlider(
    value=100,min=25,max=1000,step=25,
    description="Min 14D:",continuous_update=False,
    layout=widgets.Layout(width="270px")
)
top_n14=widgets.IntSlider(
    value=8,min=5,max=20,description="Top N:",
    continuous_update=False,layout=widgets.Layout(width="245px")
)
family14=widgets.Dropdown(
    description="Category:",layout=widgets.Layout(width="300px")
)

view14=widgets.ToggleButtons(
    options=[
        ("OVERVIEW","overview"),("RANGE","range"),
        ("OPPORTUNITY","opportunity"),("MATRIX","matrix"),
        ("RISK MAP","riskmap"),("TRAJECTORY","trajectory"),
        ("DETAIL","detail"),("DATA","data"),
        ("FINDINGS","findings"),("RECOMMEND","recommendations"),
        ("METHOD","method")
    ],
    value="range",layout=widgets.Layout(width="100%"),
    style={"button_width":"9%","font_weight":"700"}
)
view14.add_class("btn14")

controls14=widgets.HBox(
    [horizon14,band14,outlook14,direction14,risk14,min_n14,top_n14,family14],
    layout=widgets.Layout(padding="8px 18px",flex_flow="row wrap",gap="10px")
)

kpi14=widgets.HTML()
output14=widgets.Output(layout=widgets.Layout(padding="20px 23px"))
output14.add_class("out14")


def current14():
    d=scenario[scenario["Recent_14d"]>=min_n14.value].copy()

    for w,col in [
        (outlook14,"Outlook"),
        (direction14,"Direction"),
        (risk14,"Risk")
    ]:
        if w.value!="All":
            d=d[d[col].eq(w.value)]

    d["Base"]=d["Recent_Daily_Avg"]*horizon14.value
    d["Band_%"]=(d["Base_Band_%"]*band14.value).clip(.08,.45)
    d["Downside"]=d["Base"]*(1-d["Band_%"])
    d["Upside"]=d["Base"]*(1+d["Band_%"])

    return d


def update_families14(*_):
    d=current14()
    old=family14.value
    opts=d["job_family"].tolist()
    family14.options=opts or ["No eligible category"]

    if old in opts:
        family14.value=old


def update14(*_):
    d=current14()

    with output14:
        clear_output(wait=True)

        if d.empty:
            kpi14.value='<div class="cards14">'+card14(
                "STATUS","NO DATA","Adjust filters"
            )+"</div>"
            display(HTML('<div class="txt14">No categories meet the selected controls.</div>'))
            return

        top=d.nlargest(1,"Opportunity_Score").iloc[0]
        coverage=d["Recent_14d"].sum()/max(scenario["Recent_14d"].sum(),1)*100

        kpi14.value='<div class="cards14">'+"".join([
            card14("TOP OPPORTUNITY",top["job_family"],f"Score {top['Opportunity_Score']:.1f}"),
            card14("BASE DEMAND",f"{top['Base']:,.0f}",f"{horizon14.value}-day scenario"),
            card14("GROWTH",f"{top['Growth_%']:+.1f}%",top["Direction"]),
            card14("POSITIVE",(d["Outlook"]=="Positive").sum(),"Eligible families"),
            card14("HIGH RISK",(d["Risk"]=="High").sum(),"Scenario uncertainty"),
            card14("COVERAGE",f"{coverage:.1f}%","Recent demand")
        ])+"</div>"

        v=view14.value

        if v=="overview":
            g=d.dropna(subset=["Growth_%"])
            strong=g.nlargest(1,"Growth_%").iloc[0] if len(g) else top
            safe=d.nsmallest(1,"Band_%").iloc[0]

            display(HTML(f"""
            <div class="grid14">
            <div class="box14"><b>TOP OPPORTUNITY</b>{top['job_family']}<br>
            Score: {top['Opportunity_Score']:.1f}/100<br>Base: {top['Base']:,.0f}</div>

            <div class="box14"><b>STRONGEST MOMENTUM</b>{strong['job_family']}<br>
            Growth: {strong['Growth_%']:+.1f}%<br>Outlook: {strong['Outlook']}</div>

            <div class="box14"><b>MOST STABLE</b>{safe['job_family']}<br>
            Range: ±{safe['Band_%']*100:.1f}%<br>Risk: {safe['Risk']}</div>
            </div><br>

            <span class="badge14">{horizon14.value}-DAY HORIZON</span>
            <span class="badge14">{band14.label.upper()} RANGE</span>
            <span class="badge14">{len(d)} ELIGIBLE FAMILIES</span><br><br>

            <div class="note14">
            Opportunity Score combines demand scale, growth and stability.
            Scenario bands are planning ranges, not confidence intervals.
            </div>
            """))

        elif v=="range":
            z=d.nlargest(top_n14.value,"Base").sort_values("Base")
            fig=go.Figure()

            for _,q in z.iterrows():
                fig.add_trace(go.Scatter(
                    x=[q["Downside"],q["Upside"]],
                    y=[q["job_family"]]*2,
                    mode="lines",
                    line=dict(color="#c9ada7",width=7),
                    showlegend=False,hoverinfo="skip"
                ))

            for col,name,color,size in [
                ("Downside","Downside","#9a8c98",9),
                ("Base","Base","#22223b",14),
                ("Upside","Upside","#4a4e69",10)
            ]:
                kw={}
                if col=="Base":
                    kw["customdata"]=z[["Growth_%","Band_%","Outlook","Opportunity_Score"]]
                    kw["hovertemplate"]=(
                        "<b>%{y}</b><br>Base: %{x:,.0f}<br>"
                        "Growth: %{customdata[0]:+.1f}%<br>"
                        "Range: ±%{customdata[1]:.1%}<br>"
                        "Outlook: %{customdata[2]}<br>"
                        "Opportunity: %{customdata[3]:.1f}<extra></extra>"
                    )

                fig.add_trace(go.Scatter(
                    x=z[col],y=z["job_family"],mode="markers",
                    name=name,marker=dict(size=size,color=color),**kw
                ))

            fig.update_xaxes(
                title=f"Expected Posting Volume — Next {horizon14.value} Days"
            )
            display(layout14(fig,max(460,len(z)*42),165))

        elif v=="opportunity":
            z=d.nlargest(top_n14.value,"Opportunity_Score").sort_values("Opportunity_Score")

            fig=go.Figure(go.Bar(
                x=z["Opportunity_Score"],y=z["job_family"],orientation="h",
                marker=dict(
                    color=z["Opportunity_Score"],
                    colorscale=[[0,"#f2e9e4"],[.5,"#9a8c98"],[1,"#4a4e69"]]
                ),
                text=z["Opportunity_Score"].map(lambda q:f"{q:.1f}"),
                textposition="outside",
                customdata=z[["Base","Growth_%","Volatility","Risk","Outlook"]],
                hovertemplate=(
                    "<b>%{y}</b><br>Opportunity: %{x:.1f}/100<br>"
                    "Base: %{customdata[0]:,.0f}<br>"
                    "Growth: %{customdata[1]:+.1f}%<br>"
                    "Volatility: %{customdata[2]:.2f}<br>"
                    "Risk: %{customdata[3]}<br>"
                    "Outlook: %{customdata[4]}<extra></extra>"
                )
            ))

            fig.update_xaxes(title="Opportunity Score",range=[0,105])
            display(layout14(fig,max(450,len(z)*38),155))

        elif v=="matrix":
            colors={"Positive":"#4a4e69","Neutral":"#9a8c98","Watch":"#c9ada7"}
            fig=go.Figure()

            for s,c in colors.items():
                z=d[d["Outlook"].eq(s)]
                if z.empty:
                    continue

                fig.add_trace(go.Scatter(
                    x=z["Base"],y=z["Growth_%"],mode="markers",
                    name=s,text=z["job_family"],
                    marker=dict(
                        size=np.clip(z["Opportunity_Score"]/2.3,10,42),
                        color=c,opacity=.82,
                        line=dict(color="#22223b",width=.6)
                    ),
                    customdata=z[["Opportunity_Score","Risk","Band_%"]],
                    hovertemplate=(
                        "<b>%{text}</b><br>Base: %{x:,.0f}<br>"
                        "Growth: %{y:+.1f}%<br>"
                        "Opportunity: %{customdata[0]:.1f}<br>"
                        "Risk: %{customdata[1]}<br>"
                        "Range: ±%{customdata[2]:.1%}<extra></extra>"
                    )
                ))

            fig.add_hline(y=0,line_color="#9a8c98",line_dash="dot")
            fig.add_vline(x=d["Base"].median(),line_color="#c9ada7",line_dash="dot")
            fig.update_xaxes(title=f"{horizon14.value}-Day Base Demand")
            fig.update_yaxes(title="Recent Growth (%)")
            display(layout14(fig,520))

        elif v=="riskmap":
            z=d.nlargest(top_n14.value,"Recent_14d")

            fig=go.Figure(go.Scatter(
                x=z["Volatility"],y=z["Band_%"]*100,
                mode="markers+text",
                text=z["job_family"],textposition="top center",
                marker=dict(
                    size=np.clip(np.sqrt(z["Base"])*2.2,12,45),
                    color=z["Momentum"],cmin=-.5,cmax=.5,
                    colorscale=[[0,"#c9ada7"],[.5,"#f2e9e4"],[1,"#4a4e69"]],
                    showscale=True,colorbar=dict(title="Momentum"),
                    line=dict(color="#22223b",width=.6)
                ),
                customdata=z[["Base","Growth_%","Risk"]],
                hovertemplate=(
                    "<b>%{text}</b><br>Volatility: %{x:.2f}<br>"
                    "Range: ±%{y:.1f}%<br>"
                    "Base: %{customdata[0]:,.0f}<br>"
                    "Growth: %{customdata[1]:+.1f}%<br>"
                    "Risk: %{customdata[2]}<extra></extra>"
                )
            ))

            fig.update_xaxes(title="Demand Volatility")
            fig.update_yaxes(title="Scenario Uncertainty (%)")
            display(layout14(fig,520))

        elif v=="trajectory":
            z=d[d["job_family"].eq(family14.value)]
            if z.empty:
                return

            q=z.iloc[0]
            hd=pd.date_range(
                max(dense_start,trend_end-pd.Timedelta(days=27)),
                trend_end,freq="D",tz="UTC"
            )

            hist=(daily8[daily8["job_family"].eq(q["job_family"])]
                  .set_index("published_day")["Postings"]
                  .reindex(hd,fill_value=0))

            fd=pd.date_range(
                trend_end+pd.Timedelta(days=1),
                periods=horizon14.value,freq="D",tz="UTC"
            )

            base=q["Base"]/horizon14.value
            low=q["Downside"]/horizon14.value
            high=q["Upside"]/horizon14.value

            fig=go.Figure()
            fig.add_trace(go.Scatter(
                x=hist.index,y=hist.values,mode="lines+markers",
                name="Actual",line=dict(color="#22223b",width=2)
            ))
            fig.add_trace(go.Scatter(
                x=fd,y=[high]*len(fd),mode="lines",
                line=dict(width=0),showlegend=False,hoverinfo="skip"
            ))
            fig.add_trace(go.Scatter(
                x=fd,y=[low]*len(fd),mode="lines",
                fill="tonexty",fillcolor="rgba(201,173,167,.27)",
                line=dict(width=0),name="Scenario Range"
            ))
            fig.add_trace(go.Scatter(
                x=fd,y=[base]*len(fd),mode="lines+markers",
                name="Base Scenario",
                line=dict(color="#4a4e69",width=3,dash="dash")
            ))

            fig.add_vline(x=trend_end,line_dash="dot",line_color="#9a8c98")
            fig.update_yaxes(title="Daily Job Postings")
            display(layout14(fig,470))

        elif v=="detail":
            z=d[d["job_family"].eq(family14.value)]
            if z.empty:
                return

            q=z.iloc[0]

            display(HTML(f"""
            <div class="grid14">
            <div class="box14"><b>MARKET MOMENTUM</b>
            Recent 14D: {int(q['Recent_14d']):,}<br>
            Previous 14D: {int(q['Previous_14d']):,}<br>
            Growth: {q['Growth_%']:+.1f}%<br>
            Direction: {q['Direction']}</div>

            <div class="box14"><b>{horizon14.value}-DAY SCENARIO</b>
            Downside: {q['Downside']:,.0f}<br>
            Base: {q['Base']:,.0f}<br>
            Upside: {q['Upside']:,.0f}<br>
            Range: ±{q['Band_%']*100:.1f}%</div>

            <div class="box14"><b>DECISION PROFILE</b>
            Opportunity: {q['Opportunity_Score']:.1f}/100<br>
            Outlook: {q['Outlook']}<br>
            Risk: {q['Risk']}<br>
            Volatility: {q['Volatility']:.2f}</div>
            </div><br>

            <div class="note14">
            Opportunity Score is a transparent ranking heuristic,
            not a probability of future success.
            </div>
            """))

        elif v=="data":
            table14(d[[
                "job_family","Opportunity_Score","Outlook","Direction","Risk",
                "Recent_14d","Previous_14d","Growth_%","Volatility",
                "Downside","Base","Upside","Band_%"
            ]].sort_values("Opportunity_Score",ascending=False))

        elif v=="findings":
            a=d.nlargest(5,"Opportunity_Score")
            g=d.dropna(subset=["Growth_%"]).nlargest(5,"Growth_%")
            u=d.nlargest(5,"Band_%")

            display(HTML(f"""
            <div class="grid14">

            <div class="box14"><b>TOP OPPORTUNITIES</b>
            {"<br>".join(f"{q['job_family']}: {q['Opportunity_Score']:.1f}/100"
            for _,q in a.iterrows())}</div>

            <div class="box14"><b>FASTEST MOMENTUM</b>
            {"<br>".join(f"{q['job_family']}: {q['Growth_%']:+.1f}%"
            for _,q in g.iterrows())}</div>

            <div class="box14"><b>HIGHEST UNCERTAINTY</b>
            {"<br>".join(f"{q['job_family']}: ±{q['Band_%']*100:.1f}% ({q['Risk']})"
            for _,q in u.iterrows())}</div>

            </div>
            """))

        elif v=="recommendations":
            best=d.nlargest(1,"Opportunity_Score").iloc[0]
            biggest=d.nlargest(1,"Base").iloc[0]
            safest=d.nsmallest(1,"Band_%").iloc[0]

            display(HTML(f"""
            <div class="txt14">
            <b>01.</b> Highest combined opportunity: <b>{best['job_family']}</b>
            ({best['Opportunity_Score']:.1f}/100).<br><br>

            <b>02.</b> Largest expected demand pool: <b>{biggest['job_family']}</b>
            with about {biggest['Base']:,.0f} postings.<br><br>

            <b>03.</b> Most stable scenario: <b>{safest['job_family']}</b>
            with ±{safest['Band_%']*100:.1f}% range.<br><br>

            <b>04.</b> Prioritize strong demand + positive momentum + manageable uncertainty.<br><br>

            <b>05.</b> Use downside for conservative planning and upside for opportunity sizing.<br><br>

            <b>06.</b> Combine with compensation, geography and validated Task 3 forecasts.<br><br>

            <b>07.</b> Recalculate scenarios when new complete periods become available.
            </div>
            """))

        else:
            display(HTML(f"""
            <div class="txt14">
            <b>Reliable history:</b> {dense_start:%d %b %Y} to {trend_end:%d %b %Y}.<br><br>
            <b>Momentum:</b> latest 14 days vs previous 14 days.<br><br>
            <b>Base:</b> recent daily average × selected horizon.<br><br>
            <b>Volatility:</b> coefficient of variation of recent daily postings.<br><br>
            <b>Scenario width:</b> combines minimum uncertainty, volatility and momentum.<br><br>
            <b>Opportunity Score:</b> 45% demand + 35% growth + 20% stability percentiles.<br><br>
            <b>Risk:</b> Low ≤17%, Moderate ≤25%, High &gt;25%.<br><br>
            <b>Positive:</b> growth ≥15% and ≥100 recent postings.<br><br>
            <b>Watch:</b> growth ≤-15% and ≥100 recent postings.<br><br>
            <b>Note:</b> scenario ranges are planning heuristics, not confidence intervals or long-term forecasts.
            </div>
            """))


filter_widgets=[
    horizon14,band14,outlook14,
    direction14,risk14,min_n14
]

for w in filter_widgets:
    w.observe(update_families14,names="value")
    w.observe(update14,names="value")

for w in [top_n14,family14,view14]:
    w.observe(update14,names="value")

update_families14()
update14()

dashboard_14=widgets.VBox([
    header14,kpi14,
    widgets.HTML(
        '<div class="ctrl14"><b style="font:11px Arial;color:#22223b">'
        'SCENARIO PLANNING COCKPIT</b></div>'
    ),
    controls14,view14,output14
])

dashboard_14.add_class("d14")
display(dashboard_14)

scenario.to_csv(
    EXPORT_DIR/"task8_future_demand_scenarios.csv",index=False
)
scenario.sort_values(
    "Opportunity_Score",ascending=False
).to_csv(
    EXPORT_DIR/"task8_opportunity_ranking.csv",index=False
)
scenario[scenario["Outlook"].eq("Positive")].to_csv(
    EXPORT_DIR/"task8_positive_outlook_categories.csv",index=False
)
scenario[scenario["Outlook"].eq("Watch")].to_csv(
    EXPORT_DIR/"task8_watchlist_categories.csv",index=False
)

if "logger" in globals():
    logger.info(
        "TASK8_COMPLETE | categories=%s | positive=%s | high_risk=%s",
        len(scenario),
        (scenario["Outlook"]=="Positive").sum(),
        (scenario["Risk"]=="High").sum()
    )
```

```python
if "df_final" not in globals():
    raise NameError("Run Section 05 first.")

CONFIG_DIR=Path("config")
CONFIG_DIR.mkdir(exist_ok=True)

schema_rules={
    "title":{"required":True,"type":"text"},
    "link":{"required":True,"type":"text","unique":True},
    "published_date":{"required":True,"type":"datetime"},
    "is_hourly":{"required":False,"type":"boolean"},
    "hourly_low":{"required":False,"type":"numeric","min":0},
    "hourly_high":{"required":False,"type":"numeric","min":0},
    "budget":{"required":False,"type":"numeric","min":0},
    "country":{"required":False,"type":"text"}
}

rows=[]

for col,r in schema_rules.items():
    if col not in df_final:
        rows.append([
            col,r["required"],False,np.nan,np.nan,"N/A",
            "FAIL" if r["required"] else "WARNING"
        ])
        continue

    s=df_final[col]
    invalid=0

    if r["type"]=="numeric":
        n=pd.to_numeric(s,errors="coerce")
        invalid=int((s.notna()&n.isna()).sum())
        invalid+=int((n<r.get("min",-np.inf)).fillna(False).sum())

    elif r["type"]=="datetime":
        d=pd.to_datetime(s,errors="coerce",utc=True)
        invalid=int((s.notna()&d.isna()).sum())

    elif r["type"]=="boolean":
        invalid=int((~s.dropna().isin([True,False,0,1])).sum())

    unique=(
        "FAIL"
        if r.get("unique") and s.dropna().duplicated().any()
        else "PASS"
    )

    status=(
        "FAIL" if invalid or unique=="FAIL"
        else "WARNING" if r["required"] and s.isna().any()
        else "PASS"
    )

    rows.append([
        col,r["required"],True,
        int(s.isna().sum()),invalid,unique,status
    ])

schema_report=pd.DataFrame(
    rows,
    columns=[
        "Column","Required","Exists","Missing",
        "Invalid","Unique_Check","Status"
    ]
)

logical_checks=pd.DataFrame([
    [
        "Hourly lower ≤ upper",
        int(((df_final["hourly_low"]>df_final["hourly_high"]) &
             df_final["hourly_low"].notna() &
             df_final["hourly_high"].notna()).sum()),
        "ERROR"
    ],
    [
        "Hourly values non-negative",
        int(((df_final["hourly_low"]<0) |
             (df_final["hourly_high"]<0)).fillna(False).sum()),
        "ERROR"
    ],
    [
        "Fixed budget non-negative",
        int((df_final["budget"]<0).fillna(False).sum()),
        "ERROR"
    ],
    [
        "Duplicate links",
        int(df_final["link"].dropna().duplicated().sum()),
        "ERROR"
    ],
    [
        "Missing title",
        int(df_final["title"].isna().sum()),
        "WARNING"
    ],
    [
        "Missing link",
        int(df_final["link"].isna().sum()),
        "ERROR"
    ]
],columns=["Rule","Violations","Severity"])

logical_checks["Status"]=np.where(
    logical_checks["Violations"].eq(0),"PASS",
    np.where(logical_checks["Severity"].eq("ERROR"),"FAIL","WARNING")
)

schema_fail=int((schema_report["Status"]=="FAIL").sum())
schema_warn=int((schema_report["Status"]=="WARNING").sum())
logic_fail=int((logical_checks["Status"]=="FAIL").sum())
logic_warn=int((logical_checks["Status"]=="WARNING").sum())

overall="PASS" if schema_fail+logic_fail==0 else "REVIEW"

schema_report.to_csv(
    EXPORT_DIR/"schema_validation_report.csv",index=False
)
logical_checks.to_csv(
    EXPORT_DIR/"logical_validation_report.csv",index=False
)

with open(CONFIG_DIR/"schema_contract.json","w",encoding="utf-8") as f:
    json.dump(schema_rules,f,indent=2)

validation_summary=pd.DataFrame({
    "Metric":[
        "Rows Validated","Governed Fields",
        "Schema Failures","Schema Warnings",
        "Logical Failures","Logical Warnings",
        "Overall Status"
    ],
    "Value":[
        len(df_final),len(schema_rules),
        schema_fail,schema_warn,
        logic_fail,logic_warn,overall
    ]
})

validation_summary.to_csv(
    EXPORT_DIR/"schema_validation_summary.csv",index=False
)

def card16(a,b,c):
    return f"""
    <div class="c16">
    <small>{a}</small>
    <b>{b}</b>
    <span>{c}</span>
    </div>
    """

def table16(d):
    display(HTML(d.to_html(
        index=False,border=0,classes="t16"
    )))

display(HTML("""
<style>
.d16{
border:1px solid #ddd7d4;border-radius:18px;overflow:hidden;
background:#fff;box-shadow:0 8px 26px #22223b18
}
.h16{
background:linear-gradient(120deg,#22223b,#30324d 56%,#4a4e69);
padding:26px 29px;display:flex;justify-content:space-between;
font-family:Arial
}
.h16 .k{color:#c9ada7;font:700 8px Arial;letter-spacing:1.7px}
.h16 h2{color:#fff;margin:6px 0;font:700 25px Arial}
.h16 p,.h16 .m{color:#f2e9e4;font:10px/1.65 Arial}

.cards16{
display:grid;grid-template-columns:repeat(6,1fr);
border-bottom:1px solid #ece7e4
}
.c16{
padding:14px;border-right:1px solid #eee9e7;
font-family:Arial;min-width:0
}
.c16 small{color:#9a8c98;font-size:7px;font-weight:700}
.c16 b{
display:block;color:#22223b;font-size:17px;
margin-top:5px;overflow-wrap:anywhere
}
.c16 span{display:block;color:#4a4e69;font-size:8px;margin-top:4px}

.btn16{margin:8px 15px!important}
.out16{
border-top:1px solid #eee9e7!important;
padding:20px 23px
}

.t16{
width:100%;border-collapse:collapse;font:11px Arial
}
.t16 th{
background:#22223b;color:#fff;padding:9px;text-align:left
}
.t16 td{
padding:9px;border-bottom:1px solid #eee9e7;color:#22223b
}

.grid16{
display:grid;grid-template-columns:repeat(3,1fr);gap:12px
}
.box16{
background:#f8f6f5;border-left:4px solid #4a4e69;
padding:14px;border-radius:9px;
font:11px/1.65 Arial;color:#22223b
}
.box16 b{
display:block;color:#4a4e69;font-size:9px;margin-bottom:6px
}
.txt16{font:12px/1.8 Arial;color:#22223b}
.note16{
background:#f2e9e4;padding:12px 14px;border-radius:9px;
font:10px/1.65 Arial;color:#4a4e69
}
@media(max-width:900px){
.cards16{grid-template-columns:repeat(3,1fr)}
.grid16{grid-template-columns:1fr}
}
</style>
"""))

header16=widgets.HTML(f"""
<div class="h16">
<div>
<div class="k">SECTION 16 • DATA GOVERNANCE CONTROL</div>
<h2>Schema & Validation Intelligence</h2>
<p>Schema Contract • Data Integrity • Business Rules • Validation Evidence</p>
</div>

<div class="m">
RECORDS VALIDATED<br>
<b>{len(df_final):,}</b><br><br>
STATUS<br>
<b>{overall}</b>
</div>
</div>
""")

kpi16=widgets.HTML(
    '<div class="cards16">'+"".join([
        card16("OVERALL STATUS",overall,"Validation gate"),
        card16("RECORDS",f"{len(df_final):,}","Rows validated"),
        card16("FIELDS",len(schema_rules),"Governed fields"),
        card16("SCHEMA FAILS",schema_fail,"Blocking issues"),
        card16("LOGIC FAILS",logic_fail,"Business-rule issues"),
        card16("WARNINGS",schema_warn+logic_warn,"Review items")
    ])+"</div>"
)

view16=widgets.ToggleButtons(
    options=[
        ("OVERVIEW","overview"),
        ("STATUS","status"),
        ("SCHEMA","schema"),
        ("RULES","rules"),
        ("MISSINGNESS","missing"),
        ("FINDINGS","findings"),
        ("EVIDENCE","evidence"),
        ("METHOD","method")
    ],
    value="overview",
    layout=widgets.Layout(width="100%"),
    style={"button_width":"12%","font_weight":"700"}
)
view16.add_class("btn16")

output16=widgets.Output()
output16.add_class("out16")


def update16(*_):
    with output16:
        clear_output(wait=True)

        v=view16.value

        if v=="overview":
            pass_n=int((schema_report["Status"]=="PASS").sum())
            rule_pass=int((logical_checks["Status"]=="PASS").sum())

            display(HTML(f"""
            <div class="grid16">

            <div class="box16">
            <b>SCHEMA COVERAGE</b>
            {len(schema_rules)} source fields governed<br>
            {pass_n} currently pass schema validation
            </div>

            <div class="box16">
            <b>BUSINESS RULES</b>
            {len(logical_checks)} integrity rules tested<br>
            {rule_pass} currently pass
            </div>

            <div class="box16">
            <b>VALIDATION GATE</b>
            Overall status: {overall}<br>
            Blocking issues: {schema_fail+logic_fail}
            </div>

            </div><br>

            <div class="note16">
            Validation evidence is now exported separately from the analytical
            dataset so downstream models, APIs and dashboards can reference the
            same governed schema.
            </div>
            """))

        elif v=="status":
            status_counts=pd.concat([
                schema_report["Status"],
                logical_checks["Status"]
            ]).value_counts().reindex(
                ["PASS","WARNING","FAIL"],fill_value=0
            )

            fig=go.Figure(go.Pie(
                labels=status_counts.index,
                values=status_counts.values,
                hole=.65,
                marker=dict(
                    colors=["#4a4e69","#c9ada7","#9a8c98"]
                ),
                textinfo="label+value"
            ))

            fig.add_annotation(
                text=f"<b>{overall}</b><br>Validation",
                x=.5,y=.5,showarrow=False,
                font=dict(size=18,color="#22223b")
            )

            fig.update_layout(
                height=430,
                paper_bgcolor="#fff",
                font=dict(family="Arial",color="#22223b"),
                margin=dict(l=20,r=20,t=20,b=20)
            )

            display(fig)

        elif v=="schema":
            table16(schema_report)

        elif v=="rules":
            table16(logical_checks)

        elif v=="missing":
            z=schema_report[
                schema_report["Exists"].eq(True)
            ].copy()

            z=z.sort_values("Missing")

            fig=go.Figure(go.Bar(
                x=z["Missing"],
                y=z["Column"],
                orientation="h",
                marker_color="#4a4e69",
                customdata=z[["Required","Status"]],
                hovertemplate=(
                    "<b>%{y}</b><br>"
                    "Missing: %{x:,}<br>"
                    "Required: %{customdata[0]}<br>"
                    "Status: %{customdata[1]}"
                    "<extra></extra>"
                )
            ))

            fig.update_xaxes(title="Missing Records")

            fig.update_layout(
                height=450,
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",
                font=dict(family="Arial",color="#22223b"),
                margin=dict(l=120,r=30,t=20,b=45)
            )

            display(fig)

        elif v=="findings":
            fail_schema=schema_report[
                schema_report["Status"].eq("FAIL")
            ]
            warning_schema=schema_report[
                schema_report["Status"].eq("WARNING")
            ]
            failed_rules=logical_checks[
                logical_checks["Status"].eq("FAIL")
            ]

            display(HTML(f"""
            <div class="grid16">

            <div class="box16">
            <b>SCHEMA FAILURES</b>
            {
                "<br>".join(fail_schema["Column"])
                if len(fail_schema)
                else "No blocking schema failures."
            }
            </div>

            <div class="box16">
            <b>SCHEMA WARNINGS</b>
            {
                "<br>".join(warning_schema["Column"])
                if len(warning_schema)
                else "No schema warnings."
            }
            </div>

            <div class="box16">
            <b>FAILED BUSINESS RULES</b>
            {
                "<br>".join(failed_rules["Rule"])
                if len(failed_rules)
                else "No blocking integrity-rule failures."
            }
            </div>

            </div>
            """))

        elif v=="evidence":
            evidence=pd.DataFrame({
                "Evidence":[
                    "Schema contract",
                    "Schema validation report",
                    "Logical validation report",
                    "Validation summary"
                ],
                "File":[
                    "config/schema_contract.json",
                    "schema_validation_report.csv",
                    "logical_validation_report.csv",
                    "schema_validation_summary.csv"
                ],
                "Status":["Created"]*4
            })

            table16(evidence)

        else:
            display(HTML("""
            <div class="txt16">

            <b>Schema contract:</b>
            Defines required fields, expected data classes, uniqueness and
            numerical validity rules.<br><br>

            <b>Schema validation:</b>
            Confirms field existence, required-field completeness,
            parseability and uniqueness.<br><br>

            <b>Business validation:</b>
            Checks compensation bounds, negative values, duplicate links
            and critical-field completeness.<br><br>

            <b>Severity:</b>
            ERROR represents a blocking integrity issue; WARNING identifies
            a condition requiring review without automatically invalidating
            the processed dataset.<br><br>

            <b>Evidence:</b>
            All results are exported so validation can be reproduced outside
            notebook memory.

            </div>
            """))


view16.observe(update16,names="value")
update16()

dashboard_16=widgets.VBox([
    header16,
    kpi16,
    view16,
    output16
])

dashboard_16.add_class("d16")
display(dashboard_16)

if "logger" in globals():
    logger.info(
        "SECTION16_COMPLETE | status=%s | schema_fail=%s | logic_fail=%s",
        overall,schema_fail,logic_fail
    )
```

```python
required=["df_features","kw_tables"]
missing=[x for x in required if x not in globals()]

if missing:
    raise NameError(f"Run Section 07 first. Missing: {missing}")

title_col="title_normalized" if "title_normalized" in df_features else "title"

def bh_adjust(p):
    p=np.asarray(p,dtype=float)
    out=np.full(len(p),np.nan)
    ok=np.isfinite(p)

    if not ok.any():
        return out

    x=p[ok]
    order=np.argsort(x)
    ranked=x[order]*len(x)/np.arange(1,len(x)+1)
    ranked=np.minimum.accumulate(ranked[::-1])[::-1]
    adj=np.empty(len(x))
    adj[order]=np.clip(ranked,0,1)
    out[ok]=adj
    return out

def effect_label(r):
    a=abs(r)
    return (
        "Negligible" if a<.10 else
        "Small" if a<.30 else
        "Moderate" if a<.50 else
        "Large"
    )

tests17=[]

pay_sets={
    "Hourly":("hourly_midpoint",("hourly","original")),
    "Fixed Budget":("fixed_budget",("fixed","original"))
}

for pay_type,(pay_col,key) in pay_sets.items():

    if key not in kw_tables or pay_col not in df_features:
        continue

    base=df_features[[title_col,pay_col]].dropna().copy()
    base=base[base[pay_col]>=0]
    kt=kw_tables[key].copy()
    candidates=kt.loc[kt["Count"]>=100,"keyword"].astype(str).unique()

    tokens=(
        base[title_col].fillna("").str.lower()
        .str.findall(r"[a-z0-9+#.]+")
        .explode()
    )

    token_map=(
        pd.DataFrame({
            "idx":tokens.index,
            "keyword":tokens.values
        })
        .dropna()
    )

    token_map=token_map[token_map["keyword"].isin(candidates)].drop_duplicates()

    groups={
        k:g["idx"].to_numpy()
        for k,g in token_map.groupby("keyword")
    }

    for keyword in candidates:
        ids=groups.get(keyword,np.array([]))
        mask=base.index.isin(ids)
        yes=base.loc[mask,pay_col].dropna()
        no=base.loc[~mask,pay_col].dropna()

        if len(yes)<100 or len(no)<100:
            continue

        try:
            u,p=mannwhitneyu(
                yes,no,
                alternative="two-sided",
                method="asymptotic"
            )

            effect=2*u/(len(yes)*len(no))-1
        except Exception:
            u,p,effect=np.nan,np.nan,np.nan

        med_yes=yes.median()
        med_no=no.median()

        lift=(
            100*(med_yes-med_no)/med_no
            if med_no!=0 else np.nan
        )

        tests17.append({
            "Pay_Type":pay_type,
            "Keyword":keyword,
            "Keyword_N":len(yes),
            "Reference_N":len(no),
            "Keyword_Median":med_yes,
            "Reference_Median":med_no,
            "Median_Lift_%":lift,
            "U_Statistic":u,
            "P_Value":p,
            "Effect_Size":effect
        })

task1_tests=pd.DataFrame(tests17)
task1_tests["Adjusted_P"]=np.nan

for pay_type in task1_tests["Pay_Type"].unique():
    m=task1_tests["Pay_Type"].eq(pay_type)
    task1_tests.loc[m,"Adjusted_P"]=bh_adjust(
        task1_tests.loc[m,"P_Value"]
    )

task1_tests["Significant"]=task1_tests["Adjusted_P"]<.05
task1_tests["Effect_Strength"]=task1_tests["Effect_Size"].map(effect_label)

task1_tests["Practical_Signal"]=np.where(
    task1_tests["Significant"] &
    task1_tests["Effect_Size"].abs().ge(.10),
    "YES","NO"
)

task1_tests["Direction"]=np.select(
    [
        task1_tests["Effect_Size"]>.02,
        task1_tests["Effect_Size"]<-.02
    ],
    ["Higher Pay","Lower Pay"],
    default="Neutral"
)

task1_tests=task1_tests.sort_values(
    ["Pay_Type","Adjusted_P","Effect_Size"],
    ascending=[True,True,False]
).reset_index(drop=True)

summary17=(
    task1_tests.groupby("Pay_Type")
    .agg(
        Keywords_Tested=("Keyword","size"),
        Significant=("Significant","sum"),
        Practical_Signals=("Practical_Signal",lambda x:(x=="YES").sum()),
        Median_Abs_Effect=("Effect_Size",lambda x:x.abs().median()),
        Median_Sample=("Keyword_N","median")
    )
    .reset_index()
)

task1_tests.to_csv(
    EXPORT_DIR/"task1_statistical_tests.csv",
    index=False
)

summary17.to_csv(
    EXPORT_DIR/"task1_statistical_validation_summary.csv",
    index=False
)

def card17(a,b,c):
    return f"""
    <div class="c17">
    <small>{a}</small>
    <b>{b}</b>
    <span>{c}</span>
    </div>
    """

def table17(d):
    display(HTML(d.to_html(
        index=False,
        border=0,
        classes="t17",
        float_format=lambda x:f"{x:,.4f}"
    )))

display(HTML("""
<style>
.d17{
border:1px solid #ddd7d4;border-radius:18px;overflow:hidden;
background:#fff;box-shadow:0 8px 26px #22223b18
}
.h17{
background:linear-gradient(120deg,#22223b,#30324d 56%,#4a4e69);
padding:26px 29px;display:flex;justify-content:space-between;
font-family:Arial
}
.h17 .k{color:#c9ada7;font:700 8px Arial;letter-spacing:1.7px}
.h17 h2{color:#fff;margin:6px 0;font:700 25px Arial}
.h17 p,.h17 .m{color:#f2e9e4;font:10px/1.65 Arial}

.cards17{
display:grid;grid-template-columns:repeat(6,1fr);
border-bottom:1px solid #ece7e4
}
.c17{
padding:14px;border-right:1px solid #eee9e7;
font-family:Arial;min-width:0
}
.c17 small{color:#9a8c98;font-size:7px;font-weight:700}
.c17 b{
display:block;color:#22223b;font-size:17px;
margin-top:5px;overflow-wrap:anywhere
}
.c17 span{
display:block;color:#4a4e69;font-size:8px;margin-top:4px
}

.ctrl17{padding:12px 18px 4px;background:#faf9f8}
.btn17{margin:8px 15px!important}
.out17{border-top:1px solid #eee9e7!important;padding:20px 23px}

.t17{width:100%;border-collapse:collapse;font:10px Arial}
.t17 th{
background:#22223b;color:#fff;padding:9px;text-align:left
}
.t17 td{
padding:8px;border-bottom:1px solid #eee9e7;color:#22223b
}

.grid17{
display:grid;grid-template-columns:repeat(3,1fr);gap:12px
}
.box17{
background:#f8f6f5;border-left:4px solid #4a4e69;
padding:14px;border-radius:9px;
font:11px/1.65 Arial;color:#22223b
}
.box17 b{
display:block;color:#4a4e69;font-size:9px;margin-bottom:6px
}
.note17{
background:#f2e9e4;padding:12px 14px;border-radius:9px;
font:10px/1.65 Arial;color:#4a4e69
}
.txt17{font:12px/1.8 Arial;color:#22223b}

@media(max-width:900px){
.cards17{grid-template-columns:repeat(3,1fr)}
.grid17{grid-template-columns:1fr}
}
</style>
"""))

header17=widgets.HTML("""
<div class="h17">
<div>
<div class="k">TASK 01 • STATISTICAL VALIDATION</div>
<h2>Keyword Compensation Evidence Center</h2>
<p>Significance • Effect Size • Pay Lift • Sample Control • Multiple Testing</p>
</div>
<div class="m">
METHOD<br>
<b>MANN–WHITNEY U</b><br><br>
CORRECTION<br>
<b>BENJAMINI–HOCHBERG</b>
</div>
</div>
""")

pay17=widgets.Dropdown(
    options=task1_tests["Pay_Type"].unique().tolist(),
    description="Pay Type:",
    layout=widgets.Layout(width="240px")
)

sig17=widgets.Dropdown(
    options=["All","Significant","Practical Signal"],
    value="All",
    description="Filter:",
    layout=widgets.Layout(width="250px")
)

min_n17=widgets.IntSlider(
    value=100,min=100,max=1000,step=100,
    description="Min N:",
    continuous_update=False,
    layout=widgets.Layout(width="270px")
)

top_n17=widgets.IntSlider(
    value=12,min=5,max=30,
    description="Top N:",
    continuous_update=False,
    layout=widgets.Layout(width="260px")
)

view17=widgets.ToggleButtons(
    options=[
        ("OVERVIEW","overview"),
        ("SIGNIFICANCE","significance"),
        ("EFFECT SIZE","effect"),
        ("PAY LIFT","lift"),
        ("DATA","data"),
        ("FINDINGS","findings"),
        ("EVIDENCE","evidence"),
        ("METHOD","method")
    ],
    value="overview",
    layout=widgets.Layout(width="100%"),
    style={"button_width":"12%","font_weight":"700"}
)
view17.add_class("btn17")

controls17=widgets.HBox(
    [pay17,sig17,min_n17,top_n17],
    layout=widgets.Layout(
        padding="8px 18px",
        flex_flow="row wrap",
        gap="10px"
    )
)

kpi17=widgets.HTML()
output17=widgets.Output()
output17.add_class("out17")

def current17():
    d=task1_tests[
        task1_tests["Pay_Type"].eq(pay17.value) &
        task1_tests["Keyword_N"].ge(min_n17.value)
    ].copy()

    if sig17.value=="Significant":
        d=d[d["Significant"]]

    elif sig17.value=="Practical Signal":
        d=d[d["Practical_Signal"].eq("YES")]

    return d


def update17(*_):
    d=current17()

    with output17:
        clear_output(wait=True)

        if d.empty:
            kpi17.value='<div class="cards17">'+card17(
                "STATUS","NO DATA","Adjust filters"
            )+"</div>"

            display(HTML(
                '<div class="txt17">No keywords meet the selected criteria.</div>'
            ))
            return

        strongest=d.loc[d["Effect_Size"].abs().idxmax()]
        highest=d.nlargest(1,"Median_Lift_%").iloc[0]

        sig_n=int(d["Significant"].sum())
        practical_n=int((d["Practical_Signal"]=="YES").sum())

        kpi17.value='<div class="cards17">'+"".join([
            card17(
                "KEYWORDS",len(d),"Eligible tests"
            ),
            card17(
                "SIGNIFICANT",sig_n,"Adjusted p < 0.05"
            ),
            card17(
                "PRACTICAL",practical_n,"|Effect| ≥ 0.10"
            ),
            card17(
                "STRONGEST",strongest["Keyword"],
                f"{strongest['Effect_Size']:+.3f}"
            ),
            card17(
                "TOP PAY LIFT",highest["Keyword"],
                f"{highest['Median_Lift_%']:+.1f}%"
            ),
            card17(
                "MIN SAMPLE",f"{min_n17.value:,}",
                "Keyword observations"
            )
        ])+"</div>"

        v=view17.value

        if v=="overview":
            pos=d[d["Direction"].eq("Higher Pay")]
            neg=d[d["Direction"].eq("Lower Pay")]

            display(HTML(f"""
            <div class="grid17">

            <div class="box17">
            <b>STATISTICAL EVIDENCE</b>
            {sig_n} of {len(d)} eligible keywords remain significant
            after multiple-testing correction.
            </div>

            <div class="box17">
            <b>PRACTICAL EFFECT</b>
            {practical_n} keywords combine statistical significance
            with an absolute effect size of at least 0.10.
            </div>

            <div class="box17">
            <b>DIRECTION</b>
            Higher-pay signals: {len(pos)}<br>
            Lower-pay signals: {len(neg)}
            </div>

            </div><br>

            <div class="note17">
            Statistical significance is not treated as causation.
            Effect size, compensation lift and sample size are considered
            alongside adjusted p-values.
            </div>
            """))

        elif v=="significance":
            z=d.nlargest(
                top_n17.value,
                "Effect_Size",
                keep="all"
            ).copy()

            z["NegLogQ"]=-np.log10(
                z["Adjusted_P"].clip(lower=1e-300)
            )

            fig=go.Figure(go.Scatter(
                x=z["Effect_Size"],
                y=z["NegLogQ"],
                mode="markers+text",
                text=z["Keyword"],
                textposition="top center",
                marker=dict(
                    size=np.clip(np.sqrt(z["Keyword_N"]),10,38),
                    color=z["Median_Lift_%"],
                    colorscale=[
                        [0,"#c9ada7"],
                        [.5,"#f2e9e4"],
                        [1,"#4a4e69"]
                    ],
                    showscale=True,
                    colorbar=dict(title="Median Lift %"),
                    line=dict(color="#22223b",width=.5)
                ),
                customdata=z[[
                    "Keyword_N","Adjusted_P","Median_Lift_%"
                ]],
                hovertemplate=(
                    "<b>%{text}</b><br>"
                    "Effect: %{x:+.3f}<br>"
                    "-log10(q): %{y:.2f}<br>"
                    "N: %{customdata[0]:,}<br>"
                    "Adjusted p: %{customdata[1]:.4g}<br>"
                    "Median lift: %{customdata[2]:+.1f}%"
                    "<extra></extra>"
                )
            ))

            fig.add_hline(
                y=-np.log10(.05),
                line_dash="dot",
                line_color="#9a8c98"
            )

            fig.add_vline(
                x=0,
                line_dash="dot",
                line_color="#9a8c98"
            )

            fig.update_xaxes(title="Rank-Biserial Effect Size")
            fig.update_yaxes(title="-log10 Adjusted P")

            fig.update_layout(
                height=520,
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",
                font=dict(family="Arial",color="#22223b"),
                margin=dict(l=70,r=50,t=25,b=55)
            )

            display(fig)

        elif v=="effect":
            z=d.assign(
                Abs_Effect=d["Effect_Size"].abs()
            ).nlargest(
                top_n17.value,
                "Abs_Effect"
            ).sort_values("Effect_Size")

            fig=go.Figure(go.Bar(
                x=z["Effect_Size"],
                y=z["Keyword"],
                orientation="h",
                marker=dict(
                    color=z["Effect_Size"],
                    colorscale=[
                        [0,"#c9ada7"],
                        [.5,"#f2e9e4"],
                        [1,"#4a4e69"]
                    ],
                    cmin=-max(z["Effect_Size"].abs().max(),.01),
                    cmax=max(z["Effect_Size"].abs().max(),.01)
                ),
                customdata=z[[
                    "Keyword_N",
                    "Effect_Strength",
                    "Adjusted_P"
                ]],
                hovertemplate=(
                    "<b>%{y}</b><br>"
                    "Effect: %{x:+.3f}<br>"
                    "Strength: %{customdata[1]}<br>"
                    "N: %{customdata[0]:,}<br>"
                    "Adjusted p: %{customdata[2]:.4g}"
                    "<extra></extra>"
                )
            ))

            fig.add_vline(
                x=0,
                line_dash="dot",
                line_color="#9a8c98"
            )

            fig.update_xaxes(title="Rank-Biserial Effect Size")

            fig.update_layout(
                height=max(450,len(z)*35),
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",
                font=dict(family="Arial",color="#22223b"),
                margin=dict(l=140,r=35,t=20,b=45)
            )

            display(fig)

        elif v=="lift":
            z=d.dropna(
                subset=["Median_Lift_%"]
            ).assign(
                Abs_Lift=lambda q:q["Median_Lift_%"].abs()
            ).nlargest(
                top_n17.value,
                "Abs_Lift"
            ).sort_values("Median_Lift_%")

            fig=go.Figure(go.Bar(
                x=z["Median_Lift_%"],
                y=z["Keyword"],
                orientation="h",
                marker=dict(
                    color=z["Median_Lift_%"],
                    colorscale=[
                        [0,"#c9ada7"],
                        [.5,"#f2e9e4"],
                        [1,"#4a4e69"]
                    ]
                ),
                customdata=z[[
                    "Keyword_Median",
                    "Reference_Median",
                    "Keyword_N"
                ]],
                hovertemplate=(
                    "<b>%{y}</b><br>"
                    "Median lift: %{x:+.1f}%<br>"
                    "Keyword median: %{customdata[0]:,.2f}<br>"
                    "Reference median: %{customdata[1]:,.2f}<br>"
                    "N: %{customdata[2]:,}"
                    "<extra></extra>"
                )
            ))

            fig.add_vline(
                x=0,
                line_dash="dot",
                line_color="#9a8c98"
            )

            fig.update_xaxes(title="Median Compensation Lift (%)")

            fig.update_layout(
                height=max(450,len(z)*35),
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",
                font=dict(family="Arial",color="#22223b"),
                margin=dict(l=140,r=35,t=20,b=45)
            )

            display(fig)

        elif v=="data":
            cols=[
                "Keyword","Keyword_N","Keyword_Median",
                "Reference_Median","Median_Lift_%",
                "Effect_Size","Effect_Strength",
                "Adjusted_P","Significant",
                "Practical_Signal"
            ]

            table17(
                d[cols].sort_values(
                    ["Practical_Signal","Adjusted_P","Effect_Size"],
                    ascending=[False,True,False]
                ).head(100)
            )

        elif v=="findings":
            practical=d[d["Practical_Signal"].eq("YES")]

            positive=practical.nlargest(
                min(5,len(practical)),
                "Effect_Size"
            )

            negative=practical.nsmallest(
                min(5,len(practical)),
                "Effect_Size"
            )

            display(HTML(f"""
            <div class="grid17">

            <div class="box17">
            <b>STRONGEST POSITIVE SIGNALS</b>
            {
                "<br>".join(
                    f"{r.Keyword}: {r.Effect_Size:+.3f}"
                    for _,r in positive.iterrows()
                ) if len(positive)
                else "No practical positive signal."
            }
            </div>

            <div class="box17">
            <b>STRONGEST NEGATIVE SIGNALS</b>
            {
                "<br>".join(
                    f"{r.Keyword}: {r.Effect_Size:+.3f}"
                    for _,r in negative.iterrows()
                ) if len(negative)
                else "No practical negative signal."
            }
            </div>

            <div class="box17">
            <b>INTERPRETATION</b>
            Statistical association identifies compensation differences
            linked with title language, but does not establish that a
            keyword itself causes higher or lower pay.
            </div>

            </div>
            """))

        elif v=="evidence":
            evidence17=pd.DataFrame({
                "Evidence":[
                    "Keyword statistical tests",
                    "Validation summary",
                    "Adjusted significance",
                    "Effect-size classification"
                ],
                "Output":[
                    "task1_statistical_tests.csv",
                    "task1_statistical_validation_summary.csv",
                    "Benjamini-Hochberg adjusted p-values",
                    "Rank-biserial effect size"
                ],
                "Status":["Created"]*4
            })

            table17(evidence17)

        else:
            display(HTML("""
            <div class="txt17">

            <b>Eligibility:</b>
            Each tested keyword requires at least 100 compensation-valid
            postings and a sufficiently large reference group.<br><br>

            <b>Statistical test:</b>
            Mann–Whitney U is used because compensation distributions may
            be skewed and do not require a normality assumption.<br><br>

            <b>Effect size:</b>
            Rank-biserial correlation expresses the direction and magnitude
            of the keyword-versus-reference difference.<br><br>

            <b>Multiple testing:</b>
            Benjamini–Hochberg correction controls the false-discovery rate
            when many title keywords are evaluated simultaneously.<br><br>

            <b>Practical signal:</b>
            Requires adjusted p &lt; 0.05 and |effect size| ≥ 0.10.<br><br>

            <b>Important:</b>
            Hourly rates and fixed-price budgets remain completely separate.
            Association does not imply causation.

            </div>
            """))


for w in [pay17,sig17,min_n17,top_n17,view17]:
    w.observe(update17,names="value")

update17()

dashboard_17=widgets.VBox([
    header17,
    kpi17,
    widgets.HTML(
        '<div class="ctrl17"><b style="font:11px Arial;color:#22223b">'
        'STATISTICAL EVIDENCE CONTROLS</b></div>'
    ),
    controls17,
    view17,
    output17
])

dashboard_17.add_class("d17")
display(dashboard_17)

if "logger" in globals():
    logger.info(
        "SECTION17_COMPLETE | tests=%s | significant=%s | practical=%s",
        len(task1_tests),
        int(task1_tests["Significant"].sum()),
        int((task1_tests["Practical_Signal"]=="YES").sum())
    )
```

<div style="color:#f2e9e4;font-size:12px;line-height:1.65;margin-top:8px">Validating near-term job-demand forecasts against transparent baselines, selecting the strongest method by role, auditing forecast errors and preserving versioned model artifacts for reproducible deployment.</div>

```python
required=[
    "daily","Xmat","dense_start","trend_end",
    "horizon","forecast_summary"
]
missing=[x for x in required if x not in globals()]
if missing:
    raise NameError(f"Run Section 09 first. Missing: {missing}")

MODEL_DIR=Path("models")
MODEL_DIR.mkdir(exist_ok=True)
MODEL_VERSION="demand_v1"
FEATURES18=[
    "intercept","time_trend",
    "monday","tuesday","wednesday",
    "thursday","friday","saturday"
]

def metrics18(actual,pred):
    actual=np.asarray(actual,dtype=float)
    pred=np.asarray(pred,dtype=float)
    err=actual-pred
    mae=np.mean(np.abs(err))
    rmse=np.sqrt(np.mean(err**2))
    den=np.abs(actual)+np.abs(pred)
    valid=den>0
    smape=(
        np.mean(2*np.abs(err[valid])/den[valid])*100
        if valid.any() else 0
    )

    wape=np.abs(err).sum()/max(np.abs(actual).sum(),1)*100

    return mae,rmse,smape,wape


comparison18=[]
errors18=[]
registry18=[]
future18=[]
artifact_models18={}

for role,g in daily.groupby("job_family"):
    g=g.sort_values("published_day").reset_index(drop=True)
    train=g.iloc[:-horizon].copy()
    test=g.iloc[-horizon:].copy()
    start=train["published_day"].iloc[0]
    Xtr=Xmat(train["published_day"],start)
    Xte=Xmat(test["published_day"],start)

    beta=np.linalg.lstsq(
        Xtr,
        train["Actual"].to_numpy(),
        rcond=None
    )[0]

    trend_pred=np.clip(Xte@beta,0,None)

    lag7_pred=(
        g["Actual"]
        .shift(7)
        .iloc[-horizon:]
        .to_numpy(dtype=float)
    )

    mean7_pred=np.repeat(
        train["Actual"].tail(7).mean(),
        horizon
    )

    actual=test["Actual"].to_numpy(dtype=float)

    candidates={
        "Trend + Weekday":trend_pred,
        "Lag-7":lag7_pred,
        "7-Day Mean":mean7_pred
    }

    role_metrics=[]

    for model,pred in candidates.items():

        mae,rmse,smape,wape=metrics18(actual,pred)

        row={
            "job_family":role,
            "Model":model,
            "MAE":mae,
            "RMSE":rmse,
            "SMAPE_%":smape,
            "WAPE_%":wape
        }

        comparison18.append(row)
        role_metrics.append(row)

        for d,a,p in zip(
            test["published_day"],actual,pred
        ):
            errors18.append({
                "job_family":role,
                "Date":d,
                "Model":model,
                "Actual":a,
                "Predicted":p,
                "Residual":a-p,
                "Absolute_Error":abs(a-p)
            })

    best=min(
        role_metrics,
        key=lambda x:(x["MAE"],x["RMSE"])
    )

    selected=best["Model"]

    Xall=Xmat(
        g["published_day"],
        g["published_day"].iloc[0]
    )

    beta_full=np.linalg.lstsq(
        Xall,
        g["Actual"].to_numpy(),
        rcond=None
    )[0]

    future_dates=pd.date_range(
        trend_end+pd.Timedelta(days=1),
        periods=horizon,
        freq="D",
        tz="UTC"
    )

    Xf=Xmat(
        pd.Series(future_dates),
        g["published_day"].iloc[0]
    )

    future_candidates={
        "Trend + Weekday":np.clip(Xf@beta_full,0,None),
        "Lag-7":g["Actual"].tail(horizon).to_numpy(dtype=float),
        "7-Day Mean":np.repeat(
            g["Actual"].tail(7).mean(),
            horizon
        )
    }

    selected_future=future_candidates[selected]

    for d,p in zip(future_dates,selected_future):
        future18.append({
            "job_family":role,
            "forecast_day":d,
            "Selected_Model":selected,
            "Forecast":p
        })

    registry18.append({
        "Model_ID":f"{MODEL_VERSION}_{role.lower().replace(' ','_').replace('&','and')}",
        "job_family":role,
        "Selected_Model":selected,
        "MAE":best["MAE"],
        "RMSE":best["RMSE"],
        "SMAPE_%":best["SMAPE_%"],
        "WAPE_%":best["WAPE_%"],
        "Validation_Train_Start":train["published_day"].min(),
        "Validation_Train_End":train["published_day"].max(),
        "Validation_Test_Start":test["published_day"].min(),
        "Validation_Test_End":test["published_day"].max(),
        "Final_Refit_End":g["published_day"].max(),
        "Horizon_Days":horizon,
        "Version":MODEL_VERSION
    })

    artifact_models18[role]={
        "selected_model":selected,
        "trend_beta":beta_full,
        "start_date":g["published_day"].iloc[0],
        "last_7_actual":g["Actual"].tail(7).to_numpy(dtype=float),
        "recent_7_mean":float(g["Actual"].tail(7).mean())
    }


model_comparison18=pd.DataFrame(comparison18)
model_errors18=pd.DataFrame(errors18)
model_registry18=pd.DataFrame(registry18)
selected_forecast_daily18=pd.DataFrame(future18)

winner18=(
    model_registry18["Selected_Model"]
    .value_counts()
    .rename_axis("Model")
    .reset_index(name="Roles_Won")
)

model_summary18=(
    model_comparison18.groupby("Model")
    .agg(
        Mean_MAE=("MAE","mean"),
        Median_MAE=("MAE","median"),
        Mean_RMSE=("RMSE","mean"),
        Mean_SMAPE=("SMAPE_%","mean"),
        Mean_WAPE=("WAPE_%","mean")
    )
    .reset_index()
    .merge(winner18,on="Model",how="left")
    .fillna({"Roles_Won":0})
)

model_summary18["Roles_Won"]=model_summary18["Roles_Won"].astype(int)

selected_forecast_summary18=(
    selected_forecast_daily18.groupby(
        ["job_family","Selected_Model"]
    )
    .agg(
        Forecast_7d=("Forecast","sum"),
        Forecast_Avg_Day=("Forecast","mean")
    )
    .reset_index()
    .merge(
        model_registry18[[
            "job_family","MAE","RMSE",
            "SMAPE_%","WAPE_%"
        ]],
        on="job_family",
        how="left"
    )
    .sort_values("Forecast_7d",ascending=False)
)

model_artifact18={
    "version":MODEL_VERSION,
    "created_at":pd.Timestamp.now(tz="UTC"),
    "forecast_horizon_days":horizon,
    "dense_window_start":dense_start,
    "final_refit_end":trend_end,
    "feature_names":FEATURES18,
    "selection_rule":"Lowest chronological holdout MAE; RMSE used as tie-breaker",
    "models":artifact_models18
}

model_path18=MODEL_DIR/f"{MODEL_VERSION}.pkl"

with open(model_path18,"wb") as f:
    pickle.dump(model_artifact18,f)

metadata18={
    "model_version":MODEL_VERSION,
    "task":"High-Demand Role Forecasting",
    "frequency":"Daily",
    "horizon_days":int(horizon),
    "validation":"Chronological final 7 complete days",
    "candidate_models":[
        "Trend + Weekday",
        "Lag-7",
        "7-Day Mean"
    ],
    "selection_metric":"MAE",
    "tie_breaker":"RMSE",
    "negative_forecasts":"Clipped to zero",
    "training_window_start":str(dense_start),
    "final_refit_end":str(trend_end),
    "roles":int(model_registry18["job_family"].nunique()),
    "artifact":str(model_path18)
}

metadata_path18=MODEL_DIR/f"{MODEL_VERSION}_metadata.json"

with open(metadata_path18,"w",encoding="utf-8") as f:
    json.dump(metadata18,f,indent=2)

smoke_script18="""
import pickle,sys
p=sys.argv[1]
with open(p,"rb") as f:
    x=pickle.load(f)
assert "models" in x
assert len(x["models"])>0
assert "feature_names" in x
print(f"PASS | roles={len(x['models'])}")
"""

smoke18=subprocess.run(
    [sys.executable,"-c",smoke_script18,str(model_path18)],
    capture_output=True,
    text=True
)

smoke_status18="PASS" if smoke18.returncode==0 else "FAIL"
smoke_message18=(
    smoke18.stdout.strip()
    if smoke18.returncode==0
    else smoke18.stderr.strip()
)

model_comparison18.to_csv(
    EXPORT_DIR/"task3_model_comparison.csv",
    index=False
)

model_errors18.to_csv(
    EXPORT_DIR/"task3_error_analysis.csv",
    index=False
)

model_registry18.to_csv(
    EXPORT_DIR/"model_registry.csv",
    index=False
)

selected_forecast_daily18.to_csv(
    EXPORT_DIR/"task3_selected_daily_forecast.csv",
    index=False
)

selected_forecast_summary18.to_csv(
    EXPORT_DIR/"task3_selected_forecast_summary.csv",
    index=False
)

with open(
    EXPORT_DIR/"task3_model_smoke_test.txt",
    "w",encoding="utf-8"
) as f:
    f.write(
        f"STATUS={smoke_status18}\n"
        f"{smoke_message18}"
    )


def card18(a,b,c):
    return f"""
    <div class="c18">
    <small>{a}</small>
    <b>{b}</b>
    <span>{c}</span>
    </div>
    """

def table18(d,n=None):
    z=d.head(n) if n else d
    display(HTML(z.to_html(
        index=False,
        border=0,
        classes="t18",
        float_format=lambda x:f"{x:,.2f}"
    )))

display(HTML("""
<style>
.d18{
border:1px solid #ddd7d4;border-radius:18px;overflow:hidden;
background:#fff;box-shadow:0 8px 26px #22223b18
}
.h18{
background:linear-gradient(120deg,#22223b,#30324d 56%,#4a4e69);
padding:26px 29px;display:flex;justify-content:space-between;
font-family:Arial
}
.h18 .k{color:#c9ada7;font:700 8px Arial;letter-spacing:1.7px}
.h18 h2{color:#fff;margin:6px 0;font:700 25px Arial}
.h18 p,.h18 .m{color:#f2e9e4;font:10px/1.65 Arial}

.cards18{
display:grid;grid-template-columns:repeat(6,1fr);
border-bottom:1px solid #ece7e4
}
.c18{
padding:14px;border-right:1px solid #eee9e7;
font-family:Arial;min-width:0
}
.c18 small{color:#9a8c98;font-size:7px;font-weight:700}
.c18 b{
display:block;color:#22223b;font-size:17px;
margin-top:5px;overflow-wrap:anywhere
}
.c18 span{
display:block;color:#4a4e69;font-size:8px;margin-top:4px
}

.ctrl18{padding:12px 18px 4px;background:#faf9f8}
.btn18{margin:8px 15px!important}
.out18{
border-top:1px solid #eee9e7!important;
padding:20px 23px
}

.t18{
width:100%;border-collapse:collapse;font:10px Arial
}
.t18 th{
background:#22223b;color:#fff;padding:9px;text-align:left
}
.t18 td{
padding:8px;border-bottom:1px solid #eee9e7;color:#22223b
}

.grid18{
display:grid;grid-template-columns:repeat(3,1fr);gap:12px
}
.box18{
background:#f8f6f5;border-left:4px solid #4a4e69;
padding:14px;border-radius:9px;
font:11px/1.65 Arial;color:#22223b
}
.box18 b{
display:block;color:#4a4e69;font-size:9px;margin-bottom:6px
}
.note18{
background:#f2e9e4;padding:12px 14px;border-radius:9px;
font:10px/1.65 Arial;color:#4a4e69
}
.txt18{font:12px/1.8 Arial;color:#22223b}

@media(max-width:900px){
.cards18{grid-template-columns:repeat(3,1fr)}
.grid18{grid-template-columns:1fr}
}
</style>
"""))

header18=widgets.HTML(f"""
<div class="h18">
<div>
<div class="k">TASK 03 • MODEL GOVERNANCE</div>
<h2>Demand Forecast Validation Center</h2>
<p>Model Comparison • Error Analysis • Selection • Registry • Reproducibility</p>
</div>

<div class="m">
MODEL VERSION<br>
<b>{MODEL_VERSION}</b><br><br>
LOAD TEST<br>
<b>{smoke_status18}</b>
</div>
</div>
""")

role18=widgets.Dropdown(
    options=model_registry18["job_family"].tolist(),
    description="Role:",
    layout=widgets.Layout(width="300px")
)

metric18=widgets.Dropdown(
    options=["MAE","RMSE","SMAPE_%","WAPE_%"],
    value="MAE",
    description="Metric:",
    layout=widgets.Layout(width="220px")
)

view18=widgets.ToggleButtons(
    options=[
        ("OVERVIEW","overview"),
        ("COMPARISON","comparison"),
        ("PERFORMANCE","performance"),
        ("ERROR ANALYSIS","error"),
        ("FORECAST","forecast"),
        ("REGISTRY","registry"),
        ("ROLE DETAIL","detail"),
        ("EVIDENCE","evidence"),
        ("METHOD","method")
    ],
    value="overview",
    layout=widgets.Layout(width="100%"),
    style={"button_width":"11%","font_weight":"700"}
)
view18.add_class("btn18")

controls18=widgets.HBox(
    [role18,metric18],
    layout=widgets.Layout(
        padding="8px 18px",
        flex_flow="row wrap",
        gap="10px"
    )
)

kpi18=widgets.HTML()
output18=widgets.Output()
output18.add_class("out18")


def update18(*_):

    with output18:
        clear_output(wait=True)

        role=role18.value
        reg=model_registry18[
            model_registry18["job_family"].eq(role)
        ].iloc[0]

        best_overall=model_summary18.nsmallest(
            1,"Mean_MAE"
        ).iloc[0]

        avg_mae=model_registry18["MAE"].mean()

        trend_wins=int(
            model_registry18["Selected_Model"]
            .eq("Trend + Weekday").sum()
        )

        kpi18.value='<div class="cards18">'+"".join([
            card18(
                "ROLES",
                model_registry18["job_family"].nunique(),
                "Validated"
            ),
            card18(
                "BEST OVERALL",
                best_overall["Model"],
                f"Mean MAE {best_overall['Mean_MAE']:.1f}"
            ),
            card18(
                "AVG SELECTED MAE",
                f"{avg_mae:.1f}",
                "Holdout"
            ),
            card18(
                "TREND MODEL WINS",
                trend_wins,
                "Role-level selections"
            ),
            card18(
                "HORIZON",
                f"{horizon} days",
                "Near-term"
            ),
            card18(
                "LOAD TEST",
                smoke_status18,
                "Fresh process"
            )
        ])+"</div>"

        v=view18.value

        if v=="overview":

            counts=(
                model_registry18["Selected_Model"]
                .value_counts()
            )

            display(HTML(f"""
            <div class="grid18">

            <div class="box18">
            <b>MODEL SELECTION</b>
            Each job family is evaluated against three methods using
            the same chronological holdout period.
            </div>

            <div class="box18">
            <b>WINNING METHODS</b>
            {"<br>".join(
                f"{m}: {n} role(s)"
                for m,n in counts.items()
            )}
            </div>

            <div class="box18">
            <b>DEPLOYMENT READINESS</b>
            Saved artifact: {MODEL_VERSION}.pkl<br>
            Independent load test: {smoke_status18}
            </div>

            </div><br>

            <div class="note18">
            The selected model is determined by out-of-sample MAE,
            with RMSE used only as a tie-breaker. Training performance
            is never presented as validation performance.
            </div>
            """))

        elif v=="comparison":

            z=model_summary18.sort_values(
                metric18.value
            )

            fig=go.Figure()

            fig.add_trace(go.Bar(
                x=z["Model"],
                y=z[metric18.value],
                marker_color="#4a4e69",
                text=z[metric18.value].map(
                    lambda x:f"{x:.1f}"
                ),
                textposition="outside",
                customdata=z[[
                    "Roles_Won",
                    "Mean_MAE",
                    "Mean_RMSE"
                ]],
                hovertemplate=(
                    "<b>%{x}</b><br>"
                    f"{metric18.value}: %{{y:.2f}}<br>"
                    "Roles won: %{customdata[0]}<br>"
                    "Mean MAE: %{customdata[1]:.2f}<br>"
                    "Mean RMSE: %{customdata[2]:.2f}"
                    "<extra></extra>"
                )
            ))

            fig.update_yaxes(
                title=f"{metric18.value} — Lower is Better"
            )

            fig.update_layout(
                height=470,
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",
                font=dict(
                    family="Arial",
                    color="#22223b"
                ),
                margin=dict(
                    l=70,r=30,t=25,b=50
                )
            )

            display(fig)

        elif v=="performance":

            z=model_comparison18[
                model_comparison18[
                    "job_family"
                ].eq(role)
            ].sort_values(metric18.value)

            fig=go.Figure(go.Bar(
                x=z[metric18.value],
                y=z["Model"],
                orientation="h",
                marker_color="#4a4e69",
                text=z[metric18.value].map(
                    lambda x:f"{x:.2f}"
                ),
                textposition="outside"
            ))

            fig.update_xaxes(
                title=f"{metric18.value} — Lower is Better"
            )

            fig.update_layout(
                height=400,
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",
                font=dict(
                    family="Arial",
                    color="#22223b"
                ),
                margin=dict(
                    l=130,r=40,t=20,b=45
                )
            )

            display(fig)

        elif v=="error":

            selected=reg["Selected_Model"]

            z=model_errors18[
                (model_errors18["job_family"].eq(role)) &
                (model_errors18["Model"].eq(selected))
            ].sort_values("Date")

            fig=go.Figure()

            fig.add_trace(go.Scatter(
                x=z["Date"],
                y=z["Actual"],
                mode="lines+markers",
                name="Actual",
                line=dict(color="#22223b",width=2)
            ))

            fig.add_trace(go.Scatter(
                x=z["Date"],
                y=z["Predicted"],
                mode="lines+markers",
                name="Predicted",
                line=dict(color="#9a8c98",width=2)
            ))

            fig.update_yaxes(
                title="Daily Job Postings"
            )

            fig.update_layout(
                height=470,
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",
                font=dict(
                    family="Arial",
                    color="#22223b"
                ),
                legend=dict(
                    orientation="h",
                    y=1.08
                ),
                margin=dict(
                    l=70,r=30,t=25,b=50
                )
            )

            display(fig)

        elif v=="forecast":

            z=selected_forecast_daily18[
                selected_forecast_daily18[
                    "job_family"
                ].eq(role)
            ]

            fig=go.Figure(go.Scatter(
                x=z["forecast_day"],
                y=z["Forecast"],
                mode="lines+markers",
                fill="tozeroy",
                line=dict(
                    color="#4a4e69",
                    width=3
                ),
                hovertemplate=(
                    "<b>%{x|%d %b %Y}</b><br>"
                    "Forecast: %{y:.1f}"
                    "<extra></extra>"
                )
            ))

            fig.update_yaxes(
                title="Forecast Job Postings"
            )

            fig.update_layout(
                height=450,
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",
                font=dict(
                    family="Arial",
                    color="#22223b"
                ),
                margin=dict(
                    l=70,r=30,t=25,b=50
                )
            )

            display(fig)

        elif v=="registry":

            table18(model_registry18)

        elif v=="detail":

            display(HTML(f"""
            <div class="grid18">

            <div class="box18">
            <b>SELECTED MODEL</b>
            {reg['Selected_Model']}<br>
            Version: {reg['Version']}
            </div>

            <div class="box18">
            <b>VALIDATION PERFORMANCE</b>
            MAE: {reg['MAE']:.2f}<br>
            RMSE: {reg['RMSE']:.2f}<br>
            SMAPE: {reg['SMAPE_%']:.2f}%<br>
            WAPE: {reg['WAPE_%']:.2f}%
            </div>

            <div class="box18">
            <b>TIME SPLIT</b>
            Train end:
            {reg['Validation_Train_End']:%d %b %Y}<br>
            Test:
            {reg['Validation_Test_Start']:%d %b} –
            {reg['Validation_Test_End']:%d %b %Y}
            </div>

            </div>
            """))

        elif v=="evidence":

            evidence18=pd.DataFrame({
                "Evidence":[
                    "Model comparison",
                    "Forecast error analysis",
                    "Model registry",
                    "Selected daily forecasts",
                    "Selected forecast summary",
                    "Versioned model artifact",
                    "Model metadata",
                    "Independent load test"
                ],
                "Output":[
                    "task3_model_comparison.csv",
                    "task3_error_analysis.csv",
                    "model_registry.csv",
                    "task3_selected_daily_forecast.csv",
                    "task3_selected_forecast_summary.csv",
                    f"models/{MODEL_VERSION}.pkl",
                    f"models/{MODEL_VERSION}_metadata.json",
                    "task3_model_smoke_test.txt"
                ],
                "Status":[
                    "Created","Created","Created",
                    "Created","Created","Created",
                    "Created",smoke_status18
                ]
            })

            table18(evidence18)

        else:

            display(HTML(f"""
            <div class="txt18">

            <b>Candidate 1 — Trend + Weekday:</b>
            Least-squares daily demand model using a linear time trend
            and weekday indicators.<br><br>

            <b>Candidate 2 — Lag-7:</b>
            Transparent seasonal baseline using demand from the
            corresponding weekday one week earlier.<br><br>

            <b>Candidate 3 — 7-Day Mean:</b>
            Simple recent-demand baseline using the mean of the seven
            preceding training days.<br><br>

            <b>Validation:</b>
            The final {horizon} complete days are held out chronologically.
            No random train/test split is used.<br><br>

            <b>Metrics:</b>
            MAE is the primary selection metric. RMSE is the tie-breaker.
            SMAPE and WAPE provide additional proportional error context.<br><br>

            <b>Final refit:</b>
            After validation, the trend model is refitted on all available
            complete-day observations. Baseline models use the latest
            observed seven-day history.<br><br>

            <b>Selection:</b>
            Each job family uses the candidate with the lowest holdout MAE.
            This allows a simple baseline to remain preferred where it
            genuinely outperforms the regression model.<br><br>

            <b>Reproducibility:</b>
            The model registry records validation dates, metrics,
            selected method and version. The saved artifact is loaded in
            a separate Python process as a smoke test.<br><br>

            <b>Scope:</b>
            These are short-horizon demand forecasts. They should not be
            interpreted as long-term labour-market projections.

            </div>
            """))


for w in [role18,metric18,view18]:
    w.observe(update18,names="value")

update18()

dashboard_18=widgets.VBox([
    header18,
    kpi18,
    widgets.HTML(
        '<div class="ctrl18"><b style="font:11px Arial;color:#22223b">'
        'MODEL VALIDATION CONTROLS</b></div>'
    ),
    controls18,
    view18,
    output18
])

dashboard_18.add_class("d18")
display(dashboard_18)

if "logger" in globals():
    logger.info(
        "SECTION18_COMPLETE | roles=%s | models=%s | avg_selected_mae=%.2f | smoke=%s",
        model_registry18["job_family"].nunique(),
        model_comparison18["Model"].nunique(),
        model_registry18["MAE"].mean(),
        smoke_status18
    )
```

```python
required=["rec","vectorizer","tfidf_matrix"]
missing=[x for x in required if x not in globals()]
if missing:
    raise NameError(f"Run Section 11 first. Missing: {missing}")

REC_ASSET_DIR=Path("models/recommender_assets")
REC_ASSET_DIR.mkdir(parents=True,exist_ok=True)
RECOMMENDER_VERSION="recommender_v1"
TOP_K19=10

weight_sets19={
    "Balanced":{
        "title":.75,
        "country":.10,
        "pay":.05,
        "seniority":.05,
        "recency":.05
    },
    "Title Heavy":{
        "title":.85,
        "country":.05,
        "pay":.03,
        "seniority":.03,
        "recency":.04
    },
    "Preference Aware":{
        "title":.65,
        "country":.15,
        "pay":.07,
        "seniority":.08,
        "recency":.05
    }
}

def recommend19(
    query,
    country="Any",
    pay="Any",
    seniority="Any",
    n=10,
    mode="boost",
    weights=None
):
    weights=weights or weight_sets19["Balanced"]

    q=str(query).strip().lower()

    if len(q)<2:
        return pd.DataFrame()

    n=max(1,min(int(n),50))

    sim=cosine_similarity(
        vectorizer.transform([q]),
        tfidf_matrix
    ).ravel()

    d=rec.copy()

    d["Title_Similarity"]=sim

    d["Country_Match"]=(
        1.0 if country=="Any"
        else d["country_clean"].eq(country).astype(float)
    )

    d["Pay_Match"]=(
        1.0 if pay=="Any"
        else d["pay_type"].eq(pay).astype(float)
    )

    d["Seniority_Match"]=(
        1.0 if seniority=="Any"
        else d["seniority_level"].eq(seniority).astype(float)
    )

    if mode=="strict":
        if country!="Any":
            d=d[d["Country_Match"].eq(1)]

        if pay!="Any":
            d=d[d["Pay_Match"].eq(1)]

        if seniority!="Any":
            d=d[d["Seniority_Match"].eq(1)]

    active=[
        ("Title_Similarity",weights["title"]),
        ("recency_score",weights["recency"])
    ]

    if country!="Any":
        active.append(("Country_Match",weights["country"]))

    if pay!="Any":
        active.append(("Pay_Match",weights["pay"]))

    if seniority!="Any":
        active.append(("Seniority_Match",weights["seniority"]))

    denom=sum(w for _,w in active)

    score=sum(
        d[col]*w
        for col,w in active
    )/denom

    d["Match_Score"]=score

    d=d[
        d["Title_Similarity"]>0
    ].nlargest(
        n,
        "Match_Score"
    ).copy()

    d["Match_%"]=(d["Match_Score"]*100).round(1)

    reasons=[]

    for _,r in d.iterrows():
        x=[
            f"title similarity {r['Title_Similarity']:.0%}"
        ]

        if country!="Any" and r["Country_Match"]==1:
            x.append("country match")

        if pay!="Any" and r["Pay_Match"]==1:
            x.append("pay-type match")

        if seniority!="Any" and r["Seniority_Match"]==1:
            x.append("seniority match")

        if r["recency_score"]>=.75:
            x.append("recent posting")

        reasons.append(", ".join(x))

    d["Why_Matched"]=reasons

    return d[[
        "job_id",
        "title",
        "country_clean",
        "pay_type",
        "job_family",
        "seniority_level",
        "hourly_midpoint",
        "fixed_budget",
        "published_date",
        "link",
        "Title_Similarity",
        "Country_Match",
        "Pay_Match",
        "Seniority_Match",
        "recency_score",
        "Match_Score",
        "Match_%",
        "Why_Matched"
    ]]


query_map19={
    "Data & AI":"data analyst",
    "Software Development":"software developer",
    "Design & Creative":"graphic designer",
    "Marketing & Sales":"marketing specialist",
    "Writing & Translation":"content writer",
    "Finance & Accounting":"accountant",
    "Admin & Support":"virtual assistant",
    "Engineering & Architecture":"engineer",
    "Legal":"legal"
}

profiles19=[]

for family,query in query_map19.items():

    g=rec[
        rec["job_family"].eq(family)
    ].copy()

    if g.empty:
        continue

    def mode19(s):
        s=s.dropna()
        return s.mode().iloc[0] if len(s) else "Any"

    profiles19.append({
        "Profile":family,
        "Query":query,
        "Expected_Family":family,
        "Country":mode19(g["country_clean"]),
        "Pay_Type":mode19(g["pay_type"]),
        "Seniority":mode19(g["seniority_level"])
    })

profiles19=pd.DataFrame(profiles19)


def relevance19(results,profile):

    if results.empty:
        return pd.Series(dtype=bool)

    relevant=results[
        "job_family"
    ].eq(profile["Expected_Family"])

    if profile["Country"]!="Any":
        relevant&=results[
            "country_clean"
        ].eq(profile["Country"])

    if profile["Pay_Type"]!="Any":
        relevant&=results[
            "pay_type"
        ].eq(profile["Pay_Type"])

    if profile["Seniority"]!="Any":
        relevant&=results[
            "seniority_level"
        ].eq(profile["Seniority"])

    return relevant


def total_relevant19(profile):

    d=rec.copy()

    m=d[
        "job_family"
    ].eq(profile["Expected_Family"])

    if profile["Country"]!="Any":
        m&=d["country_clean"].eq(
            profile["Country"]
        )

    if profile["Pay_Type"]!="Any":
        m&=d["pay_type"].eq(
            profile["Pay_Type"]
        )

    if profile["Seniority"]!="Any":
        m&=d["seniority_level"].eq(
            profile["Seniority"]
        )

    return int(m.sum())


def evaluate_profile19(profile,weights,k=10):

    r=recommend19(
        profile["Query"],
        country=profile["Country"],
        pay=profile["Pay_Type"],
        seniority=profile["Seniority"],
        n=k,
        mode="boost",
        weights=weights
    )

    rel=relevance19(r,profile).to_numpy(dtype=int)

    precision=rel.sum()/k
    hit=int(rel.sum()>0)

    relevant_ranks=np.where(
        rel==1
    )[0]+1

    mrr=(
        1/relevant_ranks[0]
        if len(relevant_ranks)
        else 0
    )

    discounts=1/np.log2(
        np.arange(2,len(rel)+2)
    )

    dcg=(rel*discounts).sum()

    ideal_n=min(
        total_relevant19(profile),
        k
    )

    idcg=(
        discounts[:ideal_n].sum()
        if ideal_n else 0
    )

    ndcg=dcg/idcg if idcg else 0

    return {
        "Profile":profile["Profile"],
        "Query":profile["Query"],
        "Precision@10":precision,
        "Hit@10":hit,
        "MRR":mrr,
        "NDCG@10":ndcg,
        "Results_Returned":len(r)
    }


weight_eval19=[]

for weight_name,weights in weight_sets19.items():

    for _,profile in profiles19.iterrows():

        row=evaluate_profile19(
            profile,
            weights,
            TOP_K19
        )

        row["Weight_Set"]=weight_name
        weight_eval19.append(row)

weight_eval19=pd.DataFrame(weight_eval19)

weight_summary19=(
    weight_eval19.groupby("Weight_Set")
    .agg(
        Mean_Precision10=("Precision@10","mean"),
        Hit_Rate10=("Hit@10","mean"),
        Mean_MRR=("MRR","mean"),
        Mean_NDCG10=("NDCG@10","mean")
    )
    .reset_index()
)

priority19={
    "Balanced":0,
    "Title Heavy":1,
    "Preference Aware":2
}

weight_summary19["Priority"]=weight_summary19[
    "Weight_Set"
].map(priority19)

best_weight_name19=(
    weight_summary19.sort_values(
        [
            "Mean_NDCG10",
            "Mean_Precision10",
            "Hit_Rate10",
            "Priority"
        ],
        ascending=[
            False,
            False,
            False,
            True
        ]
    )
    .iloc[0]["Weight_Set"]
)

BEST_WEIGHTS19=weight_sets19[
    best_weight_name19
]

evaluation19=[]

for _,profile in profiles19.iterrows():
    evaluation19.append(
        evaluate_profile19(
            profile,
            BEST_WEIGHTS19,
            TOP_K19
        )
    )

evaluation19=pd.DataFrame(evaluation19)


def recommend_jobs(
    query,
    country="Any",
    pay="Any",
    seniority="Any",
    n=10,
    mode="boost"
):
    return recommend19(
        query,
        country=country,
        pay=pay,
        seniority=seniority,
        n=n,
        mode=mode,
        weights=BEST_WEIGHTS19
    )


sample_profile19=profiles19.iloc[0]

run_a19=recommend_jobs(
    sample_profile19["Query"],
    sample_profile19["Country"],
    sample_profile19["Pay_Type"],
    sample_profile19["Seniority"],
    TOP_K19
)

run_b19=recommend_jobs(
    sample_profile19["Query"],
    sample_profile19["Country"],
    sample_profile19["Pay_Type"],
    sample_profile19["Seniority"],
    TOP_K19
)

reproducible19=(
    run_a19["job_id"].tolist()==
    run_b19["job_id"].tolist()
    and
    np.allclose(
        run_a19["Match_Score"],
        run_b19["Match_Score"]
    )
)

edge_cases19=[]

tests19={
    "Empty query":
        lambda: recommend_jobs(""),

    "Single-character query":
        lambda: recommend_jobs("a"),

    "Unknown title":
        lambda: recommend_jobs("zzzxxyyqq"),

    "Large Top-K capped":
        lambda: recommend_jobs(
            "data",
            n=500
        ),

    "Impossible strict country":
        lambda: recommend_jobs(
            "data",
            country="__NO_MATCH__",
            mode="strict"
        )
}

for name,test in tests19.items():

    try:
        result=test()

        passed=(
            len(result)==0
            if name in [
                "Empty query",
                "Single-character query",
                "Unknown title",
                "Impossible strict country"
            ]
            else len(result)<=50
        )

        edge_cases19.append({
            "Test":name,
            "Status":"PASS" if passed else "FAIL",
            "Rows_Returned":len(result),
            "Error":""
        })

    except Exception as e:

        edge_cases19.append({
            "Test":name,
            "Status":"FAIL",
            "Rows_Returned":0,
            "Error":str(e)
        })

edge_cases19=pd.DataFrame(
    edge_cases19
)

example_results19=recommend_jobs(
    sample_profile19["Query"],
    sample_profile19["Country"],
    sample_profile19["Pay_Type"],
    sample_profile19["Seniority"],
    TOP_K19
)

api_request19={
    "query":sample_profile19["Query"],
    "country":sample_profile19["Country"],
    "pay_type":sample_profile19["Pay_Type"],
    "seniority":sample_profile19["Seniority"],
    "top_k":TOP_K19,
    "mode":"boost"
}

api_response19={
    "status":"success",
    "model_version":RECOMMENDER_VERSION,
    "weight_set":best_weight_name19,
    "data_as_of":str(
        rec["published_date"].max()
    ),
    "results":json.loads(
        example_results19.to_json(
            orient="records",
            date_format="iso"
        )
    )
}

rec.to_csv(
    REC_ASSET_DIR/"recommendation_catalog.csv",
    index=False
)

with open(
    REC_ASSET_DIR/"tfidf_vectorizer.pkl",
    "wb"
) as f:
    pickle.dump(vectorizer,f)

save_npz(
    REC_ASSET_DIR/"tfidf_matrix.npz",
    tfidf_matrix
)

with open(
    REC_ASSET_DIR/"weights.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        {
            "version":RECOMMENDER_VERSION,
            "selected_weight_set":best_weight_name19,
            "weights":BEST_WEIGHTS19
        },
        f,
        indent=2
    )

recommender_metadata19={
    "version":RECOMMENDER_VERSION,
    "engine":"TF-IDF title similarity + preference compatibility + recency",
    "similarity":"Cosine similarity",
    "text_fields":["title_normalized"],
    "catalogue_rows":int(len(rec)),
    "vocabulary_size":int(
        len(vectorizer.vocabulary_)
    ),
    "data_as_of":str(
        rec["published_date"].max()
    ),
    "selected_weight_set":best_weight_name19,
    "weights":BEST_WEIGHTS19,
    "max_top_k":50,
    "evaluation_note":
        "Offline metrics use job-family and selected preference consistency as proxy relevance because human relevance labels are unavailable."
}

with open(
    REC_ASSET_DIR/"metadata.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        recommender_metadata19,
        f,
        indent=2
    )

with open(
    EXPORT_DIR/"api_recommend_request.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        api_request19,
        f,
        indent=2
    )

with open(
    EXPORT_DIR/"api_recommend_response.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        api_response19,
        f,
        indent=2
    )

profiles19.to_csv(
    EXPORT_DIR/"task5_evaluation_profiles.csv",
    index=False
)

weight_eval19.to_csv(
    EXPORT_DIR/"task5_weight_evaluation.csv",
    index=False
)

weight_summary19.to_csv(
    EXPORT_DIR/"task5_weight_summary.csv",
    index=False
)

evaluation19.to_csv(
    EXPORT_DIR/"task5_recommender_evaluation.csv",
    index=False
)

edge_cases19.to_csv(
    EXPORT_DIR/"task5_edge_case_tests.csv",
    index=False
)

repro_summary19=pd.DataFrame({
    "Test":[
        "Repeated identical request"
    ],
    "Status":[
        "PASS" if reproducible19 else "FAIL"
    ],
    "Version":[
        RECOMMENDER_VERSION
    ]
})

repro_summary19.to_csv(
    EXPORT_DIR/"task5_reproducibility_test.csv",
    index=False
)


def card19(a,b,c):
    return f"""
    <div class="c19">
    <small>{a}</small>
    <b>{b}</b>
    <span>{c}</span>
    </div>
    """

def table19(d,n=None):
    z=d.head(n) if n else d

    display(HTML(z.to_html(
        index=False,
        border=0,
        classes="t19",
        float_format=lambda x:f"{x:,.3f}"
    )))

def json19(x):
    s=json.dumps(
        x,
        indent=2
    )

    s=(
        s.replace("&","&amp;")
         .replace("<","&lt;")
         .replace(">","&gt;")
    )

    display(HTML(
        f"""
        <pre class="json19">{s}</pre>
        """
    ))


display(HTML("""
<style>
.d19{
border:1px solid #ddd7d4;border-radius:18px;overflow:hidden;
background:#fff;box-shadow:0 8px 26px #22223b18
}
.h19{
background:linear-gradient(120deg,#22223b,#30324d 56%,#4a4e69);
padding:26px 29px;display:flex;justify-content:space-between;
font-family:Arial
}
.h19 .k{color:#c9ada7;font:700 8px Arial;letter-spacing:1.7px}
.h19 h2{color:#fff;margin:6px 0;font:700 25px Arial}
.h19 p,.h19 .m{color:#f2e9e4;font:10px/1.65 Arial}

.cards19{
display:grid;grid-template-columns:repeat(6,1fr);
border-bottom:1px solid #ece7e4
}
.c19{
padding:14px;border-right:1px solid #eee9e7;
font-family:Arial;min-width:0
}
.c19 small{color:#9a8c98;font-size:7px;font-weight:700}
.c19 b{
display:block;color:#22223b;font-size:17px;
margin-top:5px;overflow-wrap:anywhere
}
.c19 span{
display:block;color:#4a4e69;font-size:8px;margin-top:4px
}

.ctrl19{
padding:12px 18px 4px;background:#faf9f8
}
.btn19{margin:8px 15px!important}
.out19{
border-top:1px solid #eee9e7!important;
padding:20px 23px
}

.t19{
width:100%;border-collapse:collapse;font:10px Arial
}
.t19 th{
background:#22223b;color:#fff;padding:9px;text-align:left
}
.t19 td{
padding:8px;border-bottom:1px solid #eee9e7;color:#22223b
}

.grid19{
display:grid;grid-template-columns:repeat(3,1fr);gap:12px
}
.box19{
background:#f8f6f5;border-left:4px solid #4a4e69;
padding:14px;border-radius:9px;
font:11px/1.65 Arial;color:#22223b
}
.box19 b{
display:block;color:#4a4e69;font-size:9px;margin-bottom:6px
}
.note19{
background:#f2e9e4;padding:12px 14px;border-radius:9px;
font:10px/1.65 Arial;color:#4a4e69
}
.txt19{
font:12px/1.8 Arial;color:#22223b
}
.json19{
background:#22223b;color:#f2e9e4;
padding:16px;border-radius:10px;
font:10px/1.6 Consolas,monospace;
overflow:auto;max-height:450px
}

@media(max-width:900px){
.cards19{grid-template-columns:repeat(3,1fr)}
.grid19{grid-template-columns:1fr}
}
</style>
"""))

header19=widgets.HTML(f"""
<div class="h19">

<div>
<div class="k">TASK 05 • RECOMMENDER VALIDATION</div>
<h2>Recommendation Quality & Asset Center</h2>
<p>Ranking Quality • Weight Validation • Reproducibility • Edge Cases • API Evidence</p>
</div>

<div class="m">
MODEL VERSION<br>
<b>{RECOMMENDER_VERSION}</b><br><br>
SELECTED WEIGHTS<br>
<b>{best_weight_name19}</b>
</div>

</div>
""")

profile19=widgets.Dropdown(
    options=profiles19["Profile"].tolist(),
    description="Profile:",
    layout=widgets.Layout(width="310px")
)

view19=widgets.ToggleButtons(
    options=[
        ("OVERVIEW","overview"),
        ("WEIGHT TEST","weights"),
        ("EVALUATION","evaluation"),
        ("PROFILE DETAIL","profile"),
        ("EDGE CASES","edge"),
        ("REPRODUCIBILITY","repro"),
        ("ASSETS","assets"),
        ("API SAMPLE","api"),
        ("METHOD","method")
    ],
    value="overview",
    layout=widgets.Layout(width="100%"),
    style={
        "button_width":"11%",
        "font_weight":"700"
    }
)
view19.add_class("btn19")

controls19=widgets.HBox(
    [profile19],
    layout=widgets.Layout(
        padding="8px 18px"
    )
)

kpi19=widgets.HTML()
output19=widgets.Output()
output19.add_class("out19")


def update19(*_):

    with output19:
        clear_output(wait=True)

        avg_precision=evaluation19[
            "Precision@10"
        ].mean()

        hit_rate=evaluation19[
            "Hit@10"
        ].mean()

        avg_ndcg=evaluation19[
            "NDCG@10"
        ].mean()

        edge_pass=(
            edge_cases19["Status"]
            .eq("PASS")
            .sum()
        )

        kpi19.value='<div class="cards19">'+"".join([
            card19(
                "TEST PROFILES",
                len(evaluation19),
                "Representative queries"
            ),
            card19(
                "PRECISION@10",
                f"{avg_precision:.1%}",
                "Proxy relevance"
            ),
            card19(
                "HIT RATE@10",
                f"{hit_rate:.1%}",
                "Profiles with a hit"
            ),
            card19(
                "NDCG@10",
                f"{avg_ndcg:.3f}",
                "Ranking quality"
            ),
            card19(
                "REPRODUCIBLE",
                "PASS" if reproducible19 else "FAIL",
                "Same request"
            ),
            card19(
                "EDGE TESTS",
                f"{edge_pass}/{len(edge_cases19)}",
                "Passed"
            )
        ])+"</div>"

        v=view19.value

        if v=="overview":

            display(HTML(f"""
            <div class="grid19">

            <div class="box19">
            <b>SELECTED WEIGHT SET</b>
            {best_weight_name19}<br>
            Selected using mean proxy NDCG@10,
            Precision@10 and Hit Rate@10.
            </div>

            <div class="box19">
            <b>QUALITY CONTROL</b>
            {len(evaluation19)} representative profiles tested.<br>
            Mean Precision@10: {avg_precision:.1%}<br>
            Mean NDCG@10: {avg_ndcg:.3f}
            </div>

            <div class="box19">
            <b>DEPLOYMENT READINESS</b>
            Reproducibility:
            {"PASS" if reproducible19 else "FAIL"}<br>
            Edge tests:
            {edge_pass}/{len(edge_cases19)} passed
            </div>

            </div><br>

            <div class="note19">
            Evaluation uses job-family and preference consistency as a
            proxy relevance label because the source dataset does not
            contain human relevance judgments, user clicks or application
            outcomes.
            </div>
            """))

        elif v=="weights":

            z=weight_summary19.sort_values(
                "Mean_NDCG10"
            )

            fig=go.Figure()

            fig.add_trace(go.Bar(
                y=z["Weight_Set"],
                x=z["Mean_NDCG10"],
                orientation="h",
                name="NDCG@10",
                marker_color="#4a4e69"
            ))

            fig.add_trace(go.Bar(
                y=z["Weight_Set"],
                x=z["Mean_Precision10"],
                orientation="h",
                name="Precision@10",
                marker_color="#c9ada7"
            ))

            fig.update_xaxes(
                title="Mean Evaluation Score",
                range=[0,1]
            )

            fig.update_layout(
                barmode="group",
                height=420,
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",
                font=dict(
                    family="Arial",
                    color="#22223b"
                ),
                legend=dict(
                    orientation="h",
                    y=1.08
                ),
                margin=dict(
                    l=150,r=30,t=20,b=50
                )
            )

            display(fig)

            table19(
                weight_summary19.drop(
                    columns="Priority"
                )
            )

        elif v=="evaluation":

            z=evaluation19.sort_values(
                "NDCG@10"
            )

            fig=go.Figure()

            fig.add_trace(go.Bar(
                y=z["Profile"],
                x=z["Precision@10"],
                orientation="h",
                name="Precision@10",
                marker_color="#4a4e69"
            ))

            fig.add_trace(go.Bar(
                y=z["Profile"],
                x=z["NDCG@10"],
                orientation="h",
                name="NDCG@10",
                marker_color="#c9ada7"
            ))

            fig.update_xaxes(
                title="Proxy Ranking Score",
                range=[0,1]
            )

            fig.update_layout(
                barmode="group",
                height=max(
                    450,
                    len(z)*48
                ),
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",
                font=dict(
                    family="Arial",
                    color="#22223b"
                ),
                legend=dict(
                    orientation="h",
                    y=1.08
                ),
                margin=dict(
                    l=170,r=30,t=20,b=50
                )
            )

            display(fig)

        elif v=="profile":

            p=profiles19[
                profiles19["Profile"]
                .eq(profile19.value)
            ].iloc[0]

            r=recommend_jobs(
                p["Query"],
                p["Country"],
                p["Pay_Type"],
                p["Seniority"],
                TOP_K19
            )

            display(HTML(f"""
            <div class="grid19">

            <div class="box19">
            <b>QUERY</b>
            {p['Query']}<br>
            Expected family: {p['Expected_Family']}
            </div>

            <div class="box19">
            <b>PREFERENCES</b>
            Country: {p['Country']}<br>
            Pay: {p['Pay_Type']}<br>
            Seniority: {p['Seniority']}
            </div>

            <div class="box19">
            <b>RESULT</b>
            Returned: {len(r)}<br>
            Relevant:
            {int(relevance19(r,p).sum())}
            </div>

            </div><br>
            """))

            table19(
                r[[
                    "title",
                    "job_family",
                    "country_clean",
                    "pay_type",
                    "seniority_level",
                    "Match_%",
                    "Why_Matched"
                ]]
            )

        elif v=="edge":

            table19(edge_cases19)

        elif v=="repro":

            display(HTML(f"""
            <div class="grid19">

            <div class="box19">
            <b>TEST</b>
            Same profile submitted twice with
            identical data and model version.
            </div>

            <div class="box19">
            <b>RESULT</b>
            {'PASS' if reproducible19 else 'FAIL'}<br>
            Job ranking and scores compared.
            </div>

            <div class="box19">
            <b>VERSION</b>
            {RECOMMENDER_VERSION}<br>
            Catalogue rows: {len(rec):,}
            </div>

            </div>
            """))

        elif v=="assets":

            assets19=pd.DataFrame({
                "Asset":[
                    "Recommendation catalogue",
                    "TF-IDF vectorizer",
                    "TF-IDF sparse matrix",
                    "Selected weights",
                    "Recommender metadata",
                    "Evaluation profiles",
                    "Weight evaluation",
                    "Recommender evaluation",
                    "Edge-case tests",
                    "API request sample",
                    "API response sample"
                ],
                "Output":[
                    "models/recommender_assets/recommendation_catalog.csv",
                    "models/recommender_assets/tfidf_vectorizer.pkl",
                    "models/recommender_assets/tfidf_matrix.npz",
                    "models/recommender_assets/weights.json",
                    "models/recommender_assets/metadata.json",
                    "task5_evaluation_profiles.csv",
                    "task5_weight_evaluation.csv",
                    "task5_recommender_evaluation.csv",
                    "task5_edge_case_tests.csv",
                    "api_recommend_request.json",
                    "api_recommend_response.json"
                ],
                "Status":["Created"]*11
            })

            table19(assets19)

        elif v=="api":

            display(HTML(
                '<div class="txt19"><b>Example Request</b></div>'
            ))

            json19(api_request19)

            display(HTML(
                '<br><div class="txt19"><b>Example Response</b></div>'
            ))

            json19(api_response19)

        else:

            display(HTML(f"""
            <div class="txt19">

            <b>Recommendation approach:</b>
            TF-IDF unigram/bigram title representation with cosine
            similarity. This is lexical content similarity rather than
            embedding-based semantic similarity.<br><br>

            <b>Preferences:</b>
            Country, pay type and seniority are optional compatibility
            signals. Recency is retained as a small ranking factor.<br><br>

            <b>Active-weight normalization:</b>
            Preference weights are included only when the user actually
            selects that preference. Choosing "Any" therefore does not
            artificially increase Match %. Scores are normalized by the
            sum of active weights.<br><br>

            <b>Weight validation:</b>
            Three candidate weight strategies are compared using
            representative test profiles. The selected configuration is
            <b>{best_weight_name19}</b>.<br><br>

            <b>Offline relevance:</b>
            Precision@10, Hit Rate@10, MRR and NDCG@10 use consistency
            with the expected job family plus profile preferences as
            proxy relevance. These metrics are not human-judged relevance
            scores.<br><br>

            <b>Cold start:</b>
            Empty, one-character and unknown queries return controlled
            empty results rather than raising an exception.<br><br>

            <b>Reproducibility:</b>
            Identical requests are checked for identical ranked job IDs
            and scores under a fixed catalogue and model version.<br><br>

            <b>Dataset limitation:</b>
            The source contains job titles but no job descriptions or
            explicit skills. Therefore genuine skill-match and
            missing-skill explanations cannot be claimed from this
            dataset.<br><br>

            <b>Deployment:</b>
            Vectorizer, sparse TF-IDF matrix, catalogue, selected weights,
            metadata and example API payloads are saved for the later
            Flask service.

            </div>
            """))


for w in [
    profile19,
    view19
]:
    w.observe(
        update19,
        names="value"
    )

update19()

dashboard_19=widgets.VBox([
    header19,
    kpi19,
    widgets.HTML(
        '<div class="ctrl19">'
        '<b style="font:11px Arial;color:#22223b">'
        'RECOMMENDER VALIDATION CONTROLS'
        '</b></div>'
    ),
    controls19,
    view19,
    output19
])

dashboard_19.add_class("d19")
display(dashboard_19)
```

```python
required=["scenario","selected_forecast_summary18"]
missing=[x for x in required if x not in globals()]
if missing:
    raise NameError(f"Run Sections 14 and 18 first. Missing: {missing}")

HORIZON20=7

base20=(
    scenario[[
        "job_family","Recent_14d","Previous_14d",
        "Growth_%","Volatility","Base_Band_%",
        "Risk","Opportunity_Score"
    ]]
    .merge(
        selected_forecast_summary18[[
            "job_family","Selected_Model","Forecast_7d"
        ]],
        on="job_family",
        how="inner"
    )
    .copy()
)

base20["Band_%"]=base20["Base_Band_%"].clip(.10,.35)
base20["Base_Forecast"]=base20["Forecast_7d"]

tech_exposure20={
    "Data & AI":1.00,
    "Software Development":.80,
    "Engineering & Architecture":.50,
    "Design & Creative":.20,
    "Marketing & Sales":.10,
    "Writing & Translation":0,
    "Finance & Accounting":0,
    "Admin & Support":0,
    "Legal":0,
    "Other":0
}

base20["Technology_Exposure"]=(
    base20["job_family"]
    .map(tech_exposure20)
    .fillna(0)
)

base20["High_Growth"]=(
    base20["Base_Forecast"]*
    (1+base20["Band_%"])
)

base20["Slow_Growth"]=(
    base20["Base_Forecast"]*
    (1-base20["Band_%"])
).clip(lower=0)

base20["Technology_Raw"]=(
    base20["Base_Forecast"]*
    (
        1+
        base20["Band_%"]*
        base20["Technology_Exposure"]
    )
)

base_total20=base20["Base_Forecast"].sum()
tech_raw_total20=base20["Technology_Raw"].sum()

tech_scale20=(
    base_total20/tech_raw_total20
    if tech_raw_total20>0 else 1
)

base20["Technology_Shift"]=(
    base20["Technology_Raw"]*
    tech_scale20
)

scenario_specs20={
    "Base":"Base_Forecast",
    "High Growth":"High_Growth",
    "Slow Growth":"Slow_Growth",
    "Technology Shift":"Technology_Shift"
}

scenario_frames20=[]

for name,col in scenario_specs20.items():

    z=base20.copy()

    z["Scenario"]=name
    z["Scenario_Forecast"]=z[col]

    z["Difference_From_Base"]=(
        z["Scenario_Forecast"]-
        z["Base_Forecast"]
    )

    z["Difference_%"]=np.where(
        z["Base_Forecast"]>0,
        z["Difference_From_Base"]/
        z["Base_Forecast"]*100,
        np.nan
    )

    scenario_frames20.append(z)

scenario_detail20=pd.concat(
    scenario_frames20,
    ignore_index=True
)

scenario_summary20=(
    scenario_detail20
    .groupby("Scenario",as_index=False)
    .agg(
        Total_Forecast=("Scenario_Forecast","sum"),
        Avg_Forecast=("Scenario_Forecast","mean"),
        Categories=("job_family","nunique")
    )
)

scenario_summary20["Change_vs_Base_%"]=np.where(
    base_total20>0,
    (
        scenario_summary20["Total_Forecast"]-
        base_total20
    )/base_total20*100,
    np.nan
)

scenario_order20=[
    "Base",
    "High Growth",
    "Slow Growth",
    "Technology Shift"
]

scenario_summary20["Scenario"]=pd.Categorical(
    scenario_summary20["Scenario"],
    categories=scenario_order20,
    ordered=True
)

scenario_summary20=(
    scenario_summary20
    .sort_values("Scenario")
    .reset_index(drop=True)
)

high_total20=(
    scenario_summary20.loc[
        scenario_summary20["Scenario"].eq("High Growth"),
        "Total_Forecast"
    ].iloc[0]
)

slow_total20=(
    scenario_summary20.loc[
        scenario_summary20["Scenario"].eq("Slow Growth"),
        "Total_Forecast"
    ].iloc[0]
)

tech_total20=(
    scenario_summary20.loc[
        scenario_summary20["Scenario"].eq("Technology Shift"),
        "Total_Forecast"
    ].iloc[0]
)

assumptions20=pd.DataFrame([
    {
        "Scenario":"Base",
        "Driver":"Validated demand model",
        "Formula":"Selected Task 3 7-day forecast",
        "Market_Total_Effect":"No scenario adjustment",
        "Interpretation":"Reference case"
    },
    {
        "Scenario":"High Growth",
        "Driver":"Role-specific uncertainty band",
        "Formula":"Base × (1 + Band)",
        "Market_Total_Effect":"Expansion",
        "Interpretation":"Conditional stronger-demand case"
    },
    {
        "Scenario":"Slow Growth",
        "Driver":"Role-specific uncertainty band",
        "Formula":"Base × (1 - Band)",
        "Market_Total_Effect":"Contraction",
        "Interpretation":"Conditional weaker-demand case"
    },
    {
        "Scenario":"Technology Shift",
        "Driver":"Documented technology-exposure assumption",
        "Formula":"Base × (1 + Band × Exposure), normalized to Base total",
        "Market_Total_Effect":"Demand mix changes; total held constant",
        "Interpretation":"Reallocation toward technology-exposed families"
    }
])

tech_assumptions20=pd.DataFrame(
    list(tech_exposure20.items()),
    columns=[
        "job_family",
        "Technology_Exposure"
    ]
)

scenario_detail20.to_csv(
    EXPORT_DIR/"task_08_scenarios.csv",
    index=False
)

assumptions20.to_csv(
    EXPORT_DIR/"task_08_scenario_assumptions.csv",
    index=False
)

tech_assumptions20.to_csv(
    EXPORT_DIR/"task_08_technology_exposure_assumptions.csv",
    index=False
)

scenario_summary20.to_csv(
    EXPORT_DIR/"task_08_scenario_summary.csv",
    index=False
)

REPORT_DIR=Path("reports")
REPORT_DIR.mkdir(exist_ok=True)

future_report20=f"""
# Task 08 — Future Job-Market Scenario Outlook

## Forecast basis
The Base scenario uses the validated {HORIZON20}-day selected demand forecast from Task 3.

## Scenarios
- Base: validated model forecast with no scenario adjustment.
- High Growth: Base × (1 + role-specific uncertainty band).
- Slow Growth: Base × (1 - role-specific uncertainty band).
- Technology Shift: demand is reallocated toward technology-exposed job families while total Base market demand is held constant.

## Important interpretation
These scenarios are conditional planning cases, not statistical confidence intervals or guaranteed labour-market outcomes.

## Technology Shift assumption
Technology exposure is an explicit scenario assumption used only to test changes in market mix. It is not an observed causal effect in the source dataset.

## Horizon
{HORIZON20} days, consistent with the validated Task 3 forecast horizon.
"""

with open(
    REPORT_DIR/"task_08_future_outlook.md",
    "w",
    encoding="utf-8"
) as f:
    f.write(future_report20)


def card20(a,b,c):
    return f"""
    <div class="c20">
    <small>{a}</small>
    <b>{b}</b>
    <span>{c}</span>
    </div>
    """


def table20(d,n=None):
    z=d.head(n) if n else d

    display(HTML(z.to_html(
        index=False,
        border=0,
        classes="t20",
        float_format=lambda x:f"{x:,.2f}"
    )))


display(HTML("""
<style>
.d20{
border:1px solid #ddd7d4;border-radius:18px;overflow:hidden;
background:#fff;box-shadow:0 8px 26px #22223b18
}
.h20{
background:linear-gradient(120deg,#22223b,#30324d 56%,#4a4e69);
padding:26px 29px;display:flex;justify-content:space-between;
font-family:Arial
}
.h20 .k{
color:#c9ada7;font:700 8px Arial;letter-spacing:1.7px
}
.h20 h2{
color:#fff;margin:6px 0;font:700 25px Arial
}
.h20 p,.h20 .m{
color:#f2e9e4;font:10px/1.65 Arial
}

.cards20{
display:grid;grid-template-columns:repeat(6,1fr);
border-bottom:1px solid #ece7e4
}
.c20{
padding:14px;border-right:1px solid #eee9e7;
font-family:Arial;min-width:0
}
.c20 small{
color:#9a8c98;font-size:7px;font-weight:700
}
.c20 b{
display:block;color:#22223b;font-size:17px;
margin-top:5px;overflow-wrap:anywhere
}
.c20 span{
display:block;color:#4a4e69;font-size:8px;margin-top:4px
}

.ctrl20{
padding:12px 18px 4px;background:#faf9f8
}
.btn20{
margin:8px 15px!important
}
.out20{
border-top:1px solid #eee9e7!important;
padding:20px 23px
}

.t20{
width:100%;border-collapse:collapse;font:10px Arial
}
.t20 th{
background:#22223b;color:#fff;
padding:9px;text-align:left
}
.t20 td{
padding:8px;border-bottom:1px solid #eee9e7;
color:#22223b
}

.grid20{
display:grid;grid-template-columns:repeat(3,1fr);gap:12px
}
.box20{
background:#f8f6f5;border-left:4px solid #4a4e69;
padding:14px;border-radius:9px;
font:11px/1.65 Arial;color:#22223b
}
.box20 b{
display:block;color:#4a4e69;
font-size:9px;margin-bottom:6px
}
.note20{
background:#f2e9e4;padding:12px 14px;
border-radius:9px;font:10px/1.65 Arial;
color:#4a4e69
}
.txt20{
font:12px/1.8 Arial;color:#22223b
}

@media(max-width:900px){
.cards20{grid-template-columns:repeat(3,1fr)}
.grid20{grid-template-columns:1fr}
}
</style>
"""))

header20=widgets.HTML(f"""
<div class="h20">

<div>
<div class="k">TASK 08 • SCENARIO GOVERNANCE</div>
<h2>Future Market Scenario Intelligence</h2>
<p>Base • High Growth • Slow Growth • Technology Shift • Assumption Traceability</p>
</div>

<div class="m">
FORECAST BASIS<br>
<b>VALIDATED TASK 3</b><br><br>
HORIZON<br>
<b>{HORIZON20} DAYS</b>
</div>

</div>
""")

family20=widgets.Dropdown(
    options=base20["job_family"].tolist(),
    description="Job Family:",
    layout=widgets.Layout(width="330px")
)

top_n20=widgets.IntSlider(
    value=min(8,max(1,len(base20))),
    min=1,
    max=max(1,len(base20)),
    description="Top N:",
    continuous_update=False,
    layout=widgets.Layout(width="260px")
)

view20=widgets.ToggleButtons(
    options=[
        ("OVERVIEW","overview"),
        ("SCENARIOS","scenarios"),
        ("MARKET MIX","mix"),
        ("CATEGORY","category"),
        ("ASSUMPTIONS","assumptions"),
        ("TECH SHIFT","tech"),
        ("DATA","data"),
        ("FINDINGS","findings"),
        ("METHOD","method")
    ],
    value="overview",
    layout=widgets.Layout(width="100%"),
    style={
        "button_width":"11%",
        "font_weight":"700"
    }
)
view20.add_class("btn20")

controls20=widgets.HBox(
    [
        family20,
        top_n20
    ],
    layout=widgets.Layout(
        padding="8px 18px",
        flex_flow="row wrap",
        gap="10px"
    )
)

kpi20=widgets.HTML()
output20=widgets.Output()
output20.add_class("out20")


def update20(*_):

    with output20:
        clear_output(wait=True)

        fam=family20.value

        top_opp=base20.nlargest(
            1,
            "Opportunity_Score"
        ).iloc[0]

        kpi20.value='<div class="cards20">'+"".join([
            card20(
                "BASE MARKET",
                f"{base_total20:,.0f}",
                "7-day postings"
            ),
            card20(
                "HIGH GROWTH",
                f"{high_total20:,.0f}",
                f"{(high_total20/base_total20-1):+.1%}"
                if base_total20 else "N/A"
            ),
            card20(
                "SLOW GROWTH",
                f"{slow_total20:,.0f}",
                f"{(slow_total20/base_total20-1):+.1%}"
                if base_total20 else "N/A"
            ),
            card20(
                "TECH SHIFT",
                f"{tech_total20:,.0f}",
                "Total held constant"
            ),
            card20(
                "TOP OPPORTUNITY",
                top_opp["job_family"],
                f"{top_opp['Opportunity_Score']:.1f}/100"
            ),
            card20(
                "SCENARIOS",
                4,
                "Required cases"
            )
        ])+"</div>"

        v=view20.value

        if v=="overview":

            display(HTML("""
            <div class="grid20">

            <div class="box20">
            <b>MODEL FORECAST</b>
            Base demand comes directly from the validated Task 3
            selected-model forecast rather than from a newly invented
            scenario number.
            </div>

            <div class="box20">
            <b>DEMAND RANGE</b>
            High Growth and Slow Growth use the existing
            role-specific uncertainty band from Section 14.
            </div>

            <div class="box20">
            <b>STRUCTURAL SHIFT</b>
            Technology Shift changes category mix while preserving
            the total Base market forecast.
            </div>

            </div><br>

            <div class="note20">
            Scenario values are conditional planning assumptions.
            Only the Base case is the direct Task 3 model forecast.
            High Growth, Slow Growth and Technology Shift are
            assumption-driven transformations of that forecast.
            </div>
            """))

        elif v=="scenarios":

            z=scenario_summary20.copy()

            fig=go.Figure(go.Bar(
                x=z["Scenario"].astype(str),
                y=z["Total_Forecast"],
                marker=dict(
                    color=z["Total_Forecast"],
                    colorscale=[
                        [0,"#c9ada7"],
                        [.5,"#9a8c98"],
                        [1,"#4a4e69"]
                    ]
                ),
                text=z["Total_Forecast"].map(
                    lambda x:f"{x:,.0f}"
                ),
                textposition="outside",
                customdata=z["Change_vs_Base_%"],
                hovertemplate=(
                    "<b>%{x}</b><br>"
                    "Forecast: %{y:,.0f}<br>"
                    "Change vs Base: %{customdata:+.1f}%"
                    "<extra></extra>"
                )
            ))

            fig.update_yaxes(
                title="7-Day Forecast Posting Volume"
            )

            fig.update_layout(
                height=470,
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",
                font=dict(
                    family="Arial",
                    color="#22223b"
                ),
                margin=dict(
                    l=70,r=30,t=25,b=50
                )
            )

            display(fig)

        elif v=="mix":

            ranking=(
                base20.nlargest(
                    top_n20.value,
                    "Base_Forecast"
                )["job_family"]
                .tolist()
            )

            z=scenario_detail20[
                scenario_detail20[
                    "job_family"
                ].isin(ranking)
            ].copy()

            fig=go.Figure()

            for scenario_name in scenario_order20:

                q=z[
                    z["Scenario"].eq(
                        scenario_name
                    )
                ]

                fig.add_trace(go.Bar(
                    x=q["job_family"],
                    y=q["Scenario_Forecast"],
                    name=scenario_name
                ))

            fig.update_yaxes(
                title="Scenario Forecast"
            )

            fig.update_layout(
                barmode="group",
                height=520,
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",
                font=dict(
                    family="Arial",
                    color="#22223b"
                ),
                legend=dict(
                    orientation="h",
                    y=1.08
                ),
                margin=dict(
                    l=70,r=30,t=30,b=110
                )
            )

            display(fig)

        elif v=="category":

            z=scenario_detail20[
                scenario_detail20[
                    "job_family"
                ].eq(fam)
            ].copy()

            z["Scenario"]=pd.Categorical(
                z["Scenario"],
                categories=scenario_order20,
                ordered=True
            )

            z=z.sort_values("Scenario")

            fig=go.Figure(go.Bar(
                x=z["Scenario"].astype(str),
                y=z["Scenario_Forecast"],
                marker_color="#4a4e69",
                text=z["Scenario_Forecast"].map(
                    lambda x:f"{x:,.0f}"
                ),
                textposition="outside",
                customdata=z[
                    [
                        "Difference_%",
                        "Band_%",
                        "Risk"
                    ]
                ],
                hovertemplate=(
                    "<b>%{x}</b><br>"
                    "Forecast: %{y:,.1f}<br>"
                    "Change vs Base: %{customdata[0]:+.1f}%<br>"
                    "Band: %{customdata[1]:.1%}<br>"
                    "Risk: %{customdata[2]}"
                    "<extra></extra>"
                )
            ))

            fig.update_yaxes(
                title="7-Day Forecast"
            )

            fig.update_layout(
                height=450,
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",
                font=dict(
                    family="Arial",
                    color="#22223b"
                ),
                margin=dict(
                    l=70,r=30,t=25,b=50
                )
            )

            display(fig)

            r=base20[
                base20["job_family"].eq(fam)
            ].iloc[0]

            display(HTML(f"""
            <br>
            <div class="grid20">

            <div class="box20">
            <b>VALIDATED BASE</b>
            {r['Base_Forecast']:,.1f}<br>
            Model: {r['Selected_Model']}
            </div>

            <div class="box20">
            <b>RECENT SIGNAL</b>
            Growth: {r['Growth_%']:+.1f}%<br>
            Risk: {r['Risk']}<br>
            Band: {r['Band_%']:.1%}
            </div>

            <div class="box20">
            <b>TECH EXPOSURE</b>
            {r['Technology_Exposure']:.2f}<br>
            Scenario assumption only
            </div>

            </div>
            """))

        elif v=="assumptions":

            table20(assumptions20)

        elif v=="tech":

            z=tech_assumptions20.sort_values(
                "Technology_Exposure"
            )

            fig=go.Figure(go.Bar(
                x=z["Technology_Exposure"],
                y=z["job_family"],
                orientation="h",
                marker_color="#4a4e69",
                text=z["Technology_Exposure"].map(
                    lambda x:f"{x:.2f}"
                ),
                textposition="outside"
            ))

            fig.update_xaxes(
                title="Technology Exposure Assumption",
                range=[0,1.1]
            )

            fig.update_layout(
                height=max(
                    450,
                    len(z)*38
                ),
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",
                font=dict(
                    family="Arial",
                    color="#22223b"
                ),
                margin=dict(
                    l=190,r=40,t=20,b=45
                )
            )

            display(fig)

            display(HTML("""
            <div class="note20">
            These exposure values are scenario assumptions used to
            test a possible technology-led change in job-family mix.
            They are not estimated causal effects from the dataset.
            The Technology Shift scenario is normalized so total
            demand remains equal to the Base forecast.
            </div>
            """))

        elif v=="data":

            table20(
                scenario_detail20[[
                    "job_family",
                    "Scenario",
                    "Selected_Model",
                    "Base_Forecast",
                    "Scenario_Forecast",
                    "Difference_%",
                    "Band_%",
                    "Risk",
                    "Opportunity_Score"
                ]].sort_values(
                    [
                        "job_family",
                        "Scenario"
                    ]
                )
            )

        elif v=="findings":

            high_gain=(
                scenario_detail20[
                    scenario_detail20[
                        "Scenario"
                    ].eq("High Growth")
                ]
                .nlargest(
                    3,
                    "Difference_From_Base"
                )
            )

            tech_gain=(
                scenario_detail20[
                    scenario_detail20[
                        "Scenario"
                    ].eq("Technology Shift")
                ]
                .nlargest(
                    3,
                    "Difference_From_Base"
                )
            )

            risk=base20.nlargest(
                3,
                "Band_%"
            )

            display(HTML(f"""
            <div class="grid20">

            <div class="box20">
            <b>LARGEST HIGH-GROWTH UPSIDE</b>
            {"<br>".join(
                f"{r['job_family']}: +{r['Difference_From_Base']:,.0f}"
                for _,r in high_gain.iterrows()
            )}
            </div>

            <div class="box20">
            <b>TECH-SHIFT BENEFICIARIES</b>
            {"<br>".join(
                f"{r['job_family']}: {r['Difference_%']:+.1f}%"
                for _,r in tech_gain.iterrows()
            )}
            </div>

            <div class="box20">
            <b>HIGHEST UNCERTAINTY</b>
            {"<br>".join(
                f"{r['job_family']}: ±{r['Band_%']:.1%}"
                for _,r in risk.iterrows()
            )}
            </div>

            </div>
            """))

        else:

            display(HTML("""
            <div class="txt20">

            <b>Base:</b>
            Uses the selected seven-day forecast produced by the
            validated Task 3 model comparison. No scenario multiplier
            is applied.<br><br>

            <b>High Growth:</b>
            Applies the existing role-specific uncertainty band as an
            upside demand assumption:
            Base × (1 + Band).<br><br>

            <b>Slow Growth:</b>
            Applies the same band in the opposite direction:
            Base × (1 - Band), with forecasts floored at zero.<br><br>

            <b>Technology Shift:</b>
            Changes job-family composition rather than total market
            size. Technology-exposed categories receive higher raw
            weights and the result is normalized back to the Base
            market total.<br><br>

            <b>Technology exposure:</b>
            This is an explicit scenario assumption, not an observed
            causal estimate. Its purpose is sensitivity analysis.<br><br>

            <b>Uncertainty:</b>
            The role-level band comes from the existing Task 8
            volatility and momentum framework.<br><br>

            <b>Separation of forecast and assumption:</b>
            Base_Forecast stores the model result. Scenario_Forecast
            stores the conditional transformation. They are never
            presented as the same thing.<br><br>

            <b>Scope:</b>
            The horizon remains seven days because this is the
            validated forecasting horizon supported by the reliable
            observation window.

            </div>
            """))


for w in [
    family20,
    top_n20,
    view20
]:
    w.observe(
        update20,
        names="value"
    )

update20()

dashboard_20=widgets.VBox([
    header20,
    kpi20,
    widgets.HTML(
        '<div class="ctrl20">'
        '<b style="font:11px Arial;color:#22223b">'
        'SCENARIO INTELLIGENCE CONTROLS'
        '</b></div>'
    ),
    controls20,
    view20,
    output20
])

dashboard_20.add_class("d20")
display(dashboard_20)

if "logger" in globals():
    logger.info(
        "SECTION20_COMPLETE | scenarios=%s | families=%s | base_total=%.1f | tech_total=%.1f",
        scenario_detail20["Scenario"].nunique(),
        scenario_detail20["job_family"].nunique(),
        base_total20,
        tech_total20
    )
```

```python
required=[
    "df_final",
    "selected_forecast_daily18",
    "model_registry_master21"
]

missing=[
    x for x in required
    if x not in globals()
]

if missing:
    raise NameError(
        f"Run the previous sections first. Missing: {missing}"
    )

SQL_SERVER=r"localhost\SQLEXPRESS"
SQL_DATABASE="JobMarketDB"
installed_drivers22=pyodbc.drivers()

preferred_drivers22=[
    "ODBC Driver 18 for SQL Server",
    "ODBC Driver 17 for SQL Server"
]

SQL_DRIVER=next(
    (
        driver
        for driver in preferred_drivers22
        if driver in installed_drivers22
    ),
    None
)

if not SQL_DRIVER:
    raise RuntimeError(
        "Microsoft SQL Server ODBC Driver 18 or 17 was not found. "
        f"Installed drivers: {installed_drivers22}"
    )

if not re.fullmatch(
    r"[A-Za-z0-9_]+",
    SQL_DATABASE
):
    raise ValueError(
        "SQL_DATABASE may contain only letters, numbers and underscores."
    )

DATABASE_DIR=Path("database")
SRC_DATA_DIR=Path("src/data")
REPORT_DIR=Path("reports")

DATABASE_DIR.mkdir(
    parents=True,
    exist_ok=True
)

SRC_DATA_DIR.mkdir(
    parents=True,
    exist_ok=True
)

REPORT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

BUILD_ID22=pd.Timestamp.now(
    tz="UTC"
).strftime("%Y%m%d_%H%M%S")

def connection_string22(database):

    return (
        f"DRIVER={{{SQL_DRIVER}}};"
        f"SERVER={SQL_SERVER};"
        f"DATABASE={database};"
        "Trusted_Connection=yes;"
        "Encrypt=yes;"
        "TrustServerCertificate=yes;"
    )

test_conn22=pyodbc.connect(
    connection_string22("master"),
    timeout=10
)

server_name22=test_conn22.getinfo(
    pyodbc.SQL_SERVER_NAME
)

test_conn22.close()

print("✅ SQL Server connection successful")
print("Server   :",server_name22)
print("Database :",SQL_DATABASE)
print("Driver   :",SQL_DRIVER)
print("Auth     : Windows Authentication")
print("Build ID :",BUILD_ID22)
```

<div style="color:#c9ada7;font-size:8px;font-weight:700;letter-spacing:1.6px">SECTION 22.1 • DATABASE INITIALIZATION</div>
<div style="color:#fff;font-size:24px;font-weight:700;margin-top:6px">SQL Server Schema, Tables & Indexes</div>
<div style="color:#f2e9e4;font-size:11px;line-height:1.6;margin-top:7px">Creating the persistent JobMarketDB database and production-ready relational schema for jobs, forecasts, future scenarios, recommendations, model governance and build metadata.</div>

```python
status22 = "NOT_READY"

required221 = [
    "connection_string22",
    "SQL_DATABASE",
    "SQL_SERVER",
    "SQL_DRIVER"
]

missing221 = [
    name for name in required221
    if name not in globals()
]

if missing221:
    raise NameError(
        f"Run the Section 22 setup first. Missing: {missing221}"
    )

AUTO_CREATE_DATABASE22 = globals().get(
    "AUTO_CREATE_DATABASE22", False
)


def is_sql_memory_error221(exc):
    detail = str(exc).upper()

    return any(
        token in detail
        for token in (
            "INSUFFICIENT SYSTEM MEMORY",
            "RESOURCE POOL",
            "(701)",
            "ERROR 701",
            "(802)",
            "ERROR 802"
        )
    )


def describe_sql_error221(exc, action):
    detail = str(exc)

    if is_sql_memory_error221(exc):
        return RuntimeError(
            f"Section 22.1 stopped during {action}. "
            "SQL Server reports insufficient server memory. "
            "No further schema changes will be attempted. "
            "Check SQL Server memory availability, service health "
            "and competing workloads before retrying. "
            f"Original error: {detail}"
        )

    return RuntimeError(
        f"Section 22.1 failed during {action}. "
        "Do not continue to Section 22.2 until initialization passes. "
        f"ODBC details: {detail}"
    )


previous_conn221 = globals().get("sql_conn22")

if previous_conn221 is not None:
    try:
        previous_conn221.close()
    except Exception:
        pass

sql_conn22 = None

try:
    sql_conn22 = pyodbc.connect(
        connection_string22(SQL_DATABASE),
        timeout=20,
        autocommit=False
    )

except pyodbc.Error as first_error221:

    if is_sql_memory_error221(first_error221):
        raise describe_sql_error221(
            first_error221,
            "opening target database"
        ) from first_error221

    if not AUTO_CREATE_DATABASE22:
        raise ConnectionError(
            f"Cannot open database '{SQL_DATABASE}' "
            f"on server '{SQL_SERVER}'. "
            "Check that the database exists, is ONLINE, "
            "and is accessible through Windows Authentication. "
            "Database creation is disabled by default. "
            f"Original error: {first_error221}"
        ) from first_error221

    master221 = None

    try:
        master221 = pyodbc.connect(
            connection_string22("master"),
            autocommit=True,
            timeout=15
        )

        cm221 = master221.cursor()

        database_id221 = cm221.execute(
            "SELECT DB_ID(?)",
            SQL_DATABASE
        ).fetchone()[0]

        if database_id221 is None:

            if not SQL_DATABASE.replace("_", "").isalnum():
                raise ValueError(
                    "Unsafe database name."
                )

            cm221.execute(
                f"CREATE DATABASE [{SQL_DATABASE}]"
            )

        else:
            raise PermissionError(
                f"Database '{SQL_DATABASE}' already exists, "
                "but your login cannot open it."
            )

        cm221.close()

    except pyodbc.Error as create_error221:
        raise describe_sql_error221(
            create_error221,
            "optional database creation"
        ) from create_error221

    finally:
        if master221 is not None:
            try:
                master221.close()
            except Exception:
                pass

    try:
        sql_conn22 = pyodbc.connect(
            connection_string22(SQL_DATABASE),
            timeout=20,
            autocommit=False
        )

    except pyodbc.Error as retry_error221:
        raise describe_sql_error221(
            retry_error221,
            "opening newly created database"
        ) from retry_error221

try:
    sql_conn22.timeout = 120
except Exception:
    pass


schema22 = [
"""
IF OBJECT_ID('dbo.build_metadata','U') IS NULL
CREATE TABLE dbo.build_metadata(
    build_id NVARCHAR(40) PRIMARY KEY,
    source_rows BIGINT NOT NULL,
    data_as_of DATETIME2,
    forecast_version NVARCHAR(100),
    recommender_version NVARCHAR(100),
    loaded_at DATETIME2 DEFAULT SYSUTCDATETIME()
)
""",
"""
IF OBJECT_ID('dbo.jobs','U') IS NULL
CREATE TABLE dbo.jobs(
    job_id BIGINT IDENTITY(1,1) PRIMARY KEY,
    title NVARCHAR(500) NOT NULL,
    title_normalized NVARCHAR(500),
    link NVARCHAR(2048) NOT NULL,
    link_hash CHAR(64) UNIQUE NOT NULL,
    published_date DATETIME2,
    published_day DATE,
    country_clean NVARCHAR(150),
    country_code NVARCHAR(10),
    continent NVARCHAR(100),
    pay_type NVARCHAR(50),
    job_family NVARCHAR(150),
    seniority_level NVARCHAR(100),
    is_hourly BIT,
    hourly_low FLOAT,
    hourly_high FLOAT,
    hourly_midpoint FLOAT,
    budget FLOAT,
    fixed_budget FLOAT,
    hourly_outlier_flag BIT,
    fixed_outlier_flag BIT,
    build_id NVARCHAR(40) NOT NULL,
    loaded_at DATETIME2 DEFAULT SYSUTCDATETIME()
)
""",
"""
IF OBJECT_ID('dbo.forecast_predictions','U') IS NULL
CREATE TABLE dbo.forecast_predictions(
    prediction_id BIGINT IDENTITY(1,1) PRIMARY KEY,
    job_family NVARCHAR(150) NOT NULL,
    forecast_day DATE NOT NULL,
    selected_model NVARCHAR(100) NOT NULL,
    forecast_value FLOAT NOT NULL,
    model_version NVARCHAR(100) NOT NULL,
    build_id NVARCHAR(40) NOT NULL,
    created_at DATETIME2 DEFAULT SYSUTCDATETIME()
)
""",
"""
IF OBJECT_ID('dbo.scenario_forecasts','U') IS NULL
CREATE TABLE dbo.scenario_forecasts(
    scenario_id BIGINT IDENTITY(1,1) PRIMARY KEY,
    job_family NVARCHAR(150) NOT NULL,
    scenario_name NVARCHAR(100) NOT NULL,
    base_forecast FLOAT,
    scenario_forecast FLOAT NOT NULL,
    difference_pct FLOAT,
    risk NVARCHAR(50),
    opportunity_score FLOAT,
    build_id NVARCHAR(40) NOT NULL,
    created_at DATETIME2 DEFAULT SYSUTCDATETIME()
)
""",
"""
IF OBJECT_ID('dbo.model_registry','U') IS NULL
CREATE TABLE dbo.model_registry(
    registry_id BIGINT IDENTITY(1,1) PRIMARY KEY,
    component NVARCHAR(100) NOT NULL,
    version NVARCHAR(100) NOT NULL,
    scope NVARCHAR(200),
    method NVARCHAR(200),
    primary_metric NVARCHAR(100),
    primary_value FLOAT,
    secondary_metric NVARCHAR(100),
    secondary_value FLOAT,
    artifact NVARCHAR(500),
    reproducibility NVARCHAR(50),
    status NVARCHAR(50),
    build_id NVARCHAR(40) NOT NULL,
    loaded_at DATETIME2 DEFAULT SYSUTCDATETIME()
)
""",
"""
IF OBJECT_ID('dbo.recommendation_results','U') IS NULL
CREATE TABLE dbo.recommendation_results(
    recommendation_id BIGINT IDENTITY(1,1) PRIMARY KEY,
    run_id UNIQUEIDENTIFIER NOT NULL,
    query_text NVARCHAR(500) NOT NULL,
    job_id BIGINT NOT NULL,
    rank_no INT NOT NULL,
    match_score FLOAT NOT NULL,
    model_version NVARCHAR(100) NOT NULL,
    build_id NVARCHAR(40) NOT NULL,
    created_at DATETIME2 DEFAULT SYSUTCDATETIME(),
    FOREIGN KEY(job_id) REFERENCES dbo.jobs(job_id)
)
"""
]


indexes22 = {
    "IX_jobs_day":
        "CREATE INDEX IX_jobs_day ON dbo.jobs(published_day)",

    "IX_jobs_country":
        "CREATE INDEX IX_jobs_country ON dbo.jobs(country_clean)",

    "IX_jobs_family":
        "CREATE INDEX IX_jobs_family ON dbo.jobs(job_family)",

    "IX_jobs_family_day":
        "CREATE INDEX IX_jobs_family_day ON dbo.jobs(job_family,published_day)",

    "IX_jobs_pay":
        "CREATE INDEX IX_jobs_pay ON dbo.jobs(pay_type)",

    "IX_forecast_family_day":
        "CREATE INDEX IX_forecast_family_day ON dbo.forecast_predictions(job_family,forecast_day)",

    "IX_scenario_family":
        "CREATE INDEX IX_scenario_family ON dbo.scenario_forecasts(scenario_name,job_family)",

    "IX_recommendation_run":
        "CREATE INDEX IX_recommendation_run ON dbo.recommendation_results(run_id,rank_no)"
}


required_tables221 = {
    "build_metadata",
    "jobs",
    "forecast_predictions",
    "scenario_forecasts",
    "model_registry",
    "recommendation_results"
}


expected_columns221 = {
    "build_metadata": {
        "build_id",
        "source_rows",
        "data_as_of",
        "forecast_version",
        "recommender_version",
        "loaded_at"
    },

    "jobs": {
        "job_id",
        "title",
        "title_normalized",
        "link",
        "link_hash",
        "published_date",
        "published_day",
        "country_clean",
        "country_code",
        "continent",
        "pay_type",
        "job_family",
        "seniority_level",
        "is_hourly",
        "hourly_low",
        "hourly_high",
        "hourly_midpoint",
        "budget",
        "fixed_budget",
        "hourly_outlier_flag",
        "fixed_outlier_flag",
        "build_id",
        "loaded_at"
    },

    "forecast_predictions": {
        "prediction_id",
        "job_family",
        "forecast_day",
        "selected_model",
        "forecast_value",
        "model_version",
        "build_id",
        "created_at"
    },

    "scenario_forecasts": {
        "scenario_id",
        "job_family",
        "scenario_name",
        "base_forecast",
        "scenario_forecast",
        "difference_pct",
        "risk",
        "opportunity_score",
        "build_id",
        "created_at"
    },

    "model_registry": {
        "registry_id",
        "component",
        "version",
        "scope",
        "method",
        "primary_metric",
        "primary_value",
        "secondary_metric",
        "secondary_value",
        "artifact",
        "reproducibility",
        "status",
        "build_id",
        "loaded_at"
    },

    "recommendation_results": {
        "recommendation_id",
        "run_id",
        "query_text",
        "job_id",
        "rank_no",
        "match_score",
        "model_version",
        "build_id",
        "created_at"
    }
}


index_tables221 = {
    "IX_jobs_day": "jobs",
    "IX_jobs_country": "jobs",
    "IX_jobs_family": "jobs",
    "IX_jobs_family_day": "jobs",
    "IX_jobs_pay": "jobs",
    "IX_forecast_family_day": "forecast_predictions",
    "IX_scenario_family": "scenario_forecasts",
    "IX_recommendation_run": "recommendation_results"
}


cur = None

try:
    cur = sql_conn22.cursor()

    cur.execute("SELECT 1").fetchone()

    for ddl221 in schema22:
        cur.execute(ddl221)

    for name221, sql221 in indexes22.items():

        table_name221 = index_tables221[name221]

        exists221 = cur.execute(
            """
            SELECT COUNT(*)
            FROM sys.indexes
            WHERE object_id=OBJECT_ID(?)
              AND name=?
            """,
            f"dbo.{table_name221}",
            name221
        ).fetchone()[0]

        if not exists221:
            cur.execute(sql221)

    for table_name221, expected221 in expected_columns221.items():

        actual221 = {
            row[0]
            for row in cur.execute(
                """
                SELECT name
                FROM sys.columns
                WHERE object_id=OBJECT_ID(?)
                """,
                f"dbo.{table_name221}"
            ).fetchall()
        }

        absent221 = sorted(
            expected221 - actual221
        )

        if absent221:
            raise RuntimeError(
                f"Existing dbo.{table_name221} schema "
                "is incompatible with later cells. "
                f"Missing columns: {absent221}. "
                "Existing tables have not been modified."
            )

    missing_indexes221 = []

    for name221, table_name221 in index_tables221.items():

        exists221 = cur.execute(
            """
            SELECT COUNT(*)
            FROM sys.indexes
            WHERE object_id=OBJECT_ID(?)
              AND name=?
            """,
            f"dbo.{table_name221}",
            name221
        ).fetchone()[0]

        if not exists221:
            missing_indexes221.append(name221)

    if missing_indexes221:
        raise RuntimeError(
            f"Required SQL indexes missing: {missing_indexes221}"
        )

    sql_conn22.commit()

except Exception as exc221:

    status22 = "FAILED"

    try:
        sql_conn22.rollback()
    except Exception:
        pass

    try:
        sql_conn22.close()
    except Exception:
        pass

    sql_conn22 = None

    if isinstance(exc221, pyodbc.Error):
        raise describe_sql_error221(
            exc221,
            "schema creation or validation"
        ) from exc221

    raise


try:
    cur = sql_conn22.cursor()

    tables22 = [
        row[0]
        for row in cur.execute(
            """
            SELECT TABLE_NAME
            FROM INFORMATION_SCHEMA.TABLES
            WHERE TABLE_SCHEMA='dbo'
            ORDER BY TABLE_NAME
            """
        ).fetchall()
        if row[0] in required_tables221
    ]

    index_rows22 = [
        tuple(row)
        for row in cur.execute(
            """
            SELECT
                OBJECT_NAME(object_id),
                name
            FROM sys.indexes
            WHERE name IS NOT NULL
              AND OBJECT_NAME(object_id) IN (
                  'jobs',
                  'forecast_predictions',
                  'scenario_forecasts',
                  'recommendation_results',
                  'model_registry',
                  'build_metadata'
              )
            ORDER BY OBJECT_NAME(object_id),name
            """
        ).fetchall()
    ]

    indexes_df22 = pd.DataFrame(
        index_rows22,
        columns=["Table", "Index"]
    )

    status22 = (
        "PASS"
        if set(tables22) == required_tables221
        else "REVIEW"
    )

    server_name22 = sql_conn22.getinfo(
        pyodbc.SQL_SERVER_NAME
    )

    if status22 != "PASS":
        raise RuntimeError(
            "Section 22.1 cannot proceed: "
            f"required tables missing. Found: {tables22}"
        )

except Exception as exc221:

    status22 = "FAILED"

    if isinstance(exc221, pyodbc.Error):
        raise describe_sql_error221(
            exc221,
            "post-commit schema verification"
        ) from exc221

    raise


def card221(a, b, c):
    return f"""
    <div class="c221">
        <small>{a}</small>
        <b>{b}</b>
        <span>{c}</span>
    </div>
    """


def table221(df):
    display(
        HTML(
            df.to_html(
                index=False,
                border=0,
                classes="t221"
            )
        )
    )


display(HTML("""
<style>
.d221{
    border:1px solid #ded8d5;
    border-radius:17px;
    overflow:hidden;
    background:#fff;
    box-shadow:0 8px 24px #22223b17;
}
.h221{
    background:linear-gradient(
        120deg,
        #22223b,
        #30324d 55%,
        #4a4e69
    );
    padding:24px 27px;
    display:flex;
    justify-content:space-between;
    font-family:Arial;
}
.h221 .k{
    color:#c9ada7;
    font-size:8px;
    font-weight:700;
    letter-spacing:1.6px;
}
.h221 h2{
    color:#fff;
    margin:6px 0;
    font-size:24px;
}
.h221 p,.h221 .m{
    color:#f2e9e4;
    font-size:10px;
    line-height:1.65;
}
.cards221{
    display:grid;
    grid-template-columns:repeat(4,1fr);
    border-bottom:1px solid #e8e2df;
}
.c221{
    padding:14px;
    border-right:1px solid #ebe6e4;
    font-family:Arial;
}
.c221 small{
    display:block;
    color:#9a8c98;
    font-size:7px;
    font-weight:700;
}
.c221 b{
    display:block;
    color:#22223b;
    font-size:17px;
    margin-top:5px;
}
.c221 span{
    display:block;
    color:#4a4e69;
    font-size:8px;
    margin-top:4px;
}
.btn221{
    margin:8px 16px!important;
}
.out221{
    border-top:1px solid #eee9e7!important;
    padding:18px 22px;
}
.t221{
    width:100%;
    border-collapse:collapse;
    font:10px Arial;
}
.t221 th{
    background:#22223b;
    color:#fff;
    padding:9px;
    text-align:left;
}
.t221 td{
    padding:8px;
    border-bottom:1px solid #eee9e7;
    color:#22223b;
}
.grid221{
    display:grid;
    grid-template-columns:repeat(3,1fr);
    gap:12px;
}
.box221{
    background:#f8f6f5;
    border-left:4px solid #4a4e69;
    padding:14px;
    border-radius:8px;
    font:11px/1.65 Arial;
    color:#22223b;
}
.box221 b{
    display:block;
    color:#4a4e69;
    font-size:9px;
    margin-bottom:6px;
}
.note221{
    background:#f2e9e4;
    padding:12px 14px;
    border-radius:8px;
    font:10px/1.6 Arial;
    color:#4a4e69;
}
</style>
"""))


header221 = widgets.HTML(f"""
<div class="h221">
    <div>
        <div class="k">
            SECTION 22.1 • DATABASE INITIALIZATION
        </div>
        <h2>SQL Server Schema Readiness</h2>
        <p>
            Database • Relational Tables •
            Indexes • Integrity Controls
        </p>
    </div>
    <div class="m">
        SERVER<br>
        <b>{server_name22}</b><br><br>
        DATABASE<br>
        <b>{SQL_DATABASE}</b>
    </div>
</div>
""")


kpi221 = widgets.HTML(
    '<div class="cards221">' +
    "".join([
        card221(
            "SCHEMA STATUS",
            status22,
            "Database readiness"
        ),
        card221(
            "TABLES",
            f"{len(tables22)}/6",
            "Required tables"
        ),
        card221(
            "INDEXES",
            len(indexes_df22),
            "Keys + indexes"
        ),
        card221(
            "AUTH",
            "WINDOWS",
            "Trusted connection"
        )
    ]) +
    "</div>"
)


view221 = widgets.ToggleButtons(
    options=[
        ("OVERVIEW", "overview"),
        ("TABLES", "tables"),
        ("INDEXES", "indexes"),
        ("METHOD", "method")
    ],
    value="overview",
    layout=widgets.Layout(width="100%"),
    style={
        "button_width": "25%",
        "font_weight": "700"
    }
)

view221.add_class("btn221")

output221 = widgets.Output()
output221.add_class("out221")


def update221(*_):

    with output221:

        clear_output(wait=True)

        if view221.value == "overview":

            display(HTML(f"""
            <div class="grid221">

                <div class="box221">
                    <b>DATABASE ENGINE</b>
                    Microsoft SQL Server<br>
                    Instance: {server_name22}
                </div>

                <div class="box221">
                    <b>RELATIONAL DESIGN</b>
                    Jobs, forecasts, scenarios,
                    recommendations, registry and build metadata.
                </div>

                <div class="box221">
                    <b>READINESS</b>
                    {len(tables22)} required tables available.<br>
                    Schema status: {status22}
                </div>

            </div>
            <br>

            <div class="note221">
                The database structure is ready for the repeatable
                data-load stage. Frequently queried fields are indexed
                before Flask begins consuming the SQL layer.
            </div>
            """))

        elif view221.value == "tables":

            table221(
                pd.DataFrame({
                    "Table": tables22,
                    "Status": [
                        "Ready"
                    ] * len(tables22)}))

        elif view221.value == "indexes":
            table221(indexes_df22)
        else:
            display(HTML("""
            <div class="grid221">
                <div class="box221">
                    <b>PRIMARY DESIGN</b>
                    Identity primary keys are used for transactional
                    tables, while build metadata uses the
                    build ID as its key.
                </div>
                <div class="box221">
                    <b>INTEGRITY</b>
                    Job links are protected by a unique
                    SHA-256 hash and recommendation records
                    reference jobs through a foreign key.
                </div>
                <div class="box221">
                    <b>PERFORMANCE</b>
                    Date, country, family, pay type,
                    forecast and recommendation lookup
                    paths are indexed.
                </div>
            </div>
            """))

view221.observe(
    update221,
    names="value"
)

update221()

dashboard_221 = widgets.VBox([
    header221,
    kpi221,
    view221,
    output221
])

dashboard_221.add_class("d221")
display(dashboard_221)
```

```python
required222=["connection_string22","SQL_DATABASE","BUILD_ID22"]
missing222=[name for name in required222 if name not in globals()]
if "df_features" not in globals() and "df_final" not in globals():
    missing222.append("df_features / df_final")
if missing222:
    raise NameError(f"Run Section 22 setup first. Missing: {missing222}")

source22=df_features.copy() if "df_features" in globals() else df_final.copy()
cols22=[
    "title","title_normalized","link","published_date","published_day",
    "country_clean","country_code","continent","pay_type","job_family",
    "seniority_level","is_hourly","hourly_low","hourly_high",
    "hourly_midpoint","budget","fixed_budget","hourly_outlier_flag",
    "fixed_outlier_flag"
]
for column in cols22:
    if column not in source22.columns:
        source22[column]=np.nan
jobs22=source22[cols22].copy()

for column,limit in {"title":500,"title_normalized":500,"link":2048}.items():
    jobs22[column]=jobs22[column].fillna("").astype(str).str.strip().str[:limit]
jobs22=jobs22[jobs22["title"].ne("") & jobs22["link"].ne("")].copy()
jobs22["link_hash"]=jobs22["link"].map(lambda value:hashlib.sha256(value.encode("utf-8")).hexdigest())
jobs22=jobs22.drop_duplicates("link_hash").copy()

for column,limit in {
    "country_clean":150,"country_code":10,"continent":100,
    "pay_type":50,"job_family":150,"seniority_level":100
}.items():
    jobs22[column]=jobs22[column].map(lambda value:None if pd.isna(value) else str(value)[:limit])
for column in ["hourly_low","hourly_high","hourly_midpoint","budget","fixed_budget"]:
    jobs22[column]=pd.to_numeric(jobs22[column],errors="coerce")

def bit22(value):
    if pd.isna(value):
        return None
    if isinstance(value,str):
        return int(value.strip().lower() in {"1","true","yes"})
    return int(bool(value))

for column in ["is_hourly","hourly_outlier_flag","fixed_outlier_flag"]:
    jobs22[column]=jobs22[column].map(bit22)
for column in ["published_date","published_day"]:
    jobs22[column]=pd.to_datetime(jobs22[column],errors="coerce",utc=True)
jobs22["build_id"]=BUILD_ID22
if jobs22.empty:
    raise RuntimeError("No valid jobs remain after validation. SQL tables were not modified.")

def sql_value22(value):
    if value is None:
        return None
    try:
        if pd.isna(value):
            return None
    except (TypeError,ValueError):
        pass
    if isinstance(value,pd.Timestamp):
        if value.tzinfo is not None:
            value=value.tz_convert("UTC").tz_localize(None)
        return value.to_pydatetime()
    if isinstance(value,np.generic):
        return value.item()
    return value

insert_cols22=[
    "title","title_normalized","link","link_hash","published_date",
    "published_day","country_clean","country_code","continent","pay_type",
    "job_family","seniority_level","is_hourly","hourly_low","hourly_high",
    "hourly_midpoint","budget","fixed_budget","hourly_outlier_flag",
    "fixed_outlier_flag","build_id"
]
insert_sql22=f"INSERT INTO dbo.jobs ({','.join(insert_cols22)}) VALUES ({','.join(['?']*len(insert_cols22))})"

def ensure_sql_conn22(force=False):
    global sql_conn22
    existing=globals().get("sql_conn22")
    if existing is not None and not force:
        try:
            test=existing.cursor()
            test.execute("SELECT 1").fetchone()
            test.close()
            return existing
        except pyodbc.Error:
            pass
    if existing is not None:
        try:
            existing.close()
        except Exception:
            pass
    sql_conn22=None
    last_error=None
    for attempt in range(1,4):
        try:
            sql_conn22=pyodbc.connect(
                connection_string22(SQL_DATABASE),
                timeout=20,
                autocommit=False
            )
            sql_conn22.timeout=600  # Statement timeout, not login timeout.
            test=sql_conn22.cursor()
            actual_db=test.execute("SELECT DB_NAME()").fetchone()[0]
            test.close()
            sql_conn22.commit()
            if str(actual_db).casefold()!=str(SQL_DATABASE).casefold():
                sql_conn22.close()
                sql_conn22=None
                raise RuntimeError(
                    f"Wrong SQL database: expected {SQL_DATABASE}, got {actual_db}. "
                    "No tables have been modified."
                )
            return sql_conn22
        except pyodbc.Error as exc:
            last_error=exc
            sql_conn22=None
            if attempt<3:
                time.sleep(attempt)
    raise ConnectionError(
        f"SQL Server {SQL_DATABASE} was unreachable after three connection attempts. "
        "Confirm the SQL Server service is running and the Section 22.1 "
        f"connection settings work. ODBC error: {last_error}"
    )

def verify22(connection,exact=False):
    check=connection.cursor()
    try:
        # One SQL table scan instead of three, reducing load on SQL Express.
        result=check.execute("""
            SELECT COUNT_BIG(*),
                   SUM(CASE WHEN build_id=? THEN CAST(1 AS BIGINT)
                            ELSE CAST(0 AS BIGINT) END),
                   MIN(published_date), MAX(published_date)
            FROM dbo.jobs
        """, BUILD_ID22).fetchone()
        count=int(result[0])
        build_count=int(result[1] or 0)
        dates=(result[2],result[3])
        verified=(count==len(jobs22) and build_count==len(jobs22))
        if verified and exact:
            expected=set(jobs22["link_hash"])
            check.execute("SELECT link_hash FROM dbo.jobs")
            actual=set()
            while True:
                batch=check.fetchmany(5000)
                if not batch:
                    break
                actual.update(str(item[0]) for item in batch)
            cleared=int(check.execute(
                "SELECT COUNT_BIG(*) FROM dbo.recommendation_results"
            ).fetchone()[0])==0
            verified=(expected==actual and cleared)
        return verified,count,dates
    finally:
        check.close()

load_started222=time.perf_counter()
load_attempt222=1
batch_size222=1000
load_status22="NOT_LOADED"
phase222="opening SQL connection"
sql_rows22=None
sql_dates22=(None,None)
commit_attempt222=False
conn222=ensure_sql_conn22(force=True)

phase222="checking required tables and columns"
preflight222=conn222.cursor()
try:
    if preflight222.execute("SELECT OBJECT_ID(N'dbo.jobs',N'U')").fetchone()[0] is None:
        raise RuntimeError("dbo.jobs does not exist. Run Section 22.1 first. No rows deleted.")
    if preflight222.execute("SELECT OBJECT_ID(N'dbo.recommendation_results',N'U')").fetchone()[0] is None:
        raise RuntimeError(
            "dbo.recommendation_results does not exist. Run Section 22.1 first. "
            "No rows deleted."
        )
    preflight222.execute(f"SELECT TOP (0) {','.join(insert_cols22)} FROM dbo.jobs")
finally:
    preflight222.close()
sql_conn22.commit()

try:
    phase222="beginning transaction"
    cur=sql_conn22.cursor()
    cur.execute("SET XACT_ABORT ON")
    cur.execute("SET NOCOUNT ON")
    cur.execute("SET LOCK_TIMEOUT 30000")  # Report blocked statements quickly.
    phase222="deleting old recommendation records"
    cur.execute("DELETE FROM dbo.recommendation_results")
    phase222="deleting old jobs"
    cur.execute("DELETE FROM dbo.jobs")
    phase222="checking identity column"
    identity_exists=cur.execute("""
        SELECT COUNT(*) FROM sys.columns
        WHERE object_id=OBJECT_ID(N'dbo.jobs') AND is_identity=1
    """).fetchone()[0]
    if identity_exists:
        phase222="resetting identity sequence"
        cur.execute("DBCC CHECKIDENT ('dbo.jobs', RESEED, 0) WITH NO_INFOMSGS")
    cur.fast_executemany=True
    for start in range(0,len(jobs22),batch_size222):
        phase222=f"inserting records {start + 1:,}–{min(start + batch_size222,len(jobs22)):,}"
        chunk=jobs22.iloc[start:start+batch_size222]
        rows=[
            tuple(sql_value22(value) for value in row)
            for row in chunk[insert_cols22].itertuples(index=False,name=None)
        ]
        cur.executemany(insert_sql22,rows)
        if start == 0 or (start // batch_size222 + 1) % 20 == 0:
            print(f"Section 22.2: {min(start + batch_size222,len(jobs22)):,}/{len(jobs22):,} prepared records inserted (pending commit).")
    phase222="pre-commit reconciliation"
    verified222,prepared_sql_rows222,prepared_sql_dates222=verify22(sql_conn22)
    if not verified222:
        raise RuntimeError(
            f"Pre-commit reconciliation failed: prepared={len(jobs22):,}, "
            f"SQL={prepared_sql_rows222:,}. Nothing has been committed."
        )
    phase222="committing the validated load"
    commit_attempt222=True
    sql_conn22.commit()
except Exception as exc:
    if commit_attempt222:
        # SQL can commit just before a network drop. Verify it before deciding what happened.
        try:
            ensure_sql_conn22(force=True)
            recovered222,sql_rows22,sql_dates22=verify22(sql_conn22,exact=True)
        except Exception as verification_error:
            load_status22="UNCERTAIN"
            raise RuntimeError(
                "Connection failed during COMMIT. Final transaction status could not "
                "be verified. Stop here and inspect dbo.jobs before rerunning Section 22.2. "
                f"Commit error: {exc}; verification error: {verification_error}"
            ) from exc
        if not recovered222:
            load_status22="UNCERTAIN"
            raise RuntimeError(
                "Connection failed during COMMIT and the committed data could not "
                "be confirmed. Do not run later cells until Section 22.2 succeeds. "
                f"Original error: {exc}"
            ) from exc
    else:
        load_status22="FAILED"
        try:
            sql_conn22.rollback()
        except Exception:
            pass
        try:
            sql_conn22.close()  # Release locks and unusable sessions after failure.
        except Exception:
            pass
        sql_conn22=None
        raise RuntimeError(
            f"Section 22.2 failed at: {phase222}. "
            "No COMMIT was attempted; existing committed data should remain unchanged "
            "after rollback/recovery. Do not run later cells. "
            "HYT00 can indicate slow execution, blocking, or SQL Server resource pressure. "
            f"Database error: {exc}"
        ) from exc

# Read committed records and leave a live pyodbc connection and cursor for later cells.
phase222="post-commit reconciliation"
try:
    sql_conn22=ensure_sql_conn22()
    valid222,sql_rows22,sql_dates22=verify22(sql_conn22)
except pyodbc.Error:
    sql_conn22=ensure_sql_conn22(force=True)
    valid222,sql_rows22,sql_dates22=verify22(sql_conn22,exact=True)
if not valid222:
    load_status22="REVIEW"
    raise RuntimeError(
        "Section 22.2 post-commit reconciliation failed. Stop before Section 22.3. "
        f"Prepared={len(jobs22):,}, database={sql_rows22:,}."
    )
sql_conn22.commit()

cur=sql_conn22.cursor()
load_status22="PASS"
load_seconds222=round(time.perf_counter()-load_started222,1)
load_summary22=pd.DataFrame({
    "Metric":["Prepared Rows","SQL Rows","Removed Rows","First Posting","Latest Posting","Build ID","Status"],
    "Value":[f"{len(jobs22):,}",f"{sql_rows22:,}",f"{len(source22)-len(jobs22):,}",sql_dates22[0],sql_dates22[1],BUILD_ID22,load_status22]
})
LOGGER_ACTIVE222=(
    "logger" in globals() and logger is not None and
    getattr(logger,"isEnabledFor",lambda level:False)(20)
)

from html import escape as html_escape222

removed_rows222=len(source22)-len(jobs22)
retention_pct222=(100*len(jobs22)/len(source22)) if len(source22) else 0.0
sql_match_pct222=(100*sql_rows22/len(jobs22)) if len(jobs22) else 0.0
unique_hashes222=int(jobs22["link_hash"].nunique())
country_coverage222=int(jobs22["country_clean"].notna().sum())
date_coverage222=int(jobs22["published_date"].notna().sum())
quality_columns222=[
    "title","link","published_date","country_clean","job_family",
    "pay_type","hourly_midpoint","fixed_budget"
]
quality_summary222=pd.DataFrame([
    {
        "FIELD":column,
        "NON-NULL":f"{int(jobs22[column].notna().sum()):,}",
        "MISSING":f"{int(jobs22[column].isna().sum()):,}",
        "COMPLETENESS":f"{100*jobs22[column].notna().mean():.1f}%"
    }
    for column in quality_columns222
])


def card222(label,value,note):
    label=html_escape222(str(label))
    value=html_escape222(str(value))
    note=html_escape222(str(note))
    return f'<div class="c222"><small>{label}</small><b>{value}</b><span>{note}</span></div>'


def table222(df):
    display(HTML(
        '<div class="tablewrap222">'
        +df.to_html(index=False,border=0,classes="t222",escape=True)
        +'</div>'
    ))


def panel222(eyebrow,title,body):
    return (
        '<div class="box222">'
        f'<span class="brow222">{html_escape222(str(eyebrow))}</span>'
        f'<strong>{html_escape222(str(title))}</strong>'
        f'<p>{body}</p>'
        '</div>'
    )


def progress222(label,value,percent,note):
    pct=max(0,min(100,float(percent)))
    return (
        '<div class="progressline222">'
        '<div class="progresshead222">'
        f'<strong>{html_escape222(str(label))}</strong>'
        f'<span>{html_escape222(str(value))}</span>'
        '</div>'
        f'<div class="track222"><div class="fill222" style="width:{pct:.2f}%"></div></div>'
        f'<small>{html_escape222(str(note))}</small>'
        '</div>'
    )


display(HTML("""
<style>
.d222{font-family:Arial,Helvetica,sans-serif;border:1px solid #ded8d5;border-radius:17px;overflow:hidden;background:#fff;box-shadow:0 8px 24px #22223b17;width:100%;}
.d222 .widget-html-content,.d222 .widget-html-content *{box-sizing:border-box;}
.h222{background:linear-gradient(120deg,#22223b 0%,#30324d 55%,#4a4e69 100%);color:#fff;padding:24px 27px;display:flex;align-items:flex-start;justify-content:space-between;gap:25px;}
.h222 .k{color:#c9ada7;font-size:9px;font-weight:800;letter-spacing:1.4px;}
.h222 h2{margin:7px 0 8px;font-size:25px;letter-spacing:-.4px;font-weight:800;color:#fff!important;}
.h222 p{margin:0;color:#f2e9e4;font-size:11px;line-height:1.75;}
.h222 .headmeta222{display:flex;gap:8px;flex-wrap:wrap;justify-content:flex-end;max-width:42%;}
.h222 .headmeta222 div{border:1px solid #9a8c984c;background:#ffffff10;padding:9px 12px;border-radius:9px;min-width:90px;}
.h222 .headmeta222 small{display:block;color:#c9ada7;font-size:8px;letter-spacing:1px;font-weight:700;}
.h222 .headmeta222 b{display:block;color:#f2e9e4;font-size:11px;margin-top:4px;word-break:break-word;}
.subhead222{background:linear-gradient(100deg,#fff,#fcfaf9);padding:10px 22px;display:flex;gap:10px;align-items:center;border-bottom:1px solid #e9e3df;}
.pill222{padding:5px 10px;border-radius:40px;background:#efeaf0;color:#4a4e69;font-size:9px;font-weight:700;white-space:nowrap;}
.pill222.ok222{background:#e5f3e9;color:#28663d;}
.subhead222 .helper222{margin-left:auto;color:#7b7482;font-size:9px;}
.cards222{display:grid;grid-template-columns:repeat(6,minmax(0,1fr));background:#fff;border-bottom:1px solid #eae3df;}
.c222{padding:17px 14px;border-right:1px solid #ebe6e4;min-width:0;}
.c222:last-child{border-right:0;}
.c222 small{display:block;color:#9a8c98;letter-spacing:1px;font-size:9px;font-weight:800;}
.c222 b{display:block;color:#22223b;font-size:21px;margin:8px 0 5px;font-weight:800;overflow-wrap:anywhere;}
.c222 span{color:#777286;font-size:10px;line-height:1.5;}
.btn222{margin:11px 18px!important;}
.d222 .btn222 .widget-toggle-button{font-size:10px;font-weight:800!important;letter-spacing:.3px;border-radius:7px!important;}
.d222 .btn222 .widget-toggle-button.mod-active{background:#4a4e69!important;color:#fff!important;border-color:#4a4e69!important;}
.out222{border-top:1px solid #eee9e7!important;padding:19px 22px 15px;min-height:235px;}
.out222 .output{overflow:visible;}
.sectionlabel222{color:#9a8c98;font-size:9px;font-weight:800;letter-spacing:1.4px;margin-bottom:11px;}
.grid222{display:grid;grid-template-columns:repeat(3,minmax(0,1fr));gap:13px;}
.box222{background:#f8f6f5;border:1px solid #ede8e5;border-left:4px solid #4a4e69;padding:16px;border-radius:11px;color:#22223b;min-height:108px;}
.box222 .brow222{font-size:9px;font-weight:800;color:#9a8c98;letter-spacing:1px;display:block;margin-bottom:7px;}
.box222 strong{font-size:18px;color:#22223b;display:block;margin:3px 0 8px;}
.box222 p{font-size:11px;color:#4a4e69;line-height:1.75;margin:0;}
.note222{background:#f2e9e4;padding:13px 16px;border-radius:9px;font-size:11px;line-height:1.7;color:#4a4e69;border-left:3px solid #c9ada7;margin-top:15px;}
.note222 b{color:#22223b;}
.progressgrid222{display:grid;grid-template-columns:repeat(2,1fr);gap:12px;margin-top:16px;}
.progressline222{background:white;border:1px solid #ebe4e1;border-radius:10px;padding:12px 14px;}
.progresshead222{display:flex;justify-content:space-between;gap:9px;color:#22223b;font-size:11px;}
.progresshead222 span{font-weight:800;}
.track222{height:8px;background:#eee9e7;border-radius:5px;overflow:hidden;margin:10px 0 7px;}
.fill222{height:100%;background:linear-gradient(90deg,#9a8c98,#4a4e69);border-radius:5px;}
.progressline222 small{color:#857982;font-size:9px;}
.tablewrap222{overflow-x:auto;border:1px solid #eae4e1;border-radius:10px;}
.t222{width:100%;border-collapse:collapse;font:11px Arial,Helvetica,sans-serif;min-width:440px;}
.t222 th{background:#22223b!important;color:#fff!important;padding:11px;text-align:left;font-size:9px;letter-spacing:.6px;text-transform:uppercase;}
.t222 td{padding:10px 11px;border-bottom:1px solid #eee9e7;color:#22223b;}
.t222 tr:nth-child(even){background:#fbfaf9;}
.foot222{border-top:1px solid #eee9e7;padding:11px 22px;background:#fbfaf9;font-size:9px;color:#9a8c98;display:flex;justify-content:space-between;gap:15px;}
@media(max-width:1050px){.cards222{grid-template-columns:repeat(3,minmax(0,1fr))}.grid222{grid-template-columns:1fr 1fr}.h222 .headmeta222{max-width:52%;}}
@media(max-width:700px){.h222{display:block}.h222 .headmeta222{margin-top:16px;max-width:none;justify-content:flex-start}.cards222{grid-template-columns:repeat(2,minmax(0,1fr))}.grid222,.progressgrid222{grid-template-columns:1fr}.subhead222{flex-wrap:wrap}.subhead222 .helper222{margin-left:0}}
</style>
"""))

header222=widgets.HTML(f"""
<div class="h222">
  <div>
    <div class="k">SECTION 22.2 &nbsp;•&nbsp; SQL DATA LOAD</div>
    <h2>Job Data Load &amp; Reconciliation</h2>
    <p>Validated Dataset &nbsp;•&nbsp; Transactional Batch Insert &nbsp;•&nbsp; SHA-256 Integrity &nbsp;•&nbsp; Reconciliation</p>
  </div>
  <div class="headmeta222">
    <div><small>BUILD ID</small><b>{html_escape222(str(BUILD_ID22))}</b></div>
    <div><small>DATABASE</small><b>{html_escape222(str(SQL_DATABASE))}</b></div>
    <div><small>LOGGER</small><b>{'ACTIVE' if LOGGER_ACTIVE222 else 'CHECK'}</b></div>
  </div>
</div>
""")

status_strip222=widgets.HTML(f"""
<div class="subhead222">
  <span class="pill222 ok222">&#10003; {html_escape222(str(load_status22))} &nbsp;·&nbsp; Committed &amp; verified</span>
  <span class="pill222">SQL Server</span>
  <span class="pill222">5,000-row batches</span>
  <span class="helper222">Read-only analytics below · No additional SQL writes</span>
</div>
""")

kpi222=widgets.HTML('<div class="cards222">'+''.join([
    card222("LOAD STATUS",load_status22,"Post-commit verification"),
    card222("SOURCE ROWS",f"{len(source22):,}","Initial records"),
    card222("VALIDATED",f"{len(jobs22):,}","Nonblank + unique links"),
    card222("SQL LOADED",f"{sql_rows22:,}","Committed dbo.jobs rows"),
    card222("REMOVED",f"{removed_rows222:,}",f"{100-retention_pct222:.2f}% of source"),
    card222("DURATION",f"{load_seconds222:.1f}s",f"{batch_size222:,}-row batches")
])+'</div>')

view222=widgets.ToggleButtons(
    options=[
        ("OVERVIEW","overview"),
        ("RECONCILIATION","summary"),
        ("DATA QUALITY","quality"),
        ("SAMPLE DATA","data"),
        ("METHOD","method")
    ],
    value="overview",
    layout=widgets.Layout(width="100%"),
    style={"button_width":"20%","font_weight":"700"}
)
view222.add_class("btn222")
out222=widgets.Output()
out222.add_class("out222")


def update222(*_):
    with out222:
        clear_output(wait=True)
        if view222.value=="overview":
            display(HTML(
                '<div class="sectionlabel222">SQL LOAD AT A GLANCE</div>'
                '<div class="grid222">'
                +panel222("ROW RECONCILIATION",f"{sql_rows22:,} / {len(jobs22):,}",
                    f"Database records match prepared data. Status: <b>{html_escape222(str(load_status22))}</b>.")
                +panel222("LINK INTEGRITY",f"{unique_hashes222:,} unique",
                    "Valid posting links are deduplicated using SHA-256 hashes before insertion.")
                +panel222("TRANSACTION SAFETY",f"{load_seconds222:.1f} seconds",
                    "Deletes, inserts and row-count reconciliation are completed before commit.")
                +'</div>'
                +'<div class="progressgrid222">'
                +progress222("Validated source retention",f"{len(jobs22):,} records",retention_pct222,
                    f"{retention_pct222:.2f}% retained after invalid / duplicate exclusion")
                +progress222("SQL reconciliation",f"{sql_rows22:,} committed",sql_match_pct222,
                    f"{sql_match_pct222:.2f}% of the prepared records verified")
                +'</div>'
                +'<div class="note222"><b>Compensation control:</b> Hourly rates and fixed budgets remain separate; no invented combined pay or currency-normalized measure.</div>'
            ))
        elif view222.value=="summary":
            display(HTML('<div class="sectionlabel222">SOURCE-TO-DATABASE RECONCILIATION</div>'))
            table222(load_summary22)
            display(HTML(
                '<div class="note222">'
                f'<b>Integrity gate:</b> {sql_rows22:,} records committed out of {len(jobs22):,} prepared. '
                'Only a successful committed load is marked PASS. '
                'All values here describe the completed load and do not execute additional SQL updates.'
                '</div>'
            ))
        elif view222.value=="quality":
            display(HTML(
                '<div class="sectionlabel222">DATA QUALITY &amp; LINK INTEGRITY</div>'
                '<div class="grid222">'
                +panel222("SHA-256 LINK HASHES",f"{unique_hashes222:,}",
                    f"{len(jobs22)-unique_hashes222:,} duplicate hashes remain in prepared records.")
                +panel222("COUNTRY COVERAGE",f"{country_coverage222:,}",
                    f"{100*country_coverage222/len(jobs22):.1f}% of loaded records have a country value.")
                +panel222("DATE COVERAGE",f"{date_coverage222:,}",
                    f"{100*date_coverage222/len(jobs22):.1f}% of loaded records have a valid timestamp.")
                +'</div><br>'
            ))
            table222(quality_summary222)
        elif view222.value=="data":
            display(HTML('<div class="sectionlabel222">VALIDATED INPUT PREVIEW · FIRST 15 ROWS</div>'))
            table222(jobs22[["title","country_clean","job_family","pay_type","published_date"]].head(15))
            display(HTML('<div class="note222"><b>Preview only:</b> The table displays prepared job records; it does not alter the dataset or reload SQL Server.</div>'))
        else:
            display(HTML(
                '<div class="sectionlabel222">REPEATABILITY, SAFETY &amp; GOVERNANCE</div>'
                '<div class="grid222">'
                +panel222("ATOMIC REFRESH","One transaction",
                    "Recommendation results and existing jobs are cleared inside the transaction; the new rows are committed only after reconciliation.")
                +panel222("LINK UNIQUENESS","SHA-256",
                    "Posting links are hashed and de-duplicated before insertion, keeping the existing database key structure.")
                +panel222("NULL / DATE HANDLING","SQL-compatible",
                    "Pandas missing values become SQL NULL; UTC timestamps become Python datetimes for ODBC parameters.")
                +'</div>'
                '<div class="note222"><b>Recovery:</b> An uncertain commit outcome is verified against the database. Never proceed to Section 22.3 unless <b>load_status22 == "PASS"</b>.</div>'
            ))

view222.observe(update222,names="value")
update222()
footer222=widgets.HTML(
    '<div class="foot222"><span>SECTION 22.2 &nbsp;•&nbsp; JOB MARKET INTELLIGENCE</span>'
    '<span>Validated SQL layer &nbsp;·&nbsp; Ready for Section 22.3</span></div>'
)
dashboard_222=widgets.VBox([header222,status_strip222,kpi222,view222,out222,footer222])
dashboard_222.add_class("d222")
display(dashboard_222)

if "logger" in globals():
    logger.info(
        "SECTION22_2_COMPLETE | source=%s | loaded=%s | removed=%s | status=%s | batch=%s | attempt=%s",
        len(source22),sql_rows22,removed_rows222,load_status22,batch_size222,load_attempt222
    )
```

```python
queries225=r"""
-- Job Market Analysis | SQL Server Query Library

-- 1. Database health
SELECT
    COUNT(*) AS jobs,
    MIN(published_date) AS first_posting,
    MAX(published_date) AS latest_posting
FROM dbo.jobs;

-- 2. Top job families
SELECT TOP 10
    job_family,
    COUNT(*) AS postings
FROM dbo.jobs
GROUP BY job_family
ORDER BY postings DESC;

-- 3. Country demand
SELECT
    country_clean,
    COUNT(*) AS postings
FROM dbo.jobs
GROUP BY country_clean
ORDER BY postings DESC;

-- 4. Hourly compensation
SELECT
    country_clean,
    COUNT(hourly_midpoint) AS sample_size,
    AVG(hourly_midpoint) AS avg_hourly
FROM dbo.jobs
WHERE hourly_midpoint IS NOT NULL
GROUP BY country_clean
ORDER BY sample_size DESC;

-- 5. Fixed-budget compensation
SELECT
    country_clean,
    COUNT(fixed_budget) AS sample_size,
    AVG(fixed_budget) AS avg_fixed_budget
FROM dbo.jobs
WHERE fixed_budget IS NOT NULL
GROUP BY country_clean
ORDER BY sample_size DESC;

-- 6. Daily demand
SELECT
    published_day,
    job_family,
    COUNT(*) AS postings
FROM dbo.jobs
GROUP BY published_day,job_family
ORDER BY published_day,job_family;

-- 7. Seven-day demand comparison
WITH daily AS(
    SELECT
        published_day,
        job_family,
        COUNT(*) AS postings
    FROM dbo.jobs
    GROUP BY published_day,job_family
)
SELECT
    *,
    LAG(postings,7) OVER(
        PARTITION BY job_family
        ORDER BY published_day
    ) AS postings_7d_ago
FROM daily;

-- 8. Forecasts
SELECT
    job_family,
    forecast_day,
    selected_model,
    forecast_value,
    model_version
FROM dbo.forecast_predictions
ORDER BY job_family,forecast_day;

-- 9. Scenarios
SELECT
    scenario_name,
    SUM(scenario_forecast) AS total_forecast
FROM dbo.scenario_forecasts
GROUP BY scenario_name
ORDER BY total_forecast DESC;

-- 10. Model registry
SELECT
    component,
    version,
    scope,
    method,
    primary_metric,
    primary_value,
    status
FROM dbo.model_registry
ORDER BY component,scope;

-- Parameterized Flask query example:
-- WHERE job_family = ?
"""

(DATABASE_DIR/"queries.sql").write_text(
    queries225,
    encoding="utf-8"
)

(EXPORT_DIR/"sql_query_examples.txt").write_text(
    queries225,
    encoding="utf-8"
)


database_py225=r'''
import os
import pyodbc
import pandas as pd

SERVER=os.getenv(
    "SQLSERVER_SERVER",
    r"localhost\SQLEXPRESS"
)

DATABASE=os.getenv(
    "SQLSERVER_DATABASE",
    "JobMarketDB"
)

DRIVER=os.getenv(
    "SQLSERVER_DRIVER",
    "ODBC Driver 18 for SQL Server"
)


def get_connection():
    return pyodbc.connect(
        f"DRIVER={{{DRIVER}}};"
        f"SERVER={SERVER};"
        f"DATABASE={DATABASE};"
        "Trusted_Connection=yes;"
        "Encrypt=yes;"
        "TrustServerCertificate=yes;",
        timeout=15
    )


def query_df(sql,params=None):
    with get_connection() as conn:
        cur=conn.cursor()
        cur.execute(sql,params or [])
        cols=[c[0] for c in cur.description]
        return pd.DataFrame(
            [tuple(r) for r in cur.fetchall()],
            columns=cols
        )


def health_check():
    return query_df("""
        SELECT
            COUNT(*) AS jobs,
            MAX(published_date) AS data_as_of
        FROM dbo.jobs
    """)


def market_summary():
    return query_df("""
        SELECT
            COUNT(*) AS total_postings,
            COUNT(DISTINCT job_family) AS job_families,
            COUNT(DISTINCT country_clean) AS countries,
            MAX(published_date) AS data_as_of
        FROM dbo.jobs
    """)


def job_family_daily(job_family):
    return query_df(
        """
        SELECT
            published_day,
            COUNT(*) AS postings
        FROM dbo.jobs
        WHERE job_family=?
        GROUP BY published_day
        ORDER BY published_day
        """,
        [job_family]
    )


def country_market(country):
    return query_df(
        """
        SELECT
            job_family,
            COUNT(*) AS postings,
            COUNT(hourly_midpoint) AS hourly_sample,
            AVG(hourly_midpoint) AS avg_hourly
        FROM dbo.jobs
        WHERE country_clean=?
        GROUP BY job_family
        ORDER BY postings DESC
        """,
        [country]
    )


def forecast_by_family(job_family):
    return query_df(
        """
        SELECT
            forecast_day,
            selected_model,
            forecast_value,
            model_version
        FROM dbo.forecast_predictions
        WHERE job_family=?
        ORDER BY forecast_day
        """,
        [job_family]
    )


def scenario_by_family(job_family):
    return query_df(
        """
        SELECT
            scenario_name,
            base_forecast,
            scenario_forecast,
            difference_pct,
            risk
        FROM dbo.scenario_forecasts
        WHERE job_family=?
        ORDER BY scenario_name
        """,
        [job_family]
    )
'''

(SRC_DATA_DIR/"database.py").write_text(
    database_py225,
    encoding="utf-8"
)


test_family225=(
    jobs22["job_family"]
    .dropna()
    .value_counts()
    .index[0]
)

query_test225=sql_frame224(
    """
    SELECT
        published_day,
        COUNT(*) AS postings
    FROM dbo.jobs
    WHERE job_family=?
    GROUP BY published_day
    ORDER BY published_day
    """,
    [test_family225]
)

files225=pd.DataFrame({
    "Artifact":[
        "SQL Query Library",
        "Python Data Layer",
        "Query Evidence"
    ],
    "Path":[
        "database/queries.sql",
        "src/data/database.py",
        "exports/sql_query_examples.txt"
    ],
    "Status":["Ready"]*3
})


def card225(a,b,c):
    return f"""
    <div class="c225">
    <small>{a}</small><b>{b}</b><span>{c}</span>
    </div>
    """

def table225(df):
    display(HTML(df.to_html(
        index=False,
        border=0,
        classes="t225"
    )))


display(HTML("""
<style>
.d225{border:1px solid #ded8d5;border-radius:17px;overflow:hidden;
background:#fff;box-shadow:0 8px 24px #22223b17}
.h225{background:linear-gradient(120deg,#22223b,#30324d 55%,#4a4e69);
padding:24px 27px;display:flex;justify-content:space-between;font-family:Arial}
.h225 .k{color:#c9ada7;font-size:8px;font-weight:700;letter-spacing:1.6px}
.h225 h2{color:#fff;margin:6px 0;font-size:24px}
.h225 p,.h225 .m{color:#f2e9e4;font-size:10px;line-height:1.65}
.cards225{display:grid;grid-template-columns:repeat(4,1fr);
border-bottom:1px solid #e8e2df}
.c225{padding:14px;border-right:1px solid #ebe6e4;font-family:Arial}
.c225 small{display:block;color:#9a8c98;font-size:7px;font-weight:700}
.c225 b{display:block;color:#22223b;font-size:17px;margin-top:5px}
.c225 span{display:block;color:#4a4e69;font-size:8px;margin-top:4px}
.btn225{margin:8px 16px!important}
.out225{border-top:1px solid #eee9e7!important;padding:18px 22px}
.t225{width:100%;border-collapse:collapse;font:10px Arial}
.t225 th{background:#22223b;color:#fff;padding:9px;text-align:left}
.t225 td{padding:8px;border-bottom:1px solid #eee9e7;color:#22223b}
.grid225{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}
.box225{background:#f8f6f5;border-left:4px solid #4a4e69;
padding:14px;border-radius:8px;font:11px/1.65 Arial;color:#22223b}
.box225 b{display:block;color:#4a4e69;font-size:9px;margin-bottom:6px}
.note225{background:#f2e9e4;padding:12px 14px;border-radius:8px;
font:10px/1.6 Arial;color:#4a4e69}
</style>
"""))


header225=widgets.HTML(f"""
<div class="h225">
<div>
<div class="k">SECTION 22.5 • DATA ACCESS LAYER</div>
<h2>SQL Query & Application Access Center</h2>
<p>Parameterized Queries • Reusable Module • Flask Readiness</p>
</div>
<div class="m">
DATABASE<br><b>{SQL_DATABASE}</b><br><br>
TEST FAMILY<br><b>{test_family225}</b>
</div>
</div>
""")


kpi225=widgets.HTML(
    '<div class="cards225">'+
    "".join([
        card225("QUERY LIBRARY","10","Core SQL examples"),
        card225("ACCESS MODULE","READY","database.py"),
        card225(
            "QUERY TEST",
            f"{len(query_test225):,}",
            "Rows returned"
        ),
        card225("SECURITY","PARAMETERIZED","No raw input concat")
    ])+
    "</div>"
)


view225=widgets.ToggleButtons(
    options=[
        ("OVERVIEW","overview"),
        ("QUERY TEST","query"),
        ("FILES","files"),
        ("METHOD","method")
    ],
    value="overview",
    layout=widgets.Layout(width="100%"),
    style={
        "button_width":"25%",
        "font_weight":"700"
    }
)

view225.add_class("btn225")

out225=widgets.Output()
out225.add_class("out225")


def update225(*_):

    with out225:
        clear_output(wait=True)

        if view225.value=="overview":

            display(HTML("""
            <div class="grid225">

            <div class="box225">
            <b>QUERY LIBRARY</b>
            Reusable SQL covers demand, compensation,
            forecasts, scenarios and model governance.
            </div>

            <div class="box225">
            <b>APPLICATION MODULE</b>
            database.py provides reusable SQL access
            functions for the Flask API layer.
            </div>

            <div class="box225">
            <b>SECURITY</b>
            Application filters are passed separately
            through SQL parameters rather than concatenation.
            </div>

            </div><br>

            <div class="note225">
            Streamlit will consume Flask endpoints. Flask will use
            this database module to communicate with SQL Server.
            </div>
            """))

        elif view225.value=="query":
            table225(query_test225.head(15))

        elif view225.value=="files":
            table225(files225)

        else:

            display(HTML("""
            <div class="grid225">

            <div class="box225">
            <b>SEPARATION OF CONCERNS</b>
            Database connection and query logic are isolated
            from dashboard and API presentation code.
            </div>

            <div class="box225">
            <b>PARAMETERIZATION</b>
            User-selected country and job-family values use
            SQL placeholders such as WHERE job_family = ?.
            </div>

            <div class="box225">
            <b>COMPENSATION CONTROL</b>
            Hourly and fixed compensation remain separate
            in both SQL queries and application access.
            </div>

            </div>
            """))


view225.observe(
    update225,
    names="value"
)

update225()

dashboard_225=widgets.VBox([
    header225,
    kpi225,
    view225,
    out225
])

dashboard_225.add_class("d225")
display(dashboard_225)
```

<div style="color:#f2e9e4;font-size:11px;line-height:1.65;margin-top:8px">Loading finalized model artifacts into a standalone production inference layer, validating input and output contracts, generating forecasts and recommendations without retraining, and preparing reusable functions for later API integration.</div>

```python
required24=[
    "MODEL_VERSION",
    "RECOMMENDER_VERSION",
    "REC_ASSET_DIR"
]

missing24=[
    x for x in required24
    if x not in globals()
]

if missing24:
    raise NameError(
        f"Run Sections 18–23 first. Missing: {missing24}"
    )

SRC_DIR24=Path("src")
SRC_DIR24.mkdir(exist_ok=True)

(SRC_DIR24/"__init__.py").touch(
    exist_ok=True
)

inference_code24=r'''
from pathlib import Path
import json
import pickle

import numpy as np
import pandas as pd

from scipy.sparse import load_npz
from sklearn.metrics.pairwise import cosine_similarity


BASE_DIR=Path.cwd()

MODEL_DIR=BASE_DIR/"models"

FORECAST_PATH=MODEL_DIR/"demand_v1.pkl"

REC_DIR=MODEL_DIR/"recommender_assets"

CATALOG_PATH=REC_DIR/"recommendation_catalog.csv"
VECTORIZER_PATH=REC_DIR/"tfidf_vectorizer.pkl"
TFIDF_PATH=REC_DIR/"tfidf_matrix.npz"
WEIGHTS_PATH=REC_DIR/"weights.json"
METADATA_PATH=REC_DIR/"metadata.json"


def _require_files(paths):
    missing=[
        str(p)
        for p in paths
        if not Path(p).exists()
    ]

    if missing:
        raise FileNotFoundError(
            "Missing inference artifacts: "
            + ", ".join(missing)
        )


_require_files([
    FORECAST_PATH,
    CATALOG_PATH,
    VECTORIZER_PATH,
    TFIDF_PATH,
    WEIGHTS_PATH,
    METADATA_PATH
])


with open(
    FORECAST_PATH,
    "rb"
) as f:
    FORECAST_ARTIFACT=pickle.load(f)


with open(
    VECTORIZER_PATH,
    "rb"
) as f:
    REC_VECTORIZER=pickle.load(f)


REC_MATRIX=load_npz(
    TFIDF_PATH
)


REC_CATALOG=pd.read_csv(
    CATALOG_PATH
)

REC_CATALOG["published_date"]=pd.to_datetime(
    REC_CATALOG["published_date"],
    errors="coerce",
    utc=True
)


with open(
    WEIGHTS_PATH,
    "r",
    encoding="utf-8"
) as f:
    REC_WEIGHT_CONFIG=json.load(f)


with open(
    METADATA_PATH,
    "r",
    encoding="utf-8"
) as f:
    REC_METADATA=json.load(f)


REC_WEIGHTS=REC_WEIGHT_CONFIG["weights"]


if len(REC_CATALOG)!=REC_MATRIX.shape[0]:
    raise ValueError(
        "Recommendation catalogue and TF-IDF matrix "
        "have different row counts."
    )


def available_job_families():
    return sorted(
        FORECAST_ARTIFACT["models"].keys()
    )


def inference_metadata():
    return {
        "forecast_version":
            FORECAST_ARTIFACT.get(
                "version"
            ),

        "forecast_horizon_days":
            FORECAST_ARTIFACT.get(
                "forecast_horizon_days"
            ),

        "forecast_selection_rule":
            FORECAST_ARTIFACT.get(
                "selection_rule"
            ),

        "recommender_version":
            REC_WEIGHT_CONFIG.get(
                "version"
            ),

        "recommender_weight_set":
            REC_WEIGHT_CONFIG.get(
                "selected_weight_set"
            ),

        "catalogue_rows":
            len(REC_CATALOG),

        "job_families":
            len(
                FORECAST_ARTIFACT[
                    "models"
                ]
            )
    }


def forecast_demand(job_family):
    family=str(job_family).strip()

    models=FORECAST_ARTIFACT[
        "models"
    ]

    if family not in models:
        raise ValueError(
            f"Unknown job family: {family}"
        )

    model=models[family]

    horizon=int(
        FORECAST_ARTIFACT.get(
            "forecast_horizon_days",
            7
        )
    )

    end_date=pd.Timestamp(
        FORECAST_ARTIFACT[
            "final_refit_end"
        ]
    )

    if end_date.tzinfo is None:
        end_date=end_date.tz_localize(
            "UTC"
        )
    else:
        end_date=end_date.tz_convert(
            "UTC"
        )

    future_dates=pd.date_range(
        end_date+pd.Timedelta(days=1),
        periods=horizon,
        freq="D",
        tz="UTC"
    )

    selected=model[
        "selected_model"
    ]

    if selected=="Trend + Weekday":

        start=pd.Timestamp(
            model["start_date"]
        )

        if start.tzinfo is None:
            start=start.tz_localize("UTC")
        else:
            start=start.tz_convert("UTC")

        t=np.array(
            [
                (d-start).days
                for d in future_dates
            ],
            dtype=float
        )

        dow=np.array(
            [
                d.dayofweek
                for d in future_dates
            ]
        )

        X=np.column_stack(
            [
                np.ones(horizon),
                t
            ]
            +
            [
                (dow==i).astype(int)
                for i in range(6)
            ]
        )

        forecast=np.clip(
            X@np.asarray(
                model["trend_beta"],
                dtype=float
            ),
            0,
            None
        )

    elif selected=="Lag-7":

        forecast=np.asarray(
            model["last_7_actual"],
            dtype=float
        )[:horizon]

    elif selected=="7-Day Mean":

        forecast=np.repeat(
            float(
                model[
                    "recent_7_mean"
                ]
            ),
            horizon
        )

    else:
        raise ValueError(
            f"Unsupported forecast model: {selected}"
        )

    result=pd.DataFrame({
        "job_family":family,
        "forecast_day":future_dates,
        "selected_model":selected,
        "forecast":forecast
    })

    result["forecast"]=(
        result["forecast"]
        .clip(lower=0)
        .round(2)
    )

    return result


def recommend_jobs(
    query,
    country="Any",
    pay_type="Any",
    seniority="Any",
    n=10,
    mode="boost"
):
    query=str(query).strip().lower()

    if len(query)<2:
        raise ValueError(
            "Query must contain at least 2 characters."
        )

    mode=str(mode).lower()

    if mode not in {
        "boost",
        "strict"
    }:
        raise ValueError(
            "mode must be 'boost' or 'strict'."
        )

    try:
        n=int(n)
    except Exception:
        raise ValueError(
            "n must be an integer."
        )

    n=max(
        1,
        min(n,50)
    )

    similarity=cosine_similarity(
        REC_VECTORIZER.transform(
            [query]
        ),
        REC_MATRIX
    ).ravel()

    d=REC_CATALOG.copy()

    d[
        "Title_Similarity"
    ]=similarity


    d["Country_Match"]=(
        1.0
        if country=="Any"
        else
        d["country_clean"]
        .eq(country)
        .astype(float)
    )

    d["Pay_Match"]=(
        1.0
        if pay_type=="Any"
        else
        d["pay_type"]
        .eq(pay_type)
        .astype(float)
    )

    d["Seniority_Match"]=(
        1.0
        if seniority=="Any"
        else
        d["seniority_level"]
        .eq(seniority)
        .astype(float)
    )


    if mode=="strict":

        if country!="Any":
            d=d[
                d["Country_Match"].eq(1)
            ]

        if pay_type!="Any":
            d=d[
                d["Pay_Match"].eq(1)
            ]

        if seniority!="Any":
            d=d[
                d[
                    "Seniority_Match"
                ].eq(1)
            ]


    active=[
        (
            "Title_Similarity",
            REC_WEIGHTS["title"]
        ),
        (
            "recency_score",
            REC_WEIGHTS["recency"]
        )
    ]


    if country!="Any":
        active.append(
            (
                "Country_Match",
                REC_WEIGHTS["country"]
            )
        )

    if pay_type!="Any":
        active.append(
            (
                "Pay_Match",
                REC_WEIGHTS["pay"]
            )
        )

    if seniority!="Any":
        active.append(
            (
                "Seniority_Match",
                REC_WEIGHTS[
                    "seniority"
                ]
            )
        )


    denominator=sum(
        weight
        for _,weight in active
    )


    d["Match_Score"]=sum(
        d[column]*weight
        for column,weight
        in active
    )/denominator


    d=(
        d[
            d[
                "Title_Similarity"
            ]>0
        ]
        .nlargest(
            n,
            "Match_Score"
        )
        .copy()
    )


    d["Match_%"]=(
        d["Match_Score"]*100
    ).round(1)


    reasons=[]

    for _,row in d.iterrows():

        reason=[
            "title similarity "
            f"{row['Title_Similarity']:.0%}"
        ]

        if (
            country!="Any"
            and
            row["Country_Match"]==1
        ):
            reason.append(
                "country match"
            )

        if (
            pay_type!="Any"
            and
            row["Pay_Match"]==1
        ):
            reason.append(
                "pay-type match"
            )

        if (
            seniority!="Any"
            and
            row[
                "Seniority_Match"
            ]==1
        ):
            reason.append(
                "seniority match"
            )

        if (
            row[
                "recency_score"
            ]>=0.75
        ):
            reason.append(
                "recent posting"
            )

        reasons.append(
            ", ".join(reason)
        )


    d["Why_Matched"]=reasons


    output_columns=[
        "job_id",
        "title",
        "country_clean",
        "pay_type",
        "job_family",
        "seniority_level",
        "hourly_midpoint",
        "fixed_budget",
        "published_date",
        "link",
        "Title_Similarity",
        "Country_Match",
        "Pay_Match",
        "Seniority_Match",
        "recency_score",
        "Match_Score",
        "Match_%",
        "Why_Matched"
    ]

    existing=[
        c
        for c in output_columns
        if c in d.columns
    ]

    return d[
        existing
    ].reset_index(drop=True)
'''


(SRC_DIR24/"inference.py").write_text(
    inference_code24,
    encoding="utf-8"
)


if str(Path.cwd()) not in sys.path:
    sys.path.insert(
        0,
        str(Path.cwd())
    )

importlib.invalidate_caches()

if "src.inference" in sys.modules:
    del sys.modules[
        "src.inference"
    ]

from src.inference import (
    forecast_demand,
    recommend_jobs as production_recommend24,
    available_job_families,
    inference_metadata
)


metadata24=inference_metadata()

families24=available_job_families()

if not families24:
    raise RuntimeError(
        "No forecast job families were loaded."
    )


test_family24=(
    model_registry18
    .sort_values("MAE")
    .iloc[0]["job_family"]
    if "model_registry18" in globals()
    else families24[0]
)


forecast_start24=time.perf_counter()

forecast_example24=forecast_demand(
    test_family24
)

forecast_ms24=(
    time.perf_counter()
    - forecast_start24
)*1000


rec_query24=(
    str(
        sample_profile19["Query"]
    )
    if "sample_profile19" in globals()
    else "data analyst"
)

rec_country24=(
    str(
        sample_profile19["Country"]
    )
    if (
        "sample_profile19" in globals()
        and
        pd.notna(
            sample_profile19[
                "Country"
            ]
        )
    )
    else "Any"
)

rec_pay24=(
    str(
        sample_profile19[
            "Pay_Type"
        ]
    )
    if (
        "sample_profile19" in globals()
        and
        pd.notna(
            sample_profile19[
                "Pay_Type"
            ]
        )
    )
    else "Any"
)

rec_seniority24=(
    str(
        sample_profile19[
            "Seniority"
        ]
    )
    if (
        "sample_profile19" in globals()
        and
        pd.notna(
            sample_profile19[
                "Seniority"
            ]
        )
    )
    else "Any"
)


rec_start24=time.perf_counter()

recommendation_example24=production_recommend24(
    query=rec_query24,
    country=rec_country24,
    pay_type=rec_pay24,
    seniority=rec_seniority24,
    n=10
)

recommendation_ms24=(
    time.perf_counter()
    - rec_start24
)*1000


forecast_schema24={
    "job_family",
    "forecast_day",
    "selected_model",
    "forecast"
}

recommendation_schema24={
    "title",
    "job_family",
    "Match_Score",
    "Match_%",
    "Why_Matched"
}


tests24=pd.DataFrame({
    "Test":[
        "Inference module import",
        "Forecast artifact loaded",
        "Forecast family available",
        "Forecast output generated",
        "Forecast schema valid",
        "Forecast values non-negative",
        "Recommender assets loaded",
        "Recommendation output generated",
        "Recommendation schema valid",
        "Recommendation scores bounded",
        "No model retraining during inference"
    ],
    "Status":[
        "PASS",
        (
            "PASS"
            if metadata24.get(
                "forecast_version"
            )==MODEL_VERSION
            else "REVIEW"
        ),
        (
            "PASS"
            if test_family24
            in families24
            else "REVIEW"
        ),
        (
            "PASS"
            if len(
                forecast_example24
            )>0
            else "REVIEW"
        ),
        (
            "PASS"
            if forecast_schema24.issubset(
                forecast_example24.columns
            )
            else "REVIEW"
        ),
        (
            "PASS"
            if (
                forecast_example24[
                    "forecast"
                ]>=0
            ).all()
            else "REVIEW"
        ),
        (
            "PASS"
            if metadata24.get(
                "recommender_version"
            )==RECOMMENDER_VERSION
            else "REVIEW"
        ),
        (
            "PASS"
            if len(
                recommendation_example24
            )>0
            else "REVIEW"
        ),
        (
            "PASS"
            if recommendation_schema24.issubset(
                recommendation_example24.columns
            )
            else "REVIEW"
        ),
        (
            "PASS"
            if (
                recommendation_example24[
                    "Match_Score"
                ].between(
                    0,
                    1
                ).all()
            )
            else "REVIEW"
        ),
        "PASS"
    ]
})


inference_status24=(
    "PASS"
    if (
        tests24["Status"]
        =="PASS"
    ).all()
    else "REVIEW"
)


inference_summary24=pd.DataFrame({
    "Component":[
        "Forecast Inference",
        "Recommendation Inference"
    ],
    "Version":[
        metadata24[
            "forecast_version"
        ],
        metadata24[
            "recommender_version"
        ]
    ],
    "Method":[
        "Per-family selected forecast model",
        "TF-IDF + validated preference weighting"
    ],
    "Example Input":[
        test_family24,
        rec_query24
    ],
    "Output Rows":[
        len(
            forecast_example24
        ),
        len(
            recommendation_example24
        )
    ],
    "Latency_ms":[
        round(
            forecast_ms24,
            2
        ),
        round(
            recommendation_ms24,
            2
        )
    ]
})


tests24.to_csv(
    EXPORT_DIR/
    "production_inference_tests.csv",
    index=False
)

inference_summary24.to_csv(
    EXPORT_DIR/
    "production_inference_summary.csv",
    index=False
)

forecast_example24.to_csv(
    EXPORT_DIR/
    "production_forecast_example.csv",
    index=False
)

recommendation_example24.to_csv(
    EXPORT_DIR/
    "production_recommendation_example.csv",
    index=False
)


def card24(a,b,c):
    return f"""
    <div class="c24">
    <small>{a}</small>
    <b>{b}</b>
    <span>{c}</span>
    </div>
    """


def table24(df):
    display(
        HTML(
            df.to_html(
                index=False,
                border=0,
                classes="t24",
                float_format=lambda x:
                f"{x:,.3f}"
            )
        )
    )


display(HTML("""
<style>
.d24{
border:1px solid #ded8d5;
border-radius:17px;
overflow:hidden;
background:#fff;
box-shadow:0 8px 24px #22223b17
}

.h24{
background:linear-gradient(
120deg,#22223b,#30324d 55%,#4a4e69
);
padding:24px 27px;
display:flex;
justify-content:space-between;
font-family:Arial
}

.h24 .k{
color:#c9ada7;
font-size:8px;
font-weight:700;
letter-spacing:1.6px
}

.h24 h2{
color:#fff;
margin:6px 0;
font-size:24px
}

.h24 p,.h24 .m{
color:#f2e9e4;
font-size:10px;
line-height:1.65
}

.cards24{
display:grid;
grid-template-columns:repeat(5,1fr);
border-bottom:1px solid #e8e2df
}

.c24{
padding:14px;
border-right:1px solid #ebe6e4;
font-family:Arial
}

.c24 small{
display:block;
color:#9a8c98;
font-size:7px;
font-weight:700
}

.c24 b{
display:block;
color:#22223b;
font-size:16px;
margin-top:5px;
overflow-wrap:anywhere
}

.c24 span{
display:block;
color:#4a4e69;
font-size:8px;
margin-top:4px
}

.btn24{
margin:8px 16px!important
}

.out24{
border-top:1px solid #eee9e7!important;
padding:18px 22px
}

.t24{
width:100%;
border-collapse:collapse;
font:10px Arial
}

.t24 th{
background:#22223b;
color:#fff;
padding:9px;
text-align:left
}

.t24 td{
padding:8px;
border-bottom:1px solid #eee9e7;
color:#22223b
}

.grid24{
display:grid;
grid-template-columns:repeat(3,1fr);
gap:12px
}

.box24{
background:#f8f6f5;
border-left:4px solid #4a4e69;
padding:14px;
border-radius:8px;
font:11px/1.65 Arial;
color:#22223b
}

.box24 b{
display:block;
color:#4a4e69;
font-size:9px;
margin-bottom:6px
}

.note24{
background:#f2e9e4;
padding:12px 14px;
border-radius:8px;
font:10px/1.65 Arial;
color:#4a4e69
}
</style>
"""))


header24=widgets.HTML(f"""
<div class="h24">

<div>
<div class="k">
SECTION 24 • PRODUCTION INFERENCE
</div>

<h2>
Model Serving Readiness
</h2>

<p>
Saved Artifacts • Reusable Functions •
Schema Validation • Smoke Testing
</p>
</div>

<div class="m">
INFERENCE STATUS<br>
<b>{inference_status24}</b><br><br>

MODE<br>
<b>NO RETRAINING</b>
</div>

</div>
""")


kpi24=widgets.HTML(
    '<div class="cards24">'+
    "".join([
        card24(
            "STATUS",
            inference_status24,
            "Inference readiness"
        ),
        card24(
            "FORECAST VERSION",
            metadata24[
                "forecast_version"
            ],
            "Loaded artifact"
        ),
        card24(
            "RECOMMENDER",
            metadata24[
                "recommender_weight_set"
            ],
            "Selected weights"
        ),
        card24(
            "FORECAST LATENCY",
            f"{forecast_ms24:.1f} ms",
            "Example request"
        ),
        card24(
            "REC LATENCY",
            f"{recommendation_ms24:.1f} ms",
            "Top-10 request"
        )
    ])+
    "</div>"
)


view24=widgets.ToggleButtons(
    options=[
        ("OVERVIEW","overview"),
        ("FORECAST","forecast"),
        ("RECOMMENDATIONS","recommend"),
        ("TESTS","tests"),
        ("METADATA","metadata"),
        ("METHOD","method")
    ],
    value="overview",
    layout=widgets.Layout(
        width="100%"
    ),
    style={
        "button_width":"16%",
        "font_weight":"700"
    }
)

view24.add_class(
    "btn24"
)

out24=widgets.Output()
out24.add_class(
    "out24"
)


def update24(*_):

    with out24:
        clear_output(
            wait=True
        )

        if view24.value=="overview":

            display(
                HTML(f"""
                <div class="grid24">

                <div class="box24">
                <b>FORECAST INFERENCE</b>
                Example family:
                {test_family24}<br><br>

                Model:
                {forecast_example24[
                    "selected_model"
                ].iloc[0]}<br>

                Horizon:
                {len(forecast_example24)}
                days
                </div>

                <div class="box24">
                <b>RECOMMENDATION INFERENCE</b>
                Query:
                {rec_query24}<br><br>

                Results returned:
                {len(recommendation_example24)}<br>

                Weight set:
                {metadata24[
                    "recommender_weight_set"
                ]}
                </div>

                <div class="box24">
                <b>PRODUCTION DESIGN</b>
                Models are loaded from serialized
                artifacts. Training code is not
                executed during inference.
                </div>

                </div><br>

                <div class="note24">
                This module is now the reusable
                application-service layer that later
                Flask endpoints can call. The web API
                will not contain model-training logic.
                </div>
                """)
            )


        elif view24.value=="forecast":

            fig=go.Figure()

            fig.add_trace(
                go.Scatter(
                    x=forecast_example24[
                        "forecast_day"
                    ],
                    y=forecast_example24[
                        "forecast"
                    ],
                    mode="lines+markers",
                    name="Forecast"
                )
            )

            fig.update_layout(
                height=380,
                title=(
                    f"Production Forecast — "
                    f"{test_family24}"
                ),
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",
                font=dict(
                    family="Arial",
                    color="#22223b"
                )
            )

            display(fig)

            table24(
                forecast_example24
            )


        elif view24.value=="recommend":

            display(
                HTML(f"""
                <div class="note24">
                <b>Production query:</b>
                {rec_query24}<br>
                <b>Country:</b>
                {rec_country24} &nbsp; • &nbsp;
                <b>Pay:</b>
                {rec_pay24} &nbsp; • &nbsp;
                <b>Seniority:</b>
                {rec_seniority24}
                </div><br>
                """)
            )

            show_cols24=[
                c
                for c in [
                    "title",
                    "country_clean",
                    "job_family",
                    "pay_type",
                    "Match_%",
                    "Why_Matched"
                ]
                if c in
                recommendation_example24.columns
            ]

            table24(
                recommendation_example24[
                    show_cols24
                ]
            )


        elif view24.value=="tests":

            table24(
                tests24
            )


        elif view24.value=="metadata":

            metadata_table24=pd.DataFrame(
                {
                    "Field":
                        list(
                            metadata24.keys()
                        ),
                    "Value":
                        [
                            str(v)
                            for v in
                            metadata24.values()
                        ]
                }
            )

            table24(
                metadata_table24
            )


        else:

            display(
                HTML("""
                <div class="grid24">

                <div class="box24">
                <b>ARTIFACT LOADING</b>
                The forecast pickle, TF-IDF
                vectorizer, sparse matrix,
                catalogue and selected weight
                configuration are loaded directly
                from the models directory.
                </div>

                <div class="box24">
                <b>FORECAST SERVING</b>
                Each job family uses the model
                selected during chronological
                holdout validation: Trend +
                Weekday, Lag-7 or 7-Day Mean.
                </div>

                <div class="box24">
                <b>RECOMMENDER SERVING</b>
                TF-IDF cosine similarity is combined
                with only the active preference
                signals, with weights normalized
                across those active signals.
                </div>

                </div><br>

                <div class="note24">
                Separation of training and inference
                prevents accidental model refitting
                during application startup and makes
                later API deployment reproducible.
                </div>
                """)
            )

view24.observe(update24,names="value")

update24()

dashboard_24=widgets.VBox([
    header24,
    kpi24,
    view24,
    out24
])

dashboard_24.add_class("d24")

display(dashboard_24)
```

```python
required25=[
    "forecast_demand",
    "production_recommend24",
    "families24",
    "metadata24",
    "test_family24",
    "rec_query24"
]

missing25=[
    x for x in required25
    if x not in globals()
]

if missing25:
    raise NameError(
        f"Run Section 24 first. Missing: {missing25}"
    )


def success_test25(name,func,validator=None):
    try:
        result=func()

        valid=(
            True
            if validator is None
            else bool(validator(result))
        )

        return {
            "Test":name,
            "Expected":"Successful execution",
            "Observed":"Valid output" if valid else "Unexpected output",
            "Status":"PASS" if valid else "REVIEW"
        }

    except Exception as e:
        return {
            "Test":name,
            "Expected":"Successful execution",
            "Observed":f"{type(e).__name__}: {str(e)[:120]}",
            "Status":"REVIEW"
        }


def error_test25(name,func,error_type=ValueError):
    try:
        func()

        return {
            "Test":name,
            "Expected":f"{error_type.__name__} rejected",
            "Observed":"No error raised",
            "Status":"REVIEW"
        }

    except error_type as e:
        return {
            "Test":name,
            "Expected":f"{error_type.__name__} rejected",
            "Observed":f"{type(e).__name__}: {str(e)[:120]}",
            "Status":"PASS"
        }

    except Exception as e:
        return {
            "Test":name,
            "Expected":f"{error_type.__name__} rejected",
            "Observed":f"Unexpected {type(e).__name__}",
            "Status":"REVIEW"
        }


forecast_reference25=forecast_demand(
    test_family24
)


forecast_repeat25=forecast_demand(
    test_family24
)


forecast_tests25=[]

forecast_tests25.append(
    success_test25(
        "Valid job-family forecast",
        lambda:forecast_demand(
            test_family24
        ),
        lambda x:
            isinstance(x,pd.DataFrame)
            and len(x)>0
    )
)

forecast_tests25.append(
    success_test25(
        "Forecast horizon contract",
        lambda:forecast_demand(
            test_family24
        ),
        lambda x:
            len(x)==int(
                metadata24[
                    "forecast_horizon_days"
                ]
            )
    )
)

forecast_tests25.append(
    success_test25(
        "Forecast schema contract",
        lambda:forecast_demand(
            test_family24
        ),
        lambda x:{
            "job_family",
            "forecast_day",
            "selected_model",
            "forecast"
        }.issubset(x.columns)
    )
)

forecast_tests25.append(
    success_test25(
        "Non-negative forecasts",
        lambda:forecast_demand(
            test_family24
        ),
        lambda x:
            (
                x["forecast"]>=0
            ).all()
    )
)

forecast_tests25.append(
    success_test25(
        "Deterministic forecast output",
        lambda:forecast_demand(
            test_family24
        ),
        lambda x:
            np.allclose(
                x["forecast"].to_numpy(),
                forecast_reference25[
                    "forecast"
                ].to_numpy()
            )
    )
)

forecast_tests25.append(
    error_test25(
        "Unknown job family rejected",
        lambda:forecast_demand(
            "__UNKNOWN_JOB_FAMILY__"
        )
    )
)

forecast_tests25.append(
    error_test25(
        "Blank job family rejected",
        lambda:forecast_demand("")
    )
)


forecast_tests25=pd.DataFrame(
    forecast_tests25
)


recommendation_tests25=[]


recommendation_tests25.append(
    success_test25(
        "Valid recommendation request",
        lambda:production_recommend24(
            query=rec_query24,
            n=10
        ),
        lambda x:
            isinstance(x,pd.DataFrame)
            and 0<len(x)<=10
    )
)


recommendation_tests25.append(
    success_test25(
        "Recommendation schema contract",
        lambda:production_recommend24(
            query=rec_query24,
            n=10
        ),
        lambda x:{
            "title",
            "job_family",
            "Match_Score",
            "Match_%",
            "Why_Matched"
        }.issubset(x.columns)
    )
)


recommendation_tests25.append(
    success_test25(
        "Match scores bounded 0–1",
        lambda:production_recommend24(
            query=rec_query24,
            n=10
        ),
        lambda x:
            x["Match_Score"]
            .between(0,1)
            .all()
    )
)


recommendation_tests25.append(
    success_test25(
        "Recommendations ranked descending",
        lambda:production_recommend24(
            query=rec_query24,
            n=10
        ),
        lambda x:
            x["Match_Score"]
            .is_monotonic_decreasing
    )
)


rec_reference25=production_recommend24(
    query=rec_query24,
    n=10
)

rec_repeat25=production_recommend24(
    query=rec_query24,
    n=10
)


def reproducible_recommendation25(x):

    if len(x)!=len(rec_reference25):
        return False

    if "job_id" in x.columns:
        same_items=(
            x["job_id"].tolist()
            ==
            rec_reference25[
                "job_id"
            ].tolist()
        )
    else:
        same_items=(
            x["title"].tolist()
            ==
            rec_reference25[
                "title"
            ].tolist()
        )

    same_scores=np.allclose(
        x["Match_Score"].to_numpy(),
        rec_reference25[
            "Match_Score"
        ].to_numpy()
    )

    return (
        same_items
        and same_scores
    )


recommendation_tests25.append(
    success_test25(
        "Deterministic recommendation ranking",
        lambda:production_recommend24(
            query=rec_query24,
            n=10
        ),
        reproducible_recommendation25
    )
)


recommendation_tests25.append(
    error_test25(
        "Blank query rejected",
        lambda:production_recommend24(
            query=""
        )
    )
)


recommendation_tests25.append(
    error_test25(
        "Single-character query rejected",
        lambda:production_recommend24(
            query="a"
        )
    )
)


recommendation_tests25.append(
    error_test25(
        "Invalid mode rejected",
        lambda:production_recommend24(
            query=rec_query24,
            mode="invalid-mode"
        )
    )
)


recommendation_tests25.append(
    success_test25(
        "Top-K capped at 50",
        lambda:production_recommend24(
            query=rec_query24,
            n=500
        ),
        lambda x:
            len(x)<=50
    )
)


recommendation_tests25.append(
    success_test25(
        "Zero Top-K safely normalized",
        lambda:production_recommend24(
            query=rec_query24,
            n=0
        ),
        lambda x:
            len(x)<=1
    )
)


recommendation_tests25.append(
    success_test25(
        "Unseen text handled safely",
        lambda:production_recommend24(
            query="zzzzzzunseenphrasezzzzzz",
            n=10
        ),
        lambda x:
            isinstance(x,pd.DataFrame)
    )
)


if len(rec_reference25):

    first25=rec_reference25.iloc[0]

    strict_country25=(
        first25["country_clean"]
        if (
            "country_clean"
            in first25.index
            and pd.notna(
                first25["country_clean"]
            )
        )
        else "Any"
    )

    strict_pay25=(
        first25["pay_type"]
        if (
            "pay_type"
            in first25.index
            and pd.notna(
                first25["pay_type"]
            )
        )
        else "Any"
    )

    strict_seniority25=(
        first25["seniority_level"]
        if (
            "seniority_level"
            in first25.index
            and pd.notna(
                first25["seniority_level"]
            )
        )
        else "Any"
    )


    def strict_validator25(x):

        checks=[]

        if strict_country25!="Any":
            checks.append(
                x["country_clean"]
                .eq(strict_country25)
                .all()
            )

        if strict_pay25!="Any":
            checks.append(
                x["pay_type"]
                .eq(strict_pay25)
                .all()
            )

        if strict_seniority25!="Any":
            checks.append(
                x["seniority_level"]
                .eq(strict_seniority25)
                .all()
            )

        return (
            all(checks)
            if checks
            else True
        )


    recommendation_tests25.append(
        success_test25(
            "Strict preference filtering",
            lambda:production_recommend24(
                query=rec_query24,
                country=strict_country25,
                pay_type=strict_pay25,
                seniority=strict_seniority25,
                n=10,
                mode="strict"
            ),
            strict_validator25
        )
    )


recommendation_tests25=pd.DataFrame(
    recommendation_tests25
)


def benchmark25(func,runs):

    times=[]

    for _ in range(runs):

        start=time.perf_counter()

        func()

        times.append(
            (
                time.perf_counter()
                - start
            )*1000
        )

    return times


forecast_times25=benchmark25(
    lambda:forecast_demand(
        test_family24
    ),
    5
)


recommendation_times25=benchmark25(
    lambda:production_recommend24(
        query=rec_query24,
        n=10
    ),
    3
)


performance25=pd.DataFrame({
    "Component":[
        "Forecast Inference",
        "Recommendation Inference"
    ],

    "Runs":[
        len(forecast_times25),
        len(recommendation_times25)
    ],

    "Median_ms":[
        np.median(
            forecast_times25
        ),
        np.median(
            recommendation_times25
        )
    ],

    "Mean_ms":[
        np.mean(
            forecast_times25
        ),
        np.mean(
            recommendation_times25
        )
    ],

    "Min_ms":[
        np.min(
            forecast_times25
        ),
        np.min(
            recommendation_times25
        )
    ],

    "Max_ms":[
        np.max(
            forecast_times25
        ),
        np.max(
            recommendation_times25
        )
    ]
})


all_tests25=pd.concat(
    [
        forecast_tests25.assign(
            Component="Forecast"
        ),
        recommendation_tests25.assign(
            Component="Recommendation"
        )
    ],
    ignore_index=True
)


robustness_status25=(
    "PASS"
    if (
        all_tests25[
            "Status"
        ]=="PASS"
    ).all()
    else "REVIEW"
)


summary25=pd.DataFrame({
    "Component":[
        "Forecast",
        "Recommendation",
        "Overall"
    ],

    "Tests":[
        len(forecast_tests25),
        len(recommendation_tests25),
        len(all_tests25)
    ],

    "Passed":[
        int(
            (
                forecast_tests25[
                    "Status"
                ]=="PASS"
            ).sum()
        ),

        int(
            (
                recommendation_tests25[
                    "Status"
                ]=="PASS"
            ).sum()
        ),

        int(
            (
                all_tests25[
                    "Status"
                ]=="PASS"
            ).sum()
        )
    ]
})

summary25["Pass_Rate_%"]=(
    summary25["Passed"]
    /
    summary25["Tests"]
    *100
).round(1)


all_tests25.to_csv(
    EXPORT_DIR/
    "inference_edge_case_tests.csv",
    index=False
)

performance25.to_csv(
    EXPORT_DIR/
    "inference_latency_benchmark.csv",
    index=False
)

summary25.to_csv(
    EXPORT_DIR/
    "inference_test_summary.csv",
    index=False
)


def card25(a,b,c):
    return f"""
    <div class="c25">
    <small>{a}</small>
    <b>{b}</b>
    <span>{c}</span>
    </div>
    """


def table25(df):
    display(
        HTML(
            df.to_html(
                index=False,
                border=0,
                classes="t25",
                float_format=lambda x:
                f"{x:,.2f}"
            )
        )
    )


display(HTML("""
<style>
.d25{
border:1px solid #ded8d5;
border-radius:17px;
overflow:hidden;
background:#fff;
box-shadow:0 8px 24px #22223b17
}

.h25{
background:linear-gradient(
120deg,#22223b,#30324d 55%,#4a4e69
);
padding:24px 27px;
display:flex;
justify-content:space-between;
font-family:Arial
}

.h25 .k{
color:#c9ada7;
font-size:8px;
font-weight:700;
letter-spacing:1.6px
}

.h25 h2{
color:#fff;
margin:6px 0;
font-size:24px
}

.h25 p,.h25 .m{
color:#f2e9e4;
font-size:10px;
line-height:1.65
}

.cards25{
display:grid;
grid-template-columns:repeat(5,1fr);
border-bottom:1px solid #e8e2df
}

.c25{
padding:14px;
border-right:1px solid #ebe6e4;
font-family:Arial
}

.c25 small{
display:block;
color:#9a8c98;
font-size:7px;
font-weight:700
}

.c25 b{
display:block;
color:#22223b;
font-size:17px;
margin-top:5px
}

.c25 span{
display:block;
color:#4a4e69;
font-size:8px;
margin-top:4px
}

.btn25{
margin:8px 16px!important
}

.out25{
border-top:1px solid #eee9e7!important;
padding:18px 22px
}

.t25{
width:100%;
border-collapse:collapse;
font:10px Arial
}

.t25 th{
background:#22223b;
color:#fff;
padding:9px;
text-align:left
}

.t25 td{
padding:8px;
border-bottom:1px solid #eee9e7;
color:#22223b
}

.grid25{
display:grid;
grid-template-columns:repeat(3,1fr);
gap:12px
}

.box25{
background:#f8f6f5;
border-left:4px solid #4a4e69;
padding:14px;
border-radius:8px;
font:11px/1.65 Arial;
color:#22223b
}

.box25 b{
display:block;
color:#4a4e69;
font-size:9px;
margin-bottom:6px
}

.note25{
background:#f2e9e4;
padding:12px 14px;
border-radius:8px;
font:10px/1.65 Arial;
color:#4a4e69
}
</style>
"""))


header25=widgets.HTML(f"""
<div class="h25">

<div>

<div class="k">
SECTION 25 • ROBUSTNESS & PERFORMANCE
</div>

<h2>
Production Inference Stress Testing
</h2>

<p>
Edge Cases • Input Validation • Reproducibility •
Ranking Integrity • Runtime Benchmarking
</p>

</div>

<div class="m">
FINAL STATUS<br>
<b>{robustness_status25}</b><br><br>

TOTAL TESTS<br>
<b>{len(all_tests25)}</b>
</div>

</div>
""")


kpi25=widgets.HTML(
    '<div class="cards25">'+
    "".join([
        card25(
            "STATUS",
            robustness_status25,
            "Inference testing"
        ),

        card25(
            "TESTS PASSED",
            f"{(all_tests25['Status']=='PASS').sum()}/{len(all_tests25)}",
            "Functional checks"
        ),

        card25(
            "FORECAST",
            f"{(forecast_tests25['Status']=='PASS').sum()}/{len(forecast_tests25)}",
            "Tests passed"
        ),

        card25(
            "RECOMMENDER",
            f"{(recommendation_tests25['Status']=='PASS').sum()}/{len(recommendation_tests25)}",
            "Tests passed"
        ),

        card25(
            "REC MEDIAN",
            f"{np.median(recommendation_times25):.1f} ms",
            "Observed latency"
        )
    ])+
    "</div>"
)


view25=widgets.ToggleButtons(
    options=[
        ("OVERVIEW","overview"),
        ("FORECAST TESTS","forecast"),
        ("REC TESTS","recommend"),
        ("PERFORMANCE","performance"),
        ("METHOD","method")
    ],

    value="overview",

    layout=widgets.Layout(
        width="100%"
    ),

    style={
        "button_width":"20%",
        "font_weight":"700"
    }
)

view25.add_class(
    "btn25"
)


out25=widgets.Output()

out25.add_class(
    "out25"
)


def update25(*_):

    with out25:

        clear_output(
            wait=True
        )

        if view25.value=="overview":

            display(
                HTML(f"""
                <div class="grid25">

                <div class="box25">
                <b>FUNCTIONAL VALIDATION</b>
                {(all_tests25['Status']=='PASS').sum()}
                of {len(all_tests25)}
                inference tests passed.
                </div>

                <div class="box25">
                <b>REPRODUCIBILITY</b>
                Forecast predictions and recommendation
                rankings are repeated using identical
                inputs to verify deterministic output.
                </div>

                <div class="box25">
                <b>INPUT SAFETY</b>
                Blank queries, invalid modes and unknown
                forecast families are explicitly tested
                rather than silently accepted.
                </div>

                </div><br>

                <div class="note25">
                Runtime figures are observed notebook
                benchmarks rather than production SLAs.
                Final deployment performance will depend
                on infrastructure, concurrency and hosting.
                </div>
                """)
            )


        elif view25.value=="forecast":

            table25(
                forecast_tests25
            )


        elif view25.value=="recommend":

            table25(
                recommendation_tests25
            )


        elif view25.value=="performance":

            fig=go.Figure()

            fig.add_trace(
                go.Bar(
                    x=performance25[
                        "Component"
                    ],
                    y=performance25[
                        "Median_ms"
                    ],
                    name="Median"
                )
            )

            fig.add_trace(
                go.Bar(
                    x=performance25[
                        "Component"
                    ],
                    y=performance25[
                        "Max_ms"
                    ],
                    name="Maximum"
                )
            )

            fig.update_layout(
                barmode="group",
                height=380,
                title="Observed Inference Runtime",
                yaxis_title="Milliseconds",
                paper_bgcolor="#fff",
                plot_bgcolor="#fff",

                font=dict(
                    family="Arial",
                    color="#22223b"
                ),

                legend=dict(
                    orientation="h",
                    y=1.12
                )
            )

            display(fig)

            table25(
                performance25
            )


        else:

            display(
                HTML("""
                <div class="grid25">

                <div class="box25">
                <b>FORECAST EDGE CASES</b>
                Valid families, unknown families,
                blank inputs, output horizon,
                schema integrity, non-negative
                predictions and deterministic
                repetition are checked.
                </div>

                <div class="box25">
                <b>RECOMMENDER EDGE CASES</b>
                Query length, invalid serving mode,
                Top-K boundaries, unseen text,
                score limits, ranking order and
                strict preference filtering are tested.
                </div>

                <div class="box25">
                <b>PERFORMANCE</b>
                Multiple inference calls measure
                mean, median, minimum and maximum
                runtime without retraining the models.
                </div>

                </div><br>

                <div class="note25">
                Performance observations are not treated
                as universal pass/fail thresholds because
                notebook hardware differs from deployed
                application infrastructure.
                </div>
                """)
            )

view25.observe(
    update25,
    names="value"
)

update25()

dashboard_25=widgets.VBox([
    header25,
    kpi25,
    view25,
    out25
])

dashboard_25.add_class("d25")

display(dashboard_25)
```

```python
import shutil
required251=[
    "forecast_demand",
    "production_recommend24",
    "selected_forecast_daily18",
    "model_registry18",
    "model_summary18",
    "selected_forecast_summary18",
    "recommend_jobs",
    "weight_summary19",
    "evaluation19",
    "BEST_WEIGHTS19",
    "best_weight_name19",
    "MODEL_VERSION",
    "RECOMMENDER_VERSION",
    "model_registry_master21",
    "all_tests25",
    "performance25"
]

missing251=[
    x for x in required251
    if x not in globals()
]

if missing251:
    raise NameError(
        f"Run Sections 18–25 first. Missing: {missing251}"
    )


MODEL_DIR=Path("models")
REC_DIR251=Path("models/recommender_assets")

SNAPSHOT_ID251=(
    f"{MODEL_VERSION}__{RECOMMENDER_VERSION}"
)

SNAPSHOT_DIR251=(
    Path("predeployment_snapshot")
    /SNAPSHOT_ID251
)

SNAPSHOT_MODELS251=(
    SNAPSHOT_DIR251/"models"
)

SNAPSHOT_RESULTS251=(
    SNAPSHOT_DIR251/"results"
)

SNAPSHOT_REPORTS251=(
    SNAPSHOT_DIR251/"reports"
)

SNAPSHOT_CODE251=(
    SNAPSHOT_DIR251/"code"
)

for p in [
    SNAPSHOT_MODELS251,
    SNAPSHOT_RESULTS251,
    SNAPSHOT_REPORTS251,
    SNAPSHOT_CODE251
]:
    p.mkdir(
        parents=True,
        exist_ok=True
    )


# ---------------------------------------------------------
# 1. FORECAST PARITY
# Notebook validation output vs production inference
# ---------------------------------------------------------

forecast_parity_rows251=[]


for family in sorted(
    model_registry18[
        "job_family"
    ].dropna().unique()
):

    reference251=(
        selected_forecast_daily18[
            selected_forecast_daily18[
                "job_family"
            ].eq(family)
        ]
        .copy()
        .sort_values(
            "forecast_day"
        )
        .reset_index(drop=True)
    )

    production251=(
        forecast_demand(
            family
        )
        .copy()
        .sort_values(
            "forecast_day"
        )
        .reset_index(drop=True)
    )


    reference251[
        "forecast_day"
    ]=pd.to_datetime(
        reference251[
            "forecast_day"
        ],
        utc=True
    )

    production251[
        "forecast_day"
    ]=pd.to_datetime(
        production251[
            "forecast_day"
        ],
        utc=True
    )


    row_count_match251=(
        len(reference251)
        ==
        len(production251)
    )


    date_match251=(
        row_count_match251
        and
        reference251[
            "forecast_day"
        ].tolist()
        ==
        production251[
            "forecast_day"
        ].tolist()
    )


    expected_model251=(
        reference251[
            "Selected_Model"
        ].iloc[0]
        if len(reference251)
        else None
    )

    actual_model251=(
        production251[
            "selected_model"
        ].iloc[0]
        if len(production251)
        else None
    )

    model_match251=(
        expected_model251
        ==
        actual_model251
    )


    if row_count_match251:

        ref_values251=(
            reference251[
                "Forecast"
            ]
            .astype(float)
            .round(2)
            .to_numpy()
        )

        prod_values251=(
            production251[
                "forecast"
            ]
            .astype(float)
            .round(2)
            .to_numpy()
        )

        value_match251=np.allclose(
            ref_values251,
            prod_values251,
            atol=0.001,
            equal_nan=True
        )

        max_diff251=float(
            np.max(
                np.abs(
                    ref_values251
                    -
                    prod_values251
                )
            )
        )

    else:
        value_match251=False
        max_diff251=np.nan


    forecast_parity_rows251.append({
        "Job_Family":family,
        "Validated_Model":
            expected_model251,
        "Production_Model":
            actual_model251,
        "Rows_Expected":
            len(reference251),
        "Rows_Produced":
            len(production251),
        "Date_Match":
            date_match251,
        "Model_Match":
            model_match251,
        "Value_Match":
            value_match251,
        "Max_Absolute_Difference":
            max_diff251,
        "Status":
            "PASS"
            if (
                row_count_match251
                and date_match251
                and model_match251
                and value_match251
            )
            else "REVIEW"
    })


forecast_parity251=pd.DataFrame(
    forecast_parity_rows251
)


forecast_parity_status251=(
    "PASS"
    if (
        forecast_parity251[
            "Status"
        ]=="PASS"
    ).all()
    else "REVIEW"
)


# ---------------------------------------------------------
# 2. RECOMMENDER PARITY
# Notebook engine vs serialized production engine
# ---------------------------------------------------------

query251=(
    rec_query24
    if "rec_query24" in globals()
    else "data analyst"
)

country251=(
    rec_country24
    if "rec_country24" in globals()
    else "Any"
)

pay251=(
    rec_pay24
    if "rec_pay24" in globals()
    else "Any"
)

seniority251=(
    rec_seniority24
    if "rec_seniority24" in globals()
    else "Any"
)


notebook_rec251=recommend_jobs(
    query=query251,
    country=country251,
    pay=pay251,
    seniority=seniority251,
    n=10,
    mode="boost"
).reset_index(drop=True)


production_rec251=production_recommend24(
    query=query251,
    country=country251,
    pay_type=pay251,
    seniority=seniority251,
    n=10,
    mode="boost"
).reset_index(drop=True)


rec_count_match251=(
    len(notebook_rec251)
    ==
    len(production_rec251)
)


if (
    rec_count_match251
    and len(notebook_rec251)
):

    if (
        "job_id"
        in notebook_rec251.columns
        and
        "job_id"
        in production_rec251.columns
    ):

        rec_rank_match251=(
            notebook_rec251[
                "job_id"
            ].astype(str).tolist()
            ==
            production_rec251[
                "job_id"
            ].astype(str).tolist()
        )

    else:

        rec_rank_match251=(
            notebook_rec251[
                "title"
            ].astype(str).tolist()
            ==
            production_rec251[
                "title"
            ].astype(str).tolist()
        )


    rec_score_match251=np.allclose(
        notebook_rec251[
            "Match_Score"
        ].astype(float).to_numpy(),

        production_rec251[
            "Match_Score"
        ].astype(float).to_numpy(),

        atol=1e-8,
        equal_nan=True
    )


    rec_max_diff251=float(
        np.max(
            np.abs(
                notebook_rec251[
                    "Match_Score"
                ].astype(float).to_numpy()
                -
                production_rec251[
                    "Match_Score"
                ].astype(float).to_numpy()
            )
        )
    )

else:

    rec_rank_match251=(
        len(notebook_rec251)==0
        and
        len(production_rec251)==0
    )

    rec_score_match251=(
        rec_rank_match251
    )

    rec_max_diff251=(
        0.0
        if rec_rank_match251
        else np.nan
    )


recommender_parity251=pd.DataFrame({
    "Check":[
        "Result count",
        "Ranking order",
        "Match scores",
        "Selected weight set"
    ],

    "Expected":[
        len(notebook_rec251),
        "Notebook ranking",
        "Notebook scores",
        best_weight_name19
    ],

    "Observed":[
        len(production_rec251),
        (
            "Identical"
            if rec_rank_match251
            else "Different"
        ),
        (
            f"Max difference "
            f"{rec_max_diff251:.10f}"
            if pd.notna(
                rec_max_diff251
            )
            else "Unavailable"
        ),
        best_weight_name19
    ],

    "Status":[
        (
            "PASS"
            if rec_count_match251
            else "REVIEW"
        ),

        (
            "PASS"
            if rec_rank_match251
            else "REVIEW"
        ),

        (
            "PASS"
            if rec_score_match251
            else "REVIEW"
        ),

        "PASS"
    ]
})


recommender_parity_status251=(
    "PASS"
    if (
        recommender_parity251[
            "Status"
        ]=="PASS"
    ).all()
    else "REVIEW"
)


# ---------------------------------------------------------
# 3. SERIALIZED ARTIFACT VERSION VALIDATION
# ---------------------------------------------------------

forecast_model_path251=(
    MODEL_DIR/f"{MODEL_VERSION}.pkl"
)

forecast_metadata_path251=(
    MODEL_DIR/
    f"{MODEL_VERSION}_metadata.json"
)

weights_path251=(
    REC_DIR251/"weights.json"
)

rec_metadata_path251=(
    REC_DIR251/"metadata.json"
)


with open(
    forecast_model_path251,
    "rb"
) as f:

    saved_forecast251=pickle.load(f)


with open(
    weights_path251,
    "r",
    encoding="utf-8"
) as f:

    saved_weights251=json.load(f)


with open(
    rec_metadata_path251,
    "r",
    encoding="utf-8"
) as f:

    saved_rec_metadata251=json.load(f)


artifact_version251=pd.DataFrame({
    "Artifact":[
        "Forecast Model",
        "Recommendation Weights",
        "Recommendation Metadata"
    ],

    "Expected_Version":[
        MODEL_VERSION,
        RECOMMENDER_VERSION,
        RECOMMENDER_VERSION
    ],

    "Saved_Version":[
        saved_forecast251.get(
            "version"
        ),
        saved_weights251.get(
            "version"
        ),
        saved_rec_metadata251.get(
            "version"
        )
    ]
})


artifact_version251[
    "Status"
]=np.where(
    artifact_version251[
        "Expected_Version"
    ].eq(
        artifact_version251[
            "Saved_Version"
        ]
    ),
    "PASS",
    "REVIEW"
)


weight_match251=(
    saved_weights251.get(
        "selected_weight_set"
    )
    ==
    best_weight_name19
    and
    saved_weights251.get(
        "weights"
    )
    ==
    BEST_WEIGHTS19
)


# ---------------------------------------------------------
# 4. SQL CROSS-LAYER CONSISTENCY
# ---------------------------------------------------------

sql_available251=True
sql_error251=None


try:

    if (
        "sql_conn22"
        not in globals()
        or sql_conn22 is None
    ):

        if (
            "connection_string22"
            in globals()
            and
            "SQL_DATABASE"
            in globals()
        ):

            sql_conn22=pyodbc.connect(
                connection_string22(
                    SQL_DATABASE
                ),
                timeout=15
            )

        else:
            raise RuntimeError(
                "SQL connection configuration unavailable."
            )


    cur251=sql_conn22.cursor()


    forecast_sql_rows251=[
        tuple(r)
        for r in cur251.execute("""
            SELECT
                job_family,
                forecast_day,
                selected_model,
                forecast_value,
                model_version
            FROM dbo.forecast_predictions
            ORDER BY
                job_family,
                forecast_day
        """).fetchall()
    ]


    sql_forecasts251=pd.DataFrame(
        forecast_sql_rows251,
        columns=[
            "job_family",
            "forecast_day",
            "selected_model",
            "forecast_value",
            "model_version"
        ]
    )


    registry_sql_rows251=[
        tuple(r)
        for r in cur251.execute("""
            SELECT
                component,
                version,
                scope,
                method,
                primary_metric,
                primary_value,
                secondary_metric,
                secondary_value,
                status
            FROM dbo.model_registry
        """).fetchall()
    ]


    sql_registry251=pd.DataFrame(
        registry_sql_rows251,
        columns=[
            "component",
            "version",
            "scope",
            "method",
            "primary_metric",
            "primary_value",
            "secondary_metric",
            "secondary_value",
            "status"
        ]
    )


except Exception as e:

    sql_available251=False
    sql_error251=str(e)

    sql_forecasts251=pd.DataFrame()

    sql_registry251=pd.DataFrame()


if sql_available251:

    ref_sql251=(
        selected_forecast_daily18[
            [
                "job_family",
                "forecast_day",
                "Selected_Model",
                "Forecast"
            ]
        ]
        .copy()
    )


    ref_sql251[
        "forecast_day"
    ]=pd.to_datetime(
        ref_sql251[
            "forecast_day"
        ],
        utc=True
    )


    sql_forecasts251[
        "forecast_day"
    ]=pd.to_datetime(
        sql_forecasts251[
            "forecast_day"
        ],
        utc=True
    )


    sql_forecast_count_match251=(
        len(ref_sql251)
        ==
        len(sql_forecasts251)
    )


    merged_sql251=ref_sql251.merge(
        sql_forecasts251,
        on=[
            "job_family",
            "forecast_day"
        ],
        how="outer",
        indicator=True
    )


    sql_key_match251=(
        merged_sql251[
            "_merge"
        ].eq("both").all()
    )


    if (
        sql_key_match251
        and len(merged_sql251)
    ):

        sql_model_match251=(
            merged_sql251[
                "Selected_Model"
            ].astype(str)
            ==
            merged_sql251[
                "selected_model"
            ].astype(str)
        ).all()


        sql_value_match251=np.allclose(
            merged_sql251[
                "Forecast"
            ].astype(float),
            merged_sql251[
                "forecast_value"
            ].astype(float),
            atol=1e-6,
            equal_nan=True
        )


    else:

        sql_model_match251=False
        sql_value_match251=False


    sql_forecast_version251=(
        sql_forecasts251[
            "model_version"
        ]
        .astype(str)
        .eq(
            str(MODEL_VERSION)
        )
        .all()
        if len(
            sql_forecasts251
        )
        else False
    )


    sql_registry_versions251=set(
        sql_registry251[
            "version"
        ].dropna().astype(str)
    )


    sql_registry_version_match251=(
        str(MODEL_VERSION)
        in sql_registry_versions251
        and
        str(RECOMMENDER_VERSION)
        in sql_registry_versions251
    )


else:

    sql_forecast_count_match251=False
    sql_key_match251=False
    sql_model_match251=False
    sql_value_match251=False
    sql_forecast_version251=False
    sql_registry_version_match251=False


sql_consistency251=pd.DataFrame({
    "Check":[
        "SQL connection",
        "Forecast row count",
        "Forecast keys",
        "Selected model parity",
        "Forecast value parity",
        "Forecast model version",
        "Registry model versions"
    ],

    "Status":[
        (
            "PASS"
            if sql_available251
            else "REVIEW"
        ),

        (
            "PASS"
            if sql_forecast_count_match251
            else "REVIEW"
        ),

        (
            "PASS"
            if sql_key_match251
            else "REVIEW"
        ),

        (
            "PASS"
            if sql_model_match251
            else "REVIEW"
        ),

        (
            "PASS"
            if sql_value_match251
            else "REVIEW"
        ),

        (
            "PASS"
            if sql_forecast_version251
            else "REVIEW"
        ),

        (
            "PASS"
            if sql_registry_version_match251
            else "REVIEW"
        )
    ]
})


sql_consistency_status251=(
    "PASS"
    if (
        sql_consistency251[
            "Status"
        ]=="PASS"
    ).all()
    else "REVIEW"
)


# ---------------------------------------------------------
# 5. SAVE FINAL MODEL ARTIFACTS
# ---------------------------------------------------------

model_files251=[
    forecast_model_path251,
    forecast_metadata_path251,
    REC_DIR251/"recommendation_catalog.csv",
    REC_DIR251/"tfidf_vectorizer.pkl",
    REC_DIR251/"tfidf_matrix.npz",
    REC_DIR251/"weights.json",
    REC_DIR251/"metadata.json"
]


for source_path251 in model_files251:

    if not source_path251.exists():
        raise FileNotFoundError(
            f"Required model artifact missing: "
            f"{source_path251}"
        )

    shutil.copy2(
        source_path251,
        SNAPSHOT_MODELS251/
        source_path251.name
    )


inference_source251=(
    Path("src")/"inference.py"
)

if inference_source251.exists():

    shutil.copy2(
        inference_source251,
        SNAPSHOT_CODE251/
        "inference.py"
    )


# ---------------------------------------------------------
# 6. SAVE ALL IMPORTANT MODEL RESULTS
# ---------------------------------------------------------

model_registry18.to_csv(
    SNAPSHOT_RESULTS251/
    "forecast_selected_models.csv",
    index=False
)

model_summary18.to_csv(
    SNAPSHOT_RESULTS251/
    "forecast_model_comparison.csv",
    index=False
)

selected_forecast_daily18.to_csv(
    SNAPSHOT_RESULTS251/
    "selected_forecast_daily.csv",
    index=False
)

selected_forecast_summary18.to_csv(
    SNAPSHOT_RESULTS251/
    "selected_forecast_summary.csv",
    index=False
)

weight_summary19.to_csv(
    SNAPSHOT_RESULTS251/
    "recommender_weight_comparison.csv",
    index=False
)

evaluation19.to_csv(
    SNAPSHOT_RESULTS251/
    "recommender_evaluation.csv",
    index=False
)

model_registry_master21.to_csv(
    SNAPSHOT_RESULTS251/
    "master_model_registry.csv",
    index=False
)

all_tests25.to_csv(
    SNAPSHOT_RESULTS251/
    "inference_robustness_tests.csv",
    index=False
)

performance25.to_csv(
    SNAPSHOT_RESULTS251/
    "inference_performance.csv",
    index=False
)

forecast_parity251.to_csv(
    SNAPSHOT_RESULTS251/
    "forecast_parity.csv",
    index=False
)

recommender_parity251.to_csv(
    SNAPSHOT_RESULTS251/
    "recommender_parity.csv",
    index=False
)

artifact_version251.to_csv(
    SNAPSHOT_RESULTS251/
    "artifact_version_check.csv",
    index=False
)

sql_consistency251.to_csv(
    SNAPSHOT_RESULTS251/
    "sql_consistency_check.csv",
    index=False
)


if len(sql_forecasts251):

    sql_forecasts251.to_csv(
        SNAPSHOT_RESULTS251/
        "sql_forecast_snapshot.csv",
        index=False
    )


if len(sql_registry251):

    sql_registry251.to_csv(
        SNAPSHOT_RESULTS251/
        "sql_model_registry_snapshot.csv",
        index=False
    )


# ---------------------------------------------------------
# 7. FINAL PRE-DEPLOYMENT GATE
# ---------------------------------------------------------

gate251=pd.DataFrame({
    "Control":[
        "Forecast notebook vs production parity",
        "Recommender notebook vs production parity",
        "Serialized artifact versions",
        "Selected recommender weights",
        "Inference robustness tests",
        "SQL forecast and registry consistency",
        "Model artifacts saved",
        "Model results saved"
    ],

    "Status":[
        forecast_parity_status251,
        recommender_parity_status251,

        (
            "PASS"
            if (
                artifact_version251[
                    "Status"
                ]=="PASS"
            ).all()
            else "REVIEW"
        ),

        (
            "PASS"
            if weight_match251
            else "REVIEW"
        ),

        (
            "PASS"
            if (
                all_tests25[
                    "Status"
                ]=="PASS"
            ).all()
            else "REVIEW"
        ),

        sql_consistency_status251,

        (
            "PASS"
            if all(
                (
                    SNAPSHOT_MODELS251/
                    p.name
                ).exists()
                for p
                in model_files251
            )
            else "REVIEW"
        ),

        "PASS"
    ]
})


READY_TO_PACKAGE251=(
    "PASS"
    if (
        gate251[
            "Status"
        ]=="PASS"
    ).all()
    else "REVIEW"
)


gate251.to_csv(
    SNAPSHOT_RESULTS251/
    "predeployment_gate.csv",
    index=False
)


# ---------------------------------------------------------
# 8. SHA-256 HASHES FOR FROZEN SNAPSHOT
# ---------------------------------------------------------

def sha256_file251(path):

    h=hashlib.sha256()

    with open(
        path,
        "rb"
    ) as f:

        for block in iter(
            lambda:f.read(
                1024*1024
            ),
            b""
        ):
            h.update(block)

    return h.hexdigest()


snapshot_files251=[
    p
    for p in
    SNAPSHOT_DIR251.rglob("*")
    if p.is_file()
]


manifest_rows251=[]


for p in sorted(
    snapshot_files251
):

    manifest_rows251.append({
        "File":
            str(
                p.relative_to(
                    SNAPSHOT_DIR251
                )
            ),

        "Size_Bytes":
            p.stat().st_size,

        "SHA256":
            sha256_file251(p)
    })


snapshot_manifest251=pd.DataFrame(
    manifest_rows251
)


snapshot_manifest251.to_csv(
    SNAPSHOT_REPORTS251/
    "snapshot_file_manifest.csv",
    index=False
)


# Recalculate so manifest itself can also be included
snapshot_files251=[
    p
    for p in
    SNAPSHOT_DIR251.rglob("*")
    if p.is_file()
]


snapshot_metadata251={
    "snapshot_id":
        SNAPSHOT_ID251,

    "created_at_utc":
        str(
            pd.Timestamp.now(
                tz="UTC"
            )
        ),

    "forecast_model_version":
        MODEL_VERSION,

    "recommender_version":
        RECOMMENDER_VERSION,

    "selected_recommender_weight_set":
        best_weight_name19,

    "selected_recommender_weights":
        BEST_WEIGHTS19,

    "forecast_job_families":
        int(
            model_registry18[
                "job_family"
            ].nunique()
        ),

    "forecast_selection_rule":
        (
            "Lowest chronological holdout MAE; "
            "RMSE used as tie-breaker"
        ),

    "recommender_primary_metric":
        "NDCG@10",

    "mean_selected_forecast_mae":
        float(
            model_registry18[
                "MAE"
            ].mean()
        ),

    "mean_recommender_precision_at_10":
        float(
            evaluation19[
                "Precision@10"
            ].mean()
        ),

    "mean_recommender_ndcg_at_10":
        float(
            evaluation19[
                "NDCG@10"
            ].mean()
        ),

    "predeployment_status":
        READY_TO_PACKAGE251,

    "snapshot_directory":
        str(
            SNAPSHOT_DIR251.resolve()
        ),

    "saved_file_count":
        len(
            snapshot_files251
        )
}


with open(
    SNAPSHOT_REPORTS251/
    "approved_model_snapshot.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        snapshot_metadata251,
        f,
        indent=2
    )


# ---------------------------------------------------------
# 9. SAVE HUMAN-READABLE APPROVAL REPORT
# ---------------------------------------------------------

approval_report251=f"""
# Pre-Deployment Model Approval

Snapshot
--------
{SNAPSHOT_ID251}

Forecast Model Version
----------------------
{MODEL_VERSION}

Recommendation Version
----------------------
{RECOMMENDER_VERSION}

Selected Recommendation Weight Set
----------------------------------
{best_weight_name19}

Forecast Families
-----------------
{model_registry18['job_family'].nunique()}

Mean Selected Forecast MAE
--------------------------
{model_registry18['MAE'].mean():.4f}

Recommendation Precision@10
---------------------------
{evaluation19['Precision@10'].mean():.4f}

Recommendation NDCG@10
----------------------
{evaluation19['NDCG@10'].mean():.4f}

Forecast Parity
---------------
{forecast_parity_status251}

Recommendation Parity
---------------------
{recommender_parity_status251}

SQL Consistency
---------------
{sql_consistency_status251}

Final Pre-Deployment Gate
-------------------------
{READY_TO_PACKAGE251}

The snapshot contains the selected forecasting model,
recommender assets, model registry, model-comparison results,
evaluation metrics, selected forecasts, robustness tests,
production parity evidence, SQL consistency evidence and
SHA-256 integrity hashes.
"""


(
    SNAPSHOT_REPORTS251/
    "predeployment_approval.txt"
).write_text(
    approval_report251,
    encoding="utf-8"
)


# ---------------------------------------------------------
# 10. SNAPSHOT INVENTORY FOR DASHBOARD
# ---------------------------------------------------------

saved_inventory251=pd.DataFrame({
    "Category":[
        "Forecast model",
        "Forecast metadata",
        "Recommendation catalogue",
        "TF-IDF vectorizer",
        "TF-IDF matrix",
        "Recommendation weights",
        "Recommendation metadata",
        "Forecast model scores",
        "Recommendation evaluation",
        "Robustness tests",
        "Parity evidence",
        "SQL evidence",
        "Integrity manifest"
    ],

    "Saved":[
        (
            SNAPSHOT_MODELS251/
            forecast_model_path251.name
        ).exists(),

        (
            SNAPSHOT_MODELS251/
            forecast_metadata_path251.name
        ).exists(),

        (
            SNAPSHOT_MODELS251/
            "recommendation_catalog.csv"
        ).exists(),

        (
            SNAPSHOT_MODELS251/
            "tfidf_vectorizer.pkl"
        ).exists(),

        (
            SNAPSHOT_MODELS251/
            "tfidf_matrix.npz"
        ).exists(),

        (
            SNAPSHOT_MODELS251/
            "weights.json"
        ).exists(),

        (
            SNAPSHOT_MODELS251/
            "metadata.json"
        ).exists(),

        (
            SNAPSHOT_RESULTS251/
            "forecast_model_comparison.csv"
        ).exists(),

        (
            SNAPSHOT_RESULTS251/
            "recommender_evaluation.csv"
        ).exists(),

        (
            SNAPSHOT_RESULTS251/
            "inference_robustness_tests.csv"
        ).exists(),

        (
            SNAPSHOT_RESULTS251/
            "forecast_parity.csv"
        ).exists()
        and
        (
            SNAPSHOT_RESULTS251/
            "recommender_parity.csv"
        ).exists(),

        (
            SNAPSHOT_RESULTS251/
            "sql_consistency_check.csv"
        ).exists(),

        (
            SNAPSHOT_REPORTS251/
            "snapshot_file_manifest.csv"
        ).exists()
    ]
})


saved_inventory251[
    "Status"
]=np.where(
    saved_inventory251[
        "Saved"
    ],
    "PASS",
    "REVIEW"
)


# ---------------------------------------------------------
# 11. PROFESSIONAL DASHBOARD
# ---------------------------------------------------------

def card251(a,b,c):

    return f"""
    <div class="c251">
    <small>{a}</small>
    <b>{b}</b>
    <span>{c}</span>
    </div>
    """


def table251(df):

    display(
        HTML(
            df.to_html(
                index=False,
                border=0,
                classes="t251",
                float_format=
                    lambda x:
                    f"{x:,.6f}"
            )
        )
    )


display(HTML("""
<style>
.d251{
border:1px solid #ded8d5;
border-radius:17px;
overflow:hidden;
background:#fff;
box-shadow:0 8px 24px #22223b17
}

.h251{
background:linear-gradient(
120deg,#22223b,#30324d 55%,#4a4e69
);
padding:24px 27px;
display:flex;
justify-content:space-between;
font-family:Arial
}

.h251 .k{
color:#c9ada7;
font-size:8px;
font-weight:700;
letter-spacing:1.6px
}

.h251 h2{
color:#fff;
margin:6px 0;
font-size:24px
}

.h251 p,.h251 .m{
color:#f2e9e4;
font-size:10px;
line-height:1.65
}

.cards251{
display:grid;
grid-template-columns:repeat(5,1fr);
border-bottom:1px solid #e8e2df
}

.c251{
padding:14px;
border-right:1px solid #ebe6e4;
font-family:Arial
}

.c251 small{
display:block;
color:#9a8c98;
font-size:7px;
font-weight:700
}

.c251 b{
display:block;
color:#22223b;
font-size:16px;
margin-top:5px;
overflow-wrap:anywhere
}

.c251 span{
display:block;
color:#4a4e69;
font-size:8px;
margin-top:4px
}

.btn251{
margin:8px 16px!important
}

.out251{
border-top:1px solid #eee9e7!important;
padding:18px 22px
}

.t251{
width:100%;
border-collapse:collapse;
font:10px Arial
}

.t251 th{
background:#22223b;
color:#fff;
padding:9px;
text-align:left
}

.t251 td{
padding:8px;
border-bottom:1px solid #eee9e7;
color:#22223b
}

.grid251{
display:grid;
grid-template-columns:repeat(3,1fr);
gap:12px
}

.box251{
background:#f8f6f5;
border-left:4px solid #4a4e69;
padding:14px;
border-radius:8px;
font:11px/1.65 Arial;
color:#22223b
}

.box251 b{
display:block;
color:#4a4e69;
font-size:9px;
margin-bottom:6px
}

.note251{
background:#f2e9e4;
padding:12px 14px;
border-radius:8px;
font:10px/1.65 Arial;
color:#4a4e69
}
</style>
"""))


header251=widgets.HTML(f"""
<div class="h251">

<div>

<div class="k">
SECTION 25.1 • PRE-DEPLOYMENT GATE
</div>

<h2>
Model Parity & Approved Snapshot
</h2>

<p>
Notebook → Artifacts → Production Inference →
SQL Server → Frozen Deployment State
</p>

</div>

<div class="m">
PACKAGE STATUS<br>
<b>{READY_TO_PACKAGE251}</b><br><br>

SNAPSHOT<br>
<b>{SNAPSHOT_ID251}</b>
</div>

</div>
""")


kpi251=widgets.HTML(
    '<div class="cards251">'+
    "".join([
        card251(
            "READY TO PACKAGE",
            READY_TO_PACKAGE251,
            "Final gate"
        ),

        card251(
            "FORECAST PARITY",
            forecast_parity_status251,
            f"{len(forecast_parity251)} families"
        ),

        card251(
            "RECOMMENDER PARITY",
            recommender_parity_status251,
            "Ranking + scores"
        ),

        card251(
            "SQL SYNC",
            sql_consistency_status251,
            "Cross-layer"
        ),

        card251(
            "SNAPSHOT",
            f"{saved_inventory251['Saved'].sum()}/{len(saved_inventory251)}",
            "Evidence groups saved"
        )
    ])+
    "</div>"
)


view251=widgets.ToggleButtons(
    options=[
        (
            "OVERVIEW",
            "overview"
        ),
        (
            "FORECAST PARITY",
            "forecast"
        ),
        (
            "RECOMMENDER",
            "recommend"
        ),
        (
            "SQL SYNC",
            "sql"
        ),
        (
            "SAVED ASSETS",
            "saved"
        ),
        (
            "FINAL GATE",
            "gate"
        )
    ],

    value="overview",

    layout=widgets.Layout(
        width="100%"
    ),

    style={
        "button_width":"16%",
        "font_weight":"700"
    }
)

view251.add_class(
    "btn251"
)


out251=widgets.Output()

out251.add_class(
    "out251"
)


def update251(*_):

    with out251:

        clear_output(
            wait=True
        )


        if view251.value=="overview":

            display(
                HTML(f"""
                <div class="grid251">

                <div class="box251">
                <b>MODEL FREEZE</b>
                Forecast model:
                {MODEL_VERSION}<br>
                Recommender:
                {RECOMMENDER_VERSION}<br>
                Weight set:
                {best_weight_name19}
                </div>

                <div class="box251">
                <b>RESULTS PRESERVED</b>
                Forecast scores, selected models,
                forecasts, recommender metrics,
                robustness tests and parity
                results are saved with the model.
                </div>

                <div class="box251">
                <b>INTEGRITY</b>
                SHA-256 hashes are generated
                for the frozen snapshot so later
                file changes can be detected.
                </div>

                </div><br>

                <div class="note251">
                Approved snapshot directory:<br>
                <b>
                {SNAPSHOT_DIR251.resolve()}
                </b>
                </div>
                """)
            )


        elif view251.value=="forecast":

            table251(
                forecast_parity251
            )


        elif view251.value=="recommend":

            table251(
                recommender_parity251
            )

            display(
                HTML(f"""
                <br>
                <div class="note251">
                Selected recommender:
                <b>{best_weight_name19}</b><br>
                Precision@10:
                {evaluation19['Precision@10'].mean():.3f}<br>
                NDCG@10:
                {evaluation19['NDCG@10'].mean():.3f}
                </div>
                """)
            )


        elif view251.value=="sql":

            table251(
                sql_consistency251
            )

            if not sql_available251:

                display(
                    HTML(f"""
                    <div class="note251">
                    SQL diagnostic:
                    {sql_error251}
                    </div>
                    """)
                )


        elif view251.value=="saved":

            table251(
                saved_inventory251
            )

            display(
                HTML(f"""
                <br>
                <div class="note251">
                The saved snapshot contains both
                <b>model artifacts</b> and
                <b>model results/evidence</b>.
                It is therefore sufficient to
                identify exactly which model state
                was approved before deployment.
                </div>
                """)
            )


        else:

            table251(
                gate251
            )

            display(
                HTML(f"""
                <br>
                <div class="note251">
                Final pre-deployment decision:
                <b>{READY_TO_PACKAGE251}</b><br><br>

                Deployment packaging should proceed
                only when every control above is PASS.
                </div>
                """)
            )


view251.observe(
    update251,
    names="value"
)

update251()

dashboard_251=widgets.VBox([
    header251,
    kpi251,
    view251,
    out251
])

dashboard_251.add_class(
    "d251"
)

display(
    dashboard_251
)
```

```python
REPORT_DIR=Path("reports")
EXPORT_DIR=Path("exports")

REPORT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

EXPORT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

requirements_matrix252=pd.DataFrame([
    {
        "Requirement_ID":"TASK_01",
        "Requirement":"Job-title keywords vs offered compensation",
        "Implementation":"Keyword extraction, compensation association and statistical validation",
        "Notebook_Section":"Section 07 + Section 17",
        "Primary_Output":"Task 1 salary-keyword analysis",
        "Code_Proof":"Keyword compensation pipeline + statistical validation",
        "Export_Proof":"exports/task1_*",
        "Future_Streamlit_Proof":"Salary Intelligence",
        "Owner_Layer":"Analytics",
        "Current_Status":"COMPLETE"
    },
    {
        "Requirement_ID":"TASK_02",
        "Requirement":"Identify emerging job categories",
        "Implementation":"Posting-frequency growth, momentum and emerging-category analysis",
        "Notebook_Section":"Section 08",
        "Primary_Output":"Emerging category rankings and trend evidence",
        "Code_Proof":"Emerging category analytical pipeline",
        "Export_Proof":"exports/task2_*",
        "Future_Streamlit_Proof":"Emerging Jobs",
        "Owner_Layer":"Analytics",
        "Current_Status":"COMPLETE"
    },
    {
        "Requirement_ID":"TASK_03",
        "Requirement":"Predict future high-demand job roles",
        "Implementation":"Chronological forecasting with candidate-model comparison and family-level model selection",
        "Notebook_Section":"Sections 09, 18, 21, 23",
        "Primary_Output":"Selected demand forecast + validation metrics",
        "Code_Proof":"Trend + Weekday, Lag-7 and 7-Day Mean comparison",
        "Export_Proof":"exports/task3_* + model registry",
        "Future_Streamlit_Proof":"Demand Forecasting",
        "Owner_Layer":"Machine Learning",
        "Current_Status":"COMPLETE"
    },
    {
        "Requirement_ID":"TASK_04",
        "Requirement":"Compare hourly compensation across countries",
        "Implementation":"Country-level hourly-pay analysis with sample-size controls",
        "Notebook_Section":"Section 10",
        "Primary_Output":"Country hourly-rate comparison",
        "Code_Proof":"Country aggregation and geographic analysis",
        "Export_Proof":"exports/task4_*",
        "Future_Streamlit_Proof":"Geographic Intelligence",
        "Owner_Layer":"Analytics",
        "Current_Status":"COMPLETE"
    },
    {
        "Requirement_ID":"TASK_05",
        "Requirement":"Personalized job recommendation engine",
        "Implementation":"TF-IDF title similarity + preference compatibility + recency ranking",
        "Notebook_Section":"Sections 11, 19, 23–25.1",
        "Primary_Output":"Explainable Top-K job recommendations",
        "Code_Proof":"Validated recommendation engine and production inference",
        "Export_Proof":"exports/task5_* + models/recommender_assets/",
        "Future_Streamlit_Proof":"Job Recommender",
        "Owner_Layer":"Recommendation Engine",
        "Current_Status":"COMPLETE"
    },
    {
        "Requirement_ID":"TASK_06",
        "Requirement":"Track job-market dynamics over time",
        "Implementation":"Monthly market KPIs, demand changes and market movement analysis",
        "Notebook_Section":"Section 12",
        "Primary_Output":"Market-dynamics intelligence",
        "Code_Proof":"Monthly market dynamics pipeline",
        "Export_Proof":"exports/task6_*",
        "Future_Streamlit_Proof":"Market Dynamics",
        "Owner_Layer":"Analytics",
        "Current_Status":"COMPLETE"
    },
    {
        "Requirement_ID":"TASK_07",
        "Requirement":"Analyze remote-work landscape",
        "Implementation":"Proxy-based remote-work landscape analysis with documented dataset limitation",
        "Notebook_Section":"Section 13",
        "Primary_Output":"Remote-work analytical evidence",
        "Code_Proof":"Remote-work proxy analysis",
        "Export_Proof":"exports/task7_*",
        "Future_Streamlit_Proof":"Remote Work",
        "Owner_Layer":"Analytics",
        "Current_Status":"COMPLETE"
    },
    {
        "Requirement_ID":"TASK_08",
        "Requirement":"Future job-market trends and scenarios",
        "Implementation":"Base, High Growth, Slow Growth and Technology Shift scenario framework",
        "Notebook_Section":"Sections 14 and 20",
        "Primary_Output":"Future scenario forecasts and assumptions",
        "Code_Proof":"Scenario modeling pipeline",
        "Export_Proof":"exports/task8_*",
        "Future_Streamlit_Proof":"Future Trends",
        "Owner_Layer":"Forecasting / Analytics",
        "Current_Status":"COMPLETE"
    },

    {
        "Requirement_ID":"MODEL_GOVERNANCE",
        "Requirement":"Model comparison, explainability and registry",
        "Implementation":"Model comparison, selected-model registry, artifact validation and deployment readiness",
        "Notebook_Section":"Sections 18, 19, 21, 23",
        "Primary_Output":"Model registry + comparison + selected model evidence",
        "Code_Proof":"Forecast/recommender governance pipeline",
        "Export_Proof":"exports/model_* + models/",
        "Future_Streamlit_Proof":"Model Performance",
        "Owner_Layer":"Machine Learning",
        "Current_Status":"COMPLETE"
    },

    {
        "Requirement_ID":"SQL_LAYER",
        "Requirement":"Persistent SQL/data-access layer",
        "Implementation":"SQL Server relational schema, indexed tables, parameterized queries and data-access module",
        "Notebook_Section":"Section 22",
        "Primary_Output":"JobMarketDB + reusable data-access layer",
        "Code_Proof":"database/schema.sql + database/queries.sql + src/data/database.py",
        "Export_Proof":"exports/sql_*",
        "Future_Streamlit_Proof":"Indirect through Flask API",
        "Owner_Layer":"Data Engineering",
        "Current_Status":"COMPLETE"
    },

    {
        "Requirement_ID":"PRODUCTION_INFERENCE",
        "Requirement":"Reusable production inference",
        "Implementation":"Standalone forecast and recommendation inference without retraining",
        "Notebook_Section":"Section 24",
        "Primary_Output":"src/inference.py",
        "Code_Proof":"Serialized model loading + reusable inference functions",
        "Export_Proof":"exports/production_inference_*",
        "Future_Streamlit_Proof":"Consumed through Flask",
        "Owner_Layer":"Application Services",
        "Current_Status":"COMPLETE"
    },

    {
        "Requirement_ID":"INFERENCE_TESTING",
        "Requirement":"Inference robustness and edge-case validation",
        "Implementation":"Input validation, reproducibility, output-contract and latency testing",
        "Notebook_Section":"Section 25",
        "Primary_Output":"Inference robustness test suite",
        "Code_Proof":"Forecast and recommendation stress tests",
        "Export_Proof":"exports/inference_*",
        "Future_Streamlit_Proof":"Reliability evidence",
        "Owner_Layer":"Testing",
        "Current_Status":"COMPLETE"
    },

    {
        "Requirement_ID":"MODEL_PARITY",
        "Requirement":"Pre-deployment model parity and approved snapshot",
        "Implementation":"Notebook-to-artifact-to-production-to-SQL parity validation",
        "Notebook_Section":"Section 25.1",
        "Primary_Output":"Approved frozen model snapshot",
        "Code_Proof":"Parity + version + SQL consistency checks",
        "Export_Proof":"predeployment_snapshot/",
        "Future_Streamlit_Proof":"Model/build metadata",
        "Owner_Layer":"Model Governance",
        "Current_Status":"COMPLETE"
    },

    {
        "Requirement_ID":"FLASK_API",
        "Requirement":"Versioned Flask REST API",
        "Implementation":"Thin API routes calling reusable inference and SQL services",
        "Notebook_Section":"Planned Section 27",
        "Primary_Output":"api/app.py + versioned routes",
        "Code_Proof":"Flask REST endpoints",
        "Export_Proof":"API request/response JSON + API documentation",
        "Future_Streamlit_Proof":"API Status / all API-backed pages",
        "Owner_Layer":"Backend API",
        "Current_Status":"PLANNED"
    },

    {
        "Requirement_ID":"STREAMLIT_UI",
        "Requirement":"Client-facing Streamlit application",
        "Implementation":"Multi-page dashboard consuming Flask API responses",
        "Notebook_Section":"Planned Section 29",
        "Primary_Output":"streamlit_app/",
        "Code_Proof":"Streamlit pages + reusable components",
        "Export_Proof":"Screenshots + downloadable page outputs",
        "Future_Streamlit_Proof":"Entire client application",
        "Owner_Layer":"Frontend",
        "Current_Status":"PLANNED"
    },

    {
        "Requirement_ID":"EXPORT_FRAMEWORK",
        "Requirement":"Every important result must be exportable",
        "Implementation":"Centralized CSV/XLSX/JSON/chart evidence framework",
        "Notebook_Section":"Existing task exports + final framework pending",
        "Primary_Output":"exports/ manifest and task evidence",
        "Code_Proof":"Export utilities and task outputs",
        "Export_Proof":"exports/",
        "Future_Streamlit_Proof":"Download controls",
        "Owner_Layer":"Evidence / Reporting",
        "Current_Status":"IN PROGRESS"
    },

    {
        "Requirement_ID":"SCREENSHOT_EVIDENCE",
        "Requirement":"Systematic visual evidence",
        "Implementation":"Named screenshots mapped to client requirements",
        "Notebook_Section":"Final evidence stage",
        "Primary_Output":"screenshots/ + evidence index",
        "Code_Proof":"Screenshot evidence workflow",
        "Export_Proof":"screenshots/",
        "Future_Streamlit_Proof":"All major application pages",
        "Owner_Layer":"Evidence",
        "Current_Status":"IN PROGRESS"
    },

    {
        "Requirement_ID":"DEPLOYMENT",
        "Requirement":"Streamlit frontend + separately hosted Flask REST API",
        "Implementation":"Two-service deployment architecture",
        "Notebook_Section":"Planned deployment stages",
        "Primary_Output":"Hosted Streamlit + hosted Flask service",
        "Code_Proof":"Deployment configuration and service code",
        "Export_Proof":"Deployment guide + smoke-test evidence",
        "Future_Streamlit_Proof":"Live deployment",
        "Owner_Layer":"Deployment",
        "Current_Status":"PLANNED"
    },

    {
        "Requirement_ID":"DOCKER_SCOPE_CHANGE",
        "Requirement":"Document Docker/containerization substitution",
        "Implementation":"Record approved substitution with Streamlit frontend + hosted Flask API",
        "Notebook_Section":"Pre-deployment documentation",
        "Primary_Output":"scope_change_log.md",
        "Code_Proof":"Not applicable — governance evidence",
        "Export_Proof":"reports/scope_change_log.md",
        "Future_Streamlit_Proof":"About / Methodology",
        "Owner_Layer":"Project Governance",
        "Current_Status":"PENDING DOCUMENTATION"
    },

    {
        "Requirement_ID":"HANDOVER",
        "Requirement":"Reproducible client handover",
        "Implementation":"README, final report, run instructions, evidence index and presentation",
        "Notebook_Section":"Final handover stage",
        "Primary_Output":"Final client submission package",
        "Code_Proof":"Repository + documentation",
        "Export_Proof":"reports/ + README + presentation",
        "Future_Streamlit_Proof":"About / Methodology",
        "Owner_Layer":"Handover",
        "Current_Status":"PLANNED"
    }
])

status_order252={
    "COMPLETE":0,
    "IN PROGRESS":1,
    "PENDING DOCUMENTATION":2,
    "PLANNED":3
}

requirements_matrix252[
    "Status_Order"
]=requirements_matrix252[
    "Current_Status"
].map(
    status_order252
)

requirements_matrix252=(
    requirements_matrix252
    .sort_values(
        [
            "Status_Order",
            "Requirement_ID"
        ]
    )
    .drop(
        columns="Status_Order"
    )
    .reset_index(drop=True)
)

requirements_matrix252[
    "Evidence_Defined"
]=requirements_matrix252[
    [
        "Code_Proof",
        "Export_Proof",
        "Future_Streamlit_Proof"
    ]
].notna().all(axis=1)

requirements_matrix252[
    "Traceability_Status"
]=np.where(
    requirements_matrix252[
        "Evidence_Defined"
    ],
    "PASS",
    "REVIEW"
)

matrix_status252=(
    "PASS"
    if (
        requirements_matrix252[
            "Traceability_Status"
        ]=="PASS"
    ).all()
    else "REVIEW"
)

requirements_csv252=(
    REPORT_DIR/
    "requirements_matrix.csv"
)

requirements_xlsx252=(
    REPORT_DIR/
    "requirements_matrix.xlsx"
)

requirements_matrix252.to_csv(
    requirements_csv252,
    index=False
)

xlsx_status252="CREATED"

try:
    with pd.ExcelWriter(
        requirements_xlsx252,
        engine="openpyxl"
    ) as writer:
        requirements_matrix252.to_excel(
            writer,
            sheet_name="Requirements Matrix",
            index=False
        )
        summary252=pd.DataFrame({
            "Metric":[
                "Total Requirements",
                "Complete",
                "In Progress",
                "Pending Documentation",
                "Planned",
                "Traceability Gate"
            ],
            "Value":[
                len(requirements_matrix252),
                int(
                    requirements_matrix252[
                        "Current_Status"
                    ].eq(
                        "COMPLETE"
                    ).sum()
                ),
                int(
                    requirements_matrix252[
                        "Current_Status"
                    ].eq(
                        "IN PROGRESS"
                    ).sum()
                ),
                int(
                    requirements_matrix252[
                        "Current_Status"
                    ].eq(
                        "PENDING DOCUMENTATION"
                    ).sum()
                ),
                int(
                    requirements_matrix252[
                        "Current_Status"
                    ].eq(
                        "PLANNED"
                    ).sum()
                ),
                matrix_status252]})
        summary252.to_excel(
            writer,
            sheet_name="Summary",
            index=False
        )

except Exception as e:
    xlsx_status252=(
        f"REVIEW: {type(e).__name__}"
    )

matrix_summary252=pd.DataFrame({
    "Status":[
        "COMPLETE",
        "IN PROGRESS",
        "PENDING DOCUMENTATION",
        "PLANNED"
    ],

    "Requirements":[
        int(
            requirements_matrix252[
                "Current_Status"
            ].eq("COMPLETE").sum()
        ),

        int(
            requirements_matrix252[
                "Current_Status"
            ].eq("IN PROGRESS").sum()
        ),

        int(
            requirements_matrix252[
                "Current_Status"
            ].eq(
                "PENDING DOCUMENTATION"
            ).sum()
        ),

        int(
            requirements_matrix252[
                "Current_Status"
            ].eq("PLANNED").sum())]})

requirements_metadata252={
    "created_at_utc":
        str(
            pd.Timestamp.now(
                tz="UTC"
            )
        ),

    "total_requirements":
        int(
            len(
                requirements_matrix252
            )
        ),

    "traceability_status":
        matrix_status252,

    "csv_file":
        str(
            requirements_csv252
        ),

    "xlsx_file":
        str(
            requirements_xlsx252
        ),

    "xlsx_status":
        xlsx_status252,

    "deployment_architecture":
        (
            "Streamlit frontend + "
            "separately hosted Flask REST API "
            "+ SQL Server + ML + recommender"
        )
}

with open(
    REPORT_DIR/
    "requirements_matrix_metadata.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        requirements_metadata252,
        f,
        indent=2
    )

total252=len(requirements_matrix252)

complete252=int(
    requirements_matrix252[
        "Current_Status"
    ].eq("COMPLETE").sum()
)

progress252=(
    complete252/total252*100
    if total252
    else 0
)

trace_pass252=int(
    requirements_matrix252[
        "Traceability_Status"
    ].eq("PASS").sum()
)

trace_pct252=(
    trace_pass252/total252*100
    if total252
    else 0
)

open252=int(
    (
        requirements_matrix252[
            "Current_Status"
        ]!="COMPLETE"
    ).sum()
)

client_tasks252=(
    requirements_matrix252[
        requirements_matrix252[
            "Requirement_ID"
        ].str.startswith(
            "TASK_"
        )
    ]
    .copy()
)

cross_cutting252=(
    requirements_matrix252[
        ~requirements_matrix252[
            "Requirement_ID"
        ].str.startswith(
            "TASK_"
        )
    ]
    .copy()
)

dashboard_summary252=pd.DataFrame({
    "Metric":[
        "Total Requirements",
        "Completed",
        "Open Requirements",
        "Implementation Progress %",
        "Evidence Coverage %",
        "Client Tasks Complete"
    ],
    "Value":[
        total252,
        complete252,
        open252,
        round(progress252,1),
        round(trace_pct252,1),
        int(
            client_tasks252[
                "Current_Status"
            ].eq("COMPLETE").sum())]})

dashboard_summary252.to_csv(
    REPORT_DIR/
    "requirements_dashboard_summary.csv",
    index=False
)

status_colors252={
    "COMPLETE":"#4a4e69",
    "IN PROGRESS":"#9a8c98",
    "PENDING DOCUMENTATION":"#c9ada7",
    "PLANNED":"#ddd4d0"
}

def card252(
    label,
    value,
    note,
    accent="#4a4e69"
):
    return f"""
    <div class="kpi252"
    style="border-top:4px solid {accent}">
        <div class="klabel252">{label}</div>
        <div class="kvalue252">{value}</div>
        <div class="knote252">{note}</div>
    </div>
    """

def status_badge252(status):
    color=status_colors252.get(
        status,
        "#9a8c98"
    )
    return f"""
    <span style="
    display:inline-block;
    padding:4px 8px;
    border-radius:12px;
    background:{color}22;
    color:#22223b;
    font-weight:700;
    font-size:9px;
    border:1px solid {color}55">
    {status}
    </span>
    """

def table252(df,max_rows=50):
    z=df.copy().head(max_rows)
    display(
        HTML(
            z.to_html(
                index=False,
                border=0,
                classes="t252",
                escape=False
            )
        )
)

    if len(df)>max_rows:

        display(
            HTML(
                f"""
                <div class="note252">
                Showing first
                <b>{max_rows}</b>
                of
                <b>{len(df)}</b>
                rows.
                </div>
                """
            )
        )

display(
    HTML("""
<style>

.d252{
border:1px solid #ddd8d5;
border-radius:18px;
overflow:hidden;
background:#fff;
box-shadow:0 10px 26px rgba(34,34,59,.10)
}

.h252{
background:
linear-gradient(
120deg,
#22223b 0%,
#30324d 52%,
#4a4e69 100%
);
padding:25px 28px;
display:flex;
justify-content:space-between;
align-items:flex-start;
gap:24px;
font-family:Arial
}

.h252 .eyebrow{
color:#c9ada7;
font-size:8px;
font-weight:700;
letter-spacing:1.8px
}

.h252 h2{
color:#fff;
margin:6px 0 5px;
font-size:25px
}

.h252 p{
color:#f2e9e4;
font-size:10px;
line-height:1.6;
margin:0
}

.meta252{
color:#f2e9e4;
font:9px/1.7 Arial;
text-align:right;
min-width:170px
}

.meta252 b{
color:#fff;
font-size:13px
}

.kpis252{
display:grid;
grid-template-columns:
repeat(6,1fr);
border-bottom:
1px solid #e8e2df;
background:#fff
}

.kpi252{
padding:14px 13px;
border-right:
1px solid #eee9e7;
font-family:Arial;
min-width:0
}

.klabel252{
font-size:7px;
font-weight:700;
color:#9a8c98;
letter-spacing:.5px
}

.kvalue252{
font-size:18px;
font-weight:700;
color:#22223b;
margin-top:5px;
word-break:break-word
}

.knote252{
font-size:8px;
color:#4a4e69;
margin-top:4px
}

.control252{
padding:12px 17px 6px;
background:#fbfaf9;
border-bottom:1px solid #eee9e7
}

.filters252{
padding:9px 17px 13px;
background:#fbfaf9;
border-bottom:1px solid #eee9e7
}

.out252{
padding:19px 22px;
min-height:230px
}

.t252{
width:100%;
border-collapse:collapse;
font:9px Arial
}

.t252 th{
background:#22223b;
color:#fff;
padding:9px 8px;
text-align:left;
font-weight:700;
position:sticky;
top:0
}

.t252 td{
padding:8px;
border-bottom:1px solid #eee9e7;
color:#22223b;
vertical-align:top;
line-height:1.5
}

.t252 tr:nth-child(even){
background:#fbfaf9
}

.grid252{
display:grid;
grid-template-columns:
repeat(3,1fr);
gap:13px
}

.box252{
background:#f8f6f5;
border-left:
4px solid #4a4e69;
padding:15px;
border-radius:9px;
font:11px/1.65 Arial;
color:#22223b
}

.box252 b{
display:block;
font-size:9px;
color:#4a4e69;
margin-bottom:7px
}

.note252{
background:#f2e9e4;
padding:12px 14px;
border-radius:9px;
font:10px/1.65 Arial;
color:#4a4e69
}

.callout252{
border:1px solid #ded8d5;
border-radius:10px;
padding:14px;
font:10px/1.6 Arial;
color:#22223b;
background:#fff
}

.progress252{
height:9px;
background:#eee9e7;
border-radius:10px;
overflow:hidden;
margin-top:8px
}

.progress252 span{
display:block;
height:100%;
background:#4a4e69
}

.section-title252{
font:700 13px Arial;
color:#22223b;
margin:0 0 12px
}

.small252{
font:9px/1.55 Arial;
color:#9a8c98
}

@media(max-width:900px){

.kpis252{
grid-template-columns:
repeat(3,1fr)
}

.grid252{
grid-template-columns:1fr
}

.h252{
flex-direction:column
}

.meta252{
text-align:left
}

}

</style>
""")
)


header252=widgets.HTML(
    f"""
    <div class="h252">

        <div>

            <div class="eyebrow">
            SECTION 25.2 • REQUIREMENTS TRACEABILITY
            </div>

            <h2>
            Client Requirement & Evidence Control Center
            </h2>

            <p>
            Implementation progress • Evidence coverage •
            Open work • Ownership • Deployment readiness
            </p>

        </div>

        <div class="meta252">

            TRACEABILITY GATE<br>
            <b>{matrix_status252}</b>

            <br><br>

            LAST REFRESH<br>
            <b>
            {
                pd.Timestamp.now(
                    tz="Asia/Kolkata"
                ).strftime(
                    "%d %b %Y • %I:%M %p"
                )
            }
            </b>

        </div>

    </div>
    """
)

kpi252=widgets.HTML(
    '<div class="kpis252">'
    +
    "".join([
        card252(
            "IMPLEMENTATION",
            f"{progress252:.0f}%",
            f"{complete252}/{total252} complete",
            "#4a4e69"
        ),

        card252(
            "EVIDENCE COVERAGE",
            f"{trace_pct252:.0f}%",
            f"{trace_pass252}/{total252} mapped",
            "#9a8c98"
        ),

        card252(
            "CLIENT TASKS",
            f"{client_tasks252['Current_Status'].eq('COMPLETE').sum()}/8",
            "Analytical requirements",
            "#4a4e69"
        ),

        card252(
            "OPEN ITEMS",
            open252,
            "Remaining requirements",
            "#c9ada7"
        ),

        card252(
            "PLANNED",
            int(
                requirements_matrix252[
                    "Current_Status"
                ].eq(
                    "PLANNED"
                ).sum()
            ),
            "Later implementation",
            "#9a8c98"
        ),

        card252(
            "TRACEABILITY",
            matrix_status252,
            "Requirement → evidence",
            "#4a4e69"
        )
    ])
    +
    "</div>"
)


view252=widgets.ToggleButtons(
    options=[
        ("EXECUTIVE","overview"),
        ("STATUS","status"),
        ("CLIENT TASKS","tasks"),
        ("CROSS-CUTTING","cross"),
        ("OPEN WORK","open"),
        ("EVIDENCE MAP","evidence"),
        ("OWNERSHIP","ownership"),
        ("FILES","files")
    ],

    value="overview",

    layout=widgets.Layout(
        width="100%"
    ),

    style={
        "button_width":"12.5%",
        "font_weight":"700"
    }
)

view252.add_class(
    "control252"
)

owner_options252=[
    "All"
]+sorted(
    requirements_matrix252[
        "Owner_Layer"
    ].dropna().unique()
)

status_options252=[
    "All",
    "COMPLETE",
    "IN PROGRESS",
    "PENDING DOCUMENTATION",
    "PLANNED"
]

owner_filter252=widgets.Dropdown(
    options=owner_options252,
    value="All",
    description="Owner:",
    layout=widgets.Layout(
        width="30%"
    )
)

status_filter252=widgets.Dropdown(
    options=status_options252,
    value="All",
    description="Status:",
    layout=widgets.Layout(
        width="27%"
    )
)

search252=widgets.Text(
    value="",
    placeholder=
        "Search requirement, implementation or evidence...",
    description="Search:",
    layout=widgets.Layout(
        width="35%"
    )
)

reset252=widgets.Button(
    description="Reset",
    icon="refresh",
    button_style="",
    layout=widgets.Layout(
        width="8%"
    )
)

filters252=widgets.HBox(
    [
        owner_filter252,
        status_filter252,
        search252,
        reset252
    ],
    layout=widgets.Layout(
        width="100%"
    )
)

filters252.add_class(
    "filters252"
)

out252=widgets.Output()

out252.add_class("out252")

def filtered_requirements252():
    z=requirements_matrix252.copy()
    if owner_filter252.value!="All":
        z=z[z["Owner_Layer"].eq(owner_filter252.value)]
    if status_filter252.value!="All":
        z=z[z["Current_Status"].eq(status_filter252.value)]

    q=search252.value.strip().lower()
    if q:
        searchable252=(
            z[
                [
                    "Requirement_ID",
                    "Requirement",
                    "Implementation",
                    "Code_Proof",
                    "Export_Proof",
                    "Owner_Layer"
                ]
            ]
            .fillna("")
            .astype(str)
            .agg(
                " ".join,
                axis=1
            )
            .str.lower()
        )
        z=z[searchable252.str.contains(q,regex=False)]
    return z

def status_chart252():
    order=[
        "COMPLETE",
        "IN PROGRESS",
        "PENDING DOCUMENTATION",
        "PLANNED"
    ]
    counts=(
        requirements_matrix252[
            "Current_Status"
        ]
        .value_counts()
        .reindex(
            order,
            fill_value=0))

    fig=go.Figure(
        go.Pie(
            labels=counts.index,
            values=counts.values,
            hole=.60,
            marker=dict(
                colors=[
                    status_colors252[
                        x
                    ]
                    for x
                    in counts.index
                ]
            ),
            textinfo=
                "label+value",
            hovertemplate=
                "<b>%{label}</b><br>"
                "Requirements: %{value}"
                "<br>Share: %{percent}"
                "<extra></extra>"))

    fig.update_layout(
        title=
            "Requirement Completion Mix",
        height=390,
        margin=dict(
            l=25,
            r=25,
            t=60,
            b=25
        ),
        paper_bgcolor="#fff",
        font=dict(
            family="Arial",
            color="#22223b"
        ),
        annotations=[
            dict(
                text=
                    f"<b>{progress252:.0f}%</b>"
                    "<br>complete",
                x=.5,
                y=.5,
                showarrow=False,
                font=dict(
                    size=16,
                    color="#22223b"
                )
            )
        ],
        legend=dict(orientation="h",y=-.10))
    display(fig)

def ownership_chart252():
    owners=(
        requirements_matrix252
        .groupby(
            "Owner_Layer"
        )
        .agg(
            Total=(
                "Requirement_ID",
                "count"
            ),
            Complete=(
                "Current_Status",
                lambda s:
                s.eq(
                    "COMPLETE"
                ).sum()
            )
        )
        .reset_index()
    )
    owners[
        "Open"
    ]=(
        owners["Total"]
        -
        owners["Complete"]
    )
    owners=owners.sort_values(
        "Total",
        ascending=True
    )

    fig=go.Figure()
    fig.add_trace(
        go.Bar(
            y=owners[
                "Owner_Layer"
            ],
            x=owners[
                "Complete"
            ],
            orientation="h",
            name="Complete",
            marker_color=
                "#4a4e69"
        )
    )
    fig.add_trace(
        go.Bar(
            y=owners[
                "Owner_Layer"
            ],
            x=owners[
                "Open"
            ],
            orientation="h",
            name="Open",
            marker_color=
                "#c9ada7"
        )
    )

    fig.update_layout(
        barmode="stack",
        title=
            "Requirement Ownership & Workload",
        height=max(
            430,
            len(owners)*34
        ),
        xaxis_title=
            "Number of Requirements",
        yaxis_title="",
        paper_bgcolor="#fff",
        plot_bgcolor="#fff",
        margin=dict(
            l=30,
            r=25,
            t=60,
            b=40
        ),
        font=dict(
            family="Arial",
            color="#22223b"
        ),
        legend=dict(
            orientation="h",
            y=1.08
        )
    )
    display(fig)

def progress_chart252():
    fig=go.Figure(
        go.Indicator(
            mode=
                "gauge+number",
            value=
                progress252,
            number={
                "suffix":"%"
            },
            title={
                "text":
                "Implemented Requirements"
            },
            gauge={
                "axis":{
                    "range":[0,100]
                },

                "bar":{
                    "color":
                    "#4a4e69"
                },

                "steps":[
                    {
                        "range":[0,50],
                        "color":
                        "#f2e9e4"
                    },

                    {
                        "range":[50,80],
                        "color":
                        "#ddd4d0"
                    },

                    {
                        "range":[80,100],
                        "color":
                        "#c9ada7"}]}))

    fig.update_layout(
        height=310,
        margin=dict(
            l=35,
            r=35,
            t=55,
            b=25
        ),
        paper_bgcolor="#fff",
        font=dict(
            family="Arial",
            color="#22223b"))

    display(fig)

def update252(*_):
    with out252:
        clear_output(wait=True)
        filtered252=(filtered_requirements252())
        if view252.value=="overview":
            display(
                HTML(
                    f"""
                    <div class="grid252">
                    <div class="box252">
                    <b>ANALYTICAL DELIVERY</b>
                    All
                    <strong>
                    {
                    client_tasks252[
                        'Current_Status'
                    ].eq(
                        'COMPLETE'
                    ).sum()
                    } of 8
                    </strong>
                    client analytical tasks are
                    implemented and mapped to
                    evidence.
                    </div>

                    <div class="box252">
                    <b>IMPLEMENTATION PROGRESS</b>
                    {complete252} of
                    {total252} total project
                    requirements are complete.

                    <div class="progress252">
                    <span style="
                    width:{progress252}%">
                    </span>
                    </div>
                    </div>

                    <div class="box252">
                    <b>EVIDENCE CONTROL</b>
                    {trace_pass252} of
                    {total252} requirements
                    currently have defined code,
                    export and UI evidence paths.

                    <div class="progress252">
                    <span style="
                    width:{trace_pct252}%">
                    </span>
                    </div>
                    </div>

                    </div>
                    """
                )
            )

            progress_chart252()

            display(
                HTML(
                    f"""
                    <div class="grid252">

                    <div class="callout252">
                    <b>
                    Current milestone
                    </b><br><br>

                    Pre-deployment documentation
                    and reproducibility closure.
                    </div>

                    <div class="callout252">
                    <b>
                    Main remaining engineering
                    </b><br><br>

                    Flask API, Streamlit frontend,
                    final export/evidence framework
                    and deployment.
                    </div>

                    <div class="callout252">
                    <b>
                    Governance watch-item
                    </b><br><br>

                    Docker/containerization
                    substitution still requires
                    explicit scope-change evidence.
                    </div>

                    </div>
                    """
                )
            )


        elif view252.value=="status":

            status_chart252()

            status_table252=(
                requirements_matrix252[
                    "Current_Status"
                ]
                .value_counts()
                .rename_axis(
                    "Status"
                )
                .reset_index(
                    name="Requirements"
                )
            )

            status_table252[
                "Share_%"
            ]=(
                status_table252[
                    "Requirements"
                ]
                /total252*100
            ).round(1)

            table252(
                status_table252
            )


        elif view252.value=="tasks":

            display(
                HTML(
                    """
                    <div class="note252">
                    <b>
                    Client analytical requirements
                    </b><br>
                    These eight rows correspond
                    directly to the eight required
                    business tasks.
                    </div><br>
                    """
                )
            )

            task_view252=(
                client_tasks252[
                    [
                        "Requirement_ID",
                        "Requirement",
                        "Notebook_Section",
                        "Primary_Output",
                        "Export_Proof",
                        "Current_Status"
                    ]
                ]
                .copy()
            )

            task_view252[
                "Current_Status"
            ]=task_view252[
                "Current_Status"
            ].map(
                status_badge252
            )

            table252(
                task_view252
            )


        elif view252.value=="cross":

            cross252=(
                cross_cutting252[
                    [
                        "Requirement_ID",
                        "Requirement",
                        "Owner_Layer",
                        "Notebook_Section",
                        "Current_Status"
                    ]
                ]
                .copy()
            )

            cross252[
                "Current_Status"
            ]=cross252[
                "Current_Status"
            ].map(
                status_badge252
            )

            table252(
                cross252
            )


        elif view252.value=="open":

            open_work252=(
                requirements_matrix252[
                    requirements_matrix252[
                        "Current_Status"
                    ]!="COMPLETE"
                ][
                    [
                        "Requirement_ID",
                        "Requirement",
                        "Owner_Layer",
                        "Implementation",
                        "Primary_Output",
                        "Current_Status"
                    ]
                ]
                .copy()
            )


            priority_map252={
                "PENDING DOCUMENTATION":
                    "1 • CLOSE NOW",

                "IN PROGRESS":
                    "2 • ACTIVE",

                "PLANNED":
                    "3 • UPCOMING"
            }


            open_work252[
                "Priority"
            ]=open_work252[
                "Current_Status"
            ].map(
                priority_map252
            )


            open_work252=(
                open_work252
                .sort_values(
                    "Priority"
                )
            )


            open_work252[
                "Current_Status"
            ]=open_work252[
                "Current_Status"
            ].map(
                status_badge252
            )


            table252(
                open_work252[
                    [
                        "Priority",
                        "Requirement_ID",
                        "Requirement",
                        "Owner_Layer",
                        "Primary_Output",
                        "Current_Status"
                    ]
                ]
            )


        elif view252.value=="evidence":

            evidence252=(
                filtered252[
                    [
                        "Requirement_ID",
                        "Requirement",
                        "Code_Proof",
                        "Export_Proof",
                        "Future_Streamlit_Proof",
                        "Traceability_Status"
                    ]
                ]
                .copy()
            )


            evidence252[
                "Traceability_Status"
            ]=evidence252[
                "Traceability_Status"
            ].map(
                lambda x:
                status_badge252(
                    "COMPLETE"
                    if x=="PASS"
                    else "IN PROGRESS"
                )
            )


            display(
                HTML(
                    f"""
                    <div class="note252">
                    Current filtered evidence view:
                    <b>{len(evidence252)}</b>
                    requirements.
                    Use the filters above to inspect
                    evidence by owner, status or keyword.
                    </div><br>
                    """
                )
            )


            table252(
                evidence252
            )


        elif view252.value=="ownership":

            ownership_chart252()

            owner_summary252=(
                requirements_matrix252
                .groupby(
                    "Owner_Layer"
                )
                .agg(
                    Requirements=(
                        "Requirement_ID",
                        "count"
                    ),

                    Complete=(
                        "Current_Status",
                        lambda s:
                        s.eq(
                            "COMPLETE"
                        ).sum()
                    )
                )
                .reset_index()
            )

            owner_summary252[
                "Open"
            ]=(
                owner_summary252[
                    "Requirements"
                ]
                -
                owner_summary252[
                    "Complete"
                ]
            )

            owner_summary252[
                "Completion_%"
            ]=(
                owner_summary252[
                    "Complete"
                ]
                /
                owner_summary252[
                    "Requirements"
                ]
                *100
            ).round(1)


            table252(
                owner_summary252.sort_values(
                    [
                        "Open",
                        "Requirements"
                    ],
                    ascending=False
                )
            )


        else:

            files_view252=pd.DataFrame({
                "Artifact":[
                    "Requirements Matrix CSV",
                    "Requirements Matrix Excel",
                    "Requirements Metadata",
                    "Dashboard Summary"
                ],

                "Location":[
                    str(
                        requirements_csv252
                    ),

                    str(
                        requirements_xlsx252
                    ),

                    str(
                        REPORT_DIR/
                        "requirements_matrix_metadata.json"
                    ),

                    str(
                        REPORT_DIR/
                        "requirements_dashboard_summary.csv"
                    )
                ]
            })


            files_view252[
                "Exists"
            ]=files_view252[
                "Location"
            ].map(
                lambda p:
                Path(p).exists()
            )


            files_view252[
                "Status"
            ]=np.where(
                files_view252[
                    "Exists"
                ],
                "PASS",
                "REVIEW"
            )


            table252(
                files_view252
            )


            display(
                HTML(
                    f"""
                    <br>
                    <div class="note252">

                    <b>
                    Traceability Gate:
                    {matrix_status252}
                    </b>

                    <br><br>

                    The CSV is machine-friendly.
                    The Excel workbook is
                    reviewer/client-friendly and
                    includes the requirements
                    matrix plus summary sheet.

                    </div>
                    """
                )
            )


def reset_filters252(_):

    owner_filter252.value="All"

    status_filter252.value="All"

    search252.value=""


reset252.on_click(
    reset_filters252
)

for control in [
    view252,
    owner_filter252,
    status_filter252,
    search252
]:
    control.observe(
        update252,
        names="value"
    )

update252()

dashboard_252=widgets.VBox([
    header252,
    kpi252,
    view252,
    filters252,
    out252
])

dashboard_252.add_class("d252")

display(dashboard_252)
```

```python
REPORT_DIR=Path("reports")
EXPORT_DIR=Path("exports")
REPORT_DIR.mkdir(parents=True,exist_ok=True)
EXPORT_DIR.mkdir(parents=True,exist_ok=True)

def exists_any253(paths):
    return any(Path(p).exists() for p in paths)

def all_exist253(paths):
    return all(Path(p).exists() for p in paths)

def status253(passed=False,planned=False):
    if planned:
        return "PLANNED"
    return "PASS" if bool(passed) else "OPEN"

def normalize_pass253(value):
    if value is None:
        return False
    if isinstance(value,(bool,np.bool_)):
        return bool(value)
    if isinstance(value,str):
        return value.strip().upper() in {
            "PASS","TRUE","SUCCESS","OK"
        }
    if isinstance(value,pd.DataFrame):
        if value.empty:
            return False
        if "Status" in value.columns:
            return (
                value["Status"]
                .astype(str)
                .str.upper()
                .eq("PASS")
                .all()
            )

        return True

    if isinstance(value,pd.Series):
        if value.empty:
            return False
        return (
            value.astype(str)
            .str.upper()
            .eq("PASS")
            .all()
        )

    try:
        return bool(value)

    except Exception:
        return False

def file_contains253(path,terms):
    path=Path(path)
    if not path.exists():
        return False
    try:
        text=path.read_text(
            encoding="utf-8",
            errors="ignore"
        ).lower()
        return all(
            str(term).lower() in text
            for term in terms
        )
    except Exception:
        return False

task_status253={}

if "requirements_matrix252" in globals():
    task_status253=dict(
        zip(
            requirements_matrix252[
                "Requirement_ID"
            ].astype(str),

            requirements_matrix252[
                "Current_Status"
            ].astype(str)
        )
    )

def matrix_complete253(requirement_id):
    return (
        task_status253.get(
            requirement_id,
            ""
        ).upper()=="COMPLETE"
    )

criteria253=[]

def add253(
    gate_id,
    requirement_id,
    layer,
    criterion,
    evidence,
    status,
    blocking,
    source
):

    criteria253.append({
        "Gate_ID":gate_id,
        "Requirement_ID":requirement_id,
        "Layer":layer,
        "Acceptance_Criterion":criterion,
        "Evidence_Required":evidence,
        "Current_Status":status,
        "Blocking_Now":bool(blocking),
        "PDF_Source":source
    })

requirements_matrix_ready253=exists_any253([
    "reports/requirements_matrix.csv",
    "reports/requirements_matrix.xlsx"
])

add253(
    "GOV-01",
    "PROJECT_GOVERNANCE",
    "Governance",
    "Every client requirement has an implementation owner, output and evidence location.",
    "Requirements matrix",
    status253(requirements_matrix_ready253),
    True,
    "Step 01"
)

scope_log_ready253=Path(
    "reports/scope_change_log.md"
).exists()

add253(
    "GOV-02",
    "DOCKER_SCOPE_CHANGE",
    "Governance",
    "Docker/containerization substitution is explicitly documented rather than silently treated as equivalent to Streamlit deployment.",
    "reports/scope_change_log.md",
    status253(scope_log_ready253),
    True,
    "Step 01"
)

engineering_modules253={
    "Logger":Path("src/logger.py").exists(),
    "Exceptions":Path("src/exception.py").exists(),
    "Settings":Path("config/settings.py").exists(),
    "Environment Template":Path(".env.example").exists()
}

engineering_ready253=all(
    engineering_modules253.values()
)

add253(
    "ENG-01",
    "ENGINEERING_CONTROLS",
    "Engineering",
    "Central logger, exception handling and environment-aware configuration exist as reusable application modules.",
    "src/logger.py + src/exception.py + config/settings.py + .env.example",
    status253(engineering_ready253),
    True,
    "Steps 02–03"
)


application_log_ready253=Path(
    "logs/application.log"
).exists()

add253(
    "ENG-02",
    "APPLICATION_LOGGING",
    "Engineering",
    "Important pipeline and application events are written to a persistent application log.",
    "logs/application.log",
    status253(application_log_ready253),
    True,
    "Step 03"
)


schema_ready253=exists_any253([
    "exports/schema_validation_report.csv",
    "schema_validation_report.csv"
])

logical_validation_ready253=exists_any253([
    "exports/logical_validation_report.csv",
    "logical_validation_report.csv"
])

add253(
    "DATA-01",
    "DATA_VALIDATION",
    "Data Engineering",
    "Blocking schema errors stop the pipeline and validation warnings remain visible in saved evidence.",
    "Schema validation + logical validation evidence",
    status253(
        schema_ready253 and
        logical_validation_ready253
    ),
    True,
    "Step 05"
)


processed_data_ready253=(
    "df_final" in globals()
)

add253(
    "DATA-02",
    "DATA_INTEGRITY",
    "Data Engineering",
    "Cleaning, missing-value, duplicate and anomaly decisions remain auditable and result in a validated processed dataset.",
    "Processed dataset + cleaning evidence",
    status253(processed_data_ready253),
    True,
    "Steps 06–10"
)


feature_dictionary_ready253=exists_any253([
    "reports/feature_dictionary.xlsx",
    "reports/feature_dictionary.csv"
])

add253(
    "FEAT-01",
    "FEATURE_ENGINEERING",
    "Data Engineering",
    "Reusable engineered features are documented with source, transformation and business meaning.",
    "reports/feature_dictionary.xlsx or CSV",
    status253(feature_dictionary_ready253),
    True,
    "Step 11"
)


add253(
    "T01-01",
    "TASK_01",
    "Analytics",
    "Keyword compensation rankings include sufficient frequency or sample-size context.",
    "Task 1 salary-keyword analysis",
    status253(
        matrix_complete253("TASK_01")
    ),
    True,
    "Step 13"
)

add253(
    "T01-02",
    "TASK_01",
    "Analytics",
    "Keyword-compensation association is not incorrectly presented as causal.",
    "Task 1 methodology and interpretation",
    status253(
        matrix_complete253("TASK_01")
    ),
    True,
    "Step 13"
)


add253(
    "T02-01",
    "TASK_02",
    "Analytics",
    "An emerging role must demonstrate meaningful growth while maintaining sufficient posting volume.",
    "Emerging category rankings",
    status253(
        matrix_complete253("TASK_02")
    ),
    True,
    "Step 14"
)

add253(
    "T02-02",
    "TASK_02",
    "Analytics",
    "Emerging-job scoring logic and thresholds are documented and reproducible.",
    "Emerging score methodology",
    status253(
        matrix_complete253("TASK_02")
    ),
    True,
    "Step 14"
)


forecast_model_ready253=all([
    "model_summary18" in globals(),
    "model_registry18" in globals(),
    "selected_forecast_daily18" in globals()
])

add253(
    "T03-01",
    "TASK_03",
    "Machine Learning",
    "Forecast models are evaluated using chronological holdout data rather than training performance.",
    "Model comparison + chronological validation periods",
    status253(forecast_model_ready253),
    True,
    "Step 15"
)

add253(
    "T03-02",
    "TASK_03",
    "Machine Learning",
    "The selected forecasting model is justified against baseline candidates using out-of-sample metrics.",
    "MAE + RMSE + sMAPE + WAPE + model registry",
    status253(forecast_model_ready253),
    True,
    "Step 15"
)


actual_pred_csv253=exists_any253([
    "exports/task_03_actual_vs_predicted.csv",
    "exports/task3_actual_vs_predicted.csv"
])

actual_pred_chart253=exists_any253([
    "charts/task_03_actual_vs_predicted.png",
    "charts/task3_actual_vs_predicted.png"
])

add253(
    "T03-03",
    "TASK_03",
    "Machine Learning",
    "Actual-vs-predicted validation evidence is preserved as both a data export and static visualization.",
    "Actual-vs-predicted CSV + PNG",
    status253(
        actual_pred_csv253 and
        actual_pred_chart253
    ),
    True,
    "Step 15"
)


add253(
    "T04-01",
    "TASK_04",
    "Analytics",
    "Country compensation comparisons display sample size and clearly disclose compensation-unit limitations.",
    "Country hourly-rate evidence",
    status253(
        matrix_complete253("TASK_04")
    ),
    True,
    "Step 16"
)


recommender_ready253=all([
    "evaluation19" in globals(),
    "BEST_WEIGHTS19" in globals(),
    "RECOMMENDER_VERSION" in globals()
])

add253(
    "T05-01",
    "TASK_05",
    "Recommendation",
    "Recommendations provide an overall score together with human-readable matching reasons.",
    "Recommendation examples + explanation fields",
    status253(recommender_ready253),
    True,
    "Step 17"
)


reproducible_ready253=normalize_pass253(
    globals().get(
        "reproducible19",
        False
    )
)

add253(
    "T05-02",
    "TASK_05",
    "Recommendation",
    "The same recommendation request produces reproducible output for a fixed catalogue and model version.",
    "Recommendation reproducibility test",
    status253(reproducible_ready253),
    True,
    "Step 17"
)


inference_tests_ready253=False

if (
    "all_tests25" in globals()
    and
    isinstance(all_tests25,pd.DataFrame)
    and
    "Status" in all_tests25.columns
):

    inference_tests_ready253=(
        all_tests25[
            "Status"
        ]
        .astype(str)
        .str.upper()
        .eq("PASS")
        .all()
    )

add253(
    "T05-03",
    "TASK_05",
    "Recommendation",
    "Invalid, empty and edge-case recommendation inputs are handled without uncontrolled failure.",
    "Inference robustness tests",
    status253(inference_tests_ready253),
    True,
    "Step 17"
)


add253(
    "T06-01",
    "TASK_06",
    "Analytics",
    "Month-over-month market metrics reconcile with underlying posting counts.",
    "Monthly KPI evidence",
    status253(
        matrix_complete253("TASK_06")
    ),
    True,
    "Step 18"
)

add253(
    "T06-02",
    "TASK_06",
    "Analytics",
    "Incomplete latest periods are identified and prevented from creating misleading comparisons.",
    "Market-dynamics methodology",
    status253(
        matrix_complete253("TASK_06")
    ),
    True,
    "Step 18"
)


add253(
    "T07-01",
    "TASK_07",
    "Analytics",
    "Unknown work-mode records and dataset limitations remain visible rather than being silently inferred.",
    "Remote-work methodology",
    status253(
        matrix_complete253("TASK_07")
    ),
    True,
    "Step 19"
)

add253(
    "T07-02",
    "TASK_07",
    "Analytics",
    "Any remote-work proxy or forecast includes explicit uncertainty and data-availability limitations.",
    "Remote-work limitations",
    status253(
        matrix_complete253("TASK_07")
    ),
    True,
    "Step 19"
)


scenario_ready253=(
    "scenario_detail20" in globals()
    and
    isinstance(
        scenario_detail20,
        pd.DataFrame
    )
    and
    len(scenario_detail20)>0
)

add253(
    "T08-01",
    "TASK_08",
    "Forecasting",
    "Future scenarios are explicitly labeled as conditional projections rather than certainties.",
    "Scenario methodology",
    status253(scenario_ready253),
    True,
    "Step 20"
)

add253(
    "T08-02",
    "TASK_08",
    "Forecasting",
    "Every scenario result is traceable to a validated forecast or documented scenario assumption.",
    "Scenario assumptions + outputs",
    status253(scenario_ready253),
    True,
    "Step 20"
)


model_governance_ready253=all([
    "model_registry_master21" in globals(),
    "MODEL_VERSION" in globals(),
    "RECOMMENDER_VERSION" in globals()
])

add253(
    "MODEL-01",
    "MODEL_GOVERNANCE",
    "Model Governance",
    "Selected model decisions, metrics, versions and selection rationale are explicitly documented.",
    "Model registry + comparison + rationale",
    status253(model_governance_ready253),
    True,
    "Step 21"
)


approved_parity253=(
    str(
        globals().get(
            "READY_TO_PACKAGE251",
            "REVIEW"
        )
    ).upper()=="PASS"
)

add253(
    "MODEL-02",
    "MODEL_GOVERNANCE",
    "Model Governance",
    "Saved model artifacts can load independently and reproduce approved inference.",
    "Smoke test + model parity gate",
    status253(approved_parity253),
    True,
    "Step 21"
)


sql_files_ready253=all_exist253([
    "database/schema.sql",
    "database/queries.sql",
    "src/data/database.py"
])

add253(
    "SQL-01",
    "SQL_LAYER",
    "SQL / Data Layer",
    "The SQL database can be recreated from documented schema and reusable data-access files.",
    "schema.sql + queries.sql + database.py",
    status253(sql_files_ready253),
    True,
    "Step 22"
)


database_module253=Path(
    "src/data/database.py"
)

sql_parameterized_ready253=(
    file_contains253(
        database_module253,
        ["execute"]
    )
    and
    (
        file_contains253(
            database_module253,
            ["?"]
        )
        or
        file_contains253(
            database_module253,
            ["parameter"]
        )
    )
)

add253(
    "SQL-02",
    "SQL_LAYER",
    "SQL / Data Layer",
    "Application-facing SQL uses parameterized queries rather than concatenating untrusted input.",
    "Parameterized database query implementation",
    status253(sql_parameterized_ready253),
    True,
    "Step 22"
)


add253(
    "API-01",
    "FLASK_API",
    "Backend API",
    "Every POST endpoint rejects invalid input cleanly and returns structured error responses.",
    "Flask API tests",
    "PLANNED",
    False,
    "Step 23"
)

add253(
    "API-02",
    "FLASK_API",
    "Backend API",
    "API results remain consistent with the validated offline service and inference logic.",
    "API-to-offline parity tests",
    "PLANNED",
    False,
    "Step 23"
)


add253(
    "UI-01",
    "STREAMLIT_UI",
    "Frontend",
    "A client can navigate from executive overview to supporting evidence without opening the notebook.",
    "Streamlit page structure",
    "PLANNED",
    False,
    "Steps 24–25"
)

add253(
    "UI-02",
    "STREAMLIT_UI",
    "Frontend",
    "No raw traceback is shown to users and application filters update the correct results consistently.",
    "UI testing evidence",
    "PLANNED",
    False,
    "Step 25"
)


export_manifest_ready253=Path(
    "exports/manifest.csv"
).exists()

add253(
    "EXP-01",
    "EXPORT_FRAMEWORK",
    "Evidence",
    "No important analytical or model result exists only in notebook memory.",
    "exports/manifest.csv",
    status253(export_manifest_ready253),
    True,
    "Step 26"
)


add253(
    "SHOT-01",
    "SCREENSHOT_EVIDENCE",
    "Evidence",
    "Every final application page has screenshot evidence that reconciles with exported results.",
    "screenshots/ + evidence index",
    "PLANNED",
    False,
    "Step 27"
)


add253(
    "DEP-01",
    "DEPLOYMENT",
    "Deployment",
    "Hosted Flask and Streamlit services pass production smoke tests and retain required model/database assets.",
    "Deployment smoke-test report",
    "PLANNED",
    False,
    "Step 30"
)


acceptance_criteria253=pd.DataFrame(
    criteria253
)

status_rank253={
    "OPEN":0,
    "PASS":1,
    "PLANNED":2
}

acceptance_criteria253[
    "Status_Rank"
]=acceptance_criteria253[
    "Current_Status"
].map(status_rank253)

acceptance_criteria253=(
    acceptance_criteria253
    .sort_values(
        [
            "Status_Rank",
            "Blocking_Now",
            "Layer",
            "Gate_ID"
        ],
        ascending=[
            True,
            False,
            True,
            True
        ]
    )
    .drop(
        columns="Status_Rank"
    )
    .reset_index(drop=True)
)


total_gates253=len(
    acceptance_criteria253
)

pass_gates253=int(
    acceptance_criteria253[
        "Current_Status"
    ].eq("PASS").sum()
)

open_gates253=int(
    acceptance_criteria253[
        "Current_Status"
    ].eq("OPEN").sum()
)

planned_gates253=int(
    acceptance_criteria253[
        "Current_Status"
    ].eq("PLANNED").sum()
)

blocking_open253=int(
    (
        acceptance_criteria253[
            "Current_Status"
        ].eq("OPEN")
        &
        acceptance_criteria253[
            "Blocking_Now"
        ]
    ).sum()
)

current_scope_gates253=(
    pass_gates253+
    open_gates253
)

current_pass_rate253=(
    pass_gates253/
    current_scope_gates253*100
    if current_scope_gates253
    else 0
)

PREDEPLOYMENT_ACCEPTANCE_STATUS253=(
    "PASS"
    if blocking_open253==0
    else "REVIEW"
)


acceptance_csv253=(
    REPORT_DIR/
    "acceptance_criteria.csv"
)

acceptance_xlsx253=(
    REPORT_DIR/
    "acceptance_criteria.xlsx"
)

acceptance_md253=(
    REPORT_DIR/
    "acceptance_criteria.md"
)


acceptance_criteria253.to_csv(
    acceptance_csv253,
    index=False
)


summary253=pd.DataFrame({
    "Metric":[
        "Total Gates",
        "Current-Scope Gates",
        "Passed",
        "Open",
        "Planned",
        "Blocking Open",
        "Current Pass Rate %",
        "Pre-Deployment Status"
    ],

    "Value":[
        total_gates253,
        current_scope_gates253,
        pass_gates253,
        open_gates253,
        planned_gates253,
        blocking_open253,
        round(
            current_pass_rate253,
            1
        ),
        PREDEPLOYMENT_ACCEPTANCE_STATUS253
    ]
})


summary253.to_csv(
    REPORT_DIR/
    "acceptance_criteria_summary.csv",
    index=False
)


try:

    with pd.ExcelWriter(
        acceptance_xlsx253,
        engine="openpyxl"
    ) as writer:

        acceptance_criteria253.to_excel(
            writer,
            sheet_name="Acceptance Criteria",
            index=False
        )

        summary253.to_excel(
            writer,
            sheet_name="Summary",
            index=False
        )

        acceptance_criteria253[
            acceptance_criteria253[
                "Blocking_Now"
            ]
            &
            acceptance_criteria253[
                "Current_Status"
            ].eq("OPEN")
        ].to_excel(
            writer,
            sheet_name="Blocking Gates",
            index=False
        )

except Exception as e:

    print(
        "Excel export warning:",
        type(e).__name__,
        str(e)
    )


md253=[
    "# Project Acceptance Criteria",
    "",
    f"Generated: {pd.Timestamp.now(tz='UTC')}",
    "",
    "## Current Pre-Deployment Decision",
    "",
    f"**Status: {PREDEPLOYMENT_ACCEPTANCE_STATUS253}**",
    "",
    f"- Total acceptance gates: {total_gates253}",
    f"- Current-scope gates: {current_scope_gates253}",
    f"- Passed: {pass_gates253}",
    f"- Open: {open_gates253}",
    f"- Blocking open: {blocking_open253}",
    f"- Planned for later stages: {planned_gates253}",
    f"- Current pass rate: {current_pass_rate253:.1f}%",
    "",
    "## Acceptance Register",
    ""
]


for _,r in acceptance_criteria253.iterrows():

    md253.extend([
        f"### {r['Gate_ID']} — {r['Acceptance_Criterion']}",
        "",
        f"- Requirement: {r['Requirement_ID']}",
        f"- Layer: {r['Layer']}",
        f"- Evidence required: {r['Evidence_Required']}",
        f"- Current status: **{r['Current_Status']}**",
        f"- Blocking now: {r['Blocking_Now']}",
        f"- Workflow source: {r['PDF_Source']}",
        ""
    ])


acceptance_md253.write_text(
    "\n".join(md253),
    encoding="utf-8"
)


layer_summary253=(
    acceptance_criteria253
    .groupby("Layer")
    .agg(
        Total=(
            "Gate_ID",
            "count"
        ),

        Passed=(
            "Current_Status",
            lambda x:
            x.eq("PASS").sum()
        ),

        Open=(
            "Current_Status",
            lambda x:
            x.eq("OPEN").sum()
        ),

        Planned=(
            "Current_Status",
            lambda x:
            x.eq("PLANNED").sum()
        )
    )
    .reset_index()
)


layer_summary253[
    "Current_Completion_%"
]=np.where(
    (
        layer_summary253["Passed"]+
        layer_summary253["Open"]
    )>0,

    (
        layer_summary253["Passed"]/
        (
            layer_summary253["Passed"]+
            layer_summary253["Open"]
        )*100
    ),

    np.nan
)


layer_summary253[
    "Current_Completion_%"
]=layer_summary253[
    "Current_Completion_%"
].round(1)


layer_summary253.to_csv(
    REPORT_DIR/
    "acceptance_layer_summary.csv",
    index=False
)


status_colors253={
    "PASS":"#4a4e69",
    "OPEN":"#c9ada7",
    "PLANNED":"#9a8c98"
}


def card253(
    label,
    value,
    note,
    accent="#4a4e69"
):

    return f"""
    <div class="c253"
    style="border-top:4px solid {accent}">
        <small>{label}</small>
        <b>{value}</b>
        <span>{note}</span>
    </div>
    """


def table253(df,rows=50):

    display(
        HTML(
            df.head(rows).to_html(
                index=False,
                border=0,
                classes="t253"
            )
        )
    )


display(
    HTML("""
<style>

.d253{
border:1px solid #ded8d5;
border-radius:18px;
overflow:hidden;
background:#fff;
box-shadow:0 9px 25px rgba(34,34,59,.10)
}

.h253{
background:linear-gradient(
120deg,#22223b,#30324d 55%,#4a4e69
);
padding:25px 28px;
display:flex;
justify-content:space-between;
align-items:flex-start;
gap:20px;
font-family:Arial
}

.h253 .k{
color:#c9ada7;
font-size:8px;
font-weight:700;
letter-spacing:1.7px
}

.h253 h2{
color:#fff;
margin:6px 0;
font-size:25px
}

.h253 p,
.h253 .m{
color:#f2e9e4;
font-size:10px;
line-height:1.65
}

.cards253{
display:grid;
grid-template-columns:repeat(6,1fr);
border-bottom:1px solid #e8e2df
}

.c253{
padding:14px;
border-right:1px solid #ebe6e4;
font-family:Arial;
min-width:0
}

.c253 small{
display:block;
color:#9a8c98;
font-size:7px;
font-weight:700
}

.c253 b{
display:block;
color:#22223b;
font-size:17px;
margin-top:5px
}

.c253 span{
display:block;
color:#4a4e69;
font-size:8px;
margin-top:4px
}

.out253{
padding:19px 22px;
border-top:1px solid #eee9e7;
min-height:230px
}

.t253{
width:100%;
border-collapse:collapse;
font:9px Arial
}

.t253 th{
background:#22223b;
color:#fff;
padding:9px;
text-align:left
}

.t253 td{
padding:8px;
border-bottom:1px solid #eee9e7;
vertical-align:top;
color:#22223b;
line-height:1.5
}

.t253 tr:nth-child(even){
background:#fbfaf9
}

.grid253{
display:grid;
grid-template-columns:repeat(3,1fr);
gap:12px
}

.box253{
background:#f8f6f5;
border-left:4px solid #4a4e69;
padding:14px;
border-radius:9px;
font:11px/1.65 Arial;
color:#22223b
}

.box253 b{
display:block;
color:#4a4e69;
font-size:9px;
margin-bottom:5px
}

.note253{
background:#f2e9e4;
padding:12px 14px;
border-radius:8px;
font:10px/1.65 Arial;
color:#4a4e69
}

</style>
""")
)


header253=widgets.HTML(
    f"""
<div class="h253">

<div>

<div class="k">
SECTION 25.2 • ACCEPTANCE CONTROL
</div>

<h2>
Project Quality Gate Register
</h2>

<p>
Acceptance Criteria • Evidence • Blockers •
Model Governance • Deployment Readiness
</p>

</div>

<div class="m">

PRE-DEPLOYMENT<br>
<b>
{PREDEPLOYMENT_ACCEPTANCE_STATUS253}
</b>

<br><br>

BLOCKING GATES<br>
<b>
{blocking_open253}
</b>

</div>

</div>
"""
)


kpi253=widgets.HTML(
    '<div class="cards253">'
    +
    "".join([
        card253(
            "CURRENT PASS RATE",
            f"{current_pass_rate253:.0f}%",
            "Current-scope gates",
            "#4a4e69"
        ),

        card253(
            "PASS",
            pass_gates253,
            "Verified controls",
            "#4a4e69"
        ),

        card253(
            "OPEN",
            open_gates253,
            "Require closure",
            "#c9ada7"
        ),

        card253(
            "BLOCKING",
            blocking_open253,
            "Before packaging",
            "#c9ada7"
        ),

        card253(
            "PLANNED",
            planned_gates253,
            "Future stages",
            "#9a8c98"
        ),

        card253(
            "STATUS",
            PREDEPLOYMENT_ACCEPTANCE_STATUS253,
            "Current gate",
            "#4a4e69"
        )
    ])
    +
    "</div>"
)


view253=widgets.ToggleButtons(
    options=[
        ("EXECUTIVE","overview"),
        ("BLOCKERS","blockers"),
        ("TASK GATES","tasks"),
        ("ENGINEERING","engineering"),
        ("MODEL + SQL","model"),
        ("LAYER VIEW","layers"),
        ("FUTURE","future"),
        ("FILES","files")
    ],

    value="overview",

    layout=widgets.Layout(
        width="100%"
    ),

    style={
        "button_width":"12.5%",
        "font_weight":"700"
    }
)


layer_filter253=widgets.Dropdown(
    options=[
        "All"
    ]+sorted(
        acceptance_criteria253[
            "Layer"
        ].unique()
    ),

    value="All",
    description="Layer:",

    layout=widgets.Layout(
        width="32%"
    )
)


status_filter253=widgets.Dropdown(
    options=[
        "All",
        "PASS",
        "OPEN",
        "PLANNED"
    ],

    value="All",
    description="Status:",

    layout=widgets.Layout(
        width="28%"
    )
)


search253=widgets.Text(
    value="",

    placeholder=
        "Search criterion or evidence...",

    description="Search:",

    layout=widgets.Layout(
        width="32%"
    )
)


reset253=widgets.Button(
    description="Reset",
    icon="refresh",

    layout=widgets.Layout(
        width="8%"
    )
)


filters253=widgets.HBox([
    layer_filter253,
    status_filter253,
    search253,
    reset253
])


out253=widgets.Output()
out253.add_class("out253")


def filtered253():

    d=acceptance_criteria253.copy()

    if layer_filter253.value!="All":

        d=d[
            d["Layer"].eq(
                layer_filter253.value
            )
        ]

    if status_filter253.value!="All":

        d=d[
            d[
                "Current_Status"
            ].eq(
                status_filter253.value
            )
        ]

    q=search253.value.strip().lower()

    if q:

        search_text253=(
            d[
                [
                    "Gate_ID",
                    "Requirement_ID",
                    "Layer",
                    "Acceptance_Criterion",
                    "Evidence_Required"
                ]
            ]
            .fillna("")
            .astype(str)
            .agg(" ".join,axis=1)
            .str.lower()
        )

        d=d[
            search_text253.str.contains(
                q,
                regex=False
            )
        ]

    return d


def status_chart253():

    counts253=(
        acceptance_criteria253[
            "Current_Status"
        ]
        .value_counts()
        .reindex(
            [
                "PASS",
                "OPEN",
                "PLANNED"
            ],
            fill_value=0
        )
    )

    fig=go.Figure(
        go.Pie(
            labels=counts253.index,
            values=counts253.values,
            hole=.58,

            marker=dict(
                colors=[
                    status_colors253[x]
                    for x in
                    counts253.index
                ]
            ),

            textinfo="label+value",

            hovertemplate=
                "<b>%{label}</b>"
                "<br>Gates: %{value}"
                "<br>Share: %{percent}"
                "<extra></extra>"
        )
    )

    fig.update_layout(
        title="Acceptance Gate Portfolio",
        height=370,

        margin=dict(
            l=25,
            r=25,
            t=60,
            b=25
        ),

        paper_bgcolor="#fff",

        font=dict(
            family="Arial",
            color="#22223b"
        ),

        annotations=[
            dict(
                text=
                    f"<b>{current_pass_rate253:.0f}%</b>"
                    "<br>current pass",
                x=.5,
                y=.5,
                showarrow=False
            )
        ],

        legend=dict(
            orientation="h",
            y=-.08
        )
    )

    display(fig)


def layer_chart253():

    d=layer_summary253.sort_values(
        "Total",
        ascending=True
    )

    fig=go.Figure()

    fig.add_trace(
        go.Bar(
            y=d["Layer"],
            x=d["Passed"],
            orientation="h",
            name="PASS",
            marker_color="#4a4e69"
        )
    )

    fig.add_trace(
        go.Bar(
            y=d["Layer"],
            x=d["Open"],
            orientation="h",
            name="OPEN",
            marker_color="#c9ada7"
        )
    )

    fig.add_trace(
        go.Bar(
            y=d["Layer"],
            x=d["Planned"],
            orientation="h",
            name="PLANNED",
            marker_color="#9a8c98"
        )
    )

    fig.update_layout(
        barmode="stack",
        title="Quality Gates by Project Layer",

        height=max(
            430,
            len(d)*34
        ),

        xaxis_title="Acceptance Gates",
        yaxis_title="",

        paper_bgcolor="#fff",
        plot_bgcolor="#fff",

        font=dict(
            family="Arial",
            color="#22223b"
        ),

        legend=dict(
            orientation="h",
            y=1.08
        )
    )

    display(fig)


def update253(*_):

    with out253:

        clear_output(wait=True)

        if view253.value=="overview":

            display(
                HTML(
                    f"""
<div class="grid253">

<div class="box253">
<b>CURRENT QUALITY POSITION</b>
{pass_gates253} of
{current_scope_gates253}
currently applicable gates pass.
</div>

<div class="box253">
<b>PRE-PACKAGING BLOCKERS</b>
{blocking_open253}
blocking controls remain open.
</div>

<div class="box253">
<b>FUTURE CONTROLS</b>
{planned_gates253}
gates are intentionally deferred to
Flask, Streamlit, screenshots and deployment.
</div>

</div>
"""
                )
            )

            status_chart253()

            if blocking_open253>0:

                blocker_names253=(
                    acceptance_criteria253[
                        acceptance_criteria253[
                            "Current_Status"
                        ].eq("OPEN")
                        &
                        acceptance_criteria253[
                            "Blocking_Now"
                        ]
                    ][
                        "Gate_ID"
                    ]
                    .tolist()
                )

                display(
                    HTML(
                        f"""
<div class="note253">
<b>Immediate closure queue:</b><br>
{" • ".join(blocker_names253)}
</div>
"""
                    )
                )


        elif view253.value=="blockers":

            blockers253=(
                acceptance_criteria253[
                    acceptance_criteria253[
                        "Blocking_Now"
                    ]
                    &
                    acceptance_criteria253[
                        "Current_Status"
                    ].eq("OPEN")
                ]
                .copy()
            )

            if blockers253.empty:

                display(
                    HTML(
                        """
<div class="note253">
<b>No current blocking gates.</b><br>
The pre-deployment acceptance layer is ready to proceed.
</div>
"""
                    )
                )

            else:

                blockers253[
                    "Priority"
                ]=np.arange(
                    1,
                    len(blockers253)+1
                )

                table253(
                    blockers253[
                        [
                            "Priority",
                            "Gate_ID",
                            "Layer",
                            "Acceptance_Criterion",
                            "Evidence_Required",
                            "PDF_Source"
                        ]
                    ]
                )


        elif view253.value=="tasks":

            table253(
                acceptance_criteria253[
                    acceptance_criteria253[
                        "Requirement_ID"
                    ].str.startswith(
                        "TASK_"
                    )
                ][
                    [
                        "Gate_ID",
                        "Requirement_ID",
                        "Layer",
                        "Acceptance_Criterion",
                        "Evidence_Required",
                        "Current_Status"
                    ]
                ]
            )


        elif view253.value=="engineering":

            table253(
                acceptance_criteria253[
                    acceptance_criteria253[
                        "Layer"
                    ].isin([
                        "Governance",
                        "Engineering",
                        "Data Engineering"
                    ])
                ]
            )


        elif view253.value=="model":

            table253(
                acceptance_criteria253[
                    acceptance_criteria253[
                        "Layer"
                    ].isin([
                        "Machine Learning",
                        "Recommendation",
                        "Model Governance",
                        "SQL / Data Layer",
                        "Forecasting"
                    ])
                ]
            )


        elif view253.value=="layers":

            layer_chart253()
            table253(layer_summary253)


        elif view253.value=="future":

            table253(
                acceptance_criteria253[
                    acceptance_criteria253[
                        "Current_Status"
                    ].eq("PLANNED")
                ][
                    [
                        "Gate_ID",
                        "Requirement_ID",
                        "Layer",
                        "Acceptance_Criterion",
                        "Evidence_Required",
                        "PDF_Source"
                    ]
                ]
            )


        else:

            files253=pd.DataFrame({
                "Artifact":[
                    "Acceptance Criteria Markdown",
                    "Acceptance Criteria CSV",
                    "Acceptance Criteria Excel",
                    "Acceptance Summary",
                    "Layer Summary"
                ],

                "Location":[
                    str(acceptance_md253),
                    str(acceptance_csv253),
                    str(acceptance_xlsx253),
                    str(
                        REPORT_DIR/
                        "acceptance_criteria_summary.csv"
                    ),
                    str(
                        REPORT_DIR/
                        "acceptance_layer_summary.csv"
                    )
                ]
            })

            files253[
                "Exists"
            ]=files253[
                "Location"
            ].map(
                lambda p:
                Path(p).exists()
            )

            files253[
                "Status"
            ]=np.where(
                files253["Exists"],
                "PASS",
                "REVIEW"
            )

            table253(files253)

            display(
                HTML(
                    f"""
<br>
<div class="note253">

Official acceptance document:<br>
<b>{acceptance_md253}</b>

<br><br>

Current gate:
<b>
{PREDEPLOYMENT_ACCEPTANCE_STATUS253}
</b>

</div>
"""
                )
            )


def reset_filters253(_):

    layer_filter253.value="All"
    status_filter253.value="All"
    search253.value=""


reset253.on_click(
    reset_filters253
)


for control253 in [
    view253,
    layer_filter253,
    status_filter253,
    search253
]:

    control253.observe(
        update253,
        names="value"
    )


update253()


dashboard_253=widgets.VBox([
    header253,
    kpi253,
    view253,
    filters253,
    out253
])

dashboard_253.add_class("d253")

display(dashboard_253)
```

```python
REPORT_DIR=Path("reports")
REPORT_DIR.mkdir(parents=True,exist_ok=True)
CLIENT_APPROVAL_STATUS254="PENDING EVIDENCE"
CLIENT_APPROVAL_REFERENCE254=""
CLIENT_APPROVAL_NOTES254=(
    "Written client approval should be retained before final submission "
    "if Docker/containerization is explicitly scored in the original rubric."
)

scope_changes254=pd.DataFrame([
    {
        "Change_ID":"SC-001",
        "Area":"Deployment Architecture",
        "Original_Requirement":
            "Docker/containerization included in the original project brief.",
        "Selected_Implementation":
            "Streamlit frontend + separately hosted Flask REST API + "
            "SQL Server/data layer + saved ML/recommender artifacts.",
        "Change_Type":"Scope substitution",
        "Reason":
            "The selected implementation prioritizes a client-facing Streamlit "
            "application backed by a separately hosted Flask REST API. "
            "Streamlit and Docker are not technical equivalents, so this is "
            "recorded as a deliberate scope substitution.",
        "Technical_Impact":
            "Deployment uses dependency locking, environment configuration, "
            "hosted services and reproducible run instructions instead of "
            "container images.",
        "Client_Impact":
            "The client receives a working frontend and backend service, but "
            "container-based isolation and portability are not delivered.",
        "Scoring_Risk":
            "HIGH if Docker/containerization is independently scored and no "
            "written approval for the substitution is available.",
        "Mitigation":
            "Preserve written approval, requirements.txt, environment template, "
            "deployment guide, hosted Flask evidence, Streamlit evidence and "
            "this formal scope-change record.",
        "Approval_Status":CLIENT_APPROVAL_STATUS254,
        "Approval_Reference":CLIENT_APPROVAL_REFERENCE254,
        "Owner":"Project Governance",
        "Decision_Status":"ACTIVE",
        "Evidence_Location":"reports/scope_change_log.md",
        "Last_Updated_UTC":str(
            pd.Timestamp.now(tz="UTC")
        )
    }
])

architecture_decision254=pd.DataFrame([
    {
        "Layer":"Client Interface",
        "Technology":"Streamlit",
        "Purpose":"Interactive client-facing analytics and recommendation application",
        "Deployment_Mode":"Separate frontend service",
        "Status":"SELECTED"
    },
    {
        "Layer":"Backend API",
        "Technology":"Flask REST API",
        "Purpose":"Validation, inference, analytics services and SQL access",
        "Deployment_Mode":"Separate backend service",
        "Status":"SELECTED"
    },
    {
        "Layer":"Database",
        "Technology":"SQL Server",
        "Purpose":"Persistent structured data and reusable application queries",
        "Deployment_Mode":"Persistent database service",
        "Status":"SELECTED"
    },
    {
        "Layer":"Demand Forecasting",
        "Technology":globals().get(
            "MODEL_VERSION",
            "Versioned saved forecasting model"
        ),
        "Purpose":"Production demand prediction",
        "Deployment_Mode":"Loaded once by backend inference service",
        "Status":"SELECTED"
    },
    {
        "Layer":"Recommendation Engine",
        "Technology":
            f"{globals().get('RECOMMENDER_VERSION','Versioned recommender')} "
            "+ TF-IDF ranking",
        "Purpose":"Explainable Top-K job recommendations",
        "Deployment_Mode":"Loaded once by backend inference service",
        "Status":"SELECTED"
    },
    {
        "Layer":"Containerization",
        "Technology":"Docker",
        "Purpose":"Containerized runtime and environment isolation",
        "Deployment_Mode":"Not included in selected architecture",
        "Status":"SUBSTITUTED"
    }
])

approval_received254=(
    CLIENT_APPROVAL_STATUS254
    .strip()
    .upper()
    in {
        "APPROVED",
        "APPROVED IN WRITING",
        "CONFIRMED"
    }
)

approval_reference_present254=bool(
    CLIENT_APPROVAL_REFERENCE254.strip()
)

if approval_received254 and approval_reference_present254:
    approval_gate254="PASS"

elif approval_received254:
    approval_gate254="REVIEW"

else:
    approval_gate254="OPEN"

scope_controls254=pd.DataFrame([
    {
        "Control":"Scope substitution formally documented",
        "Status":"PASS",
        "Blocking_Now":False,
        "Evidence":"reports/scope_change_log.md"
    },
    {
        "Control":"Original Docker requirement retained in record",
        "Status":"PASS",
        "Blocking_Now":False,
        "Evidence":"SC-001"
    },
    {
        "Control":"Selected replacement architecture explicitly documented",
        "Status":"PASS",
        "Blocking_Now":False,
        "Evidence":"deployment_architecture_decision.csv"
    },
    {
        "Control":"Technical and scoring risk disclosed",
        "Status":"PASS",
        "Blocking_Now":False,
        "Evidence":"SC-001 risk + mitigation fields"
    },
    {
        "Control":"Written client approval available",
        "Status":approval_gate254,
        "Blocking_Now":True,
        "Evidence":
            CLIENT_APPROVAL_REFERENCE254
            if approval_reference_present254
            else "Approval evidence not yet referenced"
    }
])

scope_documentation_status254=(
    "PASS"
    if (
        scope_controls254[
            "Status"
        ].isin([
            "PASS",
            "OPEN",
            "REVIEW"
        ]).all()
    )
    else "REVIEW"
)

scope_approval_status254=(
    "PASS"
    if approval_gate254=="PASS"
    else "OPEN"
)

scope_csv254=REPORT_DIR/"scope_change_log.csv"
architecture_csv254=REPORT_DIR/"deployment_architecture_decision.csv"
scope_controls_csv254=REPORT_DIR/"scope_change_controls.csv"

scope_changes254.to_csv(
    scope_csv254,
    index=False
)

architecture_decision254.to_csv(
    architecture_csv254,
    index=False
)

scope_controls254.to_csv(
    scope_controls_csv254,
    index=False
)

scope_xlsx254=REPORT_DIR/"scope_change_log.xlsx"

try:
    with pd.ExcelWriter(
        scope_xlsx254,
        engine="openpyxl"
    ) as writer:
        scope_changes254.to_excel(
            writer,
            sheet_name="Scope Changes",
            index=False
        )
        architecture_decision254.to_excel(
            writer,
            sheet_name="Architecture",
            index=False
        )
        scope_controls254.to_excel(
            writer,
            sheet_name="Controls",
            index=False
        )

except Exception as e:

    print(
        "Excel export warning:",
        type(e).__name__,
        str(e)
    )

scope_md254=REPORT_DIR/"scope_change_log.md"
scope_md_text254=f"""# Scope Change Log
Generated: {pd.Timestamp.now(tz="UTC")}
## SC-001 — Deployment Architecture Substitution
### Original requirement
Docker/containerization was included in the original project brief.
### Selected implementation
Streamlit frontend + separately hosted Flask REST API + SQL Server/data layer + saved machine-learning and recommendation artifacts.
### Change classification
Scope substitution.
### Rationale
The selected implementation prioritizes a client-facing Streamlit application backed by a separately hosted Flask REST API.
Streamlit and Docker are not technical equivalents. Therefore, this implementation must not be represented as having fulfilled the Docker/containerization requirement.
### Technical impact
The deployment package uses:
- dependency locking
- environment configuration
- saved model artifacts
- reproducible runtime instructions
- separately hosted Flask backend
- separately deployed Streamlit frontend
- persistent SQL/data layer
Container images are not part of the selected implementation.
### Client impact
The client receives a functional deployed frontend and backend service. Container-based runtime isolation and portability are not included.
### Scoring risk
HIGH if Docker/containerization is independently scored in the original rubric and written approval for the substitution cannot be demonstrated.
### Mitigation
- Preserve written client approval for the substitution.
- Include this scope-change log in final handover.
- Provide requirements.txt.
- Provide .env.example.
- Provide reproducible run instructions.
- Provide deployment guide.
- Preserve hosted Flask API evidence.
- Preserve Streamlit deployment evidence.
- Preserve final screenshots and evidence index.

## Approval

Status: {CLIENT_APPROVAL_STATUS254}

Reference: {CLIENT_APPROVAL_REFERENCE254 or "Not yet supplied"}

Notes: {CLIENT_APPROVAL_NOTES254}

## Final architecture

Streamlit Frontend
→ Flask REST API
→ Production Inference + Application Services
→ SQL Server / Saved Model Assets

## Governance rule

The Docker requirement remains visible in the project record even though it is not part of the selected implementation. It must not be silently marked as completed without approval evidence.
"""
scope_md254.write_text(
    scope_md_text254,
    encoding="utf-8"
)

scope_summary254=pd.DataFrame({
    "Metric":[
        "Scope Changes",
        "Architecture Layers",
        "Documented Controls",
        "Controls Passed",
        "Approval Status",
        "Approval Evidence",
        "Documentation Status"
    ],
    "Value":[
        len(scope_changes254),
        len(architecture_decision254),
        len(scope_controls254),
        int(
            scope_controls254[
                "Status"
            ].eq("PASS").sum()
        ),
        CLIENT_APPROVAL_STATUS254,
        (
            CLIENT_APPROVAL_REFERENCE254
            if approval_reference_present254
            else "Not supplied"
        ),
        scope_documentation_status254
    ]
})

scope_summary254.to_csv(
    REPORT_DIR/"scope_change_summary.csv",
    index=False
)

risk_register254=pd.DataFrame([
    {
        "Risk_ID":"R-SCOPE-01",
        "Risk":
            "Docker/containerization may remain an explicitly scored requirement.",
        "Likelihood":"MEDIUM",
        "Impact":"HIGH",
        "Risk_Level":"HIGH",
        "Mitigation":
            "Retain written approval for the Streamlit + Flask substitution.",
        "Owner":"Project Governance",
        "Status":
            "OPEN"
            if approval_gate254!="PASS"
            else "CONTROLLED"
    },
    {
        "Risk_ID":"R-SCOPE-02",
        "Risk":
            "Reviewer may assume Streamlit deployment is being presented as equivalent to Docker.",
        "Likelihood":"LOW",
        "Impact":"HIGH",
        "Risk_Level":"MEDIUM",
        "Mitigation":
            "Explicitly state that the technologies are not equivalent and preserve the scope-change record.",
        "Owner":"Project Governance",
        "Status":"CONTROLLED"
    },
    {
        "Risk_ID":"R-SCOPE-03",
        "Risk":
            "Separate Flask and Streamlit services may develop configuration mismatch.",
        "Likelihood":"MEDIUM",
        "Impact":"MEDIUM",
        "Risk_Level":"MEDIUM",
        "Mitigation":
            "Centralize environment configuration and run end-to-end deployment validation.",
        "Owner":"Deployment",
        "Status":"CONTROLLED"
    }
])

risk_register254.to_csv(
    REPORT_DIR/"scope_change_risk_register.csv",
    index=False
)

def table254(df,rows=40):
    display(
        HTML(
            df.head(rows).to_html(
                index=False,
                border=0,
                classes="t254")))

def card254(
    label,
    value,
    note,
    accent="#4a4e69"
):
    return f"""
    <div class="c254"
    style="border-top:4px solid {accent}">
    <small>{label}</small>
    <b>{value}</b>
    <span>{note}</span>
    </div>
    """

display(
    HTML("""
<style>
.d254{
border:1px solid #ded8d5;
border-radius:18px;
overflow:hidden;
background:#fff;
box-shadow:0 9px 25px rgba(34,34,59,.10)
}

.h254{
background:linear-gradient(
120deg,#22223b,#30324d 55%,#4a4e69
);
padding:25px 28px;
display:flex;
justify-content:space-between;
align-items:flex-start;
gap:20px;
font-family:Arial
}

.h254 .k{
color:#c9ada7;
font-size:8px;
font-weight:700;
letter-spacing:1.7px
}

.h254 h2{
color:#fff;
margin:6px 0;
font-size:25px
}

.h254 p,
.h254 .m{
color:#f2e9e4;
font-size:10px;
line-height:1.65
}

.cards254{
display:grid;
grid-template-columns:repeat(6,1fr);
border-bottom:1px solid #e8e2df
}

.c254{
padding:14px;
border-right:1px solid #ebe6e4;
font-family:Arial;
min-width:0
}

.c254 small{
display:block;
color:#9a8c98;
font-size:7px;
font-weight:700
}

.c254 b{
display:block;
color:#22223b;
font-size:16px;
margin-top:5px
}

.c254 span{
display:block;
color:#4a4e69;
font-size:8px;
margin-top:4px
}

.out254{
padding:19px 22px;
border-top:1px solid #eee9e7;
min-height:230px
}

.t254{
width:100%;
border-collapse:collapse;
font:9px Arial
}

.t254 th{
background:#22223b;
color:#fff;
padding:9px;
text-align:left
}

.t254 td{
padding:8px;
border-bottom:1px solid #eee9e7;
vertical-align:top;
color:#22223b;
line-height:1.5
}

.t254 tr:nth-child(even){
background:#fbfaf9
}

.grid254{
display:grid;
grid-template-columns:repeat(3,1fr);
gap:12px
}

.box254{
background:#f8f6f5;
border-left:4px solid #4a4e69;
padding:14px;
border-radius:9px;
font:11px/1.65 Arial;
color:#22223b
}

.box254 b{
display:block;
color:#4a4e69;
font-size:9px;
margin-bottom:5px
}

.note254{
background:#f2e9e4;
padding:12px 14px;
border-radius:8px;
font:10px/1.65 Arial;
color:#4a4e69
}

.flow254{
padding:20px;
background:#faf8f7;
border:1px solid #eee9e7;
border-radius:12px;
font-family:Arial
}

.flowrow254{
display:flex;
align-items:center;
justify-content:center;
gap:10px;
flex-wrap:wrap
}

.node254{
padding:11px 14px;
background:#fff;
border:1px solid #c9ada7;
border-radius:9px;
font-size:10px;
font-weight:700;
color:#22223b
}

.arrow254{
font-size:18px;
color:#9a8c98
}

</style>
""")
)

header254=widgets.HTML(
    f"""
<div class="h254">

<div>

<div class="k">
SECTION 25.2 • SCOPE GOVERNANCE
</div>

<h2>
Deployment Architecture Decision Register
</h2>

<p>
Original Scope • Selected Architecture •
Risk • Mitigation • Approval • Evidence
</p>

</div>

<div class="m">

DOCUMENTATION<br>
<b>{scope_documentation_status254}</b>

<br><br>

CLIENT APPROVAL<br>
<b>{CLIENT_APPROVAL_STATUS254}</b>

</div>
</div>""")

kpi254=widgets.HTML(
    '<div class="cards254">'
    +
    "".join([
        card254(
            "SCOPE CHANGES",
            len(scope_changes254),
            "Formal decisions",
            "#4a4e69"
        ),
        card254(
            "ARCHITECTURE",
            "2-SERVICE",
            "Streamlit + Flask",
            "#4a4e69"
        ),
        card254(
            "DOCKER",
            "SUBSTITUTED",
            "Not equivalent",
            "#c9ada7"
        ),
        card254(
            "RISK",
            "HIGH",
            "Without approval",
            "#c9ada7"
        ),
        card254(
            "APPROVAL",
            approval_gate254,
            "Written evidence",
            "#9a8c98"
        ),

        card254(
            "DOCUMENT",
            "SAVED",
            "Scope evidence",
            "#4a4e69"
        )
    ])
    +
    "</div>"
)

view254=widgets.ToggleButtons(
    options=[
        ("EXECUTIVE","overview"),
        ("CHANGE LOG","change"),
        ("ARCHITECTURE","architecture"),
        ("RISKS","risks"),
        ("CONTROLS","controls"),
        ("APPROVAL","approval"),
        ("FILES","files")
    ],

    value="overview",

    layout=widgets.Layout(
        width="100%"
    ),

    style={
        "button_width":"14.2%",
        "font_weight":"700"
    }
)

out254=widgets.Output()
out254.add_class("out254")

def risk_chart254():
    risk_counts254=(
        risk_register254[
            "Risk_Level"
        ]
        .value_counts()
        .reindex(
            [
                "HIGH",
                "MEDIUM",
                "LOW"
            ],
            fill_value=0
        )
    )

    fig=go.Figure(
        go.Bar(
            x=risk_counts254.index,
            y=risk_counts254.values,
            text=risk_counts254.values,
            textposition="auto",
            marker_color=[
                "#c9ada7",
                "#9a8c98",
                "#4a4e69"]))

    fig.update_layout(
        title="Scope Change Risk Profile",
        height=330,
        xaxis_title="Risk Level",
        yaxis_title="Number of Risks",
        paper_bgcolor="#fff",
        plot_bgcolor="#fff",
        font=dict(
            family="Arial",
            color="#22223b"
        ),
        margin=dict(
            l=40,
            r=20,
            t=55,
            b=40
        )
    )

    display(fig)

def update254(*_):
    with out254:
        clear_output(wait=True)
        if view254.value=="overview":
            display(
                HTML(
                    """
<div class="grid254">
<div class="box254">
<b>WHAT CHANGED</b>
Docker/containerization remains part of the original requirement record, but is not part of the selected implementation.
</div>

<div class="box254">
<b>WHAT REPLACES IT</b>
A separately deployed Streamlit frontend communicates with a separately hosted Flask REST API.
</div>

<div class="box254">
<b>WHAT STILL MATTERS</b>
Written approval should be retained if Docker/containerization receives separate scoring.
</div>

</div>

<br>

<div class="flow254">

<div class="flowrow254">

<div class="node254">Streamlit Frontend</div>
<div class="arrow254">→</div>

<div class="node254">Flask REST API</div>
<div class="arrow254">→</div>

<div class="node254">Inference + Services</div>
<div class="arrow254">→</div>

<div class="node254">SQL Server + Models</div>

</div>

</div>
"""
                )
            )

        elif view254.value=="change":
            table254(scope_changes254)

        elif view254.value=="architecture":
            table254(architecture_decision254)
            display(
                HTML(
                    """
<br>
<div class="note254">
<b>Architecture principle:</b>
Streamlit is the presentation/client layer. Flask owns reusable application and inference services. SQL remains the persistent data layer.
</div>
"""
                )
            )

        elif view254.value=="risks":
            risk_chart254()
            table254(risk_register254)

        elif view254.value=="controls":
            table254(scope_controls254)

        elif view254.value=="approval":
            display(
                HTML(
                    f"""
<div class="grid254">

<div class="box254">
<b>APPROVAL STATUS</b>
{CLIENT_APPROVAL_STATUS254}
</div>

<div class="box254">
<b>REFERENCE</b>
{
CLIENT_APPROVAL_REFERENCE254
if approval_reference_present254
else "No written evidence referenced yet"
}
</div>
<div class="box254">
<b>APPROVAL GATE</b>
{approval_gate254}
</div>
</div>
<br>
<div class="note254">
<b>Important:</b>
The notebook intentionally does not mark client approval as PASS unless explicit evidence is available.
</div>
"""))

        else:
            files254=pd.DataFrame({
                "Artifact":[
                    "Scope Change Markdown",
                    "Scope Change CSV",
                    "Scope Change Excel",
                    "Architecture Decision",
                    "Scope Controls",
                    "Scope Risk Register",
                    "Scope Summary"
                ],
                "Location":[
                    str(scope_md254),
                    str(scope_csv254),
                    str(scope_xlsx254),
                    str(architecture_csv254),
                    str(scope_controls_csv254),
                    str(REPORT_DIR/"scope_change_risk_register.csv"),
                    str(REPORT_DIR/"scope_change_summary.csv")]})

            files254[
                "Exists"
            ]=files254[
                "Location"
            ].map(
                lambda x:
                Path(x).exists()
            )
            files254[
                "Status"
            ]=np.where(
                files254["Exists"],
                "PASS",
                "REVIEW"
            )
            table254(files254)

view254.observe(
    update254,
    names="value"
)

update254()

dashboard_254=widgets.VBox([
    header254,
    kpi254,
    view254,
    out254
])

dashboard_254.add_class("d254")
display(dashboard_254)
```

```python
SRC_DIR=Path("src")
LOG_DIR=Path("logs")
SRC_DIR.mkdir(parents=True,exist_ok=True)
LOG_DIR.mkdir(parents=True,exist_ok=True)
LOGGER_PATH255=SRC_DIR/"logger.py"
APPLICATION_LOG255=LOG_DIR/"application.log"

logger_code255=r'''
import logging
from logging.handlers import RotatingFileHandler
from pathlib import Path

BASE_DIR=Path(__file__).resolve().parents[1]
LOG_DIR=BASE_DIR/"logs"
LOG_FILE=LOG_DIR/"application.log"

LOG_DIR.mkdir(parents=True,exist_ok=True)

FILE_FORMAT=(
    "%(asctime)s | %(levelname)s | "
    "%(name)s | %(module)s:%(lineno)d | %(message)s"
)

CONSOLE_FORMAT=(
    "%(levelname)s | %(name)s | %(message)s"
)

class MaxLevelFilter(logging.Filter):

    def __init__(self,max_level):
        super().__init__()
        self.max_level=max_level

    def filter(self,record):
        return record.levelno<=self.max_level

def get_logger(
    name="job_market",
    file_level=logging.DEBUG,
    console_level=logging.WARNING,
    max_bytes=5_000_000,
    backup_count=3
):

    logger=logging.getLogger(name)

    logger.setLevel(logging.DEBUG)
    logger.propagate=False

    if getattr(
        logger,
        "_jobmarket_configured",
        False
    ):
        return logger

    file_handler=RotatingFileHandler(
        LOG_FILE,
        maxBytes=max_bytes,
        backupCount=backup_count,
        encoding="utf-8"
    )

    file_handler.setLevel(file_level)

    file_handler.setFormatter(
        logging.Formatter(FILE_FORMAT)
    )


    console_handler=logging.StreamHandler()

    console_handler.setLevel(
        console_level
    )

    console_handler.setFormatter(
        logging.Formatter(
            CONSOLE_FORMAT
        )
    )

    logger.addHandler(
        file_handler
    )

    logger.addHandler(
        console_handler
    )

    logger._jobmarket_configured=True

    return logger

def log_system_event(
    logger,
    event,
    **details
):

    detail_text=" | ".join(
        f"{key}={value}"
        for key,value in details.items()
    )

    message=(
        event
        if not detail_text
        else f"{event} | {detail_text}"
    )

    logger.info(message)


def log_exception(
    logger,
    event,
    exc,
    **details
):

    detail_text=" | ".join(
        f"{key}={value}"
        for key,value in details.items()
    )

    message=(
        f"{event} | "
        f"error_type={type(exc).__name__} | "
        f"error={exc}"
    )

    if detail_text:
        message+=f" | {detail_text}"

    logger.exception(message)
'''


LOGGER_PATH255.write_text(
    logger_code255,
    encoding="utf-8"
)


init_path255=SRC_DIR/"__init__.py"

if not init_path255.exists():
    init_path255.write_text(
        "",
        encoding="utf-8"
    )


import importlib
import src.logger as logger_module255

importlib.reload(
    logger_module255
)

from src.logger import (
    get_logger,
    log_system_event,
    log_exception
)


logger=get_logger(
    "job_market"
)


logger.info(
    "LOGGER_INITIALIZED | module=src.logger"
)

logger.debug(
    "LOGGER_DEBUG_TEST | status=PASS"
)

logger.warning(
    "LOGGER_WARNING_TEST | controlled_test=True"
)


log_system_event(
    logger,
    "LOGGER_STRUCTURED_EVENT_TEST",
    section="25.2",
    component="logger",
    status="PASS"
)


logger_file_exists255=LOGGER_PATH255.exists()
log_file_exists255=APPLICATION_LOG255.exists()


logger_handlers255=pd.DataFrame([
    {
        "Handler":
            type(handler).__name__,

        "Level":
            logging.getLevelName(
                handler.level
            ),

        "Destination":
            (
                str(
                    getattr(
                        handler,
                        "baseFilename",
                        "Console"
                    )
                )
            ),

        "Formatter":
            (
                handler.formatter._fmt
                if handler.formatter
                else ""
            )
    }

    for handler in logger.handlers
])


file_handler_count255=sum(
    isinstance(
        h,
        logging.handlers.RotatingFileHandler
    )
    for h in logger.handlers
)


console_handler_count255=sum(
    (
        isinstance(
            h,
            logging.StreamHandler
        )
        and
        not isinstance(
            h,
            logging.handlers.RotatingFileHandler
        )
    )
    for h in logger.handlers
)


duplicate_handler_test255=(
    get_logger(
        "job_market"
    )
    is logger
    and
    len(
        get_logger(
            "job_market"
        ).handlers
    )
    ==
    len(
        logger.handlers
    )
)


log_content255=""

if APPLICATION_LOG255.exists():

    try:

        log_content255=APPLICATION_LOG255.read_text(
            encoding="utf-8",
            errors="ignore"
        )

    except Exception:

        log_content255=""


info_written255=(
    "LOGGER_INITIALIZED"
    in log_content255
)

debug_written255=(
    "LOGGER_DEBUG_TEST"
    in log_content255
)

structured_written255=(
    "LOGGER_STRUCTURED_EVENT_TEST"
    in log_content255
)


logger_tests255=pd.DataFrame([
    {
        "Test":
            "Logger module created",

        "Status":
            "PASS"
            if logger_file_exists255
            else "FAIL",

        "Evidence":
            str(LOGGER_PATH255)
    },

    {
        "Test":
            "Application log created",

        "Status":
            "PASS"
            if log_file_exists255
            else "FAIL",

        "Evidence":
            str(APPLICATION_LOG255)
    },

    {
        "Test":
            "Rotating file handler active",

        "Status":
            "PASS"
            if file_handler_count255==1
            else "FAIL",

        "Evidence":
            f"handlers={file_handler_count255}"
    },

    {
        "Test":
            "Console handler active",

        "Status":
            "PASS"
            if console_handler_count255==1
            else "FAIL",

        "Evidence":
            f"handlers={console_handler_count255}"
    },

    {
        "Test":
            "INFO event written to file",

        "Status":
            "PASS"
            if info_written255
            else "FAIL",

        "Evidence":
            "LOGGER_INITIALIZED"
    },

    {
        "Test":
            "DEBUG event written to file",

        "Status":
            "PASS"
            if debug_written255
            else "FAIL",

        "Evidence":
            "LOGGER_DEBUG_TEST"
    },

    {
        "Test":
            "Structured event written",

        "Status":
            "PASS"
            if structured_written255
            else "FAIL",

        "Evidence":
            "LOGGER_STRUCTURED_EVENT_TEST"
    },

    {
        "Test":
            "Duplicate-handler protection",

        "Status":
            "PASS"
            if duplicate_handler_test255
            else "FAIL",

        "Evidence":
            f"active_handlers={len(logger.handlers)}"
    }
])


LOGGER_STATUS255=(
    "PASS"
    if (
        logger_tests255[
            "Status"
        ]=="PASS"
    ).all()
    else "REVIEW"
)


logger_tests255.to_csv(
    REPORT_DIR/
    "logger_validation.csv",
    index=False
)


logger_handlers255.to_csv(
    REPORT_DIR/
    "logger_handler_configuration.csv",
    index=False
)


logger_metadata255=pd.DataFrame([
    {
        "Component":
            "Centralized Logger",

        "Module":
            str(LOGGER_PATH255),

        "Log_File":
            str(APPLICATION_LOG255),

        "File_Level":
            "INFO",

        "Console_Level":
            "WARNING",

        "Rotation_Max_Bytes":
            5_000_000,

        "Backup_Count":
            3,

        "Duplicate_Handler_Protection":
            duplicate_handler_test255,

        "Validation_Status":
            LOGGER_STATUS255
    }
])


logger_metadata255.to_csv(
    REPORT_DIR/
    "logger_metadata.csv",
    index=False
)


def recent_log_lines255(
    n=25
):

    if not APPLICATION_LOG255.exists():
        return []

    try:

        lines=APPLICATION_LOG255.read_text(
            encoding="utf-8",
            errors="ignore"
        ).splitlines()

        return lines[-n:]

    except Exception:

        return []


def table255(df,rows=50):

    display(
        HTML(
            df.head(rows).to_html(
                index=False,
                border=0,
                classes="t255"
            )
        )
    )


def card255(
    label,
    value,
    note,
    accent="#4a4e69"
):

    return f"""
    <div class="c255"
    style="border-top:4px solid {accent}">
        <small>{label}</small>
        <b>{value}</b>
        <span>{note}</span>
    </div>
    """


display(
    HTML("""
<style>

.d255{
border:1px solid #ded8d5;
border-radius:18px;
overflow:hidden;
background:#fff;
box-shadow:0 9px 25px rgba(34,34,59,.10)
}

.h255{
background:linear-gradient(
120deg,#22223b,#30324d 55%,#4a4e69
);
padding:25px 28px;
display:flex;
justify-content:space-between;
align-items:flex-start;
gap:20px;
font-family:Arial
}

.h255 .k{
color:#c9ada7;
font-size:8px;
font-weight:700;
letter-spacing:1.7px
}

.h255 h2{
color:#fff;
margin:6px 0;
font-size:25px
}

.h255 p,
.h255 .m{
color:#f2e9e4;
font-size:10px;
line-height:1.65
}

.cards255{
display:grid;
grid-template-columns:repeat(6,1fr);
border-bottom:1px solid #e8e2df
}

.c255{
padding:14px;
border-right:1px solid #ebe6e4;
font-family:Arial;
min-width:0
}

.c255 small{
display:block;
color:#9a8c98;
font-size:7px;
font-weight:700
}

.c255 b{
display:block;
color:#22223b;
font-size:16px;
margin-top:5px
}

.c255 span{
display:block;
color:#4a4e69;
font-size:8px;
margin-top:4px
}

.out255{
padding:19px 22px;
border-top:1px solid #eee9e7;
min-height:230px
}

.t255{
width:100%;
border-collapse:collapse;
font:9px Arial
}

.t255 th{
background:#22223b;
color:#fff;
padding:9px;
text-align:left
}

.t255 td{
padding:8px;
border-bottom:1px solid #eee9e7;
vertical-align:top;
color:#22223b;
line-height:1.5
}

.t255 tr:nth-child(even){
background:#fbfaf9
}

.grid255{
display:grid;
grid-template-columns:repeat(3,1fr);
gap:12px
}

.box255{
background:#f8f6f5;
border-left:4px solid #4a4e69;
padding:14px;
border-radius:9px;
font:11px/1.65 Arial;
color:#22223b
}

.box255 b{
display:block;
color:#4a4e69;
font-size:9px;
margin-bottom:5px
}

.note255{
background:#f2e9e4;
padding:12px 14px;
border-radius:8px;
font:10px/1.65 Arial;
color:#4a4e69
}

.log255{
background:#22223b;
color:#f2e9e4;
padding:14px;
border-radius:9px;
font:9px/1.6 Consolas,monospace;
white-space:pre-wrap;
overflow-x:auto
}

</style>
""")
)


passed_tests255=int(
    logger_tests255[
        "Status"
    ].eq("PASS").sum()
)


header255=widgets.HTML(
    f"""
<div class="h255">

<div>

<div class="k">
SECTION 25.2 • ENGINEERING CONTROL
</div>

<h2>
Centralized Application Logger
</h2>

<p>
Rotating File Logs • Controlled Console Output •
Structured Events • Exception Logging • Diagnostics
</p>

</div>

<div class="m">

LOGGER STATUS<br>
<b>{LOGGER_STATUS255}</b>

<br><br>

VALIDATION<br>
<b>
{passed_tests255}/{len(logger_tests255)}
</b>

</div>

</div>
"""
)


kpi255=widgets.HTML(
    '<div class="cards255">'
    +
    "".join([
        card255(
            "STATUS",
            LOGGER_STATUS255,
            "Engineering gate",
            "#4a4e69"
        ),

        card255(
            "FILE LEVEL",
            "INFO",
            "Persistent logging",
            "#4a4e69"
        ),

        card255(
            "CONSOLE",
            "WARNING+",
            "Reduced notebook noise",
            "#9a8c98"
        ),

        card255(
            "ROTATION",
            "5 MB",
            "3 backups",
            "#c9ada7"
        ),

        card255(
            "HANDLERS",
            len(logger.handlers),
            "Active handlers",
            "#4a4e69"
        ),

        card255(
            "TESTS",
            f"{passed_tests255}/{len(logger_tests255)}",
            "Validation passed",
            "#4a4e69"
        )
    ])
    +
    "</div>"
)


view255=widgets.ToggleButtons(
    options=[
        ("OVERVIEW","overview"),
        ("VALIDATION","validation"),
        ("HANDLERS","handlers"),
        ("LIVE LOG","log"),
        ("MODULE","module"),
        ("FILES","files")
    ],

    value="overview",

    layout=widgets.Layout(
        width="100%"
    ),

    style={
        "button_width":"16.6%",
        "font_weight":"700"
    }
)


refresh255=widgets.Button(
    description="Refresh Log",
    icon="refresh"
)


test_event255=widgets.Button(
    description="Write Test Event",
    icon="check"
)


controls255=widgets.HBox([
    refresh255,
    test_event255
])


out255=widgets.Output()
out255.add_class("out255")


def update255(*_):

    with out255:

        clear_output(wait=True)

        if view255.value=="overview":

            display(
                HTML(
                    f"""
<div class="grid255">

<div class="box255">
<b>FILE LOGGING</b>
INFO, WARNING and ERROR events are preserved in
<code>logs/application.log</code>.
</div>

<div class="box255">
<b>CONSOLE CONTROL</b>
Only WARNING and ERROR messages are sent to the console, keeping notebook and production output cleaner.
</div>

<div class="box255">
<b>REUSABLE SERVICE</b>
Notebook, Flask, inference and database modules can all call the same <code>get_logger()</code> function.
</div>

</div>

<br>

<div class="note255">
<b>Current logger:</b> job_market<br>
<b>Persistent file:</b> {APPLICATION_LOG255}<br>
<b>Rotation:</b> 5 MB per file with 3 backup files<br>
<b>Duplicate handlers:</b> protected
</div>
"""
                )
            )


        elif view255.value=="validation":

            table255(
                logger_tests255
            )


        elif view255.value=="handlers":

            table255(
                logger_handlers255
            )


        elif view255.value=="log":

            lines255=recent_log_lines255(
                30
            )

            content255=(
                "\n".join(lines255)
                if lines255
                else "No log entries available."
            )

            display(
                HTML(
                    f"""
<div class="note255">
Showing the most recent application log entries.
</div>

<br>

<div class="log255">
{html.escape(content255)}
</div>
"""
                )
            )


        elif view255.value=="module":

            module_preview255=LOGGER_PATH255.read_text(
                encoding="utf-8",
                errors="ignore"
            )

            display(
                HTML(
                    f"""
<div class="note255">
Reusable source module:
<b>{LOGGER_PATH255}</b>
</div>

<br>

<div class="log255">
{html.escape(module_preview255)}
</div>
"""
                )
            )


        else:

            files255=pd.DataFrame({
                "Artifact":[
                    "Logger Module",
                    "Application Log",
                    "Validation Report",
                    "Handler Configuration",
                    "Logger Metadata"
                ],

                "Location":[
                    str(LOGGER_PATH255),
                    str(APPLICATION_LOG255),
                    str(
                        REPORT_DIR/
                        "logger_validation.csv"
                    ),
                    str(
                        REPORT_DIR/
                        "logger_handler_configuration.csv"
                    ),
                    str(
                        REPORT_DIR/
                        "logger_metadata.csv"
                    )]})

            files255[
                "Exists"
            ]=files255[
                "Location"
            ].map(
                lambda p:
                Path(p).exists()
            )

            files255[
                "Status"
            ]=np.where(
                files255["Exists"],
                "PASS",
                "REVIEW"
            )
            table255(files255)

def write_test_event255(_):
    log_system_event(
        logger,
        "MANUAL_LOGGER_TEST",
        source="interactive_dashboard",
        status="PASS"
    )
    update255()

def refresh_log255(_):
    update255()

test_event255.on_click(
    write_test_event255
)

refresh255.on_click(
    refresh_log255
)

view255.observe(
    update255,
    names="value"
)

update255()

dashboard_255=widgets.VBox([
    header255,
    kpi255,
    view255,
    controls255,
    out255
])

dashboard_255.add_class("d255")

display(dashboard_255)
```

```python
SRC_DIR=Path("src")
REPORT_DIR=Path("reports")
SRC_DIR.mkdir(parents=True,exist_ok=True)
REPORT_DIR.mkdir(parents=True,exist_ok=True)
EXCEPTION_PATH256=SRC_DIR/"exception.py"

exception_code256=r'''
from datetime import datetime, timezone

class JobMarketError(Exception):
    error_code="JOB_MARKET_ERROR"
    http_status=500
    user_message="An unexpected application error occurred."
    def __init__(
        self,
        message=None,
        *,
        user_message=None,
        details=None,
        original_exception=None
    ):

        self.message=message or self.user_message
        self.safe_message=(
            user_message
            or
            self.user_message
        )
        self.details=(
            details
            if isinstance(details,dict)
            else {})

        self.original_exception=original_exception
        self.timestamp_utc=datetime.now(
            timezone.utc
        ).isoformat()
        super().__init__(
            self.message
        )

    def to_dict(
        self,
        include_internal=False
    ):
        payload={
            "success":False,
            "error":{
                "code":self.error_code,
                "message":self.safe_message,
                "status":self.http_status}}

        if include_internal:
            payload["error"][
                "internal_message"
            ]=self.message

            payload["error"][
                "details"
            ]=self.details

            payload["error"][
                "timestamp_utc"
            ]=self.timestamp_utc

            if self.original_exception is not None:

                payload["error"][
                    "original_exception"
                ]=type(
                    self.original_exception
                ).__name__

        return payload


class ConfigurationError(JobMarketError):

    error_code="CONFIGURATION_ERROR"
    http_status=500
    user_message="Application configuration is incomplete or invalid."


class DataValidationError(JobMarketError):

    error_code="DATA_VALIDATION_ERROR"
    http_status=400
    user_message="The supplied data failed validation."


class PreprocessingError(JobMarketError):

    error_code="PREPROCESSING_ERROR"
    http_status=422
    user_message="The input could not be processed."


class ModelLoadingError(JobMarketError):

    error_code="MODEL_LOADING_ERROR"
    http_status=503
    user_message="The requested model is temporarily unavailable."


class InferenceError(JobMarketError):

    error_code="INFERENCE_ERROR"
    http_status=500
    user_message="The prediction could not be completed."


class RecommendationError(JobMarketError):

    error_code="RECOMMENDATION_ERROR"
    http_status=500
    user_message="Recommendations could not be generated."


class DatabaseError(JobMarketError):

    error_code="DATABASE_ERROR"
    http_status=503
    user_message="The data service is temporarily unavailable."


class ResourceNotFoundError(JobMarketError):

    error_code="RESOURCE_NOT_FOUND"
    http_status=404
    user_message="The requested resource was not found."


class InvalidRequestError(JobMarketError):

    error_code="INVALID_REQUEST"
    http_status=400
    user_message="The request contains invalid or missing information."


class ServiceUnavailableError(JobMarketError):

    error_code="SERVICE_UNAVAILABLE"
    http_status=503
    user_message="The requested service is temporarily unavailable."


def exception_response(
    exc,
    include_internal=False
):

    if isinstance(
        exc,
        JobMarketError
    ):

        return (
            exc.to_dict(
                include_internal=
                    include_internal
            ),
            exc.http_status
        )

    fallback=JobMarketError(
        message=str(exc),
        original_exception=exc
    )

    return (
        fallback.to_dict(
            include_internal=
                include_internal
        ),
        500
    )
'''


EXCEPTION_PATH256.write_text(
    exception_code256,
    encoding="utf-8"
)


import importlib
import src.exception as exception_module256

importlib.reload(
    exception_module256
)

from src.exception import (
    JobMarketError,
    ConfigurationError,
    DataValidationError,
    PreprocessingError,
    ModelLoadingError,
    InferenceError,
    RecommendationError,
    DatabaseError,
    ResourceNotFoundError,
    InvalidRequestError,
    ServiceUnavailableError,
    exception_response
)


exception_classes256=[
    ConfigurationError,
    DataValidationError,
    PreprocessingError,
    ModelLoadingError,
    InferenceError,
    RecommendationError,
    DatabaseError,
    ResourceNotFoundError,
    InvalidRequestError,
    ServiceUnavailableError
]


exception_registry256=pd.DataFrame([
    {
        "Exception":
            cls.__name__,

        "Error_Code":
            cls.error_code,

        "HTTP_Status":
            cls.http_status,

        "User_Message":
            cls.user_message,

        "Intended_Use":{
            "ConfigurationError":
                "Missing or invalid environment/configuration",

            "DataValidationError":
                "Schema, field or logical validation",

            "PreprocessingError":
                "Input normalization or transformation failure",

            "ModelLoadingError":
                "Missing/corrupt/unavailable model artifact",

            "InferenceError":
                "Forecast or production inference failure",

            "RecommendationError":
                "Recommendation engine failure",

            "DatabaseError":
                "SQL/database connection or query failure",

            "ResourceNotFoundError":
                "Unknown job family/resource",

            "InvalidRequestError":
                "Invalid API/user request",

            "ServiceUnavailableError":
                "Dependency or downstream service unavailable"
        }.get(
            cls.__name__,
            ""
        )
    }

    for cls in exception_classes256
])


exception_registry256.to_csv(
    REPORT_DIR/
    "exception_registry.csv",
    index=False
)


test_error256=DataValidationError(
    message=
        "Internal validation test: "
        "required field 'title' was missing.",

    user_message=
        "The supplied job data is incomplete.",

    details={
        "field":"title",
        "test":"controlled"
    }
)


safe_payload256,status_code256=(
    exception_response(
        test_error256,
        include_internal=False
    )
)


internal_payload256,_=(
    exception_response(
        test_error256,
        include_internal=True
    )
)


unexpected_test256=ValueError(
    "Internal controlled ValueError"
)

unexpected_payload256,unexpected_status256=(
    exception_response(
        unexpected_test256,
        include_internal=False
    )
)


exception_tests256=pd.DataFrame([
    {
        "Test":
            "Exception module created",

        "Status":
            "PASS"
            if EXCEPTION_PATH256.exists()
            else "FAIL",

        "Evidence":
            str(EXCEPTION_PATH256)
    },

    {
        "Test":
            "Base exception available",

        "Status":
            "PASS"
            if issubclass(
                DataValidationError,
                JobMarketError
            )
            else "FAIL",

        "Evidence":
            "DataValidationError → JobMarketError"
    },

    {
        "Test":
            "Safe message hides internal detail",

        "Status":
            "PASS"
            if (
                "required field"
                not in str(
                    safe_payload256
                ).lower()
            )
            else "FAIL",

        "Evidence":
            safe_payload256[
                "error"
            ][
                "message"
            ]
    },

    {
        "Test":
            "Internal detail available for logging",

        "Status":
            "PASS"
            if (
                "internal_message"
                in internal_payload256[
                    "error"
                ]
            )
            else "FAIL",

        "Evidence":
            "include_internal=True"
    },

    {
        "Test":
            "HTTP status mapped correctly",

        "Status":
            "PASS"
            if status_code256==400
            else "FAIL",

        "Evidence":
            status_code256
    },

    {
        "Test":
            "Unexpected errors use safe fallback",

        "Status":
            "PASS"
            if (
                unexpected_status256==500
                and
                unexpected_payload256[
                    "error"
                ][
                    "code"
                ]
                =="JOB_MARKET_ERROR"
            )
            else "FAIL",

        "Evidence":
            unexpected_payload256[
                "error"
            ][
                "message"
            ]
    },

    {
        "Test":
            "Every exception has error code",

        "Status":
            "PASS"
            if exception_registry256[
                "Error_Code"
            ].notna().all()
            else "FAIL",

        "Evidence":
            f"{len(exception_registry256)} exception classes"
    },

    {
        "Test":
            "Every exception has HTTP mapping",

        "Status":
            "PASS"
            if exception_registry256[
                "HTTP_Status"
            ].between(
                400,
                599
            ).all()
            else "FAIL",

        "Evidence":
            "400–599 contract"
    }
])


EXCEPTION_STATUS256=(
    "PASS"
    if exception_tests256[
        "Status"
    ].eq("PASS").all()
    else "REVIEW"
)


exception_tests256.to_csv(
    REPORT_DIR/
    "exception_validation.csv",
    index=False
)


exception_examples256=pd.DataFrame([
    {
        "Scenario":
            "Invalid API request",

        "Exception":
            "InvalidRequestError",

        "User_Response":
            InvalidRequestError.user_message,

        "HTTP_Status":
            InvalidRequestError.http_status
    },

    {
        "Scenario":
            "Unknown job family",

        "Exception":
            "ResourceNotFoundError",

        "User_Response":
            ResourceNotFoundError.user_message,

        "HTTP_Status":
            ResourceNotFoundError.http_status
    },

    {
        "Scenario":
            "Forecast model unavailable",

        "Exception":
            "ModelLoadingError",

        "User_Response":
            ModelLoadingError.user_message,

        "HTTP_Status":
            ModelLoadingError.http_status
    },

    {
        "Scenario":
            "SQL connection unavailable",

        "Exception":
            "DatabaseError",

        "User_Response":
            DatabaseError.user_message,

        "HTTP_Status":
            DatabaseError.http_status
    },

    {
        "Scenario":
            "Recommendation engine failure",

        "Exception":
            "RecommendationError",

        "User_Response":
            RecommendationError.user_message,

        "HTTP_Status":
            RecommendationError.http_status
    }
])


exception_examples256.to_csv(
    REPORT_DIR/
    "exception_response_examples.csv",
    index=False
)


if "logger" in globals():

    logger.info(
        "EXCEPTION_FRAMEWORK_TEST | status=%s | classes=%s",
        EXCEPTION_STATUS256,
        len(exception_registry256)
    )

    try:

        raise DataValidationError(
            message=
                "Controlled logger integration test",

            user_message=
                "The supplied data failed validation.",

            details={
                "section":"25.2",
                "component":"exceptions"
            }
        )

    except DataValidationError as e:

        logger.error(
            "CONTROLLED_EXCEPTION_TEST | "
            "code=%s | internal=%s",
            e.error_code,
            e.message
        )


integration_status256=(
    "PASS"
    if "logger" in globals()
    else "NOT TESTED"
)


exception_metadata256=pd.DataFrame([
    {
        "Component":
            "Structured Exception Framework",

        "Module":
            str(EXCEPTION_PATH256),

        "Exception_Classes":
            len(exception_registry256),

        "Safe_Response_Function":
            "exception_response",

        "Logger_Integration":
            integration_status256,

        "Validation_Status":
            EXCEPTION_STATUS256
    }
])


exception_metadata256.to_csv(
    REPORT_DIR/
    "exception_metadata.csv",
    index=False
)


def table256(df,rows=50):

    display(
        HTML(
            df.head(rows).to_html(
                index=False,
                border=0,
                classes="t256"
            )
        )
    )


def card256(
    label,
    value,
    note,
    accent="#4a4e69"
):

    return f"""
    <div class="c256"
    style="border-top:4px solid {accent}">
        <small>{label}</small>
        <b>{value}</b>
        <span>{note}</span>
    </div>
    """


display(
    HTML("""
<style>

.d256{
border:1px solid #ded8d5;
border-radius:18px;
overflow:hidden;
background:#fff;
box-shadow:0 9px 25px rgba(34,34,59,.10)
}

.h256{
background:linear-gradient(
120deg,#22223b,#30324d 55%,#4a4e69
);
padding:25px 28px;
display:flex;
justify-content:space-between;
align-items:flex-start;
gap:20px;
font-family:Arial
}

.h256 .k{
color:#c9ada7;
font-size:8px;
font-weight:700;
letter-spacing:1.7px
}

.h256 h2{
color:#fff;
margin:6px 0;
font-size:25px
}

.h256 p,
.h256 .m{
color:#f2e9e4;
font-size:10px;
line-height:1.65
}

.cards256{
display:grid;
grid-template-columns:repeat(6,1fr);
border-bottom:1px solid #e8e2df
}

.c256{
padding:14px;
border-right:1px solid #ebe6e4;
font-family:Arial;
min-width:0
}

.c256 small{
display:block;
color:#9a8c98;
font-size:7px;
font-weight:700
}

.c256 b{
display:block;
color:#22223b;
font-size:16px;
margin-top:5px
}

.c256 span{
display:block;
color:#4a4e69;
font-size:8px;
margin-top:4px
}

.out256{
padding:19px 22px;
border-top:1px solid #eee9e7;
min-height:230px
}

.t256{
width:100%;
border-collapse:collapse;
font:9px Arial
}

.t256 th{
background:#22223b;
color:#fff;
padding:9px;
text-align:left
}

.t256 td{
padding:8px;
border-bottom:1px solid #eee9e7;
vertical-align:top;
color:#22223b;
line-height:1.5
}

.t256 tr:nth-child(even){
background:#fbfaf9
}

.grid256{
display:grid;
grid-template-columns:repeat(3,1fr);
gap:12px
}

.box256{
background:#f8f6f5;
border-left:4px solid #4a4e69;
padding:14px;
border-radius:9px;
font:11px/1.65 Arial;
color:#22223b
}

.box256 b{
display:block;
color:#4a4e69;
font-size:9px;
margin-bottom:5px
}

.note256{
background:#f2e9e4;
padding:12px 14px;
border-radius:8px;
font:10px/1.65 Arial;
color:#4a4e69
}

.payload256{
background:#22223b;
color:#f2e9e4;
padding:14px;
border-radius:9px;
font:9px/1.65 Consolas,monospace;
white-space:pre-wrap
}

</style>
""")
)


passed_tests256=int(
    exception_tests256[
        "Status"
    ].eq("PASS").sum()
)


header256=widgets.HTML(
    f"""
<div class="h256">

<div>

<div class="k">
SECTION 25.2 • ERROR GOVERNANCE
</div>

<h2>
Application Exception Framework
</h2>

<p>
Safe User Errors • HTTP Mapping •
Internal Diagnostics • Logging Integration
</p>

</div>

<div class="m">

STATUS<br>
<b>{EXCEPTION_STATUS256}</b>

<br><br>

VALIDATION<br>
<b>
{passed_tests256}/{len(exception_tests256)}
</b>

</div>

</div>
"""
)


kpi256=widgets.HTML(
    '<div class="cards256">'
    +
    "".join([
        card256(
            "STATUS",
            EXCEPTION_STATUS256,
            "Engineering gate"
        ),

        card256(
            "EXCEPTIONS",
            len(exception_registry256),
            "Typed failures"
        ),

        card256(
            "SAFE ERRORS",
            "YES",
            "No stack trace to users"
        ),

        card256(
            "HTTP MAPPING",
            "YES",
            "400 / 404 / 422 / 503"
        ),

        card256(
            "LOGGER",
            integration_status256,
            "Error traceability"
        ),

        card256(
            "TESTS",
            f"{passed_tests256}/{len(exception_tests256)}",
            "Validation passed"
        )
    ])
    +
    "</div>"
)


view256=widgets.ToggleButtons(
    options=[
        ("OVERVIEW","overview"),
        ("REGISTRY","registry"),
        ("VALIDATION","validation"),
        ("API CONTRACT","contract"),
        ("EXAMPLES","examples"),
        ("FILES","files")
    ],

    value="overview",

    layout=widgets.Layout(
        width="100%"
    ),

    style={
        "button_width":"16.6%",
        "font_weight":"700"
    }
)

out256=widgets.Output()
out256.add_class("out256")

def update256(*_):
    with out256:
        clear_output(wait=True)
        if view256.value=="overview":
            display(
                HTML(
                    """
<div class="grid256">

<div class="box256">
<b>SAFE USER EXPERIENCE</b>
End users receive controlled, understandable messages instead of Python tracebacks or database/model internals.
</div>

<div class="box256">
<b>TECHNICAL TRACEABILITY</b>
Internal messages, metadata and original exception type remain available for logs and diagnostics.
</div>

<div class="box256">
<b>API READY</b>
Each exception carries a stable error code and HTTP status that Flask can convert directly into structured JSON.
</div>

</div>
"""
                )
            )


        elif view256.value=="registry":

            table256(
                exception_registry256
            )


        elif view256.value=="validation":

            table256(
                exception_tests256
            )


        elif view256.value=="contract":

            contract256=pd.DataFrame([
                {
                    "HTTP_Status":400,
                    "Meaning":"Invalid request / validation",
                    "Exceptions":
                        "InvalidRequestError, DataValidationError"
                },
                {
                    "HTTP_Status":404,
                    "Meaning":"Requested resource unavailable",
                    "Exceptions":
                        "ResourceNotFoundError"
                },
                {
                    "HTTP_Status":422,
                    "Meaning":"Input understood but cannot be processed",
                    "Exceptions":
                        "PreprocessingError"
                },
                {
                    "HTTP_Status":500,
                    "Meaning":"Controlled internal inference/application error",
                    "Exceptions":
                        "JobMarketError, InferenceError, RecommendationError"
                },
                {
                    "HTTP_Status":503,
                    "Meaning":"Required dependency temporarily unavailable",
                    "Exceptions":
                        "ModelLoadingError, DatabaseError, ServiceUnavailableError"
                }
            ])

            table256(contract256)

        elif view256.value=="examples":
            table256(
                exception_examples256
            )
            display(
                HTML(
                    f"""
<br>
<div class="note256">
<b>Safe API-style payload</b>
</div>

<div class="payload256">
{json.dumps(
    safe_payload256,
    indent=2
)}
</div>
"""
                )
            )


        else:

            files256=pd.DataFrame({
                "Artifact":[
                    "Exception Module",
                    "Exception Registry",
                    "Validation Report",
                    "Response Examples",
                    "Exception Metadata"
                ],

                "Location":[
                    str(EXCEPTION_PATH256),
                    str(
                        REPORT_DIR/
                        "exception_registry.csv"
                    ),
                    str(
                        REPORT_DIR/
                        "exception_validation.csv"
                    ),
                    str(
                        REPORT_DIR/
                        "exception_response_examples.csv"
                    ),
                    str(
                        REPORT_DIR/
                        "exception_metadata.csv"
                    )
                ]
            })

            files256[
                "Exists"
            ]=files256[
                "Location"
            ].map(
                lambda p:
                Path(p).exists()
            )

            files256[
                "Status"
            ]=np.where(
                files256["Exists"],
                "PASS",
                "REVIEW"
            )

            table256(files256)

view256.observe(
    update256,
    names="value"
)

update256()

dashboard_256=widgets.VBox([
    header256,
    kpi256,
    view256,
    out256
])

dashboard_256.add_class("d256")

display(dashboard_256)
```

```python
CONFIG_DIR=Path("config")
REPORT_DIR=Path("reports")
CONFIG_DIR.mkdir(parents=True,exist_ok=True)
REPORT_DIR.mkdir(parents=True,exist_ok=True)
SETTINGS_PATH257=CONFIG_DIR/"settings.py"
ENV_EXAMPLE257=Path(".env.example")

sql_server_default257=str(
    globals().get(
        "SQL_SERVER",
        r"localhost\SQLEXPRESS"
    )
)

sql_database_default257=str(
    globals().get(
        "SQL_DATABASE",
        "JobMarketDB"
    )
)

sql_driver_default257=str(
    globals().get(
        "SQL_DRIVER",
        "ODBC Driver 18 for SQL Server"
    )
)

forecast_version257=str(
    globals().get(
        "MODEL_VERSION",
        "demand_v1"
    )
)

recommender_version257=str(
    globals().get(
        "RECOMMENDER_VERSION",
        "recommender_v1"
    )
)


settings_code257=f'''
import os
from pathlib import Path

BASE_DIR=Path(__file__).resolve().parents[1]

DATA_DIR=BASE_DIR/"data"
MODEL_DIR=BASE_DIR/"models"
REPORT_DIR=BASE_DIR/"reports"
EXPORT_DIR=BASE_DIR/"exports"
LOG_DIR=BASE_DIR/"logs"
DATABASE_DIR=BASE_DIR/"database"
SCREENSHOT_DIR=BASE_DIR/"screenshots"

FORECAST_MODEL_VERSION=os.getenv(
    "FORECAST_MODEL_VERSION",
    "{forecast_version257}"
)

RECOMMENDER_VERSION=os.getenv(
    "RECOMMENDER_VERSION",
    "{recommender_version257}"
)

FORECAST_MODEL_PATH=Path(
    os.getenv(
        "FORECAST_MODEL_PATH",
        str(
            MODEL_DIR/
            f"{{FORECAST_MODEL_VERSION}}.pkl"
        )
    )
)

FORECAST_METADATA_PATH=Path(
    os.getenv(
        "FORECAST_METADATA_PATH",
        str(
            MODEL_DIR/
            f"{{FORECAST_MODEL_VERSION}}_metadata.json"
        )
    )
)

RECOMMENDER_DIR=Path(
    os.getenv(
        "RECOMMENDER_DIR",
        str(
            MODEL_DIR/
            "recommender_assets"
        )
    )
)

RECOMMENDATION_CATALOG_PATH=(
    RECOMMENDER_DIR/
    "recommendation_catalog.csv"
)

TFIDF_VECTORIZER_PATH=(
    RECOMMENDER_DIR/
    "tfidf_vectorizer.pkl"
)

TFIDF_MATRIX_PATH=(
    RECOMMENDER_DIR/
    "tfidf_matrix.npz"
)

RECOMMENDER_WEIGHTS_PATH=(
    RECOMMENDER_DIR/
    "weights.json"
)

RECOMMENDER_METADATA_PATH=(
    RECOMMENDER_DIR/
    "metadata.json"
)

SQL_SERVER=os.getenv(
    "SQL_SERVER",
    r"{sql_server_default257}"
)

SQL_DATABASE=os.getenv(
    "SQL_DATABASE",
    "{sql_database_default257}"
)

SQL_DRIVER=os.getenv(
    "SQL_DRIVER",
    "{sql_driver_default257}"
)

SQL_TRUSTED_CONNECTION=(
    os.getenv(
        "SQL_TRUSTED_CONNECTION",
        "yes"
    ).lower()
    in {{
        "1",
        "true",
        "yes",
        "y"
    }}
)

SQL_ENCRYPT=os.getenv(
    "SQL_ENCRYPT",
    "yes"
)

SQL_TRUST_SERVER_CERTIFICATE=os.getenv(
    "SQL_TRUST_SERVER_CERTIFICATE",
    "yes"
)

API_HOST=os.getenv(
    "API_HOST",
    "127.0.0.1"
)

API_PORT=int(
    os.getenv(
        "API_PORT",
        "5000"
    )
)

API_DEBUG=(
    os.getenv(
        "API_DEBUG",
        "false"
    ).lower()
    in {{
        "1",
        "true",
        "yes",
        "y"
    }}
)

API_VERSION=os.getenv(
    "API_VERSION",
    "v1"
)

API_BASE_PATH=f"/api/{{API_VERSION}}"

STREAMLIT_API_URL=os.getenv(
    "JOBMARKET_API_URL",
    f"http://{{API_HOST}}:{{API_PORT}}"
)

MAX_RECOMMENDATIONS=int(
    os.getenv(
        "MAX_RECOMMENDATIONS",
        "50"
    )
)

DEFAULT_RECOMMENDATIONS=int(
    os.getenv(
        "DEFAULT_RECOMMENDATIONS",
        "10"
    )
)

REQUEST_TIMEOUT_SECONDS=float(
    os.getenv(
        "REQUEST_TIMEOUT_SECONDS",
        "15"
    )
)


def ensure_directories():

    for path in [
        DATA_DIR,
        MODEL_DIR,
        REPORT_DIR,
        EXPORT_DIR,
        LOG_DIR,
        DATABASE_DIR,
        SCREENSHOT_DIR
    ]:
        path.mkdir(
            parents=True,
            exist_ok=True
        )


def sql_connection_string():

    parts=[
        f"DRIVER={{{{SQL_DRIVER}}}}",
        f"SERVER={{SQL_SERVER}}",
        f"DATABASE={{SQL_DATABASE}}"
    ]

    if SQL_TRUSTED_CONNECTION:
        parts.append(
            "Trusted_Connection=yes"
        )

    parts.extend([
        f"Encrypt={{SQL_ENCRYPT}}",
        (
            "TrustServerCertificate="
            f"{{SQL_TRUST_SERVER_CERTIFICATE}}"
        )
    ])

    return ";".join(parts)+";"


def validate_settings(
    require_models=False
):

    checks={{
        "base_directory":
            BASE_DIR.exists(),

        "api_port_valid":
            1<=API_PORT<=65535,

        "max_recommendations_valid":
            MAX_RECOMMENDATIONS>=1,

        "default_recommendations_valid":
            1<=DEFAULT_RECOMMENDATIONS
            <=MAX_RECOMMENDATIONS,

        "request_timeout_valid":
            REQUEST_TIMEOUT_SECONDS>0,

        "sql_server_defined":
            bool(
                str(SQL_SERVER).strip()
            ),

        "sql_database_defined":
            bool(
                str(SQL_DATABASE).strip()
            ),

        "sql_driver_defined":
            bool(
                str(SQL_DRIVER).strip()
            )
    }}

    if require_models:

        checks.update({{
            "forecast_model_exists":
                FORECAST_MODEL_PATH.exists(),

            "forecast_metadata_exists":
                FORECAST_METADATA_PATH.exists(),

            "recommendation_catalog_exists":
                RECOMMENDATION_CATALOG_PATH.exists(),

            "vectorizer_exists":
                TFIDF_VECTORIZER_PATH.exists(),

            "tfidf_matrix_exists":
                TFIDF_MATRIX_PATH.exists(),

            "weights_exists":
                RECOMMENDER_WEIGHTS_PATH.exists(),

            "recommender_metadata_exists":
                RECOMMENDER_METADATA_PATH.exists()
        }})

    return checks


def masked_settings():

    return {{
        "BASE_DIR":
            str(BASE_DIR),

        "FORECAST_MODEL_VERSION":
            FORECAST_MODEL_VERSION,

        "RECOMMENDER_VERSION":
            RECOMMENDER_VERSION,

        "SQL_SERVER":
            SQL_SERVER,

        "SQL_DATABASE":
            SQL_DATABASE,

        "SQL_DRIVER":
            SQL_DRIVER,

        "SQL_TRUSTED_CONNECTION":
            SQL_TRUSTED_CONNECTION,

        "API_HOST":
            API_HOST,

        "API_PORT":
            API_PORT,

        "API_DEBUG":
            API_DEBUG,

        "API_BASE_PATH":
            API_BASE_PATH,

        "STREAMLIT_API_URL":
            STREAMLIT_API_URL,

        "MAX_RECOMMENDATIONS":
            MAX_RECOMMENDATIONS,

        "DEFAULT_RECOMMENDATIONS":
            DEFAULT_RECOMMENDATIONS,

        "REQUEST_TIMEOUT_SECONDS":
            REQUEST_TIMEOUT_SECONDS
    }}


ensure_directories()
'''


SETTINGS_PATH257.write_text(
    settings_code257,
    encoding="utf-8"
)


env_example257=f"""SQL_SERVER={sql_server_default257}
SQL_DATABASE={sql_database_default257}
SQL_DRIVER={sql_driver_default257}
SQL_TRUSTED_CONNECTION=yes
SQL_ENCRYPT=yes
SQL_TRUST_SERVER_CERTIFICATE=yes

FORECAST_MODEL_VERSION={forecast_version257}
RECOMMENDER_VERSION={recommender_version257}

API_HOST=127.0.0.1
API_PORT=5000
API_DEBUG=false
API_VERSION=v1

JOBMARKET_API_URL=http://127.0.0.1:5000

MAX_RECOMMENDATIONS=50
DEFAULT_RECOMMENDATIONS=10
REQUEST_TIMEOUT_SECONDS=15
"""


ENV_EXAMPLE257.write_text(
    env_example257,
    encoding="utf-8"
)


import importlib
import config.settings as settings_module257

importlib.reload(
    settings_module257
)

from config.settings import (
    BASE_DIR as SETTINGS_BASE_DIR257,
    FORECAST_MODEL_PATH as SETTINGS_FORECAST_PATH257,
    RECOMMENDER_DIR as SETTINGS_REC_DIR257,
    SQL_SERVER as SETTINGS_SQL_SERVER257,
    SQL_DATABASE as SETTINGS_SQL_DATABASE257,
    SQL_DRIVER as SETTINGS_SQL_DRIVER257,
    API_HOST as SETTINGS_API_HOST257,
    API_PORT as SETTINGS_API_PORT257,
    API_BASE_PATH as SETTINGS_API_BASE_PATH257,
    STREAMLIT_API_URL as SETTINGS_STREAMLIT_URL257,
    MAX_RECOMMENDATIONS as SETTINGS_MAX_REC257,
    DEFAULT_RECOMMENDATIONS as SETTINGS_DEFAULT_REC257,
    REQUEST_TIMEOUT_SECONDS as SETTINGS_TIMEOUT257,
    sql_connection_string as settings_connection_string257,
    validate_settings as validate_settings257,
    masked_settings as masked_settings257
)


base_checks257=validate_settings257(
    require_models=False
)

model_checks257=validate_settings257(
    require_models=True
)


settings_validation257=pd.DataFrame([
    {
        "Check":
            key,
        "Area":
            "Runtime",
        "Required_Now":
            True,
        "Status":
            "PASS"
            if value
            else "FAIL"
    }

    for key,value
    in base_checks257.items()
])


model_validation257=pd.DataFrame([
    {
        "Check":
            key,
        "Area":
            "Model Artifact",
        "Required_Now":
            True,
        "Status":
            "PASS"
            if value
            else "FAIL"
    }

    for key,value
    in model_checks257.items()

    if key not in base_checks257
])


settings_validation257=pd.concat(
    [
        settings_validation257,
        model_validation257
    ],
    ignore_index=True
)


connection_string_test257=(
    settings_connection_string257()
)


connection_safe257=all([
    "DRIVER=" in connection_string_test257,
    "SERVER=" in connection_string_test257,
    "DATABASE=" in connection_string_test257,
    "Trusted_Connection=yes"
    in connection_string_test257,
    "PWD=" not in connection_string_test257.upper(),
    "PASSWORD="
    not in connection_string_test257.upper()
])


settings_validation257.loc[
    len(settings_validation257)
]={
    "Check":"sql_connection_string_safe",
    "Area":"SQL",
    "Required_Now":True,
    "Status":
        "PASS"
        if connection_safe257
        else "FAIL"
}


env_text257=ENV_EXAMPLE257.read_text(
    encoding="utf-8",
    errors="ignore"
)


env_safe257=(
    "PASSWORD="
    not in env_text257.upper()
    and
    "PWD="
    not in env_text257.upper()
)


settings_validation257.loc[
    len(settings_validation257)
]={
    "Check":"env_template_contains_no_password",
    "Area":"Security",
    "Required_Now":True,
    "Status":
        "PASS"
        if env_safe257
        else "FAIL"
}


settings_validation257.loc[
    len(settings_validation257)
]={
    "Check":"settings_module_created",
    "Area":"Configuration",
    "Required_Now":True,
    "Status":
        "PASS"
        if SETTINGS_PATH257.exists()
        else "FAIL"
}


settings_validation257.loc[
    len(settings_validation257)
]={
    "Check":"environment_template_created",
    "Area":"Configuration",
    "Required_Now":True,
    "Status":
        "PASS"
        if ENV_EXAMPLE257.exists()
        else "FAIL"
}


SETTINGS_STATUS257=(
    "PASS"
    if settings_validation257[
        "Status"
    ].eq("PASS").all()
    else "REVIEW"
)


settings_validation257.to_csv(
    REPORT_DIR/
    "settings_validation.csv",
    index=False
)


runtime_settings257=pd.DataFrame([
    {
        "Setting":
            key,
        "Value":
            value
    }

    for key,value
    in masked_settings257().items()
])


runtime_settings257.to_csv(
    REPORT_DIR/
    "runtime_settings_snapshot.csv",
    index=False
)


settings_metadata257=pd.DataFrame([
    {
        "Component":
            "Central Runtime Settings",

        "Module":
            str(SETTINGS_PATH257),

        "Environment_Template":
            str(ENV_EXAMPLE257),

        "SQL_Server":
            SETTINGS_SQL_SERVER257,

        "SQL_Database":
            SETTINGS_SQL_DATABASE257,

        "Forecast_Model":
            str(
                SETTINGS_FORECAST_PATH257
            ),

        "Recommender_Directory":
            str(
                SETTINGS_REC_DIR257
            ),

        "API_Base":
            (
                f"http://"
                f"{SETTINGS_API_HOST257}:"
                f"{SETTINGS_API_PORT257}"
                f"{SETTINGS_API_BASE_PATH257}"
            ),

        "Max_Recommendations":
            SETTINGS_MAX_REC257,

        "Validation_Status":
            SETTINGS_STATUS257
    }
])


settings_metadata257.to_csv(
    REPORT_DIR/
    "settings_metadata.csv",
    index=False
)


def table257(df,rows=50):

    display(
        HTML(
            df.head(rows).to_html(
                index=False,
                border=0,
                classes="t257"
            )
        )
    )


def card257(
    label,
    value,
    note,
    accent="#4a4e69"
):

    return f"""
    <div class="c257"
    style="border-top:4px solid {accent}">
        <small>{label}</small>
        <b>{value}</b>
        <span>{note}</span>
    </div>
    """


display(
    HTML("""
<style>

.d257{
border:1px solid #ded8d5;
border-radius:18px;
overflow:hidden;
background:#fff;
box-shadow:0 9px 25px rgba(34,34,59,.10)
}

.h257{
background:linear-gradient(
120deg,#22223b,#30324d 55%,#4a4e69
);
padding:25px 28px;
display:flex;
justify-content:space-between;
align-items:flex-start;
gap:20px;
font-family:Arial
}

.h257 .k{
color:#c9ada7;
font-size:8px;
font-weight:700;
letter-spacing:1.7px
}

.h257 h2{
color:#fff;
margin:6px 0;
font-size:25px
}

.h257 p,
.h257 .m{
color:#f2e9e4;
font-size:10px;
line-height:1.65
}

.cards257{
display:grid;
grid-template-columns:repeat(6,1fr);
border-bottom:1px solid #e8e2df
}

.c257{
padding:14px;
border-right:1px solid #ebe6e4;
font-family:Arial;
min-width:0
}

.c257 small{
display:block;
color:#9a8c98;
font-size:7px;
font-weight:700
}

.c257 b{
display:block;
color:#22223b;
font-size:15px;
margin-top:5px;
word-break:break-word
}

.c257 span{
display:block;
color:#4a4e69;
font-size:8px;
margin-top:4px
}

.out257{
padding:19px 22px;
border-top:1px solid #eee9e7;
min-height:230px
}

.t257{
width:100%;
border-collapse:collapse;
font:9px Arial
}

.t257 th{
background:#22223b;
color:#fff;
padding:9px;
text-align:left
}

.t257 td{
padding:8px;
border-bottom:1px solid #eee9e7;
vertical-align:top;
color:#22223b;
line-height:1.5
}

.t257 tr:nth-child(even){
background:#fbfaf9
}

.grid257{
display:grid;
grid-template-columns:repeat(3,1fr);
gap:12px
}

.box257{
background:#f8f6f5;
border-left:4px solid #4a4e69;
padding:14px;
border-radius:9px;
font:11px/1.65 Arial;
color:#22223b
}

.box257 b{
display:block;
color:#4a4e69;
font-size:9px;
margin-bottom:5px
}

.note257{
background:#f2e9e4;
padding:12px 14px;
border-radius:8px;
font:10px/1.65 Arial;
color:#4a4e69
}

.code257{
background:#22223b;
color:#f2e9e4;
padding:14px;
border-radius:9px;
font:9px/1.65 Consolas,monospace;
white-space:pre-wrap;
overflow-x:auto
}

</style>
""")
)


passed_checks257=int(
    settings_validation257[
        "Status"
    ].eq("PASS").sum()
)


header257=widgets.HTML(
    f"""
<div class="h257">

<div>

<div class="k">
SECTION 25.2 • CONFIGURATION CONTROL
</div>

<h2>
Central Runtime Settings
</h2>

<p>
Paths • SQL Server • Model Assets •
API Configuration • Environment Overrides • Security
</p>

</div>

<div class="m">

STATUS<br>
<b>{SETTINGS_STATUS257}</b>

<br><br>

VALIDATION<br>
<b>
{passed_checks257}/{len(settings_validation257)}
</b>

</div>

</div>
"""
)


kpi257=widgets.HTML(
    '<div class="cards257">'
    +
    "".join([
        card257(
            "STATUS",
            SETTINGS_STATUS257,
            "Configuration gate"
        ),

        card257(
            "DATABASE",
            SETTINGS_SQL_DATABASE257,
            "SQL Server"
        ),

        card257(
            "FORECAST",
            forecast_version257,
            "Model version"
        ),

        card257(
            "RECOMMENDER",
            recommender_version257,
            "Asset version"
        ),

        card257(
            "API PORT",
            SETTINGS_API_PORT257,
            "Flask runtime"
        ),

        card257(
            "TOP-K LIMIT",
            SETTINGS_MAX_REC257,
            "Recommendations"
        )
    ])
    +
    "</div>"
)


view257=widgets.ToggleButtons(
    options=[
        ("OVERVIEW","overview"),
        ("SETTINGS","settings"),
        ("VALIDATION","validation"),
        ("PATHS","paths"),
        ("SQL","sql"),
        ("API","api"),
        ("FILES","files")
    ],

    value="overview",

    layout=widgets.Layout(
        width="100%"
    ),

    style={
        "button_width":"14.2%",
        "font_weight":"700"
    }
)


out257=widgets.Output()
out257.add_class("out257")


def update257(*_):

    with out257:

        clear_output(wait=True)

        if view257.value=="overview":

            display(
                HTML(
                    f"""
<div class="grid257">

<div class="box257">
<b>SINGLE SOURCE OF TRUTH</b>
Paths, model locations, SQL settings and API configuration now live in one reusable module rather than being duplicated across notebook, Flask and Streamlit code.
</div>

<div class="box257">
<b>ENVIRONMENT READY</b>
Runtime values can be overridden through environment variables without changing source code.
</div>

<div class="box257">
<b>SECURITY CONTROL</b>
The environment template contains configuration examples but no database password or application secret.
</div>

</div>

<br>

<div class="note257">
<b>Runtime module:</b> {SETTINGS_PATH257}<br>
<b>Environment template:</b> {ENV_EXAMPLE257}<br>
<b>Validation:</b> {SETTINGS_STATUS257}
</div>
"""
                )
            )


        elif view257.value=="settings":

            table257(
                runtime_settings257
            )


        elif view257.value=="validation":

            table257(
                settings_validation257
            )


        elif view257.value=="paths":

            paths257=pd.DataFrame([
                {
                    "Resource":
                        "Project Root",
                    "Path":
                        SETTINGS_BASE_DIR257,
                    "Exists":
                        Path(
                            SETTINGS_BASE_DIR257
                        ).exists()
                },
                {
                    "Resource":
                        "Forecast Model",
                    "Path":
                        SETTINGS_FORECAST_PATH257,
                    "Exists":
                        Path(
                            SETTINGS_FORECAST_PATH257
                        ).exists()
                },
                {
                    "Resource":
                        "Recommender Assets",
                    "Path":
                        SETTINGS_REC_DIR257,
                    "Exists":
                        Path(
                            SETTINGS_REC_DIR257
                        ).exists()
                },
                {
                    "Resource":
                        "Reports",
                    "Path":
                        REPORT_DIR,
                    "Exists":
                        REPORT_DIR.exists()
                },
                {
                    "Resource":
                        "Logs",
                    "Path":
                        Path("logs"),
                    "Exists":
                        Path("logs").exists()
                }
            ])

            table257(paths257)


        elif view257.value=="sql":

            sql_display257=pd.DataFrame([
                {
                    "Setting":
                        "Server",
                    "Value":
                        SETTINGS_SQL_SERVER257
                },
                {
                    "Setting":
                        "Database",
                    "Value":
                        SETTINGS_SQL_DATABASE257
                },
                {
                    "Setting":
                        "Driver",
                    "Value":
                        SETTINGS_SQL_DRIVER257
                },
                {
                    "Setting":
                        "Authentication",
                    "Value":
                        "Windows Trusted Connection"
                },
                {
                    "Setting":
                        "Encrypt",
                    "Value":
                        "yes"
                },
                {
                    "Setting":
                        "Trust Server Certificate",
                    "Value":
                        "yes"
                }
            ])

            table257(
                sql_display257
            )

            safe_conn257=(
                connection_string_test257
                .replace(
                    SETTINGS_SQL_SERVER257,
                    "<SERVER>"
                )
                .replace(
                    SETTINGS_SQL_DATABASE257,
                    "<DATABASE>"
                )
            )

            display(
                HTML(
                    f"""
<br>
<div class="code257">
{html.escape(safe_conn257)}
</div>
"""
                )
            )


        elif view257.value=="api":

            api257=pd.DataFrame([
                {
                    "Setting":
                        "Host",
                    "Value":
                        SETTINGS_API_HOST257
                },
                {
                    "Setting":
                        "Port",
                    "Value":
                        SETTINGS_API_PORT257
                },
                {
                    "Setting":
                        "Base Path",
                    "Value":
                        SETTINGS_API_BASE_PATH257
                },
                {
                    "Setting":
                        "Streamlit API URL",
                    "Value":
                        SETTINGS_STREAMLIT_URL257
                },
                {
                    "Setting":
                        "Default Recommendations",
                    "Value":
                        SETTINGS_DEFAULT_REC257
                },
                {
                    "Setting":
                        "Maximum Recommendations",
                    "Value":
                        SETTINGS_MAX_REC257
                },
                {
                    "Setting":
                        "Request Timeout",
                    "Value":
                        f"{SETTINGS_TIMEOUT257} seconds"
                }
            ])

            table257(api257)


        else:

            files257=pd.DataFrame({
                "Artifact":[
                    "Settings Module",
                    "Environment Template",
                    "Validation Report",
                    "Runtime Snapshot",
                    "Settings Metadata"
                ],

                "Location":[
                    str(SETTINGS_PATH257),
                    str(ENV_EXAMPLE257),
                    str(
                        REPORT_DIR/
                        "settings_validation.csv"
                    ),
                    str(
                        REPORT_DIR/
                        "runtime_settings_snapshot.csv"
                    ),
                    str(
                        REPORT_DIR/
                        "settings_metadata.csv"
                    )
                ]
            })

            files257[
                "Exists"
            ]=files257[
                "Location"
            ].map(
                lambda p:
                Path(p).exists()
            )

            files257[
                "Status"
            ]=np.where(
                files257["Exists"],
                "PASS",
                "REVIEW"
            )

            table257(files257)

view257.observe(
    update257,
    names="value"
)

update257()

dashboard_257=widgets.VBox([
    header257,
    kpi257,
    view257,
    out257
])

dashboard_257.add_class("d257")

display(dashboard_257)
```

```python
LOG_DIR,REPORT_DIR=Path("logs"),Path("reports")
LOG_DIR.mkdir(exist_ok=True); REPORT_DIR.mkdir(exist_ok=True)
APPLICATION_LOG258=LOG_DIR/"application.log"

controlled_patterns258=[
    "LOGGER_WARNING_TEST","LOGGER_DEBUG_TEST","LOGGER_STRUCTURED_EVENT_TEST",
    "CONTROLLED_EXCEPTION_TEST","MANUAL_LOGGER_TEST"
]
secret_patterns258=["password=","pwd=","api_key=","apikey=","access_token=","secret_key=","client_secret="]

category_rules258={
    "Logger Initialization":["logger_initialized","logger_structured_event_test"],
    "Validation / Testing":["validation","test","status=pass"],
    "Model / Inference":["model","forecast","recommend","inference"],
    "SQL / Database":["sql","database"],
    "Configuration":["settings","config"],
    "Exception Handling":["exception","error"],
    "Flask API Requests":["api_request","flask","/api/"],
    "Streamlit / Deployment":["streamlit","deployment","deploy"]
}
future_categories258={"Flask API Requests","Streamlit / Deployment"}

def parse_logs258():
    lines=APPLICATION_LOG258.read_text(encoding="utf-8",errors="ignore").splitlines() if APPLICATION_LOG258.exists() else []
    rows=[]

    for line in lines:
        p=line.split(" | ",4)
        rows.append({
            "Timestamp":p[0].strip() if len(p)>0 else "",
            "Level":p[1].strip().upper() if len(p)>1 else "UNKNOWN",
            "Logger":p[2].strip() if len(p)>2 else "",
            "Source":p[3].strip() if len(p)>3 else "",
            "Message":p[4].strip() if len(p)>4 else line,
            "Parsed":len(p)==5
        })

    d=pd.DataFrame(rows,columns=["Timestamp","Level","Logger","Source","Message","Parsed"])
    d["Parsed_Timestamp"]=pd.to_datetime(d["Timestamp"],errors="coerce")
    d["Is_Controlled_Test"]=d["Message"].fillna("").apply(
        lambda x:any(p.lower() in x.lower() for p in controlled_patterns258)
    )

    def category(x):
        x=str(x).lower()
        if any(p.lower() in x for p in controlled_patterns258): return "Controlled Test"
        for c,sigs in category_rules258.items():
            if any(s in x for s in sigs): return c
        return "General"

    d["Event_Category"]=d["Message"].apply(category)
    return lines,d

def rebuild258():
    global log_lines258,log_events258,operational_events258,controlled_test_events258
    global warning_error_events258,level_summary258,event_coverage258,timeline258
    global logging_audit258,logging_metadata258,warning_count258,error_count258
    global critical_count258,controlled_count258,operational_count258,parse_rate258
    global secret_safe258,rotation_ready258,log_file_size258,log_modified258
    global APPLICATION_LOG_STATUS258,OPERATIONAL_HEALTH258,audit_pass_count258
    log_lines258,log_events258=parse_logs258()
    operational_events258=log_events258[~log_events258["Is_Controlled_Test"]].copy()
    controlled_test_events258=log_events258[log_events258["Is_Controlled_Test"]].copy()
    warning_error_events258=operational_events258[
        operational_events258["Level"].isin(["WARNING","ERROR","CRITICAL"])
    ].copy()

    levels=["DEBUG","INFO","WARNING","ERROR","CRITICAL","UNKNOWN"]
    level_summary258=operational_events258["Level"].value_counts().reindex(levels,fill_value=0).rename_axis("Level").reset_index(name="Events")
    level_summary258["Share_%"]=(level_summary258["Events"]/max(len(operational_events258),1)*100).round(1)
    text="\n".join(log_lines258).lower()
    event_coverage258=pd.DataFrame([
        {
            "Category":c,
            "Required_Now":c not in future_categories258,
            "Signals_Observed":", ".join(s for s in sigs if s in text),
            "Status":"OBSERVED" if any(s in text for s in sigs) else ("PLANNED" if c in future_categories258 else "NOT YET OBSERVED")
        }
        for c,sigs in category_rules258.items()
    ])

    valid=operational_events258.dropna(subset=["Parsed_Timestamp"]).copy()
    if not valid.empty:
        valid["Time_Bucket"]=valid["Parsed_Timestamp"].dt.floor("h")
        timeline258=valid.groupby("Time_Bucket").size().reset_index(name="Events")
    else:
        timeline258=pd.DataFrame(columns=["Time_Bucket","Events"])

    warning_count258=int((operational_events258["Level"]=="WARNING").sum())
    error_count258=int((operational_events258["Level"]=="ERROR").sum())
    critical_count258=int((operational_events258["Level"]=="CRITICAL").sum())
    controlled_count258=len(controlled_test_events258)
    operational_count258=len(operational_events258)
    parse_rate258=round(log_events258["Parsed"].mean()*100,1) if len(log_events258) else 0
    secret_safe258=not any(x in text for x in secret_patterns258)
    rotation_ready258="logger" in globals() and any("RotatingFileHandler" in type(h).__name__ for h in logger.handlers)
    log_file_size258=APPLICATION_LOG258.stat().st_size if APPLICATION_LOG258.exists() else 0
    log_modified258=pd.Timestamp.fromtimestamp(APPLICATION_LOG258.stat().st_mtime,tz="UTC") if APPLICATION_LOG258.exists() else pd.NaT

    logging_audit258=pd.DataFrame([
        ["Application log exists","PASS" if APPLICATION_LOG258.exists() else "FAIL",APPLICATION_LOG258],
        ["Application log contains events","PASS" if len(log_lines258) else "FAIL",f"{len(log_lines258)} lines"],
        ["Structured format parseable","PASS" if parse_rate258>=80 else "REVIEW",f"{parse_rate258}%"],
        ["INFO operational evidence","PASS" if (operational_events258["Level"]=="INFO").any() else "REVIEW",int((operational_events258["Level"]=="INFO").sum())],
        ["Rotating handler configured","PASS" if rotation_ready258 else "REVIEW",rotation_ready258],
        ["Controlled tests separated","PASS",f"{controlled_count258} tests"],
        ["No obvious secrets detected","PASS" if secret_safe258 else "FAIL","No sensitive patterns" if secret_safe258 else "Review required"]
    ],columns=["Control","Status","Evidence"])

    audit_pass_count258=int((logging_audit258["Status"]=="PASS").sum())
    APPLICATION_LOG_STATUS258="PASS" if (logging_audit258["Status"]=="PASS").all() else "REVIEW"
    OPERATIONAL_HEALTH258="HEALTHY" if error_count258==0 and critical_count258==0 else "ATTENTION"

    logging_metadata258=pd.DataFrame([{
        "Log_File":str(APPLICATION_LOG258),
        "Operational_Events":operational_count258,
        "Controlled_Test_Events":controlled_count258,
        "Warnings":warning_count258,
        "Errors":error_count258,
        "Critical":critical_count258,
        "Parse_Rate_%":parse_rate258,
        "Rotation_Configured":rotation_ready258,
        "Secret_Scan":"PASS" if secret_safe258 else "FAIL",
        "Operational_Health":OPERATIONAL_HEALTH258,
        "Audit_Status":APPLICATION_LOG_STATUS258
    }])

def save258():
    log_events258.to_csv(REPORT_DIR/"application_log_all_events.csv",index=False)
    operational_events258.to_csv(REPORT_DIR/"application_log_operational_events.csv",index=False)
    controlled_test_events258.to_csv(REPORT_DIR/"application_log_controlled_tests.csv",index=False)
    warning_error_events258.to_csv(REPORT_DIR/"application_log_operational_incidents.csv",index=False)
    level_summary258.to_csv(REPORT_DIR/"application_log_level_summary.csv",index=False)
    event_coverage258.to_csv(REPORT_DIR/"application_log_event_coverage.csv",index=False)
    timeline258.to_csv(REPORT_DIR/"application_log_timeline.csv",index=False)
    logging_audit258.to_csv(REPORT_DIR/"application_log_audit.csv",index=False)
    logging_metadata258.to_csv(REPORT_DIR/"application_log_metadata.csv",index=False)

rebuild258(); save258()

display(HTML("""
<style>
.d258{border:1px solid #ddd;border-radius:17px;overflow:hidden;background:white;box-shadow:0 8px 22px #22223b18}
.h258{background:linear-gradient(120deg,#22223b,#4a4e69);padding:23px 27px;font-family:Arial;color:white}
.h258 small{color:#c9ada7;font-weight:700;letter-spacing:1.5px}.h258 h2{margin:6px 0}.h258 p{font-size:10px;color:#f2e9e4}
.cards258{display:grid;grid-template-columns:repeat(7,1fr)}.c258{padding:13px;border-right:1px solid #eee;font-family:Arial}
.c258 small{font-size:7px;color:#9a8c98;font-weight:700}.c258 b{display:block;font-size:15px;color:#22223b;margin-top:5px}.c258 span{font-size:8px;color:#4a4e69}
.out258{padding:20px}.t258{width:100%;border-collapse:collapse;font:9px Arial}.t258 th{background:#22223b;color:white;padding:8px;text-align:left}.t258 td{padding:8px;border-bottom:1px solid #eee}
.box258{background:#f8f6f5;border-left:4px solid #4a4e69;padding:13px;border-radius:8px;font:10px/1.6 Arial}
.grid258{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}.note258{background:#f2e9e4;padding:12px;border-radius:8px;font:10px Arial}
</style>
"""))

def card258(a,b,c):
    return f'<div class="c258"><small>{a}</small><b>{b}</b><span>{c}</span></div>'

def header258_html():
    return f"""
    <div class="h258">
    <small>SECTION 25.2 • LOGGING EVIDENCE</small>
    <h2>Application Log Intelligence Center</h2>
    <p>Operational Health • Controlled Tests • Severity • Timeline • Coverage • Security • Audit</p>
    <b>{APPLICATION_LOG_STATUS258} • {OPERATIONAL_HEALTH258}</b>
    </div>"""

def kpi258_html():
    return '<div class="cards258">'+''.join([
        card258("AUDIT",APPLICATION_LOG_STATUS258,f"{audit_pass_count258}/{len(logging_audit258)} controls"),
        card258("OPERATIONAL",operational_count258,"Real events"),
        card258("CONTROLLED",controlled_count258,"Test evidence"),
        card258("WARNINGS",warning_count258,"Operational"),
        card258("ERRORS",error_count258,"Operational"),
        card258("PARSE",f"{parse_rate258}%","Structured"),
        card258("SECRET SCAN","PASS" if secret_safe258 else "REVIEW","Security")
    ])+"</div>"

header258=widgets.HTML(header258_html())
kpi258=widgets.HTML(kpi258_html())

view258=widgets.ToggleButtons(
    options=["EXECUTIVE","OPERATIONAL","CONTROLLED TESTS","SEVERITY","TIMELINE","COVERAGE","AUDIT","FILES"],
    value="EXECUTIVE",layout=widgets.Layout(width="100%")
)

level258=widgets.Dropdown(options=["ALL","DEBUG","INFO","WARNING","ERROR","CRITICAL"],value="ALL",description="Level:")
search258=widgets.Text(placeholder="Search logs...",description="Search:")
rows258=widgets.Dropdown(options=[25,50,100,250],value=50,description="Rows:")
refresh258=widgets.Button(description="Refresh",icon="refresh")
export258=widgets.Button(description="Export View",icon="save")
out258=widgets.Output()

def table258(d,n=100):
    display(HTML(d.head(n).to_html(index=False,border=0,classes="t258")))

def severity258():
    d=level_summary258[level_summary258.Events>0].sort_values("Events")
    if d.empty: return

    fig=go.Figure()
    for _,r in d.iterrows():
        fig.add_shape(type="line",x0=0,x1=r.Events,y0=r.Level,y1=r.Level,line=dict(color="#c9ada7",width=4))

    fig.add_trace(go.Scatter(
        x=d.Events,y=d.Level,mode="markers+text",text=d.Events,textposition="middle right",
        marker=dict(size=15,color="#4a4e69")
    ))
    fig.update_layout(title="Operational Severity Profile",height=340,showlegend=False,
                      paper_bgcolor="white",plot_bgcolor="white",xaxis_title="Operational Events")
    display(fig)

def timeline_chart258():
    if timeline258.empty: return

    fig=go.Figure(go.Scatter(
        x=timeline258.Time_Bucket,y=timeline258.Events,
        mode="lines",fill="tozeroy",
        line=dict(color="#4a4e69",width=2.5,shape="spline"),
        fillcolor="rgba(154,140,152,.20)"
    ))
    fig.update_layout(title="Operational Logging Activity",height=370,
                      paper_bgcolor="white",plot_bgcolor="white",
                      xaxis_title="Time",yaxis_title="Events")
    display(fig)

def coverage258():
    html258=""
    for _,r in event_coverage258.iterrows():
        pct=100 if r.Status=="OBSERVED" else 25 if r.Status=="PLANNED" else 10
        color="#4a4e69" if r.Status=="OBSERVED" else "#9a8c98" if r.Status=="PLANNED" else "#c9ada7"
        html258+=f"""
        <div style="display:grid;grid-template-columns:180px 1fr 120px;gap:12px;margin:12px;font:9px Arial">
        <b>{r.Category}</b>
        <div style="background:#eee;border-radius:8px;height:9px">
        <div style="background:{color};width:{pct}%;height:100%;border-radius:8px"></div></div>
        <span>{r.Status}</span></div>"""

    display(HTML(html258))

def filtered258():
    d=log_events258.copy()
    if level258.value!="ALL":
        d=d[d.Level==level258.value]
    q=search258.value.strip().lower()
    if q:
        mask=d[["Logger","Source","Message","Event_Category"]].fillna("").astype(str).agg(" ".join,axis=1).str.lower().str.contains(q,regex=False)
        d=d[mask]

    return d

def update258(*_):
    with out258:
        clear_output(wait=True)
        if view258.value=="EXECUTIVE":
            display(HTML(f"""
            <div class="grid258">
            <div class="box258"><b>Operational Health</b><br>Warnings: {warning_count258}<br>Errors: {error_count258}<br>Critical: {critical_count258}</div>
            <div class="box258"><b>Controlled Tests</b><br>{controlled_count258} validation events preserved separately.</div>
            <div class="box258"><b>Security</b><br>Secret Scan: {'PASS' if secret_safe258 else 'REVIEW'}<br>Parse Rate: {parse_rate258}%</div>
            </div>"""))
        elif view258.value=="OPERATIONAL":
            d=operational_events258
            if level258.value!="ALL": d=d[d.Level==level258.value]
            table258(d.tail(rows258.value),rows258.value)

        elif view258.value=="CONTROLLED TESTS":
            display(HTML(f'<div class="note258">{controlled_count258} controlled validation events preserved and excluded from operational KPIs.</div><br>'))
            table258(controlled_test_events258.tail(rows258.value),rows258.value)

        elif view258.value=="SEVERITY":
            severity258()
            table258(level_summary258)

        elif view258.value=="TIMELINE":
            timeline_chart258()
            table258(timeline258.tail(rows258.value))

        elif view258.value=="COVERAGE":
            coverage258()
            display(HTML("<br>"))
            table258(event_coverage258)

        elif view258.value=="AUDIT":
            table258(logging_audit258)

        else:
            files258=pd.DataFrame({
                "Artifact":[
                    "Application Log","All Events","Operational Events","Controlled Tests",
                    "Incidents","Severity","Coverage","Timeline","Audit","Metadata"
                ],
                "Location":[
                    APPLICATION_LOG258,
                    REPORT_DIR/"application_log_all_events.csv",
                    REPORT_DIR/"application_log_operational_events.csv",
                    REPORT_DIR/"application_log_controlled_tests.csv",
                    REPORT_DIR/"application_log_operational_incidents.csv",
                    REPORT_DIR/"application_log_level_summary.csv",
                    REPORT_DIR/"application_log_event_coverage.csv",
                    REPORT_DIR/"application_log_timeline.csv",
                    REPORT_DIR/"application_log_audit.csv",
                    REPORT_DIR/"application_log_metadata.csv"
                ]
            })
            files258["Status"]=files258.Location.apply(lambda x:"PASS" if Path(x).exists() else "REVIEW")
            table258(files258)

def refresh_click258(_):
    rebuild258(); save258()
    header258.value=header258_html()
    kpi258.value=kpi258_html()
    update258()

def export_click258(_):
    filtered258().to_csv(REPORT_DIR/"application_log_filtered_view.csv",index=False)

refresh258.on_click(refresh_click258)
export258.on_click(export_click258)

for w in [view258,level258,search258,rows258]:
    w.observe(update258,names="value")

update258()

dashboard_258=widgets.VBox([
    header258,kpi258,view258,
    widgets.HBox([level258,search258,rows258]),
    widgets.HBox([refresh258,export258]),
    out258
])

dashboard_258.add_class("d258")
display(dashboard_258)
```

<div style="color:#c9ada7;font-size:8px;font-weight:700;letter-spacing:1.7px">SECTION 25.2 • PRE-DEPLOYMENT DOCUMENTATION & REPRODUCIBILITY — CONTINUED</div>

<div style="color:#f2e9e4;font-size:11px;line-height:1.65;margin-top:8px">Validating the reusable environment template for required SQL, model, Flask and Streamlit configuration while checking that passwords, tokens and application secrets are not exposed in the committed template.</div>
</div>

```python
ENV_PATH259=Path(".env.example")
REPORT_DIR=Path("reports")
REPORT_DIR.mkdir(exist_ok=True)

required_keys259=[
    "SQL_SERVER",
    "SQL_DATABASE",
    "SQL_DRIVER",
    "SQL_TRUSTED_CONNECTION",
    "SQL_ENCRYPT",
    "SQL_TRUST_SERVER_CERTIFICATE",
    "FORECAST_MODEL_VERSION",
    "RECOMMENDER_VERSION",
    "API_HOST",
    "API_PORT",
    "API_DEBUG",
    "API_VERSION",
    "JOBMARKET_API_URL",
    "MAX_RECOMMENDATIONS",
    "DEFAULT_RECOMMENDATIONS",
    "REQUEST_TIMEOUT_SECONDS"
]

secret_keys259=[
    "PASSWORD",
    "PWD",
    "API_KEY",
    "SECRET_KEY",
    "ACCESS_TOKEN",
    "CLIENT_SECRET"
]


env_text259=(
    ENV_PATH259.read_text(
        encoding="utf-8",
        errors="ignore"
    )
    if ENV_PATH259.exists()
    else ""
)


env_values259={}

for line in env_text259.splitlines():

    line=line.strip()

    if (
        line
        and not line.startswith("#")
        and "=" in line
    ):

        key,value=line.split(
            "=",
            1
        )

        env_values259[
            key.strip()
        ]=value.strip()


env_validation259=pd.DataFrame([
    {
        "Variable":key,
        "Present":key in env_values259,
        "Has_Value":bool(
            env_values259.get(
                key,
                ""
            ).strip()
        ),
        "Status":
            "PASS"
            if (
                key in env_values259
                and bool(
                    env_values259.get(
                        key,
                        ""
                    ).strip()
                )
            )
            else "REVIEW"
    }
    for key in required_keys259
])


secret_hits259=[]

for key,value in env_values259.items():

    if any(
        secret in key.upper()
        for secret in secret_keys259
    ) and value.strip():

        secret_hits259.append({
            "Variable":key,
            "Value_Present":True
        })


secret_scan259=pd.DataFrame(
    secret_hits259,
    columns=[
        "Variable",
        "Value_Present"
    ]
)


ENV_SECRET_STATUS259=(
    "PASS"
    if secret_scan259.empty
    else "FAIL"
)


ENV_COMPLETENESS259=round(
    env_validation259[
        "Status"
    ].eq(
        "PASS"
    ).mean()*100,
    1
)


ENV_STATUS259=(
    "PASS"
    if (
        ENV_PATH259.exists()
        and
        ENV_COMPLETENESS259==100
        and
        ENV_SECRET_STATUS259=="PASS"
    )
    else "REVIEW"
)


env_audit259=pd.DataFrame([
    {
        "Control":"Environment template exists",
        "Status":
            "PASS"
            if ENV_PATH259.exists()
            else "FAIL",
        "Evidence":str(ENV_PATH259)
    },
    {
        "Control":"Required variables documented",
        "Status":
            "PASS"
            if ENV_COMPLETENESS259==100
            else "REVIEW",
        "Evidence":
            f"{ENV_COMPLETENESS259}% complete"
    },
    {
        "Control":"No secret-bearing variables populated",
        "Status":ENV_SECRET_STATUS259,
        "Evidence":
            "No populated secret variables"
            if secret_scan259.empty
            else f"{len(secret_scan259)} potential secret variable(s)"
    },
    {
        "Control":"SQL configuration documented",
        "Status":
            "PASS"
            if all(
                k in env_values259
                for k in [
                    "SQL_SERVER",
                    "SQL_DATABASE",
                    "SQL_DRIVER"
                ]
            )
            else "REVIEW",
        "Evidence":"SQL Server environment keys"
    },
    {
        "Control":"Flask configuration documented",
        "Status":
            "PASS"
            if all(
                k in env_values259
                for k in [
                    "API_HOST",
                    "API_PORT",
                    "API_VERSION"
                ]
            )
            else "REVIEW",
        "Evidence":"API environment keys"
    },
    {
        "Control":"Streamlit API endpoint documented",
        "Status":
            "PASS"
            if "JOBMARKET_API_URL"
            in env_values259
            else "REVIEW",
        "Evidence":"JOBMARKET_API_URL"
    }
])


env_validation259.to_csv(
    REPORT_DIR/
    "environment_template_validation.csv",
    index=False
)

env_audit259.to_csv(
    REPORT_DIR/
    "environment_template_audit.csv",
    index=False
)

secret_scan259.to_csv(
    REPORT_DIR/
    "environment_secret_scan.csv",
    index=False
)


env_metadata259=pd.DataFrame([{
    "Template":str(ENV_PATH259),
    "Required_Variables":len(required_keys259),
    "Variables_Present":
        int(
            env_validation259[
                "Present"
            ].sum()
        ),
    "Completeness_%":
        ENV_COMPLETENESS259,
    "Secret_Scan":
        ENV_SECRET_STATUS259,
    "Validation_Status":
        ENV_STATUS259
}])


env_metadata259.to_csv(
    REPORT_DIR/
    "environment_template_metadata.csv",
    index=False
)


def table259(d):
    display(
        HTML(
            d.to_html(
                index=False,
                border=0,
                classes="t259"
            )
        )
    )


display(
    HTML("""
<style>
.d259{border:1px solid #ddd;border-radius:17px;overflow:hidden;background:#fff;box-shadow:0 8px 22px #22223b18}
.h259{background:linear-gradient(120deg,#22223b,#4a4e69);padding:23px 27px;color:#fff;font-family:Arial}
.h259 small{color:#c9ada7;font-weight:700;letter-spacing:1.4px}.h259 h2{margin:6px 0}.h259 p{font-size:10px;color:#f2e9e4}
.cards259{display:grid;grid-template-columns:repeat(4,1fr)}.c259{padding:14px;border-right:1px solid #eee;font-family:Arial}
.c259 small{font-size:7px;color:#9a8c98;font-weight:700}.c259 b{display:block;font-size:16px;color:#22223b;margin-top:5px}.c259 span{font-size:8px;color:#4a4e69}
.out259{padding:20px}.t259{width:100%;border-collapse:collapse;font:9px Arial}.t259 th{background:#22223b;color:#fff;padding:8px;text-align:left}.t259 td{padding:8px;border-bottom:1px solid #eee}
.note259{background:#f2e9e4;padding:12px;border-radius:8px;font:10px/1.6 Arial;color:#4a4e69}
</style>
""")
)


header259=widgets.HTML(
    f"""
<div class="h259">
<small>SECTION 25.2 • CONFIGURATION SECURITY</small>
<h2>Environment Configuration Audit</h2>
<p>Completeness • SQL • Flask • Streamlit • Secret Safety</p>
<b>{ENV_STATUS259}</b>
</div>
"""
)


kpi259=widgets.HTML(
    '<div class="cards259">'+
    ''.join([
        f'<div class="c259"><small>STATUS</small><b>{ENV_STATUS259}</b><span>Environment gate</span></div>',
        f'<div class="c259"><small>COMPLETENESS</small><b>{ENV_COMPLETENESS259}%</b><span>Required keys</span></div>',
        f'<div class="c259"><small>VARIABLES</small><b>{len(env_values259)}</b><span>Documented</span></div>',
        f'<div class="c259"><small>SECRET SCAN</small><b>{ENV_SECRET_STATUS259}</b><span>Template safety</span></div>'
    ])+
    '</div>'
)


view259=widgets.ToggleButtons(
    options=[
        "OVERVIEW",
        "VARIABLES",
        "AUDIT",
        "SECRET SCAN",
        "FILES"
    ],
    value="OVERVIEW",
    layout=widgets.Layout(width="100%")
)


out259=widgets.Output()


def update259(*_):

    with out259:

        clear_output(
            wait=True
        )

        if view259.value=="OVERVIEW":

            display(
                HTML(
                    f"""
<div class="note259">
<b>Environment status:</b> {ENV_STATUS259}<br><br>
Required configuration documented:
<b>{ENV_COMPLETENESS259}%</b><br>
Secret-bearing values detected:
<b>{len(secret_scan259)}</b><br><br>
The template is intended for configuration examples only. Real credentials should be provided through runtime environment variables and should not be committed to source control.
</div>
"""
                )
            )

        elif view259.value=="VARIABLES":

            table259(
                env_validation259
            )

        elif view259.value=="AUDIT":

            table259(
                env_audit259
            )

        elif view259.value=="SECRET SCAN":

            if secret_scan259.empty:

                display(
                    HTML(
                        """
<div class="note259">
<b>PASS</b><br><br>
No populated password, token, API-key or secret-key variables were detected in <code>.env.example</code>.
</div>
"""
                    )
                )

            else:

                table259(
                    secret_scan259
                )

        else:

            files259=pd.DataFrame({
                "Artifact":[
                    "Environment Template",
                    "Validation Report",
                    "Audit Report",
                    "Secret Scan",
                    "Metadata"
                ],
                "Location":[
                    ENV_PATH259,
                    REPORT_DIR/"environment_template_validation.csv",
                    REPORT_DIR/"environment_template_audit.csv",
                    REPORT_DIR/"environment_secret_scan.csv",
                    REPORT_DIR/"environment_template_metadata.csv"
                ]
            })

            files259[
                "Status"
            ]=files259[
                "Location"
            ].apply(
                lambda x:
                "PASS"
                if Path(x).exists()
                else "REVIEW"
            )

            table259(
                files259
            )


view259.observe(
    update259,
    names="value"
)

update259()
dashboard_259=widgets.VBox([
    header259,
    kpi259,
    view259,
    out259
])

dashboard_259.add_class("d259")
display(dashboard_259)
```

<div style="color:#c9ada7;font-size:8px;font-weight:700;letter-spacing:1.7px">SECTION 25.2 • PRE-DEPLOYMENT DOCUMENTATION & REPRODUCIBILITY — CONTINUED</div>
<div style="color:#fff;font-size:26px;font-weight:700;margin-top:6px">Repository Hygiene & Git Safety Controls</div>
<div style="color:#f2e9e4;font-size:11px;line-height:1.65;margin-top:8px">Protecting local environments, credentials, caches and runtime files while preserving reproducible source code, configuration templates, model assets and client evidence.</div>
</div>

```python
GITIGNORE_PATH260=Path(".gitignore")
REPORT_DIR=Path("reports")
REPORT_DIR.mkdir(exist_ok=True)

required_rules260=[
    ".venv/",
    "venv/",
    "__pycache__/",
    "*.py[cod]",
    ".ipynb_checkpoints/",
    ".env",
    ".env.*",
    "!.env.example",
    "logs/*.log",
    "*.tmp",
    "*.temp",
    "*.bak",
    ".DS_Store",
    "Thumbs.db",
    ".vscode/",
    ".idea/"
]

existing260=(
    GITIGNORE_PATH260.read_text(
        encoding="utf-8",
        errors="ignore"
    ).splitlines()
    if GITIGNORE_PATH260.exists()
    else []
)

clean_existing260=[
    x.rstrip()
    for x in existing260
]

missing260=[
    rule
    for rule in required_rules260
    if rule not in clean_existing260
]

if missing260:

    with GITIGNORE_PATH260.open(
        "a",
        encoding="utf-8"
    ) as f:

        if existing260 and existing260[-1].strip():
            f.write("\n")

        f.write(
            "\n# Local environment and runtime safety\n"
        )

        for rule in missing260:
            f.write(
                rule+"\n"
            )


final_rules260=GITIGNORE_PATH260.read_text(
    encoding="utf-8",
    errors="ignore"
).splitlines()


rule_audit260=pd.DataFrame([
    {
        "Rule":rule,
        "Present":
            rule in final_rules260,
        "Status":
            "PASS"
            if rule in final_rules260
            else "FAIL"
    }
    for rule in required_rules260
])


sensitive_candidates260=[
    Path(".env"),
    Path("secrets.json"),
    Path("credentials.json"),
    Path("config/secrets.json"),
    Path("config/credentials.json")
]


sensitive_audit260=pd.DataFrame([
    {
        "Potential_Sensitive_File":
            str(p),

        "Exists":
            p.exists(),

        "Expected_Action":
            "Keep local / do not commit"
    }
    for p in sensitive_candidates260
])


protected_template260=(
    Path(".env.example").exists()
    and
    "!.env.example" in final_rules260
)


GITIGNORE_STATUS260=(
    "PASS"
    if (
        rule_audit260[
            "Status"
        ].eq("PASS").all()
        and protected_template260
    )
    else "REVIEW"
)


repo_audit260=pd.DataFrame([
    {
        "Control":
            ".gitignore exists",
        "Status":
            "PASS"
            if GITIGNORE_PATH260.exists()
            else "FAIL",
        "Evidence":
            str(GITIGNORE_PATH260)
    },
    {
        "Control":
            "Virtual environments excluded",
        "Status":
            "PASS"
            if all(
                x in final_rules260
                for x in [
                    ".venv/",
                    "venv/"
                ]
            )
            else "REVIEW",
        "Evidence":
            ".venv/ + venv/"
    },
    {
        "Control":
            "Local .env excluded",
        "Status":
            "PASS"
            if ".env" in final_rules260
            else "REVIEW",
        "Evidence":
            ".env"
    },
    {
        "Control":
            ".env.example retained",
        "Status":
            "PASS"
            if protected_template260
            else "REVIEW",
        "Evidence":
            "!.env.example"
    },
    {
        "Control":
            "Python/Jupyter caches excluded",
        "Status":
            "PASS"
            if all(
                x in final_rules260
                for x in [
                    "__pycache__/",
                    ".ipynb_checkpoints/"
                ]
            )
            else "REVIEW",
        "Evidence":
            "Cache rules"
    },
    {
        "Control":
            "Runtime logs excluded from Git",
        "Status":
            "PASS"
            if "logs/*.log"
            in final_rules260
            else "REVIEW",
        "Evidence":
            "logs/*.log"
    }
])


repo_audit260.to_csv(
    REPORT_DIR/
    "repository_hygiene_audit.csv",
    index=False
)

rule_audit260.to_csv(
    REPORT_DIR/
    "gitignore_rule_validation.csv",
    index=False
)

sensitive_audit260.to_csv(
    REPORT_DIR/
    "sensitive_file_presence_audit.csv",
    index=False
)


repo_metadata260=pd.DataFrame([{
    "Gitignore":str(GITIGNORE_PATH260),
    "Required_Rules":len(required_rules260),
    "Rules_Present":
        int(
            rule_audit260[
                "Present"
            ].sum()
        ),
    "Rules_Added":
        len(missing260),
    "Env_Example_Preserved":
        protected_template260,
    "Status":
        GITIGNORE_STATUS260
}])


repo_metadata260.to_csv(
    REPORT_DIR/
    "repository_hygiene_metadata.csv",
    index=False
)


display(
    HTML("""
<style>
.d260{border:1px solid #ddd;border-radius:17px;overflow:hidden;background:#fff;box-shadow:0 8px 22px #22223b18}
.h260{background:linear-gradient(120deg,#22223b,#4a4e69);padding:23px 27px;color:#fff;font-family:Arial}
.h260 small{color:#c9ada7;font-weight:700;letter-spacing:1.4px}.h260 h2{margin:6px 0}.h260 p{font-size:10px;color:#f2e9e4}
.cards260{display:grid;grid-template-columns:repeat(4,1fr)}.c260{padding:14px;border-right:1px solid #eee;font-family:Arial}
.c260 small{font-size:7px;color:#9a8c98;font-weight:700}.c260 b{display:block;font-size:16px;color:#22223b;margin-top:5px}.c260 span{font-size:8px;color:#4a4e69}
.out260{padding:20px}.t260{width:100%;border-collapse:collapse;font:9px Arial}.t260 th{background:#22223b;color:#fff;padding:8px;text-align:left}.t260 td{padding:8px;border-bottom:1px solid #eee}
.note260{background:#f2e9e4;padding:12px;border-radius:8px;font:10px/1.6 Arial;color:#4a4e69}
</style>
""")
)


header260=widgets.HTML(
    f"""
<div class="h260">
<small>SECTION 25.2 • REPOSITORY CONTROL</small>
<h2>Git & Repository Hygiene</h2>
<p>Environment Safety • Cache Exclusion • Runtime Protection • Reproducibility</p>
<b>{GITIGNORE_STATUS260}</b>
</div>
"""
)


kpi260=widgets.HTML(
    '<div class="cards260">'+
    f'<div class="c260"><small>STATUS</small><b>{GITIGNORE_STATUS260}</b><span>Repository gate</span></div>'+
    f'<div class="c260"><small>RULES</small><b>{len(required_rules260)}</b><span>Required</span></div>'+
    f'<div class="c260"><small>ADDED</small><b>{len(missing260)}</b><span>This run</span></div>'+
    f'<div class="c260"><small>ENV TEMPLATE</small><b>{"PASS" if protected_template260 else "REVIEW"}</b><span>.env.example retained</span></div>'+
    '</div>'
)


view260=widgets.ToggleButtons(
    options=[
        "OVERVIEW",
        "RULES",
        "AUDIT",
        "SENSITIVE FILES",
        "FILES"
    ],
    value="OVERVIEW",
    layout=widgets.Layout(width="100%")
)

out260=widgets.Output()


def table260(d):
    display(
        HTML(
            d.to_html(
                index=False,
                border=0,
                classes="t260"
            )
        )
    )


def update260(*_):

    with out260:

        clear_output(wait=True)

        if view260.value=="OVERVIEW":

            display(
                HTML(
                    f"""
<div class="note260">
<b>Repository status:</b> {GITIGNORE_STATUS260}<br><br>

The project now excludes local virtual environments, Python/Jupyter caches,
runtime logs, temporary files and the real <code>.env</code> file.

The reusable <code>.env.example</code> remains available for reproducible deployment configuration.

<br><br>

<b>Important:</b> model artifacts, reports, source code and client evidence are not globally excluded by this control.
</div>
"""
                )
            )

        elif view260.value=="RULES":

            table260(
                rule_audit260
            )

        elif view260.value=="AUDIT":

            table260(
                repo_audit260
            )

        elif view260.value=="SENSITIVE FILES":

            table260(
                sensitive_audit260
            )

        else:

            files260=pd.DataFrame({
                "Artifact":[
                    "Gitignore",
                    "Repository Audit",
                    "Rule Validation",
                    "Sensitive File Audit",
                    "Metadata"
                ],
                "Location":[
                    GITIGNORE_PATH260,
                    REPORT_DIR/"repository_hygiene_audit.csv",
                    REPORT_DIR/"gitignore_rule_validation.csv",
                    REPORT_DIR/"sensitive_file_presence_audit.csv",
                    REPORT_DIR/"repository_hygiene_metadata.csv"
                ]
            })

            files260[
                "Status"
            ]=files260[
                "Location"
            ].apply(
                lambda x:
                "PASS"
                if Path(x).exists()
                else "REVIEW"
            )

            table260(
                files260
            )


view260.observe(
    update260,
    names="value"
)

update260()


dashboard_260=widgets.VBox([
    header260,
    kpi260,
    view260,
    out260
])

dashboard_260.add_class("d260")

display(dashboard_260)
```

```python
FEATURE_XLSX263=REPORT_DIR/"feature_dictionary.xlsx"
FEATURE_CSV263=REPORT_DIR/"feature_dictionary.csv"

meta263={
    "title":["Source","Original job-posting title.","Raw dataset","None","Reference / display"],
    "link":["Source","Job-posting URL.","Raw dataset","None","Job navigation / traceability"],
    "published_date":["Time","Job publication timestamp.","Raw dataset","Datetime parsing","Trend / forecasting"],
    "country":["Geography","Original country value.","Raw dataset","None","Source reference"],
    "country_clean":["Geography","Standardized country label.","country","Missing/label standardization","Country analysis"],
    "is_hourly":["Compensation","Hourly-pay indicator.","Raw dataset","Boolean normalization","Pay-type separation"],
    "hourly_low":["Compensation","Lower hourly pay bound.","Raw dataset","Numeric cleaning","Hourly analysis only"],
    "hourly_high":["Compensation","Upper hourly pay bound.","Raw dataset","Numeric cleaning","Hourly analysis only"],
    "hourly_midpoint":["Compensation","Midpoint of valid hourly bounds.","hourly_low + hourly_high","(low + high) / 2","Hourly analysis only"],
    "budget":["Compensation","Fixed-price job budget.","Raw dataset","Numeric cleaning","Fixed-pay analysis only"],
    "fixed_budget":["Compensation","Validated fixed-price budget.","budget","Pay-type filtering / cleaning","Fixed-pay analysis only"],
    "pay_type":["Compensation","Hourly versus fixed classification.","is_hourly","Derived category","Compensation separation"],
    "title_normalized":["Text","Normalized job-title text.","title","Text normalization","Recommendation / role extraction"],
    "job_family":["Role","Derived broad job-family classification.","title_normalized","Title-based classification","Forecasting / segmentation"],
    "seniority":["Role","Title-derived seniority indicator.","title_normalized","Keyword classification","Recommendation preference"],
    "job_id":["Identifier","Stable job-level identifier.","Processed dataset","Generated identifier","Recommendation traceability"]
}

def group263(c):
    x=c.lower()
    if any(k in x for k in ["date","day","month","year"]): return "Time"
    if any(k in x for k in ["hourly","budget","pay"]): return "Compensation"
    if "country" in x: return "Geography"
    if any(k in x for k in ["title","text"]): return "Text"
    if any(k in x for k in ["family","seniority"]): return "Role"
    if any(k in x for k in ["id","link"]): return "Identifier"
    return "Derived"

feature_dictionary263=pd.DataFrame([
    {
        "Feature":c,
        "Data_Type":str(df_final[c].dtype),
        "Group":meta263.get(c,[group263(c)])[0],
        "Description":meta263.get(c,[None,f"Processed analytical field: {c}."])[1],
        "Source":meta263.get(c,[None,None,"Derived / processed dataset"])[2],
        "Transformation":meta263.get(c,[None,None,None,"See feature-engineering section"])[3],
        "Analytical_Use":meta263.get(c,[None,None,None,None,"Supporting analytical feature"])[4],
        "Missing_%":round(df_final[c].isna().mean()*100,2),
        "Unique_Values":int(df_final[c].nunique(dropna=True)),
        "Nullable":"YES" if df_final[c].isna().any() else "NO"
    }
    for c in df_final.columns
])

feature_dictionary263["Governance_Note"]=feature_dictionary263.Feature.map({
    "hourly_low":"Do not combine with fixed budgets.",
    "hourly_high":"Do not combine with fixed budgets.",
    "hourly_midpoint":"Hourly compensation only.",
    "budget":"Fixed-price compensation only.",
    "fixed_budget":"Fixed-price compensation only.",
    "country":"Missing values exist; do not infer currency.",
    "country_clean":"Geography is not currency evidence.",
    "job_family":"Title-derived; not employer-verified.",
    "seniority":"Title-derived; not verified experience."
}).fillna("—")

governance263=pd.DataFrame([
    ["Hourly vs Fixed Pay","Never combine hourly compensation with fixed-price budgets."],
    ["Currency","No currency normalization because currency is unavailable."],
    ["Job Family","Derived from job-title text."],
    ["Seniority","Derived from title keywords; not verified experience."],
    ["Remote Status","Not directly supported by source data."],
    ["Skills","Explicit structured skills are unavailable."]
],columns=["Rule","Interpretation"])

feature_dictionary263.to_csv(
    FEATURE_CSV263,
    index=False
)

with pd.ExcelWriter(
    FEATURE_XLSX263,
    engine="openpyxl"
) as writer:

    feature_dictionary263.to_excel(
        writer,
        sheet_name="Feature Dictionary",
        index=False
    )

    governance263.to_excel(
        writer,
        sheet_name="Governance Rules",
        index=False
    )

dictionary_audit263=pd.DataFrame([
    ["Feature dictionary created",FEATURE_CSV263.exists() and FEATURE_XLSX263.exists()],
    ["Every df_final column documented",set(df_final.columns)==set(feature_dictionary263.Feature)],
    ["Data type documented",feature_dictionary263.Data_Type.notna().all()],
    ["Source documented",feature_dictionary263.Source.notna().all()],
    ["Transformation documented",feature_dictionary263.Transformation.notna().all()],
    ["Missingness documented",feature_dictionary263["Missing_%"].notna().all()]
],columns=["Control","Passed"])

dictionary_audit263["Status"]=np.where(
    dictionary_audit263.Passed,
    "PASS",
    "REVIEW"
)

FEATURE_DICTIONARY_STATUS263=(
    "PASS"
    if dictionary_audit263.Passed.all()
    else "REVIEW"
)

dictionary_audit263.to_csv(
    REPORT_DIR/"feature_dictionary_validation.csv",
    index=False
)

group_summary263=(
    feature_dictionary263
    .groupby("Group",as_index=False)
    .agg(
        Features=("Feature","count"),
        Avg_Missing_Pct=("Missing_%","mean")
    )
)

group_summary263["Avg_Missing_Pct"]=group_summary263[
    "Avg_Missing_Pct"
].round(2)

nullable263=int(
    feature_dictionary263.Nullable.eq("YES").sum()
)

complete263=int(
    feature_dictionary263[
        ["Data_Type","Description","Source","Transformation","Analytical_Use"]
    ].notna().all(axis=1).sum()
)

coverage263=round(
    complete263/max(len(feature_dictionary263),1)*100,
    1
)

max_missing263=float(
    feature_dictionary263["Missing_%"].max()
)

display(HTML("""
<style>
.d263{border:1px solid #ded8d5;border-radius:18px;overflow:hidden;background:#fff;box-shadow:0 10px 28px #22223b18}
.h263{background:linear-gradient(120deg,#22223b,#30324d 55%,#4a4e69);padding:24px 28px;color:#fff;font-family:Arial;display:flex;justify-content:space-between}
.h263 small{color:#c9ada7;font-weight:700;letter-spacing:1.5px;font-size:8px}.h263 h2{margin:6px 0;font-size:25px}.h263 p{font-size:10px;color:#f2e9e4;margin:0}
.s263{text-align:right;font:9px Arial;color:#f2e9e4}.s263 b{display:block;font-size:19px;color:#fff;margin-top:4px}
.cards263{display:grid;grid-template-columns:repeat(6,1fr);border-bottom:1px solid #eee}.c263{padding:14px;border-right:1px solid #eee;font-family:Arial}
.c263 small{font-size:7px;color:#9a8c98;font-weight:700}.c263 b{display:block;font-size:17px;color:#22223b;margin-top:5px}.c263 span{font-size:8px;color:#4a4e69}
.out263{padding:20px}.t263{width:100%;border-collapse:collapse;font:9px Arial}.t263 th{background:#22223b;color:#fff;padding:9px;text-align:left}.t263 td{padding:8px;border-bottom:1px solid #eee;vertical-align:top}.t263 tr:nth-child(even){background:#fbfaf9}
.grid263{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}.box263{background:#f8f6f5;border-left:4px solid #4a4e69;padding:14px;border-radius:9px;font:10px/1.65 Arial;color:#22223b}
.box263 b{display:block;color:#4a4e69;font-size:9px;margin-bottom:5px}.note263{background:#f2e9e4;padding:12px 14px;border-radius:9px;font:10px/1.65 Arial;color:#4a4e69}
.group263{display:grid;grid-template-columns:160px 60px 1fr 90px;gap:12px;align-items:center;padding:11px;border-bottom:1px solid #eee;font:9px Arial}
.bar263{height:8px;background:#eee9e7;border-radius:8px;overflow:hidden}.fill263{height:100%;background:#4a4e69;border-radius:8px}
.badge263{display:inline-block;padding:4px 8px;border-radius:12px;background:#fff;font:8px Arial;color:#4a4e69;margin:3px}
</style>
"""))

header263=widgets.HTML(f"""
<div class="h263">

<div>
<small>SECTION 25.2 • DATA GOVERNANCE</small>
<h2>Feature Dictionary Control Center</h2>
<p>Lineage • Types • Transformations • Missingness • Usage • Governance</p>
</div>

<div class="s263">
DICTIONARY STATUS
<b>{FEATURE_DICTIONARY_STATUS263}</b>
Coverage {coverage263}%
</div>

</div>
""")


kpi263=widgets.HTML(
    '<div class="cards263">'+
    f'<div class="c263"><small>STATUS</small><b>{FEATURE_DICTIONARY_STATUS263}</b><span>Governance gate</span></div>'+
    f'<div class="c263"><small>FEATURES</small><b>{len(feature_dictionary263)}</b><span>Documented</span></div>'+
    f'<div class="c263"><small>GROUPS</small><b>{feature_dictionary263.Group.nunique()}</b><span>Domains</span></div>'+
    f'<div class="c263"><small>COVERAGE</small><b>{coverage263}%</b><span>Metadata</span></div>'+
    f'<div class="c263"><small>NULLABLE</small><b>{nullable263}</b><span>Missing-data fields</span></div>'+
    f'<div class="c263"><small>MAX MISSING</small><b>{max_missing263:.1f}%</b><span>Highest field</span></div>'+
    '</div>'
)


view263=widgets.ToggleButtons(
    options=[
        ("EXECUTIVE","overview"),
        ("FEATURES","features"),
        ("GROUP PROFILE","groups"),
        ("DATA QUALITY","quality"),
        ("GOVERNANCE","governance"),
        ("VALIDATION","validation"),
        ("FILES","files")
    ],
    value="overview",
    layout=widgets.Layout(width="100%"),
    style={"button_width":"14.2%"}
)


group_filter263=widgets.Dropdown(
    options=["ALL"]+sorted(
        feature_dictionary263.Group.unique()
    ),
    value="ALL",
    description="Group:"
)

search263=widgets.Text(
    placeholder="Search feature, source or use...",
    description="Search:"
)

out263=widgets.Output()


def table263(d):
    display(
        HTML(
            d.to_html(
                index=False,
                border=0,
                classes="t263"
            )
        )
    )


def update263(*_):

    with out263:

        clear_output(wait=True)

        if view263.value=="overview":

            display(HTML(f"""
<div class="grid263">

<div class="box263">
<b>DATA CONTRACT</b>
All {len(feature_dictionary263)} analytical fields are documented with lineage,
datatype, transformation and intended use.
</div>

<div class="box263">
<b>COMPENSATION CONTROL</b>
Hourly and fixed-price compensation remain separate analytical domains.
</div>

<div class="box263">
<b>DERIVED ATTRIBUTES</b>
Role family and seniority are title-derived analytical classifications,
not employer-verified attributes.
</div>

</div>

<br>

<div class="note263">
<b>Feature domains</b><br>
{"".join(f'<span class="badge263">{x}</span>' for x in sorted(feature_dictionary263.Group.unique()))}
</div>
"""))


        elif view263.value=="features":

            d=feature_dictionary263.copy()

            if group_filter263.value!="ALL":
                d=d[d.Group==group_filter263.value]

            q=search263.value.strip().lower()

            if q:
                d=d[
                    d.astype(str)
                    .agg(" ".join,axis=1)
                    .str.lower()
                    .str.contains(q,regex=False)
                ]

            display(
                HTML(
                    f'<div class="note263">Showing <b>{len(d)}</b> of <b>{len(feature_dictionary263)}</b> features.</div><br>'
                )
            )

            table263(d)


        elif view263.value=="groups":

            m=max(
                group_summary263.Features.max(),
                1
            )

            display(
                HTML(
                    "".join(
                        f"""
<div class="group263">
<b>{r.Group}</b>
<span>{r.Features}</span>
<div class="bar263">
<div class="fill263" style="width:{r.Features/m*100:.0f}%"></div>
</div>
<span>Missing {r.Avg_Missing_Pct}%</span>
</div>
"""
                        for _,r
                        in group_summary263.iterrows()
                    )
                )
            )


        elif view263.value=="quality":

            quality263=feature_dictionary263[
                [
                    "Feature",
                    "Group",
                    "Missing_%",
                    "Unique_Values",
                    "Nullable"
                ]
            ].sort_values(
                "Missing_%",
                ascending=False
            )

            display(HTML(f"""
<div class="grid263">

<div class="box263">
<b>HIGHEST MISSINGNESS</b>
{quality263.iloc[0].Feature if len(quality263) else "—"}:
{max_missing263:.2f}%
</div>

<div class="box263">
<b>COMPLETE FIELDS</b>
{int((feature_dictionary263["Missing_%"]==0).sum())}
of {len(feature_dictionary263)}
</div>

<div class="box263">
<b>NULLABLE FIELDS</b>
{nullable263}
</div>

</div>

<br>
"""))

            table263(quality263)


        elif view263.value=="governance":

            display(
                HTML(
                    """
<div class="note263">
<b>Interpretation controls:</b>
The following fields carry explicit restrictions to prevent misleading
compensation, geography, role or experience claims.
</div>
<br>
"""
                )
            )

            table263(
                feature_dictionary263[
                    feature_dictionary263.Governance_Note!="—"
                ][
                    [
                        "Feature",
                        "Group",
                        "Analytical_Use",
                        "Governance_Note"
                    ]
                ]
            )


        elif view263.value=="validation":

            table263(
                dictionary_audit263[
                    [
                        "Control",
                        "Status"
                    ]
                ]
            )


        else:

            files263=pd.DataFrame({
                "Artifact":[
                    "Feature Dictionary CSV",
                    "Feature Dictionary Excel",
                    "Validation Evidence"
                ],
                "Location":[
                    FEATURE_CSV263,
                    FEATURE_XLSX263,
                    REPORT_DIR/"feature_dictionary_validation.csv"
                ]
            })

            files263["Status"]=files263.Location.apply(
                lambda p:
                "PASS"
                if Path(p).exists()
                else "REVIEW"
            )

            table263(files263)


for w in [
    view263,
    group_filter263,
    search263
]:
    w.observe(
        update263,
        names="value"
    )

update263()

dashboard_263=widgets.VBox([
    header263,
    kpi263,
    view263,
    widgets.HBox([
        group_filter263,
        search263
    ]),
    out263
])

dashboard_263.add_class(
    "d263"
)

display(
    dashboard_263
)
```

```python
AP_CSV264=REPORT_DIR/"task_03_actual_vs_predicted.csv"
AP_PNG264=Path("charts")/"task_03_actual_vs_predicted.png"
AP_PNG264.parent.mkdir(exist_ok=True)

actual_pred264=model_errors18.copy()

def pick264(names):
    return next((c for c in names if c in actual_pred264.columns),None)

date264=pick264(["date","Date","forecast_day","test_date","published_date"])
role264=pick264(["job_family","Job_Family","role","Role"])
actual264=pick264(["Actual","actual","y_true","Observed","observed"])
pred264=pick264(["Predicted","predicted","y_pred","Forecast","forecast"])
model264=pick264(["Selected_Model","Model","model"])

if not all([date264,role264,actual264,pred264]):
    raise ValueError("model_errors18 must contain date, job-family, actual and predicted fields.")

actual_pred264=actual_pred264.rename(columns={
    date264:"Date",
    role264:"Job_Family",
    actual264:"Actual",
    pred264:"Predicted"
})

if model264 and model264!="Selected_Model":
    actual_pred264=actual_pred264.rename(columns={model264:"Selected_Model"})

actual_pred264["Date"]=pd.to_datetime(actual_pred264["Date"],errors="coerce")
actual_pred264=actual_pred264.dropna(subset=["Date","Job_Family","Actual","Predicted"]).copy()
actual_pred264["Error"]=actual_pred264["Actual"]-actual_pred264["Predicted"]
actual_pred264["Absolute_Error"]=actual_pred264["Error"].abs()
actual_pred264.to_csv(AP_CSV264,index=False)

role_metrics264=(
    actual_pred264.groupby("Job_Family")
    .apply(lambda d:pd.Series({
        "Observations":len(d),
        "MAE":d["Absolute_Error"].mean(),
        "RMSE":np.sqrt(np.mean(d["Error"]**2)),
        "WAPE_%":d["Absolute_Error"].sum()/max(d["Actual"].abs().sum(),1)*100
    }),include_groups=False)
    .reset_index()
)

role_metrics264[["MAE","RMSE","WAPE_%"]]=role_metrics264[["MAE","RMSE","WAPE_%"]].round(2)

overall_mae264=actual_pred264["Absolute_Error"].mean()
overall_rmse264=np.sqrt(np.mean(actual_pred264["Error"]**2))
overall_wape264=actual_pred264["Absolute_Error"].sum()/max(actual_pred264["Actual"].abs().sum(),1)*100

plot264=actual_pred264.groupby("Date",as_index=False)[["Actual","Predicted"]].sum()

fig264,ax264=plt.subplots(figsize=(11,5))
ax264.plot(plot264["Date"],plot264["Actual"],marker="o",label="Actual")
ax264.plot(plot264["Date"],plot264["Predicted"],marker="o",label="Predicted")
ax264.set_title("Task 3 — Actual vs Predicted Demand")
ax264.set_xlabel("Validation Date")
ax264.set_ylabel("Job Postings")
ax264.legend()
ax264.grid(alpha=.2)
fig264.tight_layout()
fig264.savefig(AP_PNG264,dpi=180,bbox_inches="tight")
plt.close(fig264)

forecast_evidence264=pd.DataFrame([
    ["Actual-vs-predicted CSV",AP_CSV264.exists()],
    ["Actual-vs-predicted PNG",AP_PNG264.exists()],
    ["Chronological dates available",actual_pred264["Date"].notna().all()],
    ["Actual values available",actual_pred264["Actual"].notna().all()],
    ["Predictions available",actual_pred264["Predicted"].notna().all()]
],columns=["Control","Passed"])

forecast_evidence264["Status"]=np.where(forecast_evidence264.Passed,"PASS","REVIEW")
FORECAST_EVIDENCE_STATUS264="PASS" if forecast_evidence264.Passed.all() else "REVIEW"

role_metrics264.to_csv(REPORT_DIR/"task_03_actual_vs_predicted_metrics.csv",index=False)
forecast_evidence264.to_csv(REPORT_DIR/"task_03_actual_vs_predicted_validation.csv",index=False)

display(HTML("""
<style>
.d264{border:1px solid #ded8d5;border-radius:18px;overflow:hidden;background:#fff;box-shadow:0 10px 28px #22223b18}
.h264{background:linear-gradient(120deg,#22223b,#30324d 55%,#4a4e69);padding:24px 28px;color:#fff;font-family:Arial;display:flex;justify-content:space-between}
.h264 small{color:#c9ada7;font-weight:700;letter-spacing:1.5px;font-size:8px}.h264 h2{margin:6px 0;font-size:25px}.h264 p{font-size:10px;color:#f2e9e4;margin:0}
.s264{text-align:right;font:9px Arial}.s264 b{display:block;font-size:19px;margin-top:4px}
.cards264{display:grid;grid-template-columns:repeat(6,1fr);border-bottom:1px solid #eee}.c264{padding:14px;border-right:1px solid #eee;font-family:Arial}
.c264 small{font-size:7px;color:#9a8c98;font-weight:700}.c264 b{display:block;font-size:17px;color:#22223b;margin-top:5px}.c264 span{font-size:8px;color:#4a4e69}
.out264{padding:20px}.t264{width:100%;border-collapse:collapse;font:9px Arial}.t264 th{background:#22223b;color:#fff;padding:9px;text-align:left}.t264 td{padding:8px;border-bottom:1px solid #eee}
.note264{background:#f2e9e4;padding:12px 14px;border-radius:9px;font:10px/1.6 Arial;color:#4a4e69}
.grid264{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}.box264{background:#f8f6f5;border-left:4px solid #4a4e69;padding:14px;border-radius:9px;font:10px/1.6 Arial;color:#22223b}
.box264 b{display:block;color:#4a4e69;font-size:9px;margin-bottom:5px}
</style>
"""))

header264=widgets.HTML(f"""
<div class="h264">
<div>
<small>SECTION 25.2 • FORECAST VALIDATION</small>
<h2>Actual vs Predicted Intelligence</h2>
<p>Chronological Validation • Role Accuracy • Residuals • Evidence</p>
</div>
<div class="s264">
EVIDENCE STATUS
<b>{FORECAST_EVIDENCE_STATUS264}</b>
{len(actual_pred264):,} observations
</div>
</div>
""")

kpi264=widgets.HTML(
    '<div class="cards264">'+
    f'<div class="c264"><small>STATUS</small><b>{FORECAST_EVIDENCE_STATUS264}</b><span>Evidence gate</span></div>'+
    f'<div class="c264"><small>MAE</small><b>{overall_mae264:.2f}</b><span>Mean absolute error</span></div>'+
    f'<div class="c264"><small>RMSE</small><b>{overall_rmse264:.2f}</b><span>Large-error sensitivity</span></div>'+
    f'<div class="c264"><small>WAPE</small><b>{overall_wape264:.1f}%</b><span>Weighted error</span></div>'+
    f'<div class="c264"><small>ROLES</small><b>{actual_pred264.Job_Family.nunique()}</b><span>Validated families</span></div>'+
    f'<div class="c264"><small>DAYS</small><b>{actual_pred264.Date.nunique()}</b><span>Validation dates</span></div>'+
    '</div>'
)

view264=widgets.ToggleButtons(
    options=[
        ("EXECUTIVE","overview"),
        ("ACTUAL vs PREDICTED","trend"),
        ("RESIDUALS","residual"),
        ("ROLE ACCURACY","roles"),
        ("VALIDATION","validation"),
        ("FILES","files")
    ],
    value="overview",
    layout=widgets.Layout(width="100%"),
    style={"button_width":"16.6%"}
)

role_filter264=widgets.Dropdown(
    options=["ALL"]+sorted(actual_pred264.Job_Family.unique()),
    value="ALL",
    description="Role:"
)

out264=widgets.Output()

def table264(d):
    display(HTML(d.to_html(index=False,border=0,classes="t264")))

def role_data264():
    d=actual_pred264.copy()
    if role_filter264.value!="ALL":
        d=d[d.Job_Family==role_filter264.value]
    return d.groupby("Date",as_index=False)[["Actual","Predicted"]].sum()

def trend264():
    d=role_data264()
    fig=go.Figure()
    fig.add_trace(go.Scatter(x=d.Date,y=d.Actual,mode="lines+markers",name="Actual"))
    fig.add_trace(go.Scatter(x=d.Date,y=d.Predicted,mode="lines+markers",name="Predicted"))
    fig.update_layout(
        title=f"Actual vs Predicted — {role_filter264.value}",
        height=390,
        hovermode="x unified",
        paper_bgcolor="#fff",
        plot_bgcolor="#fff",
        xaxis_title="Validation Date",
        yaxis_title="Job Postings"
    )
    display(fig)

def residual264():
    d=actual_pred264.copy()
    if role_filter264.value!="ALL":
        d=d[d.Job_Family==role_filter264.value]

    fig=go.Figure(go.Scatter(
        x=d.Predicted,
        y=d.Error,
        mode="markers",
        text=d.Job_Family,
        hovertemplate="<b>%{text}</b><br>Predicted: %{x:.1f}<br>Error: %{y:.1f}<extra></extra>"
    ))
    fig.add_hline(y=0,line_dash="dash")
    fig.update_layout(
        title=f"Residual Diagnostic — {role_filter264.value}",
        height=390,
        paper_bgcolor="#fff",
        plot_bgcolor="#fff",
        xaxis_title="Predicted Demand",
        yaxis_title="Actual − Predicted"
    )
    display(fig)

def update264(*_):
    with out264:
        clear_output(wait=True)

        if view264.value=="overview":
            best264=role_metrics264.sort_values("MAE").iloc[0]
            worst264=role_metrics264.sort_values("MAE",ascending=False).iloc[0]

            display(HTML(f"""
<div class="grid264">
<div class="box264"><b>VALIDATION DESIGN</b>Predictions are evaluated against held-out chronological observations rather than random train/test mixing.</div>
<div class="box264"><b>BEST ROLE BY MAE</b>{best264.Job_Family}<br>MAE: {best264.MAE:.2f}</div>
<div class="box264"><b>HIGHEST ROLE ERROR</b>{worst264.Job_Family}<br>MAE: {worst264.MAE:.2f}</div>
</div>
<br>
<div class="note264"><b>Interpretation:</b> MAE is the primary forecasting metric, RMSE highlights larger misses, and WAPE expresses absolute error relative to observed demand.</div>
"""))

        elif view264.value=="trend":
            trend264()

        elif view264.value=="residual":
            residual264()

        elif view264.value=="roles":
            table264(role_metrics264.sort_values("MAE"))

        elif view264.value=="validation":
            table264(forecast_evidence264[["Control","Status"]])

        else:
            files264=pd.DataFrame({
                "Artifact":[
                    "Actual vs Predicted CSV",
                    "Actual vs Predicted PNG",
                    "Role Metrics",
                    "Validation Evidence"
                ],
                "Location":[
                    AP_CSV264,
                    AP_PNG264,
                    REPORT_DIR/"task_03_actual_vs_predicted_metrics.csv",
                    REPORT_DIR/"task_03_actual_vs_predicted_validation.csv"
                ]
            })
            files264["Status"]=files264.Location.apply(
                lambda p:"PASS" if Path(p).exists() else "REVIEW"
            )
            table264(files264)

for w in [view264,role_filter264]:
    w.observe(update264,names="value")

update264()

dashboard_264=widgets.VBox([
    header264,
    kpi264,
    view264,
    role_filter264,
    out264
])
dashboard_264.add_class("d264")
display(dashboard_264)
```

```python
MASTER_EXPORT265=Path("exports/master_results")
SCREENSHOT265=Path("screenshots/results")
CHART_SHOTS265=Path("screenshots/charts")
for p in [MASTER_EXPORT265,SCREENSHOT265,CHART_SHOTS265]: p.mkdir(parents=True,exist_ok=True)
shutil265=__import__("shutil"); hashlib265=__import__("hashlib")
LOGGER_ACTIVE265="ACTIVE" if "logger" in globals() else "NOT ACTIVE"

def shot265(name,d):
    d=d.head(16).iloc[:,:10].copy()
    if d.empty:return None
    d=d.apply(lambda s:s.astype(str).str.replace("\n"," ",regex=False).str.slice(0,30))
    p=SCREENSHOT265/f"{name}.png"
    fig,ax=plt.subplots(figsize=(min(18,max(8,len(d.columns)*1.6)),min(9,max(3,len(d)*.36+1.6))))
    ax.axis("off")
    t=ax.table(cellText=d.values,colLabels=d.columns,loc="center",cellLoc="left")
    t.auto_set_font_size(False);t.set_fontsize(7);t.scale(1,1.22)
    fig.tight_layout();fig.savefig(p,dpi=160,bbox_inches="tight");plt.close(fig)
    return p

def build_archive265(rebuild_screens=True):
    global result_registry265,evidence_manifest265,screenshot_files265,chart_files265
    global RESULTS_SAVED265,SCREENSHOTS_SAVED265,TOTAL_EVIDENCE265,EXPORT_SIZE265
    global RESULT_ARCHIVE_STATUS265,evidence_profile265,validation265
    rows=[]
    for name,obj in list(globals().items()):
        if name.startswith("_") or name.endswith("265"):continue
        try:
            if isinstance(obj,pd.Series):d=obj.to_frame(name=obj.name or "value").reset_index()
            elif isinstance(obj,pd.DataFrame):d=obj
            else:continue
            p=MASTER_EXPORT265/f"{name}.csv";d.to_csv(p,index=False)
            s=shot265(name,d) if rebuild_screens or not (SCREENSHOT265/f"{name}.png").exists() else SCREENSHOT265/f"{name}.png"
            rows.append([name,type(obj).__name__,len(d),len(d.columns),str(p),str(s) if s else "","PASS"])
        except Exception as e:
            rows.append([name,type(obj).__name__,None,None,"","",f"REVIEW: {type(e).__name__}"])

    result_registry265=pd.DataFrame(rows,columns=["Result","Type","Rows","Columns","Saved_File","Screenshot","Status"])
    for root in [Path("charts"),Path("reports")]:
        if root.exists():
            for f in root.rglob("*"):
                if f.is_file() and f.suffix.lower() in {".png",".jpg",".jpeg"}:
                    dst=CHART_SHOTS265/f"{root.name}__{f.name}"
                    if rebuild_screens or not dst.exists():shutil265.copy2(f,dst)

    roots=[Path("reports"),Path("charts"),Path("exports"),Path("screenshots"),Path("models"),Path("database"),Path("logs")]
    files=[]
    for root in roots:
        if not root.exists():continue
        for f in root.rglob("*"):
            if f.is_file() and f.name!="master_evidence_manifest.csv":
                files.append([
                    str(f).replace("\\","/"),root.name,f.suffix.lower() or "file",
                    f.stat().st_size,hashlib265.sha256(f.read_bytes()).hexdigest()
                ])

    evidence_manifest265=pd.DataFrame(files,columns=["File","Root","Type","Size_Bytes","SHA256"]).drop_duplicates("File")
    screenshot_files265=list(Path("screenshots").rglob("*.png"))
    chart_files265=list(CHART_SHOTS265.glob("*"))
    RESULTS_SAVED265=int(result_registry265.Status.eq("PASS").sum())
    SCREENSHOTS_SAVED265=len(screenshot_files265)
    TOTAL_EVIDENCE265=len(evidence_manifest265)
    EXPORT_SIZE265=int(evidence_manifest265.Size_Bytes.sum()) if TOTAL_EVIDENCE265 else 0
    RESULT_ARCHIVE_STATUS265="PASS" if RESULTS_SAVED265 and SCREENSHOTS_SAVED265 and TOTAL_EVIDENCE265 else "REVIEW"
    evidence_profile265=(evidence_manifest265.groupby("Root",as_index=False)
        .agg(Files=("File","count"),Size_Bytes=("Size_Bytes","sum")))
    evidence_profile265["Size_MB"]=(evidence_profile265.Size_Bytes/1024**2).round(2)

    validation265=pd.DataFrame([
        ["Result exports created",RESULTS_SAVED265>0],
        ["Screenshot previews created",SCREENSHOTS_SAVED265>0],
        ["Existing charts archived",len(chart_files265)>0],
        ["Evidence indexed",TOTAL_EVIDENCE265>0],
        ["SHA-256 recorded",evidence_manifest265.SHA256.notna().all() if TOTAL_EVIDENCE265 else False],
        ["Result registry complete",result_registry265.Saved_File.ne("").any()],
        ["Logger available",LOGGER_ACTIVE265=="ACTIVE"]
    ],columns=["Control","Passed"])

    validation265["Status"]=np.where(validation265.Passed,"PASS","REVIEW")
    result_registry265.to_csv(REPORT_DIR/"master_result_registry.csv",index=False)
    evidence_manifest265.to_csv(REPORT_DIR/"master_evidence_manifest.csv",index=False)
    evidence_profile265.to_csv(REPORT_DIR/"master_evidence_profile.csv",index=False)
    validation265.to_csv(REPORT_DIR/"master_evidence_validation.csv",index=False)
    
    pd.DataFrame([{
        "Status":RESULT_ARCHIVE_STATUS265,"Results_Saved":RESULTS_SAVED265,
        "Screenshots_Saved":SCREENSHOTS_SAVED265,"Charts_Archived":len(chart_files265),
        "Evidence_Files":TOTAL_EVIDENCE265,"Evidence_Size_MB":round(EXPORT_SIZE265/1024**2,2),
        "Logger":LOGGER_ACTIVE265
    }]).to_csv(REPORT_DIR/"master_evidence_summary.csv",index=False)

build_archive265()

display(HTML("""
<style>
.d265{border:1px solid #ded8d5;border-radius:18px;overflow:hidden;background:#fff;box-shadow:0 10px 28px #22223b18}
.h265{background:linear-gradient(120deg,#22223b,#30324d 55%,#4a4e69);padding:24px 28px;color:#fff;font-family:Arial;display:flex;justify-content:space-between}
.h265 small{color:#c9ada7;font-weight:700;letter-spacing:1.5px;font-size:8px}.h265 h2{margin:6px 0;font-size:25px}.h265 p{font-size:10px;color:#f2e9e4;margin:0}
.s265{text-align:right;font:9px Arial;color:#f2e9e4}.s265 b{display:block;font-size:18px;color:#fff;margin:4px 0 10px}
.cards265{display:grid;grid-template-columns:repeat(6,1fr);border-bottom:1px solid #eee}.c265{padding:14px;border-right:1px solid #eee;font-family:Arial}
.c265 small{font-size:7px;color:#9a8c98;font-weight:700}.c265 b{display:block;font-size:17px;color:#22223b;margin-top:5px}.c265 span{font-size:8px;color:#4a4e69}
.out265{padding:20px}.t265{width:100%;border-collapse:collapse;font:9px Arial}.t265 th{background:#22223b;color:#fff;padding:9px;text-align:left}.t265 td{padding:8px;border-bottom:1px solid #eee;vertical-align:top}.t265 tr:nth-child(even){background:#fbfaf9}
.grid265{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}.box265{background:#f8f6f5;border-left:4px solid #4a4e69;padding:14px;border-radius:9px;font:10px/1.65 Arial;color:#22223b}
.box265 b{display:block;color:#4a4e69;font-size:9px;margin-bottom:5px}.note265{background:#f2e9e4;padding:12px 14px;border-radius:9px;font:10px/1.65 Arial;color:#4a4e69}
.progress265{height:9px;background:#eee9e7;border-radius:9px;overflow:hidden}.fill265{height:100%;background:#4a4e69}
</style>
"""))

def header265_html():
    return f"""
<div class="h265">
<div>
<small>SECTION 25.3 • MASTER EVIDENCE CONTROL</small>
<h2>Results & Screenshot Archive</h2>
<p>Results • Screenshots • Charts • Integrity • Evidence • Handover</p>
</div>
<div class="s265">
ARCHIVE STATUS
<b>{RESULT_ARCHIVE_STATUS265}</b>
LOGGER
<b>{LOGGER_ACTIVE265}</b>
{TOTAL_EVIDENCE265:,} evidence files
</div>
</div>"""

def kpi265_html():
    vals=[
        ("STATUS",RESULT_ARCHIVE_STATUS265,"Evidence gate"),
        ("RESULTS",RESULTS_SAVED265,"Full exports"),
        ("SCREENSHOTS",SCREENSHOTS_SAVED265,"PNG evidence"),
        ("CHARTS",len(chart_files265),"Archived visuals"),
        ("EVIDENCE",TOTAL_EVIDENCE265,"Indexed files"),
        ("SIZE",f"{EXPORT_SIZE265/1024**2:.1f} MB","Archive footprint")
    ]
    return '<div class="cards265">'+''.join(
        f'<div class="c265"><small>{a}</small><b>{b}</b><span>{c}</span></div>' for a,b,c in vals
    )+'</div>'

header265=widgets.HTML(header265_html())
kpi265=widgets.HTML(kpi265_html())

view265=widgets.ToggleButtons(
    options=[
        ("EXECUTIVE","overview"),("RESULTS","results"),("SCREENSHOTS","screens"),
        ("CHARTS","charts"),("ARCHIVE MAP","map"),("EVIDENCE INDEX","evidence"),
        ("INTEGRITY","integrity"),("VALIDATION","validation"),("FILES","files")
    ],
    value="overview",layout=widgets.Layout(width="100%")
)

type265=widgets.Dropdown(options=["ALL"]+sorted(evidence_manifest265.Type.unique().tolist()),value="ALL",description="Type:")
search265=widgets.Text(placeholder="Search result, file or evidence...",description="Search:")
rows265=widgets.Dropdown(options=[25,50,100,250],value=50,description="Rows:")
refresh265=widgets.Button(description="Re-scan Evidence",icon="refresh")
rebuild265=widgets.Button(description="Rebuild Screenshots",icon="camera")
export265=widgets.Button(description="Export View",icon="save")
reset265=widgets.Button(description="Reset Filters",icon="undo")
action265=widgets.HTML("<span style='font:9px Arial;color:#4a4e69'>Ready</span>")
out265=widgets.Output()

def table265(d):
    display(HTML(d.head(rows265.value).to_html(index=False,border=0,classes="t265")))

def filtered_evidence265():
    d=evidence_manifest265.copy()
    if type265.value!="ALL":d=d[d.Type==type265.value]
    q=search265.value.strip().lower()
    if q:d=d[d.astype(str).agg(" ".join,axis=1).str.lower().str.contains(q,regex=False)]
    return d

def update265(*_):
    with out265:
        clear_output(wait=True);q=search265.value.strip().lower()

        if view265.value=="overview":
            pass_rate265=round(validation265.Passed.mean()*100,1)
            display(HTML(f"""
<div class="grid265">
<div class="box265"><b>RESULT ARCHIVE</b>{RESULTS_SAVED265} analytical DataFrames/Series are saved in full as CSV exports.</div>
<div class="box265"><b>SCREENSHOT EVIDENCE</b>{SCREENSHOTS_SAVED265} PNG files preserve readable result previews and visual evidence.</div>
<div class="box265"><b>INTEGRITY</b>{TOTAL_EVIDENCE265} evidence files are indexed with SHA-256 hashes.</div>
</div><br>
<div class="note265"><b>Evidence controls: {pass_rate265}%</b>
<div class="progress265" style="margin-top:8px"><div class="fill265" style="width:{pass_rate265}%"></div></div>
<br>Full tables remain in CSV. PNG screenshots are presentation evidence and do not replace the complete exported results.</div>
"""))

        elif view265.value=="results":
            d=result_registry265.copy()
            if q:d=d[d.astype(str).agg(" ".join,axis=1).str.lower().str.contains(q,regex=False)]
            table265(d)

        elif view265.value=="screens":
            d=pd.DataFrame({"Screenshot":[str(f).replace("\\","/") for f in screenshot_files265]})
            d["Size_KB"]=d.Screenshot.apply(lambda x:round(Path(x).stat().st_size/1024,1) if Path(x).exists() else 0)
            if q:d=d[d.Screenshot.str.lower().str.contains(q,regex=False)]
            table265(d)

        elif view265.value=="charts":
            d=pd.DataFrame({"Chart_Evidence":[str(f).replace("\\","/") for f in chart_files265]})
            d["Status"]=d.Chart_Evidence.apply(lambda x:"PASS" if Path(x).exists() else "REVIEW")
            if q:d=d[d.Chart_Evidence.str.lower().str.contains(q,regex=False)]
            table265(d)

        elif view265.value=="map":
            fig=go.Figure(go.Treemap(
                labels=evidence_profile265.Root,
                parents=[""]*len(evidence_profile265),
                values=evidence_profile265.Size_Bytes,
                customdata=evidence_profile265.Files,
                hovertemplate="<b>%{label}</b><br>Files: %{customdata}<br>Size: %{value:,} bytes<extra></extra>"
            ))
            fig.update_layout(title="Evidence Archive Composition",height=430,margin=dict(t=55,l=10,r=10,b=10))
            display(fig);table265(evidence_profile265)

        elif view265.value=="evidence":
            table265(filtered_evidence265())

        elif view265.value=="integrity":
            d=filtered_evidence265()[["File","Size_Bytes","SHA256"]]
            display(HTML(f'<div class="note265"><b>{len(d)}</b> evidence files shown with cryptographic integrity hashes.</div><br>'))
            table265(d)

        elif view265.value=="validation":
            table265(validation265[["Control","Status"]])

        else:
            files=pd.DataFrame({
                "Artifact":["Master Result Archive","Result Screenshots","Chart Screenshots","Result Registry","Evidence Manifest","Evidence Profile","Validation","Summary"],
                "Location":[MASTER_EXPORT265,SCREENSHOT265,CHART_SHOTS265,REPORT_DIR/"master_result_registry.csv",REPORT_DIR/"master_evidence_manifest.csv",REPORT_DIR/"master_evidence_profile.csv",REPORT_DIR/"master_evidence_validation.csv",REPORT_DIR/"master_evidence_summary.csv"]
            })
            files["Status"]=files.Location.apply(lambda p:"PASS" if Path(p).exists() else "REVIEW")
            table265(files)

def refresh_click265(_):
    build_archive265(False);header265.value=header265_html();kpi265.value=kpi265_html()
    type265.options=["ALL"]+sorted(evidence_manifest265.Type.unique().tolist());action265.value="<span style='font:9px Arial;color:#4a4e69'>Evidence re-scanned</span>";update265()

def rebuild_click265(_):
    build_archive265(True);header265.value=header265_html();kpi265.value=kpi265_html()
    action265.value="<span style='font:9px Arial;color:#4a4e69'>Screenshots rebuilt</span>";update265()

def export_click265(_):
    d=filtered_evidence265();d.to_csv(REPORT_DIR/"master_evidence_filtered_view.csv",index=False)
    action265.value=f"<span style='font:9px Arial;color:#4a4e69'>Exported {len(d)} rows</span>"

def reset_click265(_):
    type265.value="ALL";search265.value="";rows265.value=50
    action265.value="<span style='font:9px Arial;color:#4a4e69'>Filters reset</span>"

refresh265.on_click(refresh_click265);rebuild265.on_click(rebuild_click265)
export265.on_click(export_click265);reset265.on_click(reset_click265)
for w in [view265,type265,search265,rows265]:w.observe(update265,names="value")

update265()
dashboard_265=widgets.VBox([header265,kpi265,view265,widgets.HBox([type265,search265,rows265]),widgets.HBox([refresh265,rebuild265,export265,reset265,action265]),out265])
dashboard_265.add_class("d265")
display(dashboard_265)
```

<div style="color:#f2e9e4;font-size:11px;line-height:1.65;margin-top:8px">Building a portable deployment package with correct model structure, centralized configuration, runtime-safe inference paths, integrity hashes and a compressed handover bundle before Flask and Streamlit integration.</div>

```python
PACKAGE_ROOT266=Path("deployment_package/job_market_system")
PACKAGE_ZIP266=Path("deployment_package/job_market_system.zip")
shutil266=__import__("shutil");zipfile266=__import__("zipfile");hashlib266=__import__("hashlib")
LOGGER_ACTIVE266="ACTIVE" if "logger" in globals() else "NOT ACTIVE"
MODEL_VERSION266=str(globals().get("MODEL_VERSION","demand_v1"))

def copy266(src,dst):
    src,dst=Path(src),Path(dst)
    if not src.exists():return
    if src.is_file():
        dst.parent.mkdir(parents=True,exist_ok=True);shutil266.copy2(src,dst);return
    shutil266.copytree(src,dst,dirs_exist_ok=True,ignore=shutil266.ignore_patterns("__pycache__","*.pyc"))

def build_package266():
    global package_validation266,manifest266,profile266,PACKAGE_STATUS266,PACKAGE_COMPLETENESS266
    global PACKAGE_SIZE266,PACKAGE_FILES266,ZIP_SIZE266,ZIP_HASH266,PORTABLE_PATH266
    if PACKAGE_ROOT266.exists():shutil266.rmtree(PACKAGE_ROOT266)
    PACKAGE_ROOT266.mkdir(parents=True)

    inference266=Path("src/inference.py")
    if inference266.exists():
        s=inference266.read_text(encoding="utf-8")
        s=s.replace("BASE_DIR=Path.cwd()","BASE_DIR=Path(__file__).resolve().parents[1]")
        s=s.replace("BASE_DIR = Path.cwd()","BASE_DIR = Path(__file__).resolve().parents[1]")
        inference266.write_text(s,encoding="utf-8")

    for x in ["src","config","database"]:copy266(x,PACKAGE_ROOT266/x)
    for x in [".env.example",".gitignore","README.md","requirements.txt","naming_conventions.md"]:copy266(x,PACKAGE_ROOT266/x)
    for x in [f"{MODEL_VERSION266}.pkl",f"{MODEL_VERSION266}_metadata.json"]:
        copy266(Path("models")/x,PACKAGE_ROOT266/"models"/x)
    copy266("models/recommender_assets",PACKAGE_ROOT266/"models/recommender_assets")

    for x in ["master_evidence_summary.csv","master_evidence_validation.csv","master_evidence_manifest.csv"]:
        copy266(REPORT_DIR/x,PACKAGE_ROOT266/"evidence"/x)

    runbook266=f"""# Deployment Runbook

## Architecture
Streamlit → Flask REST API → Inference / Services → SQL Server + Model Assets

## Setup
1. Create and activate the virtual environment.
2. Run `python -m pip install -r requirements.txt`.
3. Configure runtime values using `.env.example`.
4. Confirm SQL Server and JobMarketDB connectivity.
5. Run packaged-runtime validation before starting Flask.
6. Start Flask, validate `/health`, then start Streamlit.

## Model Structure
- Forecast: `models/{MODEL_VERSION266}.pkl`
- Forecast metadata: `models/{MODEL_VERSION266}_metadata.json`
- Recommender: `models/recommender_assets/`

## Configuration
Runtime settings are centralized in `config/settings.py`.

## Security
Do not commit populated `.env` files, passwords, tokens or API keys.

## Scope
Docker/containerization remains a documented scope substitution and must not be represented as completed without written approval evidence.
"""
    (PACKAGE_ROOT266/"DEPLOYMENT_RUNBOOK.md").write_text(runbook266,encoding="utf-8")

    required266=[
        "src/inference.py","src/logger.py","src/exception.py","config/settings.py",
        "database/schema.sql","database/queries.sql","src/data/database.py",
        f"models/{MODEL_VERSION266}.pkl",f"models/{MODEL_VERSION266}_metadata.json",
        "models/recommender_assets/recommendation_catalog.csv",
        "models/recommender_assets/tfidf_vectorizer.pkl",
        "models/recommender_assets/tfidf_matrix.npz",
        "models/recommender_assets/weights.json",
        "models/recommender_assets/metadata.json",
        ".env.example","README.md","requirements.txt","DEPLOYMENT_RUNBOOK.md"
    ]

    package_validation266=pd.DataFrame({
        "Required_Asset":required266,
        "Exists":[(PACKAGE_ROOT266/x).exists() for x in required266]
    })
    package_validation266["Status"]=np.where(package_validation266.Exists,"PASS","REVIEW")

    files=[f for f in PACKAGE_ROOT266.rglob("*") if f.is_file() and f.name!="deployment_manifest.csv"]
    manifest266=pd.DataFrame([{
        "File":str(f.relative_to(PACKAGE_ROOT266)).replace("\\","/"),
        "Category":str(f.relative_to(PACKAGE_ROOT266)).split("\\")[0].split("/")[0],
        "Size_Bytes":f.stat().st_size,
        "SHA256":hashlib266.sha256(f.read_bytes()).hexdigest()
    } for f in files])

    manifest266.to_csv(PACKAGE_ROOT266/"deployment_manifest.csv",index=False)
    package_validation266.to_csv(REPORT_DIR/"deployment_package_validation.csv",index=False)

    profile266=manifest266.groupby("Category",as_index=False).agg(Files=("File","count"),Size_Bytes=("Size_Bytes","sum"))
    profile266["Size_MB"]=(profile266.Size_Bytes/1024**2).round(2)

    PORTABLE_PATH266=(PACKAGE_ROOT266/"src/inference.py").exists() and "Path(__file__).resolve().parents[1]" in (PACKAGE_ROOT266/"src/inference.py").read_text(encoding="utf-8")
    PACKAGE_COMPLETENESS266=round(package_validation266.Exists.mean()*100,1)
    PACKAGE_STATUS266="PASS" if package_validation266.Exists.all() and PORTABLE_PATH266 else "REVIEW"
    PACKAGE_FILES266=len(list(PACKAGE_ROOT266.rglob("*")))
    PACKAGE_SIZE266=sum(f.stat().st_size for f in PACKAGE_ROOT266.rglob("*") if f.is_file())

    if PACKAGE_ZIP266.exists():PACKAGE_ZIP266.unlink()
    with zipfile266.ZipFile(PACKAGE_ZIP266,"w",zipfile266.ZIP_DEFLATED) as z:
        for f in PACKAGE_ROOT266.rglob("*"):
            if f.is_file():z.write(f,Path("job_market_system")/f.relative_to(PACKAGE_ROOT266))

    ZIP_SIZE266=PACKAGE_ZIP266.stat().st_size
    ZIP_HASH266=hashlib266.sha256(PACKAGE_ZIP266.read_bytes()).hexdigest()

    pd.DataFrame([{
        "Status":PACKAGE_STATUS266,"Completeness_%":PACKAGE_COMPLETENESS266,
        "Required_Assets":len(package_validation266),"Assets_Present":int(package_validation266.Exists.sum()),
        "Package_Files":PACKAGE_FILES266,"Package_Size_MB":round(PACKAGE_SIZE266/1024**2,2),
        "Zip_Size_MB":round(ZIP_SIZE266/1024**2,2),"Zip_SHA256":ZIP_HASH266,
        "Forecast_Version":MODEL_VERSION266,"Portable_Inference":"PASS" if PORTABLE_PATH266 else "REVIEW",
        "Logger":LOGGER_ACTIVE266
    }]).to_csv(REPORT_DIR/"deployment_package_summary.csv",index=False)

build_package266()

display(HTML("""
<style>
.d266{border:1px solid #ded8d5;border-radius:18px;overflow:hidden;background:#fff;box-shadow:0 10px 28px #22223b18}
.h266{background:linear-gradient(120deg,#22223b,#30324d 55%,#4a4e69);padding:24px 28px;color:#fff;font-family:Arial;display:flex;justify-content:space-between}
.h266 small{color:#c9ada7;font-weight:700;letter-spacing:1.5px;font-size:8px}.h266 h2{margin:6px 0;font-size:25px}.h266 p{font-size:10px;color:#f2e9e4;margin:0}
.s266{text-align:right;font:9px Arial;color:#f2e9e4}.s266 b{display:block;font-size:17px;color:#fff;margin:4px 0 9px}
.cards266{display:grid;grid-template-columns:repeat(6,1fr);border-bottom:1px solid #eee}.c266{padding:14px;border-right:1px solid #eee;font-family:Arial}
.c266 small{font-size:7px;color:#9a8c98;font-weight:700}.c266 b{display:block;font-size:16px;color:#22223b;margin-top:5px}.c266 span{font-size:8px;color:#4a4e69}
.out266{padding:20px}.t266{width:100%;border-collapse:collapse;font:9px Arial}.t266 th{background:#22223b;color:#fff;padding:9px;text-align:left}.t266 td{padding:8px;border-bottom:1px solid #eee;vertical-align:top}.t266 tr:nth-child(even){background:#fbfaf9}
.grid266{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}.box266{background:#f8f6f5;border-left:4px solid #4a4e69;padding:14px;border-radius:9px;font:10px/1.65 Arial;color:#22223b}
.box266 b{display:block;color:#4a4e69;font-size:9px;margin-bottom:5px}.note266{background:#f2e9e4;padding:12px 14px;border-radius:9px;font:10px/1.65 Arial;color:#4a4e69}
.flow266{display:flex;align-items:center;justify-content:center;gap:9px;flex-wrap:wrap;margin:15px 0}.node266{padding:12px 15px;border:1px solid #c9ada7;border-radius:9px;font:9px Arial;font-weight:700;color:#22223b;background:#fff}.arrow266{color:#9a8c98;font-size:18px}
.progress266{height:9px;background:#eee9e7;border-radius:9px;overflow:hidden}.fill266{height:100%;background:#4a4e69}
</style>
"""))

def header266_html():
    return f"""
<div class="h266">
<div><small>SECTION 26 • DEPLOYMENT ENGINEERING</small><h2>Deployment Readiness Control Center</h2><p>Package • Environment • Models • Configuration • Integrity • Handover</p></div>
<div class="s266">PACKAGE STATUS<b>{PACKAGE_STATUS266}</b>LOGGER<b>{LOGGER_ACTIVE266}</b>{PACKAGE_COMPLETENESS266}% complete</div>
</div>"""

def kpi266_html():
    vals=[
        ("STATUS",PACKAGE_STATUS266,"Deployment gate"),
        ("COMPLETENESS",f"{PACKAGE_COMPLETENESS266}%","Required assets"),
        ("FILES",PACKAGE_FILES266,"Package contents"),
        ("PACKAGE",f"{PACKAGE_SIZE266/1024**2:.1f} MB","Uncompressed"),
        ("ZIP",f"{ZIP_SIZE266/1024**2:.1f} MB","Handover bundle"),
        ("PATHS","PORTABLE" if PORTABLE_PATH266 else "REVIEW","Inference root")
    ]
    return '<div class="cards266">'+''.join(f'<div class="c266"><small>{a}</small><b>{b}</b><span>{c}</span></div>' for a,b,c in vals)+'</div>'

header266=widgets.HTML(header266_html());kpi266=widgets.HTML(kpi266_html())

view266=widgets.ToggleButtons(
    options=[
        ("EXECUTIVE","overview"),("ARCHITECTURE","architecture"),("REQUIRED ASSETS","validation"),
        ("PACKAGE MAP","map"),("INTEGRITY","integrity"),("CONFIG","config"),
        ("EVIDENCE","evidence"),("FILES","files")
    ],value="overview",layout=widgets.Layout(width="100%")
)

search266=widgets.Text(placeholder="Search asset or file...",description="Search:")
rows266=widgets.Dropdown(options=[25,50,100,250],value=50,description="Rows:")
rebuild266=widgets.Button(description="Rebuild Package",icon="refresh")
verify266=widgets.Button(description="Verify Integrity",icon="check")
missing266=widgets.Button(description="Export Missing",icon="save")
action266=widgets.HTML("<span style='font:9px Arial;color:#4a4e69'>Ready</span>")
out266=widgets.Output()

def table266(d):
    display(HTML(d.head(rows266.value).to_html(index=False,border=0,classes="t266")))

def update266(*_):
    with out266:
        clear_output(wait=True);q=search266.value.strip().lower()

        if view266.value=="overview":
            miss=int((~package_validation266.Exists).sum())
            display(HTML(f"""
<div class="grid266">
<div class="box266"><b>PORTABLE INFERENCE</b>Runtime paths resolve from <code>src/inference.py</code> instead of the notebook working directory.</div>
<div class="box266"><b>MODEL STRUCTURE</b>Forecast assets remain in <code>models/</code>; recommender assets remain correctly nested in <code>models/recommender_assets/</code>.</div>
<div class="box266"><b>DEPLOYMENT GATE</b>{int(package_validation266.Exists.sum())}/{len(package_validation266)} required assets present. Missing: {miss}.</div>
</div><br>
<div class="note266"><b>Package readiness: {PACKAGE_COMPLETENESS266}%</b>
<div class="progress266" style="margin-top:8px"><div class="fill266" style="width:{PACKAGE_COMPLETENESS266}%"></div></div>
<br>The Flask and Streamlit applications are intentionally added in subsequent sections and are not required by this pre-API package gate.</div>
"""))

        elif view266.value=="architecture":
            display(HTML("""
<div class="flow266">
<div class="node266">Streamlit Frontend</div><div class="arrow266">→</div>
<div class="node266">Flask REST API</div><div class="arrow266">→</div>
<div class="node266">Inference + Services</div><div class="arrow266">→</div>
<div class="node266">SQL Server + Models</div>
</div>
<div class="note266"><b>Architecture rule:</b> Streamlit owns presentation; Flask owns reusable production services and inference; SQL Server remains the persistent data layer.</div>
"""))

        elif view266.value=="validation":
            d=package_validation266.copy()
            if q:d=d[d.Required_Asset.str.lower().str.contains(q,regex=False)]
            table266(d)

        elif view266.value=="map":
            fig=go.Figure(go.Sunburst(
                labels=profile266.Category,parents=[""]*len(profile266),
                values=profile266.Size_Bytes,customdata=profile266.Files,
                hovertemplate="<b>%{label}</b><br>Files: %{customdata}<br>Size: %{value:,} bytes<extra></extra>"
            ))
            fig.update_layout(title="Deployment Package Composition",height=430,margin=dict(t=55,l=10,r=10,b=10))
            display(fig);table266(profile266)

        elif view266.value=="integrity":
            d=manifest266.copy()
            if q:d=d[d.File.str.lower().str.contains(q,regex=False)]
            display(HTML(f'<div class="note266"><b>ZIP SHA-256:</b><br>{ZIP_HASH266}</div><br>'))
            table266(d)

        elif view266.value=="config":
            d=pd.DataFrame([
                ["Forecast Model",MODEL_VERSION266],
                ["Recommender Path","models/recommender_assets/"],
                ["Settings","config/settings.py"],
                ["Environment",".env.example"],
                ["Database","database/schema.sql + queries.sql"],
                ["Portable Root","Path(__file__).resolve().parents[1]"],
                ["ZIP",str(PACKAGE_ZIP266)]
            ],columns=["Configuration","Value"])
            table266(d)

        elif view266.value=="evidence":
            d=manifest266[manifest266.File.str.startswith("evidence/")].copy()
            table266(d)

        else:
            d=manifest266.copy()
            if q:d=d[d.File.str.lower().str.contains(q,regex=False)]
            table266(d[["File","Category","Size_Bytes"]].sort_values("File"))

def rebuild_click266(_):
    build_package266();header266.value=header266_html();kpi266.value=kpi266_html()
    action266.value="<span style='font:9px Arial;color:#4a4e69'>Package rebuilt</span>";update266()

def verify_click266(_):
    checks=[]
    for _,r in manifest266.iterrows():
        p=PACKAGE_ROOT266/r.File
        checks.append(p.exists() and hashlib266.sha256(p.read_bytes()).hexdigest()==r.SHA256)
    action266.value=f"<span style='font:9px Arial;color:#4a4e69'>Integrity {'PASS' if all(checks) else 'REVIEW'} • {sum(checks)}/{len(checks)}</span>"

def missing_click266(_):
    d=package_validation266[~package_validation266.Exists]
    d.to_csv(REPORT_DIR/"deployment_missing_assets.csv",index=False)
    action266.value=f"<span style='font:9px Arial;color:#4a4e69'>{len(d)} missing asset(s) exported</span>"

rebuild266.on_click(rebuild_click266);verify266.on_click(verify_click266);missing266.on_click(missing_click266)
for w in [view266,search266,rows266]:w.observe(update266,names="value")

update266()
dashboard_266=widgets.VBox([
    header266,kpi266,view266,
    widgets.HBox([search266,rows266]),
    widgets.HBox([rebuild266,verify266,missing266,action266]),
    out266
])
dashboard_266.add_class("d266")
display(dashboard_266)
```

```python
FINAL_ROOT272=Path("deployment_package/job_market_backend_final")
FINAL_ZIP272=Path("deployment_package/job_market_backend_final.zip")
PLANNED_UI_FILE272="stream.py"
LOGGER_ACTIVE272="ACTIVE" if "logger" in globals() else "NOT ACTIVE"
shutil272=globals().get("shutil266",__import__("shutil"))
zipfile272=globals().get("zipfile266",__import__("zipfile"))
hashlib272=globals().get("hashlib266",__import__("hashlib"))

config_path272=Path("frontend_handoff/frontend_integration_config.json")
if config_path272.exists():
    cfg272=json.loads(config_path272.read_text(encoding="utf-8"))
    cfg272["planned_frontend_file"]=PLANNED_UI_FILE272
    config_path272.write_text(json.dumps(cfg272,indent=2),encoding="utf-8")

handoff_path272=Path("frontend_handoff/STREAMLIT_HANDOFF.md")
if handoff_path272.exists():
    s=handoff_path272.read_text(encoding="utf-8").replace("streamlit.py",PLANNED_UI_FILE272).replace("streamlit_app.py",PLANNED_UI_FILE272)
    handoff_path272.write_text(s,encoding="utf-8")

def copy272(src,dst):
    src,dst=Path(src),Path(dst)
    if not src.exists():return
    if src.is_file():
        dst.parent.mkdir(parents=True,exist_ok=True);shutil272.copy2(src,dst);return
    shutil272.copytree(src,dst,dirs_exist_ok=True,ignore=shutil272.ignore_patterns("__pycache__","*.pyc"))

gate272=pd.DataFrame([
    ["Evidence Archive",globals().get("RESULT_ARCHIVE_STATUS265","UNKNOWN")],
    ["Deployment Package",globals().get("PACKAGE_STATUS266","UNKNOWN")],
    ["Packaged Runtime",globals().get("FINAL_DEPLOYMENT_STATUS267","UNKNOWN")],
    ["Flask API Build",globals().get("API_BUILD_STATUS268","UNKNOWN")],
    ["Flask API Tests",globals().get("API_TEST_STATUS269","UNKNOWN")],
    ["Frontend Handoff",globals().get("FRONTEND_HANDOFF_STATUS271","UNKNOWN")],
    ["SQL Layer","PASS" if all(Path(x).exists() for x in ["database/schema.sql","database/queries.sql","src/data/database.py"]) else "REVIEW"],
    ["Logger","PASS" if LOGGER_ACTIVE272=="ACTIVE" else "REVIEW"],
    ["Forecast Model","PASS" if Path(f"models/{globals().get('MODEL_VERSION','demand_v1')}.pkl").exists() else "REVIEW"],
    ["Recommender","PASS" if Path("models/recommender_assets/metadata.json").exists() else "REVIEW"]
],columns=["Gate","Status"])

TECHNICAL_STATUS272="PASS" if gate272.Status.eq("PASS").all() else "REVIEW"
APPROVAL_STATUS272=globals().get("CLIENT_APPROVAL_STATUS254","PENDING EVIDENCE")
HANDOVER_STATUS272="PASS" if TECHNICAL_STATUS272=="PASS" and str(APPROVAL_STATUS272).upper()=="APPROVED" else "REVIEW"
PASS_RATE272=round(gate272.Status.eq("PASS").mean()*100,1)

def build_final272():
    global manifest272,FINAL_FILES272,FINAL_SIZE272,ZIP_SIZE272,ZIP_HASH272

    if FINAL_ROOT272.exists():shutil272.rmtree(FINAL_ROOT272)
    FINAL_ROOT272.mkdir(parents=True)

    for x in ["src","config","database","models","api","api_assets","frontend_handoff"]:
        copy272(x,FINAL_ROOT272/x)

    for x in [".env.example",".gitignore","README.md","requirements.txt","naming_conventions.md"]:
        copy272(x,FINAL_ROOT272/x)

    evidence272=[
        "master_evidence_summary.csv","deployment_package_summary.csv",
        "runtime_smoke_summary.csv","flask_api_build_summary.csv",
        "flask_api_test_summary.csv","flask_api_contract.csv",
        "frontend_handoff_summary.csv","frontend_handoff_validation.csv"
    ]
    for x in evidence272:copy272(REPORT_DIR/x,FINAL_ROOT272/"evidence"/x)

    runbook272=f"""# Final Backend Handover

## Architecture
stream.py → Flask REST API → Services / Inference → Models + SQL Server

## Frontend
The client application will be created separately as `{PLANNED_UI_FILE272}`.
It must communicate with Flask through `JOBMARKET_API_URL`.

## Startup Order
1. Confirm SQL Server and JobMarketDB.
2. Activate the project virtual environment.
3. Install `requirements.txt`.
4. Configure `.env`.
5. Start Flask with `python api/app.py`.
6. Verify `/api/health`.
7. Later start the separate frontend with `streamlit run {PLANNED_UI_FILE272}`.

## Rules
- Do not load production models directly in stream.py.
- Keep hourly and fixed compensation separate.
- Do not infer currency.
- Do not fabricate remote-work status.
- Treat job family and seniority as title-derived.
- Present results as a historical snapshot.

## Governance
Technical backend status: {TECHNICAL_STATUS272}
Docker substitution approval: {APPROVAL_STATUS272}
"""
    (FINAL_ROOT272/"BACKEND_HANDOVER.md").write_text(runbook272,encoding="utf-8")

    files=[f for f in FINAL_ROOT272.rglob("*") if f.is_file() and f.name!="backend_manifest.csv"]
    manifest272=pd.DataFrame([{
        "File":str(f.relative_to(FINAL_ROOT272)).replace("\\","/"),
        "Category":f.relative_to(FINAL_ROOT272).parts[0],
        "Size_Bytes":f.stat().st_size,
        "SHA256":hashlib272.sha256(f.read_bytes()).hexdigest()
    } for f in files])
    manifest272.to_csv(FINAL_ROOT272/"backend_manifest.csv",index=False)

    FINAL_FILES272=len(files)+1
    FINAL_SIZE272=sum(f.stat().st_size for f in FINAL_ROOT272.rglob("*") if f.is_file())

    if FINAL_ZIP272.exists():FINAL_ZIP272.unlink()
    with zipfile272.ZipFile(FINAL_ZIP272,"w",zipfile272.ZIP_DEFLATED) as z:
        for f in FINAL_ROOT272.rglob("*"):
            if f.is_file():z.write(f,Path("job_market_backend_final")/f.relative_to(FINAL_ROOT272))

    ZIP_SIZE272=FINAL_ZIP272.stat().st_size
    ZIP_HASH272=hashlib272.sha256(FINAL_ZIP272.read_bytes()).hexdigest()

build_final272()

required272=[
    "api/app.py","src/api_services.py","src/inference.py","src/logger.py","src/exception.py",
    "src/data/database.py","config/settings.py","database/schema.sql","database/queries.sql",
    f"models/{globals().get('MODEL_VERSION','demand_v1')}.pkl",
    "models/recommender_assets/recommendation_catalog.csv",
    "models/recommender_assets/tfidf_vectorizer.pkl",
    "models/recommender_assets/tfidf_matrix.npz",
    "frontend_handoff/frontend_api_contract.csv",
    "frontend_handoff/frontend_integration_config.json",
    "BACKEND_HANDOVER.md","requirements.txt",".env.example"
]

artifact272=pd.DataFrame({
    "Artifact":required272,
    "Exists":[(FINAL_ROOT272/x).exists() for x in required272]
})
artifact272["Status"]=np.where(artifact272.Exists,"PASS","REVIEW")
ARTIFACT_STATUS272="PASS" if artifact272.Exists.all() else "REVIEW"

governance272=pd.DataFrame([
    ["Technical backend",TECHNICAL_STATUS272],
    ["Required artifacts",ARTIFACT_STATUS272],
    ["Frontend file planned",PLANNED_UI_FILE272],
    ["Frontend packaged","NO"],
    ["Direct frontend model access","DISALLOWED"],
    ["Docker substitution approval",APPROVAL_STATUS272],
    ["Overall handover",HANDOVER_STATUS272]
],columns=["Control","Status"])

gate272.to_csv(REPORT_DIR/"final_backend_gate.csv",index=False)
artifact272.to_csv(REPORT_DIR/"final_backend_artifacts.csv",index=False)
governance272.to_csv(REPORT_DIR/"final_backend_governance.csv",index=False)
manifest272.to_csv(REPORT_DIR/"final_backend_manifest.csv",index=False)

pd.DataFrame([{
    "Technical_Status":TECHNICAL_STATUS272,
    "Artifact_Status":ARTIFACT_STATUS272,
    "Handover_Status":HANDOVER_STATUS272,
    "Gate_Pass_Rate_%":PASS_RATE272,
    "Package_Files":FINAL_FILES272,
    "Package_Size_MB":round(FINAL_SIZE272/1024**2,2),
    "Zip_Size_MB":round(ZIP_SIZE272/1024**2,2),
    "Zip_SHA256":ZIP_HASH272,
    "Planned_Frontend":PLANNED_UI_FILE272,
    "Logger":LOGGER_ACTIVE272
}]).to_csv(REPORT_DIR/"final_backend_handover_summary.csv",index=False)

display(HTML("""
<style>
.d272{border:1px solid #ded8d5;border-radius:18px;overflow:hidden;background:#fff;box-shadow:0 10px 28px #22223b18}
.h272{background:linear-gradient(120deg,#22223b,#30324d 55%,#4a4e69);padding:24px 28px;color:#fff;font-family:Arial;display:flex;justify-content:space-between}
.h272 small{color:#c9ada7;font-weight:700;letter-spacing:1.5px;font-size:8px}.h272 h2{margin:6px 0;font-size:25px}.h272 p{font-size:10px;color:#f2e9e4;margin:0}
.s272{text-align:right;font:9px Arial;color:#f2e9e4}.s272 b{display:block;font-size:17px;color:#fff;margin:4px 0 9px}
.cards272{display:grid;grid-template-columns:repeat(6,1fr);border-bottom:1px solid #eee}.c272{padding:14px;border-right:1px solid #eee;font-family:Arial}
.c272 small{font-size:7px;color:#9a8c98;font-weight:700}.c272 b{display:block;font-size:16px;color:#22223b;margin-top:5px}.c272 span{font-size:8px;color:#4a4e69}
.out272{padding:20px}.t272{width:100%;border-collapse:collapse;font:9px Arial}.t272 th{background:#22223b;color:#fff;padding:9px;text-align:left}.t272 td{padding:8px;border-bottom:1px solid #eee}.t272 tr:nth-child(even){background:#fbfaf9}
.grid272{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}.box272{background:#f8f6f5;border-left:4px solid #4a4e69;padding:14px;border-radius:9px;font:10px/1.65 Arial;color:#22223b}
.box272 b{display:block;color:#4a4e69;font-size:9px;margin-bottom:5px}.note272{background:#f2e9e4;padding:12px 14px;border-radius:9px;font:10px/1.65 Arial;color:#4a4e69}
.flow272{display:flex;align-items:center;justify-content:center;gap:8px;flex-wrap:wrap;margin:18px}.node272{padding:12px 15px;border:1px solid #c9ada7;border-radius:9px;font:9px Arial;font-weight:700}.arrow272{font-size:17px;color:#9a8c98}
.gates272{display:grid;grid-template-columns:repeat(5,1fr);gap:9px}.g272{border:1px solid #e6dfdc;border-radius:9px;padding:11px;font:9px Arial}.g272 b{display:block;color:#22223b;margin-bottom:5px}.g272 span{font-weight:700;color:#4a4e69}
</style>
"""))

def header272():
    return f"""
<div class="h272">
<div><small>SECTION 31 • FINAL BACKEND RELEASE</small><h2>Backend Handover Control Center</h2><p>Models • SQL • Flask • Tests • Evidence • Frontend Contract • Release</p></div>
<div class="s272">TECHNICAL STATUS<b>{TECHNICAL_STATUS272}</b>LOGGER<b>{LOGGER_ACTIVE272}</b>HANDOVER<b>{HANDOVER_STATUS272}</b></div>
</div>"""

def kpi272():
    vals=[
        ("TECHNICAL",TECHNICAL_STATUS272,"Backend controls"),
        ("PASS RATE",f"{PASS_RATE272}%","Release gates"),
        ("ARTIFACTS",f"{artifact272.Exists.sum()}/{len(artifact272)}","Required files"),
        ("PACKAGE",f"{FINAL_SIZE272/1024**2:.1f} MB","Backend bundle"),
        ("ZIP",f"{ZIP_SIZE272/1024**2:.1f} MB","Handover archive"),
        ("FRONTEND",PLANNED_UI_FILE272,"Created later")
    ]
    return '<div class="cards272">'+''.join(f'<div class="c272"><small>{a}</small><b>{b}</b><span>{c}</span></div>' for a,b,c in vals)+'</div>'

header_widget272=widgets.HTML(header272());kpi_widget272=widgets.HTML(kpi272())

view272=widgets.ToggleButtons(
    options=[
        ("EXECUTIVE","overview"),("RELEASE GATES","gates"),("ARCHITECTURE","architecture"),
        ("ARTIFACTS","artifacts"),("PACKAGE MAP","map"),("API HANDOFF","api"),
        ("GOVERNANCE","governance"),("BLOCKERS","blockers"),("FILES","files")
    ],value="overview",layout=widgets.Layout(width="100%")
)
search272=widgets.Text(placeholder="Search gate, artifact or file...",description="Search:")
rebuild272=widgets.Button(description="Rebuild Backend",icon="refresh")
verify272=widgets.Button(description="Verify Integrity",icon="check")
export272=widgets.Button(description="Export Blockers",icon="save")
reset272=widgets.Button(description="Reset",icon="undo")
action272=widgets.HTML("<span style='font:9px Arial;color:#4a4e69'>Ready</span>")
out272=widgets.Output()

def table272(d):display(HTML(d.to_html(index=False,border=0,classes="t272")))

def update272(*_):
    with out272:
        clear_output(wait=True);q=search272.value.strip().lower()

        if view272.value=="overview":
            display(HTML(f"""
<div class="grid272">
<div class="box272"><b>BACKEND COMPLETE</b>Models, SQL access, reusable inference, Flask services, API tests and evidence are consolidated into one release package.</div>
<div class="box272"><b>FRONTEND SEPARATION</b><code>{PLANNED_UI_FILE272}</code> is intentionally excluded and will be created later as the independent Streamlit client.</div>
<div class="box272"><b>RELEASE GOVERNANCE</b>Technical readiness is <strong>{TECHNICAL_STATUS272}</strong>. Overall handover remains <strong>{HANDOVER_STATUS272}</strong> while external approval evidence is {APPROVAL_STATUS272}.</div>
</div>
"""))

        elif view272.value=="gates":
            display(HTML('<div class="gates272">'+''.join(
                f'<div class="g272"><b>{r.Gate}</b><span>{r.Status}</span></div>' for _,r in gate272.iterrows()
            )+'</div>'))

        elif view272.value=="architecture":
            display(HTML(f"""
<div class="flow272">
<div class="node272">{PLANNED_UI_FILE272}</div><div class="arrow272">→</div>
<div class="node272">FLASK API</div><div class="arrow272">→</div>
<div class="node272">SERVICES / INFERENCE</div><div class="arrow272">→</div>
<div class="node272">MODELS + SQL SERVER</div>
</div>
<div class="note272"><b>Boundary:</b> {PLANNED_UI_FILE272} will communicate with Flask through HTTP only. It must not load models directly.</div>
"""))

        elif view272.value=="artifacts":
            d=artifact272.copy()
            if q:d=d[d.Artifact.str.lower().str.contains(q,regex=False)]
            table272(d)

        elif view272.value=="map":
            profile272=manifest272.groupby("Category",as_index=False).agg(Files=("File","count"),Size=("Size_Bytes","sum"))
            fig=go.Figure(go.Treemap(
                labels=profile272.Category,parents=[""]*len(profile272),
                values=profile272.Size,customdata=profile272.Files,
                hovertemplate="<b>%{label}</b><br>Files: %{customdata}<br>Size: %{value:,} bytes<extra></extra>"
            ))
            fig.update_layout(title="Final Backend Package Composition",height=430,margin=dict(t=55,l=10,r=10,b=10))
            display(fig);table272(profile272)

        elif view272.value=="api":
            table272(frontend_map271[["Frontend_Module","Method","Endpoint","Purpose"]] if "frontend_map271" in globals() else pd.DataFrame())

        elif view272.value=="governance":
            table272(governance272)

        elif view272.value=="blockers":
            blocks=pd.concat([
                gate272[gate272.Status!="PASS"].rename(columns={"Gate":"Item"})[["Item","Status"]],
                artifact272[~artifact272.Exists].rename(columns={"Artifact":"Item"})[["Item","Status"]]
            ],ignore_index=True)
            if str(APPROVAL_STATUS272).upper()!="APPROVED":
                blocks=pd.concat([blocks,pd.DataFrame([["Docker substitution approval",APPROVAL_STATUS272]],columns=["Item","Status"])],ignore_index=True)
            if blocks.empty:display(HTML('<div class="note272"><b>PASS:</b> No backend blockers remain.</div>'))
            else:table272(blocks)

        else:
            d=manifest272.copy()
            if q:d=d[d.File.str.lower().str.contains(q,regex=False)]
            table272(d[["File","Category","Size_Bytes","SHA256"]])

def rebuild_click272(_):
    build_final272();header_widget272.value=header272();kpi_widget272.value=kpi272()
    action272.value="<span style='font:9px Arial;color:#4a4e69'>Backend package rebuilt</span>";update272()

def verify_click272(_):
    checks=[(FINAL_ROOT272/r.File).exists() and hashlib272.sha256((FINAL_ROOT272/r.File).read_bytes()).hexdigest()==r.SHA256 for _,r in manifest272.iterrows()]
    action272.value=f"<span style='font:9px Arial;color:#4a4e69'>Integrity {'PASS' if all(checks) else 'REVIEW'} · {sum(checks)}/{len(checks)}</span>"

def export_click272(_):
    b=pd.concat([
        gate272[gate272.Status!="PASS"].rename(columns={"Gate":"Item"})[["Item","Status"]],
        artifact272[~artifact272.Exists].rename(columns={"Artifact":"Item"})[["Item","Status"]]
    ],ignore_index=True)
    b.to_csv(REPORT_DIR/"final_backend_blockers.csv",index=False)
    action272.value=f"<span style='font:9px Arial;color:#4a4e69'>{len(b)} blocker(s) exported</span>"

def reset_click272(_):
    search272.value="";action272.value="<span style='font:9px Arial;color:#4a4e69'>Reset</span>"

rebuild272.on_click(rebuild_click272);verify272.on_click(verify_click272)
export272.on_click(export_click272);reset272.on_click(reset_click272)
for w in [view272,search272]:w.observe(update272,names="value")

update272()
dashboard_272=widgets.VBox([
    header_widget272,kpi_widget272,view272,search272,
    widgets.HBox([rebuild272,verify272,export272,reset272,action272]),out272
])
dashboard_272.add_class("d272")
display(dashboard_272)

if "logger" in globals():
    logger.info(
        "SECTION31_FINAL_BACKEND | technical=%s | handover=%s | pass_rate=%.1f | artifacts=%s/%s | files=%s | zip_mb=%.2f | frontend=%s",
        TECHNICAL_STATUS272,HANDOVER_STATUS272,PASS_RATE272,
        int(artifact272.Exists.sum()),len(artifact272),FINAL_FILES272,
        ZIP_SIZE272/1024**2,PLANNED_UI_FILE272
    )
```

```python
AUDIT_DIR273=Path("reports/final_audit");AUDIT_DIR273.mkdir(parents=True,exist_ok=True)
LOGGER_ACTIVE273="ACTIVE" if "logger" in globals() else "NOT ACTIVE"
FRONTEND_FILE273=Path("stream.py")

def load273(name):
    p=REPORT_DIR/name
    try:return pd.read_csv(p) if p.exists() else pd.DataFrame()
    except:return pd.DataFrame()

requirements_file273=Path("requirements.txt")
required_deps273=[
    "Flask>=3.0,<4","streamlit>=1.40,<2","requests>=2.31,<3",
    "pandas","numpy","scikit-learn","scipy","joblib","plotly","pyodbc","openpyxl"
]

if not requirements_file273.exists():
    requirements_file273.write_text("\n".join(required_deps273)+"\n",encoding="utf-8")

copy272(requirements_file273,FINAL_ROOT272/"requirements.txt")

files272=[f for f in FINAL_ROOT272.rglob("*") if f.is_file() and f.name!="backend_manifest.csv"]

manifest272=pd.DataFrame([{
    "File":str(f.relative_to(FINAL_ROOT272)).replace("\\","/"),
    "Category":f.relative_to(FINAL_ROOT272).parts[0],
    "Size_Bytes":f.stat().st_size,
    "SHA256":hashlib272.sha256(f.read_bytes()).hexdigest()
} for f in files272])

manifest272.to_csv(FINAL_ROOT272/"backend_manifest.csv",index=False)

if FINAL_ZIP272.exists():FINAL_ZIP272.unlink()

with zipfile272.ZipFile(FINAL_ZIP272,"w",zipfile272.ZIP_DEFLATED) as z:
    for f in FINAL_ROOT272.rglob("*"):
        if f.is_file():
            z.write(f,Path("job_market_backend_final")/f.relative_to(FINAL_ROOT272))

requirements_matrix273=globals().get(
    "requirements_matrix252",load273("requirements_matrix.csv")
).copy()

acceptance273=globals().get(
    "acceptance_criteria253",load273("acceptance_criteria.csv")
).copy()

screens273=list(Path("screenshots").rglob("*.png")) if Path("screenshots").exists() else []
ui_screens273=[p for p in screens273 if any(x in p.name.lower() for x in ["stream","streamlit","ui"])]
ssms_screens273=[p for p in screens273 if any(x in p.name.lower() for x in ["ssms","sql","database"])]
model_version273=globals().get("MODEL_VERSION","demand_v1")
forecast_model273=Path(f"models/{model_version273}.pkl").exists()
forecast_meta273=any(Path("models").glob(f"{model_version273}*metadata*.json"))

rec_assets273=[
    Path("models/recommender_assets/recommendation_catalog.csv"),
    Path("models/recommender_assets/tfidf_vectorizer.pkl"),
    Path("models/recommender_assets/tfidf_matrix.npz"),
    Path("models/recommender_assets/weights.json"),
    Path("models/recommender_assets/metadata.json")
]

model_governance273=pd.DataFrame([
    ["Forecast model",forecast_model273],
    ["Forecast metadata",forecast_meta273],
    ["Forecast validation",(REPORT_DIR/"task_03_actual_vs_predicted.csv").exists()],
    ["Model registry",all(x in globals() for x in ["model_registry_master21","model_comparison_master21"])],
    ["Recommendation catalog",rec_assets273[0].exists()],
    ["TF-IDF vectorizer",rec_assets273[1].exists()],
    ["TF-IDF matrix",rec_assets273[2].exists()],
    ["Recommendation weights",rec_assets273[3].exists()],
    ["Recommender metadata",rec_assets273[4].exists()]
],columns=["Control","Passed"])

model_governance273["Status"]=np.where(model_governance273.Passed,"PASS","REVIEW")
MODEL_GOVERNANCE_STATUS273="PASS" if model_governance273.Passed.all() else "REVIEW"

# BACKEND PACKAGE
BACKEND_ROOT273=FINAL_ROOT272
BACKEND_ZIP273=FINAL_ZIP272
BACKEND_MANIFEST273=BACKEND_ROOT273/"backend_manifest.csv"

backend_required273=[
    "api/app.py",
    "src/api_services.py",
    "src/inference.py",
    "src/logger.py",
    "src/exception.py",
    "src/data/database.py",
    "config/settings.py",
    "database/schema.sql",
    "database/queries.sql",
    f"models/{model_version273}.pkl",
    "models/recommender_assets/recommendation_catalog.csv",
    "models/recommender_assets/tfidf_vectorizer.pkl",
    "models/recommender_assets/tfidf_matrix.npz",
    "models/recommender_assets/weights.json",
    "models/recommender_assets/metadata.json",
    "frontend_handoff/frontend_api_contract.csv",
    "frontend_handoff/frontend_integration_config.json",
    "BACKEND_HANDOVER.md",
    "requirements.txt",
    ".env.example"
]

backend_package273=pd.DataFrame({
    "Artifact":backend_required273,
    "Exists":[(BACKEND_ROOT273/x).exists() for x in backend_required273]
})

backend_package273["Status"]=np.where(
    backend_package273.Exists,"PASS","REVIEW"
)

zip_valid273=False
try:
    with zipfile272.ZipFile(BACKEND_ZIP273) as z:
        zip_valid273=z.testzip() is None
except:pass

manifest_valid273=False
manifest_hash_valid273=False

try:
    m273=pd.read_csv(BACKEND_MANIFEST273)
    manifest_valid273=not m273.empty and {"File","SHA256"}.issubset(m273.columns)

    if manifest_valid273:
        hash_checks273=[]
        for _,r in m273.iterrows():
            p=BACKEND_ROOT273/r.File
            hash_checks273.append(
                p.exists() and hashlib272.sha256(p.read_bytes()).hexdigest()==r.SHA256
            )
        manifest_hash_valid273=all(hash_checks273)
except:pass

BACKEND_PACKAGE_STATUS273="PASS" if (
    BACKEND_ROOT273.exists() and
    BACKEND_ZIP273.exists() and
    zip_valid273 and
    manifest_valid273 and
    manifest_hash_valid273 and
    backend_package273.Exists.all()
) else "REVIEW"

# FINAL CONTROLS
controls273=pd.DataFrame([
    ["Model Governance",MODEL_GOVERNANCE_STATUS273,"Forecast + recommender governance evidence"],
    ["SQL/Data Layer","PASS" if all(Path(x).exists() for x in [
        "database/schema.sql","database/queries.sql","src/data/database.py"
    ]) else "REVIEW","Reusable SQL layer"],
    ["Flask API Build",globals().get("API_BUILD_STATUS268","UNKNOWN"),"Production API"],
    ["Flask API QA",globals().get("API_TEST_STATUS269","UNKNOWN"),"Endpoint + failure-path tests"],
    ["Backend Package",BACKEND_PACKAGE_STATUS273,"Final backend bundle + ZIP + manifest"],
    ["Evidence Archive",globals().get("RESULT_ARCHIVE_STATUS265","UNKNOWN"),"Exports + hashes"],
    ["Frontend Contract",globals().get("FRONTEND_HANDOFF_STATUS271","UNKNOWN"),"stream.py integration contract"],
    ["stream.py Client","PASS" if FRONTEND_FILE273.exists() else "PENDING","Separate frontend not built yet"],
    ["UI Screenshots","PASS" if ui_screens273 else "PENDING","Required after stream.py"],
    ["SQL/SSMS Screenshot","PASS" if ssms_screens273 else "PENDING","Database evidence"],
    ["Logger","PASS" if LOGGER_ACTIVE273=="ACTIVE" else "REVIEW","Operational logging"]
],columns=["Control","Status","Evidence"])

backend_names273=[
    "Model Governance","SQL/Data Layer","Flask API Build",
    "Flask API QA","Backend Package","Evidence Archive",
    "Frontend Contract","Logger"
]

backend_controls273=controls273[
    controls273.Control.isin(backend_names273)
]

BACKEND_AUDIT_STATUS273="PASS" if backend_controls273.Status.eq("PASS").all() else "REVIEW"
SUBMISSION_STATUS273="PASS" if controls273.Status.eq("PASS").all() else "REVIEW"

PASS273=int(controls273.Status.eq("PASS").sum())
PENDING273=int(controls273.Status.eq("PENDING").sum())
REVIEW273=int(controls273.Status.eq("REVIEW").sum())
READINESS273=round(PASS273/len(controls273)*100,1)

blockers273=controls273[controls273.Status!="PASS"].copy()

blockers273["Severity"]=blockers273.Control.map({
    "stream.py Client":"HIGH",
    "UI Screenshots":"HIGH",
    "Backend Package":"HIGH",
    "Flask API QA":"HIGH",
    "SQL/SSMS Screenshot":"MEDIUM"
}).fillna("MEDIUM")

# EXPORTS
model_governance273.to_csv(
    AUDIT_DIR273/"model_governance_audit.csv",index=False
)

backend_package273.to_csv(
    AUDIT_DIR273/"backend_package_audit.csv",index=False
)

controls273.to_csv(
    AUDIT_DIR273/"final_compliance_controls.csv",index=False
)

blockers273.to_csv(
    AUDIT_DIR273/"final_submission_blockers.csv",index=False
)

evidence273=pd.DataFrame([
    ["Master Evidence Manifest",REPORT_DIR/"master_evidence_manifest.csv"],
    ["Requirements Matrix",REPORT_DIR/"requirements_matrix.csv"],
    ["Acceptance Criteria",REPORT_DIR/"acceptance_criteria.csv"],
    ["Feature Dictionary",REPORT_DIR/"feature_dictionary.xlsx"],
    ["Forecast Evidence",REPORT_DIR/"task_03_actual_vs_predicted.csv"],
    ["Model Governance Audit",AUDIT_DIR273/"model_governance_audit.csv"],
    ["Backend Package Audit",AUDIT_DIR273/"backend_package_audit.csv"],
    ["API Contract",REPORT_DIR/"flask_api_contract.csv"],
    ["API Tests",REPORT_DIR/"flask_api_test_results.csv"],
    ["Frontend Contract",REPORT_DIR/"frontend_api_contract.csv"],
    ["Backend Handover",REPORT_DIR/"final_backend_handover_summary.csv"],
    ["Requirements",requirements_file273],
    ["Backend ZIP",BACKEND_ZIP273]
],columns=["Evidence","Location"])

evidence273["Exists"]=evidence273.Location.apply(
    lambda p:Path(p).exists()
)

evidence273["Status"]=np.where(
    evidence273.Exists,"PASS","REVIEW"
)

evidence273.to_csv(
    AUDIT_DIR273/"final_evidence_check.csv",index=False
)

submission273=pd.DataFrame([
    ["Backend implementation",BACKEND_AUDIT_STATUS273],
    ["Model governance",MODEL_GOVERNANCE_STATUS273],
    ["Backend package",BACKEND_PACKAGE_STATUS273],
    ["Flask API",globals().get("API_TEST_STATUS269","UNKNOWN")],
    ["Frontend stream.py","COMPLETE" if FRONTEND_FILE273.exists() else "PENDING"],
    ["UI screenshots","COMPLETE" if ui_screens273 else "PENDING"],
    ["SSMS evidence","COMPLETE" if ssms_screens273 else "PENDING"],
    ["Final submission",SUBMISSION_STATUS273]
],columns=["Area","Status"])

submission273.to_csv(
    AUDIT_DIR273/"submission_readiness.csv",index=False
)

pd.DataFrame([{
    "Backend_Audit":BACKEND_AUDIT_STATUS273,
    "Model_Governance":MODEL_GOVERNANCE_STATUS273,
    "Backend_Package":BACKEND_PACKAGE_STATUS273,
    "Submission_Status":SUBMISSION_STATUS273,
    "Readiness_%":READINESS273,
    "Pass":PASS273,
    "Pending":PENDING273,
    "Review":REVIEW273,
    "Frontend_File":FRONTEND_FILE273.exists(),
    "UI_Screenshots":len(ui_screens273),
    "SSMS_Screenshots":len(ssms_screens273),
    "Requirements":(BACKEND_ROOT273/"requirements.txt").exists(),
    "ZIP_Integrity":zip_valid273,
    "Manifest_Integrity":manifest_hash_valid273,
    "Logger":LOGGER_ACTIVE273
}]).to_csv(
    AUDIT_DIR273/"final_audit_summary.csv",index=False
)

# DASHBOARD STYLE
display(HTML("""
<style>
.d273{border:1px solid #ded8d5;border-radius:18px;overflow:hidden;background:#fff;box-shadow:0 10px 28px #22223b18}
.h273{background:linear-gradient(120deg,#22223b,#30324d 55%,#4a4e69);padding:24px 28px;color:#fff;font-family:Arial;display:flex;justify-content:space-between}
.h273 small{color:#c9ada7;font-weight:700;letter-spacing:1.5px;font-size:8px}
.h273 h2{margin:6px 0;font-size:25px}
.h273 p{font-size:10px;color:#f2e9e4;margin:0}
.s273{text-align:right;font:9px Arial;color:#f2e9e4}
.s273 b{display:block;font-size:17px;color:#fff;margin:4px 0 9px}
.cards273{display:grid;grid-template-columns:repeat(6,1fr);border-bottom:1px solid #eee}
.c273{padding:14px;border-right:1px solid #eee;font-family:Arial}
.c273 small{font-size:7px;color:#9a8c98;font-weight:700}
.c273 b{display:block;font-size:16px;color:#22223b;margin-top:5px}
.c273 span{font-size:8px;color:#4a4e69}
.out273{padding:20px}
.t273{width:100%;border-collapse:collapse;font:9px Arial}
.t273 th{background:#22223b;color:#fff;padding:9px;text-align:left}
.t273 td{padding:8px;border-bottom:1px solid #eee;vertical-align:top}
.t273 tr:nth-child(even){background:#fbfaf9}
.grid273{display:grid;grid-template-columns:repeat(3,1fr);gap:12px}
.box273{background:#f8f6f5;border-left:4px solid #4a4e69;padding:14px;border-radius:9px;font:10px/1.65 Arial;color:#22223b}
.box273 b{display:block;color:#4a4e69;font-size:9px;margin-bottom:5px}
.note273{background:#f2e9e4;padding:12px 14px;border-radius:9px;font:10px/1.65 Arial;color:#4a4e69}
.progress273{height:9px;background:#eee9e7;border-radius:9px;overflow:hidden}
.fill273{height:100%;background:#4a4e69}
.flag273{display:grid;grid-template-columns:repeat(3,1fr);gap:10px}
.f273{border:1px solid #e6dfdc;border-radius:9px;padding:12px;font:9px/1.5 Arial}
.f273 b{display:block;color:#22223b;margin-bottom:5px}
.f273 span{font-weight:700;color:#4a4e69}
</style>
"""))

def header273():
    return f"""
<div class="h273">
<div>
<small>SECTION 32 • FINAL RED-TEAM AUDIT</small>
<h2>Submission Readiness Control Center</h2>
<p>Compliance • Model Governance • Backend Package • Evidence • Screenshots • Handover</p>
</div>
<div class="s273">
BACKEND AUDIT
<b>{BACKEND_AUDIT_STATUS273}</b>
BACKEND PACKAGE
<b>{BACKEND_PACKAGE_STATUS273}</b>
LOGGER
<b>{LOGGER_ACTIVE273}</b>
</div>
</div>"""

def kpi273():
    vals=[
        ("READINESS",f"{READINESS273}%","Submission controls"),
        ("PASS",PASS273,"Completed"),
        ("PENDING",PENDING273,"Outstanding"),
        ("MODEL GOV",MODEL_GOVERNANCE_STATUS273,"Forecast + recommender"),
        ("BACKEND PKG",BACKEND_PACKAGE_STATUS273,"ZIP + manifest"),
        ("FRONTEND","READY" if FRONTEND_FILE273.exists() else "PENDING","stream.py")
    ]

    return '<div class="cards273">'+''.join(
        f'<div class="c273"><small>{a}</small><b>{b}</b><span>{c}</span></div>'
        for a,b,c in vals
    )+'</div>'

header_widget273=widgets.HTML(header273())
kpi_widget273=widgets.HTML(kpi273())

view273=widgets.ToggleButtons(
    options=[
        ("EXECUTIVE","overview"),
        ("COMPLIANCE","controls"),
        ("MODEL GOVERNANCE","models"),
        ("BACKEND PACKAGE","backend"),
        ("RED FLAGS","blockers"),
        ("REQUIREMENTS","requirements"),
        ("ACCEPTANCE","acceptance"),
        ("EVIDENCE","evidence"),
        ("SCREENSHOTS","screens"),
        ("SUBMISSION","submission"),
        ("FILES","files")
    ],
    value="overview",
    layout=widgets.Layout(width="100%")
)

status273=widgets.Dropdown(
    options=["ALL","PASS","PENDING","REVIEW"],
    value="ALL",
    description="Status:"
)

search273=widgets.Text(
    placeholder="Search control, artifact or evidence...",
    description="Search:"
)

refresh273=widgets.Button(description="Recheck Audit",icon="refresh")
export273=widgets.Button(description="Export Blockers",icon="save")
evidence_btn273=widgets.Button(description="Export Evidence",icon="folder-open")
reset273=widgets.Button(description="Reset Filters",icon="undo")

action273=widgets.HTML(
    "<span style='font:9px Arial;color:#4a4e69'>Ready</span>"
)

out273=widgets.Output()

def table273(d):
    display(HTML(
        d.to_html(index=False,border=0,classes="t273")
    ))

def update273(*_):
    with out273:
        clear_output(wait=True)
        q=search273.value.strip().lower()

        if view273.value=="overview":
            display(HTML(f"""
<div class="grid273">
<div class="box273">
<b>BACKEND POSITION</b>
Backend audit is <strong>{BACKEND_AUDIT_STATUS273}</strong>. Models, SQL, inference, Flask, tests and backend evidence are independently validated.
</div>

<div class="box273">
<b>MODEL GOVERNANCE</b>
Forecast and recommender governance is <strong>{MODEL_GOVERNANCE_STATUS273}</strong> based on actual saved models, metadata, registry and validation evidence.
</div>

<div class="box273">
<b>PACKAGE POSITION</b>
Backend package is <strong>{BACKEND_PACKAGE_STATUS273}</strong>. Required files, ZIP integrity and manifest integrity are checked directly.
</div>
</div>

<br>

<div class="note273">
<b>Submission readiness: {READINESS273}%</b>
<div class="progress273" style="margin-top:8px">
<div class="fill273" style="width:{READINESS273}%"></div>
</div>
</div>
"""))

        elif view273.value=="controls":
            d=controls273.copy()

            if status273.value!="ALL":
                d=d[d.Status==status273.value]

            if q:
                d=d[
                    d.astype(str)
                    .agg(" ".join,axis=1)
                    .str.lower()
                    .str.contains(q,regex=False)
                ]

            table273(d)

        elif view273.value=="models":
            table273(
                model_governance273[["Control","Status"]]
            )

        elif view273.value=="backend":
            d=backend_package273.copy()

            if q:
                d=d[
                    d.Artifact.str.lower()
                    .str.contains(q,regex=False)
                ]

            table273(d)

            display(HTML(f"""
<div class="note273">
<b>Package Integrity</b><br>
requirements.txt: {"PASS" if (BACKEND_ROOT273/"requirements.txt").exists() else "REVIEW"}<br>
Manifest structure: {"PASS" if manifest_valid273 else "REVIEW"}<br>
Manifest hashes: {"PASS" if manifest_hash_valid273 else "REVIEW"}<br>
ZIP integrity: {"PASS" if zip_valid273 else "REVIEW"}<br>
Final package: <strong>{BACKEND_PACKAGE_STATUS273}</strong>
</div>
"""))

        elif view273.value=="blockers":
            if blockers273.empty:
                display(HTML(
                    '<div class="note273"><b>PASS:</b> No submission blockers remain.</div>'
                ))
            else:
                display(HTML(
                    '<div class="flag273">'+''.join(
                        f'<div class="f273"><b>{r.Control}</b><span>{r.Status} · {r.Severity}</span><br>{r.Evidence}</div>'
                        for _,r in blockers273.iterrows()
                    )+'</div>'
                ))

        elif view273.value=="requirements":
            table273(
                requirements_matrix273
                if not requirements_matrix273.empty
                else pd.DataFrame({"Status":["Requirements matrix unavailable."]})
            )

        elif view273.value=="acceptance":
            table273(
                acceptance273
                if not acceptance273.empty
                else pd.DataFrame({"Status":["Acceptance criteria unavailable."]})
            )

        elif view273.value=="evidence":
            table273(evidence273)

        elif view273.value=="screens":
            d=pd.DataFrame({
                "Screenshot":[str(p).replace("\\","/") for p in screens273]
            })

            if not d.empty:
                d["Type"]=d.Screenshot.apply(
                    lambda x:
                    "UI" if any(k in x.lower() for k in ["stream","streamlit","ui"])
                    else "SQL/SSMS" if any(k in x.lower() for k in ["sql","ssms","database"])
                    else "Analysis"
                )

                if q:
                    d=d[
                        d.Screenshot.str.lower()
                        .str.contains(q,regex=False)
                    ]

            table273(d)

        elif view273.value=="submission":
            table273(submission273)

        else:
            files273=pd.DataFrame({
                "Artifact":[
                    "Compliance Controls",
                    "Model Governance Audit",
                    "Backend Package Audit",
                    "Submission Blockers",
                    "Evidence Check",
                    "Submission Readiness",
                    "Audit Summary"
                ],
                "Location":[
                    AUDIT_DIR273/"final_compliance_controls.csv",
                    AUDIT_DIR273/"model_governance_audit.csv",
                    AUDIT_DIR273/"backend_package_audit.csv",
                    AUDIT_DIR273/"final_submission_blockers.csv",
                    AUDIT_DIR273/"final_evidence_check.csv",
                    AUDIT_DIR273/"submission_readiness.csv",
                    AUDIT_DIR273/"final_audit_summary.csv"
                ]
            })

            files273["Status"]=files273.Location.apply(
                lambda p:"PASS" if Path(p).exists() else "REVIEW"
            )

            table273(files273)

def refresh_click273(_):
    action273.value=(
        f"<span style='font:9px Arial;color:#4a4e69'>"
        f"Models: {MODEL_GOVERNANCE_STATUS273} · "
        f"Package: {BACKEND_PACKAGE_STATUS273} · "
        f"Frontend: {'FOUND' if FRONTEND_FILE273.exists() else 'PENDING'}"
        f"</span>"
    )

def export_click273(_):
    blockers273.to_csv(
        AUDIT_DIR273/"final_submission_blockers.csv",
        index=False
    )

    action273.value=(
        f"<span style='font:9px Arial;color:#4a4e69'>"
        f"{len(blockers273)} blocker(s) exported"
        f"</span>"
    )

def evidence_click273(_):
    evidence273.to_csv(
        AUDIT_DIR273/"final_evidence_check.csv",
        index=False
    )

    action273.value=(
        "<span style='font:9px Arial;color:#4a4e69'>"
        "Evidence index exported"
        "</span>"
    )

def reset_click273(_):
    status273.value="ALL"
    search273.value=""

    action273.value=(
        "<span style='font:9px Arial;color:#4a4e69'>"
        "Filters reset"
        "</span>"
    )

refresh273.on_click(refresh_click273)
export273.on_click(export_click273)
evidence_btn273.on_click(evidence_click273)
reset273.on_click(reset_click273)

for w in [view273,status273,search273]:
    w.observe(update273,names="value")

update273()

dashboard_273=widgets.VBox([
    header_widget273,
    kpi_widget273,
    view273,
    widgets.HBox([status273,search273]),
    widgets.HBox([
        refresh273,
        export273,
        evidence_btn273,
        reset273,
        action273
    ]),
    out273
])

dashboard_273.add_class("d273")
display(dashboard_273)

if "logger" in globals():
    logger.info(
        "SECTION32_FINAL_AUDIT | backend=%s | models=%s | package=%s | submission=%s | readiness=%.1f | pass=%s | pending=%s | review=%s",
        BACKEND_AUDIT_STATUS273,
        MODEL_GOVERNANCE_STATUS273,
        BACKEND_PACKAGE_STATUS273,
        SUBMISSION_STATUS273,
        READINESS273,
        PASS273,
        PENDING273,
        REVIEW273
    )
```

```python
ROOT33 = Path.cwd()
PROJECT_ROOT33 = ROOT33
APP33 = ROOT33 / 'stream.py'
PREVIEW_APP33 = ROOT33 / 'stream_section33_dashboard.py'
STREAMLIT_FILE33 = APP33
STREAMLIT_CANDIDATE33 = PREVIEW_APP33
STREAMLIT_THEME33 = ROOT33 / '.streamlit' / 'config.toml'
REPORT33 = ROOT33 / 'reports' / 'frontend'
REPORT33.mkdir(parents=True, exist_ok=True)

FRONTEND_SOURCE33 = '''# Generated by Section 33 | Interactive client dashboard
import os, logging
from pathlib import Path
from logging.handlers import RotatingFileHandler
import requests, pandas as pd, streamlit as st
import plotly.express as px

st.set_page_config(page_title="Job Market Intelligence", page_icon="📊", layout="wide")
BASE = os.getenv("JOBMARKET_API_URL", "http://127.0.0.1:5000/api").rstrip("/")
logfile = Path("logs/streamlit_frontend.log"); logfile.parent.mkdir(parents=True, exist_ok=True)
log = logging.getLogger("jobmarket.frontend"); log.setLevel(logging.INFO)
if not log.handlers:
    h = RotatingFileHandler(logfile, maxBytes=2000000, backupCount=3)
    h.setFormatter(logging.Formatter("%(asctime)s | %(levelname)s | %(message)s")); log.addHandler(h)

st.markdown("""<style>
.stApp{background:#faf8f7}.block-container{padding-top:1.3rem;max-width:1450px}
[data-testid="stSidebar"]{background:#22223b}
[data-testid="stSidebar"] *{color:#f2e9e4}
.hero33{padding:27px 30px;border-radius:16px;background:linear-gradient(125deg,#22223b,#4a4e69);color:white;margin-bottom:20px}
.hero33 small{color:#c9ada7;font-weight:750;letter-spacing:2px}
.hero33 h1{color:#fff;font-size:30px;margin:8px 0}.hero33 p{color:#f2e9e4;font-size:13px;margin:0}
[data-testid="stMetric"]{background:#fff;padding:18px;border-radius:12px;border:1px solid #e6dfdc}
""",unsafe_allow_html=True)

@st.cache_data(ttl=45,show_spinner=False)
def api_get(route):
    try:
        r=requests.get(f"{BASE}/{route}",timeout=6); r.raise_for_status()
        b=r.json()
        if not isinstance(b,dict) or b.get("status")!="success": raise ValueError(str(b))
        return b.get("data",b)
    except Exception as e:
        log.warning("GET %s: %s",route,e)
        return None

def api_post(route,payload):
    try:
        r=requests.post(f"{BASE}/{route}",json=payload,timeout=25); r.raise_for_status()
        b=r.json()
        if not isinstance(b,dict) or b.get("status")!="success": raise ValueError(str(b))
        return b.get("data",b)
    except Exception as e:
        log.warning("POST %s: %s",route,e); st.error(f"API request failed: {e}"); return None

def frame(data):
    if isinstance(data,list): return pd.DataFrame(data)
    if isinstance(data,dict):
        for key in ("records","rows","results","items","data"):
            if isinstance(data.get(key),list): return pd.DataFrame(data[key])
        if all(not isinstance(v,(dict,list)) for v in data.values()): return pd.DataFrame([data])
    return pd.DataFrame()

def as_frame(data): return frame(data)

def show_chart(fig): st.plotly_chart(fig,use_container_width=True)
def show_title(title,subtitle=""):
    st.title(title)
    if subtitle: st.caption(subtitle)
def show_metric(label,value): st.metric(label,value if value is not None else "—")
def show_table(data):
    df=as_frame(data)
    if not df.empty: st.dataframe(df,use_container_width=True,hide_index=True)
    else: st.info("No records available.")

def panel(data,title):
    df=frame(data)
    if df.empty: st.info("No records received from this endpoint."); return
    numeric=df.select_dtypes(include="number").columns.tolist()
    names=[c for c in df.columns if c not in numeric]
    if numeric:
        metric=numeric[0]
        label=names[0] if names else None
        chart=df[[label,metric]].dropna().head(20) if label else pd.DataFrame()
        if not chart.empty:
            fig=px.bar(chart,x=label,y=metric,title=f"{title} · {metric}",color_discrete_sequence=["#4a4e69"])
            fig.update_layout(plot_bgcolor="white",paper_bgcolor="white",height=350)
            st.plotly_chart(fig,use_container_width=True)
    st.subheader("Detailed records")
    st.dataframe(df,use_container_width=True,hide_index=True)
    st.download_button("Download CSV",df.to_csv(index=False).encode("utf-8"),f"{title.lower().replace(' ','_')}.csv","text/csv")
    if numeric:
        values=pd.to_numeric(df[numeric[0]],errors="coerce").dropna()
        if len(values):
            st.info(f"Insight: {len(df):,} records shown. {numeric[0]} ranges from {values.min():,.2f} to {values.max():,.2f}. Interpret values within the dataset's historical coverage.")

PAGES=["Client Summary","Executive Overview","Salary Intelligence","Emerging Jobs","Demand Forecast","Geographic Intelligence","Market Dynamics","Future Scenarios","Job Recommender","Model Performance","CSV Workspace","System Diagnostics"]
ROUTES={"Executive Overview":"market/trends","Salary Intelligence":"market/salary","Emerging Jobs":"market/emerging-jobs","Geographic Intelligence":"market/countries","Market Dynamics":"market/trends","Model Performance":"models/performance"}
POST={"Demand Forecast":"predict-demand","Future Scenarios":"predict-future","Job Recommender":"recommend"}
with st.sidebar:
    st.title("◈ JOB MARKET")
    st.caption("INTELLIGENCE STUDIO")
    selected_page=st.radio("Navigate",PAGES)
    st.divider()
    if st.button("Refresh data",use_container_width=True): st.cache_data.clear(); st.rerun()
    st.caption("Historical dataset · Not a live job feed")

st.markdown(f'<div class="hero33"><small>JOB MARKET INTELLIGENCE · SECTION 33</small><h1>{selected_page}</h1><p>Interactive analytics • Exportable results • Evidence-led insights and recommendations</p></div>',unsafe_allow_html=True)
health=api_get("health")
if selected_page=="Client Summary":
    summary=api_get("market/summary")
    if summary is None:
        st.warning("Flask or SQL is unavailable. Dashboard is ready; no market figures are invented.")
    elif isinstance(summary,dict):
        metrics=[("Job records",summary.get("total_jobs")),("Job families",summary.get("job_families")),("Countries",summary.get("countries")),("Latest record",summary.get("end_date"))]
        for col,(label,value) in zip(st.columns(4),metrics): col.metric(label,str(value) if value is not None else "—")
    st.subheader("Market trends")
    result=api_get("market/trends")
    if result is not None: panel(result,"Market Trends")
    st.caption("Historical data only. Hourly rates and fixed budgets remain distinct; currencies are not normalized.")
elif selected_page=="System Diagnostics":
    st.metric("Flask API", "CONNECTED" if health is not None else "OFFLINE")
    st.code(BASE)
    st.dataframe(pd.DataFrame([{"Method":"GET","Route":x} for x in ["health","market/summary","market/salary","market/emerging-jobs","market/countries","market/trends","models/performance"]]+[{"Method":"POST","Route":x} for x in POST.values()]),hide_index=True)
elif selected_page=="CSV Workspace":
    file=st.file_uploader("Upload a CSV for local inspection",type="csv")
    if file:
        df=pd.read_csv(file); a,b,c=st.columns(3)
        a.metric("Rows",f"{len(df):,}"); b.metric("Columns",len(df.columns)); c.metric("Missing cells",f"{int(df.isna().sum().sum()):,}")
        st.dataframe(df.head(200),use_container_width=True,hide_index=True)
        st.download_button("Export uploaded data",df.to_csv(index=False),"uploaded_preview.csv","text/csv")
elif selected_page in POST:
    st.subheader("Run a model request")
    st.caption("Use a payload matching the Flask endpoint contract. Subsequent model sections can replace this with client input forms.")
    raw=st.text_area("JSON request payload",value="{}",height=105)
    if st.button("Run prediction",type="primary"):
        import json
        try:
            payload=json.loads(raw)
            if not isinstance(payload,dict): raise ValueError("Payload must be a JSON object")
            answer=api_post(POST[selected_page],payload)
            if answer is not None: st.json(answer)
        except (ValueError,TypeError) as e: st.error(str(e))
else:
    result=api_get(ROUTES[selected_page])
    if result is None: st.warning("The Flask endpoint or SQL database is unavailable.")
    else: panel(result,selected_page)
'''
ast.parse(FRONTEND_SOURCE33)
if not APP33.exists():
    APP33.write_text(FRONTEND_SOURCE33, encoding='utf-8')
    FRONTEND_ACTION33 = 'CREATED stream.py'
else:
    PREVIEW_APP33.write_text(FRONTEND_SOURCE33, encoding='utf-8')
    FRONTEND_ACTION33 = 'PRESERVED stream.py · created stream_section33_dashboard.py'
config33=ROOT33/'.streamlit'/'config.toml'; config33.parent.mkdir(exist_ok=True)
if not config33.exists():
    config33.write_text('[theme]\nprimaryColor="#4a4e69"\nbackgroundColor="#faf8f7"\nsecondaryBackgroundColor="#f2e9e4"\ntextColor="#22223b"\n',encoding='utf-8')

DATA33=next((globals()[name] for name in ('df_features','df_final','jobs22') if isinstance(globals().get(name),pd.DataFrame)),pd.DataFrame())
DATE33=next((c for c in ('published_date','published_day','date') if c in DATA33.columns),None)
COUNTRY33=next((c for c in ('country_clean','country') if c in DATA33.columns),None)
FAMILY33=next((c for c in ('job_family','category') if c in DATA33.columns),None)
DATES33=pd.to_datetime(DATA33[DATE33],errors='coerce',utc=True) if DATE33 else pd.Series(dtype='datetime64[ns, UTC]')

CSS33='''<style>
.d33{font-family:Arial,sans-serif;background:white;border:1px solid #e6dfdc;border-radius:17px;overflow:hidden;box-shadow:0 8px 24px #22223b15;margin:14px 0}
.hero33{background:linear-gradient(125deg,#22223b,#32344f 60%,#4a4e69);padding:24px;color:white}
.hero33 small{color:#c9ada7;letter-spacing:1.8px;font-size:10px}.hero33 h2{color:white!important;margin:7px 0;font-size:26px}.hero33 p{color:#f2e9e4;font-size:12px;margin:0}
.stats33{display:grid;grid-template-columns:repeat(4,minmax(0,1fr));gap:1px;background:#e6dfdc}
.stats33>div{padding:18px;background:white}.stats33 span{display:block;color:#9a8c98;font-size:10px;font-weight:bold}.stats33 b{display:block;color:#22223b;font-size:23px;margin:7px 0}.stats33 em{color:#6b6575;font-size:10px;font-style:normal}
.content33{padding:20px}.tile33{background:#f8f5f3;border-left:4px solid #9a8c98;border-radius:9px;padding:15px;margin:8px 0;color:#4a4e69;font-size:12px;line-height:1.6}
.tag33{font-size:10px;font-weight:bold;color:#9a8c98;letter-spacing:1.4px}.tbl33{border-collapse:collapse;width:100%;font-size:12px}.tbl33 th{background:#22223b;color:white;padding:10px;text-align:left}.tbl33 td{padding:9px;border-bottom:1px solid #ece7e4}
</style>'''
display(HTML(CSS33))

rows33=len(DATA33)
metrics33=[('JOB RECORDS',f'{rows33:,}','Validated notebook data'),('COUNTRIES',str(DATA33[COUNTRY33].nunique()) if COUNTRY33 else '—','Geographic coverage'),('JOB FAMILIES',str(DATA33[FAMILY33].nunique()) if FAMILY33 else '—','Role distribution'),('LATEST POST',str(DATES33.max().date()) if DATES33.notna().any() else '—','Historical data')]
metric_html33=''.join(f'<div><span>{escape(a)}</span><b>{escape(b)}</b><em>{escape(c)}</em></div>' for a,b,c in metrics33)
header33=widgets.HTML('<div class="d33"><div class="hero33"><small>SECTION 33 · STREAMLIT FRONTEND</small><h2>Job Market Intelligence Studio</h2><p>Interactive navigation · Market trends · Geography · Data quality · Exports · Flask connection</p></div><div class="stats33">'+metric_html33+'</div></div>')
nav33=widgets.ToggleButtons(options=[('OVERVIEW','overview'),('TRENDS','trends'),('GEOGRAPHY','geo'),('DATA QUALITY','quality'),('SYSTEM','system')],value='overview',layout=widgets.Layout(width='100%'))
out33=widgets.Output(layout=widgets.Layout(width='100%'))

def render33(_=None):
    with out33:
        clear_output(wait=True)
        section=nav33.value
        if section=='overview':
            label='Validated dataset loaded' if rows33 else 'No notebook dataframe detected'
            display(HTML('<div class="tile33"><b>MARKET OVERVIEW</b><p>'+escape(label)+'. Interactive pages below use notebook data rather than the failed SQL connection.</p></div>'))
            if FAMILY33 and rows33:
                display(HTML('<div class="tag33">TOP JOB FAMILIES</div>'+DATA33[FAMILY33].fillna('Unknown').value_counts().head(7).rename_axis('Job Family').reset_index(name='Jobs').to_html(index=False,classes='tbl33',border=0,escape=True)))
            display(HTML('<div class="tile33"><b>INTERPRETATION</b><p>This dashboard reports historical records, not live openings. Use the geographic and trend views to identify data concentration.</p></div>'))
        elif section=='trends':
            if not DATES33.notna().any(): display(HTML('<div class="tile33">No valid date column available for a trend chart.</div>'))
            else:
                trend=DATES33.dropna().dt.to_period('M').astype(str).value_counts().sort_index().rename_axis('Month').reset_index(name='Jobs')
                import plotly.express as px
                fig=px.bar(trend,x='Month',y='Jobs',title='Historical job postings by month',color_discrete_sequence=['#4a4e69'])
                fig.update_layout(paper_bgcolor='white',plot_bgcolor='white',height=340)
                display(fig)
                trend.to_csv(REPORT33/'section33_monthly_trends.csv',index=False)
                display(HTML('<div class="tile33">Saved: reports/frontend/section33_monthly_trends.csv. Monthly counts reflect dataset coverage, not the full labour market.</div>'))
        elif section=='geo':
            if not COUNTRY33 or not rows33: display(HTML('<div class="tile33">Country data unavailable in the notebook.</div>'))
            else:
                geo=DATA33[COUNTRY33].fillna('Unknown').value_counts().head(20).rename_axis('Country').reset_index(name='Jobs')
                geo.to_csv(REPORT33/'section33_geography.csv',index=False)
                display(HTML('<div class="tag33">TOP COUNTRIES · JOB POSTINGS</div>'+geo.to_html(index=False,border=0,classes='tbl33',escape=True)))
        elif section=='quality':
            if not rows33: display(HTML('<div class="tile33">No dataframe available.</div>'))
            else:
                q=DATA33.isna().sum().sort_values(ascending=False).head(12).rename_axis('Field').reset_index(name='Missing')
                q['Completeness %']=(100*(1-q['Missing']/rows33)).round(1)
                q.to_csv(REPORT33/'section33_data_quality.csv',index=False)
                display(HTML('<div class="tag33">DATA COMPLETENESS</div>'+q.to_html(index=False,border=0,classes='tbl33',escape=True)))
        else:
            checks=pd.DataFrame([('App written',True),('Theme configured',config33.exists()),('Flask app found',(ROOT33/'api'/'app.py').exists()),('Streamlit installed',importlib.util.find_spec('streamlit') is not None)],columns=['Check','Passed'])
            display(HTML('<div class="tile33"><b>FILE STATUS:</b> '+escape(FRONTEND_ACTION33)+'<p>Flask and SQL may be unavailable while JobMarketDB is in RECOVERY_PENDING. No database changes performed.</p></div>'))
            display(HTML(checks.to_html(index=False,border=0,classes='tbl33',escape=True)))
            display(HTML('<div class="tile33">Run: <code>python -m streamlit run '+escape((PREVIEW_APP33 if APP33.exists() and FRONTEND_ACTION33.startswith('PRESERVED') else APP33).name)+'</code></div>'))
nav33.observe(render33,names='value')
dashboard_33=widgets.VBox([header33,nav33,out33]); display(dashboard_33); render33()
FRONTEND_STATUS33 = 'PASS' if (APP33.exists() and importlib.util.find_spec('streamlit') and importlib.util.find_spec('plotly')) else 'REVIEW'
API_BASE33 = os.getenv('JOBMARKET_API_URL','http://127.0.0.1:5000/api')
print('Section 33 dashboard ready |',FRONTEND_ACTION33)
```

```python
ROOT34 = Path(globals().get("ROOT33", Path.cwd())).resolve()
MODULE34 = ROOT34 / "frontend_pages" / "section34_overview.py"
STREAM34 = ROOT34 / "stream.py"
REPORT34 = ROOT34 / "reports" / "frontend"
BACKUP34 = ROOT34 / "backups"

for folder34 in (MODULE34.parent, REPORT34, BACKUP34):
    folder34.mkdir(parents=True, exist_ok=True)

INIT34 = MODULE34.parent / "__init__.py"
if not INIT34.exists():
    INIT34.write_text("", encoding="utf-8")

FRONTEND_MODULE34 = r'''
# Section 34 | Executive Market Overview

from pathlib import Path
import pandas as pd
import plotly.express as px
import streamlit as st

INK = "#22223b"
GRAPE = "#4a4e69"
ASH = "#9a8c98"
ROSE = "#c9ada7"
CREAM = "#f2e9e4"


def _frame(data):
    if isinstance(data, pd.DataFrame):
        return data.copy()
    if isinstance(data, list):
        return pd.DataFrame(data)
    if isinstance(data, dict):
        for key in ("records", "rows", "results", "items", "data"):
            if isinstance(data.get(key), list):
                return pd.DataFrame(data[key])
        if all(not isinstance(v, (list, dict)) for v in data.values()):
            return pd.DataFrame([data])
    return pd.DataFrame()


def _fetch(route, filename, api_get):
    try:
        table = _frame(api_get(route))
        if not table.empty:
            return table, "Flask API"
    except Exception:
        pass

    root = Path(__file__).resolve().parents[1]
    candidates = [
        root / "api_assets" / filename,
        root / "reports" / "frontend" / filename
    ]

    for file in candidates:
        if file.is_file():
            try:
                table = pd.read_csv(file)
                if not table.empty:
                    return table, "Historical CSV snapshot"
            except (OSError, ValueError, pd.errors.ParserError):
                pass

    return pd.DataFrame(), "Unavailable"


def _trends(df):
    columns = ["Month", "Jobs"]
    # market_trends.csv and the API name the count column "Job_Postings"
    if "Jobs" not in df.columns:
        df = df.rename(columns={"Job_Postings": "Jobs"})
    if not set(columns).issubset(df.columns):
        return pd.DataFrame(columns=columns)

    result = df[columns].copy()
    result["Month"] = pd.to_datetime(
        result["Month"].astype(str), errors="coerce", utc=True
    )
    result["Jobs"] = pd.to_numeric(result["Jobs"], errors="coerce")
    result = result.dropna(subset=columns)
    result["Month"] = result["Month"].dt.strftime("%Y-%m")

    return (
        result.groupby("Month", as_index=False)["Jobs"]
        .sum().sort_values("Month")
    )


def _emerging(df, threshold=5):
    cols = ["Job_Family", "Recent_28D", "Previous_28D"]
    if not set(cols).issubset(df.columns):
        return pd.DataFrame(columns=cols + ["Growth_%"])

    result = df[cols].copy()

    for col in cols[1:]:
        result[col] = pd.to_numeric(result[col], errors="coerce")

    result = result.dropna(subset=cols)
    result = result[
        (result["Previous_28D"] >= threshold) &
        (result["Recent_28D"] >= 0)
    ].copy()

    result["Growth_%"] = (
        result["Recent_28D"] / result["Previous_28D"] - 1
    ) * 100

    return result.sort_values(
        ["Growth_%", "Recent_28D"], ascending=False
    )


def _countries(df):
    if not {"Country", "Jobs"}.issubset(df.columns):
        return pd.DataFrame(columns=["Country", "Jobs"])

    result = df.copy()
    result["Jobs"] = pd.to_numeric(result["Jobs"], errors="coerce")

    return (
        result.dropna(subset=["Country", "Jobs"])
        .sort_values("Jobs", ascending=False)
    )


def _chart(fig, key):
    fig.update_layout(
        template="plotly_white",
        height=410,
        font=dict(color=INK),
        margin=dict(l=15, r=15, t=45, b=15),
        paper_bgcolor="white",
        plot_bgcolor="white"
    )
    st.plotly_chart(
        fig, use_container_width=True,
        config={"displaylogo": False}, key=key
    )


def _download(df, filename, key):
    st.download_button(
        "Download filtered CSV",
        data=df.to_csv(index=False).encode("utf-8-sig"),
        file_name=filename,
        mime="text/csv",
        key=key,
        use_container_width=True
    )


def render_overview(page="Executive Overview", api_get=None, as_frame=None):

    st.markdown("""
    <style>
    .s34hero {
        background:linear-gradient(115deg,#22223b,#4a4e69);
        color:white;border-radius:17px;
        padding:27px 30px;margin-bottom:20px;
    }
    .s34hero small {
        color:#c9ada7;letter-spacing:2px;font-weight:700;
    }
    .s34hero h2 {
        color:white;font-size:28px;margin:8px 0;
    }
    .s34hero p {
        color:#f2e9e4;font-size:13px;margin:0;
    }
    .s34note {
        background:#f2e9e4;border-left:4px solid #9a8c98;
        border-radius:9px;padding:15px;color:#4a4e69;
        margin:12px 0;
    }
    </style>
    """, unsafe_allow_html=True)

    title = (
        "Executive Market Overview"
        if page == "Executive Overview"
        else "Client Decision Summary"
    )

    st.markdown(
        f"""<div class="s34hero">
        <small>SECTION 34 · JOB MARKET INTELLIGENCE</small>
        <h2>{title}</h2>
        <p>Historical Demand · Emerging Jobs · Geography · Decisions</p>
        </div>""",
        unsafe_allow_html=True
    )

    if api_get is None:
        st.error("The existing Flask API helper is required.")
        return

    try:
        summary = api_get("market/summary")
    except Exception:
        summary = {}

    if not isinstance(summary, dict):
        summary = {}

    t_raw, t_src = _fetch(
        "market/trends", "market_trends.csv", api_get
    )
    e_raw, e_src = _fetch(
        "market/emerging-jobs", "emerging_jobs.csv", api_get
    )
    g_raw, g_src = _fetch(
        "market/countries", "market_countries.csv", api_get
    )

    monthly = _trends(t_raw)
    emerging = _emerging(e_raw)
    geo = _countries(g_raw)

    metrics = [
        ("Job Postings", "total_jobs"),
        ("Job Families", "job_families"),
        ("Countries", "countries"),
        ("Latest Record", "end_date")
    ]

    for col, (label, field) in zip(st.columns(4), metrics):
        value = summary.get(field)
        if isinstance(value, (int, float)) and not isinstance(value, bool):
            value = f"{value:,.0f}"
        col.metric(label, str(value) if value is not None else "—")

    if not summary:
        st.warning("Summary API unavailable. Missing KPIs are not estimated.")

    st.caption(
        f"Trends: {t_src} | Emerging: {e_src} | "
        f"Geography: {g_src} | Historical records, not live vacancies"
    )

    tabs = st.tabs([
        "Overview",
        "Market Trends",
        "Emerging Jobs",
        "Geography",
        "Insights & Exports"
    ])

    with tabs[0]:
        st.subheader("Market at a Glance")
        a, b, c = st.columns(3)

        a.metric("Monthly Periods", len(monthly) if not monthly.empty else "—")
        b.metric("Emerging Families", len(emerging) if not emerging.empty else "—")
        c.metric("Country Rows", len(geo) if not geo.empty else "—")

        st.markdown(
            """<div class="s34note">
            <b>Executive interpretation</b><p>
            Market activity represents historical dataset coverage.
            Emerging-job growth compares consecutive 28-day periods.
            Geographic API counts represent hourly-paid listings only.
            </p></div>""",
            unsafe_allow_html=True
        )

        st.markdown("### Strategic Priorities")
        st.write("1. Evaluate historical demand using comparable periods.")
        st.write("2. Identify growing job families with meaningful sample sizes.")
        st.write("3. Review geographic distribution and coverage limitations.")
        st.write("4. Validate evidence before making client recommendations.")

        st.markdown("### Recommendation")
        st.info(
            "Prioritize roles with sustained posting activity and "
            "reliable growth evidence rather than isolated spikes."
        )

    with tabs[1]:
        st.subheader("Historical Market Activity")

        if monthly.empty:
            st.info("Monthly trend records are unavailable.")
        else:
            months = st.multiselect(
                "Select months",
                monthly["Month"].tolist(),
                default=monthly["Month"].tolist(),
                key="s34_months"
            )

            view = monthly[monthly["Month"].isin(months)].copy()

            if view.empty:
                st.info("Select at least one month.")
            else:
                fig = px.line(
                    view, x="Month", y="Jobs", markers=True,
                    title="Historical Monthly Job Postings",
                    color_discrete_sequence=[GRAPE]
                )
                _chart(fig, "s34_trends")
                st.dataframe(view, use_container_width=True, hide_index=True)

                leader = view.loc[view["Jobs"].idxmax()]

                st.markdown("### Insights")
                st.write(
                    f"The highest observed monthly count is "
                    f"{leader['Jobs']:,.0f} in {leader['Month']}."
                )

                st.markdown("### Interpretation")
                st.write(
                    "Month-to-month variation reflects observed posting "
                    "activity and may be affected by incomplete boundary months."
                )

                st.markdown("### Recommendations")
                st.write(
                    "Compare equivalent observation windows before drawing "
                    "conclusions about overall demand."
                )

                _download(
                    view, "section34_market_trends.csv",
                    "s34_export_trends"
                )

    with tabs[2]:
        st.subheader("Emerging Job Families")

        threshold = st.slider(
            "Minimum previous 28-day postings",
            1, 50, 5, key="s34_threshold"
        )
        top_n = st.slider(
            "Families to display",
            1, 20, 10, key="s34_top"
        )

        ranked = _emerging(e_raw, threshold).head(top_n)

        if ranked.empty:
            st.info("No eligible emerging-job records.")
        else:
            fig = px.bar(
                ranked.sort_values("Growth_%"),
                x="Growth_%", y="Job_Family",
                orientation="h",
                title="Job Family Growth (%)",
                color_discrete_sequence=[ASH]
            )
            _chart(fig, "s34_emerging")
            st.dataframe(ranked, use_container_width=True, hide_index=True)

            leader = ranked.iloc[0]

            st.markdown("### Insights")
            st.write(
                f"{leader['Job_Family']} leads the selected ranking "
                f"with {leader['Growth_%']:.1f}% growth."
            )

            st.markdown("### Interpretation")
            st.write(
                "Growth compares two consecutive 28-day windows. "
                "Small prior counts can exaggerate percentage increases."
            )

            st.markdown("### Recommendations")
            st.write(
                "Prioritize job families only after examining "
                "absolute counts and the consistency of demand."
            )

            _download(
                ranked, "section34_emerging_jobs.csv",
                "s34_export_emerging"
            )

    with tabs[3]:
        st.subheader("Geographic Intelligence")

        st.caption(
            "Country API results cover hourly-paid postings only. "
            "Average hourly pay is not currency-adjusted."
        )

        if geo.empty:
            st.info("Geographic records are unavailable.")
        else:
            n = st.slider(
                "Countries to display",
                1, 30, 12, key="s34_country_count"
            )

            shown = geo.head(n)

            fig = px.bar(
                shown.sort_values("Jobs"),
                x="Jobs", y="Country",
                orientation="h",
                title="Hourly-Paid Postings by Country",
                color_discrete_sequence=[GRAPE]
            )
            _chart(fig, "s34_geography")
            st.dataframe(shown, use_container_width=True, hide_index=True)

            leader = shown.iloc[0]

            st.markdown("### Insights")
            st.write(
                f"{leader['Country']} has the highest observed "
                f"count: {leader['Jobs']:,.0f} hourly-paid postings."
            )

            st.markdown("### Interpretation")
            st.write(
                "Country-level counts reflect the available hourly-paid "
                "dataset rather than every vacancy in each country."
            )

            st.markdown("### Recommendations")
            st.write(
                "Compare geographic opportunities using consistent "
                "filters and avoid comparing unnormalized compensation."
            )

            _download(
                shown, "section34_hourly_countries.csv",
                "s34_export_countries"
            )

    with tabs[4]:
        st.subheader("Executive Insights & Client Recommendations")

        st.markdown("""
        **Demand:** Compare equivalent historical periods.

        **Emerging Jobs:** Evaluate growth percentages alongside raw counts.

        **Geography:** Distinguish dataset coverage from total market demand.

        **Compensation:** Keep hourly rates separate from fixed budgets.

        **Decision Making:** Prioritize validated and sustained patterns.
        """)

        st.markdown("### Data Provenance")
        st.json({
            "summary": "Flask API" if summary else "Unavailable",
            "trends": t_src,
            "emerging_jobs": e_src,
            "countries": g_src
        })

        exports = {
            "Monthly Trends": monthly,
            "Emerging Jobs": emerging,
            "Hourly Countries": geo
        }

        selected = st.selectbox(
            "Select dataset to export",
            list(exports),
            key="s34_export_select"
        )

        if exports[selected].empty:
            st.info("The selected dataset is unavailable.")
        else:
            _download(
                exports[selected],
                "section34_" + selected.lower().replace(" ", "_") + ".csv",
                "s34_export_selected"
            )
'''

ast.parse(FRONTEND_MODULE34, filename=str(MODULE34))

def _backup34(path):
    if not path.exists():
        return None
    from datetime import datetime
    stamp = datetime.now().strftime("%Y%m%d_%H%M%S_%f")
    target = BACKUP34 / f"{path.stem}_before_s34_{stamp}{path.suffix}"
    target.write_bytes(path.read_bytes())
    return target

def _write34(path, content):
    temporary = path.with_name(path.name + ".section34.tmp")
    try:
        temporary.write_text(content, encoding="utf-8")
        temporary.replace(path)
    finally:
        if temporary.exists():
            temporary.unlink()

if (
    not MODULE34.exists()
    or MODULE34.read_text(encoding="utf-8-sig") != FRONTEND_MODULE34
):
    if MODULE34.exists():
        print("Module backup:", _backup34(MODULE34))
    _write34(MODULE34, FRONTEND_MODULE34)

print("Section 34 module created:", MODULE34)

STREAM_INTEGRATED34 = False

if STREAM34.is_file():
    import re

    stream_source34 = STREAM34.read_text(encoding="utf-8-sig")

    if "frontend_pages.section34_overview import render_overview" in stream_source34:
        STREAM_INTEGRATED34 = True
        print("Section 34 routing already present.")

    else:
        route_pattern34 = re.compile(
            r'(?m)^(?P<indent>[ \t]*)(?P<branch>if|elif)\s+'
            r'selected_page\s*==\s*'
            r'(?P<quote>["\'])Client Summary(?P=quote)\s*:\s*(?:#.*)?$'
        )

        matches34 = list(route_pattern34.finditer(stream_source34))

        if len(matches34) == 1:
            match34 = matches34[0]
            indent34 = match34.group("indent")
            branch34 = match34.group("branch")

            following34 = stream_source34[match34.end():]
            next_pattern34 = re.compile(
                r'(?m)^' + re.escape(indent34) +
                r'(?:elif\b|else\s*:)'
            )
            next_match34 = next_pattern34.search(following34)

            if next_match34:
                start34 = match34.start()
                end34 = match34.end() + next_match34.start()

                replacement34 = (
                    f'{indent34}{branch34} selected_page in '
                    '("Client Summary", "Executive Overview"):\n'
                    f'{indent34}    from frontend_pages.section34_overview '
                    'import render_overview\n'
                    f'{indent34}    render_overview('
                    'selected_page, api_get, as_frame)\n'
                )

                updated34 = (
                    stream_source34[:start34]
                    + replacement34
                    + stream_source34[end34:]
                )

                ast.parse(updated34, filename=str(STREAM34))
                backup34 = _backup34(STREAM34)
                _write34(STREAM34, updated34)

                STREAM_INTEGRATED34 = True
                print("Streamlit backup:", backup34)
                print("Section 34 integrated into existing stream.py.")
            else:
                print("Routing unchanged: next page boundary not found.")
        else:
            print("Routing unchanged: Client Summary branch not uniquely found.")
else:
    print("stream.py not found; module created without routing changes.")

DATA34 = globals().get("DATA33")

if not isinstance(DATA34, pd.DataFrame) or DATA34.empty:
    DATA34 = pd.DataFrame()
    for name34 in ("df_features", "df_final", "jobs22"):
        value34 = globals().get(name34)
        if isinstance(value34, pd.DataFrame) and not value34.empty:
            DATA34 = value34.copy()
            break

def _column34(*names):
    lookup = {str(c).casefold(): c for c in DATA34.columns}
    return next(
        (lookup[n.casefold()] for n in names if n.casefold() in lookup),
        None
    )

DATE_COL34 = _column34("published_date", "published_day", "date")
FAMILY_COL34 = _column34("job_family", "category")
COUNTRY_COL34 = _column34("country_clean", "country")

DATES34 = (
    pd.to_datetime(DATA34[DATE_COL34], errors="coerce", utc=True)
    if DATE_COL34 is not None
    else pd.Series(pd.NaT, index=DATA34.index, dtype="datetime64[ns, UTC]")
)

MONTHLY34 = pd.DataFrame(columns=["Month", "Jobs"])
EMERGING34 = pd.DataFrame(
    columns=["Job_Family", "Recent_28D", "Previous_28D", "Growth_%"]
)
COUNTRIES34 = pd.DataFrame(columns=["Country", "Jobs"])

if DATES34.notna().any():
    MONTHLY34 = (
        DATES34.dropna()
        .dt.strftime("%Y-%m")
        .value_counts()
        .sort_index()
        .rename_axis("Month")
        .reset_index(name="Jobs")
    )

if COUNTRY_COL34 is not None:
    COUNTRIES34 = (
        DATA34[COUNTRY_COL34]
        .fillna("Unknown")
        .astype(str)
        .value_counts()
        .rename_axis("Country")
        .reset_index(name="Jobs")
    )

if FAMILY_COL34 is not None and DATES34.notna().any():
    latest34 = DATES34.max()

    recent_mask34 = (
        (DATES34 > latest34 - pd.Timedelta(days=28))
        & (DATES34 <= latest34)
    )
    previous_mask34 = (
        (DATES34 > latest34 - pd.Timedelta(days=56))
        & (DATES34 <= latest34 - pd.Timedelta(days=28))
    )

    families34 = DATA34[FAMILY_COL34].fillna("Unknown").astype(str)
    recent_counts34 = families34[recent_mask34].value_counts()
    previous_counts34 = families34[previous_mask34].value_counts()

    EMERGING34 = pd.DataFrame({
        "Recent_28D": recent_counts34,
        "Previous_28D": previous_counts34
    }).fillna(0)

    EMERGING34 = (
        EMERGING34.rename_axis("Job_Family").reset_index()
    )
    EMERGING34 = EMERGING34[
        EMERGING34["Previous_28D"] >= 5
    ].copy()
    EMERGING34["Growth_%"] = (
        EMERGING34["Recent_28D"] / EMERGING34["Previous_28D"] - 1
    ) * 100
    EMERGING34 = EMERGING34.sort_values("Growth_%", ascending=False)

MONTHLY34.to_csv(
    REPORT34 / "section34_monthly_trends.csv", index=False
)
EMERGING34.to_csv(
    REPORT34 / "section34_emerging_families.csv", index=False
)
COUNTRIES34.to_csv(
    REPORT34 / "section34_country_coverage.csv", index=False
)

CONTRACT34 = pd.DataFrame([
    ("GET", "/api/market/summary", "Executive KPIs"),
    ("GET", "/api/market/trends", "Historical monthly demand"),
    ("GET", "/api/market/emerging-jobs", "28-day job-family growth"),
    ("GET", "/api/market/countries", "Hourly-paid geographic coverage")
], columns=["Method", "Endpoint", "Purpose"])

CONTRACT34.to_csv(
    REPORT34 / "section34_api_contract.csv", index=False
)

INK34, GRAPE34, ASH34 = "#22223b", "#4a4e69", "#9a8c98"

CSS34 = '''
<style>
.s34wrap{font-family:Arial;background:white;border:1px solid #e6dfdc;
border-radius:16px;overflow:hidden;margin:10px 0}
.s34hero{background:linear-gradient(115deg,#22223b,#4a4e69);
padding:27px;color:white}
.s34hero h2{color:white!important;margin:9px 0;font-size:27px}
.s34hero p{color:#f2e9e4;font-size:12px;margin:0}
.s34kpi{display:grid;grid-template-columns:repeat(4,1fr);
gap:1px;background:#e6dfdc}
.s34kpi div{padding:17px;background:white}
.s34kpi span{font-size:10px;color:#9a8c98;display:block}
.s34kpi b{font-size:21px;color:#22223b;display:block;margin-top:7px}
.s34note{background:#f2e9e4;border-left:4px solid #9a8c98;
padding:14px;border-radius:9px;margin:12px 0;color:#4a4e69;
font:12px Arial;line-height:1.6}
</style>
'''

display(HTML(CSS34))

METRICS34 = [
    ("JOB RECORDS", f"{len(DATA34):,}" if not DATA34.empty else "—"),
    ("JOB FAMILIES",
     f"{DATA34[FAMILY_COL34].nunique():,}"
     if FAMILY_COL34 is not None else "—"),
    ("COUNTRIES",
     f"{DATA34[COUNTRY_COL34].nunique():,}"
     if COUNTRY_COL34 is not None else "—"),
    ("LATEST RECORD",
     str(DATES34.max().date()) if DATES34.notna().any() else "—")
]

from html import escape

kpis_html34 = "".join(
    f"<div><span>{escape(label)}</span><b>{escape(value)}</b></div>"
    for label, value in METRICS34
)

display(HTML(
    '<div class="s34wrap"><div class="s34hero">'
    '<small>SECTION 34 · JOB MARKET INTELLIGENCE</small>'
    '<h2>Executive Overview & Decision Dashboard</h2>'
    '<p>Historical Demand · Emerging Roles · Geographic Intelligence</p>'
    '</div><div class="s34kpi">'
    + kpis_html34 + '</div></div>'
))

def _table34(df, n=15):
    if df.empty:
        display(HTML('<div class="s34note">No records available.</div>'))
    else:
        display(HTML(df.head(n).to_html(index=False, escape=True, border=0)))

def _note34(message):
    display(HTML('<div class="s34note">' + escape(message) + '</div>'))

def _draw34(view):
    if view == "Overview":
        _note34(
            f"Available notebook records: {len(DATA34):,}. "
            "Market metrics are historical and should not be treated "
            "as current job vacancies."
        )
        _table34(CONTRACT34)
        _note34(
            "Recommendation: review posting trends, emerging families "
            "and geographic coverage before deciding market priorities."
        )

    elif view == "Market Trends":
        if MONTHLY34.empty:
            _note34("Published-date records unavailable.")
        else:
            import plotly.express as px
            fig = px.line(
                MONTHLY34, x="Month", y="Jobs", markers=True,
                title="Historical Monthly Job Postings",
                color_discrete_sequence=[GRAPE34]
            )
            fig.update_layout(template="plotly_white", height=340)
            display(fig)
            _table34(MONTHLY34)

            leader = MONTHLY34.loc[MONTHLY34["Jobs"].idxmax()]
            _note34(
                f"Insight: {leader['Month']} has the highest observed "
                f"monthly volume ({leader['Jobs']:,.0f} postings)."
            )
            _note34(
                "Interpretation: partial months can distort "
                "month-to-month comparisons."
            )
            _note34(
                "Recommendation: compare equal, complete observation periods."
            )

    elif view == "Emerging Jobs":
        if EMERGING34.empty:
            _note34("No job families meet the growth baseline.")
        else:
            import plotly.express as px
            ranked = EMERGING34.head(12).sort_values("Growth_%")
            fig = px.bar(
                ranked, x="Growth_%", y="Job_Family",
                orientation="h", title="28-Day Job-Family Growth (%)",
                color_discrete_sequence=[ASH34]
            )
            fig.update_layout(template="plotly_white", height=410)
            display(fig)
            _table34(EMERGING34)
            leader = EMERGING34.iloc[0]
            _note34(
                f"Insight: {leader['Job_Family']} leads at "
                f"{leader['Growth_%']:.1f}% growth."
            )
            _note34(
                "Interpretation: percentages depend on the "
                "previous-period posting baseline."
            )
            _note34(
                "Recommendation: assess absolute counts and "
                "sustained demand alongside growth."
            )

    elif view == "Geography":
        if COUNTRIES34.empty:
            _note34("Country-level records unavailable.")
        else:
            import plotly.express as px
            shown = COUNTRIES34.head(12).sort_values("Jobs")
            fig = px.bar(
                shown, x="Jobs", y="Country",
                orientation="h", title="Notebook Geographic Coverage",
                color_discrete_sequence=[GRAPE34]
            )
            fig.update_layout(template="plotly_white", height=380)
            display(fig)
            _table34(COUNTRIES34)
            leader = COUNTRIES34.iloc[0]
            _note34(
                f"Insight: {leader['Country']} has the highest observed "
                f"country count ({leader['Jobs']:,.0f} records)."
            )
            _note34(
                "Interpretation: this notebook distribution counts "
                "all included records; the country API is hourly-only."
            )
            _note34(
                "Recommendation: compare consistent posting types "
                "before identifying geographic priorities."
            )

    elif view == "Insights":
        for message in (
            "Demand: historical records do not equal current vacancies.",
            "Emerging jobs: evaluate relative growth with raw counts.",
            "Geography: distinguish dataset coverage from market demand.",
            "Compensation: keep hourly and fixed budgets separate.",
            "Client action: prioritize validated, sustained signals."
        ):
            _note34(message)

    elif view == "Exports":
        _table34(pd.DataFrame({
            "Report": [
                "Monthly Trends",
                "Emerging Families",
                "Country Coverage",
                "API Contract"
            ],
            "Filename": [
                "section34_monthly_trends.csv",
                "section34_emerging_families.csv",
                "section34_country_coverage.csv",
                "section34_api_contract.csv"
            ]
        }))
        _note34(f"All available notebook reports are saved in {REPORT34}")

    elif view == "Run Guide":
        _note34(
            "Main application: stream.py. "
            "Section 34 is a reusable module, not a second application."
        )
        _note34(
            "Routing integration: "
            + ("Present" if STREAM_INTEGRATED34 else "Requires review")
        )

try:
    import ipywidgets as widgets

    NAV34 = widgets.ToggleButtons(
        options=[
            "Overview", "Market Trends", "Emerging Jobs",
            "Geography", "Insights", "Exports", "Run Guide"
        ],
        value="Overview",
        layout=widgets.Layout(width="100%")
    )
    OUTPUT34 = widgets.Output(layout=widgets.Layout(width="100%"))

    def render34(change=None):
        with OUTPUT34:
            clear_output(wait=True)
            _draw34(NAV34.value)

    NAV34.observe(render34, names="value")
    DASHBOARD34 = widgets.VBox([NAV34, OUTPUT34])
    display(DASHBOARD34)
    render34()

except ImportError:
    _draw34("Overview")
    print("Install ipywidgets for interactive notebook navigation.")

SECTION34_STATUS = (
    "READY (STATIC VALIDATION)"
    if STREAM_INTEGRATED34
    else "MODULE READY - ROUTING REQUIRES REVIEW"
)
```

```python
from urllib.request import urlopen
from urllib.error import URLError, HTTPError
ROOT_FIX = Path(
    globals().get("ROOT35", globals().get("ROOT34", Path.cwd()))
).resolve()

REPORT_FIX = ROOT_FIX / "reports" / "validation"
REPORT_FIX.mkdir(parents=True, exist_ok=True)
STREAM_FIX = ROOT_FIX / "stream.py"
MODULE_FIX = ROOT_FIX / "frontend_pages" / "section34_overview.py"
ASSETS_FIX = ROOT_FIX / "api_assets"

BASE_URL_FIX = globals().get(
    "API_BASE_URL", "http://127.0.0.1:5000"
).rstrip("/")

results_fix = []

def check_fix(category, name, status, detail):
    results_fix.append({
        "Category": category,
        "Check": name,
        "Status": status,
        "Details": str(detail)
    })

def syntax_fix(path):
    if not path.is_file():
        return "FAIL", "Python file not found"
    try:
        ast.parse(
            path.read_text(encoding="utf-8-sig"),
            filename=str(path)
        )
        return "PASS", "Python syntax verified"
    except (SyntaxError, UnicodeError, OSError) as exc:
        return "FAIL", str(exc)

for label, path in [
    ("Existing Streamlit application", STREAM_FIX),
    ("Section 34 dashboard module", MODULE_FIX)
]:
    status, detail = syntax_fix(path)
    check_fix("Python", label, status, detail)

if STREAM_FIX.is_file():
    source_fix = STREAM_FIX.read_text(encoding="utf-8-sig")

    import_detected_fix = (
        "frontend_pages.section34_overview import render_overview"
        in source_fix
    )
    render_detected_fix = (
        "render_overview(" in source_fix
    )

    check_fix(
        "Integration",
        "Section 34 renderer",
        "PASS" if import_detected_fix and render_detected_fix else "FAIL",
        "Import and renderer detected"
        if import_detected_fix and render_detected_fix
        else "Review Section 34 routing in stream.py"
    )

    for page in ["Client Summary", "Executive Overview"]:
        check_fix(
            "Navigation",
            page,
            "PASS" if page in source_fix else "WARN",
            "Page label detected"
            if page in source_fix
            else "Page label not found"
        )

ROUTES_FIX = {
    "Market Summary": "/api/market/summary",
    "Market Trends": "/api/market/trends",
    "Emerging Jobs": "/api/market/emerging-jobs",
    "Country Intelligence": "/api/market/countries"
}

API_RESPONSES_FIX = {}

for name, route in ROUTES_FIX.items():
    url = BASE_URL_FIX + route

    try:
        with urlopen(url, timeout=3) as response:
            content_type = response.headers.get("Content-Type", "")
            raw = response.read()

            if response.status == 200 and "json" in content_type.lower():
                data = json.loads(raw.decode("utf-8"))
                API_RESPONSES_FIX[name] = data

                check_fix(
                    "Flask API", name, "PASS",
                    f"HTTP 200 · Valid JSON · {route}"
                )
            else:
                check_fix(
                    "Flask API", name, "FAIL",
                    f"HTTP {response.status} · {content_type}"
                )

    except (URLError, HTTPError, TimeoutError, OSError, ValueError) as exc:
        check_fix(
            "Flask API", name, "WARN",
            f"Endpoint not verified: {exc}"
        )

EXPECTED_FIX = {
    "market_trends.csv": ["Month", "Job_Postings"],
    "emerging_jobs.csv": [
        "Job_Family", "Recent_28D", "Previous_28D"
    ],
    "market_countries.csv": ["Country", "Jobs"]
}

ASSET_FRAMES_FIX = {}

for filename, required in EXPECTED_FIX.items():
    path = ASSETS_FIX / filename

    if not path.is_file():
        check_fix(
            "CSV Assets", filename, "FAIL", "File missing"
        )
        continue

    try:
        df = pd.read_csv(path)
        ASSET_FRAMES_FIX[filename] = df
        missing_cols = sorted(set(required) - set(df.columns))

        if missing_cols:
            check_fix(
                "CSV Assets", filename, "FAIL",
                "Missing: " + ", ".join(missing_cols)
            )
        elif df.empty:
            check_fix(
                "CSV Assets", filename, "WARN",
                "Headers valid, but no records"
            )
        else:
            check_fix(
                "CSV Assets", filename, "PASS",
                f"{len(df):,} records · Schema verified"
            )
    except (OSError, ValueError, pd.errors.ParserError) as exc:
        check_fix("CSV Assets", filename, "FAIL", str(exc))

DATA_FIX = globals().get("DATA34")

if not isinstance(DATA_FIX, pd.DataFrame) or DATA_FIX.empty:
    DATA_FIX = globals().get("CHECK_DATA35")

if not isinstance(DATA_FIX, pd.DataFrame):
    DATA_FIX = pd.DataFrame()

if not DATA_FIX.empty:
    missing_counts = DATA_FIX.isna().sum()

    MISSING_REPORT_FIX = pd.DataFrame({
        "Column": missing_counts.index,
        "Missing_Count": missing_counts.values,
        "Missing_Percent": (
            missing_counts.values / len(DATA_FIX) * 100
        ).round(2)
    }).sort_values("Missing_Count", ascending=False)

    total_missing_fix = int(missing_counts.sum())
    total_duplicates_fix = int(DATA_FIX.duplicated().sum())

    check_fix(
        "Data Quality",
        "Dataset availability",
        "PASS",
        f"{len(DATA_FIX):,} records"
    )

    check_fix(
        "Data Quality",
        "Exact duplicates",
        "PASS" if total_duplicates_fix == 0 else "WARN",
        f"{total_duplicates_fix:,} exact duplicate rows"
    )

    check_fix(
        "Data Quality",
        "Missing-value assessment",
        "PASS",
        f"{total_missing_fix:,} missing cells counted; "
        "column-level report generated. "
        "Treatment policy still requires review."
    )

else:
    MISSING_REPORT_FIX = pd.DataFrame(
        columns=["Column", "Missing_Count", "Missing_Percent"]
    )
    total_missing_fix = 0
    total_duplicates_fix = 0

    check_fix(
        "Data Quality", "Dataset availability",
        "WARN", "Notebook DataFrame unavailable"
    )

EXPORTS_FIX = {
    "Monthly Trends": ROOT_FIX / "reports" / "frontend" /
                      "section34_monthly_trends.csv",
    "Emerging Families": ROOT_FIX / "reports" / "frontend" /
                         "section34_emerging_families.csv",
    "Country Coverage": ROOT_FIX / "reports" / "frontend" /
                        "section34_country_coverage.csv",
    "API Contract": ROOT_FIX / "reports" / "frontend" /
                    "section34_api_contract.csv"
}

for label, path in EXPORTS_FIX.items():
    if path.is_file():
        try:
            df = pd.read_csv(path)
            check_fix(
                "Exports", label,
                "PASS" if not df.empty else "WARN",
                f"{len(df):,} exported rows"
            )
        except (OSError, ValueError, pd.errors.ParserError):
            check_fix("Exports", label, "FAIL", "CSV unreadable")
    else:
        check_fix("Exports", label, "FAIL", "Export missing")

check_fix(
    "Runtime",
    "Streamlit page rendering",
    "WARN",
    "Requires running stream.py and checking the pages"
)

VALIDATION_FIX = pd.DataFrame(results_fix)

pass_fix = int((VALIDATION_FIX["Status"] == "PASS").sum())
warn_fix = int((VALIDATION_FIX["Status"] == "WARN").sum())
fail_fix = int((VALIDATION_FIX["Status"] == "FAIL").sum())
total_fix = len(VALIDATION_FIX)

VALIDATION_FIX.to_csv(
    REPORT_FIX / "section35_corrected_validation.csv",
    index=False
)

MISSING_REPORT_FIX.to_csv(
    REPORT_FIX / "section35_missingness_detail.csv",
    index=False
)

SUMMARY_FIX = pd.DataFrame({
    "Metric": [
        "Total Checks", "Passed", "Warnings",
        "Failed", "Pass Percentage"
    ],
    "Value": [
        total_fix, pass_fix, warn_fix, fail_fix,
        round(pass_fix / total_fix * 100, 2)
        if total_fix else 0
    ]
})

SUMMARY_FIX.to_csv(
    REPORT_FIX / "section35_corrected_summary.csv",
    index=False
)

CSS_FIX = """
<style>
.fix-hero{
    background:linear-gradient(135deg,#22223b,#32344f 52%,#4a4e69);
    padding:25px 28px;
    border-radius:15px;
    font-family:Arial;
    color:white;
    margin:10px 0 17px;
}
.fix-hero h2{
    color:white!important;
    font-size:27px;
    margin:8px 0;
}
.fix-hero p{
    color:#f2e9e4;
    font-size:12px;
}
.fix-kpis{
    display:grid;
    grid-template-columns:repeat(4,1fr);
    gap:10px;
    margin:15px 0;
}
.fix-kpi{
    background:white;
    border:1px solid #e6dfdc;
    border-radius:12px;
    padding:18px;
    font-family:Arial;
}
.fix-kpi span{
    color:#9a8c98;
    font-size:11px;
}
.fix-kpi b{
    color:#22223b;
    font-size:25px;
    display:block;
    margin-top:8px;
}
.fix-note{
    background:#f2e9e4;
    color:#4a4e69;
    border-left:4px solid #9a8c98;
    border-radius:9px;
    padding:15px;
    margin:13px 0;
    font:12px Arial;
    line-height:1.6;
}
</style>
"""

display(HTML(CSS_FIX))

display(HTML("""
<div class="fix-hero">
  <small style="color:#c9ada7;letter-spacing:1.7px">
    SECTION 35 · CLIENT QUALITY ASSURANCE
  </small>
  <h2>System Validation & Dashboard Readiness</h2>
  <p>
    Streamlit Integration · Flask API · Data Quality ·
    Visual Analytics · Export Verification
  </p>
</div>
"""))

display(HTML(
    '<div class="fix-kpis">' +
    "".join(
        f'<div class="fix-kpi"><span>{label}</span>'
        f'<b>{value}</b></div>'
        for label, value in [
            ("TOTAL CHECKS", total_fix),
            ("PASSED", pass_fix),
            ("WARNINGS", warn_fix),
            ("FAILED", fail_fix)
        ]
    ) + '</div>'
))

def note_fix(text):
    display(HTML(
        '<div class="fix-note">' +
        escape(str(text)) + '</div>'
    ))

def table_fix(df, limit=40):
    if df.empty:
        note_fix("No records available.")
    else:
        display(HTML(
            df.head(limit).to_html(
                index=False, escape=True, border=0
            )
        ))

def chart_fix(fig):
    fig.update_layout(
        template="plotly_white",
        height=390,
        font=dict(color="#22223b"),
        margin=dict(l=20, r=20, t=50, b=20)
    )
    display(fig)

def draw_fix(page):
    if page == "Executive Overview":
        chart_fix(
            px.pie(
                pd.DataFrame({
                    "Status": ["PASS", "WARN", "FAIL"],
                    "Count": [pass_fix, warn_fix, fail_fix]
                }),
                names="Status", values="Count",
                title="Validation Status Distribution",
                color="Status",
                color_discrete_map={
                    "PASS": "#4a4e69",
                    "WARN": "#c9ada7",
                    "FAIL": "#9a8c98"
                },
                hole=0.57
            )
        )
        table_fix(SUMMARY_FIX)
        note_fix(
            "Validation results reflect the checks actually "
            "completed. Runtime rendering remains unverified "
            "until the Streamlit application is launched."
        )

    elif page == "Validation Results":
        table_fix(VALIDATION_FIX, 100)
        note_fix(
            "PASS = verified check; WARN = incomplete verification "
            "or issue requiring review; FAIL = detected problem."
        )

    elif page == "API Connectivity":
        api_rows = VALIDATION_FIX[
            VALIDATION_FIX["Category"] == "Flask API"
        ]
        table_fix(api_rows)
        note_fix(
            f"Flask API base URL: {BASE_URL_FIX}. "
            "An unavailable server is not reported as a successful test."
        )

    elif page == "Data Quality":
        if MISSING_REPORT_FIX.empty:
            note_fix("No DataFrame available for analysis.")
        else:
            top = MISSING_REPORT_FIX[
                MISSING_REPORT_FIX["Missing_Count"] > 0
            ].head(15)

            if not top.empty:
                chart_fix(
                    px.bar(
                        top.sort_values("Missing_Count"),
                        x="Missing_Count",
                        y="Column",
                        orientation="h",
                        title="Top 15 Columns with Missing Values",
                        color_discrete_sequence=["#9a8c98"]
                    )
                )

            table_fix(MISSING_REPORT_FIX, 30)
            note_fix(
                f"Missing cells: {total_missing_fix:,}. "
                "This analysis identifies missingness but does not "
                "establish that all missing values are acceptable "
                "or that imputation has been completed."
            )

    elif page == "Integration":
        table_fix(
            VALIDATION_FIX[
                VALIDATION_FIX["Category"].isin(
                    ["Python", "Integration", "Navigation", "Runtime"]
                )
            ]
        )
        note_fix(
            "The existing stream.py remains the main application. "
            "Successful syntax and import checks do not guarantee "
            "that every dashboard renders correctly."
        )

    elif page == "Exports":
        table_fix(
            VALIDATION_FIX[
                VALIDATION_FIX["Category"].isin(
                    ["Exports", "CSV Assets"]
                )
            ]
        )
        note_fix(
            "Validation and missingness reports have been saved "
            f"under {REPORT_FIX}."
        )

    elif page == "Recommendations":
        issues = VALIDATION_FIX[
            VALIDATION_FIX["Status"] != "PASS"
        ]

        if issues.empty:
            note_fix("No outstanding issues detected in these checks.")
        else:
            for _, item in issues.iterrows():
                note_fix(
                    f"{item['Category']} · {item['Check']}: "
                    f"{item['Details']}"
                )

        note_fix(
            "Before client delivery, verify the actual dashboard "
            "pages, API data, filters and downloads."
        )

NAV_FIX = widgets.ToggleButtons(
    options=[
        "Executive Overview",
        "Validation Results",
        "API Connectivity",
        "Data Quality",
        "Integration",
        "Exports",
        "Recommendations"
    ],
    value="Executive Overview",
    layout=widgets.Layout(width="100%")
)

OUTPUT_FIX = widgets.Output(
    layout=widgets.Layout(width="100%")
)

def render_fix(change=None):
    with OUTPUT_FIX:
        clear_output(wait=True)
        draw_fix(NAV_FIX.value)

NAV_FIX.observe(render_fix, names="value")

DASHBOARD_FIX = widgets.VBox([
    NAV_FIX,
    OUTPUT_FIX
])

display(DASHBOARD_FIX)
render_fix()
```

```python
ROOT39 = Path(globals().get("ROOT38", globals().get("ROOT37", Path.cwd()))).resolve()
REPORT39 = ROOT39 / "reports" / "submission"
REPORT39.mkdir(parents=True, exist_ok=True)
STREAM39 = ROOT39 / "stream.py"
MODULE39 = ROOT39 / "frontend_pages" / "section34_overview.py"
ZIP39 = REPORT39 / "section39_submission_evidence.zip"

ARTIFACTS39 = [
    ("Application", "Main Streamlit Application", "stream.py"),
    ("Application", "Executive Dashboard Module", "frontend_pages/section34_overview.py"),
    ("Deployment", "Dependency Requirements", "requirements.txt"),
    ("Analytics", "Monthly Market Trends", "reports/frontend/section34_monthly_trends.csv"),
    ("Analytics", "Emerging Job Families", "reports/frontend/section34_emerging_families.csv"),
    ("Analytics", "Geographic Coverage", "reports/frontend/section34_country_coverage.csv"),
    ("Validation", "Section 35 Validation", "reports/validation/section35_corrected_validation.csv"),
    ("Validation", "Section 36 Validation", "reports/deployment/section36_final_validation.csv"),
    ("Documentation", "Section 37 Client Handover", "reports/handover/section37_client_handover.md"),
    ("Acceptance", "Section 38 UAT Register", "reports/acceptance/section38_acceptance_register.csv"),
    ("Acceptance", "Section 38 Evidence Inventory", "reports/acceptance/section38_evidence_inventory.csv")
]

INVENTORY39 = pd.DataFrame([
    {
        "Category": category,
        "Deliverable": name,
        "File": relative,
        "Status": "PRESENT" if (ROOT39 / relative).is_file() else "MISSING",
        "Size_KB": (
            round((ROOT39 / relative).stat().st_size / 1024, 2)
            if (ROOT39 / relative).is_file() else None
        )
    }
    for category, name, relative in ARTIFACTS39
])

CHECKS39 = []

def check39(category, name, status, evidence):
    CHECKS39.append({
        "Category": category,
        "Check": name,
        "Status": status,
        "Evidence": str(evidence)
    })

for item in INVENTORY39.itertuples():
    check39(
        item.Category,
        item.Deliverable,
        "PASS" if item.Status == "PRESENT" else "FAIL",
        item.File
    )

for label, path in [
    ("Main Streamlit Syntax", STREAM39),
    ("Section 34 Module Syntax", MODULE39)
]:
    if not path.is_file():
        check39("Syntax", label, "FAIL", "File not found")
    else:
        try:
            ast.parse(
                path.read_text(encoding="utf-8-sig"),
                filename=str(path)
            )
            check39("Syntax", label, "PASS", "Python syntax verified")
        except (SyntaxError, UnicodeError, OSError) as exc:
            check39("Syntax", label, "FAIL", str(exc))

UAT_PATH39 = ROOT39 / "reports" / "acceptance" / "section38_acceptance_register.csv"
UAT39 = pd.DataFrame()

if UAT_PATH39.is_file():
    try:
        UAT39 = pd.read_csv(UAT_PATH39).fillna("")
        required39 = {"Test_ID", "Status", "Evidence"}

        if not required39.issubset(UAT39.columns):
            check39(
                "Acceptance", "UAT Register Schema",
                "FAIL", "Required columns missing"
            )
            UAT39 = pd.DataFrame()

        elif UAT39.empty:
            check39(
                "Acceptance", "UAT Completion",
                "REVIEW", "Acceptance register contains no tests"
            )

        else:
            valid_pass39 = (
                UAT39["Status"].astype(str).eq("PASS")
                & UAT39["Evidence"].astype(str).str.strip().ne("")
            )

            passed_uat39 = int(valid_pass39.sum())
            total_uat39 = len(UAT39)

            check39(
                "Acceptance",
                "UAT Completion",
                "PASS" if passed_uat39 == total_uat39 else "REVIEW",
                f"{passed_uat39}/{total_uat39} tests marked PASS with evidence references"
            )

    except (OSError, ValueError, pd.errors.ParserError) as exc:
        UAT39 = pd.DataFrame()
        check39("Acceptance", "UAT Register", "FAIL", str(exc))
else:
    check39(
        "Acceptance", "UAT Register",
        "FAIL", "Acceptance register not found"
    )

PRIOR39 = {}

for section, relative in [
    ("Section 35", "reports/validation/section35_corrected_validation.csv"),
    ("Section 36", "reports/deployment/section36_final_validation.csv")
]:
    path = ROOT39 / relative

    if not path.is_file():
        check39(
            "Prior Validation", section,
            "REVIEW", "Validation report unavailable"
        )
        continue

    try:
        frame = pd.read_csv(path)

        if frame.empty or "Status" not in frame.columns:
            check39(
                "Prior Validation", section,
                "REVIEW", "No usable validation statuses"
            )
            continue

        PRIOR39[section] = frame

        unresolved = int(
            (~frame["Status"].astype(str).str.upper().eq("PASS")).sum()
        )

        check39(
            "Prior Validation", section,
            "PASS" if unresolved == 0 else "REVIEW",
            f"{unresolved} outstanding validation entries"
        )

    except (OSError, ValueError, pd.errors.ParserError) as exc:
        check39("Prior Validation", section, "FAIL", str(exc))

EVIDENCE_DIR39 = ROOT39 / "reports" / "acceptance" / "evidence"

IMAGE_FILES39 = (
    sorted([
        p for p in EVIDENCE_DIR39.iterdir()
        if p.is_file()
        and p.suffix.lower() in (".png", ".jpg", ".jpeg", ".webp")
    ])
    if EVIDENCE_DIR39.is_dir() else []
)

EVIDENCE39 = pd.DataFrame([
    {
        "Filename": p.name,
        "Size_KB": round(p.stat().st_size / 1024, 2),
        "Path": str(p.relative_to(ROOT39))
    }
    for p in IMAGE_FILES39
], columns=["Filename", "Size_KB", "Path"])

check39(
    "Evidence",
    "Screenshot Availability",
    "REVIEW",
    f"{len(IMAGE_FILES39)} screenshots found; manual completeness and privacy review required"
)

AUDIT39 = pd.DataFrame(CHECKS39)

PASSES39 = int((AUDIT39["Status"] == "PASS").sum())
REVIEWS39 = int((AUDIT39["Status"] == "REVIEW").sum())
FAILS39 = int((AUDIT39["Status"] == "FAIL").sum())
TOTAL39 = len(AUDIT39)

SUMMARY39 = pd.DataFrame({
    "Metric": [
        "Total Checks", "Passed",
        "Review Required", "Failed", "Audit Pass Rate"
    ],
    "Value": [
        TOTAL39,
        PASSES39,
        REVIEWS39,
        FAILS39,
        round(PASSES39 / TOTAL39 * 100, 2) if TOTAL39 else 0
    ]
})

ISSUES39 = AUDIT39[AUDIT39["Status"] != "PASS"].copy()

SIGNOFF39 = (
    "AUTOMATED AUDIT PASSED - MANUAL APPROVAL REQUIRED"
    if REVIEWS39 == 0 and FAILS39 == 0
    else "NOT READY FOR FINAL SIGN-OFF"
)

INVENTORY39.to_csv(REPORT39 / "section39_deliverables.csv", index=False)
AUDIT39.to_csv(REPORT39 / "section39_audit_results.csv", index=False)
SUMMARY39.to_csv(REPORT39 / "section39_audit_summary.csv", index=False)
ISSUES39.to_csv(REPORT39 / "section39_outstanding_issues.csv", index=False)
EVIDENCE39.to_csv(REPORT39 / "section39_evidence_inventory.csv", index=False)

MANIFEST39 = {
    "project": "Job Market Analysis and Recommendation System",
    "generated_at": datetime.now().astimezone().isoformat(),
    "main_application": "stream.py",
    "audit_status": SIGNOFF39,
    "total_checks": TOTAL39,
    "passed": PASSES39,
    "reviews": REVIEWS39,
    "failed": FAILS39,
    "note": "Automated checks do not constitute client approval or runtime certification."
}

MANIFEST_PATH39 = REPORT39 / "section39_manifest.json"
MANIFEST_PATH39.write_text(
    json.dumps(MANIFEST39, indent=2),
    encoding="utf-8"
)

PACKAGE_FILES39 = [
    "section39_deliverables.csv",
    "section39_audit_results.csv",
    "section39_audit_summary.csv",
    "section39_outstanding_issues.csv",
    "section39_evidence_inventory.csv",
    "section39_manifest.json"
]

HANDOVER_FILE39 = (
    ROOT39 / "reports" / "handover" / "section37_client_handover.md"
)

with ZipFile(ZIP39, "w", ZIP_DEFLATED) as bundle:
    for filename in PACKAGE_FILES39:
        bundle.write(REPORT39 / filename, arcname=filename)

    if HANDOVER_FILE39.is_file():
        bundle.write(
            HANDOVER_FILE39,
            arcname="section37_client_handover.md"
        )

with ZipFile(ZIP39, "r") as bundle:
    ZIP_VALID39 = bundle.testzip() is None
    PACKAGE_CONTENTS39 = bundle.namelist()

PACKAGE39 = pd.DataFrame({
    "Packaged File": PACKAGE_CONTENTS39
})

CSS39 = """
<style>
.s39hero{
    background:linear-gradient(135deg,#22223b,#32344f 52%,#4a4e69);
    padding:25px 29px;border-radius:16px;color:white;
    font-family:Arial;margin:10px 0 16px;
}
.s39hero small{
    color:#c9ada7;letter-spacing:1.7px;
    font-size:10px;font-weight:700;
}
.s39hero h2{
    color:white!important;font-size:26px;margin:8px 0;
}
.s39hero p{
    color:#f2e9e4;font-size:12px;margin:0;
}
.s39metrics{
    display:grid;grid-template-columns:repeat(4,1fr);
    gap:10px;margin:14px 0;
}
.s39metric{
    background:white;border:1px solid #e6dfdc;
    border-radius:11px;padding:17px;font-family:Arial;
}
.s39metric span{
    display:block;color:#9a8c98;font-size:10px;
}
.s39metric b{
    display:block;color:#22223b;font-size:24px;
    margin-top:7px;
}
.s39note{
    background:#f2e9e4;border-left:4px solid #9a8c98;
    padding:14px;border-radius:9px;color:#4a4e69;
    margin:12px 0;font:12px Arial;line-height:1.7;
}
</style>
"""

display(HTML(CSS39))

display(HTML("""
<div class="s39hero">
<small>SECTION 39 · FINAL SUBMISSION AUDIT</small>
<h2>Evidence Packaging & Delivery Verification</h2>
<p>Deliverables · Acceptance · Validation · Submission Readiness</p>
</div>
"""))

metric_html39 = "".join(
    '<div class="s39metric">'
    f'<span>{label}</span><b>{value}</b>'
    '</div>'
    for label, value in [
        ("TOTAL CHECKS", TOTAL39),
        ("PASSED", PASSES39),
        ("REVIEWS", REVIEWS39),
        ("FAILED", FAILS39)
    ]
)

display(HTML(
    '<div class="s39metrics">' + metric_html39 + '</div>'
))

def _note39(message):
    display(HTML(
        '<div class="s39note">' + escape(str(message)) + '</div>'
    ))

def _table39(df, n=50):
    if df.empty:
        _note39("No records available.")
    else:
        display(HTML(
            df.head(n).to_html(index=False, escape=True, border=0)
        ))

def _draw39(view):

    if view == "Executive Overview":

        flow39 = (
            INVENTORY39.groupby(
                ["Category", "Status"],
                as_index=False
            )
            .size()
            .rename(columns={"size": "Count"})
        )

        categories39 = sorted(flow39["Category"].unique())
        statuses39 = ["PRESENT", "MISSING"]

        labels39 = categories39 + statuses39

        sources39 = [
            categories39.index(category)
            for category in flow39["Category"]
        ]

        targets39 = [
            len(categories39) + statuses39.index(status)
            for status in flow39["Status"]
        ]

        values39 = flow39["Count"].tolist()

        node_colors39 = (
            ["#4a4e69"] * len(categories39)
            + ["#9a8c98", "#c9ada7"]
        )

        link_colors39 = [
            "rgba(74,78,105,0.45)"
            if status == "PRESENT"
            else "rgba(201,173,167,0.65)"
            for status in flow39["Status"]
        ]

        fig39 = go.Figure(
            data=[
                go.Sankey(
                    arrangement="snap",
                    node=dict(
                        pad=20,
                        thickness=22,
                        line=dict(
                            color="#22223b",
                            width=0.5
                        ),
                        label=labels39,
                        color=node_colors39
                    ),
                    link=dict(
                        source=sources39,
                        target=targets39,
                        value=values39,
                        color=link_colors39
                    )
                )
            ]
        )

        fig39.update_layout(
            title="Submission Deliverables — Category to Verification Status",
            font=dict(
                family="Arial",
                size=12,
                color="#22223b"
            ),
            height=510,
            paper_bgcolor="white",
            margin=dict(l=20, r=20, t=65, b=20)
        )

        display(fig39)
        _table39(SUMMARY39)

        _note39(
            "Insight: Flow thickness represents the number of "
            "required deliverables in each category and status."
        )

        _note39(
            "Interpretation: PRESENT confirms that a file exists, "
            "not that its contents, functionality or security "
            "have been fully verified."
        )

        _note39(
            "Recommendation: Resolve missing files and outstanding "
            "validation issues before final submission."
        )

    elif view == "Deliverables":
        _table39(INVENTORY39)

        present = int((INVENTORY39["Status"] == "PRESENT").sum())
        missing = int((INVENTORY39["Status"] == "MISSING").sum())

        _note39(
            f"Insight: {present} required deliverables are present "
            f"and {missing} are missing."
        )

        _note39(
            "Recommendation: Review the actual contents of every "
            "present file before including it in the submission."
        )

    elif view == "Validation Audit":
        _table39(AUDIT39, 100)

        _note39(
            f"Insight: {PASSES39} passed checks, {REVIEWS39} review "
            f"items and {FAILS39} failed checks."
        )

        _note39(
            "Recommendation: Resolve failed checks first, followed "
            "by review items and runtime verification."
        )

    elif view == "Client Acceptance":
        if UAT39.empty:
            _note39("No valid UAT register is available.")
        else:
            _table39(UAT39, 30)

            _note39(
                "Interpretation: A recorded PASS requires genuine "
                "supporting evidence and independent verification."
            )

            _note39(
                "Recommendation: Complete untested requirements and "
                "obtain authorized client acceptance."
            )

    elif view == "Evidence":
        _table39(EVIDENCE39)

        _note39(
            f"Insight: {len(IMAGE_FILES39)} screenshot evidence "
            "files were detected."
        )

        _note39(
            "Recommendation: Check evidence completeness, "
            "readability, privacy and alignment with UAT results."
        )

    elif view == "Outstanding Issues":
        if ISSUES39.empty:
            _note39(
                "No unresolved issues were detected by the "
                "automated audit."
            )
        else:
            _table39(ISSUES39)

            _note39(
                f"{len(ISSUES39)} checks remain unresolved "
                "or require verification."
            )

            _note39(
                "Recommendation: Address outstanding items "
                "and rerun the audit before submission."
            )

    elif view == "Submission Package":
        _table39(PACKAGE39)

        _note39(
            "ZIP integrity: "
            + ("Verified" if ZIP_VALID39 else "Failed")
        )

        _note39(
            "Submission package location: " + str(ZIP39)
        )

        _note39(
            "The package contains audit documentation only. "
            "Application source files, credentials and raw data "
            "are intentionally excluded."
        )

    elif view == "Recommendations":

        recommendations39 = []

        if FAILS39:
            recommendations39.append(
                "Correct failed checks and regenerate affected outputs."
            )

        if REVIEWS39:
            recommendations39.append(
                "Verify outstanding validation items and UAT evidence."
            )

        recommendations39.extend([
            "Verify Flask API responses in the running environment.",
            "Test all dashboard pages using the existing stream.py.",
            "Check charts, interactive filters and CSV downloads.",
            "Review screenshots and reports for sensitive information.",
            "Obtain explicit client approval before final sign-off."
        ])

        for recommendation in recommendations39:
            _note39(recommendation)

NAV39 = widgets.ToggleButtons(
    options=[
        "Executive Overview",
        "Deliverables",
        "Validation Audit",
        "Client Acceptance",
        "Evidence",
        "Outstanding Issues",
        "Submission Package",
        "Recommendations"
    ],
    value="Executive Overview",
    layout=widgets.Layout(width="100%")
)

OUTPUT39 = widgets.Output(
    layout=widgets.Layout(width="100%")
)

def render39(change=None):
    with OUTPUT39:
        clear_output(wait=True)
        _draw39(NAV39.value)

NAV39.observe(render39, names="value")

DASHBOARD39 = widgets.VBox([
    NAV39,
    OUTPUT39
])

display(DASHBOARD39)
render39()

SECTION39_STATUS = SIGNOFF39
```

```python
ROOT40 = Path(globals().get("ROOT39", globals().get("ROOT38", Path.cwd()))).resolve()
REPORT40 = ROOT40 / "reports" / "executive"
REPORT40.mkdir(parents=True, exist_ok=True)
STREAM40 = ROOT40 / "stream.py"
AUDIT_PATH40 = ROOT40 / "reports" / "submission" / "section39_audit_results.csv"
UAT_PATH40 = ROOT40 / "reports" / "acceptance" / "section38_acceptance_register.csv"
INVENTORY_PATH40 = ROOT40 / "reports" / "submission" / "section39_deliverables.csv"

def _load40(path, required):
    if not path.is_file():
        return pd.DataFrame(), "File missing"
    try:
        data = pd.read_csv(path).fillna("")
        missing = sorted(set(required) - set(data.columns))
        if missing:
            return pd.DataFrame(), "Missing columns: " + ", ".join(missing)
        if data.empty:
            return pd.DataFrame(), "No records available"
        return data, "Available"
    except (OSError, ValueError, pd.errors.ParserError) as exc:
        return pd.DataFrame(), str(exc)

AUDIT40, AUDIT_SOURCE40 = _load40(
    AUDIT_PATH40,
    ["Category", "Check", "Status"]
)

UAT40, UAT_SOURCE40 = _load40(
    UAT_PATH40,
    ["Test_ID", "Category", "Status", "Evidence"]
)

INVENTORY40, INVENTORY_SOURCE40 = _load40(
    INVENTORY_PATH40,
    ["Category", "Deliverable", "Status"]
)

STATUS_ORDER40 = [
    "PASS", "REVIEW", "FAIL",
    "BLOCKED", "NOT TESTED", "UNVERIFIED"
]

RECORDS40 = []

if not AUDIT40.empty:
    for item in AUDIT40.itertuples(index=False):
        status = str(item.Status).strip().upper()
        if status == "WARN":
            status = "REVIEW"
        if status not in STATUS_ORDER40:
            status = "UNVERIFIED"

        RECORDS40.append({
            "Source": "Submission Audit",
            "Category": "Audit · " + str(item.Category),
            "Item": str(item.Check),
            "Status": status,
            "Evidence": str(getattr(item, "Evidence", ""))
        })
else:
    RECORDS40.append({
        "Source": "Submission Audit",
        "Category": "Audit · Source",
        "Item": "Submission audit availability",
        "Status": "UNVERIFIED",
        "Evidence": AUDIT_SOURCE40
    })

if not UAT40.empty:
    for item in UAT40.itertuples(index=False):
        status = str(item.Status).strip().upper()
        evidence = str(item.Evidence).strip()

        if status == "PASS" and not evidence:
            status = "UNVERIFIED"

        if status not in STATUS_ORDER40:
            status = "UNVERIFIED"

        RECORDS40.append({
            "Source": "Client Acceptance",
            "Category": "UAT · " + str(item.Category),
            "Item": str(item.Test_ID),
            "Status": status,
            "Evidence": evidence
        })
else:
    RECORDS40.append({
        "Source": "Client Acceptance",
        "Category": "UAT · Source",
        "Item": "Acceptance register availability",
        "Status": "UNVERIFIED",
        "Evidence": UAT_SOURCE40
    })

EVALUATION40 = pd.DataFrame(RECORDS40)

AUDIT_PASSES40 = int(
    ((EVALUATION40["Source"] == "Submission Audit") &
     (EVALUATION40["Status"] == "PASS")).sum()
)

UAT_PASSES40 = int(
    ((EVALUATION40["Source"] == "Client Acceptance") &
     (EVALUATION40["Status"] == "PASS")).sum()
)

UNRESOLVED40 = int(
    (EVALUATION40["Status"] != "PASS").sum()
)

TOTAL40 = len(EVALUATION40)
PASSED40 = int((EVALUATION40["Status"] == "PASS").sum())
COVERAGE40 = round(PASSED40 / TOTAL40 * 100, 2) if TOTAL40 else 0

SCORECARD40 = (
    EVALUATION40.groupby("Category")
    .agg(
        Total=("Status", "size"),
        Passed=("Status", lambda x: int((x == "PASS").sum())),
        Unresolved=("Status", lambda x: int((x != "PASS").sum()))
    )
    .reset_index()
)

SCORECARD40["Recorded_Pass_Rate_%"] = (
    SCORECARD40["Passed"] / SCORECARD40["Total"] * 100
).round(2)

SCORECARD40["Assessment"] = SCORECARD40["Unresolved"].apply(
    lambda x: "Recorded checks passed" if x == 0 else "Attention required"
)

HEATMAP40 = pd.crosstab(
    EVALUATION40["Category"],
    EVALUATION40["Status"]
).reindex(columns=STATUS_ORDER40, fill_value=0)

HEATMAP40 = HEATMAP40.loc[
    :,
    HEATMAP40.sum(axis=0).gt(0)
]

ISSUES40 = EVALUATION40[
    EVALUATION40["Status"] != "PASS"
].copy()

DELIVERABLES40 = pd.DataFrame(
    columns=["Category", "Deliverable", "Status"]
)

if not INVENTORY40.empty:
    DELIVERABLES40 = INVENTORY40.copy()

    DELIVERABLES40["Status"] = (
        DELIVERABLES40["Status"]
        .astype(str)
        .str.upper()
        .str.strip()
    )

PRESENT40 = int(
    (DELIVERABLES40["Status"] == "PRESENT").sum()
) if not DELIVERABLES40.empty else 0

MISSING40 = int(
    (DELIVERABLES40["Status"] == "MISSING").sum()
) if not DELIVERABLES40.empty else 0

ACTIONS40 = []

for item in ISSUES40.itertuples(index=False):
    ACTIONS40.append({
        "Priority": (
            "High" if item.Status in ("FAIL", "BLOCKED")
            else "Review"
        ),
        "Category": item.Category,
        "Issue": item.Item,
        "Status": item.Status,
        "Next_Action": (
            "Correct the failure and rerun verification"
            if item.Status == "FAIL"
            else "Resolve the blocker and repeat the acceptance test"
            if item.Status == "BLOCKED"
            else "Complete verification and retain supporting evidence"
        )
    })

ACTION_PLAN40 = pd.DataFrame(
    ACTIONS40,
    columns=["Priority", "Category", "Issue", "Status", "Next_Action"]
)

if not ACTION_PLAN40.empty:
    ACTION_PLAN40["Priority_Order"] = (
        ACTION_PLAN40["Priority"].map({
            "High": 0,
            "Review": 1
        })
    )

    ACTION_PLAN40 = (
        ACTION_PLAN40
        .sort_values(["Priority_Order", "Category"])
        .drop(columns="Priority_Order")
    )

SUMMARY40 = pd.DataFrame({
    "Metric": [
        "Total Evaluation Records",
        "Recorded Passes",
        "Submission Audit Passes",
        "UAT Passes With Evidence References",
        "Unresolved Items",
        "Recorded Pass Rate (%)",
        "Deliverables Present",
        "Deliverables Missing"
    ],
    "Value": [
        TOTAL40,
        PASSED40,
        AUDIT_PASSES40,
        UAT_PASSES40,
        UNRESOLVED40,
        COVERAGE40,
        PRESENT40,
        MISSING40
    ]
})

SECTION40_STATUS = (
    "RECORDED CHECKS PASSED - MANUAL SIGN-OFF REQUIRED"
    if UNRESOLVED40 == 0
    else "FINAL VERIFICATION INCOMPLETE"
)

EVALUATION40.to_csv(
    REPORT40 / "section40_evaluation_register.csv",
    index=False
)

SCORECARD40.to_csv(
    REPORT40 / "section40_category_scorecard.csv",
    index=False
)

HEATMAP40.to_csv(
    REPORT40 / "section40_verification_heatmap.csv"
)

SUMMARY40.to_csv(
    REPORT40 / "section40_executive_summary.csv",
    index=False
)

ACTION_PLAN40.to_csv(
    REPORT40 / "section40_action_plan.csv",
    index=False
)

DELIVERABLES40.to_csv(
    REPORT40 / "section40_deliverables.csv",
    index=False
)

PRESENTATION40 = f"""# Job Market Analysis & Recommendation System

## Final Executive Evaluation

Generated: {datetime.now().astimezone().strftime('%Y-%m-%d %H:%M')}

### Overall Assessment

Status: {SECTION40_STATUS}

- Evaluation records: {TOTAL40}
- Recorded passes: {PASSED40}
- Unresolved items: {UNRESOLVED40}
- Recorded pass rate: {COVERAGE40:.2f}%
- Submission audit passes: {AUDIT_PASSES40}
- UAT passes with evidence references: {UAT_PASSES40}
- Deliverables present: {PRESENT40}
- Deliverables missing: {MISSING40}

### Project Capabilities

The project is intended to provide historical job-market analysis,
emerging-job-family comparisons, geographic insights, dashboard
visualizations and analytical CSV exports.

### Architecture

Streamlit provides the user interface through the existing stream.py.
Flask provides the intended API interface to backend operations.
Database and model access should remain behind the backend.

### Evaluation Interpretation

The recorded pass rate reflects test-register statuses only.
It is not a machine-learning accuracy score or a measurement
of actual runtime availability.

A recorded PASS does not independently verify the authenticity
or completeness of its supporting evidence.

### Remaining Work

{len(ACTION_PLAN40)} unresolved evaluation records require attention.

### Final Recommendation

Correct failed and blocked checks, verify live Flask and Streamlit
behavior, review security and data handling, and obtain authorized
client approval before final delivery.

### Source Availability

- Section 39 audit: {AUDIT_SOURCE40}
- Section 38 acceptance: {UAT_SOURCE40}
- Deliverables inventory: {INVENTORY_SOURCE40}

### Final Status

{SECTION40_STATUS}
"""

PRESENTATION_PATH40 = (
    REPORT40 / "section40_client_presentation_summary.md"
)

PRESENTATION_PATH40.write_text(
    PRESENTATION40,
    encoding="utf-8"
)

CSS40 = """
<style>
.s40hero{
    background:linear-gradient(135deg,#22223b,#32344f 52%,#4a4e69);
    padding:25px 29px;
    border-radius:16px;
    color:white;
    font-family:Arial;
    margin:10px 0 16px;
}
.s40hero small{
    color:#c9ada7;
    font-size:10px;
    font-weight:700;
    letter-spacing:1.7px;
}
.s40hero h2{
    color:white!important;
    font-size:26px;
    margin:8px 0;
}
.s40hero p{
    color:#f2e9e4;
    font-size:12px;
    margin:0;
}
.s40metrics{
    display:grid;
    grid-template-columns:repeat(4,1fr);
    gap:10px;
    margin:15px 0;
}
.s40metric{
    padding:17px;
    border:1px solid #e6dfdc;
    border-radius:11px;
    background:white;
    font-family:Arial;
}
.s40metric span{
    color:#9a8c98;
    font-size:10px;
    display:block;
}
.s40metric b{
    color:#22223b;
    font-size:24px;
    display:block;
    margin-top:7px;
}
.s40note{
    background:#f2e9e4;
    border-left:4px solid #9a8c98;
    padding:14px;
    border-radius:9px;
    color:#4a4e69;
    font:12px Arial;
    line-height:1.7;
    margin:12px 0;
}
</style>
"""

display(HTML(CSS40))

display(HTML("""
<div class="s40hero">
<small>SECTION 40 · EXECUTIVE EVALUATION</small>
<h2>Project Performance & Client Presentation Scorecard</h2>
<p>Validation · Acceptance · Deliverables · Decision Readiness</p>
</div>
"""))

KPI_HTML40 = "".join(
    '<div class="s40metric">'
    f'<span>{escape(str(label))}</span>'
    f'<b>{escape(str(value))}</b>'
    '</div>'
    for label, value in [
        ("TOTAL CHECKS", TOTAL40),
        ("RECORDED PASSES", PASSED40),
        ("UNRESOLVED", UNRESOLVED40),
        ("PASS RATE", f"{COVERAGE40:.1f}%")
    ]
)

display(HTML(
    '<div class="s40metrics">'
    + KPI_HTML40
    + '</div>'
))

def _note40(message):
    display(HTML(
        '<div class="s40note">'
        + escape(str(message))
        + '</div>'
    ))

def _table40(df, n=60):
    if df.empty:
        _note40("No records available.")
    else:
        display(HTML(
            df.head(n).to_html(
                index=False,
                escape=True,
                border=0
            )
        ))

def _draw40(view):

    if view == "Executive Overview":

        colorscale40 = [
            [0.0, "#f2e9e4"],
            [0.25, "#c9ada7"],
            [0.55, "#9a8c98"],
            [1.0, "#4a4e69"]
        ]

        fig40 = go.Figure(
            data=go.Heatmap(
                z=HEATMAP40.values,
                x=HEATMAP40.columns.tolist(),
                y=HEATMAP40.index.tolist(),
                colorscale=colorscale40,
                zmin=0,
                zmax=max(
                    1,
                    int(HEATMAP40.values.max())
                ),
                text=HEATMAP40.values,
                texttemplate="%{text}",
                textfont=dict(size=11),
                hovertemplate=(
                    "Category: %{y}<br>"
                    "Status: %{x}<br>"
                    "Recorded checks: %{z}<extra></extra>"
                ),
                colorbar=dict(
                    title="Checks"
                )
            )
        )

        fig40.update_layout(
            title="Final Verification Heatmap — Category vs Status",
            template="plotly_white",
            height=max(
                450,
                35 * len(HEATMAP40) + 150
            ),
            font=dict(
                family="Arial",
                size=11,
                color="#22223b"
            ),
            margin=dict(
                l=20,
                r=20,
                t=65,
                b=45
            ),
            paper_bgcolor="white",
            yaxis=dict(autorange="reversed")
        )

        display(fig40)

        _table40(SUMMARY40)

        _note40(
            "Insight: The heatmap identifies categories with "
            "the highest concentrations of recorded verification "
            "results and outstanding issues."
        )

        _note40(
            "Interpretation: Each cell represents a count of "
            "checks in a category and status, not a percentage "
            "or a measurement of model predictive performance."
        )

        _note40(
            "Recommendation: Address failed, blocked and "
            "unverified records before client approval."
        )

    elif view == "Category Scorecard":

        _table40(
            SCORECARD40.sort_values(
                ["Unresolved", "Category"],
                ascending=[False, True]
            )
        )

        if not SCORECARD40.empty:
            top40 = SCORECARD40.sort_values(
                "Unresolved",
                ascending=False
            ).iloc[0]

            _note40(
                f"Insight: {top40['Category']} has "
                f"{top40['Unresolved']} unresolved records."
            )

        _note40(
            "Recommendation: Prioritize categories with "
            "more unresolved checks and review their evidence."
        )

    elif view == "Client Acceptance":

        if UAT40.empty:
            _note40(
                "The client acceptance register is unavailable "
                "or does not match the expected schema."
            )
        else:
            _table40(UAT40, 35)

            _note40(
                f"Recorded UAT passes with evidence references: "
                f"{UAT_PASSES40}."
            )

            _note40(
                "Interpretation: Recorded results do not "
                "independently establish that every feature "
                "was tested successfully."
            )

            _note40(
                "Recommendation: Verify evidence and obtain "
                "authorized client sign-off."
            )

    elif view == "Deliverables":

        if DELIVERABLES40.empty:
            _note40(
                "The previous submission inventory "
                "is unavailable."
            )
        else:
            _table40(DELIVERABLES40)

            _note40(
                f"Insight: {PRESENT40} deliverables are "
                f"recorded present and {MISSING40} missing."
            )

            _note40(
                "Recommendation: Verify the current existence, "
                "contents and security of each deliverable "
                "before sharing it with the client."
            )

    elif view == "Outstanding Issues":

        if ISSUES40.empty:
            _note40(
                "No unresolved statuses were recorded in "
                "the available evaluation sources."
            )
        else:
            _table40(ISSUES40, 100)

            _note40(
                f"{UNRESOLVED40} evaluation records "
                "still require attention."
            )

            _note40(
                "Recommendation: Resolve failures first, "
                "then review blocked and unverified checks."
            )

    elif view == "Action Plan":

        if ACTION_PLAN40.empty:
            _note40(
                "No corrective actions were generated "
                "from the available statuses."
            )
        else:
            _table40(ACTION_PLAN40, 100)

            _note40(
                "Recommendation: Record corrections and "
                "rerun the relevant validation sections."
            )

    elif view == "Client Presentation":

        display(HTML(
            '<pre style="white-space:pre-wrap;'
            'font:12px Arial;color:#22223b;">'
            + escape(PRESENTATION40)
            + '</pre>'
        ))

        _note40(
            "This summary is generated from recorded "
            "project evaluation data. Review it before "
            "including it in a client presentation."
        )

    elif view == "Reports":

        _table40(pd.DataFrame({
            "Report": [
                "Evaluation Register",
                "Category Scorecard",
                "Verification Heatmap",
                "Executive Summary",
                "Corrective Action Plan",
                "Deliverables Inventory",
                "Client Presentation Summary"
            ],
            "Filename": [
                "section40_evaluation_register.csv",
                "section40_category_scorecard.csv",
                "section40_verification_heatmap.csv",
                "section40_executive_summary.csv",
                "section40_action_plan.csv",
                "section40_deliverables.csv",
                "section40_client_presentation_summary.md"
            ]
        }))

        _note40(
            "Reports saved to: " + str(REPORT40)
        )

NAV40 = widgets.ToggleButtons(
    options=[
        "Executive Overview",
        "Category Scorecard",
        "Client Acceptance",
        "Deliverables",
        "Outstanding Issues",
        "Action Plan",
        "Client Presentation",
        "Reports"
    ],
    value="Executive Overview",
    layout=widgets.Layout(width="100%")
)

OUTPUT40 = widgets.Output(
    layout=widgets.Layout(width="100%")
)

def render40(change=None):
    with OUTPUT40:
        clear_output(wait=True)
        _draw40(NAV40.value)

NAV40.observe(render40, names="value")

DASHBOARD40 = widgets.VBox([
    NAV40,
    OUTPUT40
])

display(DASHBOARD40)
render40()
```

```python
ROOT41 = Path(globals().get("ROOT40", globals().get("ROOT39", Path.cwd()))).resolve()
REPORT41 = ROOT41 / "reports" / "conclusions"
REPORT41.mkdir(parents=True, exist_ok=True)
STREAM41 = ROOT41 / "stream.py"
EVALUATION_PATH41 = ROOT41 / "reports" / "executive" / "section40_evaluation_register.csv"
SCORECARD_PATH41 = ROOT41 / "reports" / "executive" / "section40_category_scorecard.csv"

def _load41(path, required):
    if not path.is_file():
        return pd.DataFrame(), "File unavailable"
    try:
        frame = pd.read_csv(path)
        missing = set(required) - set(frame.columns)
        if missing:
            return pd.DataFrame(), "Missing columns: " + ", ".join(sorted(missing))
        if frame.empty:
            return pd.DataFrame(), "No records"
        return frame, "Available"
    except (OSError, ValueError, pd.errors.ParserError) as exc:
        return pd.DataFrame(), str(exc)

EVALUATION41, EVALUATION_SOURCE41 = _load41(
    EVALUATION_PATH41,
    ["Source", "Category", "Item", "Status"]
)

SCORECARD41, SCORECARD_SOURCE41 = _load41(
    SCORECARD_PATH41,
    ["Category", "Total", "Passed", "Unresolved"]
)

MODEL_CANDIDATES41 = [
    "model_comparison_df",
    "model_comparison",
    "model_results_df",
    "model_results",
    "model_metrics_df",
    "model_metrics",
    "results_df"
]

MODEL_COLUMNS41 = [
    "Model", "model", "Model_Name", "model_name",
    "Algorithm", "algorithm"
]

METRIC_ALIASES41 = {
    "Accuracy": ["Accuracy", "accuracy"],
    "Precision": ["Precision", "precision"],
    "Recall": ["Recall", "recall"],
    "F1": ["F1", "F1_Score", "f1", "f1_score"],
    "ROC_AUC": ["ROC_AUC", "roc_auc", "AUC", "auc"],
    "MAE": ["MAE", "mae"],
    "RMSE": ["RMSE", "rmse"],
    "R2": ["R2", "r2", "R2_Score", "r2_score"]
}

def _model_frame41(frame):
    if not isinstance(frame, pd.DataFrame) or frame.empty:
        return pd.DataFrame()

    lookup = {str(c).casefold(): c for c in frame.columns}
    model_col = next(
        (lookup[c.casefold()] for c in MODEL_COLUMNS41
         if c.casefold() in lookup),
        None
    )

    if model_col is None:
        return pd.DataFrame()

    found = {}
    for metric, aliases in METRIC_ALIASES41.items():
        original = next(
            (lookup[a.casefold()] for a in aliases
             if a.casefold() in lookup),
            None
        )
        if original is not None:
            found[metric] = original

    if not found:
        return pd.DataFrame()

    result = frame[[model_col] + list(found.values())].copy()
    result = result.rename(columns={
        model_col: "Model",
        **{original: metric for metric, original in found.items()}
    })

    result["Model"] = result["Model"].astype(str).str.strip()
    result = result[
        result["Model"].ne("")
        & result["Model"].ne("nan")
    ].copy()

    for metric in found:
        result[metric] = pd.to_numeric(
            result[metric], errors="coerce"
        )

    return result.dropna(
        subset=list(found),
        how="all"
    )

MODEL41 = pd.DataFrame()
MODEL_SOURCE41 = "No compatible model-evaluation results found"

for name41 in MODEL_CANDIDATES41:
    candidate41 = globals().get(name41)
    prepared41 = _model_frame41(candidate41)

    if not prepared41.empty:
        MODEL41 = prepared41
        MODEL_SOURCE41 = "Notebook DataFrame: " + name41
        break

if MODEL41.empty:
    model_files41 = [
        ROOT41 / "reports" / "model_comparison.csv",
        ROOT41 / "reports" / "model_metrics.csv",
        ROOT41 / "reports" / "model_evaluation.csv",
        ROOT41 / "model_comparison.csv",
        ROOT41 / "model_metrics.csv"
    ]

    for path41 in model_files41:
        if not path41.is_file():
            continue

        try:
            candidate41 = pd.read_csv(path41)
            prepared41 = _model_frame41(candidate41)

            if not prepared41.empty:
                MODEL41 = prepared41
                MODEL_SOURCE41 = str(path41.relative_to(ROOT41))
                break
        except (OSError, ValueError, pd.errors.ParserError):
            continue

MODEL_METRICS41 = [
    c for c in METRIC_ALIASES41
    if c in MODEL41.columns
]

if not EVALUATION41.empty:
    EVALUATION41["Status"] = (
        EVALUATION41["Status"]
        .fillna("UNVERIFIED")
        .astype(str)
        .str.upper()
        .str.strip()
        .replace({
            "WARN": "REVIEW",
            "": "UNVERIFIED"
        })
    )

    accepted41 = [
        "PASS", "FAIL", "REVIEW",
        "BLOCKED", "NOT TESTED", "UNVERIFIED"
    ]

    EVALUATION41.loc[
        ~EVALUATION41["Status"].isin(accepted41),
        "Status"
    ] = "UNVERIFIED"

    STATUS_COUNTS41 = (
        EVALUATION41["Status"]
        .value_counts()
        .reindex(accepted41, fill_value=0)
    )
else:
    STATUS_COUNTS41 = pd.Series(
        0,
        index=[
            "PASS", "FAIL", "REVIEW",
            "BLOCKED", "NOT TESTED", "UNVERIFIED"
        ]
    )

TOTAL41 = int(STATUS_COUNTS41.sum())
PASSED41 = int(STATUS_COUNTS41["PASS"])
FAILED41 = int(STATUS_COUNTS41["FAIL"])
REVIEW41 = int(STATUS_COUNTS41["REVIEW"])
BLOCKED41 = int(STATUS_COUNTS41["BLOCKED"])
UNTESTED41 = int(STATUS_COUNTS41["NOT TESTED"])
UNVERIFIED41 = int(STATUS_COUNTS41["UNVERIFIED"])

UNRESOLVED41 = TOTAL41 - PASSED41
PASS_RATE41 = round(
    PASSED41 / TOTAL41 * 100, 2
) if TOTAL41 else 0

WATERFALL41 = pd.DataFrame({
    "Stage": [
        "Total Checks",
        "Failed",
        "Blocked",
        "Review",
        "Not Tested",
        "Unverified",
        "Recorded Passes"
    ],
    "Contribution": [
        TOTAL41,
        -FAILED41,
        -BLOCKED41,
        -REVIEW41,
        -UNTESTED41,
        -UNVERIFIED41,
        PASSED41
    ],
    "Type": [
        "absolute",
        "relative",
        "relative",
        "relative",
        "relative",
        "relative",
        "total"
    ]
})

IMPACT41 = pd.DataFrame([
    (
        "Market Intelligence",
        "Historical demand and posting trends",
        "Helps users compare observed market activity",
        "Historical dataset coverage; not live vacancies",
        "Available as a project capability"
    ),
    (
        "Emerging Jobs",
        "28-day job-family growth comparison",
        "Highlights potential changes in role demand",
        "Small baselines and partial periods affect interpretation",
        "Available as a project capability"
    ),
    (
        "Geographic Insights",
        "Country-level posting analysis",
        "Supports geographic opportunity exploration",
        "Hourly-only API coverage; compensation not normalized",
        "Available as a project capability"
    ),
    (
        "Machine Learning",
        "Job-related predictive or recommendation models",
        "May help rank or classify opportunities",
        "Actual accuracy, generalization and fairness require evidence",
        "Review model metrics and evaluation methodology"
    ),
    (
        "Dashboard Experience",
        "Interactive charts, filters and CSV exports",
        "Supports accessible, self-service analysis",
        "Runtime usability and export correctness require testing",
        "Review UAT evidence"
    ),
    (
        "Business Outcomes",
        "Potential improvements to analytical decision-making",
        "Could reduce manual comparison effort",
        "No validated cost, time-saving or ROI measurements supplied",
        "Business impact not quantitatively established"
    )
], columns=[
    "Area", "Project_Capability", "Potential_Value",
    "Limitations", "Evidence_Status"
])

CONCLUSIONS41 = pd.DataFrame([
    (
        "Analytical Contribution",
        "The project organizes historical job-posting evidence into interpretable market views."
    ),
    (
        "Emerging Demand",
        "Job-family growth should be evaluated using percentage change and posting baselines."
    ),
    (
        "Geography",
        "Geographic comparisons require consistent filters and clear coverage definitions."
    ),
    (
        "Machine Learning",
        "Model quality must be supported by appropriate held-out evaluation and task-specific metrics."
    ),
    (
        "Client Readiness",
        "Recorded validation passes do not replace runtime testing or authorized client acceptance."
    ),
    (
        "Future Improvement",
        "Prioritize data freshness, model monitoring, evidence-based evaluation and usability improvements."
    )
], columns=["Topic", "Conclusion"])

RECOMMENDATIONS41 = []

if MODEL41.empty:
    RECOMMENDATIONS41.append(
        ("High", "Model Evaluation",
         "Locate verified held-out model evaluation results and comparison metrics.")
    )
else:
    RECOMMENDATIONS41.append(
        ("Review", "Model Evaluation",
         "Confirm dataset splits, test-set independence, metric definitions and model provenance.")
    )

if FAILED41:
    RECOMMENDATIONS41.append(
        ("High", "Validation",
         f"Resolve {FAILED41} failed evaluation checks and rerun the affected tests.")
    )

if BLOCKED41:
    RECOMMENDATIONS41.append(
        ("High", "Acceptance",
         f"Resolve {BLOCKED41} blocked acceptance checks.")
    )

if REVIEW41 + UNTESTED41 + UNVERIFIED41:
    RECOMMENDATIONS41.append(
        ("Review", "Verification",
         "Complete outstanding reviews, tests and evidence verification.")
    )

RECOMMENDATIONS41.extend([
    (
        "Review", "Data Quality",
        "Document accepted missingness, sample coverage and potential data leakage."
    ),
    (
        "Review", "Business Impact",
        "Measure actual user time savings and outcomes before making ROI claims."
    ),
    (
        "Review", "Deployment",
        "Verify live Flask responses, Streamlit navigation and export functionality."
    ),
    (
        "Review", "Client Sign-Off",
        "Obtain authorized acceptance after all required tests are completed."
    )
])

ACTION_PLAN41 = pd.DataFrame(
    RECOMMENDATIONS41,
    columns=["Priority", "Area", "Recommendation"]
)

SECTION41_STATUS = (
    "SOURCE VALIDATION REQUIRED"
    if TOTAL41 == 0
    else "RECORDED CHECKS PASSED - MANUAL SIGN-OFF REQUIRED"
    if UNRESOLVED41 == 0
    else "PROJECT VERIFICATION INCOMPLETE"
)

SUMMARY41 = pd.DataFrame({
    "Metric": [
        "Recorded Evaluation Checks",
        "Recorded Passes",
        "Unresolved Checks",
        "Recorded Pass Rate (%)",
        "Models With Available Metrics",
        "Metric Types Available"
    ],
    "Value": [
        TOTAL41,
        PASSED41,
        UNRESOLVED41,
        PASS_RATE41,
        MODEL41["Model"].nunique() if not MODEL41.empty else 0,
        len(MODEL_METRICS41)
    ]
})

SUMMARY41.to_csv(
    REPORT41 / "section41_final_summary.csv",
    index=False
)

WATERFALL41.to_csv(
    REPORT41 / "section41_verification_waterfall.csv",
    index=False
)

MODEL41.to_csv(
    REPORT41 / "section41_model_performance.csv",
    index=False
)

IMPACT41.to_csv(
    REPORT41 / "section41_business_impact.csv",
    index=False
)

CONCLUSIONS41.to_csv(
    REPORT41 / "section41_project_conclusions.csv",
    index=False
)

ACTION_PLAN41.to_csv(
    REPORT41 / "section41_recommendations.csv",
    index=False
)

FINAL_REPORT41 = f"""# Final Model Performance Review & Project Conclusions

Generated: {datetime.now().astimezone().isoformat()}

## Evaluation Summary

Status: {SECTION41_STATUS}

Total recorded checks: {TOTAL41}
Recorded passes: {PASSED41}
Unresolved checks: {UNRESOLVED41}
Recorded pass rate: {PASS_RATE41:.2f}%

## Machine-Learning Evaluation

Source: {MODEL_SOURCE41}

Models with available metrics: {MODEL41['Model'].nunique() if not MODEL41.empty else 0}
Available metric types: {', '.join(MODEL_METRICS41) if MODEL_METRICS41 else 'None'}

These values have not been independently verified for test-set
independence, leakage, class balance or prediction validity.

## Business Impact

The project supports historical market analysis, job-family
comparison, geographic exploration and interactive reporting.

Quantified revenue gains, productivity improvements and
return on investment have not been established.

## Main Limitations

Historical postings are not live vacancies.
Growth percentages require meaningful baselines.
The geographic API covers hourly-paid listings.
Unnormalized pay values are not directly comparable.
Recorded validation statuses do not certify production readiness.

## Final Recommendation

Resolve outstanding checks, verify model-evaluation
methodology, test the running application and obtain
authorized client approval before final delivery.
"""

REPORT_TEXT41 = REPORT41 / "section41_final_project_review.md"
REPORT_TEXT41.write_text(
    FINAL_REPORT41, encoding="utf-8"
)

CSS41 = """
<style>
.s41hero{
    background:linear-gradient(135deg,#22223b,#32344f 52%,#4a4e69);
    padding:25px 29px;border-radius:16px;color:white;
    font-family:Arial;margin:10px 0 16px;
}
.s41hero small{
    color:#c9ada7;letter-spacing:1.7px;
    font-size:10px;font-weight:700;
}
.s41hero h2{
    color:white!important;font-size:26px;margin:8px 0;
}
.s41hero p{
    color:#f2e9e4;font-size:12px;margin:0;
}
.s41metrics{
    display:grid;grid-template-columns:repeat(4,1fr);
    gap:10px;margin:14px 0;
}
.s41metric{
    background:white;border:1px solid #e6dfdc;
    border-radius:11px;padding:17px;font-family:Arial;
}
.s41metric span{
    display:block;color:#9a8c98;font-size:10px;
}
.s41metric b{
    display:block;color:#22223b;font-size:24px;margin-top:7px;
}
.s41note{
    background:#f2e9e4;border-left:4px solid #9a8c98;
    padding:14px;border-radius:9px;color:#4a4e69;
    margin:12px 0;font:12px Arial;line-height:1.7;
}
</style>
"""

display(HTML(CSS41))

display(HTML("""
<div class="s41hero">
<small>SECTION 41 · MODEL EVALUATION & BUSINESS IMPACT</small>
<h2>Final Model Review & Project Conclusions</h2>
<p>Model Evidence · Verification · Business Value · Final Recommendations</p>
</div>
"""))

KPI_HTML41 = "".join(
    '<div class="s41metric">'
    f'<span>{escape(str(label))}</span>'
    f'<b>{escape(str(value))}</b>'
    '</div>'
    for label, value in [
        ("TOTAL CHECKS", TOTAL41),
        ("RECORDED PASSES", PASSED41),
        ("UNRESOLVED", UNRESOLVED41),
        ("MODELS FOUND", MODEL41["Model"].nunique() if not MODEL41.empty else 0)
    ]
)

display(HTML(
    '<div class="s41metrics">' + KPI_HTML41 + '</div>'
))

def _note41(message):
    display(HTML(
        '<div class="s41note">' + escape(str(message)) + '</div>'
    ))

def _table41(df, limit=50):
    if df.empty:
        _note41("No verified records available.")
    else:
        display(HTML(
            df.head(limit).to_html(
                index=False, escape=True, border=0
            )
        ))

def _draw41(view):

    if view == "Executive Overview":

        fig41 = go.Figure(
            go.Waterfall(
                name="Evaluation",
                orientation="v",
                measure=WATERFALL41["Type"].tolist(),
                x=WATERFALL41["Stage"].tolist(),
                y=WATERFALL41["Contribution"].tolist(),
                text=[
                    str(abs(int(value)))
                    for value in WATERFALL41["Contribution"]
                ],
                textposition="outside",
                connector=dict(
                    line=dict(color="#9a8c98", width=1)
                ),
                increasing=dict(marker=dict(color="#4a4e69")),
                decreasing=dict(marker=dict(color="#c9ada7")),
                totals=dict(marker=dict(color="#9a8c98"))
            )
        )

        fig41.update_layout(
            title="Verification Waterfall — Total Checks to Recorded Passes",
            template="plotly_white",
            height=480,
            showlegend=False,
            font=dict(family="Arial", size=11, color="#22223b"),
            margin=dict(l=20, r=20, t=65, b=60),
            xaxis_title="Evaluation stage",
            yaxis_title="Number of checks",
            paper_bgcolor="white"
        )

        display(fig41)
        _table41(SUMMARY41)

        _note41(
            f"Insight: {PASSED41} of {TOTAL41} recorded "
            f"evaluation checks passed ({PASS_RATE41:.1f}%)."
        )

        _note41(
            "Interpretation: The waterfall subtracts failed, blocked, "
            "review, untested and unverified checks from the total. "
            "It measures recorded test status, not predictive accuracy."
        )

        _note41(
            "Recommendation: Resolve unresolved evaluation checks "
            "and verify the actual application before final sign-off."
        )

    elif view == "Model Performance":

        if MODEL41.empty:
            _note41(
                "No compatible model-comparison DataFrame or CSV "
                "was found. Model scores have not been invented."
            )

            _note41(
                "To populate this tab, make a DataFrame available "
                "with a Model column and task-appropriate metrics "
                "such as Accuracy, Precision, Recall, F1, ROC_AUC, "
                "MAE, RMSE or R2."
            )

        else:
            _table41(MODEL41, 50)

            _note41("Model metrics source: " + MODEL_SOURCE41)

            _note41(
                "Interpretation: Classification and regression "
                "metrics have different meanings. Model scores "
                "should only be compared for the same prediction "
                "task, dataset and evaluation protocol."
            )

            _note41(
                "Recommendation: Validate held-out test performance, "
                "data leakage, class imbalance and model stability "
                "before recommending deployment."
            )

    elif view == "Business Impact":

        _table41(IMPACT41)

        _note41(
            "Insight: The project provides analytical capabilities "
            "for comparing historical demand, emerging job families "
            "and geographic distribution."
        )

        _note41(
            "Interpretation: These are potential business benefits. "
            "Actual savings, financial returns and decision improvements "
            "require independently measured outcomes."
        )

        _note41(
            "Recommendation: Define business KPIs and collect "
            "baseline and post-deployment measurements."
        )

    elif view == "Evaluation Results":

        if EVALUATION41.empty:
            _note41(
                "The Section 40 evaluation register is unavailable."
            )
        else:
            _table41(EVALUATION41, 100)

            _note41(
                f"Recorded passes: {PASSED41}; "
                f"unresolved items: {UNRESOLVED41}."
            )

    elif view == "Project Conclusions":

        _table41(CONCLUSIONS41)

        _note41(
            "Overall conclusion: The system supports structured "
            "job-market intelligence, but deployment readiness "
            "depends on verified test results, model evaluation "
            "and client acceptance."
        )

    elif view == "Recommendations":

        _table41(ACTION_PLAN41)

        _note41(
            "Prioritize failed and blocked checks, then verify "
            "model evidence, runtime functionality and business impact."
        )

    elif view == "Final Report":

        display(HTML(
            '<pre style="white-space:pre-wrap;'
            'font:12px Arial;color:#22223b;">'
            + escape(FINAL_REPORT41)
            + '</pre>'
        ))

    elif view == "Exports":

        _table41(pd.DataFrame({
            "Report": [
                "Executive Summary",
                "Verification Waterfall",
                "Model Performance",
                "Business Impact",
                "Project Conclusions",
                "Recommendations",
                "Final Project Review"
            ],
            "Filename": [
                "section41_final_summary.csv",
                "section41_verification_waterfall.csv",
                "section41_model_performance.csv",
                "section41_business_impact.csv",
                "section41_project_conclusions.csv",
                "section41_recommendations.csv",
                "section41_final_project_review.md"
            ]
        }))

        _note41("Reports saved to: " + str(REPORT41))

NAV41 = widgets.ToggleButtons(
    options=[
        "Executive Overview",
        "Model Performance",
        "Business Impact",
        "Evaluation Results",
        "Project Conclusions",
        "Recommendations",
        "Final Report",
        "Exports"
    ],
    value="Executive Overview",
    layout=widgets.Layout(width="100%")
)

OUTPUT41 = widgets.Output(
    layout=widgets.Layout(width="100%")
)

def render41(change=None):
    with OUTPUT41:
        clear_output(wait=True)
        _draw41(NAV41.value)

NAV41.observe(render41, names="value")

DASHBOARD41 = widgets.VBox([
    NAV41,
    OUTPUT41
])

display(DASHBOARD41)
render41()
SECTION41_STATUS = SECTION41_STATUS
```

```python
ROOT42 = Path(globals().get("ROOT41", globals().get("ROOT40", Path.cwd()))).resolve()
REPORT42 = ROOT42 / "reports" / "improvement"
REPORT42.mkdir(parents=True, exist_ok=True)
STREAM42 = ROOT42 / "stream.py"
ROADMAP_PATH42 = REPORT42 / "section42_improvement_roadmap.csv"
START42 = pd.Timestamp.today().normalize()

TASKS42 = [
    ("IMP-001", "Validation", "Resolve outstanding validation findings",
     "Critical", 0, 14, "Close verified failures and outstanding review items"),
    ("IMP-002", "Data Quality", "Review missing-value treatment and data integrity",
     "High", 7, 21, "Document field-level null policies and duplicate handling"),
    ("IMP-003", "Data Quality", "Improve data freshness and source monitoring",
     "High", 14, 42, "Track collection dates, coverage and source failures"),
    ("IMP-004", "Model Monitoring", "Establish predictive performance baselines",
     "High", 14, 35, "Validate held-out metrics and reproducible evaluations"),
    ("IMP-005", "Model Monitoring", "Implement data and prediction drift checks",
     "High", 28, 56, "Define reference distributions and alert thresholds"),
    ("IMP-006", "Model Monitoring", "Design a controlled retraining workflow",
     "Medium", 49, 84, "Specify retraining triggers, evaluation and rollback"),
    ("IMP-007", "Flask API", "Improve API reliability and error handling",
     "High", 14, 42, "Review response schemas, error handling and health checks"),
    ("IMP-008", "Flask API", "Add API performance and latency monitoring",
     "Medium", 35, 63, "Measure request errors, throughput and response time"),
    ("IMP-009", "Streamlit", "Improve dashboard navigation and usability",
     "Medium", 21, 49, "Review filters, chart clarity and user workflows"),
    ("IMP-010", "Streamlit", "Optimize dashboard loading and caching",
     "Medium", 42, 70, "Profile data requests and improve response efficiency"),
    ("IMP-011", "Security", "Review secrets, access controls and data exposure",
     "High", 7, 28, "Verify secret management and least-privilege access"),
    ("IMP-012", "Testing", "Expand automated integration and regression tests",
     "High", 28, 63, "Test API contracts, calculations and dashboard modules"),
    ("IMP-013", "Business Intelligence", "Improve geographic and demand insights",
     "Medium", 56, 91, "Add validated filters and clearer interpretation"),
    ("IMP-014", "Business Intelligence", "Design decision-support evaluation",
     "Medium", 70, 105, "Measure usefulness and user decision outcomes"),
    ("IMP-015", "Deployment", "Prepare deployment and recovery procedures",
     "High", 35, 70, "Document environment, health checks and recovery"),
    ("IMP-016", "Documentation", "Finalize maintenance and ownership guidance",
     "Medium", 84, 112, "Document maintenance, responsibilities and future work")
]

ROADMAP42 = pd.DataFrame(
    TASKS42,
    columns=[
        "Task_ID", "Category", "Task", "Priority",
        "Start_Day", "End_Day", "Success_Criteria"
    ]
)

ROADMAP42["Start"] = (
    START42 + pd.to_timedelta(ROADMAP42["Start_Day"], unit="D")
)

ROADMAP42["Finish"] = (
    START42 + pd.to_timedelta(ROADMAP42["End_Day"], unit="D")
)

ROADMAP42["Status"] = "PLANNED"
ROADMAP42["Progress_%"] = 0
ROADMAP42["Evidence"] = ""
ROADMAP42["Updated_At"] = ""

if ROADMAP_PATH42.is_file():
    try:
        previous42 = pd.read_csv(ROADMAP_PATH42)

        if "Task_ID" in previous42.columns:
            previous42 = (
                previous42
                .drop_duplicates("Task_ID", keep="last")
                .set_index("Task_ID")
            )

            for field42 in [
                "Status", "Progress_%",
                "Evidence", "Updated_At"
            ]:
                if field42 in previous42.columns:
                    ROADMAP42[field42] = (
                        ROADMAP42["Task_ID"]
                        .map(previous42[field42])
                        .fillna(ROADMAP42[field42])
                    )

            for field42 in ["Start", "Finish"]:
                if field42 in previous42.columns:
                    restored42 = pd.to_datetime(
                        ROADMAP42["Task_ID"].map(previous42[field42]),
                        errors="coerce"
                    )
                    ROADMAP42[field42] = restored42.fillna(
                        ROADMAP42[field42]
                    )

    except (OSError, ValueError, pd.errors.ParserError):
        print("Previous roadmap could not be read; proposed schedule initialized.")

VALID_STATUSES42 = [
    "PLANNED", "IN PROGRESS", "BLOCKED", "COMPLETED"
]

ROADMAP42["Status"] = (
    ROADMAP42["Status"]
    .astype(str)
    .str.upper()
    .str.strip()
)

ROADMAP42.loc[
    ~ROADMAP42["Status"].isin(VALID_STATUSES42),
    "Status"
] = "PLANNED"

ROADMAP42["Progress_%"] = (
    pd.to_numeric(ROADMAP42["Progress_%"], errors="coerce")
    .fillna(0)
    .clip(0, 100)
    .astype(int)
)

PRIORITY_ORDER42 = {
    "Critical": 0,
    "High": 1,
    "Medium": 2,
    "Low": 3
}

STATUS_COLORS42 = {
    "PLANNED": "#9a8c98",
    "IN PROGRESS": "#4a4e69",
    "BLOCKED": "#c9ada7",
    "COMPLETED": "#22223b"
}

EVALUATION42 = globals().get("EVALUATION41", pd.DataFrame())

if isinstance(EVALUATION42, pd.DataFrame) and not EVALUATION42.empty:
    if "Status" in EVALUATION42.columns:
        UNRESOLVED_SOURCE42 = int(
            EVALUATION42["Status"]
            .astype(str)
            .str.upper()
            .ne("PASS")
            .sum()
        )
    else:
        UNRESOLVED_SOURCE42 = None
else:
    UNRESOLVED_SOURCE42 = None

def _summary42():
    statuses = (
        ROADMAP42["Status"]
        .value_counts()
        .reindex(VALID_STATUSES42, fill_value=0)
    )

    return {
        "Total": len(ROADMAP42),
        "Planned": int(statuses["PLANNED"]),
        "In Progress": int(statuses["IN PROGRESS"]),
        "Blocked": int(statuses["BLOCKED"]),
        "Completed": int(statuses["COMPLETED"]),
        "Average Progress": round(
            ROADMAP42["Progress_%"].mean(), 1
        )
    }

def _exports42():
    ROADMAP42.to_csv(
        ROADMAP_PATH42,
        index=False,
        encoding="utf-8-sig"
    )

    summary42 = _summary42()

    pd.DataFrame({
        "Metric": list(summary42),
        "Value": list(summary42.values())
    }).to_csv(
        REPORT42 / "section42_roadmap_summary.csv",
        index=False
    )

    category42 = (
        ROADMAP42.groupby("Category", as_index=False)
        .agg(
            Tasks=("Task_ID", "count"),
            Average_Progress=("Progress_%", "mean"),
            Completed=("Status", lambda x: int((x == "COMPLETED").sum()))
        )
    )

    category42["Average_Progress"] = (
        category42["Average_Progress"].round(1)
    )

    category42.to_csv(
        REPORT42 / "section42_category_progress.csv",
        index=False
    )

    outstanding42 = ROADMAP42[
        ROADMAP42["Status"] != "COMPLETED"
    ].copy()

    outstanding42["Priority_Order"] = (
        outstanding42["Priority"].map(PRIORITY_ORDER42)
    )

    outstanding42 = (
        outstanding42
        .sort_values(["Priority_Order", "Start"])
        .drop(columns="Priority_Order")
    )

    outstanding42.to_csv(
        REPORT42 / "section42_outstanding_actions.csv",
        index=False
    )

    monitoring42 = pd.DataFrame([
        ("Data Freshness", "Latest successful ingestion timestamp",
         "Set threshold from approved data refresh requirements"),
        ("Data Quality", "Missingness and schema-change rate",
         "Establish accepted baseline and alert policy"),
        ("Model Performance", "Held-out task-specific metrics",
         "Select acceptable performance and degradation thresholds"),
        ("Data Drift", "Feature-distribution change",
         "Choose reference population and appropriate tests"),
        ("Prediction Drift", "Prediction-score distribution change",
         "Review output distribution and relevant business context"),
        ("API Reliability", "Error rate and request latency",
         "Define an agreed service-level objective"),
        ("Dashboard Usability", "Loading time and task completion",
         "Measure performance and feedback from actual users"),
        ("Business Outcomes", "Validated decision usefulness",
         "Collect baseline and post-deployment evidence")
    ], columns=["Monitoring_Area", "Metric", "Implementation_Guidance"])

    monitoring42.to_csv(
        REPORT42 / "section42_monitoring_framework.csv",
        index=False
    )

    return summary42, category42, outstanding42, monitoring42

SUMMARY42, CATEGORY42, OUTSTANDING42, MONITORING42 = _exports42()

CSS42 = """
<style>
.s42hero{
    background:linear-gradient(135deg,#22223b,#32344f 52%,#4a4e69);
    padding:25px 29px;border-radius:16px;color:white;
    font-family:Arial;margin:10px 0 16px;
}
.s42hero small{
    color:#c9ada7;letter-spacing:1.7px;
    font-size:10px;font-weight:700;
}
.s42hero h2{
    color:white!important;font-size:26px;margin:8px 0;
}
.s42hero p{
    color:#f2e9e4;font-size:12px;margin:0;
}
.s42metrics{
    display:grid;grid-template-columns:repeat(4,1fr);
    gap:10px;margin:15px 0;
}
.s42metric{
    background:white;border:1px solid #e6dfdc;
    border-radius:11px;padding:17px;font-family:Arial;
}
.s42metric span{
    display:block;color:#9a8c98;font-size:10px;
}
.s42metric b{
    display:block;color:#22223b;font-size:24px;margin-top:7px;
}
.s42note{
    background:#f2e9e4;border-left:4px solid #9a8c98;
    padding:14px;border-radius:9px;color:#4a4e69;
    margin:12px 0;font:12px Arial;line-height:1.7;
}
</style>
"""

display(HTML(CSS42))

display(HTML("""
<div class="s42hero">
<small>SECTION 42 · CONTINUOUS IMPROVEMENT ROADMAP</small>
<h2>Future Enhancements & Model Monitoring</h2>
<p>Development Timeline · Model Reliability · API Performance · Strategic Improvements</p>
</div>
"""))

METRIC_OUTPUT42 = widgets.Output()

def _render_metrics42():
    summary = _summary42()

    values = [
        ("TOTAL TASKS", summary["Total"]),
        ("IN PROGRESS", summary["In Progress"]),
        ("COMPLETED", summary["Completed"]),
        ("AVG. PROGRESS", f'{summary["Average Progress"]:.1f}%')
    ]

    html42 = "".join(
        '<div class="s42metric">'
        f'<span>{escape(str(label))}</span>'
        f'<b>{escape(str(value))}</b>'
        '</div>'
        for label, value in values
    )

    with METRIC_OUTPUT42:
        clear_output(wait=True)
        display(HTML(
            '<div class="s42metrics">'
            + html42
            + '</div>'
        ))

_render_metrics42()
display(METRIC_OUTPUT42)

def _note42(message):
    display(HTML(
        '<div class="s42note">'
        + escape(str(message))
        + '</div>'
    ))

def _table42(df, n=50):
    if df.empty:
        _note42("No records available.")
    else:
        display(HTML(
            df.head(n).to_html(
                index=False, escape=True, border=0
            )
        ))

TASK_SELECT42 = widgets.Dropdown(
    options=[
        (f"{r.Task_ID} | {r.Task}", r.Task_ID)
        for r in ROADMAP42.itertuples()
    ],
    description="Task:",
    layout=widgets.Layout(width="95%")
)

STATUS_SELECT42 = widgets.Dropdown(
    options=VALID_STATUSES42,
    description="Status:"
)

PROGRESS_SLIDER42 = widgets.IntSlider(
    value=0,
    min=0,
    max=100,
    step=5,
    description="Progress:",
    continuous_update=False,
    layout=widgets.Layout(width="95%")
)

EVIDENCE_TEXT42 = widgets.Textarea(
    description="Evidence:",
    placeholder="Completed work, test results or evidence reference",
    layout=widgets.Layout(width="95%", height="75px")
)

SAVE_BUTTON42 = widgets.Button(
    description="Save Roadmap Progress",
    button_style="primary",
    icon="save"
)

SAVE_OUTPUT42 = widgets.Output()

def _load_task42(change=None):
    row = ROADMAP42[
        ROADMAP42["Task_ID"] == TASK_SELECT42.value
    ]

    if row.empty:
        return

    item = row.iloc[0]

    STATUS_SELECT42.value = item["Status"]
    PROGRESS_SLIDER42.value = int(item["Progress_%"])
    EVIDENCE_TEXT42.value = (
        str(item["Evidence"])
        if pd.notna(item["Evidence"]) else ""
    )

TASK_SELECT42.observe(_load_task42, names="value")
_load_task42()

def _save_task42(button):
    selected = TASK_SELECT42.value
    matches = ROADMAP42.index[
        ROADMAP42["Task_ID"] == selected
    ]

    if not len(matches):
        return

    index = matches[0]
    status = STATUS_SELECT42.value
    progress = PROGRESS_SLIDER42.value
    evidence = EVIDENCE_TEXT42.value.strip()

    with SAVE_OUTPUT42:
        clear_output(wait=True)

        if status == "COMPLETED" and (
            progress != 100 or not evidence
        ):
            print(
                "Completed tasks require 100% progress "
                "and a supporting evidence reference."
            )
            return

        if status == "PLANNED" and progress > 0:
            print(
                "A planned task cannot have recorded progress. "
                "Choose IN PROGRESS instead."
            )
            return

        ROADMAP42.at[index, "Status"] = status
        ROADMAP42.at[index, "Progress_%"] = progress
        ROADMAP42.at[index, "Evidence"] = evidence
        ROADMAP42.at[index, "Updated_At"] = (
            datetime.now().astimezone().isoformat(
                timespec="seconds"
            )
        )

        globals()["SUMMARY42"], globals()["CATEGORY42"], globals()["OUTSTANDING42"], globals()["MONITORING42"] = _exports42()

        _render_metrics42()
        print(f"Saved {selected}: {status}, {progress}%")

    _render42()

SAVE_BUTTON42.on_click(_save_task42)

def _draw42(view):

    if view == "Executive Overview":

        timeline = ROADMAP42.copy()

        timeline["Category_Task"] = (
            timeline["Category"] + " · " + timeline["Task_ID"]
        )

        timeline = timeline.sort_values(
            ["Start", "Priority_Order"]
        ) if "Priority_Order" in timeline.columns else (
            timeline.sort_values(["Start", "Category"])
        )

        fig = px.timeline(
            timeline,
            x_start="Start",
            x_end="Finish",
            y="Category_Task",
            color="Status",
            color_discrete_map=STATUS_COLORS42,
            hover_name="Task",
            hover_data={
                "Priority": True,
                "Progress_%": True,
                "Start": "|%d %b %Y",
                "Finish": "|%d %b %Y",
                "Category_Task": False
            },
            title="Proposed Continuous Improvement Roadmap"
        )

        fig.update_yaxes(autorange="reversed")

        fig.update_layout(
            template="plotly_white",
            height=max(520, len(timeline) * 36),
            font=dict(
                family="Arial",
                size=11,
                color="#22223b"
            ),
            margin=dict(l=20, r=20, t=65, b=35),
            paper_bgcolor="white",
            legend_title_text="Task Status"
        )

        display(fig)

        _table42(pd.DataFrame({
            "Metric": list(_summary42()),
            "Value": list(_summary42().values())
        }))

        _note42(
            "Insight: The timeline maps proposed enhancement "
            "tasks across data quality, model monitoring, API "
            "reliability, security and dashboard improvements."
        )

        _note42(
            "Interpretation: Dates are illustrative planning targets "
            "generated from the notebook's first run. They are not "
            "approved deadlines or evidence that work has started."
        )

        _note42(
            "Recommendation: Resolve outstanding validation "
            "and security concerns before lower-priority enhancements."
        )

    elif view == "Roadmap Tasks":

        _table42(
            ROADMAP42.sort_values(
                ["Start", "Priority"]
            ),
            100
        )

        _note42(
            "Tasks are proposed improvements. Progress should be "
            "updated only when the underlying work has actually "
            "been performed."
        )

    elif view == "Update Progress":

        display(widgets.VBox([
            TASK_SELECT42,
            STATUS_SELECT42,
            PROGRESS_SLIDER42,
            EVIDENCE_TEXT42,
            SAVE_BUTTON42,
            SAVE_OUTPUT42
        ]))

        _note42(
            "Changes are saved to the roadmap CSV and retained "
            "when Section 42 is rerun."
        )

    elif view == "Model Monitoring":

        model_monitoring = MONITORING42[
            MONITORING42["Monitoring_Area"].isin([
                "Model Performance",
                "Data Drift",
                "Prediction Drift",
                "Data Quality"
            ])
        ]

        _table42(model_monitoring)

        _note42(
            "Insight: Model reliability depends on validated "
            "performance baselines, representative data and "
            "monitoring of distribution changes."
        )

        _note42(
            "Recommendation: Establish baseline metrics and "
            "approved drift thresholds before implementing "
            "automatic monitoring or retraining triggers."
        )

    elif view == "API & Dashboard":

        filtered = ROADMAP42[
            ROADMAP42["Category"].isin([
                "Flask API", "Streamlit", "Deployment"
            ])
        ]

        _table42(filtered)

        _note42(
            "Recommendation: Profile response times, test "
            "failure handling, review caching and verify "
            "that existing dashboard pages remain functional."
        )

    elif view == "Priority Actions":

        high = OUTSTANDING42[
            OUTSTANDING42["Priority"].isin([
                "Critical", "High"
            ])
        ]

        _table42(high)

        if UNRESOLVED_SOURCE42 is not None:
            _note42(
                f"Previous evaluation records included "
                f"{UNRESOLVED_SOURCE42} unresolved checks. "
                "Review the latest validation reports to confirm "
                "which findings still require action."
            )

        _note42(
            "Recommendation: Address blockers, security risks "
            "and validation issues before optional enhancements."
        )

    elif view == "Business Enhancements":

        filtered = ROADMAP42[
            ROADMAP42["Category"] == "Business Intelligence"
        ]

        _table42(filtered)

        _note42(
            "Potential enhancements include richer market "
            "comparison and decision-support evaluation. "
            "Business value must be measured, not assumed."
        )

    elif view == "Reports":

        _table42(pd.DataFrame({
            "Report": [
                "Improvement Roadmap",
                "Roadmap Summary",
                "Category Progress",
                "Outstanding Actions",
                "Monitoring Framework"
            ],
            "Filename": [
                "section42_improvement_roadmap.csv",
                "section42_roadmap_summary.csv",
                "section42_category_progress.csv",
                "section42_outstanding_actions.csv",
                "section42_monitoring_framework.csv"
            ]
        }))

        _note42(
            "Reports saved to: " + str(REPORT42)
        )

NAV42 = widgets.ToggleButtons(
    options=[
        "Executive Overview",
        "Roadmap Tasks",
        "Update Progress",
        "Model Monitoring",
        "API & Dashboard",
        "Priority Actions",
        "Business Enhancements",
        "Reports"
    ],
    value="Executive Overview",
    layout=widgets.Layout(width="100%")
)

OUTPUT42 = widgets.Output(
    layout=widgets.Layout(width="100%")
)

def _render42(change=None):
    with OUTPUT42:
        clear_output(wait=True)
        _draw42(NAV42.value)

NAV42.observe(_render42, names="value")

DASHBOARD42 = widgets.VBox([
    NAV42,
    OUTPUT42
])

display(DASHBOARD42)
_render42()
SECTION42_STATUS = "PROPOSED ROADMAP CREATED"
```

```python
ROOT43 = Path(globals().get("ROOT42", globals().get("ROOT41", Path.cwd()))).resolve()
REPORT43 = ROOT43 / "reports" / "monitoring"
REPORT43.mkdir(parents=True, exist_ok=True)
STREAM43 = ROOT43 / "stream.py"
DATA43 = pd.DataFrame()
SOURCE43 = "Unavailable"

for name43 in ("DATA34", "DATA33", "df_features", "df_final", "jobs22"):
    candidate43 = globals().get(name43)
    if isinstance(candidate43, pd.DataFrame) and not candidate43.empty:
        DATA43 = candidate43.copy()
        SOURCE43 = name43
        break

DATE_CANDIDATES43 = [
    "published_date", "published_day",
    "date", "posting_date", "created_at"
]

date_lookup43 = {
    str(c).casefold(): c for c in DATA43.columns
}

DATE_COL43 = next(
    (
        date_lookup43[name.casefold()]
        for name in DATE_CANDIDATES43
        if name.casefold() in date_lookup43
    ),
    None
)

if DATE_COL43 is not None:
    DATES43 = pd.to_datetime(
        DATA43[DATE_COL43],
        errors="coerce",
        utc=True
    ).dt.floor("D")
else:
    DATES43 = pd.Series(
        pd.NaT,
        index=DATA43.index,
        dtype="datetime64[ns, UTC]"
    )

WINDOW_VALID43 = False
BASELINE43 = pd.DataFrame()
CURRENT43 = pd.DataFrame()
WINDOW_LABEL43 = "Unavailable"

if DATES43.notna().any():
    END43 = DATES43.max()

    BASE_START43 = END43 - pd.Timedelta(days=55)
    BASE_END43 = END43 - pd.Timedelta(days=28)
    CURRENT_START43 = END43 - pd.Timedelta(days=27)

    baseline_mask43 = (
        (DATES43 >= BASE_START43)
        & (DATES43 <= BASE_END43)
    )

    current_mask43 = (
        (DATES43 >= CURRENT_START43)
        & (DATES43 <= END43)
    )

    BASELINE43 = DATA43.loc[baseline_mask43].copy()
    CURRENT43 = DATA43.loc[current_mask43].copy()

    WINDOW_VALID43 = (
        len(BASELINE43) >= 30
        and len(CURRENT43) >= 30
    )

    WINDOW_LABEL43 = (
        f"Baseline: {BASE_START43.date()} to {BASE_END43.date()} | "
        f"Comparison: {CURRENT_START43.date()} to {END43.date()}"
    )

PSI_RESULTS43 = []
QUALITY_RESULTS43 = []
NUMERIC_FEATURES43 = []
CATEGORY_FEATURES43 = []

EXCLUDED43 = {
    str(DATE_COL43).casefold() if DATE_COL43 is not None else "",
    "id", "job_id", "record_id", "index",
    "unnamed: 0"
}

def _risk43(value):
    if pd.isna(value):
        return "INSUFFICIENT DATA"
    if value < 0.10:
        return "LOW"
    if value < 0.25:
        return "MODERATE"
    return "HIGH"

def _psi_numeric43(reference, comparison, bins=10):
    reference = pd.to_numeric(
        reference, errors="coerce"
    ).dropna().to_numpy(dtype=float)

    comparison = pd.to_numeric(
        comparison, errors="coerce"
    ).dropna().to_numpy(dtype=float)

    reference = reference[np.isfinite(reference)]
    comparison = comparison[np.isfinite(comparison)]

    if len(reference) < 30 or len(comparison) < 30:
        return float("nan")

    edges = np.unique(
        np.quantile(
            reference,
            np.linspace(0, 1, bins + 1)
        )
    )

    if len(edges) < 3:
        return float("nan")

    edges[0] = -np.inf
    edges[-1] = np.inf

    ref_counts = np.histogram(
        reference, bins=edges
    )[0].astype(float)

    cur_counts = np.histogram(
        comparison, bins=edges
    )[0].astype(float)

    epsilon = 1e-6

    ref_rate = np.clip(
        ref_counts / ref_counts.sum(),
        epsilon, None
    )

    cur_rate = np.clip(
        cur_counts / cur_counts.sum(),
        epsilon, None
    )

    ref_rate = ref_rate / ref_rate.sum()
    cur_rate = cur_rate / cur_rate.sum()

    return float(
        np.sum(
            (cur_rate - ref_rate)
            * np.log(cur_rate / ref_rate)
        )
    )

def _psi_category43(reference, comparison, top=10):
    ref = reference.astype("object").fillna("<Missing>").astype(str)
    cur = comparison.astype("object").fillna("<Missing>").astype(str)

    if len(ref) < 30 or len(cur) < 30:
        return float("nan"), float("nan")

    major = ref.value_counts().head(top).index.tolist()

    ref_grouped = ref.where(ref.isin(major), "<Other>")
    cur_grouped = cur.where(cur.isin(major), "<Other>")

    categories = sorted(
        set(ref_grouped.unique())
        | set(cur_grouped.unique())
    )

    a = ref_grouped.value_counts().reindex(
        categories, fill_value=0
    ).to_numpy(dtype=float)

    b = cur_grouped.value_counts().reindex(
        categories, fill_value=0
    ).to_numpy(dtype=float)

    epsilon = 1e-6

    p = np.clip(a / a.sum(), epsilon, None)
    q = np.clip(b / b.sum(), epsilon, None)

    p = p / p.sum()
    q = q / q.sum()

    psi = float(
        np.sum((q - p) * np.log(q / p))
    )

    unseen = (
        ~cur.isin(ref.unique())
    ).mean() * 100

    return psi, float(unseen)

if WINDOW_VALID43:

    for col43 in DATA43.columns:
        if str(col43).casefold() in EXCLUDED43:
            continue

        base43 = BASELINE43[col43]
        recent43 = CURRENT43[col43]

        base_missing43 = float(
            base43.isna().mean() * 100
        )

        recent_missing43 = float(
            recent43.isna().mean() * 100
        )

        missing_shift43 = (
            recent_missing43 - base_missing43
        )

        QUALITY_RESULTS43.append({
            "Feature": str(col43),
            "Baseline_Missing_%": round(base_missing43, 2),
            "Recent_Missing_%": round(recent_missing43, 2),
            "Missing_Shift_pp": round(missing_shift43, 2),
            "Abs_Shift_pp": round(abs(missing_shift43), 2)
        })

        if pd.api.types.is_numeric_dtype(DATA43[col43]):

            ref_numeric43 = pd.to_numeric(
                base43, errors="coerce"
            )
            cur_numeric43 = pd.to_numeric(
                recent43, errors="coerce"
            )

            if (
                ref_numeric43.notna().sum() < 30
                or cur_numeric43.notna().sum() < 30
                or ref_numeric43.nunique() < 3
            ):
                continue

            psi43 = _psi_numeric43(
                ref_numeric43, cur_numeric43
            )

            if pd.notna(psi43):
                NUMERIC_FEATURES43.append(col43)

                PSI_RESULTS43.append({
                    "Feature": str(col43),
                    "Feature_Type": "Numerical",
                    "PSI": round(psi43, 4),
                    "Drift_Level": _risk43(psi43),
                    "Baseline_Count": int(ref_numeric43.notna().sum()),
                    "Recent_Count": int(cur_numeric43.notna().sum()),
                    "Unseen_Category_%": None
                })

        elif (
            pd.api.types.is_object_dtype(DATA43[col43])
            or isinstance(DATA43[col43].dtype, pd.CategoricalDtype)
            or pd.api.types.is_string_dtype(DATA43[col43])
        ):

            if (
                base43.nunique(dropna=True) < 2
                or base43.nunique(dropna=True) > 100
            ):
                continue

            psi43, unseen43 = _psi_category43(
                base43, recent43
            )

            if pd.notna(psi43):
                CATEGORY_FEATURES43.append(col43)

                PSI_RESULTS43.append({
                    "Feature": str(col43),
                    "Feature_Type": "Categorical",
                    "PSI": round(psi43, 4),
                    "Drift_Level": _risk43(psi43),
                    "Baseline_Count": len(base43),
                    "Recent_Count": len(recent43),
                    "Unseen_Category_%": round(unseen43, 2)
                })

PSI43 = pd.DataFrame(
    PSI_RESULTS43,
    columns=[
        "Feature", "Feature_Type", "PSI",
        "Drift_Level", "Baseline_Count",
        "Recent_Count", "Unseen_Category_%"
    ]
)

QUALITY43 = pd.DataFrame(
    QUALITY_RESULTS43,
    columns=[
        "Feature", "Baseline_Missing_%",
        "Recent_Missing_%", "Missing_Shift_pp",
        "Abs_Shift_pp"
    ]
)

if not PSI43.empty:
    PSI43 = PSI43.sort_values(
        "PSI", ascending=False
    ).reset_index(drop=True)

if not QUALITY43.empty:
    QUALITY43 = QUALITY43.sort_values(
        "Abs_Shift_pp", ascending=False
    ).reset_index(drop=True)

HIGH43 = int(
    (PSI43["Drift_Level"] == "HIGH").sum()
)

MODERATE43 = int(
    (PSI43["Drift_Level"] == "MODERATE").sum()
)

LOW43 = int(
    (PSI43["Drift_Level"] == "LOW").sum()
)

TOTAL_FEATURES43 = len(PSI43)

MONITORING_STATUS43 = (
    "INSUFFICIENT HISTORICAL DATA"
    if not WINDOW_VALID43
    else "NO ELIGIBLE FEATURES"
    if PSI43.empty
    else "DISTRIBUTION SHIFT FLAGGED"
    if HIGH43 > 0
    else "MODERATE SHIFT FLAGGED"
    if MODERATE43 > 0
    else "LOW PSI IN ELIGIBLE FEATURES"
)

SUMMARY43 = pd.DataFrame({
    "Metric": [
        "Historical Data Records",
        "Baseline Records",
        "Comparison Records",
        "Evaluated Features",
        "Low PSI Features",
        "Moderate PSI Features",
        "High PSI Features"
    ],
    "Value": [
        len(DATA43),
        len(BASELINE43),
        len(CURRENT43),
        TOTAL_FEATURES43,
        LOW43,
        MODERATE43,
        HIGH43
    ]
})

RECOMMENDATIONS43 = []

if not WINDOW_VALID43:
    RECOMMENDATIONS43.append(
        ("High", "Historical Coverage",
         "Provide two comparable 28-day observation windows with at least 30 records each.")
    )

if HIGH43:
    RECOMMENDATIONS43.append(
        ("High", "Distribution Shift",
         "Investigate high-PSI features, source changes, seasonality and possible model effects.")
    )

if MODERATE43:
    RECOMMENDATIONS43.append(
        ("Review", "Distribution Shift",
         "Review moderate feature shifts and compare against additional historical windows.")
    )

if not QUALITY43.empty:
    flagged43 = QUALITY43[
        QUALITY43["Abs_Shift_pp"] >= 5
    ]

    if not flagged43.empty:
        RECOMMENDATIONS43.append(
            ("Review", "Missingness",
             f"Investigate missing-value changes of at least 5 percentage points in {len(flagged43)} features.")
        )

RECOMMENDATIONS43.extend([
    (
        "Review", "Model Impact",
        "Use held-out outcomes and model performance metrics to determine whether feature shifts affect predictions."
    ),
    (
        "Review", "Monitoring Baseline",
        "Review reference-window suitability, data volume and approved alert thresholds."
    ),
    (
        "Review", "Retraining",
        "Do not retrain models automatically from PSI alerts alone; validate performance degradation first."
    )
])

ACTION_PLAN43 = pd.DataFrame(
    RECOMMENDATIONS43,
    columns=["Priority", "Area", "Recommendation"]
)

BASELINE_METADATA43 = {
    "source_dataframe": SOURCE43,
    "generated_at": pd.Timestamp.now(tz="UTC").isoformat(),
    "date_column": str(DATE_COL43) if DATE_COL43 is not None else None,
    "historical_window": WINDOW_LABEL43,
    "baseline_records": len(BASELINE43),
    "comparison_records": len(CURRENT43),
    "minimum_records_per_window": 30,
    "psi_screening_thresholds": {
        "low": "PSI < 0.10",
        "moderate": "0.10 <= PSI < 0.25",
        "high": "PSI >= 0.25"
    },
    "evaluated_features": TOTAL_FEATURES43,
    "status": MONITORING_STATUS43,
    "limitations": [
        "Historical comparison only; not live monitoring",
        "PSI thresholds are screening heuristics, not universal statistical significance tests",
        "PSI does not establish predictive performance degradation",
        "Numerical PSI excludes missing values; missingness is analysed separately",
        "Categorical PSI groups infrequent reference categories into Other",
        "Small samples and seasonal changes can affect results"
    ]
}

PSI43.to_csv(
    REPORT43 / "section43_feature_drift.csv",
    index=False
)

QUALITY43.to_csv(
    REPORT43 / "section43_missingness_shift.csv",
    index=False
)

SUMMARY43.to_csv(
    REPORT43 / "section43_monitoring_summary.csv",
    index=False
)

ACTION_PLAN43.to_csv(
    REPORT43 / "section43_monitoring_recommendations.csv",
    index=False
)

(
    REPORT43 / "section43_baseline_metadata.json"
).write_text(
    json.dumps(BASELINE_METADATA43, indent=2),
    encoding="utf-8"
)

CSS43 = """
<style>
.s43hero{
    background:linear-gradient(135deg,#22223b,#32344f 52%,#4a4e69);
    padding:25px 29px;border-radius:16px;color:white;
    font-family:Arial;margin:10px 0 16px;
}
.s43hero small{
    color:#c9ada7;letter-spacing:1.7px;
    font-size:10px;font-weight:700;
}
.s43hero h2{
    color:white!important;font-size:26px;margin:8px 0;
}
.s43hero p{
    color:#f2e9e4;font-size:12px;margin:0;
}
.s43metrics{
    display:grid;grid-template-columns:repeat(4,1fr);
    gap:10px;margin:15px 0;
}
.s43metric{
    background:white;border:1px solid #e6dfdc;
    border-radius:11px;padding:17px;font-family:Arial;
}
.s43metric span{
    display:block;color:#9a8c98;font-size:10px;
}
.s43metric b{
    display:block;color:#22223b;font-size:24px;margin-top:7px;
}
.s43note{
    background:#f2e9e4;border-left:4px solid #9a8c98;
    padding:14px;border-radius:9px;color:#4a4e69;
    margin:12px 0;font:12px Arial;line-height:1.7;
}
</style>
"""

display(HTML(CSS43))

display(HTML("""
<div class="s43hero">
<small>SECTION 43 · HISTORICAL DATA DRIFT MONITORING</small>
<h2>Feature Stability & Data Quality Surveillance</h2>
<p>Historical Baselines · PSI · Missingness · Distribution Comparison</p>
</div>
"""))

METRIC_HTML43 = "".join(
    '<div class="s43metric">'
    f'<span>{escape(str(label))}</span>'
    f'<b>{escape(str(value))}</b>'
    '</div>'
    for label, value in [
        ("BASELINE RECORDS", f"{len(BASELINE43):,}"),
        ("RECENT RECORDS", f"{len(CURRENT43):,}"),
        ("FEATURES TESTED", TOTAL_FEATURES43),
        ("HIGH PSI", HIGH43)
    ]
)

display(HTML(
    '<div class="s43metrics">'
    + METRIC_HTML43
    + '</div>'
))

def _note43(message):
    display(HTML(
        '<div class="s43note">'
        + escape(str(message))
        + '</div>'
    ))

def _table43(frame, limit=40):
    if frame.empty:
        _note43("No eligible records available.")
    else:
        display(HTML(
            frame.head(limit).to_html(
                index=False, escape=True, border=0
            )
        ))

FEATURE_SELECT43 = widgets.Dropdown(
    options=[
        (str(col), col)
        for col in NUMERIC_FEATURES43
    ],
    description="Feature:",
    layout=widgets.Layout(width="90%"),
    disabled=not bool(NUMERIC_FEATURES43)
)

def _violin43(feature):
    baseline_values = pd.to_numeric(
        BASELINE43[feature], errors="coerce"
    ).replace([np.inf, -np.inf], np.nan).dropna()

    comparison_values = pd.to_numeric(
        CURRENT43[feature], errors="coerce"
    ).replace([np.inf, -np.inf], np.nan).dropna()

    if baseline_values.empty or comparison_values.empty:
        _note43("No comparable numeric values available.")
        return

    baseline_sample = baseline_values.sample(
        n=min(len(baseline_values), 2000),
        random_state=43
    )

    comparison_sample = comparison_values.sample(
        n=min(len(comparison_values), 2000),
        random_state=43
    )

    plot43 = pd.DataFrame({
        "Value": pd.concat(
            [baseline_sample, comparison_sample],
            ignore_index=True
        ),
        "Period": (
            ["Baseline"] * len(baseline_sample)
            + ["Recent"] * len(comparison_sample)
        )
    })

    fig43 = px.violin(
        plot43,
        x="Period",
        y="Value",
        color="Period",
        box=True,
        points=False,
        color_discrete_map={
            "Baseline": "#4a4e69",
            "Recent": "#c9ada7"
        },
        title=f"Feature Distribution Comparison — {feature}"
    )

    fig43.update_layout(
        template="plotly_white",
        height=480,
        font=dict(
            family="Arial",
            size=12,
            color="#22223b"
        ),
        showlegend=False,
        margin=dict(l=25, r=20, t=65, b=30),
        paper_bgcolor="white"
    )

    display(fig43)

def _draw43(view):

    if view == "Executive Overview":

        if not WINDOW_VALID43:
            _note43(
                "Historical drift screening could not proceed. "
                "At least 30 records are required in each "
                "of two consecutive 28-day periods."
            )
            _note43("Data source: " + SOURCE43)
        elif not NUMERIC_FEATURES43:
            _note43(
                "No eligible numerical features were found "
                "for violin-plot comparison."
            )
        else:
            display(FEATURE_SELECT43)
            _violin43(FEATURE_SELECT43.value)

        _table43(SUMMARY43)

        _note43("Observation windows: " + WINDOW_LABEL43)

        _note43(
            "Insight: Violin plots compare the shapes, spread "
            "and central tendencies of numeric feature distributions."
        )

        _note43(
            "Interpretation: Differences between historical "
            "periods indicate a possible distribution shift. "
            "They do not establish degraded model accuracy."
        )

        _note43(
            "Recommendation: Investigate material shifts "
            "with data-source checks, additional periods "
            "and model-performance evidence."
        )

    elif view == "Feature Drift":

        _table43(PSI43, 100)

        _note43(
            f"High PSI features: {HIGH43}; "
            f"moderate PSI features: {MODERATE43}; "
            f"low PSI features: {LOW43}."
        )

        _note43(
            "PSI thresholds of 0.10 and 0.25 are "
            "screening conventions. They should be "
            "calibrated to the project and feature distributions."
        )

    elif view == "Data Quality":

        _table43(QUALITY43, 100)

        if not QUALITY43.empty:
            leader43 = QUALITY43.iloc[0]

            _note43(
                f"Largest observed missingness shift: "
                f"{leader43['Feature']} "
                f"({leader43['Missing_Shift_pp']:+.2f} "
                "percentage points)."
            )

        _note43(
            "Recommendation: Investigate changes in missingness "
            "before applying imputation or dropping records. "
            "Some missing fields may be legitimate."
        )

    elif view == "Categorical Drift":

        category_results43 = PSI43[
            PSI43["Feature_Type"] == "Categorical"
        ]

        _table43(category_results43)

        _note43(
            "Categorical PSI groups infrequent reference "
            "categories into an Other category. Unseen "
            "category percentage is reported separately."
        )

    elif view == "Monitoring Baseline":

        _table43(pd.DataFrame({
            "Setting": [
                "Data Source",
                "Date Column",
                "Baseline Records",
                "Comparison Records",
                "Minimum Sample",
                "Drift Classification",
                "Monitoring Mode"
            ],
            "Value": [
                SOURCE43,
                str(DATE_COL43),
                len(BASELINE43),
                len(CURRENT43),
                "30 per window",
                "PSI heuristic thresholds",
                "Historical screening only"
            ]
        }))

        _note43(
            "The saved JSON documents the comparison "
            "configuration and aggregate findings. "
            "It does not establish a live monitoring service."
        )

    elif view == "Recommendations":

        _table43(ACTION_PLAN43)

        _note43(
            "Priority: investigate high distribution shifts, "
            "meaningful missingness changes and changes in "
            "data collection before considering retraining."
        )

    elif view == "Reports":

        _table43(pd.DataFrame({
            "Report": [
                "Feature Drift",
                "Missingness Shift",
                "Monitoring Summary",
                "Recommendations",
                "Baseline Metadata"
            ],
            "Filename": [
                "section43_feature_drift.csv",
                "section43_missingness_shift.csv",
                "section43_monitoring_summary.csv",
                "section43_monitoring_recommendations.csv",
                "section43_baseline_metadata.json"
            ]
        }))

        _note43(
            "Reports saved to: " + str(REPORT43)
        )

NAV43 = widgets.ToggleButtons(
    options=[
        "Executive Overview",
        "Feature Drift",
        "Data Quality",
        "Categorical Drift",
        "Monitoring Baseline",
        "Recommendations",
        "Reports"
    ],
    value="Executive Overview",
    layout=widgets.Layout(width="100%")
)

OUTPUT43 = widgets.Output(
    layout=widgets.Layout(width="100%")
)

def render43(change=None):
    with OUTPUT43:
        clear_output(wait=True)
        _draw43(NAV43.value)

NAV43.observe(render43, names="value")

FEATURE_SELECT43.observe(
    lambda change: render43(),
    names="value"
)

DASHBOARD43 = widgets.VBox([
    NAV43,
    OUTPUT43
])

display(DASHBOARD43)
render43()

SECTION43_STATUS = MONITORING_STATUS43
```

```python
ROOT44 = Path(globals().get("ROOT43", globals().get("ROOT42", Path.cwd()))).resolve()
REPORT44 = ROOT44 / "reports" / "model_reliability"
MODEL_DIR44 = ROOT44 / "models"
REPORT44.mkdir(parents=True, exist_ok=True)
MODEL_DIR44.mkdir(parents=True, exist_ok=True)
STREAM44 = ROOT44 / "stream.py"

MODEL_NAMES44 = [
    "best_model", "final_model", "trained_model",
    "best_estimator", "best_pipeline", "final_pipeline",
    "model_pipeline", "pipeline", "model", "clf", "regressor"
]

MODEL44 = None
MODEL_NAME44 = "Unavailable"
MODEL_STATUS44 = "NO FITTED MODEL FOUND"

try:
    from sklearn.utils.validation import check_is_fitted
    from sklearn.utils.multiclass import type_of_target
    import joblib
    import hashlib
    RETENTION_AVAILABLE44 = True
except ImportError:
    RETENTION_AVAILABLE44 = False

if RETENTION_AVAILABLE44:
    for name44 in MODEL_NAMES44:
        candidate44 = globals().get(name44)

        if candidate44 is None or not callable(
            getattr(candidate44, "predict", None)
        ):
            continue

        try:
            check_is_fitted(candidate44)
            MODEL44 = candidate44
            MODEL_NAME44 = name44
            MODEL_STATUS44 = "FITTED MODEL RETAINED"
            break
        except (TypeError, AttributeError, ValueError):
            continue

X_TEST_NAMES44 = [
    "X_test", "X_test_scaled", "X_test_final",
    "X_test_processed", "x_test"
]

Y_TEST_NAMES44 = [
    "y_test", "Y_test", "y_test_final",
    "test_labels", "target_test"
]

X_TEST44 = None
Y_TEST44 = None
X_NAME44 = "Unavailable"
Y_NAME44 = "Unavailable"

for name44 in X_TEST_NAMES44:
    value44 = globals().get(name44)
    if isinstance(value44, (pd.DataFrame, pd.Series, np.ndarray)):
        if len(value44) > 0:
            X_TEST44 = value44
            X_NAME44 = name44
            break

for name44 in Y_TEST_NAMES44:
    value44 = globals().get(name44)
    if isinstance(value44, (pd.Series, pd.DataFrame, np.ndarray, list)):
        if len(value44) > 0:
            Y_TEST44 = value44
            Y_NAME44 = name44
            break

MODEL_FILE44 = MODEL_DIR44 / "section44_retained_model.joblib"
MODEL_SAVED44 = False
MODEL_SAVE_NOTE44 = "No new model artifact created"

if MODEL44 is not None and RETENTION_AVAILABLE44:
    if MODEL_FILE44.exists():
        MODEL_SAVED44 = True
        MODEL_SAVE_NOTE44 = (
            "Existing saved artifact preserved; "
            "its identity has not been verified against the current model"
        )
    else:
        try:
            joblib.dump(MODEL44, MODEL_FILE44)
            MODEL_SAVED44 = MODEL_FILE44.is_file()
            MODEL_SAVE_NOTE44 = (
                "Fitted model saved without retraining"
                if MODEL_SAVED44 else "Model save failed"
            )
        except Exception as exc44:
            MODEL_SAVE_NOTE44 = f"Model could not be saved: {exc44}"

PREDICTIONS44 = pd.DataFrame()
PREDICTION_SOURCE44 = "Unavailable"
PREDICTION_ERROR44 = ""
CONFIDENCE44 = None
TASK_TYPE44 = "Unknown"

if Y_TEST44 is not None:
    true44 = np.asarray(Y_TEST44).reshape(-1)
else:
    true44 = np.array([])

pred44 = None

if MODEL44 is not None and X_TEST44 is not None and len(true44):
    if len(X_TEST44) == len(true44):
        try:
            pred44 = np.asarray(
                MODEL44.predict(X_TEST44)
            ).reshape(-1)

            PREDICTION_SOURCE44 = (
                f"{MODEL_NAME44}.predict({X_NAME44})"
            )
        except Exception as exc44:
            PREDICTION_ERROR44 = str(exc44)

if pred44 is None and len(true44):
    for name44 in [
        "y_pred_test", "y_test_pred",
        "test_predictions", "y_pred"
    ]:
        candidate44 = globals().get(name44)

        if candidate44 is None:
            continue

        candidate44 = np.asarray(candidate44).reshape(-1)

        if len(candidate44) == len(true44):
            pred44 = candidate44
            PREDICTION_SOURCE44 = (
                f"Existing notebook predictions: {name44}; "
                "model linkage not independently verified"
            )
            break

if pred44 is not None and len(pred44) == len(true44):
    try:
        target_type44 = type_of_target(true44)

        TASK_TYPE44 = (
            "Regression"
            if target_type44 == "continuous"
            else "Classification"
            if target_type44 in ("binary", "multiclass")
            else "Unsupported target type"
        )
    except Exception:
        TASK_TYPE44 = "Unknown"

    if TASK_TYPE44 in ("Regression", "Classification"):
        PREDICTIONS44 = pd.DataFrame({
            "Actual": true44,
            "Predicted": pred44
        })

        if TASK_TYPE44 == "Regression":
            PREDICTIONS44["Actual"] = pd.to_numeric(
                PREDICTIONS44["Actual"], errors="coerce"
            )
            PREDICTIONS44["Predicted"] = pd.to_numeric(
                PREDICTIONS44["Predicted"], errors="coerce"
            )
            PREDICTIONS44 = PREDICTIONS44.dropna(
                subset=["Actual", "Predicted"]
            )
            PREDICTIONS44["Absolute_Error"] = (
                PREDICTIONS44["Actual"]
                - PREDICTIONS44["Predicted"]
            ).abs()
        else:
            PREDICTIONS44["Correct"] = (
                PREDICTIONS44["Actual"]
                == PREDICTIONS44["Predicted"]
            )
            PREDICTIONS44["Error"] = (
                ~PREDICTIONS44["Correct"]
            ).astype(int)

if (
    TASK_TYPE44 == "Classification"
    and MODEL44 is not None
    and X_TEST44 is not None
    and len(PREDICTIONS44) == len(X_TEST44)
    and callable(getattr(MODEL44, "predict_proba", None))
    and PREDICTION_SOURCE44.startswith(MODEL_NAME44 + ".predict")
):
    try:
        probabilities44 = np.asarray(
            MODEL44.predict_proba(X_TEST44)
        )

        if (
            probabilities44.ndim == 2
            and probabilities44.shape[0] == len(PREDICTIONS44)
            and np.isfinite(probabilities44).all()
        ):
            CONFIDENCE44 = probabilities44.max(axis=1)

            PREDICTIONS44["Confidence"] = CONFIDENCE44
            PREDICTIONS44["Uncertainty"] = 1 - CONFIDENCE44
    except Exception as exc44:
        PREDICTION_ERROR44 = str(exc44)

METRIC_ROWS44 = []

if not PREDICTIONS44.empty:
    from sklearn import metrics as skmetrics44

    actual44 = PREDICTIONS44["Actual"]
    predicted44 = PREDICTIONS44["Predicted"]

    if TASK_TYPE44 == "Regression":
        METRIC_ROWS44 = [
            ("MAE", skmetrics44.mean_absolute_error(actual44, predicted44)),
            ("RMSE", np.sqrt(skmetrics44.mean_squared_error(actual44, predicted44))),
            ("R2", skmetrics44.r2_score(actual44, predicted44))
        ]
    elif TASK_TYPE44 == "Classification":
        METRIC_ROWS44 = [
            ("Accuracy", skmetrics44.accuracy_score(actual44, predicted44)),
            ("Balanced Accuracy", skmetrics44.balanced_accuracy_score(actual44, predicted44)),
            ("Macro Precision", skmetrics44.precision_score(
                actual44, predicted44, average="macro", zero_division=0
            )),
            ("Macro Recall", skmetrics44.recall_score(
                actual44, predicted44, average="macro", zero_division=0
            )),
            ("Macro F1", skmetrics44.f1_score(
                actual44, predicted44, average="macro", zero_division=0
            ))
        ]

METRICS44 = pd.DataFrame(
    METRIC_ROWS44,
    columns=["Metric", "Value"]
)

if not METRICS44.empty:
    METRICS44["Value"] = METRICS44["Value"].round(5)

ERROR_SUMMARY44 = pd.DataFrame()

if not PREDICTIONS44.empty:
    if TASK_TYPE44 == "Regression":
        ERROR_SUMMARY44 = pd.DataFrame([
            ("Median Absolute Error",
             PREDICTIONS44["Absolute_Error"].median()),
            ("90th Percentile Absolute Error",
             PREDICTIONS44["Absolute_Error"].quantile(0.90)),
            ("95th Percentile Absolute Error",
             PREDICTIONS44["Absolute_Error"].quantile(0.95))
        ], columns=["Indicator", "Value"])
    else:
        ERROR_SUMMARY44 = pd.DataFrame([
            ("Correct Predictions",
             int(PREDICTIONS44["Correct"].sum())),
            ("Incorrect Predictions",
             int((~PREDICTIONS44["Correct"]).sum())),
            ("Error Rate (%)",
             float(PREDICTIONS44["Error"].mean() * 100))
        ], columns=["Indicator", "Value"])

CONFIDENCE_SUMMARY44 = pd.DataFrame()

if "Confidence" in PREDICTIONS44.columns:
    bins44 = np.linspace(0, 1, 11)

    confidence_group44 = pd.cut(
        PREDICTIONS44["Confidence"],
        bins=bins44,
        include_lowest=True
    )

    CONFIDENCE_SUMMARY44 = (
        PREDICTIONS44
        .groupby(confidence_group44, observed=True)
        .agg(
            Predictions=("Correct", "size"),
            Mean_Confidence=("Confidence", "mean"),
            Observed_Accuracy=("Correct", "mean")
        )
        .reset_index()
    )

    CONFIDENCE_SUMMARY44["Confidence"] = (
        CONFIDENCE_SUMMARY44["Confidence"].astype(str)
    )

RECOMMENDATIONS44 = []

if MODEL44 is None:
    RECOMMENDATIONS44.append(
        ("High", "Model Retention",
         "Locate the fitted model from previous notebook sections.")
    )

if PREDICTIONS44.empty:
    RECOMMENDATIONS44.append(
        ("High", "Prediction Evaluation",
         "Provide matching held-out features, labels and predictions.")
    )
else:
    RECOMMENDATIONS44.append(
        ("Review", "Model Evaluation",
         "Confirm test-set independence and absence of data leakage.")
    )

if TASK_TYPE44 == "Classification":
    if "Confidence" not in PREDICTIONS44.columns:
        RECOMMENDATIONS44.append(
            ("Review", "Prediction Confidence",
             "Probability estimates are unavailable; do not invent confidence scores.")
        )
    else:
        RECOMMENDATIONS44.append(
            ("Review", "Confidence Calibration",
             "Compare reported probabilities with observed correctness before interpreting them as calibrated confidence.")
        )

if TASK_TYPE44 == "Regression":
    RECOMMENDATIONS44.append(
        ("Review", "Regression Reliability",
         "Compare absolute errors with business-defined acceptable error limits.")
    )

RECOMMENDATIONS44.append(
    ("Review", "Deployment",
     "Test the retained model together with its preprocessing pipeline before deployment.")
)

ACTION_PLAN44 = pd.DataFrame(
    RECOMMENDATIONS44,
    columns=["Priority", "Area", "Recommendation"]
)

MODEL_METADATA44 = {
    "model_variable": MODEL_NAME44,
    "model_class": (
        type(MODEL44).__name__ if MODEL44 is not None else None
    ),
    "model_status": MODEL_STATUS44,
    "artifact_path": str(MODEL_FILE44) if MODEL_SAVED44 else None,
    "artifact_note": MODEL_SAVE_NOTE44,
    "task_type": TASK_TYPE44,
    "test_features": X_NAME44,
    "test_labels": Y_NAME44,
    "prediction_source": PREDICTION_SOURCE44,
    "prediction_error": PREDICTION_ERROR44,
    "evaluated_predictions": len(PREDICTIONS44),
    "saved_at": datetime.now().astimezone().isoformat(),
    "model_retrained": False,
    "confidence_calibration_verified": False
}

if MODEL_SAVED44:
    try:
        MODEL_METADATA44["artifact_sha256"] = (
            hashlib.sha256(MODEL_FILE44.read_bytes()).hexdigest()
        )
    except OSError:
        MODEL_METADATA44["artifact_sha256"] = None

METRICS44.to_csv(
    REPORT44 / "section44_performance_metrics.csv",
    index=False
)

ERROR_SUMMARY44.to_csv(
    REPORT44 / "section44_error_summary.csv",
    index=False
)

CONFIDENCE_SUMMARY44.to_csv(
    REPORT44 / "section44_confidence_summary.csv",
    index=False
)

ACTION_PLAN44.to_csv(
    REPORT44 / "section44_reliability_recommendations.csv",
    index=False
)

(
    REPORT44 / "section44_model_metadata.json"
).write_text(
    json.dumps(MODEL_METADATA44, indent=2, default=str),
    encoding="utf-8"
)

CSS44 = """
<style>
.s44hero{
    background:linear-gradient(135deg,#22223b,#32344f 52%,#4a4e69);
    padding:25px 29px;border-radius:16px;
    color:white;font-family:Arial;margin:10px 0 16px;
}
.s44hero small{
    color:#c9ada7;letter-spacing:1.7px;
    font-size:10px;font-weight:700;
}
.s44hero h2{
    color:white!important;font-size:26px;margin:8px 0;
}
.s44hero p{
    color:#f2e9e4;font-size:12px;margin:0;
}
.s44metrics{
    display:grid;grid-template-columns:repeat(4,1fr);
    gap:10px;margin:15px 0;
}
.s44metric{
    background:white;border:1px solid #e6dfdc;
    border-radius:11px;padding:17px;font-family:Arial;
}
.s44metric span{
    color:#9a8c98;font-size:10px;display:block;
}
.s44metric b{
    color:#22223b;font-size:21px;display:block;margin-top:8px;
}
.s44note{
    background:#f2e9e4;border-left:4px solid #9a8c98;
    padding:14px;border-radius:9px;color:#4a4e69;
    margin:12px 0;font:12px Arial;line-height:1.7;
}
</style>
"""

display(HTML(CSS44))

display(HTML("""
<div class="s44hero">
<small>SECTION 44 · MODEL RETENTION & PREDICTION RELIABILITY</small>
<h2>Retained Model & Error Analysis Dashboard</h2>
<p>Fitted Model · Prediction Accuracy · Error Distribution · Confidence</p>
</div>
"""))

KPI44 = [
    ("MODEL", MODEL_NAME44),
    ("PREDICTIONS", f"{len(PREDICTIONS44):,}"),
    ("TASK TYPE", TASK_TYPE44),
    ("MODEL FILE", "PRESENT" if MODEL_SAVED44 else "NOT SAVED")
]

KPI_HTML44 = "".join(
    '<div class="s44metric">'
    f'<span>{escape(str(name))}</span>'
    f'<b>{escape(str(value))}</b>'
    '</div>'
    for name, value in KPI44
)

display(HTML(
    '<div class="s44metrics">' + KPI_HTML44 + '</div>'
))

def _note44(message):
    display(HTML(
        '<div class="s44note">' + escape(str(message)) + '</div>'
    ))

def _table44(df, n=40):
    if df.empty:
        _note44("No evaluated records available.")
    else:
        display(HTML(
            df.head(n).to_html(
                index=False, escape=True, border=0
            )
        ))

def _draw44(view):

    if view == "Executive Overview":
        if PREDICTIONS44.empty:
            _note44(
                "No compatible held-out prediction data was found. "
                "The original model and existing files remain unchanged."
            )
        else:
            if TASK_TYPE44 == "Regression":
                chart44 = PREDICTIONS44[
                    ["Absolute_Error"]
                ].rename(columns={
                    "Absolute_Error": "Error"
                })
                x_label44 = "Absolute prediction error"

            elif "Uncertainty" in PREDICTIONS44.columns:
                chart44 = PREDICTIONS44[
                    ["Uncertainty"]
                ].rename(columns={
                    "Uncertainty": "Error"
                })
                x_label44 = "Prediction uncertainty (1 - maximum probability)"

            else:
                chart44 = PREDICTIONS44[
                    ["Error"]
                ].copy()
                x_label44 = "Misclassification indicator (0 or 1)"

            fig44 = px.ecdf(
                chart44,
                x="Error",
                title="Empirical Cumulative Distribution of Prediction Errors",
                color_discrete_sequence=["#4a4e69"]
            )

            fig44.update_layout(
                template="plotly_white",
                height=470,
                font=dict(family="Arial", size=12, color="#22223b"),
                margin=dict(l=25, r=20, t=65, b=35),
                xaxis_title=x_label44,
                yaxis_title="Cumulative proportion",
                paper_bgcolor="white"
            )

            display(fig44)

        _table44(METRICS44)

        _note44(
            "Insight: The ECDF shows the proportion of evaluated "
            "predictions at or below each error or uncertainty level."
        )

        _note44(
            "Interpretation: For regression, smaller absolute errors "
            "indicate closer predictions. For classification, probability "
            "uncertainty is not equivalent to model error or calibration."
        )

        _note44(
            "Recommendation: Validate performance against task-specific "
            "requirements before model deployment."
        )

    elif view == "Retained Model":
        _table44(pd.DataFrame([
            ("Model Variable", MODEL_NAME44),
            ("Model Status", MODEL_STATUS44),
            ("Model Class", MODEL_METADATA44["model_class"]),
            ("Artifact", str(MODEL_FILE44) if MODEL_SAVED44 else "Unavailable"),
            ("Save Result", MODEL_SAVE_NOTE44),
            ("Model Retrained", "No"),
            ("Prediction Source", PREDICTION_SOURCE44)
        ], columns=["Property", "Value"]))

        _note44(
            "Model retention preserves a previously fitted estimator. "
            "The exact preprocessing pipeline and feature schema must "
            "also be available for future predictions."
        )

    elif view == "Performance Metrics":
        _table44(METRICS44)
        _table44(ERROR_SUMMARY44)

        _note44(
            "Metrics are calculated from the available actual and "
            "predicted values. Verify that these correspond to a "
            "genuine held-out test set."
        )

    elif view == "Confidence Analysis":
        _table44(CONFIDENCE_SUMMARY44)

        if CONFIDENCE_SUMMARY44.empty:
            _note44(
                "Probability-based confidence results are unavailable."
            )
        else:
            _note44(
                "Confidence is the model's maximum predicted class "
                "probability. Compare confidence against actual correctness "
                "before claiming reliable probabilities."
            )

    elif view == "Error Analysis":
        _table44(ERROR_SUMMARY44)

        _note44(
            "Error distributions should be reviewed alongside feature "
            "coverage, target imbalance and important user subgroups."
        )

    elif view == "Recommendations":
        _table44(ACTION_PLAN44)

    elif view == "Exports":
        _table44(pd.DataFrame({
            "Report": [
                "Performance Metrics",
                "Error Summary",
                "Confidence Summary",
                "Reliability Recommendations",
                "Model Metadata"
            ],
            "Filename": [
                "section44_performance_metrics.csv",
                "section44_error_summary.csv",
                "section44_confidence_summary.csv",
                "section44_reliability_recommendations.csv",
                "section44_model_metadata.json"
            ]
        }))

        _note44(
            "Reports saved to: " + str(REPORT44)
        )

NAV44 = widgets.ToggleButtons(
    options=[
        "Executive Overview",
        "Retained Model",
        "Performance Metrics",
        "Confidence Analysis",
        "Error Analysis",
        "Recommendations",
        "Exports"
    ],
    value="Executive Overview",
    layout=widgets.Layout(width="100%")
)

OUTPUT44 = widgets.Output(layout=widgets.Layout(width="100%"))

def render44(change=None):
    with OUTPUT44:
        clear_output(wait=True)
        _draw44(NAV44.value)

NAV44.observe(render44, names="value")

DASHBOARD44 = widgets.VBox([NAV44, OUTPUT44])
display(DASHBOARD44)
render44()

SECTION44_STATUS = (
    "FITTED MODEL AND PREDICTIONS AVAILABLE"
    if MODEL44 is not None and not PREDICTIONS44.empty
    else "MODEL OR PREDICTION EVIDENCE REQUIRES REVIEW"
)
```

```python
ROOT46 = Path(globals().get("ROOT45", globals().get("ROOT44", Path.cwd()))).resolve()
REPORT46 = ROOT46 / "reports" / "integration_testing"
REPORT46.mkdir(parents=True, exist_ok=True)
STREAM46 = ROOT46 / "stream.py"
MODEL_PATH46 = ROOT46 / "models" / "section44_retained_model.joblib"
MANIFEST_PATH46 = ROOT46 / "reports" / "model_deployment" / "section45_model_manifest.json"
MODEL46 = globals().get("MODEL44")
X_TEST46 = globals().get("X_TEST44")
TEST_RESULTS46 = []

def _set_check46(group, check, status, evidence):
    for item in TEST_RESULTS46:
        if item["Group"] == group and item["Check"] == check:
            item.update(Status=status, Evidence=str(evidence),
                        Checked_At=datetime.now().astimezone().isoformat(timespec="seconds"))
            return
    TEST_RESULTS46.append({
        "Group": group, "Check": check, "Status": status,
        "Evidence": str(evidence),
        "Checked_At": datetime.now().astimezone().isoformat(timespec="seconds")
    })

def _file_syntax46(label, path):
    if not path.is_file():
        _set_check46("Application", label, "FAIL", "File not found")
        return
    try:
        ast.parse(path.read_text(encoding="utf-8-sig"), filename=str(path))
        _set_check46("Application", label, "PASS", "Syntax parsed successfully; runtime not tested")
    except (SyntaxError, OSError, UnicodeError) as exc:
        _set_check46("Application", label, "FAIL", str(exc))

_file_syntax46("Streamlit entry point syntax", STREAM46)
_file_syntax46("Executive overview module syntax", ROOT46 / "frontend_pages" / "section34_overview.py")

if MODEL_PATH46.is_file():
    _set_check46("Model Artifact", "Retained artifact exists", "PASS",
                 str(MODEL_PATH46.relative_to(ROOT46)))
else:
    _set_check46("Model Artifact", "Retained artifact exists", "FAIL", "Run model retention in Section 44")

manifest46 = {}
if MANIFEST_PATH46.is_file():
    try:
        manifest46 = json.loads(MANIFEST_PATH46.read_text(encoding="utf-8"))
        _set_check46("Model Artifact", "Version manifest available", "PASS", "Section 45 manifest parsed")
    except (OSError, ValueError) as exc:
        _set_check46("Model Artifact", "Version manifest available", "FAIL", str(exc))
else:
    _set_check46("Model Artifact", "Version manifest available", "REVIEW", "Section 45 manifest unavailable")

if MODEL_PATH46.is_file() and manifest46.get("sha256"):
    if "hashlib" in globals():
        digest46 = hashlib.sha256()
        with MODEL_PATH46.open("rb") as fh46:
            for chunk46 in iter(lambda: fh46.read(1048576), b""):
                digest46.update(chunk46)
        checksum46 = digest46.hexdigest()
        _set_check46("Model Artifact", "Artifact matches manifest", 
                     "PASS" if checksum46 == manifest46["sha256"] else "FAIL",
                     "SHA-256 matches" if checksum46 == manifest46["sha256"] else "SHA-256 mismatch")
    else:
        _set_check46("Model Artifact", "Artifact matches manifest", "REVIEW", "hashlib was not initialized in the notebook")
else:
    _set_check46("Model Artifact", "Artifact matches manifest", "BLOCKED", "Artifact or prior checksum unavailable")

if MODEL46 is not None and callable(getattr(MODEL46, "predict", None)):
    _set_check46("Model Runtime", "In-memory estimator available", "PASS", type(MODEL46).__name__)
else:
    _set_check46("Model Runtime", "In-memory estimator available", "BLOCKED", "No estimator from Section 44")

if X_TEST46 is not None and hasattr(X_TEST46, "__len__") and len(X_TEST46) > 0:
    _set_check46("Model Runtime", "Existing test features available", "PASS", f"{len(X_TEST46):,} rows")
else:
    _set_check46("Model Runtime", "Existing test features available", "BLOCKED", "No nonempty X_TEST44")

feature_names46 = manifest46.get("input_features", [])
if feature_names46 and isinstance(X_TEST46, pd.DataFrame):
    missing46 = [c for c in feature_names46 if c not in X_TEST46.columns]
    _set_check46("Feature Contract", "Named features match test inputs",
                 "PASS" if not missing46 else "FAIL",
                 "All manifest features available" if not missing46 else "Missing columns: " + ", ".join(missing46[:12]))
else:
    _set_check46("Feature Contract", "Named features match test inputs", "REVIEW",
                 "Named feature contract or DataFrame test inputs unavailable")

_set_check46("Inference", "Single-row model inference", "NOT TESTED", "Run explicit test in the Model Inference tab")
_set_check46("Flask API", "Local read-only endpoint response", "NOT TESTED", "Run explicit API check in the Flask API tab")
_set_check46("Flask API", "Live model inference through API", "NOT TESTED", "Requires a documented inference endpoint, schema and controlled sample")
_set_check46("Streamlit", "Interactive dashboard loads", "NOT TESTED", "Manual browser test required")
_set_check46("Streamlit", "Prediction shown in Streamlit", "NOT TESTED", "Manual end-to-end browser test required")
_set_check46("Acceptance", "Client deployment sign-off", "NOT TESTED", "Authorized client approval required")

def _results46():
    return pd.DataFrame(TEST_RESULTS46)

def _export46():
    frame46 = _results46()
    frame46.to_csv(REPORT46 / "section46_test_register.csv", index=False)
    summary46 = (frame46.groupby(["Group", "Status"]).size()
                 .rename("Count").reset_index())
    summary46.to_csv(REPORT46 / "section46_status_by_group.csv", index=False)
    frame46.loc[frame46["Status"] != "PASS"].to_csv(
        REPORT46 / "section46_outstanding_checks.csv", index=False)
    pd.DataFrame([{"Model": str(globals().get("MODEL_NAME44", "Unavailable")),
                   "Artifact": str(MODEL_PATH46),
                   "Manifest": str(MANIFEST_PATH46),
                   "Model_retrained": False,
                   "Runtime_API_test_is_model_inference_test": False}]).to_csv(
        REPORT46 / "section46_integration_context.csv", index=False)
    return frame46, summary46

_export46()
CSS46 = """
<style>
.s46hero{background:linear-gradient(135deg,#22223b,#32344f 52%,#4a4e69);padding:24px 28px;border-radius:15px;font-family:Arial;margin:12px 0 15px;color:#fff}
.s46hero small{color:#c9ada7;letter-spacing:1.6px;font-size:10px;font-weight:700}
.s46hero h2{color:#fff!important;font-size:25px;margin:7px 0}
.s46hero p{color:#f2e9e4;font-size:12px;margin:0}
.s46kpis{display:grid;grid-template-columns:repeat(4,1fr);gap:9px;margin:14px 0}
.s46kpi{border:1px solid #e6dfdc;background:#fff;padding:14px;border-radius:11px;font-family:Arial}
.s46kpi span{display:block;font-size:10px;color:#9a8c98}
.s46kpi b{display:block;color:#22223b;font-size:21px;margin-top:5px}
.s46note{background:#f2e9e4;border-left:4px solid #9a8c98;border-radius:9px;padding:13px;margin:12px 0;color:#4a4e69;font:12px Arial;line-height:1.65}
</style>
"""
display(HTML(CSS46))
display(HTML("""
<div class="s46hero"><small>SECTION 46 · END-TO-END INTEGRATION VALIDATION</small>
<h2>Retained Model, Flask API & Streamlit Test Dashboard</h2>
<p>Artifact Integrity · Prediction Smoke Test · API Connectivity · Acceptance Evidence</p></div>
"""))

def _note46(message):
    display(HTML('<div class="s46note">' + escape(str(message)) + '</div>'))

def _table46(frame):
    if frame.empty:
        _note46("No records available")
    else:
        display(HTML(frame.to_html(index=False, escape=True, border=0)))

INFERENCE_BUTTON46 = widgets.Button(description="Run One-Row Prediction Test", button_style="primary", icon="play")
INFERENCE_OUTPUT46 = widgets.Output()
API_URL46 = widgets.Text(value="http://127.0.0.1:5000/api/market/summary",
                         description="Local URL:", layout=widgets.Layout(width="95%"))
API_BUTTON46 = widgets.Button(description="Test Read-Only GET Endpoint", button_style="primary", icon="wifi")
API_OUTPUT46 = widgets.Output()

def _test_inference46(button):
    with INFERENCE_OUTPUT46:
        clear_output(wait=True)
        if MODEL46 is None or not callable(getattr(MODEL46, "predict", None)):
            _set_check46("Inference", "Single-row model inference", "BLOCKED", "Fitted model not available")
            print("BLOCKED: No fitted in-memory model available")
        elif X_TEST46 is None or len(X_TEST46) == 0:
            _set_check46("Inference", "Single-row model inference", "BLOCKED", "Test features unavailable")
            print("BLOCKED: No test inputs available")
        else:
            try:
                sample46 = X_TEST46.iloc[[0]] if hasattr(X_TEST46, "iloc") else X_TEST46[:1]
                output46 = np.asarray(MODEL46.predict(sample46)).reshape(-1)
                if len(output46) != 1:
                    raise ValueError("Model did not return a single prediction")
                _set_check46("Inference", "Single-row model inference", "PASS",
                             "One prediction returned for first existing test row; correctness not certified")
                print("PASS: Single-row inference completed. Predicted value:", str(output46[0])[:120])
            except Exception as exc46:
                _set_check46("Inference", "Single-row model inference", "FAIL", repr(exc46))
                print("FAILED:", repr(exc46))
        _export46()
    _render46()

def _test_api46(button):
    with API_OUTPUT46:
        clear_output(wait=True)
        from urllib.parse import urlsplit
        from urllib.request import urlopen
        parsed46 = urlsplit(API_URL46.value.strip())
        if (parsed46.scheme != "http" or parsed46.hostname not in ("localhost", "127.0.0.1", "::1")
                or parsed46.username is not None or parsed46.password is not None):
            print("Only plain HTTP loopback URLs are allowed for this read-only test")
            return
        try:
            with urlopen(API_URL46.value.strip(), timeout=3) as response46:
                status46 = response46.status
                content_type46 = response46.headers.get("Content-Type", "")
                body46 = response46.read(65536)
            if status46 == 200 and "json" in content_type46.lower():
                json.loads(body46.decode("utf-8"))
                _set_check46("Flask API", "Local read-only endpoint response", "PASS",
                             "HTTP 200 and parseable JSON from local GET endpoint; model inference not tested")
                print("PASS: Local endpoint returned HTTP 200 and valid JSON")
            else:
                _set_check46("Flask API", "Local read-only endpoint response", "FAIL",
                             f"HTTP {status46}; content type {content_type46}")
                print("FAILED: Unexpected response", status46, content_type46)
        except Exception as exc46:
            _set_check46("Flask API", "Local read-only endpoint response", "FAIL", repr(exc46))
            print("FAILED:", repr(exc46))
        _export46()
    _render46()

INFERENCE_BUTTON46.on_click(_test_inference46)
API_BUTTON46.on_click(_test_api46)

def _radar46():
    frame46 = _results46()
    grouped46 = frame46.groupby("Group").agg(
        total=("Status", "size"),
        passed=("Status", lambda s: int((s == "PASS").sum()))
    ).reset_index()
    grouped46["Percent"] = (100 * grouped46["passed"] / grouped46["total"]).round(1)
    labels46 = grouped46["Group"].tolist()
    values46 = grouped46["Percent"].tolist()
    if not labels46:
        _note46("No checks have been registered")
        return
    fig46 = go.Figure(go.Scatterpolar(
        r=values46 + values46[:1],
        theta=labels46 + labels46[:1],
        fill="toself",
        fillcolor="rgba(74,78,105,0.24)",
        line=dict(color="#4a4e69", width=2),
        name="Recorded PASS share"
    ))
    fig46.update_layout(
        title="Technical Verification Radar — Recorded PASS Share by Group",
        polar=dict(radialaxis=dict(range=[0, 100], ticksuffix="%", visible=True)),
        showlegend=False, height=520, paper_bgcolor="white",
        font=dict(family="Arial", size=11, color="#22223b"),
        margin=dict(l=55, r=55, t=70, b=35)
    )
    display(fig46)
    _table46(grouped46[["Group", "total", "passed", "Percent"]])
    _note46("The radar measures the share of recorded PASS checks per area, not predictive accuracy or live production readiness. An untested check is not a pass.")

def _show46(view):
    frame46 = _results46()
    if view == "Executive Overview":
        _radar46()
        _note46("Recommendation: Run the explicit in-memory prediction smoke test, then check your local Flask summary endpoint. These do not by themselves demonstrate model inference via Flask or Streamlit.")
    elif view == "Model Inference":
        display(widgets.VBox([INFERENCE_BUTTON46, INFERENCE_OUTPUT46]))
        _table46(frame46[frame46["Group"].isin(["Model Runtime", "Inference", "Feature Contract"])])
        _note46("Predict uses one existing test record; it does not retrain or modify the saved model. A successful response is only a smoke test.")
    elif view == "Flask API":
        display(widgets.VBox([API_URL46, API_BUTTON46, API_OUTPUT46]))
        _table46(frame46[frame46["Group"] == "Flask API"])
        _note46("This GET request checks local JSON API connectivity. It does not test the model prediction endpoint or prove end-to-end ML inference.")
    elif view == "Artifact & Schema":
        _table46(frame46[frame46["Group"].isin(["Model Artifact", "Feature Contract"])])
        _note46("Artifact hash agreement confirms byte consistency with the manifest, not safety, preprocessing completeness or behavioral compatibility.")
    elif view == "Application Checks":
        _table46(frame46[frame46["Group"].isin(["Application", "Streamlit", "Acceptance"])])
        _note46("Dashboard opening, user workflows and client approval require manual evidence and remain NOT TESTED.")
    elif view == "Outstanding Issues":
        _table46(frame46.loc[frame46["Status"] != "PASS"])
    elif view == "Reports":
        _table46(pd.DataFrame({"Report": ["Test register", "Status summary", "Outstanding checks", "Integration context"],
                              "Filename": ["section46_test_register.csv", "section46_status_by_group.csv",
                                           "section46_outstanding_checks.csv", "section46_integration_context.csv"]}))
        _note46("All reports are written to: " + str(REPORT46))

NAV46 = widgets.ToggleButtons(
    options=["Executive Overview", "Model Inference", "Flask API", "Artifact & Schema",
             "Application Checks", "Outstanding Issues", "Reports"],
    value="Executive Overview", layout=widgets.Layout(width="100%")
)
OUTPUT46 = widgets.Output(layout=widgets.Layout(width="100%"))

def _render46(change=None):
    with OUTPUT46:
        clear_output(wait=True)
        _show46(NAV46.value)

NAV46.observe(_render46, names="value")
display(widgets.VBox([NAV46, OUTPUT46]))
_render46()
STATUS46 = _results46()["Status"].value_counts()
```

```python
ROOT55 = Path(globals().get('ROOT54', globals().get('ROOT53', Path.cwd()))).resolve()
OUT55 = ROOT55 / 'reports' / 'retrospective'
OUT55.mkdir(parents=True, exist_ok=True)

SOURCES55 = [
    ('Submission Audit', 'reports/submission/section39_audit_results.csv', 'Category', 'Check'),
    ('Stakeholder Review', 'reports/stakeholder_review/section53_demonstration_review.csv', None, 'Demonstration'),
    ('Project Closure', 'reports/project_closure/section51_closure_checklist.csv', None, 'Requirement')
]

parts55, source_notes55 = [], []

for source, relative, group_col, item_col in SOURCES55:
    path = ROOT55 / relative

    if not path.is_file():
        source_notes55.append((source, 'MISSING', relative))
        continue

    try:
        frame = pd.read_csv(path).fillna('')

        if frame.empty or not {'Status', item_col}.issubset(frame.columns):
            source_notes55.append((source, 'UNUSABLE', 'Empty or missing required columns'))
            continue

        status = frame['Status'].astype(str).str.upper().str.strip()
        mapped = status.map(
            lambda s: 'PASS' if s in ('PASS', 'PASSED')
            else 'REVIEW' if s in ('REVIEW', 'WARN', 'RECORDED')
            else 'OPEN'
        )

        group = (
            frame[group_col].astype(str)
            if group_col and group_col in frame
            else pd.Series([source] * len(frame), index=frame.index)
        )

        parts55.append(pd.DataFrame({
            'Category': group,
            'Item': frame[item_col].astype(str),
            'Status': mapped,
            'Original_Status': status,
            'Source': source
        }))

        source_notes55.append((source, 'LOADED', f'{len(frame)} records'))

    except (OSError, ValueError, pd.errors.ParserError) as exc:
        source_notes55.append((source, 'UNUSABLE', str(exc)))

RESULTS55 = (
    pd.concat(parts55, ignore_index=True)
    if parts55
    else pd.DataFrame(
        columns=['Category', 'Item', 'Status', 'Original_Status', 'Source']
    )
)

SOURCES_STATUS55 = pd.DataFrame(
    source_notes55,
    columns=['Source', 'Load_Status', 'Detail']
)

COUNT55 = len(RESULTS55)
PASSED55 = int(RESULTS55['Status'].eq('PASS').sum())
OPEN55 = int(RESULTS55['Status'].eq('OPEN').sum())
REVIEW55 = int(RESULTS55['Status'].eq('REVIEW').sum())

ISSUES55 = RESULTS55[RESULTS55['Status'].ne('PASS')].copy()

LESSONS55 = pd.DataFrame([
    ('Data integrity', 'Document missingness and schema assumptions before analytics', 'Proposed improvement'),
    ('Model preservation', 'Retain a verified artifact and matching preprocessing contract', 'Proposed improvement'),
    ('Integration', 'Record live API and browser evidence, not only source-code checks', 'Proposed improvement'),
    ('Client acceptance', 'Tie completed demonstrations to authentic reviewer evidence', 'Proposed improvement'),
    ('Delivery', 'Separate file availability from functional and business-outcome verification', 'Proposed improvement')
], columns=['Theme', 'Lesson', 'Evidence_Type'])

ACTIONS55 = (
    ISSUES55.groupby(['Category', 'Status'], as_index=False)
    .size()
    .rename(columns={'size': 'Count'})
    .sort_values('Count', ascending=False)
    if not ISSUES55.empty
    else pd.DataFrame(columns=['Category', 'Status', 'Count'])
)

RESULTS55.to_csv(OUT55 / 'section55_review_register.csv', index=False)
ISSUES55.to_csv(OUT55 / 'section55_outstanding_items.csv', index=False)
LESSONS55.to_csv(OUT55 / 'section55_lessons_learned.csv', index=False)
ACTIONS55.to_csv(OUT55 / 'section55_action_priorities.csv', index=False)
SOURCES_STATUS55.to_csv(OUT55 / 'section55_source_status.csv', index=False)

display(HTML("""
<div style="background:linear-gradient(135deg,#22223b,#4a4e69);
padding:23px;border-radius:14px;font-family:Arial;color:white;margin:12px 0">
<div style="color:#c9ada7;font-size:10px;letter-spacing:1.5px">
SECTION 55 · PROJECT RETROSPECTIVE</div>
<div style="font-size:24px;font-weight:bold;margin-top:7px">
Lessons Learned & Continuous Improvement</div>
<div style="color:#f2e9e4;font-size:12px;margin-top:7px">
Evidence Review · Unresolved Work · Practical Recommendations</div>
</div>
"""))

display(HTML(
    '<div style="display:flex;gap:10px;flex-wrap:wrap;margin:14px 0">'
    + ''.join(
        '<div style="flex:1;min-width:110px;border:1px solid #e6dfdc;'
        'border-radius:11px;padding:15px">'
        f'<div style="font:10px Arial;color:#9a8c98">{label}</div>'
        f'<div style="font:bold 24px Arial;color:#22223b">{value}</div>'
        '</div>'
        for label, value in [
            ('REVIEW RECORDS', COUNT55),
            ('RECORDED PASSES', PASSED55),
            ('OPEN ITEMS', OPEN55),
            ('REVIEW NEEDED', REVIEW55)
        ]
    )
    + '</div>'
))

NAV55 = widgets.ToggleButtons(
    options=[
        'Overview',
        'Outstanding Work',
        'Lessons Learned',
        'Source Evidence',
        'Exports'
    ],
    value='Overview',
    layout=widgets.Layout(width='100%')
)

PANEL55 = widgets.Output()

def show55(view):

    if view == 'Overview':

        if RESULTS55.empty:
            print(
                'No usable source reports. Run the relevant '
                'validation and review sections first.'
            )
        else:
            mix = (
                RESULTS55.groupby(['Category', 'Status'])
                .size()
                .unstack(fill_value=0)
            )

            mix = mix.reindex(
                columns=['PASS', 'REVIEW', 'OPEN'],
                fill_value=0
            )

            totals = mix.sum(axis=1)
            widths = totals / totals.sum()
            centers = widths.cumsum() - widths / 2

            fig = go.Figure()
            base = pd.Series(0.0, index=mix.index)

            for status, color in [
                ('PASS', '#4a4e69'),
                ('REVIEW', '#9a8c98'),
                ('OPEN', '#c9ada7')
            ]:
                heights = mix[status] / totals * 100

                fig.add_trace(go.Bar(
                    x=centers,
                    y=heights,
                    width=widths * 0.98,
                    base=base,
                    name=status,
                    marker_color=color,
                    customdata=pd.DataFrame({
                        'Category': mix.index,
                        'Count': mix[status]
                    }).to_numpy(),
                    hovertemplate=(
                        '%{customdata[0]}<br>'
                        + status
                        + ': %{customdata[1]} checks<extra></extra>'
                    )
                ))

                base = base + heights

            fig.update_layout(
                title='Review Outcomes by Category — Marimekko Chart',
                template='plotly_white',
                height=440,
                xaxis=dict(
                    title='Category width proportional to records',
                    range=[0, 1],
                    tickvals=centers.tolist(),
                    ticktext=mix.index.tolist(),
                    tickangle=-35
                ),
                yaxis=dict(
                    title='Category composition (%)',
                    range=[0, 100]
                ),
                barmode='overlay',
                font=dict(family='Arial', color='#22223b'),
                margin=dict(l=35, r=25, t=60, b=110),
                legend=dict(orientation='h', y=1.14)
            )

            display(fig)

        print(
            'Insight: Width shows review volume; segment height '
            'shows each category’s outcome mix.'
        )
        print(
            'Important: REVIEW and RECORDED are not independently '
            'verified passes.'
        )

    elif view == 'Outstanding Work':

        display(
            ACTIONS55 if not ACTIONS55.empty
            else pd.DataFrame({
                'Finding': ['No unresolved entries in available reports']
            })
        )

        display(ISSUES55.head(40))
        print(
            'Recommendation: Resolve OPEN items, '
            'then verify REVIEW evidence.'
        )

    elif view == 'Lessons Learned':

        display(LESSONS55)

        print(
            'These are proposed process lessons, '
            'not independently measured business outcomes.'
        )

    elif view == 'Source Evidence':

        display(SOURCES_STATUS55)

        display(
            RESULTS55.groupby(['Source', 'Status'])
            .size()
            .reset_index(name='Count')
            if not RESULTS55.empty
            else RESULTS55
        )

    elif view == 'Exports':

        display(pd.DataFrame({
            'Report': [
                'Review Register',
                'Outstanding Items',
                'Lessons Learned',
                'Action Priorities',
                'Source Status'
            ],
            'Filename': [
                'section55_review_register.csv',
                'section55_outstanding_items.csv',
                'section55_lessons_learned.csv',
                'section55_action_priorities.csv',
                'section55_source_status.csv'
            ]
        }))

        print('Reports saved to:', OUT55)

def render55(change=None):
    with PANEL55:
        clear_output(wait=True)
        show55(NAV55.value)

NAV55.observe(render55, names='value')

display(widgets.VBox([NAV55, PANEL55]))
render55()
```

```python
ROOT57 = Path(globals().get('ROOT56', globals().get('ROOT55', Path.cwd()))).resolve()
OUT57 = ROOT57 / 'reports' / 'risk_management'
OUT57.mkdir(parents=True, exist_ok=True)
REGISTER_PATH57 = OUT57 / 'section57_risk_register.csv'
SOURCE57 = ROOT57 / 'reports' / 'quality_assessment' / 'section56_open_findings.csv'

try:
    issues57 = pd.read_csv(SOURCE57).fillna('') if SOURCE57.is_file() else pd.DataFrame()
    if not {'Category', 'Check', 'Status'}.issubset(issues57.columns):
        issues57 = pd.DataFrame()
except (OSError, ValueError, pd.errors.ParserError):
    issues57 = pd.DataFrame()

entries57 = [
    (str(r.Category), str(r.Check), str(r.Status))
    for r in issues57.itertuples(index=False)
]
entries57 += [
    ('Deployment', 'Verify live Flask-to-model predictions', 'NOT VERIFIED'),
    ('Deployment', 'Verify Streamlit pages and exports in browser', 'NOT VERIFIED'),
    ('Security', 'Review secrets and sensitive data in submission', 'NOT VERIFIED'),
    ('Handover', 'Verify authorized client approval evidence', 'NOT VERIFIED')
]
entries57 = list(dict.fromkeys(entries57))
risk57 = pd.DataFrame([
    {'Risk_ID': f'R-{i:03d}', 'Category': category, 'Risk': item,
     'Source_Status': status, 'Likelihood': 3, 'Impact': 3,
     'Owner': '', 'Mitigation': '', 'Treatment': 'OPEN',
     'Assessment': 'PROVISIONAL'}
    for i, (category, item, status) in enumerate(entries57, 1)
])

if REGISTER_PATH57.is_file():
    try:
        old57 = pd.read_csv(REGISTER_PATH57).fillna('')
        if {'Category', 'Risk'}.issubset(old57.columns):
            prior57 = old57.drop_duplicates(['Category', 'Risk'], keep='last').set_index(['Category', 'Risk'])
            for col57 in ['Likelihood', 'Impact', 'Owner', 'Mitigation', 'Treatment', 'Assessment']:
                if col57 in prior57:
                    risk57[col57] = [
                        prior57.at[(r.Category, r.Risk), col57]
                        if (r.Category, r.Risk) in prior57.index
                        else getattr(r, col57)
                        for r in risk57.itertuples(index=False)
                    ]
    except (OSError, ValueError, pd.errors.ParserError):
        pass

for col57 in ['Likelihood', 'Impact']:
    risk57[col57] = pd.to_numeric(risk57[col57], errors='coerce').fillna(3).clip(1, 5).astype(int)
risk57['Treatment'] = risk57['Treatment'].where(risk57['Treatment'].isin(['OPEN', 'IN PROGRESS', 'MITIGATED', 'ACCEPTED']), 'OPEN')
risk57['Score'] = risk57['Likelihood'] * risk57['Impact']
risk57['Priority'] = risk57['Score'].map(lambda x: 'HIGH' if x >= 15 else 'MEDIUM' if x >= 8 else 'LOW')

def export57():
    risk57.to_csv(REGISTER_PATH57, index=False)
    risk57[risk57['Treatment'].isin(['OPEN', 'IN PROGRESS'])].to_csv(
        OUT57 / 'section57_open_risks.csv', index=False
    )
    risk57.groupby(['Category', 'Priority'], as_index=False).size().rename(
        columns={'size': 'Risks'}
    ).to_csv(OUT57 / 'section57_risk_summary.csv', index=False)

export57()
display(HTML('''<div style="background:linear-gradient(135deg,#22223b,#32344f 52%,#4a4e69);padding:22px 26px;border-radius:15px;font-family:Arial;margin:14px 0;color:white"><div style="color:#c9ada7;font-size:10px;letter-spacing:1.6px">SECTION 57 · PROJECT RISK MANAGEMENT</div><div style="font-size:24px;font-weight:bold;margin-top:7px">Risk Register & Release Decision Dashboard</div><div style="color:#f2e9e4;font-size:12px;margin-top:8px">Risk Matrix · Mitigation Tracking · Release Evidence</div></div>'''))

picker57 = widgets.Dropdown(
    options=[(f'{r.Risk_ID} · {r.Risk[:65]}', i) for i, r in risk57.iterrows()],
    description='Risk:', layout=widgets.Layout(width='95%')
)
likelihood57 = widgets.IntSlider(value=3, min=1, max=5, description='Likelihood:', continuous_update=False)
impact57 = widgets.IntSlider(value=3, min=1, max=5, description='Impact:', continuous_update=False)
owner57 = widgets.Text(description='Owner:', layout=widgets.Layout(width='95%'))
mitigation57 = widgets.Textarea(description='Mitigation:', layout=widgets.Layout(width='95%', height='75px'))
treatment57 = widgets.Dropdown(options=['OPEN', 'IN PROGRESS', 'MITIGATED', 'ACCEPTED'], description='Status:')
notice57 = widgets.Output()
submit57 = widgets.Button(description='Save Assessment', button_style='primary', icon='save')


def load57(change=None):
    r = risk57.iloc[picker57.value]
    likelihood57.value, impact57.value = int(r['Likelihood']), int(r['Impact'])
    owner57.value = str(r['Owner']) if pd.notna(r['Owner']) else ''
    mitigation57.value = str(r['Mitigation']) if pd.notna(r['Mitigation']) else ''
    treatment57.value = r['Treatment']

picker57.observe(load57, names='value')
load57()

def save57(button):
    i = picker57.value
    risk57.loc[i, ['Likelihood', 'Impact', 'Owner', 'Mitigation', 'Treatment', 'Assessment']] = [
        likelihood57.value, impact57.value, owner57.value.strip(),
        mitigation57.value.strip(), treatment57.value, 'REVIEWED'
    ]
    risk57.loc[i, 'Score'] = likelihood57.value * impact57.value
    risk57.loc[i, 'Priority'] = 'HIGH' if risk57.loc[i, 'Score'] >= 15 else 'MEDIUM' if risk57.loc[i, 'Score'] >= 8 else 'LOW'
    export57()
    with notice57:
        clear_output(wait=True)
        print('Risk assessment saved. Treatment is a recorded status, not verified resolution.')
    render57()

submit57.on_click(save57)
nav57 = widgets.ToggleButtons(
    options=['Risk Matrix', 'Risk Register', 'Assess Risks', 'Release Decision', 'Exports'],
    value='Risk Matrix', layout=widgets.Layout(width='100%')
)
panel57 = widgets.Output()

def render57(change=None):
    with panel57:
        clear_output(wait=True)
        total = len(risk57)
        high = int((risk57['Priority'] == 'HIGH').sum())
        open_count = int(risk57['Treatment'].isin(['OPEN', 'IN PROGRESS']).sum())
        reviewed = int((risk57['Assessment'] == 'REVIEWED').sum())
        cards = [('RISKS TRACKED', total), ('HIGH SCORES', high), ('ACTIVE RISKS', open_count), ('ASSESSED', f'{reviewed}/{total}')]
        display(HTML('<div style="display:flex;gap:9px;flex-wrap:wrap;margin:12px 0">' + ''.join(
            '<div style="flex:1;min-width:110px;border:1px solid #e6dfdc;border-radius:10px;padding:13px">'
            f'<div style="font:10px Arial;color:#9a8c98">{name}</div>'
            f'<div style="font:bold 22px Arial;color:#22223b">{value}</div></div>'
            for name, value in cards) + '</div>'))

        if nav57.value == 'Risk Matrix':
            chart = risk57.groupby(['Likelihood', 'Impact', 'Priority'], as_index=False).agg(
                Count=('Risk_ID', 'size')
            )
            fig = px.scatter(chart, x='Likelihood', y='Impact', size='Count', color='Priority',
                size_max=47, color_discrete_map={'HIGH': '#c97979', 'MEDIUM': '#9a8c98', 'LOW': '#4a4e69'},
                title='Likelihood × Impact — Risk Exposure Matrix', hover_data=['Count'])
            fig.update_layout(template='plotly_white', height=430,
                xaxis=dict(range=[0.5, 5.5], dtick=1), yaxis=dict(range=[0.5, 5.5], dtick=1),
                font=dict(family='Arial', color='#22223b'), margin=dict(l=25, r=20, t=60, b=35))
            display(fig)
            print('Ratings default to provisional 3/5 until reviewed; scores are planning estimates.')
        elif nav57.value == 'Risk Register':
            display(risk57.sort_values('Score', ascending=False))
        elif nav57.value == 'Assess Risks':
            display(widgets.VBox([picker57, likelihood57, impact57, owner57,
                                  mitigation57, treatment57, submit57, notice57]))
        elif nav57.value == 'Release Decision':
            display(risk57[risk57['Treatment'].isin(['OPEN', 'IN PROGRESS'])].sort_values('Score', ascending=False))
            print('Recommendation: Review high-scoring active risks, confirm mitigation evidence,')
            print('complete end-to-end testing, and obtain authorized client acceptance.')
            print('This register does not independently approve production release.')
        else:
            display(pd.DataFrame({'Report': [
                'section57_risk_register.csv', 'section57_open_risks.csv',
                'section57_risk_summary.csv']}))
            print('Saved to:', OUT57)

nav57.observe(render57, names='value')
display(widgets.VBox([nav57, panel57]))
render57()
```

<div style="color:#f2e9e4;font-size:12px;line-height:1.7;margin-top:8px">Analyze the retained trained model's available feature-importance scores, identify leading predictors through an interactive Pareto Chart, review preprocessing and interpretation limitations, and generate transparent model-explanation reports without retraining models or modifying the existing application.</div>

```python
ROOT58 = Path(globals().get('ROOT57', globals().get('ROOT56', Path.cwd()))).resolve()
OUT58 = ROOT58 / 'reports' / 'model_explainability'
OUT58.mkdir(parents=True, exist_ok=True)
MODEL58 = globals().get('MODEL44')
NAME58 = 'MODEL44'

if MODEL58 is None:
    for name58 in [
        'best_model', 'final_model', 'trained_model',
        'best_pipeline', 'final_pipeline', 'model'
    ]:
        candidate58 = globals().get(name58)
        if candidate58 is not None and callable(
            getattr(candidate58, 'predict', None)
        ):
            MODEL58, NAME58 = candidate58, name58
            break

IMPORTANCE58 = pd.DataFrame(
    columns=['Feature', 'Magnitude', 'Share_%', 'Cumulative_%']
)
METHOD58 = 'Not available'
NOTE58 = 'No fitted in-memory estimator with supported native importance was found.'

if MODEL58 is not None:
    estimator58 = (
        MODEL58.steps[-1][1]
        if hasattr(MODEL58, 'steps') else MODEL58
    )
    values58 = None

    if hasattr(estimator58, 'feature_importances_'):
        values58 = np.asarray(
            estimator58.feature_importances_, dtype=float
        ).ravel()
        METHOD58 = 'Native tree importance'
        NOTE58 = (
            'Tree importances can favor high-cardinality variables; '
            'they are not causal effects.'
        )

    elif hasattr(estimator58, 'coef_'):
        coeff58 = np.asarray(estimator58.coef_, dtype=float)
        values58 = (
            np.abs(coeff58).mean(axis=0)
            if coeff58.ndim == 2
            else np.abs(coeff58).ravel()
        )
        METHOD58 = 'Absolute model coefficients'
        NOTE58 = (
            'Coefficient magnitudes depend on feature scaling '
            'and encoding; class signs are not shown.'
        )

    if (
        values58 is not None
        and values58.size
        and np.isfinite(values58).all()
        and (values58 >= 0).all()
        and values58.sum() > 0
    ):
        names58 = None

        if hasattr(MODEL58, 'steps'):
            try:
                names58 = list(
                    MODEL58[:-1].get_feature_names_out()
                )
            except (AttributeError, ValueError, TypeError):
                pass

        if names58 is None and hasattr(
            estimator58, 'feature_names_in_'
        ):
            names58 = list(estimator58.feature_names_in_)

        if (
            names58 is None
            and hasattr(MODEL58, 'feature_names_in_')
            and not hasattr(MODEL58, 'steps')
        ):
            names58 = list(MODEL58.feature_names_in_)

        if names58 is None or len(names58) != len(values58):
            names58 = [
                f'Transformed feature {i+1}'
                for i in range(len(values58))
            ]
            NOTE58 += (
                ' Feature names could not be verified '
                'against the transformed input schema.'
            )

        IMPORTANCE58 = pd.DataFrame({
            'Feature': names58,
            'Magnitude': values58
        })

        IMPORTANCE58 = (
            IMPORTANCE58
            .sort_values('Magnitude', ascending=False)
            .reset_index(drop=True)
        )

        IMPORTANCE58['Share_%'] = (
            100 * IMPORTANCE58['Magnitude']
            / IMPORTANCE58['Magnitude'].sum()
        )

        IMPORTANCE58['Cumulative_%'] = (
            IMPORTANCE58['Share_%'].cumsum()
        )

        IMPORTANCE58[
            ['Magnitude', 'Share_%', 'Cumulative_%']
        ] = IMPORTANCE58[
            ['Magnitude', 'Share_%', 'Cumulative_%']
        ].round(4)

    else:
        NOTE58 = (
            'This estimator does not expose usable native '
            'importance values. No importance scores were invented.'
        )

IMPORTANCE58.to_csv(
    OUT58 / 'section58_feature_importance.csv',
    index=False
)

META58 = pd.DataFrame([
    (
        'Model variable',
        NAME58 if MODEL58 is not None else 'Unavailable'
    ),
    (
        'Estimator',
        type(MODEL58).__name__
        if MODEL58 is not None else 'Unavailable'
    ),
    ('Explanation method', METHOD58),
    ('Features with usable scores', len(IMPORTANCE58)),
    ('Interpretation', NOTE58),
    ('Model retrained', 'No')
], columns=['Property', 'Value'])

META58.to_csv(
    OUT58 / 'section58_explanation_metadata.csv',
    index=False
)

RECOMMEND58 = pd.DataFrame({
    'Recommendation': [
        'Validate preprocessing and feature-name alignment before sharing explanations.',
        'Compare feature rankings across held-out samples and time periods.',
        'For unsupported models, use a separately validated permutation-importance or SHAP workflow.',
        'Do not interpret native feature importance as causality or individual prediction explanations.',
        'Review predictive performance, fairness and model deployment evidence independently.'
    ]
})

RECOMMEND58.to_csv(
    OUT58 / 'section58_recommendations.csv',
    index=False
)

display(HTML("""
<div style="background:linear-gradient(135deg,#22223b,#32344f 52%,#4a4e69);
padding:22px 26px;border-radius:15px;font-family:Arial;color:white;margin:12px 0">
<div style="color:#c9ada7;font-size:10px;letter-spacing:1.6px">
SECTION 58 · MODEL EXPLAINABILITY</div>
<div style="font-size:24px;font-weight:bold;margin-top:7px">
Feature Importance & Prediction Transparency</div>
<div style="color:#f2e9e4;font-size:12px;margin-top:8px">
Retained Model · Pareto Analysis · Interpretation · Recommendations</div>
</div>
"""))

cards58 = [
    ('MODEL', NAME58 if MODEL58 is not None else 'Unavailable'),
    ('FEATURES', len(IMPORTANCE58)),
    ('METHOD', METHOD58),
    ('RETRAINED', 'NO')
]

display(HTML(
    '<div style="display:flex;gap:10px;flex-wrap:wrap;margin:12px 0">'
    + ''.join(
        '<div style="flex:1;min-width:110px;border:1px solid #e6dfdc;'
        'border-radius:10px;padding:14px">'
        f'<div style="font:10px Arial;color:#9a8c98">{escape(str(k))}</div>'
        f'<div style="font:bold 17px Arial;color:#22223b;'
        f'margin-top:7px">{escape(str(v))}</div></div>'
        for k, v in cards58
    )
    + '</div>'
))

limit58 = widgets.IntSlider(
    value=min(12, max(1, len(IMPORTANCE58))),
    min=1,
    max=max(1, len(IMPORTANCE58)),
    step=1,
    description='Top features:',
    continuous_update=False
)

nav58 = widgets.ToggleButtons(
    options=[
        'Pareto Overview',
        'Feature Ranking',
        'Method & Caveats',
        'Recommendations',
        'Exports'
    ],
    value='Pareto Overview',
    layout=widgets.Layout(width='100%')
)

panel58 = widgets.Output()

def render58(change=None):
    with panel58:
        clear_output(wait=True)

        if nav58.value == 'Pareto Overview':

            if IMPORTANCE58.empty:
                print(NOTE58)
            else:
                display(limit58)

                view58 = IMPORTANCE58.head(limit58.value)

                fig58 = go.Figure()

                fig58.add_trace(go.Bar(
                    x=view58['Feature'],
                    y=view58['Share_%'],
                    name='Importance share',
                    marker_color='#4a4e69'
                ))

                fig58.add_trace(go.Scatter(
                    x=view58['Feature'],
                    y=view58['Cumulative_%'],
                    yaxis='y2',
                    mode='lines+markers',
                    name='Cumulative share',
                    line=dict(color='#c9ada7', width=3),
                    marker=dict(size=8)
                ))

                fig58.update_layout(
                    title='Feature Importance Pareto Chart',
                    template='plotly_white',
                    height=450,
                    yaxis=dict(
                        title='Importance share (%)'
                    ),
                    yaxis2=dict(
                        title='Cumulative share (%)',
                        overlaying='y',
                        side='right',
                        range=[0, 105]
                    ),
                    xaxis=dict(tickangle=-40),
                    legend=dict(orientation='h', y=1.15),
                    margin=dict(l=45, r=45, t=65, b=100),
                    font=dict(
                        family='Arial',
                        color='#22223b'
                    )
                )

                display(fig58)

                print(
                    'Insight: The highest-ranked transformed features '
                    'account for the largest share of recorded native importance.'
                )
                print('Interpretation:', NOTE58)

        elif nav58.value == 'Feature Ranking':
            display(IMPORTANCE58)

        elif nav58.value == 'Method & Caveats':
            display(META58)
            print(
                'Global importance is not an explanation '
                'for a specific prediction.'
            )

        elif nav58.value == 'Recommendations':
            display(RECOMMEND58)

        else:
            display(pd.DataFrame({
                'File': [
                    'section58_feature_importance.csv',
                    'section58_explanation_metadata.csv',
                    'section58_recommendations.csv'
                ]
            }))
            print('Saved to:', OUT58)

nav58.observe(render58, names='value')
limit58.observe(render58, names='value')
display(widgets.VBox([nav58, panel58]))
render58()
```

<div style="color:#f2e9e4;font-size:12px;line-height:1.7;margin-top:8px">Assess available prediction-error evidence across job-market segments, examine group-level reliability and error dispersion using an interactive beeswarm-style plot, identify potential performance disparities and produce responsible-AI recommendations without retraining the existing model or inferring protected characteristics.</div>
</div>


```python
ROOT59 = Path(globals().get("ROOT58", globals().get("ROOT57", Path.cwd()))).resolve()
OUT59 = ROOT59 / "reports" / "responsible_ai"
OUT59.mkdir(parents=True, exist_ok=True)
PRED59 = globals().get("PREDICTIONS44", pd.DataFrame())
FEATURES59 = globals().get("X_TEST44")
TASK59 = globals().get("TASK_TYPE44", "Unknown")

PRED59 = PRED59.copy() if isinstance(PRED59, pd.DataFrame) else pd.DataFrame()

SENSITIVE59 = (
    "gender", "sex", "race", "ethnicity", "religion",
    "disability", "age", "caste", "nationality",
    "sexual_orientation", "political"
)

PREFERRED59 = [
    "employment_type", "job_type", "work_type",
    "industry", "experience_level", "company_size",
    "remote_type", "country", "region"
]

SEGMENTS59 = {}
DATA59 = pd.DataFrame()

if (
    not PRED59.empty
    and isinstance(FEATURES59, pd.DataFrame)
    and len(PRED59) == len(FEATURES59)
):
    candidates59 = [
        c for c in FEATURES59.columns
        if not any(s in str(c).lower() for s in SENSITIVE59)
        and 2 <= FEATURES59[c].nunique(dropna=True) <= 12
        and (
            not pd.api.types.is_numeric_dtype(FEATURES59[c])
            or str(c).lower() in PREFERRED59
        )
    ]

    ordered59 = sorted(
        candidates59,
        key=lambda c: (
            str(c).lower() not in PREFERRED59,
            str(c).lower()
        )
    )

    for col59 in ordered59[:12]:
        SEGMENTS59[str(col59)] = (
            FEATURES59[col59]
            .reset_index(drop=True)
            .fillna("Missing")
            .astype(str)
        )

    DATA59 = PRED59.reset_index(drop=True).copy()

    if TASK59 == "Classification" and "Error" in DATA59:
        DATA59["Diagnostic_Error"] = pd.to_numeric(
            DATA59["Error"], errors="coerce"
        )
        UNIT59 = "Error rate (%)"

    elif TASK59 == "Regression" and "Absolute_Error" in DATA59:
        DATA59["Diagnostic_Error"] = pd.to_numeric(
            DATA59["Absolute_Error"], errors="coerce"
        )
        UNIT59 = "Mean absolute error"

    else:
        DATA59 = pd.DataFrame()
        UNIT59 = "Unavailable"
else:
    UNIT59 = "Unavailable"

if not DATA59.empty:
    DATA59 = DATA59[
        DATA59["Diagnostic_Error"].notna()
        & np.isfinite(DATA59["Diagnostic_Error"])
    ].copy()

def evaluate59(segment):
    if DATA59.empty or segment not in SEGMENTS59:
        return pd.DataFrame(), pd.DataFrame()

    data = DATA59.copy()
    data["Segment"] = SEGMENTS59[segment].loc[data.index].values

    counts = data["Segment"].value_counts()
    eligible = counts[counts >= 20].index

    data = data[data["Segment"].isin(eligible)].copy()

    if data.empty:
        return pd.DataFrame(), pd.DataFrame()

    summary = data.groupby("Segment", as_index=False).agg(
        Observations=("Diagnostic_Error", "size"),
        Mean_Error=("Diagnostic_Error", "mean"),
        Median_Error=("Diagnostic_Error", "median"),
        P90_Error=("Diagnostic_Error", lambda s: s.quantile(.90))
    )

    if TASK59 == "Classification":
        for col in ["Mean_Error", "Median_Error", "P90_Error"]:
            summary[col] = summary[col] * 100

    summary = summary.sort_values(
        "Mean_Error", ascending=False
    ).round(3)

    return data, summary

def diagnostics59(segment):
    data, summary = evaluate59(segment)

    if summary.empty or len(summary) < 2:
        return "INSUFFICIENT COMPARABLE GROUPS", None

    largest = float(summary["Mean_Error"].max())
    smallest = float(summary["Mean_Error"].min())
    gap = largest - smallest

    return "GROUP DIFFERENCES OBSERVED", round(gap, 3)

SELECT59 = widgets.Dropdown(
    options=list(SEGMENTS59),
    description="Segment:",
    layout=widgets.Layout(width="85%"),
    disabled=not bool(SEGMENTS59)
)

NAV59 = widgets.ToggleButtons(
    options=[
        "Error Distribution",
        "Group Comparison",
        "Responsible AI Review",
        "Recommendations",
        "Exports"
    ],
    value="Error Distribution",
    layout=widgets.Layout(width="100%")
)

PANEL59 = widgets.Output()

def show59(view):
    selected = SELECT59.value
    data, summary = evaluate59(selected)
    status, gap = diagnostics59(selected)

    if view == "Error Distribution":
        if data.empty:
            print(
                "No suitable segment-aligned prediction data. "
                "The retained model has not been modified."
            )
            return

        display(SELECT59)

        sample = data.sample(
            n=min(1500, len(data)),
            random_state=59
        )

        fig = px.strip(
            sample,
            x="Segment",
            y="Diagnostic_Error",
            color="Segment",
            stripmode="overlay",
            title=f"Prediction Error Distribution — {selected}",
            color_discrete_sequence=[
                "#4a4e69", "#9a8c98", "#c9ada7",
                "#6d687d", "#b8a6ae"
            ]
        )

        fig.update_traces(jitter=0.65, opacity=0.48)

        fig.update_layout(
            template="plotly_white",
            height=450,
            showlegend=False,
            xaxis_title=selected,
            yaxis_title=(
                "Misclassification (0 = correct, 1 = incorrect)"
                if TASK59 == "Classification"
                else "Absolute prediction error"
            ),
            font=dict(family="Arial", color="#22223b"),
            margin=dict(l=30, r=20, t=65, b=65)
        )

        display(fig)

        print("Insight:", status)
        print("Maximum group mean-error difference:", gap, UNIT59)
        print(
            "Interpretation: Each dot represents an observed "
            "prediction error. Differences may reflect sample "
            "composition, not discrimination or causality."
        )

    elif view == "Group Comparison":
        display(SELECT59)
        display(summary)

        if not summary.empty:
            print("Largest observed mean-error gap:", gap, UNIT59)
            print(
                "Groups with fewer than 20 observations "
                "are excluded from comparison."
            )

    elif view == "Responsible AI Review":
        display(pd.DataFrame([
            ("Prediction evidence", "Available" if not PRED59.empty else "Unavailable"),
            ("Model task", TASK59),
            ("Selected segment", selected or "Unavailable"),
            ("Eligible groups", len(summary)),
            ("Diagnostic", status),
            ("Protected-attribute audit", "Not performed"),
            ("Causal fairness conclusion", "Not supported"),
            ("Model retrained", "No")
        ], columns=["Assessment", "Result"]))

    elif view == "Recommendations":
        display(pd.DataFrame({
            "Recommended Action": [
                "Verify predictions originate from a held-out evaluation set.",
                "Investigate groups with elevated error and adequate sample sizes.",
                "Check group composition, label quality and data leakage.",
                "Use confidence intervals before interpreting small differences.",
                "Conduct a separately authorized protected-group fairness audit where appropriate.",
                "Do not equate equal error rates with proven fairness."
            ]
        }))

    else:
        data.to_csv(
            OUT59 / "section59_segment_predictions.csv",
            index=False
        )
        summary.to_csv(
            OUT59 / "section59_group_error_comparison.csv",
            index=False
        )
        print("Reports exported to:", OUT59)

def render59(change=None):
    with PANEL59:
        clear_output(wait=True)
        show59(NAV59.value)

NAV59.observe(render59, names="value")
SELECT59.observe(render59, names="value")

display(HTML("""
<div style="background:linear-gradient(135deg,#22223b,#4a4e69);
padding:22px;border-radius:14px;color:white;font-family:Arial">
<div style="color:#c9ada7;font-size:10px;letter-spacing:1.5px">
SECTION 59 · RESPONSIBLE AI DIAGNOSTICS</div>
<div style="font-size:24px;font-weight:bold;margin-top:7px">
Prediction Reliability & Group-Level Error Review</div>
<div style="color:#f2e9e4;font-size:12px;margin-top:8px">
Retained Model · Segment Analysis · Error Distribution</div>
</div>
"""))

display(HTML(
    '<div style="display:flex;gap:10px;flex-wrap:wrap;margin:14px 0">'
    + "".join(
        '<div style="flex:1;min-width:110px;border:1px solid #e6dfdc;'
        'border-radius:11px;padding:14px">'
        f'<div style="font:10px Arial;color:#9a8c98">{label}</div>'
        f'<div style="font:bold 22px Arial;color:#22223b">{value}</div>'
        '</div>'
        for label, value in [
            ("PREDICTIONS", len(PRED59)),
            ("SEGMENTS", len(SEGMENTS59)),
            ("TASK", TASK59),
            ("RETRAINED", "NO")
        ]
    )
    + '</div>'
))

display(widgets.VBox([NAV59, PANEL59]))
render59()

if SEGMENTS59:
    initial_data59, initial_summary59 = evaluate59(SELECT59.value)
    initial_summary59.to_csv(
        OUT59 / "section59_group_error_comparison.csv",
        index=False
    )
```
